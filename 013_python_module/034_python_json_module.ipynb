The `json` module in Python is part of the standard library and provides functions for working with JSON (JavaScript Object Notation) data. JSON is a lightweight data interchange format that is widely used for storing and exchanging data, particularly in web services and APIs. The `json` module allows you to serialize (convert Python objects to JSON) and deserialize (convert JSON to Python objects) data efficiently.

### 1. **Introduction to JSON**

JSON is a text-based format that represents data as a series of key-value pairs. It is easy to read and write for humans and machines alike, and it is language-independent. JSON is often used for exchanging data between a client (e.g., web browser) and server.

JSON supports the following basic types:

- **Objects** (called `dict` in Python)
- **Arrays** (called `list` in Python)
- **String**
- **Number** (integer or floating-point)
- **Boolean** (`true` or `false`)
- **Null** (`None` in Python)

A typical JSON object might look like this:

```json
{
  "name": "John",
  "age": 30,
  "is_student": false,
  "courses": ["Math", "Science"],
  "address": {
    "street": "123 Main St",
    "city": "Somewhere"
  }
}
```

### 2. **Functions in the `json` Module**

The `json` module provides several important functions for working with JSON data.

#### **json.dumps()** (Serialize Python Objects to JSON)

The `json.dumps()` function converts a Python object into a JSON string. This is known as **serialization** or **encoding**.

```python
import json

# Python dictionary
person = {
    "name": "Alice",
    "age": 25,
    "is_student": True,
    "courses": ["Math", "Art"]
}

# Serialize to JSON
json_string = json.dumps(person)
print(json_string)
```

Output:

```json
{ "name": "Alice", "age": 25, "is_student": true, "courses": ["Math", "Art"] }
```

- **Indentation**: You can pretty-print JSON with indentation for better readability.

```python
json_string = json.dumps(person, indent=4)
print(json_string)
```

Output:

```json
{
  "name": "Alice",
  "age": 25,
  "is_student": true,
  "courses": ["Math", "Art"]
}
```

- **Other Options for `dumps()`**:
  - **`separators`**: To control the separators between items in the JSON output.
  - **`ensure_ascii`**: By default, `ensure_ascii=True` ensures all characters are escaped using ASCII encoding. You can set it to `False` to allow non-ASCII characters.

```python
json_string = json.dumps(person, separators=(',', ': '), ensure_ascii=False)
```

#### **json.dump()** (Serialize to a File)

The `json.dump()` function is used to serialize a Python object and write it directly to a file.

```python
import json

# Python dictionary
person = {
    "name": "Bob",
    "age": 32,
    "is_student": False,
    "courses": ["History", "Philosophy"]
}

# Serialize and write to file
with open('person.json', 'w') as file:
    json.dump(person, file, indent=4)
```

This will create a file named `person.json` with the serialized JSON content.

#### **json.loads()** (Deserialize JSON to Python Object)

The `json.loads()` function deserializes (or decodes) a JSON string into a Python object, such as a dictionary or list.

```python
import json

# JSON string
json_string = '{"name": "John", "age": 30, "is_student": false, "courses": ["Math", "Science"]}'

# Deserialize JSON string to Python dictionary
person = json.loads(json_string)
print(person)
```

Output:

```python
{'name': 'John', 'age': 30, 'is_student': False, 'courses': ['Math', 'Science']}
```

#### **json.load()** (Deserialize JSON from a File)

The `json.load()` function is used to read JSON data from a file and deserialize it into a Python object.

```python
import json

# Read JSON from file and deserialize
with open('person.json', 'r') as file:
    person = json.load(file)
    print(person)
```

This assumes there is a `person.json` file that contains valid JSON data.

### 4. **Customizing Serialization and Deserialization**

You may need to customize how Python objects are converted to JSON or how JSON data is converted back into Python objects. The `json` module allows you to define custom behavior for both serialization and deserialization.

#### **Custom Serialization with `default` Parameter**

You can specify a custom function to convert objects that are not serializable by default.

```python
import json

# Custom object
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Custom serialization function
def person_serializer(obj):
    if isinstance(obj, Person):
        return {'name': obj.name, 'age': obj.age}
    raise TypeError("Type not serializable")

# Creating a Person object
person = Person("Eve", 28)

# Serialize using custom function
json_string = json.dumps(person, default=person_serializer)
print(json_string)
```

Output:

```json
{ "name": "Eve", "age": 28 }
```

#### **Custom Deserialization with `object_hook` Parameter**

You can specify a custom function for deserializing JSON objects back into Python objects.

```python
import json

# Custom deserialization function
def person_decoder(dct):
    if 'name' in dct and 'age' in dct:
        return Person(dct['name'], dct['age'])
    return dct

# JSON string
json_string = '{"name": "Tom", "age": 35}'

# Deserialize using custom function
person = json.loads(json_string, object_hook=person_decoder)
print(person.name, person.age)
```

### 5. **Handling Errors in JSON Parsing**

When working with JSON, errors may occur if the JSON is malformed or invalid. The `json` module raises exceptions for such errors.

#### **`json.JSONDecodeError`**

This exception is raised when there’s an error in decoding (deserializing) the JSON string.

```python
import json

invalid_json_string = '{"name": "John", "age": 30'  # Missing closing bracket

try:
    person = json.loads(invalid_json_string)
except json.JSONDecodeError as e:
    print(f"Error decoding JSON: {e}")
```

#### **`TypeError`**

This exception occurs if an object is passed to `json.dumps()` that is not serializable by default.

```python
import json

# Passing a custom object that is not serializable by default
try:
    json.dumps(object())  # This will raise TypeError
except TypeError as e:
    print(f"Error: {e}")
```

````python

import json
import os # For file operations cleanup

# --- 1. Python to JSON (Serialization / Encoding) ---
# Goal: Convert Python objects (dict, list, etc.) into JSON formatted strings or files.

# 1.1 json.dumps() - Python object to JSON string
# This is the most common way to convert a Python dictionary to a JSON string.
print("--- 1.1 json.dumps() - Python object to JSON string ---")

# A Python dictionary representing some data
python_dict = {
    "name": "Alice",
    "age": 30,
    "isStudent": False,
    "courses": ["History", "Physics"],
    "address": {
        "street": "456 Oak Ave",
        "city": "Villagetown"
    },
    "notes": None, # Python's None maps to JSON's null
    "id": 12345
}

print("Original Python Dictionary:")
print(python_dict)
print(f"Type: {type(python_dict)}\n")

# Convert the Python dictionary to a JSON string
json_string_simple = json.dumps(python_dict)
print("JSON String (simple):")
print(json_string_simple)
print(f"Type: {type(json_string_simple)}\n")

# Pretty printing: make the JSON string human-readable with indentation
json_string_pretty = json.dumps(python_dict, indent=4)
print("JSON String (pretty printed with indent=4):")
print(json_string_pretty)
print(f"Type: {type(json_string_pretty)}\n")

# Sorting keys: output JSON with keys sorted alphabetically
json_string_sorted_keys = json.dumps(python_dict, indent=4, sort_keys=True)
print("JSON String (pretty printed with sorted keys):")
print(json_string_sorted_keys)
print(f"Type: {type(json_string_sorted_keys)}\n")

# Compact output: remove whitespace after separators for minimal size
json_string_compact = json.dumps(python_dict, separators=(',', ':'))
print("JSON String (compact):")
print(json_string_compact)
print(f"Type: {type(json_string_compact)}\n")

# Encoding a Python list
python_list = [
    {"item": "Laptop", "price": 1200},
    {"item": "Mouse", "price": 25},
    "Keyboard"
]
json_list_string = json.dumps(python_list, indent=2)
print("Python List to JSON String:")
print(json_list_string)
print(f"Type: {type(json_list_string)}\n")

# 1.2 json.dump() - Python object to JSON file
# Writes the JSON representation of a Python object directly to a file.
print("--- 1.2 json.dump() - Python object to JSON file ---")

file_name_dump = "data_output.json"
try:
    with open(file_name_dump, 'w') as f:
        # Use indent for readability in the file
        json.dump(python_dict, f, indent=4)
    print(f"Python dictionary successfully dumped to '{file_name_dump}'")
except IOError as e:
    print(f"Error writing to file '{file_name_dump}': {e}")
print("-" * 50 + "\n")


# --- 2. JSON to Python (Deserialization / Decoding) ---
# Goal: Convert JSON formatted strings or files back into Python objects.

# 2.1 json.loads() - JSON string to Python object
# Converts a JSON formatted string into a Python dictionary or list.
print("--- 2.1 json.loads() - JSON string to Python object ---")

# A JSON string (can be obtained from a file or network request)
json_string_to_load = """
{
    "product": "Smartphone",
    "brand": "TechCorp",
    "price": 699.99,
    "inStock": true,
    "features": ["5G", "OLED Display", "Dual Camera"],
    "dimensions": null
}
"""

print("Original JSON String:")
print(json_string_to_load)

# Convert the JSON string to a Python dictionary
try:
    loaded_python_dict = json.loads(json_string_to_load)
    print("\nLoaded Python Dictionary:")
    print(loaded_python_dict)
    print(f"Type: {type(loaded_python_dict)}")
    print(f"Accessing data: Brand = {loaded_python_dict['brand']}")
    print(f"Accessing list element: First feature = {loaded_python_dict['features'][0]}\n")
except json.JSONDecodeError as e:
    print(f"Error decoding JSON string: {e}\n")

# Loading a JSON array string into a Python list
json_array_string_to_load = '[{"id": 1, "name": "Item A"}, {"id": 2, "name": "Item B"}]'
print("Original JSON Array String:")
print(json_array_string_to_load)
try:
    loaded_python_list = json.loads(json_array_string_to_load)
    print("\nLoaded Python List:")
    print(loaded_python_list)
    print(f"Type: {type(loaded_python_list)}")
    print(f"Accessing list element: First item name = {loaded_python_list[0]['name']}\n")
except json.JSONDecodeError as e:
    print(f"Error decoding JSON array string: {e}\n")


# 2.2 json.load() - JSON file to Python object
# Reads JSON data directly from a file and converts it into a Python object.
print("--- 2.2 json.load() - JSON file to Python object ---")

# Using the file created earlier by json.dump()
file_name_load = "data_output.json"
try:
    with open(file_name_load, 'r') as f:
        loaded_data_from_file = json.load(f)
    print(f"Data successfully loaded from '{file_name_load}':")
    print(loaded_data_from_file)
    print(f"Type: {type(loaded_data_from_file)}\n")
except FileNotFoundError:
    print(f"Error: File '{file_name_load}' not found. Please ensure it was created by the 'json.dump()' section.")
except json.JSONDecodeError as e:
    print(f"Error decoding JSON from file '{file_name_load}': {e}\n")
print("-" * 50 + "\n")


# --- 3. Custom Encoders and Decoders (Advanced) ---
# For handling custom Python objects that aren't natively supported by JSON.

# 3.1 Custom JSONEncoder
# Extend JSONEncoder to define how your custom objects should be serialized.
print("--- 3.1 Custom JSONEncoder ---")

class Product:
    def __init__(self, name, price, stock_date):
        self.name = name
        self.price = price
        self.stock_date = stock_date # datetime object, not directly JSON serializable

    def __repr__(self):
        return f"Product({self.name}, {self.price}, {self.stock_date})"

import datetime

class ProductEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, Product):
            # Return a dictionary representation that is JSON-serializable
            return {
                "__product__": True, # Marker to help with decoding
                "name": obj.name,
                "price": obj.price,
                "stock_date": obj.stock_date.isoformat() # Convert datetime to ISO 8601 string
            }
        # Let the base class default method raise the TypeError for other types
        return super().default(obj)

# Create an instance of our custom object
my_product = Product("Smartphone X", 799.99, datetime.date(2025, 6, 24))
print(f"Original Product object: {my_product}")

# Attempt to dump without custom encoder (will fail)
try:
    json.dumps(my_product)
except TypeError as e:
    print(f"Attempting to dump Product without custom encoder (expected error): {e}")

# Dump with custom encoder
json_product_string = json.dumps(my_product, indent=4, cls=ProductEncoder)
print("\nProduct object serialized with CustomEncoder:")
print(json_product_string)
print("-" * 50 + "\n")


# 3.2 Custom Object Hook for Decoding
# Use `object_hook` in json.loads() or json.load() to reconstruct custom objects.
print("--- 3.2 Custom Object Hook for Decoding ---")

def product_object_hook(dct):
    # Check for the custom marker added during encoding
    if "__product__" in dct:
        # Reconstruct the Product object
        return Product(
            dct['name'],
            dct['price'],
            datetime.date.fromisoformat(dct['stock_date']) # Convert ISO 8601 string back to datetime.date
        )
    return dct # Return the dictionary as is if it's not our custom object

# Decode the JSON string back into our custom object
loaded_product = json.loads(json_product_string, object_hook=product_object_hook)
print("Product object deserialized with object_hook:")
print(loaded_product)
print(f"Type of loaded object: {type(loaded_product)}")
print(f"Stock Date Type: {type(loaded_product.stock_date)}")
print("-" * 50 + "\n")


# --- 4. JSON Data Type Mapping Summary ---
print("--- 4. JSON Data Type Mapping Summary ---")
print("Python -> JSON:")
print("  dict         -> object")
print("  list, tuple  -> array")
print("  str          -> string")
print("  int, float   -> number")
print("  True         -> true")
print("  False        -> false")
print("  None         -> null")

print("\nJSON -> Python:")
print("  object       -> dict")
print("  array        -> list")
print("  string       -> str")
print("  number       -> int (if integer), float (if decimal)")
print("  true         -> True")
print("  false        -> False")
print("  null         -> None")
print("-" * 50 + "\n")


# --- 5. Cleanup ---
# Remove the generated file.
print("--- 5. Cleanup ---")
if os.path.exists(file_name_dump):
    os.remove(file_name_dump)
    print(f"Cleaned up: Removed '{file_name_dump}'")
else:
    print(f"No file '{file_name_dump}' to clean up.")
print("-" * 50 + "\n")
````

````python
import json

# --- json.dumps() Syntax and Usage ---

# json.dumps(obj, *, skipkeys=False, ensure_ascii=True, check_circular=True,
#            allow_nan=True, cls=None, indent=None, separators=None,
#            default=None, sort_keys=False)

# Parameters:
# 1. obj: The Python object to be serialized (converted to a JSON string).
#    Common types: dict, list, str, int, float, bool, None.

# 2. skipkeys (bool): If True, dict keys that are not of type str, int, float,
#    or bool will be skipped. Otherwise (default False), a TypeError will be raised.

# 3. ensure_ascii (bool): If True (default), all non-ASCII characters in the output
#    are escaped (e.g., '€' becomes '\u20ac'). If False, these characters are output
#    directly. Setting to False can make the output more readable and smaller
#    if non-ASCII characters are common and the target system supports UTF-8.

# 4. check_circular (bool): If True (default), `dumps` checks for circular
#    references in lists, dictionaries, and custom encoded objects. If a circular
#    reference is found, a `ValueError` is raised. If False, circular references
#    can lead to an infinite recursion.

# 5. allow_nan (bool): If True (default), `NaN`, `Infinity`, and `-Infinity`
#    (Not a Number, positive/negative infinity for floats) will be serialized
#    as `NaN`, `Infinity`, and `-Infinity` in JSON, which is technically
#    non-standard JSON but commonly accepted. If False, a `ValueError` is raised.

# 6. cls (class): A custom JSONEncoder subclass. If provided, instances of `cls`
#    will be used to serialize objects that are not natively serializable by `json`.
#    This is used for custom object serialization.

# 7. indent (int or str): If `indent` is a non-negative integer or a string,
#    then JSON array elements and object members will be pretty-printed with
#    that indent level. An `indent` of 0, negative, or `''` (empty string) will
#    only insert newlines. If `indent` is a positive integer, that many spaces
#    per indentation level. If `indent` is a string (e.g., '\t'), that string
#    is used for indentation. Default is `None` (most compact output).

# 8. separators (tuple): A `(item_separator, key_separator)` tuple. Default is
#    `(', ', ': ')`. Used to specify the strings inserted between items and keys.
#    To get the most compact JSON, use `(',', ':')`.

# 9. default (function): A function that will be called for objects that `dumps`
#    doesn't know how to serialize. It should return a JSON-serializable
#    representation of `obj`, or raise a `TypeError`. This is an alternative
#    to `cls` for simple custom type serialization.

# 10. sort_keys (bool): If True, the output of dictionaries will be sorted by key.
#     This can make output canonical and easier to compare, but might be slower.


# --- Practical Examples ---

# Example data
data = {
    "name": "Chandra",
    "age": 35,
    "is_employed": True,
    "skills": ["Python", "JSON", "APIs"],
    "address": {"city": "Dhaka", "country": "Bangladesh"},
    "notes": None,
    "temperature": 25.5,
    "symbol": "€",
    "timestamp": 1719213316.0 # Example timestamp
}

print("--- 1. Basic dumps (default behavior) ---")
# Most compact output, no special characters escaped by default (if all ASCII)
json_output = json.dumps(data)
print(json_output)
print(f"Type of output: {type(json_output)}\n")


print("--- 2. dumps with indent (pretty print) ---")
# Makes the JSON string readable for humans
json_output_pretty = json.dumps(data, indent=4) # 4 spaces for indentation
print(json_output_pretty)
print("\n")


print("--- 3. dumps with sort_keys (alphabetical order of keys) ---")
# Useful for canonical representation or easier comparison
json_output_sorted = json.dumps(data, indent=4, sort_keys=True)
print(json_output_sorted)
print("\n")


print("--- 4. dumps with ensure_ascii=False (for non-ASCII characters) ---")
# Outputs non-ASCII characters directly, requires UTF-8 handling when reading
json_output_unicode = json.dumps(data, indent=4, ensure_ascii=False)
print(json_output_unicode)
print("\n")


print("--- 5. dumps with separators (for compact output) ---")
# Overrides default separators (", ", ": ") to (",", ":")
# Removes whitespace for smallest possible string representation
json_output_compact_sep = json.dumps(data, separators=(',', ':'))
print(json_output_compact_sep)
print("\n")


print("--- 6. dumps with skipkeys=True (handling non-string/non-numeric keys) ---")
# Example data with a non-string key (tuple)
data_with_bad_key = {
    "good_key": "value",
    (1, 2): "this key is a tuple" # Not a valid JSON key type
}

try:
    # This will raise a TypeError by default because (1,2) is not a string/number
    json.dumps(data_with_bad_key)
except TypeError as e:
    print(f"Error without skipkeys=True (expected): {e}")

# With skipkeys=True, the invalid key-value pair is skipped
json_output_skipkeys = json.dumps(data_with_bad_key, skipkeys=True, indent=2)
print("Output with skipkeys=True (invalid key skipped):")
print(json_output_skipkeys)
print("\n")


print("--- 7. dumps with default callable (custom object serialization) ---")
import datetime

class MyComplexObject:
    def __init__(self, id, created_at):
        self.id = id
        self.created_at = created_at # datetime object

    def __repr__(self):
        return f"MyComplexObject(id={self.id}, created_at={self.created_at})"

def custom_serializer(obj):
    # This function is called if json.dumps doesn't know how to serialize `obj`
    if isinstance(obj, MyComplexObject):
        # Return a dictionary representation that is JSON-serializable
        return {
            "object_id": obj.id,
            "created_at_iso": obj.created_at.isoformat(), # Convert datetime to string
            "__type__": "MyComplexObject" # Add a marker for potential decoding
        }
    # For any other unsupported type, let the default TypeError be raised
    raise TypeError(f"Object of type {obj.__class__.__name__} is not JSON serializable")

complex_data = {
    "item_code": "XYZ789",
    "details": MyComplexObject(101, datetime.datetime.now()) # Custom object here
}

# Attempt to dump without default function (will raise TypeError)
try:
    json.dumps(complex_data)
except TypeError as e:
    print(f"Error dumping complex_data without default function (expected): {e}")

# Dump with the custom `default` function
json_output_custom_default = json.dumps(complex_data, indent=4, default=custom_serializer)
print("Output with custom 'default' function:")
print(json_output_custom_default)
print("\n")


print("--- 8. dumps with cls (custom JSONEncoder subclass) ---")
# More powerful for complex custom type hierarchies and reusable encoders

class AnotherComplexObject:
    def __init__(self, name, value):
        self.name = name
        self.value = value

    def __repr__(self):
        return f"AnotherComplexObject(name='{self.name}', value={self.value})"

class CustomEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, MyComplexObject):
            return {
                "object_id": obj.id,
                "created_at_iso": obj.created_at.isoformat(),
                "__type__": "MyComplexObject"
            }
        elif isinstance(obj, AnotherComplexObject):
            return {
                "custom_name": obj.name,
                "custom_value": obj.value,
                "__type__": "AnotherComplexObject"
            }
        # Let the base class default method handle other types or raise TypeError
        return super().default(obj)

multi_object_data = {
    "item1": MyComplexObject(201, datetime.datetime(2024, 1, 15)),
    "item2": AnotherComplexObject("DataPoint", 99.5),
    "status": "processed"
}

json_output_custom_cls = json.dumps(multi_object_data, indent=4, cls=CustomEncoder)
print("Output with custom 'cls' (JSONEncoder subclass):")
print(json_output_custom_cls)
print("\n")


print("--- 9. dumps with allow_nan=False (handling NaN/Infinity) ---")
data_with_nan = {
    "value": float('nan'), # Not a Number
    "pos_inf": float('inf'), # Positive Infinity
    "neg_inf": float('-inf') # Negative Infinity
}

# Default behavior (allow_nan=True)
json_output_nan_allowed = json.dumps(data_with_nan, indent=4)
print("Output with allow_nan=True (default):")
print(json_output_nan_allowed)

# With allow_nan=False (will raise ValueError)
try:
    json.dumps(data_with_nan, allow_nan=False)
except ValueError as e:
    print(f"\nError with allow_nan=False (expected): {e}")
print("\n")
````

````python

import json
import os
import datetime

# --- json.load() Syntax and Usage ---

# json.load(fp, *, cls=None, object_hook=None, parse_float=None,
#           parse_int=None, parse_constant=None, object_pairs_hook=None, **kw)

# Parameters:
# 1. fp: A file-like object (e.g., an opened file in read mode `f = open('file.json', 'r')`).
#    `json.load()` reads the entire JSON document from this file-like object.

# 2. cls (class): A custom JSONDecoder subclass. If provided, instances of `cls`
#    will be used to decode the JSON. This is for advanced custom type deserialization.

# 3. object_hook (function): An optional function that will be called with the result
#    of any object literal (JSON object) decoded. This function can be used to
#    implement custom decoders for custom Python objects. If specified, the return
#    value of `object_hook` will be used instead of the `dict` that would otherwise be produced.

# 4. parse_float (function): An optional function that will be called with the
#    string of every JSON float. By default, this is `float()`. This can be used
#    to parse JSON floats into a different numeric type or to handle specific precision.

# 5. parse_int (function): An optional function that will be called with the
#    string of every JSON int. By default, this is `int()`. Can be used to parse
#    large integers into a different type (e.g., a custom BigInt class).

# 6. parse_constant (function): An optional function that will be called with the
#    string of any JSON constant (e.g., `"-Infinity"`, `"Infinity"`, `"NaN"`).
#    By default, it raises `ValueError` for invalid JSON values, but allows `Infinity`,
#    `-Infinity`, `NaN` if `allow_nan` (a dumps parameter, not load) is `True`.
#    This can be used to convert these constants into specific Python values.

# 7. object_pairs_hook (function): An optional function that will be called with
#    the result of any object literal (JSON object) decoded, as a list of
#    `(key, value)` pairs. This allows for custom handling of duplicate keys
#    or preserving key order, as default `dict`s don't guarantee order (though
#    Python 3.7+ dicts preserve insertion order). If `object_hook` is also defined,
#    `object_pairs_hook` takes precedence.

# 8. **kw (arbitrary keyword arguments): Additional keyword arguments are passed
#    to the constructor of the `cls` (JSONDecoder subclass) if provided.


# --- Preparation: Create a sample JSON file for loading ---
print("--- Preparation: Creating a sample JSON file (data_input.json) ---")
sample_data = {
    "name": "Fahim",
    "city": "Chittagong",
    "occupation": "Engineer",
    "age": 28,
    "is_active": True,
    "skills": ["Python", "Networking", "Linux"],
    "preferences": {
        "theme": "dark",
        "notifications": "email"
    },
    "last_login": "2025-06-23T10:30:00", # ISO 8601 string
    "null_field": None,
    "version": 1.5
}

file_name = "data_input.json"
try:
    with open(file_name, 'w', encoding='utf-8') as f:
        json.dump(sample_data, f, indent=4, ensure_ascii=False)
    print(f"Sample JSON file '{file_name}' created successfully.\n")
except IOError as e:
    print(f"Error creating sample file: {e}")
    # Exit if file creation fails, as subsequent examples depend on it
    exit(1)


# --- Practical Examples of json.load() ---

print("--- 1. Basic json.load() ---")
# Reads the entire JSON document from the specified file and converts it
# into a Python dictionary or list.
try:
    with open(file_name, 'r', encoding='utf-8') as f:
        loaded_data = json.load(f)
    print(f"Data loaded from '{file_name}':")
    print(loaded_data)
    print(f"Type of loaded_data: {type(loaded_data)}")
    print(f"Accessing a value: Name = {loaded_data['name']}\n")
except FileNotFoundError:
    print(f"Error: File '{file_name}' not found. Please ensure it was created.")
except json.JSONDecodeError as e:
    print(f"Error decoding JSON from file: {e}\n")
print("-" * 50 + "\n")


print("--- 2. json.load() with object_hook (Custom Object Deserialization) ---")
# Use object_hook to convert specific JSON objects back into custom Python objects.

class UserProfile:
    def __init__(self, name, city, occupation, age, is_active, skills, preferences, last_login, null_field, version):
        self.name = name
        self.city = city
        self.occupation = occupation
        self.age = age
        self.is_active = is_active
        self.skills = skills
        self.preferences = preferences
        self.last_login = last_login # Will be converted from string to datetime object
        self.null_field = null_field
        self.version = version

    def __repr__(self):
        return (f"UserProfile(name='{self.name}', city='{self.city}', "
                f"occupation='{self.occupation}', age={self.age}, "
                f"last_login={self.last_login})")

def user_profile_object_hook(dct):
    # This hook is called for every JSON object (dictionary) that is decoded.
    # We check if it contains keys that uniquely identify our UserProfile.
    if "name" in dct and "occupation" in dct and "last_login" in dct:
        # Convert ISO 8601 string to datetime object
        if dct["last_login"]: # Ensure it's not None
            try:
                dct["last_login"] = datetime.datetime.fromisoformat(dct["last_login"])
            except ValueError:
                # Handle cases where the format might not be correct
                pass
        return UserProfile(
            dct["name"],
            dct["city"],
            dct["occupation"],
            dct["age"],
            dct["is_active"],
            dct["skills"],
            dct["preferences"],
            dct["last_login"],
            dct["null_field"],
            dct["version"]
        )
    return dct # Return the dictionary as is if it's not a UserProfile

try:
    with open(file_name, 'r', encoding='utf-8') as f:
        loaded_user_profile = json.load(f, object_hook=user_profile_object_hook)
    print(f"Data loaded with object_hook from '{file_name}':")
    print(loaded_user_profile)
    print(f"Type of loaded_user_profile: {type(loaded_user_profile)}")
    if isinstance(loaded_user_profile, UserProfile):
        print(f"Accessing object attribute: User's occupation = {loaded_user_profile.occupation}")
        print(f"Type of last_login attribute: {type(loaded_user_profile.last_login)}\n")
except json.JSONDecodeError as e:
    print(f"Error decoding JSON with object_hook from file: {e}\n")
print("-" * 50 + "\n")


print("--- 3. json.load() with parse_int and parse_float ---")
# Use `parse_int` to convert integers, and `parse_float` for floats.
# Useful if you need to use a custom numeric type or enforce specific behavior.

def parse_large_int_as_string(s):
    # Example: always keep integers as strings
    return str(s)

def parse_float_to_decimal(s):
    # Example: convert floats to Python's Decimal type for precision
    from decimal import Decimal
    return Decimal(s)

# Create a JSON file with specific int/float values
file_name_numeric = "numeric_data.json"
numeric_data = {
    "large_id": 9876543210987654321,
    "exact_pi": 3.141592653589793,
    "item_count": 100
}
with open(file_name_numeric, 'w', encoding='utf-8') as f:
    json.dump(numeric_data, f, indent=4)

try:
    with open(file_name_numeric, 'r', encoding='utf-8') as f:
        loaded_numeric_data = json.load(f,
                                        parse_int=parse_large_int_as_string,
                                        parse_float=parse_float_to_decimal)
    print(f"Data loaded from '{file_name_numeric}' with custom parsers:")
    print(loaded_numeric_data)
    print(f"Type of 'large_id': {type(loaded_numeric_data['large_id'])}")
    print(f"Type of 'exact_pi': {type(loaded_numeric_data['exact_pi'])}")
    print(f"Type of 'item_count': {type(loaded_numeric_data['item_count'])}\n") # Still int if parse_int not affecting small ints
except Exception as e:
    print(f"Error loading numeric data: {e}\n")

# Clean up the temporary numeric file
if os.path.exists(file_name_numeric):
    os.remove(file_name_numeric)
print("-" * 50 + "\n")


print("--- 4. json.load() with object_pairs_hook (Preserving Key Order) ---")
# By default, Python dicts (from 3.7+) preserve insertion order.
# `object_pairs_hook` can be useful for older Python versions or specific use cases
# where you need the key-value pairs as an ordered list of tuples.

from collections import OrderedDict

# Create a JSON with a specific key order (dict itself will maintain order in modern Python)
file_name_order = "ordered_data.json"
ordered_data = {
    "first": 1,
    "second": 2,
    "third": {"nested_a": 10, "nested_b": 20},
    "fourth": 4
}
with open(file_name_order, 'w', encoding='utf-8') as f:
    json.dump(ordered_data, f, indent=4)

try:
    with open(file_name_order, 'r', encoding='utf-8') as f:
        # Using OrderedDict to explicitly show preservation of order (less crucial 3.7+)
        loaded_ordered_data = json.load(f, object_pairs_hook=OrderedDict)
    print(f"Data loaded from '{file_name_order}' with object_pairs_hook=OrderedDict:")
    print(loaded_ordered_data)
    print(f"Type of loaded_ordered_data: {type(loaded_ordered_data)}")
    print(f"Keys in order: {list(loaded_ordered_data.keys())}\n")
except Exception as e:
    print(f"Error loading ordered data: {e}\n")

# Clean up the temporary ordered file
if os.path.exists(file_name_order):
    os.remove(file_name_order)
print("-" * 50 + "\n")


print("--- 5. Handling JSONDecodeError (Malformed JSON) ---")
# Important to catch this error when working with external JSON files.

malformed_file_name = "malformed.json"
# Create a malformed JSON file (missing a closing brace)
with open(malformed_file_name, 'w', encoding='utf-8') as f:
    f.write('{"key": "value", "another_key": 123')

try:
    with open(malformed_file_name, 'r', encoding='utf-8') as f:
        json.load(f)
except json.JSONDecodeError as e:
    print(f"Caught JSONDecodeError (expected): {e}")
    print(f"Error message indicates the parsing issue and position.\n")
except FileNotFoundError:
    print(f"Error: Malformed file '{malformed_file_name}' not found.")

# Clean up the malformed file
if os.path.exists(malformed_file_name):
    os.remove(malformed_file_name)
print("-" * 50 + "\n")


# --- 6. Cleanup of the main sample file ---
print("--- 6. Final Cleanup ---")
if os.path.exists(file_name):
    os.remove(file_name)
    print(f"Cleaned up: Removed '{file_name}'")
else:
    print(f"No file '{file_name}' to clean up.")
print("-" * 50 + "\n")
````

````python
import json
import os
import datetime

# --- json.dump() Syntax and Usage ---

# json.dump(obj, fp, *, skipkeys=False, ensure_ascii=True, check_circular=True,
#           allow_nan=True, cls=None, indent=None, separators=None,
#           default=None, sort_keys=False)

# Parameters:
# 1. obj: The Python object to be serialized (converted to JSON).
#    This is the data you want to write to the file.
#    Common types: dict, list, str, int, float, bool, None.

# 2. fp: A file-like object opened in write mode (e.g., `f = open('file.json', 'w')`).
#    `json.dump()` writes the JSON representation of `obj` directly to this file stream.

# 3. skipkeys (bool): If True, dict keys that are not of type str, int, float,
#    or bool will be skipped. Otherwise (default False), a TypeError will be raised.

# 4. ensure_ascii (bool): If True (default), all non-ASCII characters in the output
#    are escaped (e.g., '€' becomes '\u20ac'). If False, these characters are output
#    directly. Setting to False can make the output more readable and smaller
#    if non-ASCII characters are common and the target system supports UTF-8.

# 5. check_circular (bool): If True (default), `dump` checks for circular
#    references in lists, dictionaries, and custom encoded objects. If a circular
#    reference is found, a `ValueError` is raised. If False, circular references
#    can lead to an infinite recursion when dumping.

# 6. allow_nan (bool): If True (default), `NaN`, `Infinity`, and `-Infinity`
#    (Not a Number, positive/negative infinity for floats) will be serialized
#    as `NaN`, `Infinity`, and `-Infinity` in JSON, which is technically
#    non-standard JSON but commonly accepted. If False, a `ValueError` is raised.

# 7. cls (class): A custom JSONEncoder subclass. If provided, instances of `cls`
#    will be used to serialize objects that are not natively serializable by `json`.
#    This is used for advanced custom object serialization.

# 8. indent (int or str): If `indent` is a non-negative integer or a string,
#    then JSON array elements and object members will be pretty-printed with
#    that indent level. An `indent` of 0, negative, or `''` (empty string) will
#    only insert newlines. If `indent` is a positive integer, that many spaces
#    per indentation level. If `indent` is a string (e.g., '\t'), that string
#    is used for indentation. Default is `None` (most compact output).

# 9. separators (tuple): A `(item_separator, key_separator)` tuple. Default is
#    `(', ', ': ')`. Used to specify the strings inserted between items and keys.
#    To get the most compact JSON, use `(',', ':')`.

# 10. default (function): A function that will be called for objects that `dump`
#     doesn't know how to serialize. It should return a JSON-serializable
#     representation of `obj`, or raise a `TypeError`. This is an alternative
#     to `cls` for simple custom type serialization.


# --- Practical Examples ---

# Example Python data to be written to JSON files
data = {
    "product_name": "Wireless Headphones",
    "model": "X-100",
    "price": 99.99,
    "available": True,
    "colors": ["Black", "Silver", "Red"],
    "specifications": {
        "weight_grams": 250,
        "battery_hours": 20,
        "bluetooth_version": 5.2
    },
    "notes": None,
    "country_of_origin": "China", # Example of non-ASCII characters if not escaped
    "release_date": "2025-06-24T16:40:11+06:00" # ISO 8601 string, example with timezone
}

# Define filenames for output
file_simple = "product_simple.json"
file_pretty = "product_pretty.json"
file_sorted = "product_sorted.json"
file_unicode = "product_unicode.json"
file_compact = "product_compact.json"
file_skipkeys = "product_skipkeys.json"
file_custom_default = "product_custom_default.json"
file_custom_cls = "product_custom_cls.json"
file_nan_issue = "product_nan_issue.json"


print("--- 1. Basic json.dump() (default behavior) ---")
# Writes the most compact JSON output to a file.
try:
    with open(file_simple, 'w', encoding='utf-8') as f:
        json.dump(data, f)
    print(f"Data dumped to '{file_simple}' (default compact format).")
    # Content of product_simple.json will be a single line:
    # {"product_name": "Wireless Headphones", "model": "X-100", ...}
except IOError as e:
    print(f"Error writing to {file_simple}: {e}")
print("\n")


print("--- 2. json.dump() with indent (pretty print to file) ---")
# Writes human-readable JSON with indentation.
try:
    with open(file_pretty, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=4) # Use 4 spaces for indentation
    print(f"Data dumped to '{file_pretty}' (pretty printed with indent=4).")
    # Content of product_pretty.json will be nicely formatted:
    # {
    #     "product_name": "Wireless Headphones",
    #     ...
    # }
except IOError as e:
    print(f"Error writing to {file_pretty}: {e}")
print("\n")


print("--- 3. json.dump() with sort_keys=True (alphabetical order of keys) ---")
# Ensures the keys in JSON objects are sorted alphabetically.
try:
    with open(file_sorted, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=4, sort_keys=True)
    print(f"Data dumped to '{file_sorted}' (keys sorted alphabetically).")
    # Keys like "age", "colors", "country_of_origin", etc., will be sorted.
except IOError as e:
    print(f"Error writing to {file_sorted}: {e}")
print("\n")


print("--- 4. json.dump() with ensure_ascii=False (for non-ASCII characters) ---")
# Prevents escaping of non-ASCII characters, making the file potentially smaller
# and more readable for non-English speakers, assuming UTF-8 encoding is used.
try:
    with open(file_unicode, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=4, ensure_ascii=False)
    print(f"Data dumped to '{file_unicode}' (non-ASCII characters unescaped).")
    # '€' will appear as '€' instead of '\u20ac' in the file.
    # "country_of_origin": "China" will remain as is.
except IOError as e:
    print(f"Error writing to {file_unicode}: {e}")
print("\n")


print("--- 5. json.dump() with separators (for extremely compact output) ---")
# Overrides default separators (", ", ": ") to (",", ":") to remove all unnecessary whitespace.
try:
    with open(file_compact, 'w', encoding='utf-8') as f:
        json.dump(data, f, separators=(',', ':'))
    print(f"Data dumped to '{file_compact}' (most compact format).")
    # Output will look like: {"product_name":"Wireless Headphones","model":"X-100",...}
except IOError as e:
    print(f"Error writing to {file_compact}: {e}")
print("\n")


print("--- 6. json.dump() with skipkeys=True (handling non-string/non-numeric keys) ---")
# If a dictionary has keys that are not strings, numbers, or booleans, `json.dump()`
# will raise a TypeError by default. `skipkeys=True` prevents this by ignoring such entries.
data_with_bad_key = {
    "item_id": 1001,
    (1, 2): "This key is a tuple, which is not valid in JSON", # Invalid JSON key
    "status": "active"
}

try:
    # This will cause a TypeError because (1, 2) is not a valid JSON key type.
    # The default behavior is check_circular=True, skipkeys=False.
    with open("error_file.json", 'w', encoding='utf-8') as f:
        json.dump(data_with_bad_key, f, indent=2)
    print("This line should not be printed without skipkeys=True.")
except TypeError as e:
    print(f"Caught TypeError (expected without skipkeys=True): {e}")
except IOError as e:
    print(f"Error writing to error_file.json: {e}")
finally:
    if os.path.exists("error_file.json"):
        os.remove("error_file.json") # Clean up potentially created partial file

# Now, dump with skipkeys=True
try:
    with open(file_skipkeys, 'w', encoding='utf-8') as f:
        json.dump(data_with_bad_key, f, indent=2, skipkeys=True)
    print(f"Data dumped to '{file_skipkeys}' (invalid key was skipped).")
    # Content will only have "item_id" and "status"
except IOError as e:
    print(f"Error writing to {file_skipkeys}: {e}")
print("\n")


print("--- 7. json.dump() with default callable (custom object serialization) ---")
# Use the `default` parameter to provide a function that serializes objects
# that `json.dump()` doesn't natively know how to handle.

class SensorReading:
    def __init__(self, sensor_id, value, unit, timestamp):
        self.sensor_id = sensor_id
        self.value = value
        self.unit = unit
        self.timestamp = timestamp # This is a datetime object

    def __repr__(self):
        return f"SensorReading(id={self.sensor_id}, value={self.value}, unit='{self.unit}')"

def serialize_sensor_reading(obj):
    # This function is called if `json.dump` encounters an object it cannot serialize.
    if isinstance(obj, SensorReading):
        return {
            "sensor_id": obj.sensor_id,
            "value": obj.value,
            "unit": obj.unit,
            "timestamp": obj.timestamp.isoformat() # Convert datetime to ISO 8601 string
        }
    # For any other type `json.dump` doesn't know, let it raise the default TypeError.
    raise TypeError(f"Object of type {obj.__class__.__name__} is not JSON serializable")

sensor_data = {
    "device_name": "WeatherStation-01",
    "readings": [
        SensorReading(1, 22.5, "Celsius", datetime.datetime.now()),
        SensorReading(2, 75.2, "Fahrenheit", datetime.datetime.now() - datetime.timedelta(hours=1))
    ]
}

# Attempt to dump without `default` (will fail)
try:
    with open("error_sensor.json", 'w') as f:
        json.dump(sensor_data, f)
except TypeError as e:
    print(f"Caught TypeError (expected without default function): {e}")
finally:
    if os.path.exists("error_sensor.json"):
        os.remove("error_sensor.json")

# Dump with the custom `default` function
try:
    with open(file_custom_default, 'w', encoding='utf-8') as f:
        json.dump(sensor_data, f, indent=4, default=serialize_sensor_reading)
    print(f"Data dumped to '{file_custom_default}' using 'default' function for custom objects.")
except IOError as e:
    print(f"Error writing to {file_custom_default}: {e}")
print("\n")


print("--- 8. json.dump() with cls (custom JSONEncoder subclass) ---")
# `cls` provides a more robust and reusable way to handle complex custom object serialization.
# It defines a custom encoder class that inherits from `json.JSONEncoder`.

class CustomDateTimeEncoder(json.JSONEncoder):
    def default(self, obj):
        # If the object is a datetime.datetime or datetime.date, convert it to ISO format string
        if isinstance(obj, (datetime.datetime, datetime.date)):
            return obj.isoformat()
        # If it's a SensorReading object, use our defined serialization
        elif isinstance(obj, SensorReading):
            return {
                "sensor_id": obj.sensor_id,
                "value": obj.value,
                "unit": obj.unit,
                "timestamp": obj.timestamp.isoformat()
            }
        # For other types, use the base class's default method (which might raise TypeError)
        return super().default(obj)

mixed_data = {
    "report_id": "REP-2025-06-24",
    "created_at": datetime.datetime.now(),
    "last_updated": datetime.date.today(),
    "sensor_status": SensorReading(3, "Online", "Status", datetime.datetime.now()),
    "measurements": [10.5, 20, 30.1]
}

try:
    with open(file_custom_cls, 'w', encoding='utf-8') as f:
        json.dump(mixed_data, f, indent=4, cls=CustomDateTimeEncoder)
    print(f"Data dumped to '{file_custom_cls}' using custom 'cls' (JSONEncoder subclass).")
except IOError as e:
    print(f"Error writing to {file_custom_cls}: {e}")
print("\n")


print("--- 9. json.dump() with allow_nan=False (handling NaN/Infinity) ---")
# `allow_nan=False` forces `json.dump()` to raise a ValueError if it encounters
# `NaN` (Not a Number), `Infinity`, or `-Infinity` float values. By default, it allows them.
data_with_special_floats = {
    "invalid_reading": float('nan'),
    "max_temperature": float('inf'),
    "min_temperature": float('-inf')
}

# Dump with default allow_nan=True (will succeed)
try:
    with open("nan_allowed.json", 'w', encoding='utf-8') as f:
        json.dump(data_with_special_floats, f, indent=2)
    print(f"Data with NaN/Inf dumped to 'nan_allowed.json' (allow_nan=True).")
except IOError as e:
    print(f"Error writing to nan_allowed.json: {e}")
finally:
    if os.path.exists("nan_allowed.json"):
        os.remove("nan_allowed.json")

# Dump with allow_nan=False (will raise ValueError)
try:
    with open(file_nan_issue, 'w', encoding='utf-8') as f:
        json.dump(data_with_special_floats, f, indent=2, allow_nan=False)
    print("This line should not be printed with allow_nan=False.")
except ValueError as e:
    print(f"Caught ValueError (expected with allow_nan=False): {e}")
except IOError as e:
    print(f"Error writing to {file_nan_issue}: {e}")
finally:
    if os.path.exists(file_nan_issue):
        os.remove(file_nan_issue) # Clean up potentially created partial file
print("\n")


# --- Cleanup: Remove all generated JSON files ---
print("--- Cleanup: Removing generated JSON files ---")
generated_files = [
    file_simple,
    file_pretty,
    file_sorted,
    file_unicode,
    file_compact,
    file_skipkeys,
    file_custom_default,
    file_custom_cls
]

for f_name in generated_files:
    if os.path.exists(f_name):
        os.remove(f_name)
        print(f"Removed: {f_name}")
    else:
        print(f"Skipped (not found): {f_name}")

print("\n--- All operations completed. ---")
````

````python
import json
import datetime

# --- 1. Define a Custom Python Class ---
# This class represents data that is not directly supported by JSON.
class Book:
    def __init__(self, title, author, publication_date, isbn=None):
        self.title = title
        self.author = author
        # publication_date will be a datetime.date object
        self.publication_date = publication_date
        self.isbn = isbn

    def __repr__(self):
        # A string representation for easy debugging/printing
        return (f"Book(title='{self.title}', author='{self.author}', "
                f"publication_date={self.publication_date.isoformat() if self.publication_date else None}, "
                f"isbn='{self.isbn}')")

    # Optional: A method to convert to a dictionary (useful for manual serialization or default function)
    def to_dict(self):
        return {
            "title": self.title,
            "author": self.author,
            "publication_date": self.publication_date.isoformat() if self.publication_date else None,
            "isbn": self.isbn,
            "__book__": True # A special marker for deserialization
        }


# --- 2. Custom Serialization (Python to JSON) ---
# Goal: Convert a 'Book' object (which contains a datetime.date) into a JSON string.

print("--- Custom Serialization ---")

# Create an instance of our custom object
my_book = Book(
    title="The Hitchhiker's Guide to the Galaxy",
    author="Douglas Adams",
    publication_date=datetime.date(1979, 10, 12),
    isbn="978-0345391803"
)
print(f"Original Python Book Object: {my_book}")
print(f"Type of publication_date: {type(my_book.publication_date)}\n")

# 2.1 Method 1: Using the `default` parameter in json.dumps()
# Define a function that tells json.dumps how to handle unsupported types.
def custom_serializer_default(obj):
    if isinstance(obj, Book):
        # Convert the Book object to a dictionary (JSON object)
        return obj.to_dict()
    elif isinstance(obj, (datetime.datetime, datetime.date)):
        # Convert datetime objects to ISO 8601 strings
        return obj.isoformat()
    # If the object type is not handled, raise a TypeError
    raise TypeError(f"Object of type {obj.__class__.__name__} is not JSON serializable")

try:
    json_string_default = json.dumps(my_book, indent=4, default=custom_serializer_default)
    print("Serialized JSON (using 'default' parameter):")
    print(json_string_default)
    print(f"Type of output: {type(json_string_default)}\n")
except TypeError as e:
    print(f"Serialization Error with 'default' parameter: {e}\n")


# 2.2 Method 2: Using a Custom JSONEncoder subclass (more robust for multiple custom types)
# Define a class that inherits from json.JSONEncoder and overrides the 'default' method.
class CustomBookEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, Book):
            return obj.to_dict()
        elif isinstance(obj, (datetime.datetime, datetime.date)):
            return obj.isoformat()
        # Let the base class handle other types or raise TypeError
        return super().default(obj)

try:
    json_string_cls = json.dumps(my_book, indent=4, cls=CustomBookEncoder)
    print("Serialized JSON (using 'cls' parameter with CustomBookEncoder):")
    print(json_string_cls)
    print(f"Type of output: {type(json_string_cls)}\n")
except TypeError as e:
    print(f"Serialization Error with 'cls' parameter: {e}\n")

print("-" * 50 + "\n")


# --- 3. Custom Deserialization (JSON to Python) ---
# Goal: Convert a JSON string back into a 'Book' object.

print("--- Custom Deserialization ---")

# Let's use the JSON string generated from our custom object
json_to_deserialize = json_string_cls # Or json_string_default, they are identical

print("JSON String to deserialize:")
print(json_to_deserialize)

# 3.1 Method 1: Using the `object_hook` parameter in json.loads()
# Define a function that tells json.loads how to convert specific dictionaries back to objects.
def custom_deserializer_object_hook(dct):
    # Check for the custom marker added during serialization
    if "__book__" in dct:
        # Reconstruct the datetime.date object from its ISO string
        pub_date_str = dct.get("publication_date")
        publication_date_obj = None
        if pub_date_str:
            try:
                publication_date_obj = datetime.date.fromisoformat(pub_date_str)
            except ValueError:
                print(f"Warning: Could not parse date string '{pub_date_str}' during deserialization.")

        # Reconstruct the Book object
        return Book(
            title=dct["title"],
            author=dct["author"],
            publication_date=publication_date_obj,
            isbn=dct.get("isbn")
        )
    return dct # Return the dictionary as is if it's not our custom object

try:
    deserialized_object = json.loads(json_to_deserialize, object_hook=custom_deserializer_object_hook)
    print("\nDeserialized Python Object (using 'object_hook'):")
    print(deserialized_object)
    print(f"Type of deserialized_object: {type(deserialized_object)}")

    if isinstance(deserialized_object, Book):
        print(f"  Accessing attributes: Title='{deserialized_object.title}', Author='{deserialized_object.author}'")
        print(f"  Type of deserialized publication_date: {type(deserialized_object.publication_date)}")
    else:
        print("  Deserialization failed: Expected a Book object, got a dict.")
except json.JSONDecodeError as e:
    print(f"Deserialization Error with 'object_hook': {e}\n")


# 3.2 Method 2: Using a Custom JSONDecoder subclass (less common for simple cases, but powerful)
# Define a class that inherits from json.JSONDecoder and overrides its `decode` or `parse_object` methods.
# For simple object reconstruction from dictionaries, `object_hook` is usually sufficient and simpler.
# This is more for deeply customizing the parsing process itself.
class CustomBookDecoder(json.JSONDecoder):
    def __init__(self, *args, **kwargs):
        # Pass our custom object_hook to the base class
        super().__init__(object_hook=custom_deserializer_object_hook, *args, **kwargs)

try:
    deserialized_object_cls = json.loads(json_to_deserialize, cls=CustomBookDecoder)
    print("\nDeserialized Python Object (using 'cls' parameter with CustomBookDecoder):")
    print(deserialized_object_cls)
    print(f"Type of deserialized_object_cls: {type(deserialized_object_cls)}")
    if isinstance(deserialized_object_cls, Book):
        print(f"  Successfully reconstructed Book via custom decoder class.")
except json.JSONDecodeError as e:
    print(f"Deserialization Error with 'cls' parameter: {e}\n")

print("-" * 50 + "\n")

# --- Summary ---
# Custom Serialization:
# - `json.dumps(obj, default=your_function)`: Provide a function for non-serializable types.
# - `json.dumps(obj, cls=YourJSONEncoderSubclass)`: Provide a custom encoder class.
#   `YourJSONEncoderSubclass` must override `default` method.

# Custom Deserialization:
# - `json.loads(s, object_hook=your_function)`: Provide a function to transform decoded dictionaries.
# - `json.loads(s, cls=YourJSONDecoderSubclass)`: Less common for simple cases, more for deep parsing control.
#   `YourJSONDecoderSubclass` typically uses `object_hook` internally or custom parsing logic.
````
