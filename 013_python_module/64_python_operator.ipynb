The `operator` module in Python provides a set of functions that correspond to the intrinsic operators of Python. In simpler terms, it's a library that lets you use Python's built-in operations (like addition, comparison, item access, etc.) as regular functions.

---

## Python's `operator` Module: All About

### 1. Introduction: What and Why?

**What is the `operator` module?**
It's a standard Python library that exports functions corresponding to the arithmetic, comparison, logical, bitwise, and sequence operations. For example, `operator.add(a, b)` is equivalent to `a + b`.

**Why use it instead of direct operators?**

1.  **Functional Programming:** Many higher-order functions (like `map()`, `filter()`, `functools.reduce()`, `list.sort()`, `sorted()`) expect a callable as an argument. The `operator` functions provide these callables directly, often replacing simple `lambda` functions.
2.  **Readability:** In some complex functional constructs, `operator.add` can be clearer than an anonymous `lambda x, y: x + y`.
3.  **Efficiency:** For very tight loops or performance-critical code, `operator` functions can sometimes be marginally faster than equivalent `lambda` expressions because they avoid the overhead of creating a new `lambda` object on each call.
4.  **Specialized Tools:** It includes powerful functions like `itemgetter()` and `attrgetter()` which are indispensable for sorting and grouping collections of objects.

---

### 2. Core Concepts and Categories of Functions

The functions in the `operator` module are broadly categorized based on the type of operation they perform:

- **Arithmetic Operations:** `add`, `sub`, `mul`, `truediv`, `floordiv`, `mod`, `pow`, `neg`, `pos`, etc.
- **Comparison Operations:** `eq`, `ne`, `lt`, `le`, `gt`, `ge`.
- **Logical Operations:** `and_`, `or_`, `not_`, `xor`.
- **Bitwise Operations:** `and_`, `or_`, `xor`, `invert`, `lshift`, `rshift`. (Note: `and_` and `or_` are overloaded for both logical and bitwise operations; their behavior depends on the operand types).
- **Sequence Operations:** `concat`, `contains`, `countOf`, `indexOf`, `getitem`, `setitem`, `delitem`.
- **In-place Operations:** `iadd`, `isub`, `imul`, etc. (correspond to `+=`, `-=`, etc.).
- **Attribute and Item Lookup:** `attrgetter`, `itemgetter`.
- **Method Calling:** `methodcaller`.

### 3. Code Examples

```python
import operator
import functools # For reduce, which often uses operator functions

print("--- 3. Code Examples for `operator` Module ---")

# --- 3.1. Arithmetic Operations ---
print("\n--- 3.1. Arithmetic Operations ---")

a, b = 10, 3

# Addition: operator.add(a, b) equivalent to a + b
print(f"operator.add({a}, {b}): {operator.add(a, b)}")

# Subtraction: operator.sub(a, b) equivalent to a - b
print(f"operator.sub({a}, {b}): {operator.sub(a, b)}")

# Multiplication: operator.mul(a, b) equivalent to a * b
print(f"operator.mul({a}, {b}): {operator.mul(a, b)}")

# True Division: operator.truediv(a, b) equivalent to a / b
print(f"operator.truediv({a}, {b}): {operator.truediv(a, b)}")

# Floor Division: operator.floordiv(a, b) equivalent to a // b
print(f"operator.floordiv({a}, {b}): {operator.floordiv(a, b)}")

# Modulo: operator.mod(a, b) equivalent to a % b
print(f"operator.mod({a}, {b}): {operator.mod(a, b)}")

# Power: operator.pow(a, b) equivalent to a ** b
print(f"operator.pow({a}, {b}): {operator.pow(a, b)}")

# Negation: operator.neg(a) equivalent to -a
print(f"operator.neg({a}): {operator.neg(a)}")

# Positive: operator.pos(a) equivalent to +a
print(f"operator.pos({a}): {operator.pos(a)}")

# Example with reduce: Summing a list
numbers = [1, 2, 3, 4, 5]
total_sum = functools.reduce(operator.add, numbers)
print(f"Sum of {numbers} using reduce and operator.add: {total_sum}")


# --- 3.2. Comparison Operations ---
print("\n--- 3.2. Comparison Operations ---")

x, y = 5, 8

# Equal: operator.eq(a, b) equivalent to a == b
print(f"operator.eq({x}, {y}): {operator.eq(x, y)}")

# Not Equal: operator.ne(a, b) equivalent to a != b
print(f"operator.ne({x}, {y}): {operator.ne(x, y)}")

# Less Than: operator.lt(a, b) equivalent to a < b
print(f"operator.lt({x}, {y}): {operator.lt(x, y)}")

# Less Than or Equal To: operator.le(a, b) equivalent to a <= b
print(f"operator.le({x}, {y}): {operator.le(x, y)}")

# Greater Than: operator.gt(a, b) equivalent to a > b
print(f"operator.gt({x}, {y}): {operator.gt(x, y)}")

# Greater Than or Equal To: operator.ge(a, b) equivalent to a >= b
print(f"operator.ge({x}, {y}): {operator.ge(x, y)}")

# Example with filter: Filtering numbers greater than 5
values = [1, 6, 3, 8, 2, 7]
greater_than_5 = list(filter(functools.partial(operator.gt, 5), values)) # (num > 5 is False for num=5)
# Note: functools.partial is used to fix one argument of a two-argument function
print(f"Values > 5 in {values}: {greater_than_5}")


# --- 3.3. Logical Operations ---
print("\n--- 3.3. Logical Operations ---")

p, q = True, False

# Logical NOT: operator.not_(a) equivalent to not a
print(f"operator.not_({p}): {operator.not_(p)}")

# Logical AND: operator.and_(a, b) equivalent to a and b
# Note: operator.and_ evaluates both operands, unlike the 'and' keyword's short-circuiting.
print(f"operator.and_({p}, {q}): {operator.and_(p, q)}")

# Logical OR: operator.or_(a, b) equivalent to a or b
# Note: operator.or_ evaluates both operands, unlike the 'or' keyword's short-circuiting.
print(f"operator.or_({p}, {q}): {operator.or_(p, q)}")

# Logical XOR: operator.xor(a, b)
print(f"operator.xor({p}, {q}): {operator.xor(p, q)}")


# --- 3.4. Bitwise Operations ---
print("\n--- 3.4. Bitwise Operations ---")

num1, num2 = 0b1100, 0b1010 # 12, 10 in decimal

# Bitwise AND: operator.and_(a, b) equivalent to a & b
# (Same function as logical AND, context/operand types determine behavior)
print(f"operator.and_({num1}, {num2}) (bitwise): {operator.and_(num1, num2):04b}") # Output: 1000 (8)

# Bitwise OR: operator.or_(a, b) equivalent to a | b
# (Same function as logical OR, context/operand types determine behavior)
print(f"operator.or_({num1}, {num2}) (bitwise): {operator.or_(num1, num2):04b}") # Output: 1110 (14)

# Bitwise XOR: operator.xor(a, b) equivalent to a ^ b
print(f"operator.xor({num1}, {num2}): {operator.xor(num1, num2):04b}") # Output: 0110 (6)

# Bitwise NOT (Invert): operator.invert(a) equivalent to ~a
print(f"operator.invert({num1}): {operator.invert(num1)}") # Output: -13 (two's complement)

# Left Shift: operator.lshift(a, b) equivalent to a << b
print(f"operator.lshift({num1}, 2): {operator.lshift(num1, 2):06b}") # 12 << 2 = 48 (110000)

# Right Shift: operator.rshift(a, b) equivalent to a >> b
print(f"operator.rshift({num1}, 2): {operator.rshift(num1, 2):02b}") # 12 >> 2 = 3 (0011)


# --- 3.5. Sequence Operations ---
print("\n--- 3.5. Sequence Operations ---")

my_list = [1, 2, 3, 4, 5]
my_string = "hello"

# Concatenation: operator.concat(a, b) equivalent to a + b
print(f"operator.concat({my_list}, [6, 7]): {operator.concat(my_list, [6, 7])}")
print(f"operator.concat({my_string}, ' world'): {operator.concat(my_string, ' world')}")

# Contains: operator.contains(a, b) equivalent to b in a
print(f"operator.contains({my_list}, 3): {operator.contains(my_list, 3)}")
print(f"operator.contains({my_string}, 'ell'): {operator.contains(my_string, 'ell')}")

# Get Item: operator.getitem(a, b) equivalent to a[b]
print(f"operator.getitem({my_list}, 2): {operator.getitem(my_list, 2)}")
print(f"operator.getitem({my_string}, 1): {operator.getitem(my_string, 1)}")

# Set Item: operator.setitem(a, b, c) equivalent to a[b] = c
# (Modifies the list in-place)
set_list = ['a', 'b', 'c']
operator.setitem(set_list, 1, 'B')
print(f"operator.setitem({['a', 'b', 'c']}, 1, 'B'): {set_list}")

# Delete Item: operator.delitem(a, b) equivalent to del a[b]
# (Modifies the list in-place)
del_list = ['x', 'y', 'z']
operator.delitem(del_list, 0)
print(f"operator.delitem({['x', 'y', 'z']}, 0): {del_list}")


# --- 3.6. Specialized Functions (Higher-Order Functions) ---
print("\n--- 3.6. Specialized Functions ---")

# operator.itemgetter(*items): Creates a callable that fetches items from an iterable.
# Very useful as a 'key' argument for sorting.
data = [
    ('Alice', 30, 'NY'),
    ('Bob', 25, 'LA'),
    ('Charlie', 35, 'NY')
]

# Sort by age (index 1)
sorted_by_age = sorted(data, key=operator.itemgetter(1))
print(f"Sorted by age: {sorted_by_age}")

# Sort by city (index 2), then by name (index 0)
sorted_by_city_then_name = sorted(data, key=operator.itemgetter(2, 0))
print(f"Sorted by city, then name: {sorted_by_city_then_name}")


# operator.attrgetter(*attrs): Creates a callable that fetches attributes from an object.
# Very useful as a 'key' argument for sorting custom objects.
class Person:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city
    def __repr__(self):
        return f"Person('{self.name}', {self.age}, '{self.city}')"

people = [
    Person('Alice', 30, 'NY'),
    Person('Bob', 25, 'LA'),
    Person('Charlie', 35, 'NY')
]

# Sort by age
sorted_people_by_age = sorted(people, key=operator.attrgetter('age'))
print(f"\nSorted people by age: {sorted_people_by_age}")

# Sort by city, then by name
sorted_people_by_city_name = sorted(people, key=operator.attrgetter('city', 'name'))
print(f"Sorted people by city, then name: {sorted_people_by_city_name}")


# operator.methodcaller(name, *args, **kwargs): Creates a callable that calls a named method.
s = "Hello World"
upper_func = operator.methodcaller('upper')
print(f"\nCalling 'upper' method: {upper_func(s)}") # 'HELLO WORLD'

split_func = operator.methodcaller('split', 'o') # Split by 'o'
print(f"Calling 'split' method: {split_func(s)}") # ['Hell', ' W', 'rld']

```

### 4. Why Use the `operator` Module? (Detailed Benefits)

- **Readability in Functional Contexts:**
  When passing a simple operation as a `key` to `sorted()` or a function to `reduce()`, `operator.add` or `operator.itemgetter(0)` is often more concise and immediately understandable than `lambda x, y: x + y` or `lambda item: item[0]`.

  ```python
  from functools import reduce
  import operator

  data = [(10, 'A'), (20, 'B'), (5, 'C')]

  # Preferred:
  min_value = min(data, key=operator.itemgetter(0))
  print(f"Min value by first element: {min_value}")

  # Less preferred (for simple cases):
  # min_value_lambda = min(data, key=lambda item: item[0])
  ```

- **Replacing Simple Lambdas:**
  It reduces the need for trivial `lambda` functions, making the code cleaner and potentially a tiny bit more performant as `operator` functions are implemented in C and avoid the overhead of `lambda` creation.

- **Consistency:**
  Provides a uniform interface for all Python operators, which can be useful when building libraries or generic algorithms that need to perform operations dynamically.

- **Error Handling:**
  `itemgetter` and `attrgetter` provide better error messages than a generic lambda if the index or attribute doesn't exist.

---

### 5. When NOT to Use `operator`

- **Simple, Direct Operations:** For a simple `x + y` where no functional context is required, the direct operator is always more readable and idiomatic.
- **Short-Circuiting Logic:** The `and_` and `or_` functions from `operator` do _not_ short-circuit. They evaluate both their operands. If you rely on the short-circuiting behavior of Python's `and` and `or` keywords (e.g., `if x and x.method():`), you should use the keywords, not the `operator` module functions.

  ```python
  import operator

  def risky_func():
      print("Risky function called!")
      raise ValueError("Error!")

  # Using 'and' keyword (short-circuits):
  result_keyword = True and risky_func() # risky_func() is called
  # Output: Risky function called! (then raises ValueError)

  # Using operator.and_ (DOES NOT short-circuit):
  # result_operator = operator.and_(True, risky_func())
  # This would try to evaluate risky_func() regardless of the first operand's truthiness.
  # Output: Risky function called! (then raises ValueError)
  ```

```python
import operator
import functools # Used for reduce, which commonly uses operator functions

print("--- 4. Code Examples for Arithmetic Operations in `operator` Module ---")

# --- 4.1. Core Arithmetic Functions ---
print("\n--- 4.1. Core Arithmetic Functions (Return New Value) ---")

val1 = 15
val2 = 4
val3 = -7

# Addition: operator.add(a, b) equivalent to a + b
result_add = operator.add(val1, val2)
print(f"operator.add({val1}, {val2}): {result_add} (Equivalent to {val1} + {val2})")

# Subtraction: operator.sub(a, b) equivalent to a - b
result_sub = operator.sub(val1, val2)
print(f"operator.sub({val1}, {val2}): {result_sub} (Equivalent to {val1} - {val2})")

# Multiplication: operator.mul(a, b) equivalent to a * b
result_mul = operator.mul(val1, val2)
print(f"operator.mul({val1}, {val2}): {result_mul} (Equivalent to {val1} * {val2})")

# True Division: operator.truediv(a, b) equivalent to a / b
result_truediv = operator.truediv(val1, val2)
print(f"operator.truediv({val1}, {val2}): {result_truediv} (Equivalent to {val1} / {val2})")

# Floor Division: operator.floordiv(a, b) equivalent to a // b
result_floordiv = operator.floordiv(val1, val2)
print(f"operator.floordiv({val1}, {val2}): {result_floordiv} (Equivalent to {val1} // {val2})")

# Modulo: operator.mod(a, b) equivalent to a % b
result_mod = operator.mod(val1, val2)
print(f"operator.mod({val1}, {val2}): {result_mod} (Equivalent to {val1} % {val2})")

# Power: operator.pow(a, b) equivalent to a ** b
result_pow = operator.pow(val1, 2) # 15 squared
print(f"operator.pow({val1}, 2): {result_pow} (Equivalent to {val1} ** 2)")

# Unary Negation: operator.neg(a) equivalent to -a
result_neg = operator.neg(val1)
print(f"operator.neg({val1}): {result_neg} (Equivalent to -{val1})")
result_neg_val3 = operator.neg(val3)
print(f"operator.neg({val3}): {result_neg_val3} (Equivalent to -{val3})")

# Unary Positive: operator.pos(a) equivalent to +a
result_pos = operator.pos(val3)
print(f"operator.pos({val3}): {result_pos} (Equivalent to +{val3})")

# Absolute Value: operator.abs(a) equivalent to abs(a)
result_abs = operator.abs(val3)
print(f"operator.abs({val3}): {result_abs} (Equivalent to abs({val3}))")

# --- Example of use with `functools.reduce` ---
numbers = [1, 2, 3, 4, 5]
sum_of_numbers = functools.reduce(operator.add, numbers)
print(f"\nSum of {numbers} using `reduce(operator.add, ...)`: {sum_of_numbers}")

product_of_numbers = functools.reduce(operator.mul, numbers)
print(f"Product of {numbers} using `reduce(operator.mul, ...)`: {product_of_numbers}")


# --- 4.2. In-Place Arithmetic Functions ---
print("\n--- 4.2. In-Place Arithmetic Functions (Modify First Operand) ---")

# For mutable types (like lists), these operations modify the object directly.
# For immutable types (like integers, floats, strings), they behave like their
# non-in-place counterparts by returning a new object, because the original cannot be changed.

my_list = [10, 20, 30]
print(f"Initial list: {my_list}")
initial_list_id = id(my_list)
print(f"Initial list ID: {initial_list_id}")

# In-place Addition: operator.iadd(a, b) equivalent to a += b
operator.iadd(my_list, [40, 50]) # Adds elements to the list
print(f"After operator.iadd(my_list, [40, 50]): {my_list}")
print(f"List ID after iadd: {id(my_list)} (Same ID if mutable)") # ID should be the same

# In-place Multiplication: operator.imul(a, b) equivalent to a *= b
# For lists, this repeats the list in-place.
my_list_repeat = [1, 2]
print(f"\nInitial list for repeat: {my_list_repeat}")
initial_repeat_id = id(my_list_repeat)
print(f"Initial repeat list ID: {initial_repeat_id}")
operator.imul(my_list_repeat, 3) # Repeats the list 3 times
print(f"After operator.imul(my_list_repeat, 3): {my_list_repeat}")
print(f"Repeat list ID after imul: {id(my_list_repeat)} (Same ID if mutable)") # ID should be the same


# For immutable types, the ID will change (as a new object is created and assigned)
num_val = 10
print(f"\nInitial number: {num_val}")
initial_num_id = id(num_val)
print(f"Initial number ID: {initial_num_id}")
# operator.iadd will return a new int object because int is immutable
num_val = operator.iadd(num_val, 5) # Reassigning the variable with the new object
print(f"After num_val = operator.iadd(num_val, 5): {num_val}")
print(f"Number ID after iadd: {id(num_val)} (Different ID if immutable)") # ID will be different

```

## Logical Operators in `operator` Module: Theory and Code

### 1. Theory of Logical Operators

In Boolean logic, logical operators combine or modify boolean values (`True` or `False`) to produce a new boolean value.

- **AND:** Returns `True` if _both_ operands are `True`. Otherwise, returns `False`.
- **OR:** Returns `True` if _at least one_ operand is `True`. Otherwise, returns `False`.
- **NOT:** Returns the opposite boolean value of the operand (`True` becomes `False`, `False` becomes `True`). This is a unary operator.
- **XOR (Exclusive OR):** Returns `True` if _exactly one_ of the operands is `True`. Returns `False` if both are `True` or both are `False`.

#### Key Distinction: `operator` functions vs. Python Keywords

This is the most critical theoretical point when discussing logical operators from the `operator` module:

- **Python Keywords (`and`, `or`, `not`):**

  - **Short-circuiting:** They evaluate operands from left to right and stop as soon as the result can be determined.
    - `A and B`: If `A` is `False`, `B` is not evaluated.
    - `A or B`: If `A` is `True`, `B` is not evaluated.
  - **Return values:** They return the _value of the last evaluated operand_, not necessarily `True` or `False`. This allows them to work with "truthy" and "falsy" values (e.g., `[] and "hello"` results in `[]`).

- **`operator` Module Functions (`operator.and_`, `operator.or_`, `operator.not_`, `operator.xor`):**
  - **No Short-circuiting:** They _always evaluate both operands_ (for binary operators like `and_`, `or_`, `xor`).
  - **Return values:** They _always return `True` or `False`_. They perform a strict boolean operation.
  - **Name Overload:** `operator.and_` and `operator.or_` serve dual purposes: logical AND/OR when operands are boolean, and bitwise AND/OR when operands are integers. Python determines which operation to perform based on the operand types (via dunder methods like `__and__` and `__or__`).

Therefore, you must choose between the keyword and the `operator` function based on whether short-circuiting and specific return values are important to your logic.

---

### 2. Logical Operator Functions in `operator`

- `operator.and_(a, b)`: Implements `a & b` (bitwise AND) or logical AND if operands are booleans.
- `operator.or_(a, b)`: Implements `a | b` (bitwise OR) or logical OR if operands are booleans.
- `operator.not_(a)`: Implements `not a` (logical NOT).
- `operator.xor(a, b)`: Implements `a ^ b` (bitwise XOR) or logical XOR if operands are booleans.

## Python's `operator` Module: Bitwise Operations

The `operator` module in Python provides functional equivalents for Python's built-in operators. This is particularly useful when you need to pass an operation as a callable to higher-order functions (like `functools.reduce`, `map`, or `filter`). This section focuses on the **bitwise operators** available through this module.

### 1. Introduction to Bitwise Operations in `operator`

Bitwise operations manipulate individual bits of integer numbers. They are fundamental in low-level programming, digital logic, encryption, and optimizing certain algorithms.

Using the `operator` module for bitwise operations means:

- **Functional Style:** You can use `operator.and_(a, b)` instead of `a & b`.
- **Higher-Order Functions:** These functions can be directly passed to functions that expect callables, reducing the need for `lambda` functions (e.g., `functools.reduce(operator.and_, list_of_masks)`).
- **Clarity:** In some contexts, `operator.lshift` might be more explicit about the intent than `<<`.

### 2. Theory of Bitwise Operations

Before diving into the `operator` functions, let's briefly review what each bitwise operation does:

- **AND (`&`):** Compares two bits at the corresponding position. If both bits are 1, the result is 1; otherwise, it's 0.
  ```
  1010 (10)
  & 1100 (12)
  ----
  1000 (8)
  ```
- **OR (`|`):** Compares two bits. If at least one bit is 1, the result is 1; otherwise, it's 0.
  ```
  1010 (10)
  | 1100 (12)
  ----
  1110 (14)
  ```
- **XOR (`^`):** (Exclusive OR) Compares two bits. If the bits are different, the result is 1; otherwise, it's 0.
  ```
  1010 (10)
  ^ 1100 (12)
  ----
  0110 (6)
  ```
- **NOT (`~`):** (Bitwise Inversion/Complement) Flips the bits of a single operand. 1 becomes 0, and 0 becomes 1. For signed integers, this results in the two's complement, often leading to a negative number.
  ```
  ~ 1010 (10)  -> ...1111111111110101 (-11 in two's complement)
  ```
- **Left Shift (`<<`):** Shifts the bits of the first operand to the left by the number of positions specified by the second operand. Vacated bits are filled with zeros. Effectively multiplies by powers of 2.
  ```
  1010 (10) << 2 -> 101000 (40)
  ```
- **Right Shift (`>>`):** Shifts the bits of the first operand to the right by the number of positions specified by the second operand. Vacated bits are filled with sign-extension (for signed numbers) or zeros (for unsigned numbers, but Python handles signed integers typically). Effectively divides by powers of 2 (integer division).
  ```
  1010 (10) >> 2 -> 0010 (2)
  ```

### 3. Bitwise Operator Functions in `operator`

The `operator` module provides functions for each of these bitwise operations.

- `operator.and_(a, b)`: Corresponds to `a & b`.
- `operator.or_(a, b)`: Corresponds to `a | b`.
- `operator.xor(a, b)`: Corresponds to `a ^ b`.
- `operator.invert(a)`: Corresponds to `~a`.
- `operator.lshift(a, b)`: Corresponds to `a << b`.
- `operator.rshift(a, b)`: Corresponds to `a >> b`.

Additionally, there are in-place (augmented assignment) versions:

- `operator.iand(a, b)`: Corresponds to `a &= b`.
- `operator.ior(a, b)`: Corresponds to `a |= b`.
- `operator.ixor(a, b)`: Corresponds to `a ^= b`.
- `operator.ilshift(a, b)`: Corresponds to `a <<= b`.
- `operator.irshift(a, b)`: Corresponds to `a >>= b`.

### 4. Code Examples

```python
import operator
import functools # For functools.reduce, which is a common use case

print("--- 4. Code Examples for Bitwise Operations in `operator` Module ---")

num1 = 10  # Binary: 1010
num2 = 12  # Binary: 1100
num3 = 5   # Binary: 0101
num4 = -10 # Binary: ...11110110 (two's complement)


# Helper function to print binary representation
def to_binary_padded(num, bits=8):
    """Converts an integer to a padded binary string."""
    if num >= 0:
        return bin(num)[2:].zfill(bits)
    else:
        # For negative numbers, show 8 bits of two's complement for illustration
        # This is a simplification; Python's integers have arbitrary precision.
        return bin((1 << bits) + num)[2:] if num >= -(1 << (bits - 1)) else bin(num)

print(f"Numbers in binary (8-bit representation for positive):")
print(f"num1 = {num1} ({to_binary_padded(num1)})")
print(f"num2 = {num2} ({to_binary_padded(num2)})")
print(f"num3 = {num3} ({to_binary_padded(num3)})")
print(f"num4 = {num4}") # Showing raw decimal for negative numbers


# --- 4.1. Bitwise AND (`operator.and_`) ---
print("\n--- 4.1. Bitwise AND (`operator.and_`) ---")
# Result: 1000 (8)
result_and = operator.and_(num1, num2)
print(f"operator.and_({num1}, {num2}): {result_and} ({to_binary_padded(result_and)})")

# Use case: Masking bits (e.g., checking if a specific bit is set)
mask_bit2 = 0b0100 # Binary for 4 (check if 3rd bit from right is set)
is_bit2_set = operator.and_(num1, mask_bit2) != 0
print(f"Is 3rd bit of {num1} ({to_binary_padded(num1)}) set (mask {to_binary_padded(mask_bit2)})? {is_bit2_set}") # True


# --- 4.2. Bitwise OR (`operator.or_`) ---
print("\n--- 4.2. Bitwise OR (`operator.or_`) ---")
# Result: 1110 (14)
result_or = operator.or_(num1, num2)
print(f"operator.or_({num1}, {num2}): {result_or} ({to_binary_padded(result_or)})")

# Use case: Setting bits (e.g., ensuring a specific bit is 1)
set_bit1 = 0b0010 # Binary for 2 (set 2nd bit from right)
new_num_set_bit = operator.or_(num3, set_bit1)
print(f"Set 2nd bit of {num3} ({to_binary_padded(num3)}) to 1 (mask {to_binary_padded(set_bit1)}): {new_num_set_bit} ({to_binary_padded(new_num_set_bit)})") # 0101 | 0010 = 0111 (7)


# --- 4.3. Bitwise XOR (`operator.xor`) ---
print("\n--- 4.3. Bitwise XOR (`operator.xor`) ---")
# Result: 0110 (6)
result_xor = operator.xor(num1, num2)
print(f"operator.xor_({num1}, {num2}): {result_xor} ({to_binary_padded(result_xor)})")

# Use case: Flipping bits, simple encryption (XOR with a key), swapping numbers without temp variable
# 1010 ^ 0101 = 1111
flipped_bits = operator.xor(num1, 0b1111)
print(f"Flip all bits of {num1} ({to_binary_padded(num1)}): {flipped_bits} ({to_binary_padded(flipped_bits)})")


# --- 4.4. Bitwise NOT (Invert) (`operator.invert`) ---
print("\n--- 4.4. Bitwise NOT (Invert) (`operator.invert`) ---")
# Result of ~10 (1010) is -11 in two's complement.
result_invert = operator.invert(num1)
print(f"operator.invert({num1}): {result_invert}") # -11
# For negative numbers, it behaves like ~x = -x - 1
print(f"operator.invert({num4}): {operator.invert(num4)}") # ~(-10) = -(-10) - 1 = 10 - 1 = 9


# --- 4.5. Left Shift (`operator.lshift`) ---
print("\n--- 4.5. Left Shift (`operator.lshift`) ---")
# Result: 101000 (40)
result_lshift = operator.lshift(num1, 2)
print(f"operator.lshift({num1}, 2): {result_lshift} ({to_binary_padded(result_lshift)})")

# Use case: Fast multiplication by powers of 2
print(f"5 << 3: {operator.lshift(5, 3)}") # 5 * 2^3 = 5 * 8 = 40


# --- 4.6. Right Shift (`operator.rshift`) ---
print("\n--- 4.6. Right Shift (`operator.rshift`) ---")
# Result: 0010 (2)
result_rshift = operator.rshift(num1, 2)
print(f"operator.rshift({num1}, 2): {result_rshift} ({to_binary_padded(result_rshift)})")

# Use case: Fast integer division by powers of 2
print(f"40 >> 3: {operator.rshift(40, 3)}") # 40 // 8 = 5


# --- 4.7. In-place Bitwise Operations ---
print("\n--- 4.7. In-place Bitwise Operations ---")
# These modify the variable's value in-place if possible (for integers, it's reassignment).
mutable_int = 7 # Binary: 0111
print(f"Initial mutable_int: {mutable_int} ({to_binary_padded(mutable_int)})")

operator.iand(mutable_int, 0b0101) # 0111 & 0101 = 0101 (5)
print(f"After iand(..., 0b0101): {mutable_int} ({to_binary_padded(mutable_int)})") # Python's iand/ior/etc. for int types internally reassigns.

mutable_int = 7
mutable_int = operator.ior(mutable_int, 0b1000) # 0111 | 1000 = 1111 (15)
print(f"After ior(..., 0b1000): {mutable_int} ({to_binary_padded(mutable_int)})")


# --- 4.8. Use with `functools.reduce` ---
print("\n--- 4.8. Use with `functools.reduce` ---")
# Finding the common bits (intersection of all bits)
masks = [0b1111, 0b1010, 0b1100]
common_bits = functools.reduce(operator.and_, masks)
print(f"Common bits in {masks}: {common_bits} ({to_binary_padded(common_bits)})") # 1000 (8)

# Combining all set bits
all_set_bits = functools.reduce(operator.or_, masks)
print(f"All set bits in {masks}: {all_set_bits} ({to_binary_padded(all_set_bits)})") # 1111 (15)

```

### 1. Introduction to Sequence Operations in `operator`

Sequence operations are fundamental for manipulating ordered collections of items. The `operator` module provides functional equivalents for common sequence behaviors:

- **Item Access:** Getting, setting, or deleting elements by index.
- **Membership Testing:** Checking if an item is in a sequence (`in`).
- **Concatenation:** Joining sequences (`+`).
- **Repetition:** Repeating a sequence (`*`).
- **Length:** Getting the number of items (`len()`). (Note: `len()` is a built-in function, not an operator, but conceptually related.)

**Why use them?**

- **Functional Programming:** Essential when functions like `map()`, `filter()`, `sorted()`, or `functools.reduce()` require a callable that performs a sequence operation.
- **Clarity:** In some contexts, `operator.getitem` might be more explicit than `lambda x, y: x[y]`.
- **Dynamic Operations:** Allows for selecting which sequence operation to apply at runtime.
- **Performance (minor):** Like other `operator` functions, they can be slightly faster than equivalent `lambda` expressions as they avoid `lambda` object creation overhead.

---

### 2. Core Sequence Functions

These functions operate on sequences and return a new value or perform an action.

- `operator.concat(a, b)`: Corresponds to `a + b` (concatenation).
- `operator.contains(a, b)`: Corresponds to `b in a` (membership testing).
- `operator.countOf(a, b)`: Returns the number of occurrences of `b` in `a` (similar to `a.count(b)`).
- `operator.indexOf(a, b)`: Returns the first index of `b` in `a` (similar to `a.index(b)`). Raises `ValueError` if not found.
- `operator.getitem(a, b)`: Corresponds to `a[b]` (item retrieval by index or slice).
- `operator.setitem(a, b, c)`: Corresponds to `a[b] = c` (item assignment). Modifies `a` in-place.
- `operator.delitem(a, b)`: Corresponds to `del a[b]` (item deletion). Modifies `a` in-place.
- `operator.length_hint(obj)`: Returns an estimated length of the object. Useful for iterators where `len()` might not be available or efficient. (Less commonly used).

```python
import operator
import functools # For functools.reduce, functools.partial
import collections # For deque

print("--- 3. Code Examples for Sequence Operations in `operator` Module ---")

my_list = [1, 2, 3, 2, 5]
my_tuple = ('a', 'b', 'c', 'd')
my_string = "hello world"

# --- 3.1. Concatenation and Repetition ---
print("\n--- 3.1. Concatenation and Repetition ---")

# operator.concat(a, b): Equivalent to a + b
result_concat_list = operator.concat(my_list, [6, 7])
print(f"operator.concat({my_list}, [6, 7]): {result_concat_list}")

result_concat_str = operator.concat(my_string, "!")
print(f"operator.concat('{my_string}', '!'): {result_concat_str}")

# operator.repeat(seq, n): Equivalent to seq * n
# Note: This is operator.repeat, not to be confused with string/list method 'repeat'.
result_repeat_tuple = operator.repeat(my_tuple, 2)
print(f"operator.repeat({my_tuple}, 2): {result_repeat_tuple}")

result_repeat_str = operator.repeat("abc", 3)
print(f"operator.repeat('abc', 3): {result_repeat_str}")


# --- 3.2. Membership Testing ---
print("\n--- 3.2. Membership Testing ---")

# operator.contains(a, b): Equivalent to b in a
print(f"operator.contains({my_list}, 3): {operator.contains(my_list, 3)}") # True
print(f"operator.contains({my_list}, 99): {operator.contains(my_list, 99)}") # False
print(f"operator.contains('{my_string}', 'world'): {operator.contains(my_string, 'world')}") # True


# --- 3.3. Item Access (Get, Set, Delete) ---
print("\n--- 3.3. Item Access (Get, Set, Delete) ---")

# operator.getitem(a, b): Equivalent to a[b]
print(f"operator.getitem({my_list}, 0): {operator.getitem(my_list, 0)}") # 1
print(f"operator.getitem({my_tuple}, 2): {operator.getitem(my_tuple, 2)}") # 'c'
print(f"operator.getitem('{my_string}', 6): {operator.getitem(my_string, 6)}") # 'w'

# Getitem with slicing
print(f"operator.getitem({my_list}, slice(1, 4)): {operator.getitem(my_list, slice(1, 4))}") # [2, 3, 2]
print(f"operator.getitem('{my_string}', slice(0, 5)): {operator.getitem(my_string, slice(0, 5))}") # 'hello'

# operator.setitem(a, b, c): Equivalent to a[b] = c (modifies in-place)
mutable_seq = ['x', 'y', 'z']
operator.setitem(mutable_seq, 1, 'Y')
print(f"operator.setitem({['x', 'y', 'z']}, 1, 'Y'): {mutable_seq}") # ['x', 'Y', 'z']

# Setitem with slicing (can change length)
mutable_seq_slice = [10, 20, 30, 40]
operator.setitem(mutable_seq_slice, slice(1, 3), [25, 35, 45])
print(f"operator.setitem({[10, 20, 30, 40]}, slice(1, 3), [25, 35, 45]): {mutable_seq_slice}") # [10, 25, 35, 45, 40]

# operator.delitem(a, b): Equivalent to del a[b] (modifies in-place)
mutable_seq_del = ['a', 'b', 'c', 'd']
operator.delitem(mutable_seq_del, 0)
print(f"operator.delitem({['a', 'b', 'c', 'd']}, 0): {mutable_seq_del}") # ['b', 'c', 'd']

# Delitem with slicing
mutable_seq_del_slice = [100, 200, 300, 400, 500]
operator.delitem(mutable_seq_del_slice, slice(1, 4)) # Delete elements at indices 1, 2, 3
print(f"operator.delitem({[100, 200, 300, 400, 500]}, slice(1, 4)): {mutable_seq_del_slice}") # [100, 500]


# --- 3.4. Count and Index ---
print("\n--- 3.4. Count and Index ---")

# operator.countOf(a, b): Returns the number of occurrences of b in a
list_with_duplicates = [1, 2, 3, 2, 4, 2]
print(f"operator.countOf({list_with_duplicates}, 2): {operator.countOf(list_with_duplicates, 2)}") # 3
print(f"operator.countOf('{my_string}', 'o'): {operator.countOf(my_string, 'o')}") # 2

# operator.indexOf(a, b): Returns the first index of b in a
print(f"operator.indexOf({list_with_duplicates}, 2): {operator.indexOf(list_with_duplicates, 2)}") # 1

# Raises ValueError if not found
try:
    operator.indexOf(my_list, 99)
except ValueError as e:
    print(f"Error for operator.indexOf({my_list}, 99): {e}")


# --- 3.5. Practical Use Cases ---
print("\n--- 3.5. Practical Use Cases ---")

# Example: Summing elements at a specific index across a list of tuples
data_points = [(1, 5), (2, 8), (3, 2), (4, 10)]
# Sum of second elements: 5 + 8 + 2 + 10 = 25
sum_of_second_elements = sum(map(functools.partial(operator.getitem, 1), data_points))
print(f"Sum of second elements in {data_points}: {sum_of_second_elements}")

# Example: Checking if all items in a list are present in a master list
master_list = [1, 2, 3, 4, 5, 6, 7]
items_to_check = [2, 5, 7]
all_contained = all(map(functools.partial(operator.contains, master_list), items_to_check))
print(f"Are all {items_to_check} contained in {master_list}? {all_contained}")

items_to_check_fail = [2, 5, 9]
all_contained_fail = all(map(functools.partial(operator.contains, master_list), items_to_check_fail))
print(f"Are all {items_to_check_fail} contained in {master_list}? {all_contained_fail}")

```

### 2. `operator.itemgetter(*items)`

**Theory:**
`itemgetter(*items)` returns a callable object that fetches the `item`(s) at the specified index or key from its operand.

- If a single `item` is given (e.g., `itemgetter(0)`), the returned callable will fetch that single item.
- If multiple `items` are given (e.g., `itemgetter(0, 2)`), the returned callable will fetch multiple items, and return them as a tuple.

This is extremely useful when you want to sort a list of tuples (or lists) based on a specific element, or sort a list of dictionaries based on a specific key.

**Syntax:** `operator.itemgetter(index_or_key, ...)`

**Code Examples:**

```python
import operator

print("--- 2. `operator.itemgetter(*items)` ---")

# Data: A list of tuples (name, age, city)
data_tuples = [
    ('Alice', 30, 'New York'),
    ('Bob', 25, 'Los Angeles'),
    ('Charlie', 35, 'New York'),
    ('David', 25, 'Chicago')
]

# 2.1 Fetching a single item
get_name = operator.itemgetter(0) # Callable to get the item at index 0
get_age = operator.itemgetter(1) # Callable to get the item at index 1

print(f"2.1 Name of first person: {get_name(data_tuples[0])}")
print(f"    Age of second person: {get_age(data_tuples[1])}")

# Common Use Case: Sorting a list of tuples/lists by a specific element
print(f"\nOriginal data (tuples): {data_tuples}")

# Sort by age (index 1)
sorted_by_age = sorted(data_tuples, key=operator.itemgetter(1))
print(f"    Sorted by age: {sorted_by_age}")

# Sort by city (index 2), then by name (index 0) - composite key
sorted_by_city_then_name = sorted(data_tuples, key=operator.itemgetter(2, 0))
print(f"    Sorted by city, then name: {sorted_by_city_then_name}")

# Data: A list of dictionaries
data_dicts = [
    {'name': 'Alice', 'score': 90, 'id': 101},
    {'name': 'Bob', 'score': 75, 'id': 103},
    {'name': 'Charlie', 'score': 90, 'id': 102}
]

# Sorting a list of dictionaries by a specific key
# Sort by score (key 'score')
sorted_by_score = sorted(data_dicts, key=operator.itemgetter('score'))
print(f"\nOriginal data (dicts): {data_dicts}")
print(f"    Sorted by score: {sorted_by_score}")

# Sort by score (key 'score'), then by id (key 'id')
sorted_by_score_then_id = sorted(data_dicts, key=operator.itemgetter('score', 'id'))
print(f"    Sorted by score, then ID: {sorted_by_score_then_id}")

# 2.2 Getting multiple items
get_name_city = operator.itemgetter(0, 2)
print(f"2.2 Name and city of first person: {get_name_city(data_tuples[0])}") # Returns a tuple

# Example with map: Extracting specific columns
names_and_cities = list(map(operator.itemgetter(0, 2), data_tuples))
print(f"    Names and cities: {names_and_cities}")

```

### 3. `operator.attrgetter(*attrs)`

**Theory:**
`attrgetter(*attrs)` returns a callable object that fetches the named `attribute`(s) from its operand.

- If a single `attr` is given (e.g., `attrgetter('name')`), the returned callable will fetch that single attribute.
- If multiple `attrs` are given (e.g., `attrgetter('name', 'age')`), the returned callable will fetch multiple attributes, and return them as a tuple.
- Attributes can be nested (e.g., `'address.street'`).

This is ideal for sorting or processing lists of custom objects.

**Syntax:** `operator.attrgetter(attribute_name, ...)`

**Code Examples:**

```python
import operator

print("\n--- 3. `operator.attrgetter(*attrs)` ---")

class Person:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city
    def __repr__(self): # For better printing in lists
        return f"Person(name='{self.name}', age={self.age}, city='{self.city}')"

class Address:
    def __init__(self, street, zip_code):
        self.street = street
        self.zip_code = zip_code
    def __repr__(self):
        return f"Address('{self.street}', '{self.zip_code}')"

class User:
    def __init__(self, username, email, contact_address):
        self.username = username
        self.email = email
        self.address = contact_address # This is an Address object
    def __repr__(self):
        return f"User('{self.username}', '{self.email}', {self.address})"


people_objects = [
    Person('Alice', 30, 'New York'),
    Person('Bob', 25, 'Los Angeles'),
    Person('Charlie', 35, 'New York')
]

# 3.1 Fetching a single attribute
get_person_name = operator.attrgetter('name')
get_person_age = operator.attrgetter('age')

print(f"3.1 Name of first person object: {get_person_name(people_objects[0])}")
print(f"    Age of second person object: {get_person_age(people_objects[1])}")

# Common Use Case: Sorting a list of objects by an attribute
print(f"\nOriginal people objects: {people_objects}")

# Sort by age
sorted_people_by_age = sorted(people_objects, key=operator.attrgetter('age'))
print(f"    Sorted people by age: {sorted_people_by_age}")

# Sort by city, then by name
sorted_people_by_city_name = sorted(people_objects, key=operator.attrgetter('city', 'name'))
print(f"    Sorted people by city, then name: {sorted_people_by_city_name}")


# 3.2 Getting multiple attributes
get_name_city = operator.attrgetter('name', 'city')
print(f"3.2 Name and city of first person object: {get_name_city(people_objects[0])}") # Returns a tuple

# 3.3 Accessing nested attributes
user_with_address = User('john.doe', 'john@example.com', Address('123 Main St', '90210'))
get_zip_code = operator.attrgetter('address.zip_code')
print(f"\n3.3 Zip code of user: {get_zip_code(user_with_address)}")

```

### 4. `operator.methodcaller(name, *args, **kwargs)`

**Theory:**
`methodcaller(name, *args, **kwargs)` returns a callable object that, when called with an operand, will call the method named `name` on that operand. Any additional `*args` and `**kwargs` passed to `methodcaller` itself will be passed on to the method being called.

This is useful when you want to apply the same method call to every item in an iterable.

**Syntax:** `operator.methodcaller(method_name, arg1, arg2=value, ...)`

**Code Examples:**

```python
import operator

print("\n--- 4. `operator.methodcaller(name, *args, **kwargs)` ---")

strings = ["hello world", "PYTHON programming", "example sentence"]

# 4.1 Calling a method with no arguments
upper_caller = operator.methodcaller('upper')
# Use map to apply 'upper()' to each string
uppercase_strings = list(map(upper_caller, strings))
print(f"4.1 Uppercased strings: {uppercase_strings}")

# 4.2 Calling a method with positional arguments
strip_caller = operator.methodcaller('strip', 'h') # strip 'h' from ends
stripped_strings = list(map(strip_caller, ["hello", "hi there", "Oh no"]))
print(f"4.2 Stripped strings (stripping 'h'): {stripped_strings}")

# 4.3 Calling a method with keyword arguments
replace_caller = operator.methodcaller('replace', 'o', 'X', 1) # replace first 'o' with 'X'
replaced_strings = list(map(replace_caller, ["foo", "bar", "long string"]))
print(f"4.3 Replaced strings (replace 'o' with 'X', count=1): {replaced_strings}")

# 4.4 Calling a method on custom objects
class Greeter:
    def __init__(self, name):
        self.name = name
    def say_hello(self, greeting_word="Hello"):
        return f"{greeting_word}, {self.name}!"
    def __repr__(self):
        return f"Greeter('{self.name}')"

greeters = [Greeter('Alice'), Greeter('Bob')]

# Call say_hello with default greeting
hello_caller = operator.methodcaller('say_hello')
greetings = list(map(hello_caller, greeters))
print(f"\n4.4 Greetings with default: {greetings}")

# Call say_hello with a custom greeting
custom_hello_caller = operator.methodcaller('say_hello', 'Hi there')
custom_greetings = list(map(custom_hello_caller, greeters))
print(f"    Greetings with custom: {custom_greetings}")

```

While direct attribute access (`obj.attr`) or item access (`seq[idx]`) remains the most common for individual operations, the `operator` module's specialized functions are invaluable when dealing with collections and higher-order functions.


The `operator` module in Python provides functions that correspond to the built-in bitwise operators (`&`, `|`, `^`, `~`, `<<`, `>>`). These functions are incredibly useful when you need to perform bit-level operations in a functional programming style, allowing you to pass these operations as arguments to other functions or build dynamic bitwise logic.

Bitwise operations work on the binary representations of integers. They are commonly used in low-level programming, data compression, encryption, flag manipulation, and optimizing certain calculations.

Here's a comprehensive look at the bitwise operators available in the `operator` module, with code examples:

```python
import operator

print("--- Bitwise Operators from Python's 'operator' Module ---")
print("-------------------------------------------------------\n")

# Define some integer operands for bitwise operations
x = 0b1100  # 12 in decimal
y = 0b1010  # 10 in decimal
z = 0b0011  # 3 in decimal
neg_val = -10 # Negative numbers are represented using two's complement

print(f"Operands:")
print(f"  x = {x} (binary: {bin(x)})")
print(f"  y = {y} (binary: {bin(y)})")
print(f"  z = {z} (binary: {bin(z)})")
print(f"  neg_val = {neg_val}\n")


# 1. Bitwise AND: operator.and_(a, b) equivalent to a & b
# Sets each bit to 1 if both bits are 1.
print("1. Bitwise AND (operator.and_):")
result_and = operator.and_(x, y)
print(f"operator.and_({bin(x)}, {bin(y)}) -> {bin(result_and)} ({result_and})")
# 1100 (x)
# 1010 (y)
# ----
# 1000 (result) = 8
print("\n")


# 2. Bitwise OR: operator.or_(a, b) equivalent to a | b
# Sets each bit to 1 if any of the bits is 1.
print("2. Bitwise OR (operator.or_):")
result_or = operator.or_(x, y)
print(f"operator.or_({bin(x)}, {bin(y)})  -> {bin(result_or)} ({result_or})")
# 1100 (x)
# 1010 (y)
# ----
# 1110 (result) = 14
print("\n")


# 3. Bitwise XOR (Exclusive OR): operator.xor(a, b) equivalent to a ^ b
# Sets each bit to 1 if only one of the bits is 1.
print("3. Bitwise XOR (operator.xor):")
result_xor = operator.xor(x, y)
print(f"operator.xor({bin(x)}, {bin(y)})  -> {bin(result_xor)} ({result_xor})")
# 1100 (x)
# 1010 (y)
# ----
# 0110 (result) = 6
print("\n")


# 4. Bitwise NOT (Inversion): operator.invert(obj) equivalent to ~obj
# Inverts all the bits. For integers, this is equivalent to -(x + 1).
print("4. Bitwise NOT (operator.invert):")
result_invert_x = operator.invert(x)
result_invert_neg = operator.invert(neg_val)
print(f"operator.invert({bin(x)})   -> {bin(result_invert_x)} ({result_invert_x})")
# For x = 12 (0...01100 in two's complement):
# ~12 = -13
print(f"operator.invert({neg_val}) -> {bin(result_invert_neg)} ({result_invert_neg})\n")
# For neg_val = -10 (which is represented as ...11110110 in two's complement)
# ~-10 = 9


# 5. Left Shift: operator.lshift(a, b) equivalent to a << b
# Shifts the bits of 'a' to the left by 'b' positions. Equivalent to multiplying by 2**b.
print("5. Left Shift (operator.lshift):")
result_lshift = operator.lshift(x, 2)
print(f"operator.lshift({bin(x)}, 2) -> {bin(result_lshift)} ({result_lshift})")
# 0000 1100 (12) << 2 -> 0011 0000 (48)
print(f"operator.lshift({z}, 4)     -> {bin(operator.lshift(z, 4))} ({operator.lshift(z, 4)})\n")


# 6. Right Shift: operator.rshift(a, b) equivalent to a >> b
# Shifts the bits of 'a' to the right by 'b' positions. Equivalent to floor dividing by 2**b.
# For positive numbers, it's a logical shift (fills with 0s).
# For negative numbers, it's an arithmetic shift (fills with sign bit).
print("6. Right Shift (operator.rshift):")
result_rshift_x = operator.rshift(x, 2)
result_rshift_neg = operator.rshift(neg_val, 2)
print(f"operator.rshift({bin(x)}, 2)  -> {bin(result_rshift_x)} ({result_rshift_x})")
# 0000 1100 (12) >> 2 -> 0000 0011 (3)
print(f"operator.rshift({neg_val}, 2) -> {bin(result_rshift_neg)} ({result_rshift_neg})\n")
# For -10 (binary representation ...11110110 in two's complement)
# >> 2 results in ...11111101 (-3)


# --- Practical Use Cases ---
print("--- Practical Use Cases for 'operator' bitwise functions ---\n")

# Use Case 1: Flag Management (Setting, Checking, Clearing flags)
# Common in configurations, permissions, or state representation.
ACTIVE_FLAG = 0b001 # 1
EDIT_FLAG   = 0b010 # 2
DELETE_FLAG = 0b100 # 4

user_permission = 0b000 # Start with no permissions

# Set ACTIVE_FLAG
user_permission = operator.or_(user_permission, ACTIVE_FLAG)
print(f"Permission after setting ACTIVE: {bin(user_permission)}") # 001

# Set EDIT_FLAG
user_permission = operator.or_(user_permission, EDIT_FLAG)
print(f"Permission after setting EDIT:   {bin(user_permission)}") # 011

# Check if EDIT_FLAG is set
has_edit = operator.and_(user_permission, EDIT_FLAG) == EDIT_FLAG
print(f"Does user have EDIT permission? {has_edit}")

# Check if DELETE_FLAG is set (it's not)
has_delete = operator.and_(user_permission, DELETE_FLAG) == DELETE_FLAG
print(f"Does user have DELETE permission? {has_delete}")

# Clear EDIT_FLAG
# To clear a bit, use AND with the inverted flag (mask)
user_permission = operator.and_(user_permission, operator.invert(EDIT_FLAG))
print(f"Permission after clearing EDIT: {bin(user_permission)}\n") # 001

# Use Case 2: Efficient Power-of-Two Checks
print("2. Efficient Power-of-Two Checks:")
# A number is a power of two if and only if it's positive and (number & (number - 1)) is 0.
def is_power_of_two(n):
    return n > 0 and operator.and_(n, operator.sub(n, 1)) == 0

print(f"Is 8 a power of two? {is_power_of_two(8)}")   # 1000 & 0111 = 0000 -> True
print(f"Is 6 a power of two? {is_power_of_two(6)}")   # 0110 & 0101 = 0100 -> False
print(f"Is 1 a power of two? {is_power_of_two(1)}")   # 0001 & 0000 = 0000 -> True
print(f"Is 0 a power of two? {is_power_of_two(0)}\n")  # False


# Use Case 3: Swapping Two Numbers Without a Temporary Variable (Illustrative)
print("3. Swapping Two Numbers Without a Temporary Variable (Illustrative):")
# While `x, y = y, x` is the Pythonic way, XOR swap is a classic bitwise trick.
# This works because XOR is associative and commutative, and x ^ x = 0.
val1 = 15 # 1111
val2 = 7  # 0111
print(f"Before swap: val1={val1}, val2={val2}")

val1 = operator.xor(val1, val2) # val1 = 15 ^ 7 = 1000 (8)
val2 = operator.xor(val1, val2) # val2 = 8 ^ 7  = 1111 (15)
val1 = operator.xor(val1, val2) # val1 = 8 ^ 15 = 0111 (7)

print(f"After XOR swap: val1={val1}, val2={val2}\n")


# Use Case 4: Working with Hash Values or Checksums (simplified example)
print("4. Simple Hash/Checksum Operation:")
# Bitwise operations are fundamental in hashing algorithms and checksum calculations.
# This is a very simplistic example.
data_bytes = [0xAA, 0x55, 0xF0, 0x0F] # Example byte data

checksum = 0
for byte_val in data_bytes:
    # XOR each byte with the current checksum
    checksum = operator.xor(checksum, byte_val)
    # Rotate left by a few bits (simplified: left shift then OR with right shifted part)
    # This is a basic form of mixing bits to avoid simple patterns
    checksum = operator.or_(operator.lshift(checksum, 1), operator.rshift(checksum, 7)) # Rotate left by 1 bit for an 8-bit value
    checksum = operator.and_(checksum, 0xFF) # Keep it within 8 bits
print(f"Data: {[hex(b) for b in data_bytes]}")
print(f"Calculated Checksum: {hex(checksum)}\n")


print("--- End of 'operator' Bitwise Operators Demonstration ---")
```
