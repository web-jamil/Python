  ```python
  result, error = quad(integrand, 0, 1)
  ```

- This submodule provides functions for solving optimization problems (finding the minimum or maximum of functions) and solving systems of equations.

  ```python
  from scipy.optimize import minimize

  # Defining a simple quadratic function
  def objective(x):
      return x**2 + 4*x + 4

  # Finding the minimum of the function
  result = minimize(objective, 0)
  print(result)
  ```

- **Common Functions:**
  - `minimize`: Finds the minimum of a scalar function.
  - `fsolve`: Solves a system of nonlinear equations.

  ```python
  from scipy.interpolate import interp1d

  x = [1, 2, 3]
  y = [1, 4, 9]

  # Creating a linear interpolation function
  f = interp1d(x, y, kind='linear')

  # Interpolating a new point
  print(f(2.5))  # Output: 6.5 (since it's a linear interpolation)
  ```

### 1. `scipy.optimize`: Optimization and Root Finding

**Theory:**
Optimization is the process of finding the best solution from a set of available alternatives. In mathematics, it often means finding the minimum or maximum of a function. Root finding involves finding the values where a function equals zero. `scipy.optimize` provides algorithms for:

- Minimizing (or maximizing) scalar functions (e.g., `minimize`).
- Finding roots of equations (e.g., `root`, `fsolve`).
- Curve fitting (e.g., `curve_fit`).
- Linear programming (e.g., `linprog`).

**Code Example (Minimizing a function):**

Let's minimize the Rosenbrock function, a common test function for optimization algorithms:
$f(x, y) = (1 - x)^2 + 100(y - x^2)^2$

```python
import numpy as np
from scipy.optimize import minimize
import matplotlib.pyplot as plt

print("--- 1. scipy.optimize: Minimization ---")

# Define the Rosenbrock function
def rosenbrock(x):
    return (1 - x[0])**2 + 100 * (x[1] - x[0]**2)**2

# Initial guess
x0 = np.array([0.5, 0.5]) # Start near (0,0)

# Perform minimization using the 'Nelder-Mead' simplex algorithm
res = minimize(rosenbrock, x0, method='Nelder-Mead', options={'disp': True})

print(f"\nOptimization Result:")
print(f"Success: {res.success}")
print(f"Message: {res.message}")
print(f"Optimal parameters (x, y): {res.x}")
print(f"Minimum function value: {res.fun}")

# Example: Curve fitting
print("\n--- 1. scipy.optimize: Curve Fitting ---")

# Data to fit (e.g., a noisy sine wave)
x_data = np.linspace(0, 4, 50)
y_data = 3 * np.sin(x_data) + 2 + np.random.normal(0, 0.5, size=x_data.shape)

# Define the model function (e.g., a sine wave with amplitude A, frequency B, phase C, and offset D)
def sine_model(x, A, B, C, D):
    return A * np.sin(B * x + C) + D

# Use curve_fit to find the best parameters
# Initial guess for parameters (A, B, C, D)
initial_guess = [3, 1, 0, 2]
params, covariance = optimize.curve_fit(sine_model, x_data, y_data, p0=initial_guess)

print(f"Fitted Parameters (A, B, C, D): {params}")

# Plotting the results
plt.figure(figsize=(10, 5))
plt.scatter(x_data, y_data, label='Noisy Data')
plt.plot(x_data, sine_model(x_data, *params), color='red', label='Fitted Curve')
plt.title('Curve Fitting with scipy.optimize.curve_fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

### 2. `scipy.integrate`: Numerical Integration

**Theory:**
Numerical integration (or quadrature) is the process of approximating the value of a definite integral. This is essential when analytical solutions are impossible or difficult to obtain. `scipy.integrate` provides various functions for:

- Integrating functions (e.g., `quad` for single integrals, `dblquad` for double integrals).
- Integrating sampled data (e.g., `trapz`, `simps`).
- Solving ordinary differential equations (ODEs) (e.g., `solve_ivp`).

**Code Example (Single and Double Integration):**

Let's integrate $f(x) = e^{-x^2}$ from $0$ to $1$ (related to the error function) and $f(x, y) = x \cdot y$ over a rectangular region.

```python
from scipy import integrate
import numpy as np

print("\n--- 2. scipy.integrate: Numerical Integration ---")

# Example 1: Single integral using quad
def f(x):
    return np.exp(-x**2)

result_quad, error_quad = integrate.quad(f, 0, 1)
print(f"Integral of e^(-x^2) from 0 to 1: {result_quad:.6f} (Error: {error_quad:.2e})")

# Example 2: Double integral using dblquad
# Integrate f(x, y) = x * y over x in [0, 1], y in [0, 2]
def g(y, x): # Note: argument order is y, x for dblquad
    return x * y

# dblquad(func, a, b, gfun, hfun)
# gfun and hfun are the lower and upper limits of y, which can be functions of x
result_dblquad, error_dblquad = integrate.dblquad(g, 0, 1, lambda x: 0, lambda x: 2)
print(f"Integral of x*y over [0,1]x[0,2]: {result_dblquad:.6f} (Error: {error_dblquad:.2e})")

# Example 3: Integrating sampled data using trapezoidal rule
x_data_trapz = np.array([0, 1, 2, 3])
y_data_trapz = np.array([1, 4, 9, 16]) # y = x^2 + 1
integral_trapz = integrate.trapz(y_data_trapz, x_data_trapz)
print(f"Integral of sampled data (trapz): {integral_trapz:.2f}")

# Example 4: Solving an Ordinary Differential Equation (ODE)
print("\n--- 2. scipy.integrate: Solving ODEs (solve_ivp) ---")

# Define the ODE: dy/dt = -2y
def model(t, y):
    return -2 * y

# Initial condition: y(0) = 1
y0 = [1]
# Time span for solution
t_span = (0, 4)
# Points where solution is to be evaluated
t_eval = np.linspace(0, 4, 100)

sol = integrate.solve_ivp(model, t_span, y0, t_eval=t_eval)

print(f"ODE solution at t=0: {sol.y[0,0]:.4f}")
print(f"ODE solution at t=1: {sol.y[0,25]:.4f}") # Approx t=1
print(f"ODE solution at t=4: {sol.y[0,-1]:.4f}") # At end

# plt.figure(figsize=(8, 4))
# plt.plot(sol.t, sol.y[0], label='y(t) numerical')
# plt.plot(sol.t, np.exp(-2*sol.t), '--', label='y(t) analytical (e^(-2t))')
# plt.title('Solving ODE dy/dt = -2y with solve_ivp')
# plt.xlabel('t')
# plt.ylabel('y')
# plt.legend()
# plt.grid(True)
# plt.show()
```

### 3. `scipy.interpolate`: Interpolation

**Theory:**
Interpolation is the process of constructing new data points within the range of a discrete set of known data points. It allows you to estimate values between observed measurements. `scipy.interpolate` offers various interpolation methods, including:

- One-dimensional interpolation (e.g., `interp1d`).
- Spline interpolation (e.g., `CubicSpline`, `splrep`, `splev`).
- Multi-dimensional interpolation (e.g., `griddata`).

**Code Example (1D Interpolation):**

Let's create some discrete data points and then interpolate values between them.

```python
from scipy import interpolate
import numpy as np
import matplotlib.pyplot as plt

print("\n--- 3. scipy.interpolate: Interpolation ---")

# Original data points
x_obs = np.linspace(0, 10, num=11)
y_obs = np.cos(x_obs) + 0.1 * np.random.rand(len(x_obs)) # Add some noise

# Create interpolation functions
f_linear = interpolate.interp1d(x_obs, y_obs, kind='linear')
f_cubic = interpolate.interp1d(x_obs, y_obs, kind='cubic')

# Points to evaluate the interpolated function
x_new = np.linspace(0, 10, num=100, endpoint=True)

# Evaluate
y_linear = f_linear(x_new)
y_cubic = f_cubic(x_new)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(x_obs, y_obs, 'o', label='Observed Data')
plt.plot(x_new, y_linear, '-', label='Linear Interpolation')
plt.plot(x_new, y_cubic, '--', label='Cubic Interpolation')
plt.title('1D Interpolation with scipy.interpolate')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

### 4. `scipy.signal`: Signal Processing

**Theory:**
The `scipy.signal` subpackage provides functions for signal processing, including filtering, convolution, correlation, spectral analysis, and waveform generation. It's used in various fields like audio processing, image processing, and communications.

**Code Example (Filtering a Signal):**

Let's create a noisy signal and apply a Butterworth filter to smooth it.

```python
from scipy import signal
import numpy as np
import matplotlib.pyplot as plt

print("\n--- 4. scipy.signal: Signal Processing ---")

# Parameters
fs = 1000 # Sampling frequency (Hz)
t = np.linspace(0, 1, fs, endpoint=False) # 1 second of data
freq_signal = 5 # Frequency of the main signal (Hz)
freq_noise = 50 # Frequency of the noise (Hz)

# Create a clean sine wave and add high-frequency noise
clean_signal = np.sin(2 * np.pi * freq_signal * t)
noisy_signal = clean_signal + 0.5 * np.sin(2 * np.pi * freq_noise * t) + \
               0.2 * np.random.randn(len(t)) # Add some random noise too

# Design a Butterworth low-pass filter
# N = order of the filter
# Wn = critical frequency (normalized to Nyquist frequency: Wn = cutoff_freq / (fs/2))
cutoff_freq = 10 # Hz
nyquist_freq = 0.5 * fs
Wn = cutoff_freq / nyquist_freq
N = 4 # 4th order filter

b, a = signal.butter(N, Wn, btype='low', analog=False)

# Apply the filter using filtfilt (zero-phase filter)
filtered_signal = signal.filtfilt(b, a, noisy_signal)

# Plotting
plt.figure(figsize=(12, 6))
plt.plot(t, noisy_signal, label='Noisy Signal', alpha=0.7)
plt.plot(t, filtered_signal, color='red', linewidth=2, label='Filtered Signal')
plt.plot(t, clean_signal, '--', color='green', label='Original Clean Signal')
plt.title('Signal Filtering with scipy.signal.butter and filtfilt')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```

### 5. `scipy.stats`: Statistical Functions

**Theory:**
`scipy.stats` contains a large number of probability distributions (continuous and discrete) and statistical functions. It's incredibly useful for hypothesis testing, data analysis, and random variable generation.

**Code Example (Normal Distribution and T-test):**

```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt

print("\n--- 5. scipy.stats: Statistical Functions ---")

# Example 1: Normal Distribution (PDF, CDF, sampling)
mu, sigma = 0, 1 # Mean and standard deviation
x_norm = np.linspace(-3, 3, 100)
pdf_norm = stats.norm.pdf(x_norm, loc=mu, scale=sigma) # Probability Density Function
cdf_norm = stats.norm.cdf(x_norm, loc=mu, scale=sigma) # Cumulative Distribution Function

# Generate random samples
random_samples = stats.norm.rvs(loc=mu, scale=sigma, size=1000)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(x_norm, pdf_norm, label='PDF')
plt.plot(x_norm, cdf_norm, label='CDF')
plt.title('Standard Normal Distribution')
plt.xlabel('X')
plt.ylabel('Probability/CDF')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.hist(random_samples, bins=30, density=True, alpha=0.7, label='Random Samples Histogram')
plt.plot(x_norm, pdf_norm, color='red', linewidth=2, label='Normal PDF')
plt.title('Histogram of Normal Samples')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Example 2: T-test (comparing means of two independent samples)
print("\n--- 5. scipy.stats: T-test ---")

# Sample data
# Group 1: Scores of students who used Method A
group1_scores = np.array([85, 92, 78, 88, 95, 80, 89, 91, 84, 87])
# Group 2: Scores of students who used Method B
group2_scores = np.array([75, 80, 70, 82, 79, 73, 85, 77, 81, 76])

# Perform independent t-test (assuming equal variances: equal_var=True)
t_stat, p_value = stats.ttest_ind(group1_scores, group2_scores, equal_var=True)

print(f"Group 1 Mean: {np.mean(group1_scores):.2f}")
print(f"Group 2 Mean: {np.mean(group2_scores):.2f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print(f"P-value ({p_value:.4f}) is less than alpha ({alpha}), so we reject the null hypothesis.")
    print("Conclusion: There is a statistically significant difference between the means of the two groups.")
else:
    print(f"P-value ({p_value:.4f}) is greater than alpha ({alpha}), so we fail to reject the null hypothesis.")
    print("Conclusion: There is no statistically significant difference between the means of the two groups.")
```

### 6. `scipy.linalg`: Linear Algebra

- Matrix exponentials, logarithms, and more.

**Code Example (Solving Linear System and Matrix Inverse):**

Solve the system:
$2x + y = 5$
$x + 3y = 10$

```python
from scipy import linalg
import numpy as np

print("\n--- 6. scipy.linalg: Linear Algebra ---")

# Example 1: Solving a linear system Ax = b
A = np.array([[2, 1], [1, 3]]) # Coefficient matrix
b = np.array([5, 10])         # Right-hand side vector

# Solve for x
x_solution = linalg.solve(A, b)
print(f"Solution to Ax = b: x = {x_solution}")
print(f"Check: A @ x = {A @ x_solution}") # Should be close to b

# Example 2: Matrix Inverse
matrix_to_invert = np.array([[1, 2], [3, 4]])
inverse_matrix = linalg.inv(matrix_to_invert)
print(f"\nOriginal Matrix:\n{matrix_to_invert}")
print(f"Inverse Matrix:\n{inverse_matrix}")
print(f"Check (Original @ Inverse):\n{matrix_to_invert @ inverse_matrix}") # Should be identity matrix
```

### 7. `scipy.special`: Special Functions

**Theory:**
This subpackage provides a wide range of mathematical functions that are "special" in the sense that they are not elementary functions but appear frequently in scientific computations, such as:

- Bessel functions
- Gamma functions
- Error functions
- Legendre polynomials
- And many more, often solutions to differential equations.

**Code Example (Gamma Function and Error Function):**

```python
from scipy import special
import numpy as np

print("\n--- 7. scipy.special: Special Functions ---")

# Example 1: Gamma function (generalization of factorial)
# Gamma(n) = (n-1)! for positive integers n
print(f"Gamma(5) = {special.gamma(5)}")   # Should be 4! = 24
print(f"Gamma(4.5) = {special.gamma(4.5):.4f}")

# Example 2: Error Function (erf)
# Related to the cumulative distribution function of the normal distribution
x_erf = np.linspace(-3, 3, 10)
erf_values = special.erf(x_erf)
print(f"\nX values for erf: {x_erf}")
print(f"erf(x) values: {erf_values}")

# CDF of standard normal distribution can be calculated from erf:
# CDF(x) = 0.5 * (1 + erf(x / sqrt(2)))
def normal_cdf_from_erf(x):
    return 0.5 * (1 + special.erf(x / np.sqrt(2)))

print(f"Normal CDF at x=0 (should be 0.5): {normal_cdf_from_erf(0):.4f}")
print(f"Normal CDF at x=1.96 (approx 97.5%): {normal_cdf_from_erf(1.96):.4f}")
```

#### 3.1. `scipy.stats` (Statistical Functions)

**Theory:**

- **Probability Distributions:** Access to PDFs (Probability Density Functions), CDFs (Cumulative Distribution Functions), inverse CDFs (PPFs - Percent Point Functions), and random variate generation for various distributions (e.g., `norm` for normal, `t` for t-distribution, `binom` for binomial).

**Code Examples:**

```python
print("\n--- 3.1. `scipy.stats` (Statistical Functions) ---")

# Sample data
data = np.array([1, 2, 2, 3, 4, 4, 4, 5, 6, 7, 8, 9, 10])

# Descriptive Statistics
print("\nDescriptive Statistics:")
desc_stats = stats.describe(data)
print(f"Number of observations: {desc_stats.nobs}")
print(f"Min, Max: {desc_stats.minmax}")
print(f"Mean: {desc_stats.mean}")
print(f"Variance: {desc_stats.variance}")
print(f"Skewness: {desc_stats.skewness}") # Measures asymmetry of the data distribution
print(f"Kurtosis: {desc_stats.kurtosis}") # Measures 'tailedness' of the data distribution

# Probability Distributions (Normal Distribution example)
print("\nNormal Distribution (`stats.norm`):")
# Create a normal distribution object with mean=0 and std_dev=1 (standard normal)
norm_dist = stats.norm(loc=0, scale=1)

# PDF (Probability Density Function) - height of the curve at a given point
pdf_at_0 = norm_dist.pdf(0)
pdf_at_1 = norm_dist.pdf(1)
print(f"PDF at x=0 (mean): {pdf_at_0:.4f}")
print(f"PDF at x=1 (one std dev away): {pdf_at_1:.4f}")

# CDF (Cumulative Distribution Function) - probability that a random variable is <= a value
cdf_at_0 = norm_dist.cdf(0)
cdf_at_1_96 = norm_dist.cdf(1.96) # Approx 97.5% of data is below 1.96 in std normal
print(f"CDF at x=0: {cdf_at_0:.4f}")
print(f"CDF at x=1.96: {cdf_at_1_96:.4f}")

# PPF (Percent Point Function / Inverse CDF) - value for a given probability
val_for_50_percent = norm_dist.ppf(0.5) # Should be the mean (0)
val_for_97_5_percent = norm_dist.ppf(0.975) # Should be approx 1.96
print(f"Value at 50th percentile: {val_for_50_percent:.4f}")
print(f"Value at 97.5th percentile: {val_for_97_5_percent:.4f}")

# Generate random variates (samples)
random_samples = norm_dist.rvs(size=5, random_state=42) # random_state for reproducibility
print(f"5 random samples from standard normal: {random_samples}")

# Hypothesis Testing (Independent T-test)
# Are the means of two independent samples significantly different?
print("\nHypothesis Testing (`stats.ttest_ind`):")
group1 = np.array([20, 22, 23, 25, 24, 21])
group2 = np.array([26, 27, 28, 29, 27, 25])

t_stat, p_value = stats.ttest_ind(group1, group2, equal_var=False) # Welch's t-test
print(f"Group 1 mean: {np.mean(group1):.2f}")
print(f"Group 2 mean: {np.mean(group2):.2f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")
if p_value < 0.05: # Common significance level
    print("Conclusion: Reject the null hypothesis. Means are significantly different.")
else:
    print("Conclusion: Fail to reject the null hypothesis. Means are not significantly different.")
```

#### 3.2. `scipy.optimize` (Optimization)

The `scipy.optimize` module provides algorithms for finding optimal values of functions (minimization or maximization), root finding, and curve fitting.

**Theory:**

- **Minimization:** Find the input values that produce the minimum output of a scalar function. `minimize` is a general-purpose function that supports various algorithms (e.g., 'Nelder-Mead', 'BFGS').
- **Root Finding:** Find the roots (zeros) of a function, i.e., the values of x for which $f(x) = 0$. `fsolve` is for systems of non-linear equations.
- **Curve Fitting:** Fit a function to a set of data points, typically minimizing the sum of squared residuals.

**Code Examples:**

```python
print("\n--- 3.2. `scipy.optimize` (Optimization) ---")

# 1. Minimization of a scalar function
print("\nMinimization (`optimize.minimize`):")
# Objective function: f(x) = (x - 2)^2 + 1, minimum at x=2
def objective_function(x):
    return (x - 2)**2 + 1

# Initial guess
x0 = 0.0 # Start searching from 0

# Use 'Nelder-Mead' method for simplicity with a single variable
result_minimize = optimize.minimize(objective_function, x0, method='Nelder-Mead')

print(f"Optimization result: {result_minimize}")
print(f"Minimum found at x: {result_minimize.x[0]:.4f}")
print(f"Minimum function value: {result_minimize.fun:.4f}")

# Example with multiple variables (Rosenbrock function, common optimization benchmark)
def rosenbrock(x):
    return sum(100.0 * (x[1:] - x[:-1]**2.0)**2.0 + (1 - x[:-1])**2.0)

x0_multi = np.array([0.0, 0.0]) # Initial guess for [x, y]
result_multi = optimize.minimize(rosenbrock, x0_multi, method='BFGS') # BFGS is a good general choice
print(f"Rosenbrock minimum found at x: {result_multi.x}")
print(f"Rosenbrock minimum function value: {result_multi.fun}")


# 2. Root Finding (finding x where f(x) = 0)
print("\nRoot Finding (`optimize.fsolve`):")
# Function: f(x) = x^2 - 4, roots at x = 2 and x = -2
def func_to_find_root(x):
    return x**2 - 4

# Initial guess for the root
initial_guess_root = [1.0] # If we guess near 1, it will find 2
root = optimize.fsolve(func_to_find_root, initial_guess_root)
print(f"Root found for x^2 - 4 near initial guess {initial_guess_root}: {root[0]:.4f}")

initial_guess_root_neg = [-3.0] # If we guess near -3, it will find -2
root_neg = optimize.fsolve(func_to_find_root, initial_guess_root_neg)
print(f"Root found for x^2 - 4 near initial guess {initial_guess_root_neg}: {root_neg[0]:.4f}")

# System of equations
print("\nRoot Finding for System of Equations:")
def equations(p):
    x, y = p
    return (x + y**2 - 4, x**2 - y - 2) # Two equations: x+y^2=4, x^2-y=2

initial_guess_system = (1, 1)
solution = optimize.fsolve(equations, initial_guess_system)
print(f"Solution for system of equations (x, y): {solution}")
# Verify solution:
# x + y^2 - 4 = 2.0 + (-0.236)^2 - 4 = 0 (approx)
# x^2 - y - 2 = 2.0^2 - (-0.236) - 2 = 0 (approx)


# 3. Curve Fitting (`optimize.curve_fit`)
print("\nCurve Fitting (`optimize.curve_fit`):")
# Generate some noisy data for a linear relationship: y = ax + b
x_data = np.linspace(0, 10, 50)
true_a, true_b = 2.5, 3.0
y_data = true_a * x_data + true_b + np.random.normal(0, 1.5, size=len(x_data))

# Define the function to fit (must take x as first argument, then parameters)
def linear_function(x, a, b):
    return a * x + b

# Perform the curve fit
# p_opt: optimal parameters (a, b)
# p_cov: covariance matrix of the parameters
p_opt, p_cov = optimize.curve_fit(linear_function, x_data, y_data)

fitted_a, fitted_b = p_opt
print(f"True parameters: a={true_a}, b={true_b}")
print(f"Fitted parameters: a={fitted_a:.4f}, b={fitted_b:.4f}")

# Plotting the results (requires matplotlib)
plt.figure(figsize=(8, 5))
plt.scatter(x_data, y_data, label='Noisy Data', s=15)
plt.plot(x_data, linear_function(x_data, fitted_a, fitted_b), color='red', label='Fitted Line')
plt.title('Curve Fitting with scipy.optimize.curve_fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

#### 3.3. `scipy.interpolate` (Interpolation)

The `scipy.interpolate` module is used for estimating values between known data points.

**Theory:**
Interpolation is the process of constructing new data points within the range of a discrete set of known data points. SciPy offers various interpolation methods, including linear, cubic, and spline interpolation. `interp1d` is a common choice for 1-D interpolation.

**Code Examples:**

```python
print("\n--- 3.3. `scipy.interpolate` (Interpolation) ---")

# Data points
x_points = np.array([0, 1, 2, 3, 4, 5])
y_points = np.array([0, 0.8, 0.9, 0.1, -0.8, -1.0])

# New x-values where we want to interpolate
x_new = np.linspace(0, 5, 100) # 100 points between 0 and 5

# 1. Linear Interpolation
print("\nLinear Interpolation (`interp1d(kind='linear')`):")
linear_interp_func = interpolate.interp1d(x_points, y_points, kind='linear')
y_linear = linear_interp_func(x_new)

# 2. Cubic Spline Interpolation (smoother)
print("\nCubic Spline Interpolation (`interp1d(kind='cubic')`):")
cubic_interp_func = interpolate.interp1d(x_points, y_points, kind='cubic')
y_cubic = cubic_interp_func(x_new)

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(x_points, y_points, 'o', label='Original Data Points')
plt.plot(x_new, y_linear, '-', label='Linear Interpolation')
plt.plot(x_new, y_cubic, '--', label='Cubic Spline Interpolation')
plt.title('1-D Interpolation with scipy.interpolate')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()

# 3. UnivariateSpline (more control over smoothness)
print("\nUnivariate Spline (`scipy.interpolate.UnivariateSpline`):")
# Useful when you want to balance between fitting the data exactly and smoothing it.
# 's' is a smoothing factor. Larger 's' means more smoothing.
spline_func = interpolate.UnivariateSpline(x_points, y_points, s=0.5)
y_spline = spline_func(x_new)

plt.figure(figsize=(8, 5))
plt.plot(x_points, y_points, 'o', label='Original Data Points')
plt.plot(x_new, y_spline, '-', label='Univariate Spline (s=0.5)')
plt.title('Univariate Spline Interpolation')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

#### 4.1. `scipy.linalg` (Linear Algebra)

While NumPy provides basic linear algebra operations, `scipy.linalg` offers more advanced and specialized routines, often with better performance and numerical stability for specific problems.

**Theory:**

- **Solving Linear Systems:** Efficiently solve systems of linear equations $Ax = b$. `solve` is generally preferred over computing the inverse explicitly.
- **Determinants:** Calculate the determinant of a square matrix.
- **Eigenvalues and Eigenvectors:** Compute eigenvalues and eigenvectors of a square matrix, fundamental for many scientific applications.
- **Matrix Decompositions:** LU, Cholesky, QR, SVD decompositions are crucial for numerical stability and efficiency in complex problems.

**Code Examples:**

```python
print("\n--- 4.1. `scipy.linalg` (Linear Algebra) ---")

# Define a matrix A and a vector b
A = np.array([[3, 2, 0],
              [1, -1, 0],
              [0, 5, 1]])
b = np.array([10, 0, 5])

# 1. Solving a system of linear equations (Ax = b)
print("\nSolving Linear System (`linalg.solve`):")
# This is equivalent to x = A_inverse * b, but more numerically stable and faster.
x_solution = linalg.solve(A, b)
print(f"Matrix A:\n{A}")
print(f"Vector b: {b}")
print(f"Solution x: {x_solution}") # Should be [2, 2, -5]
print(f"Check A @ x: {A @ x_solution}") # Verify: should be close to b

# 2. Determinant of a matrix
print("\nDeterminant (`linalg.det`):")
det_A = linalg.det(A)
print(f"Determinant of A: {det_A}")

# 3. Inverse of a matrix
print("\nInverse (`linalg.inv`):")
A_inv = linalg.inv(A)
print(f"Inverse of A:\n{A_inv}")
print(f"A @ A_inv (should be identity):\n{A @ A_inv}") # Should be close to identity matrix

# 4. Eigenvalues and Eigenvectors
print("\nEigenvalues and Eigenvectors (`linalg.eig`):")
eigenvalues, eigenvectors = linalg.eig(A)
print(f"Eigenvalues: {eigenvalues}") # Can be complex numbers
print(f"Eigenvectors:\n{eigenvectors}") # Columns are eigenvectors

# Verify: A @ v = lambda * v
# For the first eigenvalue and eigenvector:
# print(f"A @ eigenvectors[:, 0]: {A @ eigenvectors[:, 0]}")
# print(f"eigenvalues[0] * eigenvectors[:, 0]: {eigenvalues[0] * eigenvectors[:, 0]}")

# 5. Singular Value Decomposition (SVD)
print("\nSingular Value Decomposition (SVD) (`linalg.svd`):")
U, s, Vh = linalg.svd(A)
print(f"U (left singular vectors):\n{U}")
print(f"s (singular values): {s}") # 1D array of singular values
print(f"Vh (right singular vectors, Hermitian conjugate):\n{Vh}")

# Reconstruct the matrix from SVD components (for verification)
# Create a diagonal matrix from singular values
Sigma = np.zeros(A.shape)
Sigma[:A.shape[0], :A.shape[0]] = np.diag(s)
A_reconstructed = U @ Sigma @ Vh
print(f"Matrix A reconstructed from SVD:\n{A_reconstructed}")
print(f"Is A_reconstructed close to A? {np.allclose(A, A_reconstructed)}")
```

#### 4.2. `scipy.integrate` (Numerical Integration)

This module provides tools for computing definite integrals and solving ordinary differential equations (ODEs).

**Theory:**

- **Quadrature:** Numerical methods for computing definite integrals of functions (`quad` for single integrals, `dblquad` for double, `tplquad` for triple, `nquad` for N-dimensional).
- **ODE Solvers:** Integrate systems of ordinary differential equations (`odeint` is a common and convenient interface to a general-purpose solver).

**Code Examples:**

```python
print("\n--- 4.2. `scipy.integrate` (Numerical Integration) ---")

# 1. Numerical Integration (Quadrature)
print("\nNumerical Integration (`integrate.quad`):")
# Integrate f(x) = x^2 from 0 to 2. Analytical solution: [x^3/3]_0^2 = 8/3 ≈ 2.6667
def func_to_integrate(x):
    return x**2

integral_result, abs_error = integrate.quad(func_to_integrate, 0, 2)
print(f"Integral of x^2 from 0 to 2: {integral_result:.4f}")
print(f"Estimated absolute error: {abs_error:.2e}")

# Example with parameters
def parameterized_func(x, a, b):
    return a * x + b

integral_param, _ = integrate.quad(parameterized_func, 0, 1, args=(2, 3)) # Integrate 2x + 3 from 0 to 1
print(f"Integral of 2x + 3 from 0 to 1: {integral_param:.4f}") # Analytical: [x^2+3x]_0^1 = 1+3 = 4


# 2. Solving Ordinary Differential Equations (ODEs)
print("\nSolving ODEs (`integrate.odeint`):")
# Consider a simple harmonic oscillator: d^2y/dt^2 + k/m * y = 0
# This can be written as a system of first-order ODEs:
# dy1/dt = y2
# dy2/dt = -(k/m) * y1
# where y1 = y (position), y2 = dy/dt (velocity)

def harmonic_oscillator_ode(y, t, k, m):
    # y is a 2-element array: y[0] is position, y[1] is velocity
    # t is time
    # k and m are parameters (spring constant, mass)
    dy1_dt = y[1]
    dy2_dt = -(k / m) * y[0]
    return [dy1_dt, dy2_dt]

# Parameters
k_val = 1.0 # Spring constant
m_val = 1.0 # Mass

# Initial conditions: y(0) = 1 (position), dy/dt(0) = 0 (velocity)
initial_conditions = [1.0, 0.0]

# Time points at which to solve the ODE
time_points = np.linspace(0, 10, 100) # From t=0 to t=10

# Solve the ODE
# Pass additional arguments (k_val, m_val) to the ODE function via args
solution = integrate.odeint(harmonic_oscillator_ode, initial_conditions, time_points, args=(k_val, m_val))

# solution[:, 0] gives position y1 over time
# solution[:, 1] gives velocity y2 over time

plt.figure(figsize=(10, 6))
plt.plot(time_points, solution[:, 0], label='Position (y)')
plt.plot(time_points, solution[:, 1], label='Velocity (dy/dt)')
plt.title('Harmonic Oscillator Solution using odeint')
plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

#### 4.3. `scipy.signal` (Signal Processing)

The `scipy.signal` module provides functions for filtering, convolution, spectral analysis, and other signal processing tasks.

**Theory:**

- **Filtering:** Applying filters to modify signal characteristics (e.g., smoothing, removing noise).
- **Convolution:** A mathematical operation that combines two signals to produce a third signal (often used in filtering, image processing).
- **Fourier Transforms (`scipy.fftpack`):** Converting signals between the time/spatial domain and the frequency domain. (Note: `scipy.fftpack` is the original, `numpy.fft` is usually sufficient for general FFTs, and Python 3.9+ also introduced `scipy.fft` which is faster and more consistent).
- **Waveform Generation:** Create basic waveforms like sine, square, sawtooth.

**Code Examples:**

```python
print("\n--- 4.3. `scipy.signal` (Signal Processing) ---")

# Generate a noisy signal
sampling_rate = 100 # Hz
t = np.linspace(0, 1, sampling_rate, endpoint=False) # 1 second of data
freq1 = 5 # Hz
freq2 = 20 # Hz
clean_signal = 0.5 * np.sin(2 * np.pi * freq1 * t) + 0.2 * np.sin(2 * np.pi * freq2 * t)
noise = np.random.normal(0, 0.1, len(t))
noisy_signal = clean_signal + noise

plt.figure(figsize=(12, 6))
plt.plot(t, noisy_signal, label='Noisy Signal')
plt.title('Original Noisy Signal')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()

# 1. Filtering (e.g., a Butterworth low-pass filter to remove high-frequency noise)
print("\nFiltering (`signal.butter`, `signal.lfilter`):")
# Design a Butterworth filter (low-pass)
cutoff_freq = 10 # Hz
nyquist_freq = 0.5 * sampling_rate # Nyquist frequency (half the sampling rate)
normalized_cutoff = cutoff_freq / nyquist_freq
order = 4 # Filter order

# Get filter coefficients
b, a = signal.butter(order, normalized_cutoff, btype='low', analog=False)

# Apply the filter to the signal
filtered_signal = signal.lfilter(b, a, noisy_signal)

plt.figure(figsize=(12, 6))
plt.plot(t, noisy_signal, label='Noisy Signal', alpha=0.7)
plt.plot(t, filtered_signal, label='Filtered Signal (Low-pass)', color='red')
plt.plot(t, clean_signal, label='Clean Signal (Original)', linestyle='--', color='green')
plt.title('Signal Filtering Example')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()

# 2. Convolution
print("\nConvolution (`signal.convolve`):")
# Simple 1D convolution: smooth a signal with a moving average kernel
sig = np.array([0, 0, 1, 1, 1, 0, 0])
kernel = np.array([0.5, 0.5]) # Averaging two adjacent points

convolved_signal = signal.convolve(sig, kernel, mode='valid') # 'valid' returns only points where kernel fully overlaps
print(f"Original signal: {sig}")
print(f"Convolution kernel: {kernel}")
print(f"Convolved signal (valid mode): {convolved_signal}") # [0.5, 1.0, 1.0, 0.5, 0.0]

# 3. Fourier Transform (using NumPy's FFT for simplicity, but scipy.fftpack is also available)
print("\nFourier Transform (using `np.fft`):")
N = len(noisy_signal)
yf = np.fft.fft(noisy_signal)
xf = np.fft.fftfreq(N, 1 / sampling_rate) # Frequencies corresponding to FFT output

# Plotting magnitude spectrum
plt.figure(figsize=(10, 6))
plt.plot(xf, np.abs(yf))
plt.title('Magnitude Spectrum of Noisy Signal')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.grid(True)
plt.xlim(0, sampling_rate / 2) # Show only positive frequencies
plt.show()

# 4. Waveform Generation (e.g., square wave)
print("\nWaveform Generation (`signal.square`):")
square_wave = signal.square(2 * np.pi * 5 * t) # 5 Hz square wave
plt.figure(figsize=(8, 4))
plt.plot(t, square_wave)
plt.title('Square Wave (5 Hz)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.ylim(-1.2, 1.2)
plt.grid(True)
plt.show()
```

```python
print("\n--- 4.4. `scipy.ndimage` (N-dimensional Image Processing) ---")

# Create a simple dummy image (2D array)
# Represents a white square on a black background with some noise
image = np.zeros((64, 64), dtype=np.float32)
image[15:45, 15:45] = 255.0 # A bright square
image += np.random.normal(0, 10, image.shape) # Add some noise
image = np.clip(image, 0, 255) # Clip values to image range

plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Noisy Image')
plt.axis('off')

# 1. Gaussian Filter (Blurring/Smoothing)
print("\nGaussian Filter (`ndimage.gaussian_filter`):")
sigma = 3 # Standard deviation for Gaussian kernel (larger means more blur)
blurred_image = ndimage.gaussian_filter(image, sigma=sigma)

plt.subplot(1, 3, 2)
plt.imshow(blurred_image, cmap='gray')
plt.title(f'Gaussian Filter (sigma={sigma})')
plt.axis('off')

# 2. Rotation
print("\nRotation (`ndimage.rotate`):")
angle = 45 # degrees
rotated_image = ndimage.rotate(image, angle, reshape=False, mode='nearest')
# reshape=False keeps original dimensions, 'nearest' fills new pixels with nearest neighbor

plt.subplot(1, 3, 3)
plt.imshow(rotated_image, cmap='gray')
plt.title(f'Rotated Image ({angle} deg)')
plt.axis('off')
plt.tight_layout()
plt.show()

# 3. Median Filter (Noise Reduction, especially salt-and-pepper noise)
print("\nMedian Filter (`ndimage.median_filter`):")
# Simulate salt-and-pepper noise
salt_pepper_image = np.copy(image)
num_salt = int(0.01 * image.size)
coords_salt = [np.random.randint(0, i - 1, num_salt) for i in salt_pepper_image.shape]
salt_pepper_image[coords_salt[0], coords_salt[1]] = 255

num_pepper = int(0.01 * image.size)
coords_pepper = [np.random.randint(0, i - 1, num_pepper) for i in salt_pepper_image.shape]
salt_pepper_image[coords_pepper[0], coords_pepper[1]] = 0

median_filtered_image = ndimage.median_filter(salt_pepper_image, size=3) # 3x3 window

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(salt_pepper_image, cmap='gray')
plt.title('Image with Salt-and-Pepper Noise')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(median_filtered_image, cmap='gray')
plt.title('Median Filtered Image (size=3)')
plt.axis('off')
plt.tight_layout()
plt.show()

# 4. Morphological Operations (Erosion and Dilation on a binary image)
print("\nMorphological Operations (`ndimage.binary_erosion`, `ndimage.binary_dilation`):")
binary_image = np.zeros((30, 30), dtype=bool)
binary_image[5:25, 5:25] = True # A solid square
binary_image[10:20, 10:20] = False # A hole in the middle

eroded_image = ndimage.binary_erosion(binary_image, iterations=2)
dilated_image = ndimage.binary_dilation(binary_image, iterations=2)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(binary_image, cmap='gray')
axes[0].set_title('Original Binary Image')
axes[0].axis('off')

axes[1].imshow(eroded_image, cmap='gray')
axes[1].set_title('Eroded Image (2 iterations)')
axes[1].axis('off')

axes[2].imshow(dilated_image, cmap='gray')
axes[2].set_title('Dilated Image (2 iterations)')
axes[2].axis('off')
plt.tight_layout()
plt.show()

# 5. Labeling Connected Components
print("\nLabeling Connected Components (`ndimage.label`):")
labeled_array, num_features = ndimage.label(dilated_image)
print(f"Number of distinct features (connected components): {num_features}")

plt.figure(figsize=(6, 6))
plt.imshow(labeled_array, cmap='nipy_spectral') # Different colors for different labels
plt.title('Labeled Connected Components')
plt.axis('off')
plt.show()
```

### 5. Performance Considerations

SciPy's strength lies in its optimized C and Fortran implementations, which handle the heavy lifting of numerical computations.

- **Vectorization:** Always prefer vectorized operations provided by NumPy and SciPy over explicit Python loops. Vectorized code is significantly faster because the loops are executed in compiled C/Fortran code.
- **Choosing the Right Tool:** SciPy functions are often wrappers around highly optimized low-level libraries (e.g., LAPACK for linear algebra, MINPACK for optimization). Using these high-level functions correctly leverages these optimizations.
- **Memory:** Be mindful of memory usage, especially with large N-dimensional arrays. While optimized, large arrays still consume significant RAM.

### 6. When to use SciPy vs. Other Libraries

- **SciPy vs. NumPy:**

  - **NumPy:** Fundamental for numerical computing, provides the `ndarray` object, basic array operations, linear algebra (dot products, basic inverse/determinant), random number generation, and basic Fourier transforms.
  - **SciPy:** Builds on NumPy, offering more specialized and advanced algorithms for scientific tasks (optimization, integration, specialized linear algebra, signal processing, statistics, etc.). If NumPy doesn't have a specific algorithm for a scientific problem, SciPy is often the next place to look.

- **SciPy vs. scikit-learn:**

  - **SciPy:** Focuses on general scientific computing algorithms, mathematical tools, and fundamental building blocks.
  - **scikit-learn:** A machine learning library built on NumPy and SciPy. It provides high-level implementations of common machine learning algorithms (classification, regression, clustering, dimensionality reduction, model selection, preprocessing). If you're doing machine learning, scikit-learn is your go-to. It often uses SciPy's underlying optimization or linear algebra routines.

- **SciPy vs. Specialized Libraries:**
  - For very specific domains (e.g., Astropy for astronomy, Biopython for bioinformatics, OpenCV for advanced computer vision), you might find more tailored and advanced functionality in specialized libraries. However, these often leverage SciPy for their numerical core.
