### **Python `datetime` Module: Overview, Concepts, and Theory**

The `datetime` module in Python provides classes for manipulating dates and times in both simple and complex ways. It allows you to work with dates, times, and time intervals, and convert between different formats. This module is essential for handling and formatting date and time data, performing arithmetic on date objects, and working with time zones.

The `datetime` module is part of Python's standard library and contains several classes and functions for handling and formatting dates and times.

---

### **Key Concepts of the `datetime` Module:**

1. **Datetime Representation:**

   - The `datetime` module provides a way to represent dates and times as objects, allowing you to store, manipulate, and perform arithmetic with date and time values.

2. **Timezone Awareness:**

   - `datetime` allows you to create "naive" and "aware" datetime objects. Naive datetime objects are unaware of time zones, while aware datetime objects include time zone information.

3. **Time Arithmetic:**
   - The module allows you to perform arithmetic with `timedelta` objects, enabling date and time manipulation (e.g., adding or subtracting days, hours, etc.).

---

### **Classes and Functions in the `datetime` Module:**

#### 1. **The `datetime` Class:**

The `datetime` class combines both date and time components (year, month, day, hour, minute, second, and microsecond) into a single object.

- **`datetime.datetime(year, month, day[, hour, minute, second, microsecond, tzinfo])`**:

  - A `datetime` object represents both date and time.
  - It can be created with the specified year, month, day, and optional hour, minute, second, microsecond, and time zone information (`tzinfo`).

  ```python
  import datetime
  dt = datetime.datetime(2025, 3, 25, 12, 45, 30)
  print(dt)  # Output: 2025-03-25 12:45:30
  ```

- **Accessing the Components:**
  - You can access the individual components of the `datetime` object (e.g., year, month, day, hour, etc.) using attributes:
  ```python
  print(dt.year)   # Output: 2025
  print(dt.month)  # Output: 3
  print(dt.day)    # Output: 25
  print(dt.hour)   # Output: 12
  print(dt.minute) # Output: 45
  ```

#### 2. **The `date` Class:**

The `date` class represents the date without the time component (year, month, and day).

- **`datetime.date(year, month, day)`**:

  - You can create a `date` object with a specific year, month, and day.

  ```python
  d = datetime.date(2025, 3, 25)
  print(d)  # Output: 2025-03-25
  ```

- **Accessing the Components:**
  - Similar to `datetime`, you can access the individual components of the `date` object (year, month, and day).
  ```python
  print(d.year)  # Output: 2025
  print(d.month) # Output: 3
  print(d.day)   # Output: 25
  ```

#### 3. **The `time` Class:**

The `time` class represents a time value (hour, minute, second, and microsecond), without a date component.

- **`datetime.time(hour, minute, second, microsecond)`**:

  - A `time` object can be created with specified hour, minute, second, and microsecond.

  ```python
  t = datetime.time(12, 45, 30)
  print(t)  # Output: 12:45:30
  ```

- **Accessing the Components:**
  - You can access the individual components of the `time` object (hour, minute, second, and microsecond).
  ```python
  print(t.hour)   # Output: 12
  print(t.minute) # Output: 45
  print(t.second) # Output: 30
  ```

#### 4. **The `timedelta` Class:**

The `timedelta` class represents a difference between two dates or times, allowing you to perform arithmetic on `datetime` and `date` objects.

- **`datetime.timedelta(days, seconds, microseconds)`**:

  - You can create a `timedelta` object with specified days, seconds, and microseconds, and perform date/time arithmetic.

  ```python
  delta = datetime.timedelta(days=5, hours=3, minutes=30)
  print(delta)  # Output: 5 days, 3:30:00
  ```

- **Arithmetic with `timedelta`:**
  - You can add or subtract `timedelta` objects to/from `datetime` or `date` objects.
  ```python
  today = datetime.date(2025, 3, 25)
  new_date = today + datetime.timedelta(days=5)
  print(new_date)  # Output: 2025-03-30
  ```

#### 5. **The `timezone` Class:**

The `timezone` class represents a fixed offset from UTC (Coordinated Universal Time), enabling timezone-aware datetime objects.

- **`datetime.timezone(offset)`**:
  - Create a `timezone` object with a specific offset (e.g., UTC+1, UTC-5).
  ```python
  tz = datetime.timezone(datetime.timedelta(hours=3))
  dt = datetime.datetime(2025, 3, 25, 12, 45, 30, tzinfo=tz)
  print(dt)  # Output: 2025-03-25 12:45:30+03:00
  ```

### **Functions in the `datetime` Module:**

#### 1. **`datetime.now(tz=None)`**:

- Returns the current local date and time as a `datetime` object.
- If the optional `tz` argument is passed, it returns a timezone-aware `datetime`.

```python
now = datetime.datetime.now()
print(now)  # Output: Current local date and time
```

#### 2. **`datetime.today()`**:

- Returns the current local date and time as a `datetime` object. This is equivalent to `datetime.now()` but without the `tz` argument.

```python
today = datetime.datetime.today()
print(today)  # Output: Current local date and time
```

#### 3. **`datetime.utcnow()`**:

- Returns the current UTC date and time as a `datetime` object.

```python
utc_now = datetime.datetime.utcnow()
print(utc_now)  # Output: Current UTC date and time
```

#### 4. **`datetime.fromtimestamp(timestamp, tz=None)`**:

- Converts a POSIX timestamp (the number of seconds since the Unix epoch) into a `datetime` object in local time.
- Optionally, a `tz` argument can be provided to create an aware `datetime` object.

```python
timestamp = 1609459200  # Example timestamp (Jan 1, 2021)
dt = datetime.datetime.fromtimestamp(timestamp)
print(dt)  # Output: 2021-01-01 00:00:00
```

#### 5. **`datetime.strptime(date_string, format)`**:

- Parses a string representing a date and time into a `datetime` object using the provided format string.

```python
dt = datetime.datetime.strptime("2025-03-25 12:45:30", "%Y-%m-%d %H:%M:%S")
print(dt)  # Output: 2025-03-25 12:45:30
```

#### 6. **`datetime.strftime(format)`**:

- Formats a `datetime` object as a string according to the given format string.

```python
formatted = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
print(formatted)  # Output: Current date and time as a string
```

### Core Classes of `datetime` Module

The `datetime` module introduces five main classes:

1.  **`date`**: Represents a date (year, month, day).
2.  **`time`**: Represents a time of day (hour, minute, second, microsecond).
3.  **`datetime`**: Represents a combination of a date and a time. This is the most commonly used class.
4.  **`timedelta`**: Represents a duration, the difference between two `date`, `time`, or `datetime` instances.
5.  **`tzinfo`**: An abstract base class for time zone information. Not directly used, but subclassed by concrete time zone implementations (like `datetime.timezone` or those from `zoneinfo`/`pytz`).

### Basic Usage

Let's explore each class with practical examples.

#### 1. `date` Objects (Year, Month, Day)

A `date` object represents a calendar date (year, month, day).

- **Creation:**
  - `date.today()`: Returns the current local date.
  - `date(year, month, day)`: Constructor for a specific date.
- **Accessors:** `.year`, `.month`, `.day`
- **Operations:** Comparison (`<`, `>`, `==`), subtraction (returns `timedelta`).

```python
import datetime

print("--- 1. date Objects ---")

# Get today's date
today = datetime.date.today()
print(f"Today's date: {today}")
print(f"Year: {today.year}, Month: {today.month}, Day: {today.day}")

# Create a specific date
christmas_2024 = datetime.date(2024, 12, 25)
print(f"Christmas 2024: {christmas_2024}")

# Date comparison
if today < christmas_2024:
    print(f"Christmas 2024 is in the future relative to {today}.")
else:
    print(f"Christmas 2024 has passed relative to {today}.")

# Date arithmetic (subtraction returns a timedelta)
days_until_christmas = christmas_2024 - today
print(f"Days until Christmas 2024: {days_until_christmas.days} days")
```

#### 2. `time` Objects (Hour, Minute, Second, Microsecond)

A `time` object represents a time of day, independent of any date. It does not carry timezone information by default (it's "naive").

- **Creation:** `time(hour, minute, second, microsecond)`
- **Accessors:** `.hour`, `.minute`, `.second`, `.microsecond`
- **Limitations:** Cannot be directly subtracted (as they don't imply a date).

```python
import datetime

print("\n--- 2. time Objects ---")

# Create a specific time
noon = datetime.time(12, 0, 0)
print(f"Noon: {noon}")
print(f"Hour: {noon.hour}, Minute: {noon.minute}, Second: {noon.second}")

# Time with microseconds
exact_time = datetime.time(10, 30, 45, 123456)
print(f"Exact time: {exact_time}")

# Comparison
later_time = datetime.time(15, 0, 0)
if noon < later_time:
    print(f"{noon} is earlier than {later_time}.")
```

#### 3. `datetime` Objects (Date + Time)

This is the most powerful and frequently used class, combining both date and time information.

- **Creation:**
  - `datetime.now()`: Current local `datetime` (may or may not have timezone info, depends on system).
  - `datetime.today()`: Similar to `now()`, but for local time only.
  - `datetime.utcnow()`: Current UTC `datetime` (naive by default).
  - `datetime.fromtimestamp(timestamp)`: Converts a Unix timestamp (seconds since epoch) to local `datetime`.
  - `datetime.fromisoformat(isostring)`: Parses an ISO 8601 formatted string.
  - `datetime(year, month, day, hour, minute, second, microsecond, tzinfo)`: Constructor.
- **Accessors:** `.year`, `.month`, `.day`, `.hour`, `.minute`, `.second`, `.microsecond`, `.tzinfo`
- **Operations:** Comparison, subtraction (returns `timedelta`).
- **Conversion to/from `date` and `time`:** `.date()`, `.time()`, `.combine()`.

```python
import datetime
import time # For time.time()

print("\n--- 3. datetime Objects ---")

# Get current datetime (local)
current_dt = datetime.datetime.now()
print(f"Current datetime: {current_dt}")

# Get current datetime (UTC)
current_utc_dt = datetime.datetime.utcnow()
print(f"Current UTC datetime (naive): {current_utc_dt}") # Note: it's naive, doesn't know it's UTC

# Create a specific datetime
specific_moment = datetime.datetime(2025, 6, 15, 9, 0, 0, 123456)
print(f"Specific moment: {specific_moment}")

# From Unix timestamp
epoch_timestamp = time.time() # Current seconds since epoch
dt_from_timestamp = datetime.datetime.fromtimestamp(epoch_timestamp)
print(f"Datetime from timestamp ({epoch_timestamp:.3f}): {dt_from_timestamp}")

# Combining date and time objects
my_date = datetime.date(2025, 7, 4)
my_time = datetime.time(17, 30, 0)
combined_dt = datetime.datetime.combine(my_date, my_time)
print(f"Combined datetime: {combined_dt}")

# Accessing date and time parts
print(f"Date part: {combined_dt.date()}")
print(f"Time part: {combined_dt.time()}")
```

#### 4. `timedelta` Objects (Durations)

A `timedelta` object represents a difference between two `date`, `time`, or `datetime` instances.

- **Creation:**
  - Result of subtraction between `date`/`datetime` objects.
  - `timedelta(days=0, seconds=0, microseconds=0, milliseconds=0, minutes=0, hours=0, weeks=0)`
- **Operations:** Addition (`+`), subtraction (`-`), multiplication (`*`), division (`/`) with other `timedelta`s or numbers.

```python
import datetime

print("\n--- 4. timedelta Objects ---")

# Create timedelta objects
one_day = datetime.timedelta(days=1)
three_hours = datetime.timedelta(hours=3)
five_minutes = datetime.timedelta(minutes=5)

print(f"One day: {one_day}")
print(f"Three hours: {three_hours}")

# Add/Subtract timedelta from datetime
future_date = datetime.datetime.now() + one_day
print(f"Tomorrow: {future_date}")

past_date = datetime.datetime.now() - datetime.timedelta(weeks=2)
print(f"Two weeks ago: {past_date}")

# Add timedelta to another timedelta
total_duration = one_day + three_hours
print(f"One day and three hours: {total_duration}")

# Multiply timedelta
ten_days = one_day * 10
print(f"Ten days: {ten_days}")

# Total seconds in a timedelta
print(f"Total seconds in {total_duration}: {total_duration.total_seconds()} seconds")
```

### Advanced Topics

#### 1. Formatting and Parsing (`strftime` and `strptime`)

- **`strftime(format)`**: Converts a `date`, `time`, or `datetime` object into a string formatted according to the provided format codes.
- **`strptime(string, format)`**: Parses a string representing a date/time into a `datetime` object, given the correct format codes.

**Common Format Codes:**

- `%Y`: Year with century (e.g., 2025)
- `%m`: Month as a zero-padded decimal number (01-12)
- `%d`: Day of the month as a zero-padded decimal number (01-31)
- `%H`: Hour (24-hour clock) as a zero-padded decimal number (00-23)
- `%M`: Minute as a zero-padded decimal number (00-59)
- `%S`: Second as a zero-padded decimal number (00-59)
- `%f`: Microsecond as a decimal number (000000-999999)
- `%A`: Weekday name (e.g., 'Monday')
- `%B`: Month name (e.g., 'January')
- `%c`: Locale’s appropriate date and time representation (e.g., 'Wed Jan 20 14:00:00 2024')
- `%x`: Locale’s appropriate date representation
- `%X`: Locale’s appropriate time representation
- `%Z`: Time zone name (e.g., 'CST', 'EST')
- `%z`: UTC offset in the form +HHMM or -HHMM
- `%j`: Day of the year as a zero-padded decimal number (001-366)
- `%%`: A literal '%' character

```python
import datetime

print("\n--- 5. Formatting and Parsing ---")

now = datetime.datetime.now()

# Formatting datetime to string
formatted_iso = now.strftime("%Y-%m-%dT%H:%M:%S")
print(f"ISO Format: {formatted_iso}")

formatted_readable = now.strftime("Today is %A, %B %d, %Y at %I:%M:%S %p")
print(f"Readable Format: {formatted_readable}")

formatted_custom = now.strftime("Log Entry: %d/%m/%y %H:%M:%S.%f")
print(f"Custom Format: {formatted_custom}")

# Parsing string to datetime
date_string = "2023-10-27 14:00:00"
format_string = "%Y-%m-%d %H:%M:%S"
parsed_dt = datetime.datetime.strptime(date_string, format_string)
print(f"Parsed datetime: {parsed_dt}")
print(f"Type of parsed_dt: {type(parsed_dt)}")

# Example of a mismatch causing ValueError
try:
    datetime.datetime.strptime("October 27, 2023", "%Y-%m-%d")
except ValueError as e:
    print(f"Error parsing (format mismatch): {e}")
```

#### 2. Time Zones (`tzinfo`, `datetime.timezone`, `zoneinfo`)

This is a critical and often complex aspect.

- **Naive vs. Aware `datetime` Objects:**

  - **Naive:** Have no explicit timezone information. `datetime.now()`, `datetime.utcnow()`, and direct constructor calls without `tzinfo` result in naive objects. They assume the local system's time or UTC, but don't _know_ it.
  - **Aware:** Include timezone information (e.g., UTC offset, timezone name). These are crucial for handling different geographical locations and daylight saving changes correctly.

- **`datetime.timezone` (Fixed Offset):**
  Use `datetime.timezone` for fixed UTC offsets (e.g., UTC+5, UTC-8). It doesn't handle daylight saving time changes.

- **`zoneinfo` (Python 3.9+) / `pytz` (Older Python):**
  For proper handling of real-world time zones (which include historical data, daylight saving rules, and arbitrary offsets), you need a third-party library like `pytz` (very common) or, if using Python 3.9+, the built-in `zoneinfo` module (which relies on `tzdata` package). `zoneinfo` is the modern, preferred way.

```python
import datetime
import zoneinfo # Requires Python 3.9+ and 'tzdata' package (pip install tzdata)

print("\n--- 6. Time Zones ---")

# 1. Naive datetime objects
naive_dt = datetime.datetime(2025, 1, 1, 10, 0, 0)
print(f"Naive datetime: {naive_dt} (tzinfo: {naive_dt.tzinfo})")

# 2. Aware datetime with fixed offset (datetime.timezone)
# Fixed UTC+6 offset for Bangladesh
bd_offset = datetime.timezone(datetime.timedelta(hours=6))
aware_dt_bd = datetime.datetime(2025, 1, 1, 10, 0, 0, tzinfo=bd_offset)
print(f"Aware datetime (Fixed +06:00): {aware_dt_bd} (tzinfo: {aware_dt_bd.tzinfo})")

# 3. Aware datetime with real time zones (using zoneinfo)
# Ensure 'tzdata' is installed (pip install tzdata) for zoneinfo to work
try:
    london_tz = zoneinfo.ZoneInfo("Europe/London")
    ny_tz = zoneinfo.ZoneInfo("America/New_York")
    tokyo_tz = zoneinfo.ZoneInfo("Asia/Tokyo")
    dhaka_tz = zoneinfo.ZoneInfo("Asia/Dhaka") # Current location context

    # Get current time in local timezone (as provided in prompt)
    current_dhaka_dt = datetime.datetime.now(dhaka_tz)
    print(f"\nCurrent time in Dhaka: {current_dhaka_dt}")

    # Create an aware datetime object for a specific timezone
    dt_london = datetime.datetime(2025, 3, 20, 10, 0, 0, tzinfo=london_tz)
    print(f"March 20, 10 AM London: {dt_london}")

    # Convert between timezones
    dt_ny = dt_london.astimezone(ny_tz)
    dt_tokyo = dt_london.astimezone(tokyo_tz)

    print(f"March 20, 10 AM London in New York: {dt_ny}")
    print(f"March 20, 10 AM London in Tokyo: {dt_tokyo}")

    # Handling Daylight Saving Time transition (e.g., London in October)
    # The timezone object automatically handles the offset change
    oct_dt_london_before_dst = datetime.datetime(2025, 10, 26, 23, 0, 0, tzinfo=london_tz)
    oct_dt_london_after_dst = datetime.datetime(2025, 10, 27, 1, 0, 0, tzinfo=london_tz)

    print(f"\nLondon before DST ends (Oct 26, 23:00): {oct_dt_london_before_dst}")
    print(f"London after DST ends (Oct 27, 01:00): {oct_dt_london_after_dst}")

except ModuleNotFoundError:
    print("\n'zoneinfo' or 'tzdata' not found. Please install 'tzdata' (pip install tzdata) for full timezone examples.")
    print("For older Python, 'pytz' (pip install pytz) is needed.")
```

#### 3. Comparisons

`date`, `time`, and `datetime` objects can be compared using standard comparison operators (`<`, `<=`, `==`, `>=`, `>`, `!=`). For `datetime` objects, comparisons between naive and aware objects might lead to `TypeError`. It's best to compare objects that are either both naive or both aware (and preferably in the same timezone for aware objects, or convert them).

```python
import datetime

print("\n--- 7. Comparisons ---")

dt1 = datetime.datetime(2025, 1, 1, 10, 0, 0)
dt2 = datetime.datetime(2025, 1, 1, 10, 0, 1)
dt3 = datetime.datetime(2025, 1, 1, 10, 0, 0)

print(f"{dt1} < {dt2}: {dt1 < dt2}")
print(f"{dt1} == {dt3}: {dt1 == dt3}")
print(f"{dt1} != {dt2}: {dt1 != dt2}")

# Comparing aware vs naive (can raise TypeError or lead to subtle bugs)
# It's generally advised to convert to a common timezone or make them both aware
# before comparison.
naive_dt = datetime.datetime(2025, 5, 29, 20, 0, 0) # 8 PM
try:
    utc_tz = datetime.timezone.utc
    aware_utc_dt = datetime.datetime(2025, 5, 29, 14, 0, 0, tzinfo=utc_tz) # 2 PM UTC = 8 PM BD (assuming +6)

    print(f"\nNaive {naive_dt} vs Aware {aware_utc_dt}:")
    # This might result in a TypeError or a comparison that's not timezone-aware
    # print(naive_dt == aware_utc_dt) # Will likely be False even if they represent the same moment
    # The comparison only works if one is UTC and the other is naive UTC-like:
    # print(datetime.datetime.utcnow().replace(tzinfo=None) == datetime.datetime.now(datetime.timezone.utc).replace(tzinfo=None))

    # Correct way to compare: make them both aware and in the same timezone
    print(f"Naive {naive_dt} (converted to UTC): {naive_dt.astimezone(utc_tz)}")
    print(f"Comparing after conversion to UTC: {naive_dt.astimezone(utc_tz) == aware_utc_dt}") # True

except Exception as e:
    print(f"Comparison of naive and aware often leads to issues: {e}")

```

#### 4. Replacing Attributes (`.replace()`)

The `.replace()` method creates a _new_ `date`, `time`, or `datetime` object with some of its attributes changed. Original object remains unchanged.

```python
import datetime

print("\n--- 8. Replacing Attributes (`.replace()`) ---")

original_dt = datetime.datetime(2025, 1, 15, 10, 30, 0)

    print(f"With Dhaka timezone: {aware_dt}")

    print("zoneinfo not available for replace(tzinfo) example.")

```

### Key Takeaways

- **`datetime` is king**: Most common use case.
- **`timedelta` for durations**: Essential for date/time arithmetic.
- **`strftime`/`strptime`**: Your go-to for string conversions. Master the format codes!
- **Naive vs. Aware**: A crucial distinction. Always prefer **aware** `datetime` objects for robust applications, especially when dealing with data from different regions or involving daylight saving.
- **Time Zones**: Use `zoneinfo` (Python 3.9+) or `pytz` for real-world time zone handling. `datetime.timezone` is only for fixed UTC offsets.
- **Immutability**: `datetime` objects are immutable. Operations like addition, subtraction, or `replace()` always return new objects.

The `datetime` module is incredibly versatile and powerful. Understanding its classes and how to use them with awareness of time zones will significantly improve the robustness of any Python application dealing with time-sensitive data.


````python
import datetime

print("--- 1. Calculating Age ---")

def calculate_age(birth_date_str):
    """Calculates age in years from a birth date string."""
    try:
        # Parse the birth date string into a date object
        birth_date = datetime.datetime.strptime(birth_date_str, "%Y-%m-%d").date()
        today = datetime.date.today()

        # Calculate age
        # A simple way: (today.year - birth_date.year) - (1 if (today.month, today.day) < (birth_date.month, birth_date.day) else 0)
        # More precise: approximate years, then adjust for full years
        
        # Calculate full years passed
        age = today.year - birth_date.year
        
        # Adjust if birthday hasn't occurred yet this year
        if (today.month, today.day) < (birth_date.month, birth_date.day):
            age -= 1
        
        print(f"Birth Date: {birth_date_str}")
        print(f"Today's Date: {today}")
        print(f"Calculated Age: {age} years")

        # You can also get a timedelta for the exact duration
        duration_since_birth = today - birth_date
        print(f"Exact duration since birth: {duration_since_birth.days} days")

    except ValueError:
        print(f"Error: Invalid date format for '{birth_date_str}'. Please use YYYY-MM-DD.")

calculate_age("1990-05-30") # Birthday today (May 30, 2025)
calculate_age("1985-11-15") # Birthday already passed this year
calculate_age("2000-07-20") # Birthday not yet passed this year
calculate_age("invalid-date")
````

````python
import datetime
import time # for time.sleep()

print("\n--- 2. Event Scheduling and Reminders ---")

def set_event_reminder(event_name, event_datetime_str):
    """
    Sets a reminder for an event.
    For simplicity, it just tells you how long until the event.
    """
    try:
        event_dt = datetime.datetime.strptime(event_datetime_str, "%Y-%m-%d %H:%M:%S")
        now = datetime.datetime.now()

        if event_dt < now:
            print(f"'{event_name}' was in the past: {event_dt}")
            return

        time_until_event = event_dt - now
        
        days = time_until_event.days
        seconds_remainder = time_until_event.seconds # Remaining seconds after days
        
        hours = seconds_remainder // 3600
        minutes = (seconds_remainder % 3600) // 60
        seconds = seconds_remainder % 60

        print(f"Event: '{event_name}' scheduled for {event_dt}")
        print(f"Time until '{event_name}': {days} days, {hours} hours, {minutes} minutes, {seconds} seconds.")

        # Simulate a real-time reminder (for short durations)
        if time_until_event.total_seconds() < 10: # Only wait if less than 10 secs
            print(f"Short reminder: waiting {time_until_event.total_seconds():.1f} seconds...")
            time.sleep(time_until_event.total_seconds())
            print(f"\n--- REMINDER: '{event_name}' IS NOW! ---")

    except ValueError:
        print(f"Error: Invalid datetime format for '{event_datetime_str}'. Use YYYY-MM-DD HH:MM:SS.")

# Example event 1: A future event
set_event_reminder("Project Deadline", "2025-06-15 17:00:00")

# Example event 2: A very near future event (if you run it quickly)
# Adjust the time below to be a few seconds in the future
near_future_time = (datetime.datetime.now() + datetime.timedelta(seconds=5)).strftime("%Y-%m-%d %H:%M:%S")
set_event_reminder("Coffee Break", near_future_time)

# Example event 3: An event in the past
set_event_reminder("Last Meeting", "2025-05-20 10:00:00")
````

````python
import datetime
import zoneinfo # Python 3.9+ for proper timezones, 'pip install tzdata'
import time # For time.time()

print("\n--- 3. Working with Timestamps and Time Zones ---")

try:
    # Define some time zones
    utc_tz = datetime.timezone.utc
    bangladesh_tz = zoneinfo.ZoneInfo("Asia/Dhaka") # Your current location
    london_tz = zoneinfo.ZoneInfo("Europe/London")
    new_york_tz = zoneinfo.ZoneInfo("America/New_York")

    # 1. Get current time in different timezones
    now_utc = datetime.datetime.now(utc_tz)
    now_dhaka = datetime.datetime.now(bangladesh_tz)
    now_london = now_dhaka.astimezone(london_tz) # Convert from Dhaka to London
    now_ny = now_dhaka.astimezone(new_york_tz)   # Convert from Dhaka to New York

    print(f"Current time (UTC): {now_utc}")
    print(f"Current time (Dhaka): {now_dhaka}")
    print(f"Current time (London): {now_london}")
    print(f"Current time (New York): {now_ny}")

    # 2. Convert Unix Timestamp to aware datetime in a specific timezone
    some_timestamp = time.time() # Get a current timestamp
    dt_from_ts_london = datetime.datetime.fromtimestamp(some_timestamp, tz=london_tz)
    print(f"\nTimestamp {some_timestamp:.3f} in London time: {dt_from_ts_london}")

    # 3. Create an aware datetime and convert it to timestamp
    specific_aware_dt = datetime.datetime(2025, 6, 1, 10, 0, 0, tzinfo=new_york_tz)
    timestamp_from_aware_dt = specific_aware_dt.timestamp()
    print(f"\nAware datetime {specific_aware_dt} converted to timestamp: {timestamp_from_aware_dt}")
    
    # 4. Handling Naive vs. Aware Conversion
    naive_dt = datetime.datetime(2025, 7, 10, 15, 0, 0) # Assumed to be local time
    print(f"\nNaive datetime: {naive_dt}")
    
    # Convert naive to aware (by assigning timezone, assumes it's in that tz)
    aware_dt_from_naive = naive_dt.replace(tzinfo=bangladesh_tz)
    print(f"Naive assigned to Dhaka TZ: {aware_dt_from_naive}")

    # Convert to UTC
    utc_from_dhaka_assigned = aware_dt_from_naive.astimezone(utc_tz)
    print(f"Converted to UTC: {utc_from_dhaka_assigned}")

except ModuleNotFoundError:
    print("\nSkipping timezone examples: 'zoneinfo' or 'tzdata' not found.")
    print("Install 'tzdata' (pip install tzdata) for `zoneinfo` to work correctly on Python 3.9+.")
    print("For older Python, use 'pytz' (pip install pytz).")
except Exception as e:
    print(f"An unexpected error occurred in timezone examples: {e}")
````

````python
import datetime

print("\n--- 4. Calculating Business Days ---")

def add_business_days(start_date_str, num_days):
    """
    Adds a specified number of business days (Mon-Fri) to a start date.
    Does not account for holidays.
    """
    try:
        current_date = datetime.datetime.strptime(start_date_str, "%Y-%m-%d").date()
        days_added = 0

        while days_added < num_days:
            current_date += datetime.timedelta(days=1)
            # Weekday: Monday is 0, Sunday is 6
            if current_date.weekday() < 5: # 0-4 are Monday-Friday
                days_added += 1
        
        print(f"Starting from {start_date_str}, adding {num_days} business days, ends on: {current_date}")
        return current_date

    except ValueError:
        print(f"Error: Invalid date format for '{start_date_str}'. Please use YYYY-MM-DD.")

# Example: Starting on a Friday, add 3 business days
add_business_days("2025-05-30", 3) # Starts Fri, then Mon, Tue, Wed

# Example: Starting on a Monday, add 5 business days
add_business_days("2025-06-02", 5)

# Example: Starting on a Saturday, add 1 business day
add_business_days("2025-05-31", 1)
````

````python
import datetime

print("--- 5. Iterating Through Dates ---")

def generate_date_range(start_date_str, end_date_str, step_days=1):
    """
    Generates dates within a given range, optionally with a step.
    Yields date objects.
    """
    try:
        start_date = datetime.datetime.strptime(start_date_str, "%Y-%m-%d").date()
        end_date = datetime.datetime.strptime(end_date_str, "%Y-%m-%d").date()
        
        if start_date > end_date:
            print(f"Warning: Start date {start_date_str} is after end date {end_date_str}. No dates generated.")
            return

        current_date = start_date
        while current_date <= end_date:
            yield current_date # Use yield to make it a generator
            current_date += datetime.timedelta(days=step_days)

    except ValueError:
        print(f"Error: Invalid date format. Please use YYYY-MM-DD.")

print("Dates for June 2025 (daily):")
for d in generate_date_range("2025-06-01", "2025-06-30"):
    print(d)

print("\nDates for Q3 2025 (weekly):")
for d in generate_date_range("2025-07-01", "2025-09-30", step_days=7):
    print(d)
````

````python
import datetime
import calendar # Python's built-in calendar module is useful here

print("\n--- 6. Finding Specific Days ---")

def get_first_weekday_of_month(year, month, weekday_num):
    """
    Finds the first occurrence of a specific weekday (0=Mon, 6=Sun) in a given month.
    """
    first_day_of_month = datetime.date(year, month, 1)
    # Calculate how many days to add to reach the target weekday
    days_to_add = (weekday_num - first_day_of_month.weekday() + 7) % 7
    first_occurrence = first_day_of_month + datetime.timedelta(days=days_to_add)
    return first_occurrence

def get_last_day_of_month(year, month):
    """
    Returns the last day of the given month.
    """
    # Option 1: Using timedelta to go to next month and subtract one day
    if month == 12:
        return datetime.date(year, 12, 31)
    else:
        return datetime.date(year, month + 1, 1) - datetime.timedelta(days=1)
    
    # Option 2: Using calendar module
    # last_day_num = calendar.monthrange(year, month)[1]
    # return datetime.date(year, month, last_day_num)


print(f"First Monday of June 2025: {get_first_weekday_of_month(2025, 6, calendar.MONDAY)}")
print(f"First Friday of July 2025: {get_first_weekday_of_month(2025, 7, calendar.FRIDAY)}")

print(f"Last day of February 2024 (leap year): {get_last_day_of_month(2024, 2)}")
print(f"Last day of February 2025 (non-leap year): {get_last_day_of_month(2025, 2)}")
print(f"Last day of May 2025: {get_last_day_of_month(2025, 5)}") # Current month
````

````python
import datetime

print("\n--- 8. Handling Different Date String Formats ---")

def parse_flexible_date(date_string, possible_formats):
    """
    Attempts to parse a date string using a list of possible formats.
    Returns datetime object or None if no format matches.
    """
    for fmt in possible_formats:
        try:
            dt = datetime.datetime.strptime(date_string, fmt)
            print(f"Successfully parsed '{date_string}' with format '{fmt}': {dt}")
            return dt
        except ValueError:
            continue # Try the next format
    
    print(f"Could not parse '{date_string}' with any of the provided formats.")
    return None

# List of common formats to try
formats_to_try = [
    "%Y-%m-%d %H:%M:%S",   # 2025-05-30 14:30:00
    "%Y/%m/%d %H:%M:%S",   # 2025/05/30 14:30:00
    "%d-%m-%Y %H:%M:%S",   # 30-05-2025 14:30:00
    "%m/%d/%Y %I:%M:%S %p",# 05/30/2025 02:30:00 PM
    "%Y-%m-%d",            # 2025-05-30
    "%d %B %Y",            # 30 May 2025
    "%B %d, %Y",           # May 30, 2025
]

parse_flexible_date("2025-05-30 10:00:00", formats_to_try)
parse_flexible_date("2024/01/15 22:05:30", formats_to_try)
parse_flexible_date("25 December 2023", formats_to_try)
parse_flexible_date("April 1, 2026", formats_to_try)
parse_flexible_date("Invalid date string", formats_to_try)
````
