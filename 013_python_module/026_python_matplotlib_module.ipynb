## 1. The Pyplot Interface (`matplotlib.pyplot`)

**Theory:**
The `pyplot` module is a collection of functions that make Matplotlib work like MATLAB. Each `pyplot` function makes some change to a figure: e.g., creates a figure, creates a plotting area in a figure, plots some lines in a plotting area, decorates the plot with labels, etc. `pyplot` keeps track of the current figure and plotting area, and the plotting functions are directed to the current axes.

**Common Syntax & Code:**

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Basic Plotting
# plt.plot() is the most common function
# It automatically creates a figure and axes if none exist.

# Line plot
plt.plot([1, 2, 3, 4], [1, 4, 2, 3])
plt.show() # Displays the figure

# 2. Adding Labels and Title (Pyplot style)
plt.plot(np.array([1, 2, 3]), np.array([4, 5, 6]))
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('My First Pyplot')
plt.show()

# 3. Customizing Line Styles, Colors, Markers
x = np.linspace(0, 10, 50)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, 'r--', label='Sine Wave') # 'r--' is a shorthand for red dashed line
plt.plot(x, y_cos, 'bo:', label='Cosine Wave') # 'bo:' is shorthand for blue circles, dotted line
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.title('Sine and Cosine Waves')
plt.legend() # Displays the legend based on 'label' arguments
plt.grid(True) # Adds a grid
plt.show()

# 4. Scatter Plot
plt.scatter([1, 2, 3, 4], [4, 1, 3, 2], color='green', marker='X', s=100, alpha=0.7)
plt.title('Scatter Plot Example')
plt.show()

# 5. Bar Chart
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 12]
plt.bar(categories, values, color=['skyblue', 'lightcoral', 'lightgreen', 'gold'])
plt.title('Bar Chart Example')
plt.show()

# 6. Histogram
data = np.random.randn(1000) # 1000 random numbers from a standard normal distribution
plt.hist(data, bins=30, color='purple', alpha=0.7, edgecolor='black')
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

# 7. Multiple Plots in One Figure (Subplots with Pyplot)
# plt.subplot(nrows, ncols, index)
plt.figure(figsize=(10, 5)) # Create a figure explicitly

plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
plt.plot(x, y_sin, 'b-')
plt.title('Sine')

plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
plt.plot(x, y_cos, 'g:')
plt.title('Cosine')

plt.suptitle('Two Subplots Example') # Figure title
plt.tight_layout() # Adjust layout to prevent overlapping
plt.show()

# 8. Saving a Figure
plt.plot([1, 2], [3, 4])
plt.title('Plot to Save')
plt.savefig('my_plot.png') # Saves as PNG
plt.savefig('my_plot.pdf', dpi=300) # Saves as PDF with 300 DPI
plt.close() # Closes the current figure to free memory if not showing
```

## 2. The Object-Oriented (OO) Interface

**Theory:**
The OO interface is the recommended approach for more complex plots because it gives you explicit control over the entire figure. You create `Figure` objects and then `Axes` objects (which are the actual plot areas). All plotting functions are then called as methods of these `Axes` objects. This makes your code more readable, maintainable, and less prone to unexpected behavior in complex scenarios.

- **`Figure` (`fig`):** The top-level container for all plot elements. It's like the canvas or the paper on which you draw. A figure can contain multiple `Axes`.
- **`Axes` (`ax` or `axs`):** This is the actual plotting area where the data is drawn. It contains the x-axis, y-axis, labels, title, legend, and all the plot elements (lines, points, bars, etc.). Crucially, one `Figure` can have multiple `Axes`, each representing a separate subplot.

**Common Syntax & Code:**

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Creating Figure and Axes
# fig: The Figure object (the whole window/canvas)
# ax: The Axes object (the plot area within the figure)
fig, ax = plt.subplots(figsize=(8, 5)) # Creates a figure and a single set of axes

# Plotting on the Axes object
ax.plot([1, 2, 3, 4], [1, 4, 2, 3], color='purple', linewidth=2, linestyle='-.')

# Setting labels and title on the Axes object
ax.set_xlabel('X-axis Label (OO)')
ax.set_ylabel('Y-axis Label (OO)')
ax.set_title('My First OO Plot')

plt.show()

# 2. Multiple Subplots with OO Interface
# plt.subplots(nrows, ncols) returns a figure and an array of axes
fig, axs = plt.subplots(2, 2, figsize=(12, 8)) # 2 rows, 2 columns of subplots

x = np.linspace(0, 10, 100)

# Plot on the first Axes (top-left)
axs[0, 0].plot(x, np.sin(x), 'b-')
axs[0, 0].set_title('Sine Wave')
axs[0, 0].set_xlabel('X')
axs[0, 0].set_ylabel('sin(X)')
axs[0, 0].grid(True)

# Plot on the second Axes (top-right)
axs[0, 1].scatter(x, np.cos(x), color='red', s=10)
axs[0, 1].set_title('Cosine Scatter')
axs[0, 1].set_xlabel('X')
axs[0, 1].set_ylabel('cos(X)')

# Plot on the third Axes (bottom-left)
categories = ['Apples', 'Bananas', 'Oranges']
values = [30, 45, 20]
axs[1, 0].bar(categories, values, color=['green', 'yellow', 'orange'])
axs[1, 0].set_title('Fruit Sales')
axs[1, 0].set_ylabel('Quantity')

# Plot on the fourth Axes (bottom-right)
data = np.random.normal(0, 1, 500)
axs[1, 1].hist(data, bins=20, color='cyan', edgecolor='black')
axs[1, 1].set_title('Histogram of Normal Data')
axs[1, 1].set_xlabel('Value')
axs[1, 1].set_ylabel('Frequency')

# Set a super title for the entire figure
fig.suptitle('Comprehensive Plot Examples (OO Interface)', fontsize=18)

# Adjust layout to prevent overlapping
plt.tight_layout(rect=[0, 0, 1, 0.96]) # Adjust rect to make space for suptitle
plt.show()

# 3. Customizing Elements on an Axes Object (More granular control)
fig, ax = plt.subplots(figsize=(7, 5))
line, = ax.plot(x, np.exp(-x/2) * np.sin(2*x), label='Damped Sine', color='teal')
ax.plot(x, np.exp(-x/2), 'r--', label='Upper Envelope', alpha=0.6)
ax.plot(x, -np.exp(-x/2), 'r--', label='Lower Envelope', alpha=0.6)

# Customizing line properties after creation
line.set_linewidth(3)
line.set_linestyle(':')
line.set_marker('o')
line.set_markersize(6)

ax.set_title('Damped Oscillation', fontsize=16, color='darkgreen', pad=15)
ax.set_xlabel('Time (s)', fontsize=12, labelpad=10)
ax.set_ylabel('Amplitude', fontsize=12)

# Customize ticks
ax.tick_params(axis='x', rotation=45, colors='gray', labelsize=10)
ax.tick_params(axis='y', colors='darkblue')

# Customizing legend
ax.legend(loc='upper right', shadow=True, fancybox=True, frameon=True, facecolor='lightyellow', edgecolor='darkgray', fontsize=10)

# Customizing grid
ax.grid(True, linestyle='--', alpha=0.6, color='lightgray')

# Customizing spines (borders)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(1.5)
ax.spines['bottom'].set_linewidth(1.5)
ax.spines['left'].set_color('blue')

# Adding text annotation
ax.text(0.5, 0.7, 'Damping Effect', transform=ax.transAxes,
        fontsize=14, color='darkred', ha='center', va='center',
        bbox=dict(boxstyle='round,pad=0.5', fc='wheat', ec='k', lw=1, alpha=0.8))

plt.tight_layout()
plt.show()
```

## **2. Basic Plotting Methods & Attributes (with Code)**

### **1. Simple Line Plot**

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create figure & axes
fig, ax = plt.subplots()

# Plotting
ax.plot(x, y,
        color='red',       # Line color
        linestyle='--',    # Line style (solid, dashed, dotted)
        linewidth=2,       # Line thickness
        marker='o',        # Marker style (o, s, ^, etc.)
        markersize=8,      # Marker size
        label='Linear')    # Legend label

# Customizations
ax.set_title("Basic Line Plot")  # Title
ax.set_xlabel("X-axis")          # X-label
ax.set_ylabel("Y-axis")          # Y-label
ax.legend()                     # Show legend
ax.grid(True)                   # Enable grid

plt.show()  # Display plot
```

### **2. Scatter Plot**

```python
import numpy as np

# Generate random data
x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)
sizes = 1000 * np.random.rand(50)

fig, ax = plt.subplots()

# Scatter plot
scatter = ax.scatter(
    x, y,
    c=colors,        # Color mapping
    s=sizes,         # Size of points
    alpha=0.7,       # Transparency (0-1)
    cmap='viridis'   # Colormap
)

# Colorbar
plt.colorbar(scatter, label="Color Intensity")

ax.set_title("Scatter Plot")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
plt.show()
```

### **3. Bar Plot**

```python
categories = ['A', 'B', 'C', 'D']
values = [15, 25, 30, 20]

fig, ax = plt.subplots()

# Bar plot
ax.bar(
    categories, values,
    color=['red', 'blue', 'green', 'orange'],
    edgecolor='black',  # Bar edge color
    linewidth=1.5,      # Edge width
    alpha=0.8          # Transparency
)

ax.set_title("Bar Plot")
ax.set_xlabel("Categories")
ax.set_ylabel("Values")
plt.show()
```

### **4. Histogram**

```python
data = np.random.normal(0, 1, 1000)  # Normal distribution

fig, ax = plt.subplots()

# Histogram
ax.hist(
    data,
    bins=30,          # Number of bins
    color='skyblue',  # Fill color
    edgecolor='black', # Bin edges
    density=True      # Normalize to PDF
)

ax.set_title("Histogram")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")
plt.show()
```

### **5. Multiple Subplots**

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))  # 2x2 grid

# Plot 1: Line plot
axes[0, 0].plot(x, y, 'r-')
axes[0, 0].set_title("Line Plot")

# Plot 2: Scatter
axes[0, 1].scatter(x, y, c='b')
axes[0, 1].set_title("Scatter Plot")

# Plot 3: Bar
axes[1, 0].bar(categories, values)
axes[1, 0].set_title("Bar Plot")

# Plot 4: Histogram
axes[1, 1].hist(data, bins=20)
axes[1, 1].set_title("Histogram")

plt.tight_layout()  # Prevent overlapping
plt.show()
```

This function is the recommended way to create a plot in Matplotlib, as it generates both the **`Figure`** and **`Axes`** objects at the same time. The `Figure` is the overall window or page, while the `Axes` is the actual plot area where you draw your data.

### Single Subplot

When you want to create a single plot, `plt.subplots()` returns a tuple containing the `Figure` object and a single `Axes` object. This is the simplest and most common use case.

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Initialize: Create the figure and a single axes object.
# `fig` is the entire canvas, `ax` is the plot area.
fig, ax = plt.subplots()

# 2. Prepare: Plot data on the axes object.
x = np.linspace(0, 10, 100)
y = np.sin(x)
ax.plot(x, y, label='sin(x)')

# 3. Customize: Set labels and a title using methods of the `ax` object.
ax.set_title('My First Plot')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.legend()

# 4. Render: Show the final plot.
plt.show()
```

---

### Multiple Subplots

`plt.subplots()` is particularly powerful for creating a grid of plots, or subplots, within a single figure. You can specify the number of rows and columns, and it returns a NumPy array of `Axes` objects.

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Initialize: Create a 2x2 grid of subplots.
# `fig` is the single figure, `axs` is a 2D array of axes.
#
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# 2. Prepare & Customize: Plot on each individual axes object.
# Access each subplot using NumPy-style indexing, e.g., `axs[row, col]`.
x = np.linspace(0, 2 * np.pi, 100)

axs[0, 0].plot(x, np.sin(x))
axs[0, 0].set_title('Sine Wave')

axs[0, 1].scatter(np.random.rand(50), np.random.rand(50))
axs[0, 1].set_title('Scatter Plot')

axs[1, 0].hist(np.random.randn(1000), bins=20)
axs[1, 0].set_title('Histogram')

axs[1, 1].boxplot(np.random.rand(50, 4))
axs[1, 1].set_title('Box Plot')

# 3. Render: Adjust layout and show.
# `plt.tight_layout()` prevents titles and labels from overlapping.
plt.tight_layout()
plt.show()
```


## 🧩 `plt.subplots()` — Create Figure and Axes Together

This function returns a **Figure** and one or more **Axes** objects, allowing you to build complex plots with clean, modular control.

```python
fig, ax = plt.subplots()
```

- `fig`: the entire figure (like a canvas)
- `ax`: the subplot (like a drawing area within the canvas)

---

## ⚙️ Key Parameters

| Parameter          | Description                                                           | Example                |
| ------------------ | --------------------------------------------------------------------- | ---------------------- |
| `nrows`, `ncols`   | Number of rows and columns of subplots                                | `plt.subplots(2, 3)`   |
| `figsize`          | Size of the figure in inches `(width, height)`                        | `figsize=(10, 5)`      |
| `sharex`, `sharey` | Share x or y axis across subplots (`True`, `False`, `'row'`, `'col'`) | `sharex=True`          |
| `squeeze`          | If `True`, returns scalars if only one subplot                        | `squeeze=False`        |
| `layout`           | Auto layout engine (`'constrained'`, `'tight'`)                       | `layout='constrained'` |

---

## 🧠 Basic Example

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot([0, 1, 2], [0, 1, 4])
ax.set_title("Single Subplot")
plt.show()
```

---

## 🔲 Multiple Subplots

```python
fig, axs = plt.subplots(2, 2, figsize=(8, 6))

axs[0, 0].plot([1, 2, 3], [1, 4, 9])
axs[0, 1].plot([1, 2, 3], [1, 2, 3])
axs[1, 0].plot([1, 2, 3], [3, 2, 1])
axs[1, 1].plot([1, 2, 3], [9, 4, 1])

plt.tight_layout()
plt.show()
```

- `axs` is a 2D NumPy array of Axes objects.
- You can index into it like `axs[row, col]`.

---

## 🔗 Shared Axes

Useful for comparing signals or harmonics across time or frequency:

```python
fig, axs = plt.subplots(2, 1, sharex=True, figsize=(6, 4))

axs[0].plot([0, 1, 2], [1, 2, 3])
axs[1].plot([0, 1, 2], [3, 2, 1])

axs[0].set_title("Shared X-axis")
plt.show()
```

---

## 🧪 Advanced: Looping Through Axes

```python
fig, axs = plt.subplots(1, 3, figsize=(12, 4))
for i, ax in enumerate(axs):
    ax.plot([0, 1, 2], [j**i for j in [0, 1, 2]])
    ax.set_title(f"Power {i}")
plt.tight_layout()
plt.show()
```
