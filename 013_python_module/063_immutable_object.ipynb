````python
# --- Immutable Objects in Python: A Deep Dive with Code ---

# In Python, an object is considered immutable if its internal state cannot be changed
# after it is created. When you perform an operation that seems to modify an immutable
# object, you are actually creating a new object with the updated value.

# Let's explore the core immutable built-in types in Python:

# -------------------- 1. Numbers (int, float, complex) --------------------

# Integers
x = 10
print(f"Initial x: {x}, id(x): {id(x)}")
y = x + 5  # Creating a new integer object
print(f"y = x + 5: {y}, id(y): {id(y)}")
print(f"x remains: {x}, id(x) still: {id(x)}")
# Notice that id(x) and id(y) are different, indicating new objects.

# Floats
a = 3.14
print(f"\nInitial a: {a}, id(a): {id(a)}")
b = a * 2
print(f"b = a * 2: {b}, id(b): {id(b)}")
print(f"a remains: {a}, id(a) still: {id(a)}")

# Complex Numbers
c = 1 + 2j
print(f"\nInitial c: {c}, id(c): {id(c)}")
d = c + 1j
print(f"d = c + 1j: {d}, id(d): {id(d)}")
print(f"c remains: {c}, id(c) still: {id(c)}")

# -------------------- 2. Strings (str) --------------------

text = "hello"
print(f"\nInitial text: '{text}', id(text): {id(text)}")
new_text = text.upper()  # Creating a new string object
print(f"new_text = text.upper(): '{new_text}', id(new_text): {id(new_text)}")
print(f"Original text remains: '{text}', id(text) still: {id(text)}")

text_concat = text + " world" # Creating another new string object
print(f"text_concat = text + ' world': '{text_concat}', id(text_concat): {id(text_concat)}")

# You cannot modify a string in place:
# text[0] = 'H'  # This will raise a TypeError

# -------------------- 3. Tuples (tuple) --------------------

my_tuple = (1, 2, 3)
print(f"\nInitial tuple: {my_tuple}, id(my_tuple): {id(my_tuple)}")
another_tuple = my_tuple + (4,)  # Creating a new tuple object
print(f"another_tuple = my_tuple + (4,): {another_tuple}, id(another_tuple): {id(another_tuple)}")
print(f"Original tuple remains: {my_tuple}, id(my_tuple) still: {id(my_tuple)}")

# You cannot modify elements within a tuple:
# my_tuple[0] = 10  # This will raise a TypeError

# However, if a tuple contains mutable objects (like lists), the *contents* of those
# mutable objects can be changed, but the tuple itself (which elements it contains) cannot.
tuple_with_list = (1, [2, 3])
print(f"\nTuple with list: {tuple_with_list}, id(tuple_with_list): {id(tuple_with_list)}")
tuple_with_list[1].append(4)  # Modifying the list *inside* the tuple
print(f"Tuple with list after modification: {tuple_with_list}, id(tuple_with_list) still: {id(tuple_with_list)}")
# The id of the tuple remains the same because the tuple's structure (which list it holds) hasn't changed.

# -------------------- 4. Frozen Sets (frozenset) --------------------

my_frozenset = frozenset({1, 2, 3})
print(f"\nInitial frozen set: {my_frozenset}, id(my_frozenset): {id(my_frozenset)}")
another_frozenset = my_frozenset.union({4})  # Creating a new frozen set
print(f"another_frozenset = my_frozenset.union({{4}}): {another_frozenset}, id(another_frozenset): {id(another_frozenset)}")
print(f"Original frozen set remains: {my_frozenset}, id(my_frozenset) still: {id(my_frozenset)}")

# You cannot add or remove elements from a frozen set:
# my_frozenset.add(4)      # AttributeError
# my_frozenset.remove(1)   # AttributeError

# -------------------- Why Immutability Matters --------------------

# 1. Predictability: Immutable objects are easier to reason about because their state
#    doesn't change unexpectedly. This makes code less prone to bugs.

# 2. Thread Safety: Multiple threads can safely access immutable objects concurrently
#    without the risk of data corruption, as no thread can modify them.

# 3. Caching and Optimization: Immutable objects can be safely cached and reused,
#    leading to potential performance improvements.

# 4. Use as Dictionary Keys and Set Elements: As seen with frozen sets, immutability
#    is a requirement for objects to be hashable, which is necessary for use as
#    keys in dictionaries and elements in sets.

# -------------------- Mutable vs. Immutable --------------------

# Mutable objects (like lists, dictionaries, and sets) can be changed in place.
my_list = [1, 2, 3]
print(f"\nMutable list: {my_list}, id(my_list): {id(my_list)}")
my_list.append(4)  # Modifies the existing list object
print(f"List after append: {my_list}, id(my_list) still: {id(my_list)}")
# Notice that the id of the list remains the same after modification.

# Understanding the distinction between mutable and immutable objects is crucial
# for writing correct and efficient Python code, especially when dealing with
# assignments, function arguments, and data structures. When you pass an immutable
# object to a function, the function cannot change the original object. However, if
# you pass a mutable object, the function can modify it in place, which might have
# side effects you need to be aware of.
````

**Core Concept: State Preservation**

At its heart, the theory of immutable objects revolves around the principle of **state preservation**. Once an immutable object is created, its internal data or the value it represents **cannot be altered** throughout its lifetime. Any operation that appears to modify an immutable object actually results in the creation of a **brand new object** with the updated value. The original object remains unchanged.

**Key Characteristics and Implications:**

1.  **Unchangeable Value:** The fundamental characteristic is that the data held within the object is fixed after instantiation. For example, once a string is created with the characters "hello," those characters and their order are permanent for that specific string object.

2.  **New Object Creation for "Modification":** When you perform an operation that seems like a modification (e.g., concatenating strings, adding to a tuple), Python doesn't change the original object. Instead, it allocates memory for a new object that holds the result of the operation. The original object persists in its initial state. This is evident when you examine the object's identity (using the `id()` function) before and after such operations – the IDs will be different for the new object.

3.  **Hashability:** Immutability is a prerequisite for an object to be **hashable**. A hashable object has a hash value that remains constant during its lifetime and can be compared to other objects for equality. This property is crucial for using objects as keys in dictionaries and elements in sets, as these data structures rely on hash values for efficient lookups and uniqueness checks. Mutable objects, whose state can change, cannot guarantee a consistent hash value and are therefore not hashable.

4.  **Predictability and Reduced Side Effects:** Because immutable objects cannot be modified after creation, they contribute to more predictable code. You can be certain that once you have a reference to an immutable object, its value will not change unexpectedly by some other part of the program. This reduces the potential for subtle bugs caused by unintended side effects.

5.  **Thread Safety:** Immutability inherently supports thread safety. Multiple threads can access and operate on immutable objects concurrently without the need for explicit locking mechanisms. Since no thread can alter the object's state, there's no risk of race conditions or data corruption.

6.  **Caching and Optimization:** Immutable objects are excellent candidates for caching and optimization. Since their value is guaranteed not to change, the system can safely store and reuse instances of immutable objects, potentially improving performance and reducing memory consumption. For instance, Python often interns small integers and short strings, reusing the same object instance whenever the same value is encountered.

7.  **Simplified Reasoning:** Working with immutable objects can simplify reasoning about program state. When you pass an immutable object to a function, you don't have to worry about the function unexpectedly modifying the original object. This makes function behavior more transparent.

**Examples of Immutable Types in Python:**

- **Numbers:** `int`, `float`, `complex`
- **Strings:** `str`
- **Tuples:** `tuple` (though they can contain mutable objects)
- **Frozen Sets:** `frozenset`
- `bool` (internally represented as integers 0 and 1)
- `None` (a singleton object)

**Contrast with Mutable Objects:**

Mutable objects, on the other hand, _can_ have their internal state changed after creation. Examples include:

- **Lists:** `list`
- **Dictionaries:** `dict`
- **Sets:** `set`

When you modify a mutable object, you are typically operating on the same object in memory (its identity remains the same). This can lead to side effects if multiple parts of your program are referencing the same mutable object.

**In Summary:**

The theory of immutable objects centers on the idea of **unchangeable state**. This property leads to significant benefits in terms of predictability, thread safety, caching, and suitability for use in hash-based data structures. By understanding which Python types are immutable and the implications of this immutability, you can write more robust, efficient, and easier-to-reason-about code. The creation of a new object for seemingly modifying an immutable object is a fundamental aspect of how Python manages these types.

class ImmutableMeta(type):
    def __call__(cls, *args, **kwargs):
        instance = super().__call__(*args, **kwargs)
        def setattr(self, name, value):
            raise AttributeError(f"Cannot modify {name} - object is immutable")
        instance.__setattr__ = setattr
        return instance

class ImmutablePoint(metaclass=ImmutableMeta):
    def __init__(self, x, y):

**I. The Foundation: What is Immutability?**

At its core, immutability means that once an object is created, its internal state (the data it holds) **cannot be changed**. Any operation that appears to modify an immutable object actually creates a **new object** with the modified value. The original object remains untouched in memory.

**II. Basic Immutable Types in Python:**

1.  **Numbers (int, float, complex):**

    ```python
    # Integers
    x = 10
    print(f"Initial x: {x}, id(x): {id(x)}")
    x = x + 5  # A new integer object is created and assigned to x
    print(f"x after addition: {x}, id(x): {id(x)}") # Notice the different id

    # Floats
    pi = 3.14
    print(f"Initial pi: {pi}, id(pi): {id(pi)}")
    pi = pi * 2
    print(f"pi after multiplication: {pi}, id(pi): {id(pi)}")

    # Complex Numbers
    z = 1 + 2j
    print(f"Initial z: {z}, id(z): {id(z)}")
    z += 1j
    print(f"z after addition: {z}, id(z): {id(z)}")
    ```

    When you perform arithmetic operations, you're not changing the original number object; you're creating a new number object that holds the result.

2.  **Strings (str):**

    ```python
    greeting = "hello"
    print(f"Initial greeting: '{greeting}', id(greeting): {id(greeting)}")
    new_greeting = greeting.upper()  # A new string object is created
    print(f"Uppercase greeting: '{new_greeting}', id(new_greeting): {id(new_greeting)}")
    print(f"Original greeting remains: '{greeting}', id(greeting): {id(greeting)}")

    concatenated = greeting + " world" # Another new string object
    print(f"Concatenated string: '{concatenated}', id(concatenated): {id(concatenated)}")

    # Attempting in-place modification raises a TypeError
    # greeting[0] = 'H'  # This will cause an error
    ```

    String methods like `upper()`, `lower()`, `replace()`, and concatenation (`+`) always return new string objects. The original string is never altered.

3.  **Tuples (tuple):**

    ```python
    my_tuple = (1, 2, 3)
    print(f"Initial tuple: {my_tuple}, id(my_tuple): {id(my_tuple)}")
    another_tuple = my_tuple + (4,)  # A new tuple object is created
    print(f"New tuple: {another_tuple}, id(another_tuple): {id(another_tuple)}")
    print(f"Original tuple remains: {my_tuple}, id(my_tuple): {id(my_tuple)}")

    # Attempting to modify an element raises a TypeError
    # my_tuple[0] = 10  # This will cause an error

    # Important Note: If a tuple contains mutable objects (like lists), the *contents*
    # of those mutable objects can be changed, but the tuple itself (which objects it
    # *references*) cannot be modified. The tuple's structure is immutable.
    tuple_with_list = (1, [2, 3])
    print(f"\nTuple with list: {tuple_with_list}, id(tuple_with_list): {id(tuple_with_list)}")
    tuple_with_list[1].append(4)  # Modifies the list *inside* the tuple
    print(f"Tuple after list modification: {tuple_with_list}, id(tuple_with_list): {id(tuple_with_list)}")
    # The id of the tuple remains the same because the tuple still references the same list object.
    ```

    Tuples are immutable sequences. Once created, you cannot add, remove, or change their elements. Operations like concatenation create new tuples.

4.  **Frozen Sets (frozenset):**

    ```python
    my_set = {1, 2, 3}
    frozen_set = frozenset(my_set)
    print(f"Initial frozen set: {frozen_set}, id(frozen_set): {id(frozen_set)}")
    another_frozen_set = frozen_set.union({4})  # Creates a new frozen set
    print(f"New frozen set after union: {another_frozen_set}, id(another_frozen_set): {id(another_frozen_set)}")
    print(f"Original frozen set remains: {frozen_set}, id(frozen_set): {id(frozen_set)}")

    # Attempting to modify a frozen set raises an AttributeError
    # frozen_set.add(4)
    # frozen_set.remove(1)
    ```

    Frozen sets are immutable versions of regular sets. They support set operations (union, intersection, difference, etc.), but these operations always return new frozen set objects.

5.  **Booleans (bool):**

    ```python
    true_val = True
    false_val = False
    print(f"True value: {true_val}, id(true_val): {id(true_val)}")
    print(f"False value: {false_val}, id(false_val): {id(false_val)}")

    # Operations on booleans create new boolean objects
    another_true = not false_val
    print(f"Not false_val: {another_true}, id(another_true): {id(another_true)}")
    ```

    Booleans in Python are also immutable. `True` and `False` are singleton instances.

6.  **None:**

    ```python
    none_val = None
    print(f"None value: {none_val}, id(none_val): {id(none_val)}")
    another_none = None
    print(f"Another None value: {another_none}, id(another_none): {id(another_none)}") # Same id
    ```

    `None` is a singleton object representing the absence of a value and is immutable.

**III. Advanced Concepts and Implications of Immutability:**

1.  **Hashability and Use as Dictionary Keys/Set Elements:**

    Immutability is a fundamental requirement for an object to be hashable. A hashable object has a hash value that remains constant throughout its lifetime and can be compared to other objects for equality. Dictionaries rely on hashable keys for efficient lookups, and sets rely on hashable elements for uniqueness checks.

    ```python
    # Mutable set cannot be a dictionary key
    # my_dict = {{1, 2}: "value"}  # TypeError: unhashable type: 'set'

    # Immutable frozenset can be a dictionary key
    valid_dict = {frozenset({1, 2}): "value"}
    print(f"Dictionary with frozenset key: {valid_dict}")

    # Mutable list cannot be an element of a set
    # my_set_of_lists = {[1, 2], [3, 4]} # TypeError: unhashable type: 'list'

    # Immutable tuple can be an element of a set
    valid_set = {(1, 2), (3, 4)}
    print(f"Set with tuple elements: {valid_set}")
    ```

2.  **Predictability and Reduced Side Effects:**

    Working with immutable objects makes code more predictable. When you pass an immutable object to a function, you can be certain that the function cannot alter the original object. This reduces the chances of unexpected side effects and makes debugging easier.

    ```python
    def modify_string(s):
        s = s.upper()
        print(f"Inside function: {s}, id(s): {id(s)}")

    text = "world"
    print(f"Before function call: {text}, id(text): {id(text)}")
    modify_string(text)
    print(f"After function call: {text}, id(text): {id(text)}") # Original string remains unchanged
    ```

3.  **Thread Safety:**

    Immutable objects are inherently thread-safe. Since their state cannot change after creation, multiple threads can access and operate on them concurrently without the risk of race conditions or data corruption. No explicit locking mechanisms are needed for immutable objects in multithreaded environments.

4.  **Caching and Optimization:**

    Python can often optimize the handling of immutable objects. For example, string interning is a technique where identical string literals are stored only once in memory. When you create another string with the same value, Python might reuse the existing object. This can save memory and improve performance. Small integers are also often interned.

    ```python
    str1 = "hello"
    str2 = "hello"
    print(f"id(str1): {id(str1)}")
    print(f"id(str2): {id(str2)}") # Might be the same due to interning

    int1 = 5
    int2 = 5
    print(f"id(int1): {id(int1)}")
    print(f"id(int2): {id(int2)}") # Likely to be the same due to interning
    ```

5.  **Functional Programming Paradigm:**

    Immutability is a core concept in functional programming. Functions that operate on immutable data tend to be pure functions (their output depends only on their input and they have no side effects), which makes code easier to reason about, test, and compose.

6.  **Shallow vs. Deep Copying (Less Relevant for Purely Immutable Types):**

    The distinction between shallow and deep copying is more relevant for mutable objects that contain other objects. For purely immutable objects, a shallow copy and a deep copy would essentially result in the same outcome (a new reference to the same immutable object, or potentially the same object due to optimization). However, when immutable collections (like tuples) contain mutable objects, the copying behavior of the inner mutable objects still matters.

    ```python
    tuple_with_list = (1, [2, 3])
    import copy
    shallow_copy = copy.copy(tuple_with_list)
    deep_copy = copy.deepcopy(tuple_with_list)

    print(f"Original tuple: {tuple_with_list}, id: {id(tuple_with_list)}, id of inner list: {id(tuple_with_list[1])}")
    print(f"Shallow copy: {shallow_copy}, id: {id(shallow_copy)}, id of inner list: {id(shallow_copy[1])}") # Same inner list id
    print(f"Deep copy: {deep_copy}, id: {id(deep_copy)}, id of inner list: {id(deep_copy[1])}")     # Different inner list id

    shallow_copy[1].append(4)
    print(f"Shallow copy after append: {shallow_copy}")
    print(f"Original tuple after shallow copy append: {tuple_with_list}") # Original's list is also changed

    deep_copy[1].append(5)
    print(f"Deep copy after append: {deep_copy}")
    print(f"Original tuple after deep copy append: {tuple_with_list}")   # Original's list is unchanged
    ```

**IV. Creating Custom Immutable Objects (Advanced):**

You can create your own immutable classes in Python by:

1.  **Defining `__slots__`:** This can prevent the dynamic creation of attributes and potentially save memory, although it doesn't inherently enforce immutability.

2.  **Overriding `__setattr__` and `__delattr__`:** By raising an `AttributeError` in these methods, you can prevent the setting or deleting of attributes after the object's initialization.

3.  **Using `@property` with no setter:** This allows you to define read-only attributes.

4.  **Careful initialization (`__init__`)**: Ensure all necessary attributes are set during initialization and not modified afterward.

```python
class ImmutablePoint:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    @property
    def x(self):
        return self._x

    @property
    def y(self):
        return self._y

    def __setattr__(self, name, value):
        if not hasattr(self, name):
            super().__setattr__(name, value)
        else:
            raise AttributeError(f"Cannot modify attribute '{name}' of immutable object.")

    def __delattr__(self, name):
        raise AttributeError("Cannot delete attributes of immutable object.")

    def __repr__(self):
        return f"ImmutablePoint(x={self.x}, y={self.y})"

point = ImmutablePoint(3, 5)
print(point)
print(point.x)
print(point.y)

# Attempting to modify attributes will raise an error
# point.x = 10  # AttributeError: Cannot modify attribute 'x' of immutable object.
# del point.y  # AttributeError: Cannot delete attributes of immutable object.

# You can still create new instances based on existing ones
new_point = ImmutablePoint(point.x + 1, point.y + 1)
print(new_point)
```
