### **Python `requests` Module: Overview, Concepts, and Theory**

The `requests` module is one of the most popular libraries in Python for making HTTP requests. It abstracts the complexities of making requests and interacting with web servers, providing a simple and human-friendly interface for sending HTTP requests to web services and receiving responses. With `requests`, you can interact with REST APIs, download data, submit forms, handle JSON data, and much more.

---

### **Key Concepts of the `requests` Module:**

1. **HTTP Methods:**

   - The `requests` module provides methods that correspond to various HTTP operations, such as `GET`, `POST`, `PUT`, `DELETE`, and more. These methods are the foundation of communication between a client (Python script) and a server.

   - **GET**: Used to retrieve data from the server.
   - **POST**: Used to send data to the server (e.g., form submissions or API requests).
   - **PUT**: Used to update data on the server.
   - **DELETE**: Used to delete data from the server.
   - **HEAD**: Similar to `GET`, but does not return the body of the response, only the headers.
   - **OPTIONS**: Used to get information about the communication options available for the target resource.

2. **Request and Response:**

   - A **request** consists of a URL, optional headers, parameters, and data sent to the server.
   - A **response** is the server's reply, which typically includes a status code (indicating success or failure), headers, and data (such as JSON, HTML, etc.).

3. **URL (Uniform Resource Locator):**

   - The URL is the web address to which you send the request. It contains a domain, path, and can also include query parameters for specific data retrieval.

4. **Status Codes:**

   - HTTP status codes are returned in the server's response to indicate the outcome of the request:
     - **200**: OK – The request was successful.
     - **404**: Not Found – The resource could not be found on the server.
     - **500**: Internal Server Error – The server encountered an error while processing the request.
     - **301**: Moved Permanently – The resource has been moved to a new URL.
     - **401**: Unauthorized – Authentication is required.
     - **403**: Forbidden – You do not have permission to access the resource.

5. **Session Handling:**

   - The `requests` module provides the `Session` object, which maintains persistent connections and session-level settings across multiple requests. It helps with handling cookies, headers, and other session-based attributes.

6. **Headers:**

   - HTTP headers are metadata sent along with the request or response to convey information, such as content type, authentication, etc. You can pass custom headers in a request to influence how the server responds.

7. **Parameters:**
   - Parameters (query parameters) are used in the URL to send additional information to the server. They are often used in `GET` requests to filter, sort, or paginate data.

### **Basic Usage:**

Here are some examples to demonstrate how to use the `requests` module.

#### 1. **Sending a GET Request:**

- The `GET` method is used to retrieve data from a specified resource.

```python
import requests

response = requests.get('https://jsonplaceholder.typicode.com/posts')
print(response.status_code)  # Prints the HTTP status code
print(response.text)         # Prints the response content as text
```

- `response.status_code`: Returns the HTTP status code (e.g., 200, 404).
- `response.text`: Returns the body of the response as a string.

#### 2. **Sending a POST Request:**

- The `POST` method is used to send data to the server, often used for submitting form data or creating resources.

```python
import requests

data = {'title': 'foo', 'body': 'bar', 'userId': 1}
response = requests.post('https://jsonplaceholder.typicode.com/posts', data=data)
print(response.status_code)
print(response.json())  # The server returns JSON data, so we use .json() to parse it
```

- `response.json()`: Automatically parses the JSON response into a Python dictionary.

#### 3. **Sending a PUT Request:**

- The `PUT` method is used to update data on the server.

```python
import requests

data = {'title': 'foo', 'body': 'bar', 'userId': 1}
response = requests.put('https://jsonplaceholder.typicode.com/posts/1', data=data)
print(response.status_code)
print(response.json())
```

#### 4. **Sending a DELETE Request:**

- The `DELETE` method is used to remove data from the server.

```python
import requests

response = requests.delete('https://jsonplaceholder.typicode.com/posts/1')
print(response.status_code)
```

### **Working with Query Parameters:**

### **Working with Headers:**

Custom headers can be passed in the request using the `headers` argument. For example, you might need to pass authentication tokens or specify the content type.

```python
import requests

headers = {'Authorization': 'Bearer your_token_here'}
response = requests.get('https://api.example.com/data', headers=headers)
print(response.status_code)
print(response.json())
```

### **Handling Timeouts:**

Requests can sometimes take too long to respond. To avoid waiting indefinitely, you can set a timeout for your request. If the request takes longer than the timeout, a `Timeout` exception is raised.

```python
import requests

try:
    response = requests.get('https://jsonplaceholder.typicode.com/posts', timeout=5)
    print(response.status_code)
except requests.Timeout:
    print("The request timed out.")
```

---

### **Handling Errors:**

You can handle common errors such as connection errors, timeouts, or invalid HTTP responses using `try-except` blocks.

```python
import requests

try:
    response = requests.get('https://jsonplaceholder.typicode.com/posts')
    response.raise_for_status()  # Raises HTTPError for bad responses (4xx or 5xx)
    print(response.status_code)
except requests.exceptions.HTTPError as errh:
    print(f"HTTP error occurred: {errh}")
except requests.exceptions.ConnectionError as errc:
    print(f"Error connecting: {errc}")
except requests.exceptions.Timeout as errt:
    print(f"Timeout error occurred: {errt}")
except requests.exceptions.RequestException as err:
    print(f"An error occurred: {err}")
```

### Python `requests` Module: Theory

The `requests` library is a **third-party Python library** (meaning it's not built into Python's standard library, you need to install it) designed to make HTTP requests simpler and more human-friendly. It handles complex underlying details like connection pooling, SSL verification, content encoding, and redirects, allowing developers to focus on interacting with web services.

**Why `requests`?**

While Python's standard library has modules like `urllib.request` for making HTTP requests, `requests` is generally preferred for several reasons:

1.  **Simplicity and Readability:** Its API is intuitive and easy to use, making HTTP operations feel natural. You send `GET` requests, `POST` requests, etc., in a way that mirrors how you'd describe them.
2.  **Batteries Included (but better):** It provides sensible defaults and handles common HTTP tasks automatically (e.g., connection pooling, GZIP compression, automatic content decoding, session management).
3.  **Robust Error Handling:** It provides clear exception types and mechanisms for handling HTTP errors.
4.  **Modern and Actively Maintained:** It follows modern best practices for web interactions.
5.  **Extensibility:** Easy to add custom headers, authentication, proxies, etc.

**Core Concepts of HTTP Requests:**

Before diving into `requests`, a brief refresher on HTTP:

- **HTTP (Hypertext Transfer Protocol):** The foundation of data communication for the World Wide Web.
- **Client-Server Model:** Your Python script (the client) sends requests to a web server, which then sends back responses.
- **URL (Uniform Resource Locator):** The address of the resource you want to access (e.g., `https://api.github.com/users/octocat`).
- **HTTP Methods (Verbs):** Define the action to be performed on the resource.
  - **GET:** Retrieve data from the server. (Most common for reading data)
  - **POST:** Send data to the server to create a new resource.
  - **PUT:** Send data to the server to update an existing resource.
  - **DELETE:** Remove a resource from the server.
  - **PATCH:** Partially update an existing resource.
  - **HEAD:** Retrieve only the headers of a resource, not the body.
  - **OPTIONS:** Retrieve the communication options available for a resource.
- **Headers:** Key-value pairs sent with requests or responses, providing metadata (e.g., `Content-Type`, `User-Agent`, `Authorization`).
- **Body/Payload:** The actual data sent with `POST`, `PUT`, `PATCH` requests.
- **Status Codes:** Three-digit numbers indicating the outcome of a request (e.g., `200 OK`, `404 Not Found`, `500 Internal Server Error`).
- **Response:** The server's reply, containing a status code, headers, and often a body.

**Key Features of `requests`:**

- **Simple API for common methods:** `requests.get()`, `requests.post()`, `requests.put()`, `requests.delete()`, etc.
- **Handling of query parameters:** Easily pass parameters in URLs using the `params` argument.
- **Sending various data types:** `data` for form-encoded data, `json` for JSON data, `files` for multipart file uploads.
- **Custom Headers:** Set `headers` dictionary.
- **Authentication:** Built-in support for Basic Auth, Digest Auth, and OAuth.
- **Timeouts:** Prevent requests from hanging indefinitely.
- **Redirects:** Automatic handling of HTTP redirects.
- **SSL Verification:** Default SSL certificate verification for secure connections.
- **Sessions:** Persist parameters across multiple requests (e.g., cookies, headers).
- **Error Handling:** `response.raise_for_status()` to raise an exception for bad status codes.

### Python `requests` Module: Code Examples

Let's explore the core functionalities with code.

```python
import requests
import json # Used for pretty-printing JSON responses
import time # For demonstrating timeouts

print("--- Python Requests Module Demonstration ---")
print("------------------------------------------\n")

# --- 1. Basic GET Request ---
print("1. Basic GET Request:")
try:
    # We'll use a public API for demonstration: JSONPlaceholder (fake online REST API)
    response = requests.get('https://jsonplaceholder.typicode.com/posts/1')

    # Check if the request was successful (status code 200)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code} (OK)")
        print(f"Response Body (JSON): {json.dumps(response.json(), indent=2)}")
        print(f"Response Headers: {response.headers['Content-Type']}")
    else:
        print(f"Request failed with status code: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 2. GET Request with Query Parameters ---
print("2. GET Request with Query Parameters:")
# To fetch posts by a specific userId
params = {
    'userId': 1,
    '_limit': 3 # Limit to 3 posts for brevity
}
try:
    response = requests.get('https://jsonplaceholder.typicode.com/posts', params=params)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print("Response Body (JSON for first 3 posts by userId 1):")
        print(json.dumps(response.json(), indent=2))
        print(f"Actual URL called: {response.url}") # Shows the URL with query params
    else:
        print(f"Request failed with status code: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 3. POST Request (Creating a Resource) ---
print("3. POST Request (Creating a Resource):")
# JSONPlaceholder will "create" a post and return it with a new ID
new_post_data = {
    'title': 'foo',
    'body': 'bar',
    'userId': 1
}
try:
    response = requests.post('https://jsonplaceholder.typicode.com/posts', json=new_post_data)

    # POST requests often return 201 Created on success
    if response.status_code == 201:
        print(f"Status Code: {response.status_code} (Created)")
        print(f"New Post Created: {json.dumps(response.json(), indent=2)}")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 4. PUT Request (Updating a Resource) ---
print("4. PUT Request (Updating a Resource):")
# Updating an existing post (e.g., post with ID 1)
updated_post_data = {
    'id': 1, # Often included, though URL specifies it
    'title': 'updated title',
    'body': 'updated body',
    'userId': 1
}
try:
    response = requests.put('https://jsonplaceholder.typicode.com/posts/1', json=updated_post_data)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code} (OK)")
        print(f"Post Updated: {json.dumps(response.json(), indent=2)}")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 5. PATCH Request (Partially Updating a Resource) ---
print("5. PATCH Request (Partially Updating a Resource):")
# Updating only specific fields of a post (e.g., post with ID 1)
patch_data = {
    'title': 'patched title only'
}
try:
    response = requests.patch('https://jsonplaceholder.typicode.com/posts/1', json=patch_data)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code} (OK)")
        print(f"Post Patched: {json.dumps(response.json(), indent=2)}")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 6. DELETE Request (Deleting a Resource) ---
print("6. DELETE Request (Deleting a Resource):")
# Deleting a post (e.g., post with ID 1)
try:
    response = requests.delete('https://jsonplaceholder.typicode.com/posts/1')
    # DELETE often returns 200 OK or 204 No Content
    if response.status_code == 200 or response.status_code == 204:
        print(f"Status Code: {response.status_code}")
        print("Post deleted successfully (or indicated as such by API).")
        print(f"Response body (often empty for DELETE): '{response.text}'")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 7. Custom Headers ---
print("7. Custom Headers:")
headers = {
    'User-Agent': 'MyPythonApp/1.0',
    'Accept': 'application/json',
    'X-Custom-Header': 'HelloFromPython'
}
try:
    response = requests.get('https://httpbin.org/headers', headers=headers)
    # httpbin.org is a useful service to test HTTP requests
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print("Headers sent by client:")
        print(json.dumps(response.json()['headers'], indent=2))
    else:
        print(f"Request failed with status code: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 8. Timeouts ---
print("8. Timeouts:")
# Prevents a request from hanging indefinitely.
# timeout= (connect timeout, read timeout) or a single float for both.
try:
    # This URL introduces a delay of 3 seconds
    response = requests.get('https://httpbin.org/delay/3', timeout=2) # Set timeout to 2 seconds
    print(f"Request successful: {response.status_code}")
except requests.exceptions.Timeout as e:
    print(f"Request timed out after 2 seconds: {e}")
except requests.exceptions.RequestException as e:
    print(f"An general request error occurred: {e}")

try:
    # This request should succeed within the timeout
    response = requests.get('https://httpbin.org/delay/1', timeout=2)
    print(f"Request completed within 2 seconds: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 9. Handling Redirects ---
print("9. Handling Redirects:")
# By default, requests follows redirects. `allow_redirects=False` to disable.
try:
    # httpbin.org/redirect/1 redirects once
    response = requests.get('https://httpbin.org/redirect/1')
    print(f"Default behavior (follow redirects):")
    print(f"Final URL: {response.url}")
    print(f"Status Code: {response.status_code}") # Will be 200 OK after following
    print(f"Redirect history: {response.history}") # Shows list of redirect responses
    print("\n")

    response_no_redirect = requests.get('https://httpbin.org/redirect/1', allow_redirects=False)
    print(f"Disabled redirects (allow_redirects=False):")
    print(f"Status Code: {response_no_redirect.status_code}") # Will be 302 Found
    print(f"Redirect history: {response_no_redirect.history}") # Will be empty
    print(f"Location header for redirect: {response_no_redirect.headers.get('Location')}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 10. Sessions (Persisting Parameters Across Requests) ---
print("10. Sessions:")
# A Session object allows you to persist certain parameters across requests.
# It's useful for maintaining cookies, custom headers, and connection pooling.
# This makes requests more efficient, especially when making many requests to the same host.

session = requests.Session()

# Set a common header for all requests in this session
session.headers.update({'X-Session-ID': 'my-unique-session-token'})

# Let's use httpbin for cookies
print("First request (sets a cookie in the session):")
try:
    # This URL sets a cookie "mycookie=myvalue"
    response1 = session.get('https://httpbin.org/cookies/set?mycookie=myvalue')
    print(f"Status Code: {response1.status_code}")
    print(f"Cookies in session after first request: {session.cookies.get('mycookie')}")
    print(f"Response from server: {json.dumps(response1.json(), indent=2)}")
except requests.exceptions.RequestException as e:
    print(f"Session request 1 error: {e}")

time.sleep(1) # Small delay for demonstration

print("\nSecond request (sends the cookie from the session):")
try:
    # This URL reflects the cookies it receives
    response2 = session.get('https://httpbin.org/cookies')
    print(f"Status Code: {response2.status_code}")
    print(f"Cookies sent by session in second request: {json.dumps(response2.json(), indent=2)}")
    print(f"Headers sent in second request (check for X-Session-ID):\n{json.dumps(response2.request.headers, indent=2)}")
except requests.exceptions.RequestException as e:
    print(f"Session request 2 error: {e}")

# Always close the session to release resources
session.close()
print("-" * 40 + "\n")


# --- 11. Error Handling with raise_for_status() ---
print("11. Error Handling with `raise_for_status()`:")
# `response.raise_for_status()` raises an HTTPError for bad responses (4xx or 5xx).
try:
    # This will result in a 404 Not Found error
    response = requests.get('https://jsonplaceholder.typicode.com/nonexistent-path')
    response.raise_for_status() # This will raise an HTTPError
    print("Request successful (this line won't be reached for 404)")
except requests.exceptions.HTTPError as e:
    print(f"HTTP Error caught: {e}")
    print(f"Response content: {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error caught: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error caught: {e}")
except requests.exceptions.RequestException as e:
    # This catches all other requests-related errors
    print(f"An general RequestException occurred: {e}")
print("-" * 40 + "\n")


# --- 12. Sending Different Types of Data with POST/PUT ---
print("12. Sending Different Types of Data:")

# Sending form-encoded data (content-type: application/x-www-form-urlencoded)
print("Sending form-encoded data (`data` parameter):")
form_data = {'key1': 'value1', 'key2': 'value2'}
try:
    response = requests.post('https://httpbin.org/post', data=form_data)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print(f"Form Data received by server: {json.dumps(response.json()['form'], indent=2)}")
    else:
        print(f"Request failed: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
print("-" * 40 + "\n")

# Sending raw JSON data (content-type: application/json)
print("Sending JSON data (`json` parameter):")
json_payload = {'name': 'test_user', 'settings': {'theme': 'dark', 'notifications': True}}
try:
    response = requests.post('https://httpbin.org/post', json=json_payload)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print(f"JSON Data received by server: {json.dumps(response.json()['json'], indent=2)}")
        print(f"Content-Type header sent: {response.request.headers.get('Content-Type')}")
    else:
        print(f"Request failed: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
print("-" * 40 + "\n")

# --- 13. File Uploads (Multipart-Encoded) ---
print("13. File Uploads (`files` parameter):")

# Create a dummy file for upload
with open('test_file.txt', 'w') as f:
    f.write('This is a test file for upload.\n')
    f.write('It contains multiple lines of text.')

try:
    # `files` parameter takes a dictionary where values can be file-like objects,
    # or tuples of (filename, file_object), (filename, file_object, content_type), etc.
    with open('test_file.txt', 'rb') as f: # Open in binary read mode
        files = {'upload_file': f}
        response = requests.post('https://httpbin.org/post', files=files)
        if response.status_code == 200:
            print(f"Status Code: {response.status_code}")
            print(f"File upload successful. Server received:")
            print(json.dumps(response.json()['files'], indent=2))
            print(f"Form data (if any): {json.dumps(response.json()['form'], indent=2)}")
        else:
            print(f"Request failed: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"Error during file upload: {e}")
finally:
    import os
    if os.path.exists('test_file.txt'):
        os.remove('test_file.txt') # Clean up the dummy file
print("-" * 40 + "\n")


print("--- End of Requests Module Demo ---")
```

- **Simple API for common methods:** `requests.get()`, `requests.post()`, `requests.put()`, `requests.delete()`, etc.
- **Handling of query parameters:** Easily pass parameters in URLs using the `params` argument.
- **Sending various data types:** `data` for form-encoded data, `json` for JSON data, `files` for multipart file uploads.
- **Custom Headers:** Set `headers` dictionary.
- **Authentication:** Built-in support for Basic Auth, Digest Auth, and OAuth.
- **Timeouts:** Prevent requests from hanging indefinitely.
- **Redirects:** Automatic handling of HTTP redirects.
- **SSL Verification:** Default SSL certificate verification for secure connections.
- **Sessions:** Persist parameters across multiple requests (e.g., cookies, headers).
- **Error Handling:** `response.raise_for_status()` to raise an exception for bad status codes.

### Python `requests` Module: Code Examples

Let's explore the core functionalities with code.

```python
import requests
import json # Used for pretty-printing JSON responses
import time # For demonstrating timeouts

print("--- Python Requests Module Demonstration ---")
print("------------------------------------------\n")

# --- 1. Basic GET Request ---
print("1. Basic GET Request:")
try:
    # We'll use a public API for demonstration: JSONPlaceholder (fake online REST API)
    response = requests.get('https://jsonplaceholder.typicode.com/posts/1')

    # Check if the request was successful (status code 200)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code} (OK)")
        print(f"Response Body (JSON): {json.dumps(response.json(), indent=2)}")
        print(f"Response Headers: {response.headers['Content-Type']}")
    else:
        print(f"Request failed with status code: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 2. GET Request with Query Parameters ---
print("2. GET Request with Query Parameters:")
# To fetch posts by a specific userId
params = {
    'userId': 1,
    '_limit': 3 # Limit to 3 posts for brevity
}
try:
    response = requests.get('https://jsonplaceholder.typicode.com/posts', params=params)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print("Response Body (JSON for first 3 posts by userId 1):")
        print(json.dumps(response.json(), indent=2))
        print(f"Actual URL called: {response.url}") # Shows the URL with query params
    else:
        print(f"Request failed with status code: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 3. POST Request (Creating a Resource) ---
print("3. POST Request (Creating a Resource):")
# JSONPlaceholder will "create" a post and return it with a new ID
new_post_data = {
    'title': 'foo',
    'body': 'bar',
    'userId': 1
}
try:
    response = requests.post('https://jsonplaceholder.typicode.com/posts', json=new_post_data)

    # POST requests often return 201 Created on success
    if response.status_code == 201:
        print(f"Status Code: {response.status_code} (Created)")
        print(f"New Post Created: {json.dumps(response.json(), indent=2)}")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 4. PUT Request (Updating a Resource) ---
print("4. PUT Request (Updating a Resource):")
# Updating an existing post (e.g., post with ID 1)
updated_post_data = {
    'id': 1, # Often included, though URL specifies it
    'title': 'updated title',
    'body': 'updated body',
    'userId': 1
}
try:
    response = requests.put('https://jsonplaceholder.typicode.com/posts/1', json=updated_post_data)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code} (OK)")
        print(f"Post Updated: {json.dumps(response.json(), indent=2)}")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 5. PATCH Request (Partially Updating a Resource) ---
print("5. PATCH Request (Partially Updating a Resource):")
# Updating only specific fields of a post (e.g., post with ID 1)
patch_data = {
    'title': 'patched title only'
}
try:
    response = requests.patch('https://jsonplaceholder.typicode.com/posts/1', json=patch_data)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code} (OK)")
        print(f"Post Patched: {json.dumps(response.json(), indent=2)}")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 6. DELETE Request (Deleting a Resource) ---
print("6. DELETE Request (Deleting a Resource):")
# Deleting a post (e.g., post with ID 1)
try:
    response = requests.delete('https://jsonplaceholder.typicode.com/posts/1')
    # DELETE often returns 200 OK or 204 No Content
    if response.status_code == 200 or response.status_code == 204:
        print(f"Status Code: {response.status_code}")
        print("Post deleted successfully (or indicated as such by API).")
        print(f"Response body (often empty for DELETE): '{response.text}'")
    else:
        print(f"Request failed with status code: {response.status_code}")
        print(f"Error details: {response.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 7. Custom Headers ---
print("7. Custom Headers:")
headers = {
    'User-Agent': 'MyPythonApp/1.0',
    'Accept': 'application/json',
    'X-Custom-Header': 'HelloFromPython'
}
try:
    response = requests.get('https://httpbin.org/headers', headers=headers)
    # httpbin.org is a useful service to test HTTP requests
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print("Headers sent by client:")
        print(json.dumps(response.json()['headers'], indent=2))
    else:
        print(f"Request failed with status code: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 8. Timeouts ---
print("8. Timeouts:")
# Prevents a request from hanging indefinitely.
# timeout= (connect timeout, read timeout) or a single float for both.
try:
    # This URL introduces a delay of 3 seconds
    response = requests.get('https://httpbin.org/delay/3', timeout=2) # Set timeout to 2 seconds
    print(f"Request successful: {response.status_code}")
except requests.exceptions.Timeout as e:
    print(f"Request timed out after 2 seconds: {e}")
except requests.exceptions.RequestException as e:
    print(f"An general request error occurred: {e}")

try:
    # This request should succeed within the timeout
    response = requests.get('https://httpbin.org/delay/1', timeout=2)
    print(f"Request completed within 2 seconds: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 9. Handling Redirects ---
print("9. Handling Redirects:")
# By default, requests follows redirects. `allow_redirects=False` to disable.
try:
    # httpbin.org/redirect/1 redirects once
    response = requests.get('https://httpbin.org/redirect/1')
    print(f"Default behavior (follow redirects):")
    print(f"Final URL: {response.url}")
    print(f"Status Code: {response.status_code}") # Will be 200 OK after following
    print(f"Redirect history: {response.history}") # Shows list of redirect responses
    print("\n")

    response_no_redirect = requests.get('https://httpbin.org/redirect/1', allow_redirects=False)
    print(f"Disabled redirects (allow_redirects=False):")
    print(f"Status Code: {response_no_redirect.status_code}") # Will be 302 Found
    print(f"Redirect history: {response_no_redirect.history}") # Will be empty
    print(f"Location header for redirect: {response_no_redirect.headers.get('Location')}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred during the request: {e}")
print("-" * 40 + "\n")


# --- 10. Sessions (Persisting Parameters Across Requests) ---
print("10. Sessions:")
# A Session object allows you to persist certain parameters across requests.
# It's useful for maintaining cookies, custom headers, and connection pooling.
# This makes requests more efficient, especially when making many requests to the same host.

session = requests.Session()

# Set a common header for all requests in this session
session.headers.update({'X-Session-ID': 'my-unique-session-token'})

# Let's use httpbin for cookies
print("First request (sets a cookie in the session):")
try:
    # This URL sets a cookie "mycookie=myvalue"
    response1 = session.get('https://httpbin.org/cookies/set?mycookie=myvalue')
    print(f"Status Code: {response1.status_code}")
    print(f"Cookies in session after first request: {session.cookies.get('mycookie')}")
    print(f"Response from server: {json.dumps(response1.json(), indent=2)}")
except requests.exceptions.RequestException as e:
    print(f"Session request 1 error: {e}")

time.sleep(1) # Small delay for demonstration

print("\nSecond request (sends the cookie from the session):")
try:
    # This URL reflects the cookies it receives
    response2 = session.get('https://httpbin.org/cookies')
    print(f"Status Code: {response2.status_code}")
    print(f"Cookies sent by session in second request: {json.dumps(response2.json(), indent=2)}")
    print(f"Headers sent in second request (check for X-Session-ID):\n{json.dumps(response2.request.headers, indent=2)}")
except requests.exceptions.RequestException as e:
    print(f"Session request 2 error: {e}")

# Always close the session to release resources
session.close()
print("-" * 40 + "\n")


# --- 11. Error Handling with raise_for_status() ---
print("11. Error Handling with `raise_for_status()`:")
# `response.raise_for_status()` raises an HTTPError for bad responses (4xx or 5xx).
try:
    # This will result in a 404 Not Found error
    response = requests.get('https://jsonplaceholder.typicode.com/nonexistent-path')
    response.raise_for_status() # This will raise an HTTPError
    print("Request successful (this line won't be reached for 404)")
except requests.exceptions.HTTPError as e:
    print(f"HTTP Error caught: {e}")
    print(f"Response content: {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error caught: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error caught: {e}")
except requests.exceptions.RequestException as e:
    # This catches all other requests-related errors
    print(f"An general RequestException occurred: {e}")
print("-" * 40 + "\n")


# --- 12. Sending Different Types of Data with POST/PUT ---
print("12. Sending Different Types of Data:")

# Sending form-encoded data (content-type: application/x-www-form-urlencoded)
print("Sending form-encoded data (`data` parameter):")
form_data = {'key1': 'value1', 'key2': 'value2'}
try:
    response = requests.post('https://httpbin.org/post', data=form_data)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print(f"Form Data received by server: {json.dumps(response.json()['form'], indent=2)}")
    else:
        print(f"Request failed: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
print("-" * 40 + "\n")

# Sending raw JSON data (content-type: application/json)
print("Sending JSON data (`json` parameter):")
json_payload = {'name': 'test_user', 'settings': {'theme': 'dark', 'notifications': True}}
try:
    response = requests.post('https://httpbin.org/post', json=json_payload)
    if response.status_code == 200:
        print(f"Status Code: {response.status_code}")
        print(f"JSON Data received by server: {json.dumps(response.json()['json'], indent=2)}")
        print(f"Content-Type header sent: {response.request.headers.get('Content-Type')}")
    else:
        print(f"Request failed: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
print("-" * 40 + "\n")

# --- 13. File Uploads (Multipart-Encoded) ---
print("13. File Uploads (`files` parameter):")

# Create a dummy file for upload
with open('test_file.txt', 'w') as f:
    f.write('This is a test file for upload.\n')
    f.write('It contains multiple lines of text.')

try:
    # `files` parameter takes a dictionary where values can be file-like objects,
    # or tuples of (filename, file_object), (filename, file_object, content_type), etc.
    with open('test_file.txt', 'rb') as f: # Open in binary read mode
        files = {'upload_file': f}
        response = requests.post('https://httpbin.org/post', files=files)
        if response.status_code == 200:
            print(f"Status Code: {response.status_code}")
            print(f"File upload successful. Server received:")
            print(json.dumps(response.json()['files'], indent=2))
            print(f"Form data (if any): {json.dumps(response.json()['form'], indent=2)}")
        else:
            print(f"Request failed: {response.status_code}")
except requests.exceptions.RequestException as e:
    print(f"Error during file upload: {e}")
finally:
    import os
    if os.path.exists('test_file.txt'):
        os.remove('test_file.txt') # Clean up the dummy file
print("-" * 40 + "\n")


print("--- End of Requests Module Demo ---")
```


## HTTP Methods in the `requests` Module

HTTP methods (also known as HTTP verbs) define the type of action to be performed on a resource identified by a URL. The `requests` library in Python provides convenient functions for making requests using all the standard HTTP methods.

Here's a breakdown of the most common HTTP methods, their theory, and code examples using the `requests` module:

---

### 1. GET

**Theory:**
The `GET` method is used to _retrieve_ data from a server. It's considered a "safe" and "idempotent" method.

- **Safe:** It doesn't cause any changes on the server.
- **Idempotent:** Making the same `GET` request multiple times will have the same effect as making it once (i.e., it will always retrieve the same data, assuming the data on the server hasn't changed independently).

`GET` requests typically send parameters in the URL's query string.

**When to Use:**

- Fetching a web page.
- Retrieving a specific resource (e.g., an image, a document).
- Querying an API for data.

**Code Example:**

```python
import requests

# Example 1: Basic GET request
url = "https://www.google.com"
response = requests.get(url)

print(f"GET Request to: {url}")
print(f"Status Code: {response.status_code}")
print(f"Content Length: {len(response.text)} bytes")
# print(response.text[:200]) # Print first 200 characters of the response content

print("-" * 30)

# Example 2: GET with query parameters
api_url = "https://jsonplaceholder.typicode.com/posts"
params = {'userId': 1, 'id': 1} # Get post with userId 1 and id 1
response_with_params = requests.get(api_url, params=params)

print(f"GET Request to: {api_url} with params: {params}")
print(f"Status Code: {response_with_params.status_code}")
print("Response JSON:")
print(response_with_params.json())

print("-" * 30)

# Example 3: GET with headers
headers = {'User-Agent': 'MyCustomApp/1.0'}
response_with_headers = requests.get("https://httpbin.org/headers", headers=headers)

print(f"GET Request to: https://httpbin.org/headers with custom headers")
print(f"Status Code: {response_with_headers.status_code}")
print("Response JSON (showing headers):")
print(response_with_headers.json())
```

---

### 2. POST

**Theory:**
The `POST` method is used to _submit_ data to be processed to a specified resource. It's typically used for creating new resources or sending data that needs to be stored or processed on the server.

`POST` is neither safe nor idempotent. Repeated `POST` requests can result in duplicate resources or unintended side effects. Data is sent in the request body, not in the URL.

**When to Use:**

- Submitting a form (e.g., creating a new user, posting a comment).
- Uploading a file.
- Sending data to an API to create a new record.

**Code Example:**

```python
import requests
import json

# Example 1: Basic POST request with JSON data
api_url = "https://jsonplaceholder.typicode.com/posts"
new_post_data = {
    'title': 'foo',
    'body': 'bar',
    'userId': 1,
}

response = requests.post(api_url, json=new_post_data)

print(f"POST Request to: {api_url}")
print(f"Sent Data: {new_post_data}")
print(f"Status Code: {response.status_code}")
print("Response JSON (newly created resource):")
print(response.json())

print("-" * 30)

# Example 2: POST with form-encoded data
form_url = "https://httpbin.org/post"
form_data = {
    'name': 'John Doe',
    'email': 'john.doe@example.com'
}

response_form = requests.post(form_url, data=form_data)

print(f"POST Request to: {form_url} with form data: {form_data}")
print(f"Status Code: {response_form.status_code}")
print("Response JSON (showing form data):")
print(response_form.json())
```

---

### 3. PUT

**Theory:**
The `PUT` method is used to _update_ an existing resource or _create_ a new resource if it doesn't exist at the specified URL. It is idempotent: sending the same `PUT` request multiple times will have the same effect as sending it once (it will always result in the resource being in the same state).

**When to Use:**

- Updating a complete resource.
- Creating a resource at a specific URL if it doesn't already exist.

**Code Example:**

```python
import requests
import json

# Example: PUT request to update a resource
# Assuming a post with id=1 already exists
api_url = "https://jsonplaceholder.typicode.com/posts/1"
updated_post_data = {
    'id': 1,
    'title': 'Updated Title',
    'body': 'This is the updated body of the post.',
    'userId': 1
}

response = requests.put(api_url, json=updated_post_data)

print(f"PUT Request to: {api_url}")
print(f"Sent Data: {updated_post_data}")
print(f"Status Code: {response.status_code}")
print("Response JSON (updated resource):")
print(response.json())
```

---

### 4. DELETE

**Theory:**
The `DELETE` method is used to _delete_ a specified resource. It is idempotent: deleting a resource multiple times will have the same effect as deleting it once (the resource will remain deleted).

**When to Use:**

- Removing a specific resource from the server.

**Code Example:**

```python
import requests

# Example: DELETE request to remove a resource
# Assuming a post with id=1 already exists and we want to delete it
api_url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.delete(api_url)

print(f"DELETE Request to: {api_url}")
print(f"Status Code: {response.status_code}")
if response.status_code == 200:
    print("Resource successfully deleted.")
elif response.status_code == 204: # No Content, also indicates success
    print("Resource successfully deleted (no content returned).")
else:
    print(f"Failed to delete resource. Response: {response.text}")
```

### 5. HEAD

**Theory:**
The `HEAD` method is similar to `GET`, but it only requests the headers of a resource, not the actual body. This is useful for checking if a resource exists, its size, last modified date, or content type, without downloading the entire content. It's both safe and idempotent.

**When to Use:**

- Checking if a resource is available.
- Determining the size or last modification date of a file before downloading.
- Verifying content type.

**Code Example:**

```python
import requests

# Example: HEAD request to get headers
url = "https://www.google.com"
response = requests.head(url)

print(f"HEAD Request to: {url}")
print(f"Status Code: {response.status_code}")
print("Response Headers:")
for header, value in response.headers.items():
    print(f"  {header}: {value}")
```

---

### 6. OPTIONS

**Theory:**
The `OPTIONS` method is used to describe the communication options for the target resource. It allows a client to determine the HTTP methods and other options that are supported by a web server or an API endpoint, without actually performing a request to the resource. This is crucial for Cross-Origin Resource Sharing (CORS) preflight requests.

**When to Use:**

- Discovering supported HTTP methods for a URL.
- For CORS preflight requests (handled automatically by browsers).

**Code Example:**

```python
import requests

# Example: OPTIONS request to see allowed methods
url = "https://jsonplaceholder.typicode.com/posts" # A resource that supports multiple methods
response = requests.options(url)

print(f"OPTIONS Request to: {url}")
print(f"Status Code: {response.status_code}")
print("Allowed Methods (from 'Allow' header):")
print(response.headers.get('Allow'))
```

---

### 7. PATCH

**Theory:**
The `PATCH` method is used to apply partial modifications to a resource. Unlike `PUT`, which replaces the entire resource, `PATCH` only sends the changes that need to be made. It is not necessarily idempotent.

**When to Use:**

- Updating a specific field of a resource without sending the entire resource data.

**Code Example:**

```python
import requests
import json

# Example: PATCH request to partially update a resource
# Assuming a post with id=1 exists and we only want to change its title
api_url = "https://jsonplaceholder.typicode.com/posts/1"
patch_data = {
    'title': 'Patched Title Only'
}

response = requests.patch(api_url, json=patch_data)

print(f"PATCH Request to: {api_url}")
print(f"Sent Data: {patch_data}")
print(f"Status Code: {response.status_code}")
print("Response JSON (partially updated resource):")
print(response.json())
```

## HTTP GET Method in the `requests` Module

The `GET` method is one of the most fundamental and frequently used HTTP methods. It's designed for **retrieving data** from a specified resource on a server.

### Theory of the GET Method

1.  **Purpose: Data Retrieval**

    - The primary goal of a `GET` request is to fetch information from a server. It's like asking a server, "Give me the data located at this URL."

2.  **Safety:**

    - `GET` is considered a "safe" method. This means that a `GET` request, by definition, **should not alter the state of the server or the resource**. It should only read data.
    - If a `GET` request causes a change on the server (e.g., triggering a database update), it's generally considered a design flaw and violates the HTTP specification.

3.  **Idempotence:**

    - `GET` is an "idempotent" method. This means that making the same `GET` request multiple times will have the **exact same effect** on the server as making it once.
    - For example, requesting a web page 10 times will always return the same page content (assuming the page itself hasn't been independently updated by someone else).

4.  **No Request Body:**

    - `GET` requests typically **do not have a request body**. Any data sent to the server (like parameters for filtering or searching) is appended to the URL as a **query string**.
    - A query string starts with a `?` followed by key-value pairs separated by `&` (e.g., `https://example.com/api/products?category=electronics&price_max=500`).

5.  **Caching:**

    - Because `GET` requests are safe and idempotent, their responses are often **cacheable**. This means that browsers or proxy servers can store the response and serve it directly for subsequent identical `GET` requests, improving performance and reducing server load.

6.  **Usage Scenarios:**
    - Accessing a regular website page (e.g., `www.google.com`).
    - Fetching a specific article or image.
    - Searching for information using query parameters.
    - Retrieving data from a REST API (e.g., getting a list of users, fetching details of a single user).

### `requests.get()` in Python

The `requests` library in Python provides a simple and intuitive way to perform `GET` requests using the `requests.get()` function.

#### Basic Syntax:

```python
import requests

response = requests.get(url, params=None, headers=None, timeout=None, **kwargs)
```

- `url`: The URL of the resource you want to retrieve.
- `params` (optional): A dictionary of query parameters to be added to the URL.
- `headers` (optional): A dictionary of HTTP headers to send with the request.
- `timeout` (optional): A float or tuple specifying how long to wait for the server to send data before giving up.
- `**kwargs`: Additional options like authentication, proxies, SSL verification, etc.

### Code Examples

Let's illustrate the `GET` method with various practical examples using the `requests` module.

```python
import requests

print("--- Example 1: Basic GET Request to a Website ---")
# This is the simplest form, just retrieving the content of a web page.
url_basic = "https://www.example.com"
try:
    response_basic = requests.get(url_basic)

    print(f"URL: {url_basic}")
    print(f"Status Code: {response_basic.status_code}") # 200 usually means success
    print(f"Content Type: {response_basic.headers.get('Content-Type')}")
    print(f"Content Length: {len(response_basic.text)} characters")
    # print("\nFirst 300 characters of content:")
    # print(response_basic.text[:300]) # Print a snippet of the HTML content
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 2: GET Request with Query Parameters (API Example) ---")
# Using a public API to fetch posts by a specific user.
# Query parameters are passed using the 'params' dictionary.
api_url_params = "https://jsonplaceholder.typicode.com/posts"
query_parameters = {
    'userId': 5, # Fetch all posts by user ID 5
    'id': 45    # And specifically post ID 45 (this will effectively narrow it down to one)
}
try:
    response_params = requests.get(api_url_params, params=query_parameters)

    print(f"URL: {response_params.url} (Notice the query string added by requests)")
    print(f"Status Code: {response_params.status_code}")
    if response_params.status_code == 200:
        data = response_params.json() # Parse JSON response into a Python list/dict
        print("Response JSON (first item):")
        if data:
            print(data[0])
        else:
            print("No data found for the given parameters.")
    else:
        print(f"Error fetching data: {response_params.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 3: GET Request with Custom Headers ---")
# Sometimes you need to send custom headers, e.g., User-Agent, Accept, Authorization.
api_url_headers = "https://httpbin.org/headers" # An endpoint that echoes back request headers
custom_headers = {
    'User-Agent': 'MyPythonApp/1.0',
    'Accept': 'application/json, text/plain',
    'X-Custom-Header': 'Hello from Python'
}
try:
    response_headers = requests.get(api_url_headers, headers=custom_headers)

    print(f"URL: {api_url_headers}")
    print(f"Status Code: {response_headers.status_code}")
    if response_headers.status_code == 200:
        print("Response JSON (showing sent headers):")
        print(response_headers.json()['headers']) # httpbin.org returns headers under 'headers' key
    else:
        print(f"Error fetching data: {response_headers.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 4: Handling Errors and Timeouts ---")
# Demonstrating error handling and using the timeout parameter.
# This URL will intentionally cause a 404 Not Found error.
bad_url = "https://jsonplaceholder.typicode.com/nonexistent-path"
timeout_url = "https://www.google.com" # Using a reliable URL for timeout test

try:
    # Attempt to fetch a non-existent resource
    response_error = requests.get(bad_url)
    print(f"URL: {bad_url}")
    print(f"Status Code (expected 404): {response_error.status_code}")
    # You can raise an HTTPError for bad responses (4xx or 5xx)
    response_error.raise_for_status() # This will raise an HTTPError if status code is not 2xx
except requests.exceptions.HTTPError as e:
    print(f"HTTP Error for {bad_url}: {e}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error for {bad_url}: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error for {bad_url}: {e}")
except requests.exceptions.RequestException as e:
    print(f"An unknown error occurred for {bad_url}: {e}")

print("-" * 30)

try:
    # Demonstrate timeout
    print(f"Attempting to fetch {timeout_url} with a 0.001 second timeout...")
    response_timeout = requests.get(timeout_url, timeout=0.001)
    print(f"Status Code: {response_timeout.status_code}") # This line might not be reached if timeout occurs
except requests.exceptions.Timeout:
    print("Request timed out as expected!")
except requests.exceptions.RequestException as e:
    print(f"An error occurred during timeout test: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 5: Downloading Binary Content (Image) ---")
# GET requests are also used for downloading files like images or documents.
image_url = "https://www.python.org/static/community_logos/python-logo-only.png"
try:
    response_image = requests.get(image_url, stream=True) # Use stream=True for large files
    response_image.raise_for_status() # Check for bad status codes

    with open("python_logo.png", "wb") as f:
        for chunk in response_image.iter_content(chunk_size=8192):
            f.write(chunk)
    print(f"Image downloaded successfully to python_logo.png from {image_url}")
except requests.exceptions.RequestException as e:
    print(f"Error downloading image: {e}")

```

## HTTP POST Method in the `requests` Module

The `POST` method is a fundamental HTTP verb used for **sending data to a server to create or update a resource**. Unlike `GET`, where data is in the URL, `POST` sends data primarily in the **request body**.

### Theory of the POST Method

1.  **Purpose: Data Submission/Creation**

    - The primary goal of a `POST` request is to submit data to be processed by a specified resource. This often leads to a **change in the server's state**, such as:
      - Creating a new record in a database (e.g., a new user, a new blog post).
      - Uploading a file.
      - Submitting a form on a website.
      - Sending data to an API endpoint for some action (e.g., processing a payment).

2.  **Not Safe:**

    - `POST` is **not considered a "safe" method**. This means that a `POST` request _can and often does_ alter the state of the server or the resource. Performing a `POST` request might have side effects.

3.  **Not Idempotent:**

    - `POST` is **not an "idempotent" method**. This means that making the same `POST` request multiple times can have different or additional effects compared to making it once.
    - For example, if you `POST` data to create a new user, sending the exact same `POST` request twice will likely create two distinct user accounts (unless the server has specific logic to prevent duplicates). This is a key difference from `GET` and `PUT`.

4.  **Request Body:**

    - The data being sent with a `POST` request is typically included in the **request body**, not as part of the URL (like with `GET`).
    - This allows for sending larger amounts of data and more complex data structures (like JSON or form data).

5.  **Data Formats:**

    - `POST` requests can send data in various formats, indicated by the `Content-Type` header:
      - `application/x-www-form-urlencoded`: Standard HTML form submission data.
      - `multipart/form-data`: Used for file uploads, where the body is divided into parts.
      - `application/json`: Common for API interactions, sending data as a JSON string.
      - `text/plain`: Plain text.
      - `application/xml`: XML data.

6.  **Responses:**
    - A successful `POST` request typically returns a `200 OK`, `201 Created` (if a new resource was created), or `204 No Content` (if the request was successful but no content is returned).
    - If a new resource is created, the response often includes a `Location` header pointing to the URL of the newly created resource.

### `requests.post()` in Python

The `requests` library in Python provides the `requests.post()` function for performing `POST` requests.

#### Basic Syntax:

```python
import requests

response = requests.post(url, data=None, json=None, files=None, headers=None, timeout=None, **kwargs)
```

- `url`: The URL of the resource to which you want to send data.
- `data` (optional):
  - A dictionary, list of tuples, bytes, or file-like object for sending form-encoded data (`application/x-www-form-urlencoded`).
  - If a dictionary is provided, `requests` will automatically form-encode it.
- `json` (optional):
  - A Python dictionary or list to send as JSON in the request body (`application/json`). `requests` will automatically serialize it to a JSON string and set the `Content-Type` header.
- `files` (optional):
  - A dictionary of file objects to send as `multipart/form-data` (for file uploads).
- `headers` (optional): A dictionary of HTTP headers to send with the request.
- `timeout` (optional): A float or tuple specifying how long to wait for the server to send data.
- `**kwargs`: Additional options like authentication, proxies, SSL verification, etc.

### Code Examples

Let's illustrate the `POST` method with various practical examples using the `requests` module.

```python
import requests
import json # Used for pretty-printing JSON responses

print("--- Example 1: Basic POST Request with Form-Encoded Data ---")
# Often used for submitting HTML forms.
# requests automatically sets Content-Type: application/x-www-form-urlencoded
form_submission_url = "https://httpbin.org/post" # An endpoint that echoes back POST data
payload_form = {
    'name': 'Alice Smith',
    'email': 'alice.smith@example.com',
    'message': 'Hello from requests!'
}
try:
    response_form = requests.post(form_submission_url, data=payload_form)

    print(f"URL: {form_submission_url}")
    print(f"Status Code: {response_form.status_code}")
    if response_form.status_code == 200:
        print("Response JSON (showing 'form' data):")
        # httpbin.org puts form data under the 'form' key in its JSON response
        print(json.dumps(response_form.json()['form'], indent=2))
        print(f"Content-Type in request headers (sent by requests): {response_form.request.headers.get('Content-Type')}")
    else:
        print(f"Error: {response_form.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 2: POST Request with JSON Data (API Example) ---")
# This is extremely common for interacting with modern REST APIs.
# requests automatically sets Content-Type: application/json
api_creation_url = "https://jsonplaceholder.typicode.com/posts" # A public API for creating posts
new_post_data = {
    'title': 'My New Post from Python',
    'body': 'This is the content of my newly created post via requests.',
    'userId': 101
}
try:
    response_json = requests.post(api_creation_url, json=new_post_data)

    print(f"URL: {api_creation_url}")
    print(f"Sent Data (as JSON): {json.dumps(new_post_data)}")
    print(f"Status Code: {response_json.status_code}") # Expecting 201 Created
    if response_json.status_code == 201:
        print("Response JSON (newly created resource details):")
        print(json.dumps(response_json.json(), indent=2))
        print(f"Content-Type in request headers (sent by requests): {response_json.request.headers.get('Content-Type')}")
    else:
        print(f"Error creating post: {response_json.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 3: POST Request with File Upload (multipart/form-data) ---")
# Used for sending files (e.g., images, documents) to a server.
# requests automatically handles Content-Type: multipart/form-data
upload_url = "https://httpbin.org/post" # An endpoint that echoes back POST data
file_path = "test_upload.txt"

# Create a dummy file for testing
with open(file_path, "w") as f:
    f.write("This is a test file for upload.\n")
    f.write("It contains multiple lines of text.\n")
print(f"Created dummy file: {file_path}")

try:
    with open(file_path, 'rb') as f_obj: # Open in binary read mode
        files = {'upload_file': f_obj} # 'upload_file' is the name of the form field for the file
        data = {'description': 'A file uploaded from a Python script'} # Can also send other form data

        response_file = requests.post(upload_url, files=files, data=data)

        print(f"URL: {upload_url}")
        print(f"Status Code: {response_file.status_code}")
        if response_file.status_code == 200:
            print("Response JSON (showing 'files' and 'form' data):")
            response_data = response_file.json()
            print(f"Uploaded file name: {response_data['files'].get('upload_file')}")
            print(f"Description field: {response_data['form'].get('description')}")
            print(f"Content-Type in request headers (sent by requests): {response_file.request.headers.get('Content-Type').split(';')[0]}") # Only show main type
        else:
            print(f"Error uploading file: {response_file.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
finally:
    import os
    if os.path.exists(file_path):
        os.remove(file_path) # Clean up the dummy file
        print(f"Cleaned up dummy file: {file_path}")

print("\n" + "="*50 + "\n")

print("--- Example 4: POST with Custom Headers and Error Handling ---")
# Combining custom headers with POST, and robust error handling.
api_with_auth = "https://httpbin.org/post" # Using httpbin to echo headers
payload_auth = {'action': 'login', 'username': 'testuser'}
custom_headers_auth = {
    'Content-Type': 'application/json', # Explicitly setting, though requests.json() would do it
    'Authorization': 'Bearer a_secret_token_123',
    'Accept': 'application/json'
}
try:
    response_auth = requests.post(api_with_auth, json=payload_auth, headers=custom_headers_auth, timeout=5) # 5-second timeout
    response_auth.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

    print(f"URL: {api_with_auth}")
    print(f"Status Code: {response_auth.status_code}")
    print("Response JSON (showing sent headers and JSON data):")
    print(json.dumps(response_auth.json(), indent=2))

except requests.exceptions.HTTPError as e:
    print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error: {e}")
except requests.exceptions.RequestException as e:
    print(f"An unexpected requests error occurred: {e}")

```

### Key Takeaways for `POST` Requests with `requests`:

- **`data` vs. `json`:**
  - Use `data=your_dict` for `application/x-www-form-urlencoded` (standard web forms).
  - Use `json=your_dict` for `application/json` (most modern APIs). `requests` handles JSON serialization and setting the `Content-Type` header automatically.
- **File Uploads:** Use the `files` parameter for `multipart/form-data` uploads. Remember to open files in binary mode (`'rb'`).
- **Headers:** Always pay attention to the `Content-Type` header, especially if you're interacting with an API that expects a specific format. `requests` often sets this automatically when `json` or `files` are used.

- **Idempotency:** Remember that `POST` is _not_ idempotent. Be careful about repeating `POST` requests unless you intend to create duplicate resources.
- **Server Response:** Pay attention to the response status code (`201 Created` is common for successful resource creation) and the response body, which often contains details of the newly created resource.

Understanding and effectively using the `POST` method is critical for building applications that interact with web services, create content, or manage resources on a server.


## HTTP PUT Method in the `requests` Module

The `PUT` method is an HTTP verb used for **modifying an existing resource** on a server, or **creating a new resource** if one does not already exist at the specified URI.

### Theory of the PUT Method

1.  **Purpose: Complete Replacement or Creation**

    - The primary goal of a `PUT` request is to replace the _entire content_ of a target resource with the content provided in the request body.
    - If the resource identified by the URI does not exist, `PUT` can create it (acting as a "create or update" operation). The client determines the URI of the resource.

2.  **Safety:**

    - `PUT` is **not considered a "safe" method**. This means that a `PUT` request _will_ alter the state of the server or the resource.

3.  **Idempotence:**

    - `PUT` is an **"idempotent" method**. This is a crucial characteristic. It means that making the same `PUT` request multiple times will have the **exact same effect** on the server as making it once.
    - If you `PUT` data to a resource at a specific URL, sending that exact same `PUT` request again will simply overwrite the resource with the same data, leaving the server in the same state as after the first request. It won't create duplicates.

4.  **Request Body:**

    - Like `POST`, `PUT` requests include the data to be sent in the **request body**. This data represents the _complete new state_ of the resource.

5.  **Data Formats:**

    - `PUT` requests commonly send data in formats like `application/json`, `application/xml`, or other content types, depending on the API's requirements. The `Content-Type` header is vital.

6.  **Usage Scenarios:**

    - **Updating a resource:** Replacing all fields of an existing record (e.g., updating all details of a user profile).
    - **Creating a resource with a client-determined URI:** If the client dictates the exact URL where the resource should reside (e.g., `PUT /users/john_doe`). This is less common in typical REST APIs, where `POST` to a collection (`/users`) is usually preferred for creation, and the server assigns the ID.

7.  **Responses:**
    - A successful `PUT` request typically returns:
      - `200 OK` or `204 No Content` if an existing resource was successfully updated.
      - `201 Created` if a new resource was successfully created.
      - Other codes like `400 Bad Request`, `404 Not Found`, etc., for errors.

### `requests.put()` in Python

The `requests` library in Python provides the `requests.put()` function for performing `PUT` requests.

#### Basic Syntax:

```python
import requests

response = requests.put(url, data=None, json=None, headers=None, timeout=None, **kwargs)
```

- `url`: The URL of the resource you want to update or create. This URL should point directly to the specific resource, often including its ID (e.g., `/api/users/123`).
- `data` (optional):
  - A dictionary, list of tuples, bytes, or file-like object for sending form-encoded data or raw body content.
- `json` (optional):
  - A Python dictionary or list to send as JSON in the request body (`application/json`). `requests` will automatically serialize it to a JSON string and set the `Content-Type` header to `application/json`. This is the most common way to send data with `PUT` to modern APIs.
- `headers` (optional): A dictionary of HTTP headers to send with the request.
- `timeout` (optional): A float or tuple specifying how long to wait for the server to send data.
- `**kwargs`: Additional options like authentication, proxies, SSL verification, etc.

### Code Examples

Let's illustrate the `PUT` method with various practical examples using the `requests` module.

```python
import requests
import json

# Define a base URL for our examples (using a public API that supports PUT)
# jsonplaceholder.typicode.com/posts/{id} allows PUT to update posts
BASE_API_URL = "https://jsonplaceholder.typicode.com/posts"

print("--- Example 1: Updating an Existing Resource (Common Use Case) ---")
# We'll update post with ID 1.
# The data sent here should represent the complete new state of the resource.
post_id_to_update = 1
update_url = f"{BASE_API_URL}/{post_id_to_update}"

updated_post_data = {
    'id': post_id_to_update, # It's good practice to include the ID in the body for PUT
    'title': 'My Fully Updated Post Title',
    'body': 'This is the completely new content for post 1, replacing everything old.',
    'userId': 1 # Assuming this post belonged to userId 1
}

try:
    print(f"Attempting to PUT to: {update_url}")
    print(f"Sending data: {json.dumps(updated_post_data, indent=2)}")
    response_update = requests.put(update_url, json=updated_post_data)

    print(f"Status Code: {response_update.status_code}")
    if response_update.status_code == 200: # 200 OK is typical for successful PUT update
        print("Response JSON (showing updated resource):")
        print(json.dumps(response_update.json(), indent=2))
        print(f"Content-Type in request headers: {response_update.request.headers.get('Content-Type')}")
    else:
        print(f"Error updating resource: {response_update.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 2: Creating a Resource with PUT (Less Common, Client-Assigned ID) ---")
# Some APIs allow PUT to create a resource if the URI doesn't exist.
# For jsonplaceholder, if you PUT to a non-existent ID, it will return a 200 OK with the data,
# but it doesn't actually create a *persistent* resource on the server.
# This demonstrates the client-assigned ID concept.
new_post_id = 9999 # A large, likely non-existent ID
create_with_put_url = f"{BASE_API_URL}/{new_post_id}"

new_post_data_put = {
    'id': new_post_id,
    'title': 'Post Created with PUT (Client-Assigned ID)',
    'body': 'This content is sent to a specific, client-determined URI.',
    'userId': 500
}

try:
    print(f"Attempting to PUT to: {create_with_put_url}")
    print(f"Sending data: {json.dumps(new_post_data_put, indent=2)}")
    response_create = requests.put(create_with_put_url, json=new_post_data_put)

    print(f"Status Code: {response_create.status_code}") # jsonplaceholder returns 200 for this
    if response_create.status_code in [200, 201]: # 201 Created is typical for new resource
        print("Response JSON (showing created/echoed resource):")
        print(json.dumps(response_create.json(), indent=2))
        if response_create.status_code == 201:
            print(f"Resource created at: {response_create.headers.get('Location', 'N/A')}")
    else:
        print(f"Error creating resource with PUT: {response_create.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 3: PUT with Custom Headers and Error Handling ---")
# Essential for APIs requiring authentication or specific content types.
# Using httpbin.org/put to echo back headers and data.
httpbin_put_url = "https://httpbin.org/put"
data_with_headers = {'status': 'active', 'version': 2}
custom_headers = {
    'Content-Type': 'application/json',
    'Authorization': 'Bearer YOUR_AUTH_TOKEN', # Replace with actual token if needed
    'X-Client-ID': 'my-python-app'
}

try:
    print(f"Attempting to PUT to: {httpbin_put_url}")
    print(f"Sending data: {json.dumps(data_with_headers, indent=2)}")
    print(f"Sending headers: {json.dumps(custom_headers, indent=2)}")

    response_headers = requests.put(httpbin_put_url, json=data_with_headers, headers=custom_headers, timeout=5)
    response_headers.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

    print(f"Status Code: {response_headers.status_code}")
    print("Response JSON (echoing request details):")
    echoed_data = response_headers.json()
    print(f"Echoed JSON data: {json.dumps(echoed_data.get('json'), indent=2)}")
    print(f"Echoed headers: {json.dumps(echoed_data.get('headers'), indent=2)}")

except requests.exceptions.HTTPError as e:
    print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error: {e}")
except requests.exceptions.RequestException as e:
    print(f"An unexpected requests error occurred: {e}")

```

### Key Takeaways for `PUT` Requests with `requests`:

- **Complete Replacement:** Remember that `PUT` is meant for _replacing_ the entire resource. If you send a `PUT` request with only a few fields, and the server strictly adheres to `PUT` semantics, those few fields will become the _only_ fields in the resource, and any others will be deleted. For partial updates, use `PATCH`.
- **Idempotency:** This is `PUT`'s defining characteristic. You can safely retry a `PUT` request if it fails, knowing it won't create duplicate resources.
- **Client-Controlled URI:** `PUT` is often used when the client already knows the exact URI of the resource it wants to create or update.
- **`json` Parameter:** For sending JSON data (most common with REST APIs), the `json` parameter is the most convenient way; `requests` handles the serialization and `Content-Type` header automatically.
- **Error Handling:** Crucial, as network issues or server-side validation can lead to errors. Always use `try...except` and `response.raise_for_status()`.
- **Response Codes:** Be aware of the common success codes: `200 OK` for updates and `201 Created` for new resource creation.

The `PUT` method is essential for operations where you need to ensure a resource's state exactly matches the data you provide, making it a cornerstone of robust API design and interaction.


## HTTP DELETE Method in the `requests` Module

The `DELETE` method is an HTTP verb used for **removing a specific resource** identified by a URI from the server.

### Theory of the DELETE Method

1.  **Purpose: Resource Removal**

    - The primary goal of a `DELETE` request is to ask the server to remove the resource identified by the request-URI. It's like telling the server, "Get rid of this specific thing."

2.  **Safety:**

    - `DELETE` is **not considered a "safe" method**. This is because performing a `DELETE` request directly changes the state of the server by removing a resource.

3.  **Idempotence:**

    - `DELETE` is an **"idempotent" method**. This is a crucial characteristic. It means that making the same `DELETE` request multiple times will have the **exact same effect** on the server as making it once.
    - If you `DELETE` a resource, and then send the exact same `DELETE` request again, the resource will still be gone (or remain gone). You won't "delete it twice" or cause additional side effects beyond the initial removal. The server's state regarding that resource remains the same (it's absent).

4.  **No Request Body (Typically):**

    - `DELETE` requests typically **do not have a request body**. The resource to be deleted is fully identified by the URL. While the HTTP specification _allows_ a body, it's very rarely used and often ignored by servers, as the URI is supposed to be the sole identifier.

5.  **Usage Scenarios:**

    - Deleting a user account.
    - Removing a specific file or document.
    - Deleting an item from a shopping cart.
    - Removing a blog post or comment.

6.  **Responses:**
    - A successful `DELETE` request typically returns:
      - `200 OK` (if the response includes an entity describing the status).
      - `202 Accepted` (if the action has been queued but not yet performed).
      - `204 No Content` (most common and preferred, indicating success without returning any response body).
      - Other codes like `404 Not Found` (if the resource to be deleted doesn't exist), `403 Forbidden`, etc., for errors.

### `requests.delete()` in Python

The `requests` library in Python provides the `requests.delete()` function for performing `DELETE` requests.

#### Basic Syntax:

```python
import requests

response = requests.delete(url, headers=None, timeout=None, **kwargs)
```

- `url`: The URL of the specific resource you want to delete. This URL _must_ identify the precise resource.

### Code Examples

Let's illustrate the `DELETE` method with various practical examples using the `requests` module.

```python
import requests
import json # Not strictly needed for DELETE, but useful for context/error responses

# Define a base URL for our examples (using a public API that supports DELETE)
# jsonplaceholder.typicode.com/posts/{id} allows DELETE to remove posts
BASE_API_URL = "https://jsonplaceholder.typicode.com/posts"

print("--- Example 1: Deleting an Existing Resource ---")
# We'll attempt to delete post with ID 1.
# Note: jsonplaceholder's DELETE is simulated; it won't actually remove data persistently.
post_id_to_delete = 1
delete_url_1 = f"{BASE_API_URL}/{post_id_to_delete}"

try:
    print(f"Attempting to DELETE resource at: {delete_url_1}")
    response_delete_1 = requests.delete(delete_url_1)

    print(f"Status Code: {response_delete_1.status_code}")
    if response_delete_1.status_code == 200: # jsonplaceholder returns 200 OK for successful DELETE
        print(f"Resource {post_id_to_delete} simulated deletion successful. Response body (if any): {response_delete_1.text}")
    elif response_delete_1.status_code == 204: # 204 No Content is also common for successful DELETE
        print(f"Resource {post_id_to_delete} deleted (No Content returned).")
    elif response_delete_1.status_code == 404:
        print(f"Resource {post_id_to_delete} not found. Already deleted or never existed.")
    else:
        print(f"Error deleting resource: {response_delete_1.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 2: Demonstrating Idempotence (Deleting the Same Resource Again) ---")
# Sending the same DELETE request again should yield the same result.
# For jsonplaceholder, it will still return 200 OK, indicating the operation
# (simulated deletion) is idempotent. In a real API, it might return 404 the second time
# if the resource was genuinely removed.
post_id_to_delete_again = 1
delete_url_2 = f"{BASE_API_URL}/{post_id_to_delete_again}"

try:
    print(f"Attempting to DELETE resource again at: {delete_url_2}")
    response_delete_2 = requests.delete(delete_url_2)

    print(f"Status Code: {response_delete_2.status_code}")
    if response_delete_2.status_code == 200:
        print(f"Resource {post_id_to_delete_again} simulated deletion (again) successful. Response body (if any): {response_delete_2.text}")
    elif response_delete_2.status_code == 204:
        print(f"Resource {post_id_to_delete_again} deleted (No Content returned again).")
    elif response_delete_2.status_code == 404:
        print(f"Resource {post_id_to_delete_again} not found after previous deletion (expected for real API).")
    else:
        print(f"Error deleting resource the second time: {response_delete_2.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 3: Handling Deletion of a Non-Existent Resource ---")
# Attempting to delete a resource that doesn't exist should typically result in 404.
non_existent_id = 99999
delete_non_existent_url = f"{BASE_API_URL}/{non_existent_id}"

try:
    print(f"Attempting to DELETE non-existent resource at: {delete_non_existent_url}")
    response_non_existent = requests.delete(delete_non_existent_url)

    print(f"Status Code: {response_non_existent.status_code}")
    if response_non_existent.status_code == 404:
        print(f"Successfully received 404 Not Found for non-existent resource {non_existent_id}.")
    elif response_non_existent.status_code == 200 or response_non_existent.status_code == 204:
         print(f"Unexpected success for non-existent resource. Check API behavior. Response: {response_non_existent.text}")
    else:
        print(f"Unexpected error when deleting non-existent resource: {response_non_existent.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 4: DELETE with Custom Headers and Error Handling ---")
# APIs often require authentication headers for DELETE operations.
# Using httpbin.org/delete to echo back headers.
httpbin_delete_url = "https://httpbin.org/delete"
custom_headers_delete = {
    'Authorization': 'Bearer YOUR_ADMIN_TOKEN', # Replace with actual token
    'X-Requested-By': 'Python-Script'
}

try:
    print(f"Attempting to DELETE to: {httpbin_delete_url} with custom headers")
    response_headers_delete = requests.delete(httpbin_delete_url, headers=custom_headers_delete, timeout=5)
    response_headers_delete.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

    print(f"Status Code: {response_headers_delete.status_code}")
    print("Response JSON (echoing request details):")
    echoed_data = response_headers_delete.json()
    print(f"Echoed headers: {json.dumps(echoed_data.get('headers'), indent=2)}")
    print(f"Echoed URL: {echoed_data.get('url')}")

except requests.exceptions.HTTPError as e:
    print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error: {e}")
except requests.exceptions.RequestException as e:
    print(f"An unexpected requests error occurred: {e}")
```

### Key Takeaways for `DELETE` Requests with `requests`:

- **No Request Body:** Generally, `DELETE` requests do not carry a request body.
- **Idempotency:** A defining characteristic. Repeating a `DELETE` request has the same effect as the first one, ensuring the resource remains absent without unintended side effects.

- **Error Handling:** Essential for robust applications. Use `try...except` blocks to catch `requests.exceptions.RequestException` and `response.raise_for_status()` to immediately detect HTTP errors.

The `DELETE` method provides a clear and standardized way to manage the lifecycle of resources on a server, making it an indispensable part of interacting with RESTful APIs.


## HTTP HEAD Method in the `requests` Module

The `HEAD` method is an HTTP verb that is identical to `GET` but **without the response body**. It retrieves only the header information of a resource.

### Theory of the HEAD Method

1.  **Purpose: Fetching Metadata (Headers Only)**

    - The primary goal of a `HEAD` request is to obtain metadata about a resource without transferring the entire resource content. It's like asking, "What kind of data is at this URL, when was it last modified, how big is it, but don't send me the actual data itself?"

2.  **Safety:**

    - `HEAD` is considered a **"safe" method**. Just like `GET`, it does not cause any changes on the server. Its sole purpose is to retrieve information about the resource.

3.  **Idempotence:**

    - `HEAD` is an **"idempotent" method**. Making the same `HEAD` request multiple times will have the exact same effect as making it once. You'll always get the same set of headers (assuming the resource hasn't been independently updated).

4.  **No Response Body:**

    - The most defining characteristic: a `HEAD` response **must not contain a message body**. The server will send all the headers that it would send for a corresponding `GET` request, but then close the connection or indicate the end of headers without sending content.

5.  **Request Body:**

    - Like `GET`, `HEAD` requests typically **do not have a request body**. All parameters are usually part of the URL's query string if any are needed.

6.  **Usage Scenarios:**

    - **Checking Resource Existence:** Quickly determine if a file or page exists (`200 OK` vs. `404 Not Found`).
    - **Verifying Resource Accessibility/Permissions:** Check status codes like `401 Unauthorized` or `403 Forbidden` without downloading.
    - **Pre-download Checks:** Before initiating a large file download, use `HEAD` to check:
      - **Content-Length:** To know the file size.
      - **Content-Type:** To verify the file type.
      - **Last-Modified / ETag:** To determine if a cached version is still valid.
    - **Link Validation:** Check if external links are still valid and accessible.
    - **Debugging / Caching Headers:** Inspect specific headers like `Cache-Control`, `Expires`, `Server`, etc.

7.  **Responses:**
    - The status codes returned by a `HEAD` request are generally the same as those for a `GET` request to the same URI (e.g., `200 OK`, `404 Not Found`, `500 Internal Server Error`).

### `requests.head()` in Python

The `requests` library in Python provides the `requests.head()` function for performing `HEAD` requests.

#### Basic Syntax:

```python
import requests

response = requests.head(url, params=None, headers=None, timeout=None, **kwargs)
```

- `url`: The URL of the resource whose headers you want to retrieve.
- `params` (optional): A dictionary of query parameters to be added to the URL (less common for HEAD, but possible).
- `headers` (optional): A dictionary of HTTP headers to send with the request.
- `timeout` (optional): A float or tuple specifying how long to wait for the server to send data.
- `**kwargs`: Additional options like authentication, proxies, SSL verification, etc.

### Code Examples

Let's illustrate the `HEAD` method with various practical examples using the `requests` module.

```python
import requests
import json # For pretty printing headers if needed

print("--- Example 1: Basic HEAD Request to Check a Website ---")
# Quickly check if a website is up and get its basic headers.
url_website = "https://www.google.com"
try:
    print(f"Attempting HEAD request to: {url_website}")
    response_website = requests.head(url_website)

    print(f"Status Code: {response_website.status_code}")
    print("Response Headers:")
    # headers is a dictionary-like object
    for header, value in response_website.headers.items():
        print(f"  {header}: {value}")
    print(f"Content length (from header): {response_website.headers.get('Content-Length', 'N/A')}")
    print(f"Content of response.text (should be empty): '{response_website.text}' (length: {len(response_website.text)})")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 2: Checking File Existence and Size Before Download ---")
# Imagine you want to download an image only if it exists and is not too large.
image_url = "https://www.python.org/static/community_logos/python-logo-only.png"
non_existent_image_url = "https://www.python.org/static/community_logos/non_existent_logo.png"

# Check an existing image
try:
    print(f"Attempting HEAD request for existing image: {image_url}")
    response_image = requests.head(image_url)

    print(f"Status Code: {response_image.status_code}")
    if response_image.status_code == 200:
        content_type = response_image.headers.get('Content-Type')
        content_length = response_image.headers.get('Content-Length')
        last_modified = response_image.headers.get('Last-Modified')
        print(f"  Content-Type: {content_type}")
        print(f"  Content-Length: {content_length} bytes")
        print(f"  Last-Modified: {last_modified}")
        if content_length and int(content_length) < 100000: # Example check: less than 100KB
            print("  Image is within acceptable size for download.")
            # Here you might proceed with a GET request to download
        else:
            print("  Image is too large or size not available.")
    else:
        print("  Image resource not found or accessible.")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("-" * 30)

# Check a non-existent image
try:
    print(f"Attempting HEAD request for non-existent image: {non_existent_image_url}")
    response_non_existent = requests.head(non_existent_image_url)

    print(f"Status Code: {response_non_existent.status_code}") # Expecting 404
    if response_non_existent.status_code == 404:
        print("  Resource indeed does not exist (received 404 Not Found).")
    else:
        print(f"  Unexpected status for non-existent resource: {response_non_existent.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 3: HEAD with Custom Headers and Error Handling ---")
# Sending custom headers (e.g., authentication) and handling potential errors.
httpbin_head_url = "https://httpbin.org/headers" # Echoes back request headers
custom_headers_head = {
    'User-Agent': 'MyPythonHeadChecker/1.0',
    'Authorization': 'Bearer a_dummy_token',
    'X-Custom-Info': 'Checking-Headers'
}
try:
    print(f"Attempting HEAD request to: {httpbin_head_url} with custom headers")
    response_custom_head = requests.head(httpbin_head_url, headers=custom_headers_head, timeout=5)
    response_custom_head.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

    print(f"Status Code: {response_custom_head.status_code}")
    print("Response Headers (from httpbin.org, which echoes sent headers):")
    # httpbin.org returns actual request headers in its GET/POST response body,
    # but for HEAD, we only get the *response* headers.
    # To see our sent headers echoed, we'd need a GET/POST.
    # Here we just show the headers *received* from the HEAD response.
    for header, value in response_custom_head.headers.items():
        print(f"  {header}: {value}")
    # Note: You won't see 'User-Agent', 'Authorization', etc. from httpbin.org's *response* headers
    # because HEAD doesn't echo them in a body. You only get standard response headers.

except requests.exceptions.HTTPError as e:
    print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error: {e}")
except requests.exceptions.RequestException as e:
    print(f"An unexpected requests error occurred: {e}")
```

### Key Takeaways for `HEAD` Requests with `requests`:

- **Efficiency:** `HEAD` is highly efficient for checking resource existence or metadata because it avoids transferring the potentially large resource body.
- **No Body:** Always remember that `response.text` and `response.content` will be empty for a successful `HEAD` request. The information you need is in `response.headers`.
- **Idempotency & Safety:** `HEAD` is both safe (no server changes) and idempotent (repeated requests have the same outcome).
- **Pre-flight Checks:** Ideal for "pre-flight" checks before a full `GET` request, saving bandwidth and time.
- **Error Handling:** Standard `requests` error handling applies (`try...except` and `response.raise_for_status()`).

The `HEAD` method is a powerful but often overlooked tool for intelligent and efficient interaction with web resources, especially when dealing with large files or needing quick metadata checks.


## HTTP OPTIONS Method in the `requests` Module

The `OPTIONS` method is an HTTP verb used to **describe the communication options** for the target resource. It allows a client to determine the HTTP methods and other options that are supported by a web server or an API endpoint, without actually performing an action on the resource.

### Theory of the OPTIONS Method

1.  **Purpose: Discovering Server Capabilities**

    - The primary goal of an `OPTIONS` request is to query the server about the communication options available for a particular resource or a whole server. It's like asking, "What can I do with this URL?"
    - This is especially useful for:
      - **API Discovery:** Understanding which HTTP methods (`GET`, `POST`, `PUT`, `DELETE`, etc.) an endpoint supports.
      - **CORS (Cross-Origin Resource Sharing) Preflight Requests:** This is its most common and critical use case in web development. When a web application in a browser makes a "complex" cross-origin HTTP request (e.g., using `PUT`, `DELETE`, or custom headers), the browser first sends an `OPTIONS` request (the "preflight") to the server. This preflight request asks the server for permission to make the actual request. The server's response to the `OPTIONS` request dictates whether the browser is allowed to proceed.

2.  **Safety:**

    - `OPTIONS` is considered a **"safe" method**. It does not cause any changes to the state of the server or the resource. Its only purpose is to retrieve information about capabilities.

3.  **Idempotence:**

    - `OPTIONS` is an **"idempotent" method**. Making the same `OPTIONS` request multiple times will have the exact same effect as making it once. You'll always get the same information about supported methods and options.

4.  **No Request Body (Typically):**

    - `OPTIONS` requests typically **do not have a request body**. The information about what capabilities are being queried is implicit in the request URI.

5.  **Key Response Header: `Allow`**

    - The most important header in an `OPTIONS` response is the `Allow` header. This header lists the set of HTTP methods supported by the target resource.
    - Example: `Allow: GET, POST, PUT, DELETE, HEAD, OPTIONS`

6.  **Other Potential Response Headers:**

    - For CORS preflight requests, headers like `Access-Control-Allow-Origin`, `Access-Control-Allow-Methods`, `Access-Control-Allow-Headers`, `Access-Control-Max-Age` are crucial.

7.  **Responses:**
    - A successful `OPTIONS` request typically returns `200 OK`. The response body (if any) might contain more detailed information about the options, but it's often empty, with the `Allow` header carrying the main information.

### `requests.options()` in Python

The `requests` library in Python provides the `requests.options()` function for performing `OPTIONS` requests.

#### Basic Syntax:

```python
import requests

response = requests.options(url, headers=None, timeout=None, **kwargs)
```

- `url`: The URL of the resource for which you want to discover communication options.
- `headers` (optional): A dictionary of HTTP headers to send with the request (e.g., for CORS preflight, `Origin`, `Access-Control-Request-Method`, `Access-Control-Request-Headers` might be sent by the browser, but you'd manually set them for a Python script).
- `timeout` (optional): A float or tuple specifying how long to wait for the server to send data.
- `**kwargs`: Additional options like authentication, proxies, SSL verification, etc.

### Code Examples

Let's illustrate the `OPTIONS` method with various practical examples using the `requests` module.

```python
import requests
import json # For pretty printing headers if needed

print("--- Example 1: Discovering Supported Methods for a Public API Endpoint ---")
# Using jsonplaceholder.typicode.com/posts to see what methods it allows.
# This API generally supports GET, POST, PUT, PATCH, DELETE.
api_url = "https://jsonplaceholder.typicode.com/posts"

try:
    print(f"Attempting OPTIONS request to: {api_url}")
    response_options_api = requests.options(api_url)

    print(f"Status Code: {response_options_api.status_code}")
    if response_options_api.status_code == 200:
        allowed_methods = response_options_api.headers.get('Allow')
        print(f"Allowed Methods (from 'Allow' header): {allowed_methods}")
        if response_options_api.text:
            print("Response Body (if any):")
            try:
                print(json.dumps(response_options_api.json(), indent=2))
            except json.JSONDecodeError:
                print(response_options_api.text)
    else:
        print(f"Error fetching options: {response_options_api.status_code} - {response_options_api.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 2: Checking Options for a Root Domain ---")
# Sometimes OPTIONS requests to the root of a domain can reveal server capabilities.
website_url = "https://www.google.com"

try:
    print(f"Attempting OPTIONS request to: {website_url}")
    response_options_website = requests.options(website_url)

    print(f"Status Code: {response_options_website.status_code}")
    if response_options_website.status_code == 200:
        allowed_methods_website = response_options_website.headers.get('Allow')
        print(f"Allowed Methods (from 'Allow' header): {allowed_methods_website}")
        print("Other relevant headers:")
        for header, value in response_options_website.headers.items():
            if header.startswith('Access-Control') or header == 'Server':
                print(f"  {header}: {value}")
    else:
        print(f"Error fetching options: {response_options_website.status_code} - {response_options_website.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 3: Simulating a CORS Preflight Request ---")
# While browsers handle this automatically, you can manually construct a preflight.
# httpbin.org/anything endpoint is good for echoing requests.
cors_target_url = "https://httpbin.org/anything"
preflight_headers = {
    'Origin': 'http://my-frontend-app.com', # The origin of the client application
    'Access-Control-Request-Method': 'PUT', # The method the client wants to use
    'Access-Control-Request-Headers': 'X-Custom-Header, Authorization', # Custom headers client wants to send
}

try:
    print(f"Simulating CORS preflight OPTIONS to: {cors_target_url}")
    print(f"Sending headers: {json.dumps(preflight_headers, indent=2)}")
    response_cors_preflight = requests.options(cors_target_url, headers=preflight_headers, timeout=5)
    response_cors_preflight.raise_for_status() # Raise for bad status codes

    print(f"Status Code: {response_cors_preflight.status_code}")
    print("Response Headers (relevant to CORS):")
    for header, value in response_cors_preflight.headers.items():
        if header.startswith('Access-Control') or header == 'Allow':
            print(f"  {header}: {value}")

    # In a real scenario, you'd check these headers to see if the actual request is allowed.
    if 'Access-Control-Allow-Origin' in response_cors_preflight.headers and \
       response_cors_preflight.headers['Access-Control-Allow-Origin'] == 'http://my-frontend-app.com' and \
       'PUT' in response_cors_preflight.headers.get('Access-Control-Allow-Methods', '').split(', '):
        print("\nServer appears to allow the PUT request from this origin.")
    else:
        print("\nServer may not allow the PUT request from this origin or method.")

except requests.exceptions.HTTPError as e:
    print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error: {e}")
except requests.exceptions.RequestException as e:
    print(f"An unexpected requests error occurred: {e}")

```

### Key Takeaways for `OPTIONS` Requests with `requests`:

- **API Introspection:** `OPTIONS` is excellent for programmatically discovering what actions you can perform on a given resource, especially when interacting with unfamiliar APIs.
- **CORS Context:** While you're manually making the `OPTIONS` call in Python, remember its primary role in browsers for CORS preflight requests. The server's response to these requests determines cross-origin access.
- **`Allow` Header:** The most significant piece of information returned by a successful `OPTIONS` request is almost always in the `Allow` response header.
- **No Body (Usually):** Don't expect a meaningful response body; the information is predominantly in the headers.
- **Safety & Idempotence:** It's a safe and idempotent method, meaning you can call it repeatedly without side effects.
- **Error Handling:** Standard `requests` error handling practices are important.

The `OPTIONS` method might not be used as frequently for direct data manipulation as `GET` or `POST`, but its role in understanding server capabilities and facilitating secure cross-origin communication is fundamental to modern web architecture.


## HTTP PATCH Method in the `requests` Module

The `PATCH` method is an HTTP verb used for **applying partial modifications** to a resource. Unlike `PUT`, which is for complete replacement, `PATCH` is designed to send only the changes needed to update a resource, rather than the entire resource.

### Theory of the PATCH Method

1.  **Purpose: Partial Modification**

    - The primary goal of a `PATCH` request is to apply a _partial_ update to a resource. This means you send only the specific fields or parts of the resource that need to be changed, leaving the rest untouched.
    - This is particularly useful for large resources where sending the entire data just to change a small part would be inefficient.

2.  **Not Safe:**

    - `PATCH` is **not considered a "safe" method**. Like `POST` and `PUT`, it _will_ alter the state of the server or the resource.

3.  **Not Idempotent (Generally):**

    - `PATCH` is **generally not an "idempotent" method**. This is a key distinction from `PUT`.
    - If you apply a `PATCH` operation multiple times, the result might not be the same as applying it once. For example, if a `PATCH` operation is "increment a counter," applying it twice will increment the counter twice, leading to a different state than applying it once.
    - However, if a `PATCH` operation specifies an _absolute_ change (e.g., "set status to 'active'"), then it _can be_ idempotent. The idempotency depends on the _nature of the patch document_ and how the server applies it. The HTTP specification itself notes that `PATCH` is not necessarily idempotent.

4.  **Request Body:**

    - `PATCH` requests include the data representing the partial modifications in the **request body**.
    - The format of this "patch document" can vary and needs to be specified via the `Content-Type` header. Common formats include:
      - `application/json` (most common for APIs, sending a JSON object with only the fields to change).
      - `application/json-patch+json` (standardized JSON Patch format, more complex but very powerful for defining operations like add, remove, replace, move, copy, test).
      - `application/merge-patch+json` (JSON Merge Patch, a simpler alternative to JSON Patch).

5.  **Usage Scenarios:**

    - Updating a single field of a user profile (e.g., changing only the email address).
    - Incrementing a counter.
    - Adding an item to a list within a resource.
    - Toggling a boolean flag.

6.  **Responses:**
    - A successful `PATCH` request typically returns `200 OK` (with the updated resource in the body) or `204 No Content` (if no response body is needed).
    - Other codes like `400 Bad Request` (if the patch document is invalid), `404 Not Found`, `415 Unsupported Media Type` (if the server doesn't understand the patch format), etc., indicate errors.

### `requests.patch()` in Python

The `requests` library in Python provides the `requests.patch()` function for performing `PATCH` requests.

#### Basic Syntax:

```python
import requests

response = requests.patch(url, data=None, json=None, headers=None, timeout=None, **kwargs)
```

- `url`: The URL of the specific resource you want to partially modify.
- `data` (optional):
  - Bytes or file-like object for raw body content.
  - Less commonly used with `PATCH` for simple dictionaries, as `json` is preferred.
- `json` (optional):
  - A Python dictionary or list representing the partial changes to send as JSON in the request body (`application/json`). `requests` will automatically serialize it to a JSON string and set the `Content-Type` header to `application/json`. This is the most common way to send data with `PATCH` to modern APIs.
- `headers` (optional): A dictionary of HTTP headers to send with the request (crucial for `Content-Type` if using a specific patch format like `json-patch+json`).
- `timeout` (optional): A float or tuple specifying how long to wait for the server to send data.
- `**kwargs`: Additional options like authentication, proxies, SSL verification, etc.

### Code Examples

Let's illustrate the `PATCH` method with various practical examples using the `requests` module.

```python
import requests
import json

# Define a base URL for our examples (using jsonplaceholder.typicode.com/posts)
BASE_API_URL = "https://jsonplaceholder.typicode.com/posts"

print("--- Example 1: Basic PATCH Request with JSON Data (Common Use Case) ---")
# We'll update only the 'title' of post with ID 1.
# The server will merge this partial data with the existing resource.
post_id_to_patch = 1
patch_url = f"{BASE_API_URL}/{post_id_to_patch}"

# Only send the fields you want to change
partial_update_data = {
    'title': 'My Partially Updated Title from PATCH Request',
    # 'userId': 1 # You could also change other fields, but only send what's new.
}

try:
    print(f"Attempting to PATCH resource at: {patch_url}")
    print(f"Sending partial data: {json.dumps(partial_update_data, indent=2)}")
    response_patch = requests.patch(patch_url, json=partial_update_data)

    print(f"Status Code: {response_patch.status_code}")
    if response_patch.status_code == 200: # 200 OK is typical for successful PATCH
        print("Response JSON (showing updated resource):")
        print(json.dumps(response_patch.json(), indent=2))
        print(f"Content-Type in request headers: {response_patch.request.headers.get('Content-Type')}")
    else:
        print(f"Error patching resource: {response_patch.text}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 2: PATCH with Custom Headers and Error Handling ---")
# Using httpbin.org/patch to echo back headers and data.
httpbin_patch_url = "https://httpbin.org/patch"
patch_data_with_headers = {'status': 'archived'}
custom_headers_patch = {
    'Content-Type': 'application/json',
    'Authorization': 'Bearer ANOTHER_AUTH_TOKEN', # Replace with actual token if needed
    'X-Operation-Reason': 'Maintenance'
}

try:
    print(f"Attempting to PATCH to: {httpbin_patch_url}")
    print(f"Sending data: {json.dumps(patch_data_with_headers, indent=2)}")
    print(f"Sending headers: {json.dumps(custom_headers_patch, indent=2)}")

    response_headers_patch = requests.patch(httpbin_patch_url, json=patch_data_with_headers, headers=custom_headers_patch, timeout=5)
    response_headers_patch.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

    print(f"Status Code: {response_headers_patch.status_code}")
    print("Response JSON (echoing request details):")
    echoed_data = response_headers_patch.json()
    print(f"Echoed JSON data ('json' key): {json.dumps(echoed_data.get('json'), indent=2)}")
    print(f"Echoed headers: {json.dumps(echoed_data.get('headers'), indent=2)}")

except requests.exceptions.HTTPError as e:
    print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
except requests.exceptions.ConnectionError as e:
    print(f"Connection Error: {e}")
except requests.exceptions.Timeout as e:
    print(f"Timeout Error: {e}")
except requests.exceptions.RequestException as e:
    print(f"An unexpected requests error occurred: {e}")

print("\n" + "="*50 + "\n")

print("--- Example 3: Understanding PATCH vs. PUT Idempotency ---")
# Illustrative point, as jsonplaceholder doesn't truly handle non-idempotent PATCH
# but the concept is important.
# Imagine an API that has a 'views' counter that PATCH increments.
# If you PATCH with {'views': 1} (to set to 1) it's idempotent.
# If you PATCH with {'increment_views': True} (to add 1 to current), it's not.
# For jsonplaceholder, it's essentially a merge patch when using JSON.

# Let's get the current state of a post first to illustrate
post_id_for_idempotency = 2
get_url = f"{BASE_API_URL}/{post_id_for_idempotency}"
patch_url_idempotency = f"{BASE_API_URL}/{post_id_for_idempotency}"

# Step 1: Get initial state
print(f"GETting initial state of post {post_id_for_idempotency}:")
initial_response = requests.get(get_url)
initial_post = initial_response.json()
print(f"Initial title: {initial_post['title']}")

# Step 2: Apply a PATCH
patch_data = {'title': 'Updated by PATCH 1'}
print(f"\nApplying PATCH 1: {patch_data}")
patch_response_1 = requests.patch(patch_url_idempotency, json=patch_data)
updated_post_1 = patch_response_1.json()
print(f"Title after PATCH 1: {updated_post_1['title']}")

# Step 3: Apply the *same* PATCH again
print(f"\nApplying PATCH 2 (same data as PATCH 1): {patch_data}")
patch_response_2 = requests.patch(patch_url_idempotency, json=patch_data)
updated_post_2 = patch_response_2.json()
print(f"Title after PATCH 2: {updated_post_2['title']}")

# In this case, for a JSON PATCH setting a field, it is idempotent
# because setting 'title' to a specific value twice results in the same final title.
print("\nConclusion: For setting specific fields, PATCH can be idempotent. "
      "However, if the patch operation is 'incremental' (e.g., 'add 5'), "
      "then it is NOT idempotent.")
```

### Key Takeaways for `PATCH` Requests with `requests`:

## HTTP Status Codes in Python's `requests` Module: Theory and Code

### 1. Theory: Understanding HTTP Status Codes

Status codes are grouped into five classes, each representing a different category of response:

- **1xx Informational:** The request was received, continuing process. (Rarely seen by clients)
  - `100 Continue`
  - `101 Switching Protocols`
- **2xx Success:** The action was successfully received, understood, and accepted.
  - `200 OK`: Standard response for successful HTTP requests.
  - `201 Created`: The request has been fulfilled and resulted in a new resource being created. (Common for POST requests).
  - `204 No Content`: The server successfully processed the request, but is not returning any content. (Common for PUT/DELETE requests).
- **3xx Redirection:** Further action needs to be taken by the user agent to fulfill the request.
  - `301 Moved Permanently`: The requested resource has been assigned a new permanent URI.
  - `302 Found`: The requested resource resides temporarily under a different URI.
  - `304 Not Modified`: Indicates that the resource has not been modified since the version specified by the request headers.
- **4xx Client Error:** The request contains bad syntax or cannot be fulfilled.
  - `400 Bad Request`: The server cannot or will not process the request due to an apparent client error.
  - `401 Unauthorized`: Authentication is required and has failed or has not yet been provided.
  - `403 Forbidden`: The server understood the request but refuses to authorize it.
  - `404 Not Found`: The requested resource could not be found on the server.
- **5xx Server Error:** The server failed to fulfill an apparently valid request.
  - `500 Internal Server Error`: A generic error message, given when an unexpected condition was encountered and no more specific message is suitable.
  - `502 Bad Gateway`: The server, while acting as a gateway or proxy, received an invalid response from an inbound server it accessed in attempting to fulfill the request.
  - `503 Service Unavailable`: The server is currently unable to handle the request due to a temporary overload or scheduled maintenance.

#### 1.1 Accessing Status Code in `requests`

The `requests` module makes it very easy to access the status code of a response using the `response.status_code` attribute.

#### 1.2 Convenience Methods for Status Codes

- `response.ok`: A boolean attribute that is `True` for `2xx` status codes (successful responses) and `False` otherwise.
- `response.raise_for_status()`: A method that raises an `HTTPError` if the status code indicates a client error (4xx) or server error (5xx). This is a very common and efficient way to handle errors in your scripts.

```python
import requests
import json

print("--- Python `requests` Module: Status Code Examples ---")

# --- 2.1. Checking `response.status_code` directly ---
print("\n--- 2.1. Checking `response.status_code` directly ---")

# Example 1: Successful GET request (200 OK)
url_success = 'https://jsonplaceholder.typicode.com/posts/1'
try:
    response_success = requests.get(url_success)
    print(f"Request to {url_success}")
    print(f"  Status Code: {response_success.status_code}")
    print(f"  Is OK (response.ok): {response_success.ok}")
    if response_success.status_code == 200:
        print("  Request was successful!")
    else:
        print("  Request was not successful.")
except requests.exceptions.RequestException as e:
    print(f"  Error: {e}")

# Example 2: Resource Not Found (404)
url_not_found = 'https://jsonplaceholder.typicode.com/non-existent-page'
try:
    response_not_found = requests.get(url_not_found)
    print(f"\nRequest to {url_not_found}")
    print(f"  Status Code: {response_not_found.status_code}")
    print(f"  Is OK (response.ok): {response_not_found.ok}")
    if response_not_found.status_code == 404:
        print("  Resource not found (as expected).")
    else:
        print(f"  Unexpected status code: {response_not_found.status_code}")
except requests.exceptions.RequestException as e:
    print(f"  Error: {e}")

# Example 3: Creating a resource (201 Created) - common for POST
url_post_create = 'https://jsonplaceholder.typicode.com/posts'
new_data = {'title': 'New Entry', 'body': '...', 'userId': 1}
try:
    response_created = requests.post(url_post_create, json=new_data)
    print(f"\nPOST request to {url_post_create}")
    print(f"  Status Code: {response_created.status_code}")
    print(f"  Is OK (response.ok): {response_created.ok}")
    if response_created.status_code == 201:
        print("  Resource created successfully!")
        print(f"  New resource ID: {response_created.json().get('id')}")
    else:
        print(f"  Failed to create resource. Response: {response_created.text}")
except requests.exceptions.RequestException as e:
    print(f"  Error: {e}")


# --- 2.2. Using `response.ok` for quick success check ---
print("\n--- 2.2. Using `response.ok` for quick success check ---")

try:
    response_ok_check = requests.get('https://www.example.com')
    print(f"Request to www.example.com (Status: {response_ok_check.status_code})")
    if response_ok_check.ok:
        print("  Request was successful (2xx status code).")
    else:
        print("  Request was NOT successful (non-2xx status code).")
except requests.exceptions.RequestException as e:
    print(f"  Error: {e}")


# --- 2.3. Using `response.raise_for_status()` for robust error handling ---
print("\n--- 2.3. Using `response.raise_for_status()` for robust error handling ---")

# This is often the preferred method for handling non-success responses.
# It raises an `requests.exceptions.HTTPError` for 4xx and 5xx status codes.

# Example 1: `raise_for_status()` on a successful request (no error raised)
print("Attempting request for existing resource (200 OK)...")
try:
    response_ok_raise = requests.get('https://jsonplaceholder.typicode.com/todos/1')
    response_ok_raise.raise_for_status() # This will NOT raise an error for 200 OK
    print(f"  Request to {response_ok_raise.url} was successful (Status: {response_ok_raise.status_code}).")
    print(f"  Content snippet: {response_ok_raise.json()['title']}")
except requests.exceptions.RequestException as e:
    # This block would only catch network errors or other exceptions, not HTTP 4xx/5xx errors
    print(f"  Caught a general RequestException: {e}")


# Example 2: `raise_for_status()` on a client error (400 Bad Request)
print("\nAttempting request for a bad endpoint (400 Bad Request simulation)...")
# Note: jsonplaceholder doesn't typically return 400 for bad paths, so we'll use httpbin.org
try:
    response_bad_request = requests.get('https://httpbin.org/status/400')
    response_bad_request.raise_for_status() # This WILL raise an HTTPError
    print("  This line will not be reached.")
except requests.exceptions.HTTPError as e:
    print(f"  Caught HTTPError for status {e.response.status_code}: {e}")
    print(f"  Response content: {e.response.text}")
except requests.exceptions.RequestException as e:
    print(f"  Caught a general RequestException: {e}")


# Example 3: `raise_for_status()` on a server error (500 Internal Server Error)
print("\nAttempting request for a server error (500 Internal Server Error simulation)...")
try:
    response_server_error = requests.get('https://httpbin.org/status/500')
    response_server_error.raise_for_status() # This WILL raise an HTTPError
    print("  This line will not be reached.")
except requests.exceptions.HTTPError as e:
    print(f"  Caught HTTPError for status {e.response.status_code}: {e}")
    print(f"  Response content: {e.response.text}")
except requests.exceptions.RequestException as e:
    print(f"  Caught a general RequestException: {e}")


# --- 2.4. Customizing Error Handling with Status Codes ---
print("\n--- 2.4. Customizing Error Handling with Status Codes ---")

def fetch_data(url):
    try:
        response = requests.get(url)
        if response.status_code == 200:
            print(f"  Data successfully retrieved from {url}.")
            return response.json()
        elif response.status_code == 404:
            print(f"  Error: Resource not found at {url}. Please check the URL.")
        elif response.status_code == 401:
            print(f"  Error: Unauthorized access to {url}. Authentication required.")
        elif 500 <= response.status_code < 600:
            print(f"  Server error ({response.status_code}) from {url}. Please try again later.")
        else:
            print(f"  Unexpected status code {response.status_code} from {url}: {response.text}")
        return None
    except requests.exceptions.Timeout:
        print(f"  Timeout: Request to {url} took too long.")
        return None
    except requests.exceptions.ConnectionError:
        print(f"  Connection Error: Could not connect to {url}. Check internet or URL.")
        return None
    except requests.exceptions.RequestException as e:
        print(f"  An general error occurred during request to {url}: {e}")
        return None

print("\n--- Testing custom error handler ---")
# Test with a good URL
data = fetch_data('https://jsonplaceholder.typicode.com/users/1')
if data:
    print(f"  Fetched user name: {data.get('name')}")

# Test with a 404
fetch_data('https://jsonplaceholder.typicode.com/users/99999')

# Test with a simulated 401 (httpbin doesn't have true auth but can simulate status)
fetch_data('https://httpbin.org/status/401')

# Test with a simulated 503
fetch_data('https://httpbin.org/status/503')

```

## Session Handling in Python's `requests` Module: Theory and Code

### 1. Theory: Understanding `requests.Session()`

In the context of the `requests` module, a "session" (represented by the `requests.Session()` object) is primarily about **client-side state persistence and connection pooling**. It's important not to confuse this with a "server-side session," which is a distinct concept managed by the web server to track a user's activity over time.

#### 1.1 Why Use `requests.Session()`?

When you make individual requests using `requests.get()`, `requests.post()`, etc., each call is a new, independent request. While simple, this can be inefficient or problematic for certain scenarios:

1.  **Cookie Persistence:** Many web applications use cookies to maintain user authentication and track state. If you make separate `requests.get()` calls, cookies from one response are _not_ automatically sent with subsequent requests. A `Session` object handles this automatically: any cookies received in a response within the session are automatically stored and sent back with subsequent requests within the _same_ session. This is fundamental for logging in and then accessing protected pages.
2.  **Default Parameters:** You can set default headers, authentication credentials, proxies, or other request parameters on a `Session` object. These defaults will then apply to all requests made using that session instance, avoiding repetition.
3.  **Connection Pooling:** When you make requests to the same host using a `Session`, `requests` will reuse the underlying TCP connection if possible. This significantly reduces the overhead of establishing new connections for every request, leading to faster execution times, especially for a large number of requests or in scenarios with high latency.
4.  **Performance:** Due to connection pooling and reduced overhead from not re-evaluating headers/cookies for each request, sessions generally offer better performance for multiple requests to the same domain.
5.  **Context Manager:** `Session` objects can be used as context managers (`with requests.Session() as session:`), ensuring that resources (like connections) are properly closed and cleaned up even if errors occur.

#### 1.2 How it Works

When you create a `Session` object:

- It maintains an internal `CookieJar` to store and manage cookies.
- It has attributes for `headers`, `auth`, `proxies`, `params`, etc., which you can set as defaults.
- It uses an underlying `requests.adapters.HTTPAdapter` (and `urllib3` connection pools) to manage persistent connections.

In essence, a `Session` object is like a "browser instance" that remembers things (like cookies) and efficiently manages its network connections.

### 2. Code Examples

```python
import requests
import json # For pretty-printing JSON responses

print("--- Python `requests` Module: Session Handling Examples ---")

# Base URL for testing (httpbin.org is excellent for this)
HTTPBIN_URL = 'https://httpbin.org'

# --- 2.1. Basic Session Creation and Usage ---
print("\n--- 2.1. Basic Session Creation and Usage ---")

# Create a Session object
s = requests.Session()

# Use the Session object to make requests, just like requests directly
try:
    print("Making first request via session...")
    response_1 = s.get(f'{HTTPBIN_URL}/get')
    print(f"  First request status: {response_1.status_code}")

    print("\nMaking second request via session...")
    response_2 = s.post(f'{HTTPBIN_URL}/post', data={'key': 'value'})
    print(f"  Second request status: {response_2.status_code}")

    print("\nSession object created (but not explicitly closed yet):", s)

except requests.exceptions.RequestException as e:
    print(f"An error occurred during basic session usage: {e}")
finally:
    # It's good practice to close the session explicitly if not using 'with'
    s.close()
    print("Session explicitly closed.")


# --- 2.2. Cookie Persistence with a Session ---
print("\n--- 2.2. Cookie Persistence with a Session ---")

# Using a 'with' statement ensures the session is properly closed
try:
    with requests.Session() as session_with_cookies:
        # First request: Server sets a cookie (httpbin.org/cookies/set sets a cookie)
        print("1. First request: Setting a cookie via httpbin.org/cookies/set/mycookie/testvalue")
        set_cookie_url = f'{HTTPBIN_URL}/cookies/set/mycookie/testvalue'
        response_set = session_with_cookies.get(set_cookie_url)
        print(f"   Status Code: {response_set.status_code}")
        # The cookie is now stored in the session_with_cookies.cookies attribute
        print(f"   Cookies in session after first request: {session_with_cookies.cookies.get_dict()}")

        # Second request: Check if the cookie is sent automatically (httpbin.org/cookies reflects back received cookies)
        print("\n2. Second request: Checking if the cookie was sent automatically to httpbin.org/cookies")
        get_cookies_url = f'{HTTPBIN_URL}/cookies'
        response_get = session_with_cookies.get(get_cookies_url)
        print(f"   Status Code: {response_get.status_code}")
        received_cookies = response_get.json().get('cookies', {})
        print(f"   Cookies received by server: {received_cookies}")
        if 'mycookie' in received_cookies and received_cookies['mycookie'] == 'testvalue':
            print("   Success! Cookie 'mycookie' persisted across requests.")
        else:
            print("   Failure: Cookie 'mycookie' was NOT persisted.")

    print("\nSession (session_with_cookies) closed automatically by 'with' statement.")

except requests.exceptions.RequestException as e:
    print(f"An error occurred during cookie persistence demonstration: {e}")


# --- 2.3. Setting Default Headers for a Session ---
print("\n--- 2.3. Setting Default Headers for a Session ---")

try:
    with requests.Session() as session_with_headers:
        # Set a default User-Agent and a custom header for the session
        session_with_headers.headers.update({
            'User-Agent': 'MyCustomSessionApp/1.0',
            'X-Session-ID': '12345ABC'
        })

        print("1. First request via session with default headers...")
        response_headers_1 = session_with_headers.get(f'{HTTPBIN_URL}/headers')
        print(f"   Status Code: {response_headers_1.status_code}")
        # The headers sent by the server will include our custom ones
        sent_headers_1 = response_headers_1.json().get('headers', {})
        print(f"   Headers sent by client in first request: {json.dumps(sent_headers_1, indent=2)}")

        print("\n2. Second request via session with default headers (and overriding one)...")
        # You can override session-level headers for specific requests
        response_headers_2 = session_with_headers.get(
            f'{HTTPBIN_URL}/headers',
            headers={'X-Session-ID': 'OVERRIDDEN_ID', 'Accept': 'application/xml'}
        )
        print(f"   Status Code: {response_headers_2.status_code}")
        sent_headers_2 = response_headers_2.json().get('headers', {})
        print(f"   Headers sent by client in second request: {json.dumps(sent_headers_2, indent=2)}")
        if sent_headers_2.get('X-Session-Id') == 'OVERRIDDEN_ID': # Note: httpbin capitalizes headers
            print("   Success! X-Session-ID was overridden for the second request.")

except requests.exceptions.RequestException as e:
    print(f"An error occurred during default headers demonstration: {e}")


# --- 2.4. Default Parameters for a Session (less common but possible) ---
print("\n--- 2.4. Default Parameters for a Session ---")

try:
    with requests.Session() as session_with_params:
        # Set a default parameter for all GET requests in this session
        session_with_params.params.update({'source': 'session_default'})

        print("1. Request with default parameter:")
        response_params_1 = session_with_params.get(f'{HTTPBIN_URL}/get')
        print(f"   Request URL: {response_params_1.url}") # Should contain '?source=session_default'
        print(f"   Received args: {response_params_1.json()['args']}")

        print("\n2. Request with default parameter and an additional one:")
        response_params_2 = session_with_params.get(f'{HTTPBIN_URL}/get', params={'filter': 'active'})
        print(f"   Request URL: {response_params_2.url}") # Should contain '?source=session_default&filter=active'
        print(f"   Received args: {response_params_2.json()['args']}")

except requests.exceptions.RequestException as e:
    print(f"An error occurred during default parameters demonstration: {e}")

```

### 3. When to Use a `Session`

- **Authentication/Login:** Whenever you need to log into a website and then access other authenticated pages. The `Session` object automatically manages the authentication cookies.
- **Multiple Requests to the Same Host:** If your script makes several requests to the same domain, a `Session` will reuse connections, which improves performance.
- **Default Settings:** When you have common headers, authentication, or other parameters that need to be sent with multiple requests.
- **Stateful Interactions:** Any scenario where the server relies on client-side state (like cookies) to maintain context between requests.

For a single, isolated request, using `requests.get()` or `requests.post()` directly is perfectly fine and often simpler. However, for more complex interactions, `requests.Session()` is an indispensable tool.

## HTTP Headers in Python's `requests` Module: Theory and Code

### 1. Theory: What are HTTP Headers?

HTTP headers are lines in an HTTP message (either a request or a response) that provide information about the message itself, the sender, the recipient, or the resource being exchanged. They act like envelopes that carry instructions and context for the communication.

Each header consists of a case-insensitive name followed by a colon (`:`) and then its value.

#### 1.1 Purpose of Headers

Headers serve various purposes:

- **Content Description:** Describing the type, length, and encoding of the message body (e.g., `Content-Type`, `Content-Length`).
- **Client/Server Capabilities:** Informing about the capabilities of the client (browser/script) or server (e.g., `User-Agent`, `Accept`).
- **Authentication & Security:** Carrying credentials or security-related information (e.g., `Authorization`, `Set-Cookie`).
- **Caching Control:** Directing how responses should be cached (e.g., `Cache-Control`, `ETag`).
- **Redirection:** Indicating where a resource has moved (e.g., `Location`).
- **Connection Management:** Controlling how the network connection should be handled (e.g., `Connection`).

#### 1.2 Types of Headers

HTTP headers are broadly categorized into **Request Headers** (sent by the client) and **Response Headers** (sent by the server).

**Common Request Headers (Sent by your `requests` script to the server):**

- `User-Agent`: Identifies the client software originating the request (e.g., `Mozilla/5.0`, `Python-requests/2.28.1`). Servers use this for analytics, serving optimized content, or blocking certain clients.
- `Accept`: Informs the server about the types of content the client can understand (e.g., `application/json`, `text/html`, `image/png`).
- `Content-Type`: Specifies the media type of the request body when data is being sent (e.g., `application/json` for JSON payloads, `application/x-www-form-urlencoded` for form data, `multipart/form-data` for file uploads).
- `Authorization`: Contains authentication credentials (e.g., `Bearer <token>`, `Basic <base64_encoded_credentials>`).
- `Cookie`: Contains HTTP cookies previously received from the server, used to maintain session state.
- `Referer`: The URL of the page that linked to the requested resource.
- `If-Modified-Since` / `If-None-Match`: Used for conditional requests to leverage caching.

**Common Response Headers (Sent by the server to your `requests` script):**

- `Content-Type`: Specifies the media type of the response body (e.g., `application/json`, `text/html`). Crucial for `requests` to automatically decode the response (`response.text`, `response.json()`).
- `Content-Length`: The size of the response body in bytes.
- `Server`: Information about the web server software.
- `Date`: The date and time the response originated.
- `Set-Cookie`: Instructs the client to store a cookie.
- `Location`: Used with `3xx` redirection status codes to indicate the new URL.
- `Cache-Control`: Directives for caching mechanisms.
- `ETag`: A unique identifier for a specific version of a resource, used for caching.
- `X-Powered-By`: Indicates the technology used to build the server-side application.

#### 1.3 Headers in `requests` Module

- **Sending Request Headers:** You pass a dictionary of headers to the `headers` argument of any `requests` method (`get()`, `post()`, etc.).
- **Accessing Response Headers:** The `response.headers` attribute returns a case-insensitive dictionary-like object, allowing you to access response headers (e.g., `response.headers['Content-Type']`).
- **Automatic Headers:** `requests` automatically sets some standard headers (like `User-Agent`, `Connection: keep-alive`) and infers others based on the data you provide (e.g., `Content-Type` is set to `application/json` if you use the `json=` argument in a POST request).

### 2. Code Examples

```python
import requests
import json # For pretty-printing JSON responses

print("--- Python `requests` Module: Headers Examples ---")

# Base URL for testing (httpbin.org is excellent for reflecting requests)
HTTPBIN_URL = 'https://httpbin.org'

# --- 2.1. Sending Custom Request Headers ---
print("\n--- 2.1. Sending Custom Request Headers ---")

# Pass a dictionary of headers to the 'headers' argument
custom_headers = {
    'User-Agent': 'MyPythonScript/1.0 (Contact: user@example.com)',
    'Accept': 'application/json, text/plain', # Requesting JSON preferred, then plain text
    'X-Custom-Header': 'My-Value',
    'Authorization': 'Bearer YOUR_AUTH_TOKEN' # Example of an auth header (token not real)
}

try:
    # httpbin.org/headers reflects back the headers it received
    response_request_headers = requests.get(f'{HTTPBIN_URL}/headers', headers=custom_headers)

    if response_request_headers.status_code == 200:
        print("GET Request with custom headers successful!")
        print("Headers received by the server:")
        # httpbin.org returns headers in a 'headers' key within its JSON response
        received_headers = response_request_headers.json().get('headers', {})
        print(json.dumps(received_headers, indent=2))
    else:
        print(f"Request failed with status code: {response_request_headers.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")


# --- 2.2. Automatic `Content-Type` Header for POST/PUT Requests ---
print("\n--- 2.2. Automatic `Content-Type` Header for POST/PUT Requests ---")

# When you use the 'json' argument, requests automatically sets Content-Type: application/json
print("\nPOST request with `json` argument:")
try:
    post_json_data = {'name': 'Alice', 'age': 30}
    response_json_post = requests.post(f'{HTTPBIN_URL}/post', json=post_json_data)

    if response_json_post.status_code == 200:
        print(f"Status Code: {response_json_post.status_code}")
        sent_headers = response_json_post.json().get('headers', {})
        print(f"  Sent Content-Type: {sent_headers.get('Content-Type')}")
        print("  JSON data sent by client:", response_json_post.json().get('json'))
    else:
        print(f"Request failed with status code: {response_json_post.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")

# When you use the 'data' argument (for form data), requests sets Content-Type: application/x-www-form-urlencoded
print("\nPOST request with `data` argument (form data):")
try:
    post_form_data = {'username': 'testuser', 'password': 'password123'}
    response_form_post = requests.post(f'{HTTPBIN_URL}/post', data=post_form_data)

    if response_form_post.status_code == 200:
        print(f"Status Code: {response_form_post.status_code}")
        sent_headers = response_form_post.json().get('headers', {})
        print(f"  Sent Content-Type: {sent_headers.get('Content-Type')}")
        print("  Form data sent by client:", response_form_post.json().get('form'))
    else:
        print(f"Request failed with status code: {response_form_post.status_code}")
except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")


# --- 2.3. Accessing Response Headers ---
print("\n--- 2.3. Accessing Response Headers ---")

# Let's make a request to a common website to get typical response headers
url_response_headers = 'https://www.example.com' # A simple, reliable site

try:
    response_access = requests.get(url_response_headers)
    print(f"\nRequest to {url_response_headers} (Status: {response_access.status_code})")
    print("All Response Headers:")
    # response.headers is a case-insensitive dictionary-like object
    for key, value in response_access.headers.items():
        print(f"  {key}: {value}")

    # Accessing specific headers
    content_type = response_access.headers.get('Content-Type')
    server_info = response_access.headers.get('Server')
    date_info = response_access.headers.get('Date')

    print(f"\nSpecific Headers:")
    print(f"  Content-Type: {content_type}")
    print(f"  Server: {server_info}")
    print(f"  Date: {date_info}")

    # Checking if a header exists
    if 'Set-Cookie' in response_access.headers:
        print(f"  'Set-Cookie' header present: {response_access.headers['Set-Cookie']}")
    else:
        print(f"  'Set-Cookie' header not present.")

except requests.exceptions.RequestException as e:
    print(f"An error occurred while accessing response headers: {e}")


# --- 2.4. Using Sessions for Persistent Headers ---
print("\n--- 2.4. Using Sessions for Persistent Headers ---")

# A `requests.Session` object allows you to set default headers that will be
# sent with every request made using that session instance.
try:
    with requests.Session() as session:
        session.headers.update({
            'X-Persistent-Header': 'Session-Value',
            'User-Agent': 'MySessionAgent/1.0'
        })

        print("First request via session (should include persistent headers):")
        response_session_1 = session.get(f'{HTTPBIN_URL}/headers')
        print(f"  Status Code: {response_session_1.status_code}")
        print(f"  Headers received by server (1st request):")
        print(json.dumps(response_session_1.json().get('headers', {}), indent=2))

        print("\nSecond request via session (should also include persistent headers):")
        response_session_2 = session.post(f'{HTTPBIN_URL}/post', data={'foo': 'bar'})
        print(f"  Status Code: {response_session_2.status_code}")
        print(f"  Headers received by server (2nd request):")
        print(json.dumps(response_session_2.json().get('headers', {}), indent=2))

except requests.exceptions.RequestException as e:
    print(f"An error occurred during session header demonstration: {e}")

```

## Parameters in Python's `requests` Module: Theory and Code

### 1. Theory: Understanding Request Parameters

Parameters are key-value pairs that are included in an HTTP request. Their placement and typical use depend on the HTTP method and the nature of the data.

#### 1.1 Types of Parameters

1.  **Query Parameters (`params` argument):**

    - **Location:** Appended to the URL after a `?` symbol, with key-value pairs separated by `&`.
      - Example: `https://api.example.com/data?category=books&limit=10`
    - **Usage:** Primarily used with **GET** requests to filter, sort, paginate, or search data. They are visible in the URL and thus should _not_ contain sensitive information.
    - **Encoding:** `requests` automatically handles URL encoding for parameter keys and values (e.g., spaces become `%20`, special characters like `&` are correctly escaped). You pass them as a Python dictionary.

2.  **Request Body Parameters (`data` / `json` arguments):**

    - **Location:** Sent within the _body_ of the HTTP request.
    - **Usage:** Primarily used with **POST**, **PUT**, and **PATCH** requests to submit data to the server for creation or update. The `Content-Type` header specifies the format of the data in the body.
    - **`data` argument:** Used for sending form-encoded data (like an HTML form submission). `requests` sets the `Content-Type` header to `application/x-www-form-urlencoded`. Values can be strings or dictionaries.
    - **`json` argument:** Used for sending data in JSON format. `requests` automatically serializes the Python dictionary/list to a JSON string and sets the `Content-Type` header to `application/json`. This is very common for interacting with modern RESTful APIs.

3.  **Path Parameters (URL segments):**
    - **Location:** Part of the URL path itself, often representing specific resource identifiers.
      - Example: `https://api.example.com/users/123/profile` (here, `123` is a path parameter for a user ID).
    - **Usage:** Used to identify a specific resource. You typically construct these directly into the URL string using f-strings or string formatting in Python.
    - **Encoding:** You are responsible for ensuring these are correctly placed in the URL, but `requests` will handle any necessary encoding of characters within the path segment if you construct the URL correctly.

#### 1.2 `requests` and Parameter Handling

The `requests` module simplifies parameter handling significantly:

- It abstracts away the complexities of URL encoding for query parameters and body data.
- It automatically sets the appropriate `Content-Type` header when `data` or `json` arguments are used.
- It provides clear arguments for each type of parameter (`params`, `data`, `json`, `files`).

## Sending GET Requests with Python's `requests` Module: Theory and Code

### 1. Theory: Understanding HTTP GET Requests

The HTTP **GET** method is one of the most fundamental request methods defined by the Hypertext Transfer Protocol.

#### 1.1 Purpose of GET

- **Retrieving Data:** The primary purpose of a GET request is to retrieve (or "get") data from a specified resource on a web server.
- **Idempotent:** Making the same GET request multiple times should have the same effect as making it once. It should _not_ cause any side effects on the server (like modifying data).
- **Safe:** GET requests are considered "safe" because they are not supposed to alter the state of the server. This makes them suitable for caching and bookmarking.
- **Data via URL Parameters:** Data passed with a GET request is typically appended to the URL as query parameters (e.g., `?key=value&another_key=another_value`).

#### 1.2 When to Use `requests.get()`

You use `requests.get()` whenever your goal is to fetch information from a web server without changing anything on the server. Common scenarios include:

- Fetching a web page (e.g., scraping HTML).
- Retrieving data from a RESTful API (e.g., getting a list of users, a specific product, or search results).
- Downloading a file (e.g., an image, a PDF document).
- Checking the status or existence of a resource.

#### 1.3 Key Arguments for `requests.get()`

The `requests.get()` function accepts several important arguments to customize your request:

- **`url` (required):** The URL of the resource you want to retrieve.
- **`params` (optional):** A dictionary or list of tuples of key-value pairs to be sent as URL query parameters. `requests` automatically handles URL encoding.
  - Example: `params={'city': 'London', 'temp_unit': 'celsius'}` would become `?city=London&temp_unit=celsius`.
- **`headers` (optional):** A dictionary of HTTP headers to send with the request. This is used to provide metadata (e.g., `User-Agent`, `Accept`, `Authorization`).
- **`timeout` (optional):** A float or tuple indicating how many seconds to wait for the server to send data before giving up. Crucial for preventing your program from hanging indefinitely.
- **`allow_redirects` (optional):** A boolean (default `True`). If `False`, `requests` will not follow HTTP redirects (3xx status codes).
- **`verify` (optional):** A boolean (default `True`). If `False`, requests will skip SSL certificate verification. **Use with caution, as this can make your application vulnerable to Man-in-the-Middle attacks.**
- **`auth` (optional):** A tuple or object for HTTP authentication (e.g., `('username', 'password')` for Basic Auth).
- **`proxies` (optional):** A dictionary mapping protocol to proxy URL (e.g., `{'http': 'http://proxy.example.com:8080'}`).

#### 1.4 The `Response` Object

When `requests.get()` completes (successfully or with an error), it returns a `Response` object. This object contains all the information received from the server, including:

- **`response.status_code`**: The HTTP status code (e.g., `200`, `404`, `500`).
- **`response.ok`**: A boolean (`True` for `2xx` status codes, `False` otherwise).
- **`response.text`**: The content of the response, decoded as a Unicode string (useful for HTML, plain text).
- **`response.content`**: The raw bytes of the response body (useful for images, binary files).
- **`response.json()`**: A method that parses the response body as JSON and returns a Python dictionary or list. Raises `JSONDecodeError` if the content is not valid JSON.
- **`response.headers`**: A case-insensitive dictionary-like object of response headers.
- **`response.url`**: The final URL of the request, after any redirects.
- **`response.reason`**: The HTTP status reason phrase (e.g., 'OK', 'Not Found').
- **`response.raise_for_status()`**: A method that raises an `requests.exceptions.HTTPError` if the response status code is a 4xx (client error) or 5xx (server error). This is excellent for simple error checking.

```python
import requests
import json # For pretty-printing JSON responses

print("--- Python `requests` Module: GET Request Examples ---")

# Base URL for examples (using JSONPlaceholder, a fake online REST API)
BASE_API_URL = 'https://jsonplaceholder.typicode.com'

# --- 2.1. Basic GET Request ---
print("\n--- 2.1. Basic GET Request ---")

# Fetching a single post by ID
url_basic_get = f'{BASE_API_URL}/posts/1'

print(f"Sending GET request to: {url_basic_get}")
try:
    response = requests.get(url_basic_get)

    # Check if the request was successful (HTTP 200 OK)
    if response.status_code == 200:
        print("  Request successful (Status 200 OK)!")
        # Access the response content as JSON
        post_data = response.json()
        print("  Received data:")
        print(json.dumps(post_data, indent=2))
    else:
        print(f"  Request failed with status code: {response.status_code}")
        print(f"  Response content: {response.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred during the request: {e}")


# --- 2.2. GET Request with Query Parameters (`params` argument) ---
print("\n--- 2.2. GET Request with Query Parameters (`params` argument) ---")

# Fetching comments for a specific post ID
# The 'params' dictionary is automatically URL-encoded and appended to the URL.
url_with_params = f'{BASE_API_URL}/comments'
query_parameters = {
    'postId': 5,
    '_limit': 3,         # Limit the number of results
    'search_term': 'quia et', # Example with a space
    'category': 'science fiction' # Example with multiple words
}

print(f"Sending GET request to: {url_with_params}")
print(f"  With parameters: {query_parameters}")
try:
    response_params = requests.get(url_with_params, params=query_parameters)

    if response_params.ok: # Use .ok for any 2xx status code
        print("  Request successful (Status 2xx)!")
        print(f"  Full URL sent by requests: {response_params.url}") # Observe URL encoding
        comments = response_params.json()
        print(f"  Retrieved {len(comments)} comments.")
        if comments:
            print("  First comment snippet:")
            print(json.dumps(comments[0], indent=2))
    else:
        print(f"  Request failed with status code: {response_params.status_code}")
        print(f"  Reason: {response_params.reason}")
        print(f"  Response content: {response_params.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred during the request with parameters: {e}")


# --- 2.3. GET Request with Custom Headers ---
print("\n--- 2.3. GET Request with Custom Headers ---")

# Sending a custom User-Agent and requesting specific content type
url_with_headers = f'{BASE_API_URL}/users/1'
custom_headers = {
    'User-Agent': 'MyAwesomePythonApp/1.0',
    'Accept': 'application/json', # Explicitly ask for JSON
    'X-Requested-With': 'XMLHttpRequest' # Common header for AJAX requests
}

print(f"Sending GET request to: {url_with_headers}")
print(f"  With headers: {custom_headers}")
try:
    response_headers = requests.get(url_with_headers, headers=custom_headers)

    if response_headers.ok:
        print("  Request successful (Status 2xx)!")
        user_data = response_headers.json()
        print("  Received user data:")
        print(json.dumps(user_data, indent=2))
        print(f"  Content-Type of response: {response_headers.headers.get('Content-Type')}")
    else:
        print(f"  Request failed with status code: {response_headers.status_code}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred during the request with headers: {e}")


# --- 2.4. Error Handling with `raise_for_status()` ---
print("\n--- 2.4. Error Handling with `raise_for_status()` ---")

# `raise_for_status()` is a convenient way to check if the response was a success (2xx).
# If not, it raises an HTTPError.

# Example 1: Non-existent resource (will result in 404 Not Found)
url_error = f'{BASE_API_URL}/nonexistent-resource-12345'

print(f"Sending GET request to a non-existent URL: {url_error}")
try:
    response_error = requests.get(url_error)
    response_error.raise_for_status() # This will raise an HTTPError for 404
    print("  This line will NOT be reached if an HTTPError occurs.")
except requests.exceptions.HTTPError as e:
    print(f"  Caught HTTPError: {e}")
    print(f"  Status Code: {e.response.status_code}, Reason: {e.response.reason}")
except requests.exceptions.RequestException as e:
    # This catches other types of requests errors (e.g., network issues, timeouts)
    print(f"  Caught a general RequestException: {e}")


# Example 2: Successful request with `raise_for_status()` (no error raised)
url_success_check = f'{BASE_API_URL}/todos/1'
print(f"\nSending GET request to a valid URL: {url_success_check}")
try:
    response_success_check = requests.get(url_success_check)
    response_success_check.raise_for_status() # No error for 200 OK
    print(f"  Request successful (Status {response_success_check.status_code}), no error raised by `raise_for_status()`.")
except requests.exceptions.RequestException as e:
    print(f"  Caught an unexpected RequestException: {e}")


# --- 2.5. Handling Different Response Content Types ---
print("\n--- 2.5. Handling Different Response Content Types ---")

# Getting raw text/HTML content
url_html = 'https://www.example.com' # Returns HTML
print(f"\nSending GET request to fetch HTML from: {url_html}")
try:
    response_html = requests.get(url_html)
    if response_html.ok:
        print(f"  Content-Type: {response_html.headers.get('Content-Type')}")
        print("  HTML snippet (first 100 characters):")
        print(response_html.text[:100] + "...")
    else:
        print(f"  Request failed: {response_html.status_code}")
except requests.exceptions.RequestException as e:
    print(f"  Error: {e}")

# Getting binary content (e.g., an image)
# Note: For real image download, you'd save response_image.content to a file.
url_image = 'https://placehold.co/100x50/ff0000/ffffff?text=Image' # A small placeholder image
print(f"\nSending GET request to fetch an image from: {url_image}")
try:
    response_image = requests.get(url_image)
    if response_image.ok:
        print(f"  Content-Type: {response_image.headers.get('Content-Type')}")
        print(f"  Content length: {len(response_image.content)} bytes (raw binary data)")
        # In a real scenario, you'd save response_image.content to a file like:
        # with open('downloaded_image.png', 'wb') as f:
        #     f.write(response_image.content)
        # print("  Image content downloaded (not saved to disk in this example).")
    else:
        print(f"  Request failed: {response_image.status_code}")
except requests.exceptions.RequestException as e:
    print(f"  Error: {e}")

```

## Sending POST Requests with Python's `requests` Module: Theory and Code

### 1. Theory: Understanding HTTP POST Requests

The HTTP **POST** method is a fundamental request method used to send data to a server to create or update a resource.

#### 1.1 Purpose of POST

- **Submitting Data:** The primary purpose of a POST request is to send data to the server to be processed, often resulting in a change on the server's side.
- **Creating Resources:** Commonly used to create new resources on the server (e.g., submitting a new user registration, creating a new product listing, adding a comment).
- **Updating Resources:** Can also be used to update existing resources when the update requires sending a full body of data (though PUT is often preferred for full replacement updates).
- **Non-Idempotent (Typically):** Unlike GET, POST requests are generally _not_ idempotent. Sending the same POST request multiple times might result in multiple new resources being created or repeated actions. For example, submitting a "create user" form twice could create two users.
- **Data in Request Body:** Data sent with a POST request is typically enclosed in the _request body_, not in the URL's query string. This makes it suitable for sending larger amounts of data and sensitive information (as it's not visible in browser history or server logs by default, though still transmitted over the network).

#### 1.2 When to Use `requests.post()`

You use `requests.post()` whenever your goal is to send data to a web server that changes its state or creates a new resource. Common scenarios include:

- Submitting data from a web form (e.g., login credentials, contact form).
- Adding new entries to a database via an API.
- Uploading files.
- Sending complex, structured data (like JSON or XML) to an API.

#### 1.3 Key Arguments for `requests.post()`

The `requests.post()` function accepts several important arguments to define the data being sent and customize the request:

- **`url` (required):** The URL of the endpoint where you want to submit the data.
- **`data` (optional):**
  - A dictionary, list of tuples, bytes, or file-like object.
  - Used primarily for sending **form-encoded data** (`application/x-www-form-urlencoded`) or raw binary data.
  - If a dictionary is provided, `requests` will automatically URL-encode it and set the `Content-Type` header.
  - Example: `data={'username': 'test', 'password': '123'}`
- **`json` (optional):**
  - A Python dictionary or list.
  - Used for sending data as **JSON** (`application/json`). `requests` will automatically serialize the Python object to a JSON string and set the `Content-Type` header. This is the most common way to interact with modern RESTful APIs.
  - Example: `json={'name': 'Alice', 'email': 'alice@example.com'}`
- **`files` (optional):**
  - A dictionary of `{'field_name': file_tuple_or_object}`.
  - Used for **file uploads** (`multipart/form-data`). `requests` handles the complex encoding required for file uploads.
  - A `file_tuple_or_object` can be `('filename', file_object, 'content_type', {'headers'})`.
  - Example: `files={'image': open('my_image.png', 'rb')}`
- **`headers` (optional):** A dictionary of HTTP headers to send with the request. (e.g., `Authorization`). Note that `requests` often sets `Content-Type` automatically when using `data` or `json`.
- **`timeout` (optional):** How many seconds to wait for the server.
- **`auth` (optional):** For authentication.
- **`proxies` (optional):** For proxy configuration.
- **`allow_redirects` (optional):** Default `True`, but often set to `False` for POST requests if you want to inspect redirects manually.

#### 1.4 The `Response` Object

Just like with GET requests, `requests.post()` returns a `Response` object. Key attributes and methods of the `Response` object remain the same:

- `response.status_code`: Typically `201 Created` for successful resource creation, `200 OK` for successful updates/processing, or `4xx`/`5xx` for errors.
- `response.ok`: `True` for `2xx` status codes.
- `response.text`: Response content as a Unicode string.
- `response.json()`: Parses JSON response.
- `response.headers`: Response headers.
- `response.raise_for_status()`: Raises `HTTPError` for 4xx/5xx status codes.

### 2. Code Examples

```python
import requests
import json # For pretty-printing JSON responses
import os   # For creating a dummy file for upload demonstration

print("--- Python `requests` Module: POST Request Examples ---")

# Base URL for examples (using JSONPlaceholder and httpbin.org)
JSON_PLACEHOLDER_URL = 'https://jsonplaceholder.typicode.com'
HTTPBIN_URL = 'https://httpbin.org' # Excellent for reflecting requests back

# --- 2.1. Basic POST Request with Form Data (`data` argument) ---
print("\n--- 2.1. Basic POST Request with Form Data (`data` argument) ---")

# This mimics submitting a traditional HTML form.
# requests will automatically set the 'Content-Type' header to 'application/x-www-form-urlencoded'.
post_url_form = f'{JSON_PLACEHOLDER_URL}/posts'
form_payload = {
    'title': 'My First Post via Requests',
    'body': 'This is the content of my new post, sent as form data.',
    'userId': 1
}

print(f"Sending POST request to: {post_url_form}")
print(f"  With form data: {form_payload}")
try:
    response_form_post = requests.post(post_url_form, data=form_payload)

    if response_form_post.status_code == 201: # 201 Created for a new resource
        print("  POST Request successful (Status 201 Created)!")
        new_resource_data = response_form_post.json()
        print("  Received new resource data:")
        print(json.dumps(new_resource_data, indent=2))
        print(f"  Content-Type header sent: {response_form_post.request.headers.get('Content-Type')}")
    else:
        print(f"  POST Request failed with status code: {response_form_post.status_code}")
        print(f"  Response content: {response_form_post.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred: {e}")


# --- 2.2. POST Request with JSON Data (`json` argument) ---
print("\n--- 2.2. POST Request with JSON Data (`json` argument) ---")

# This is the most common way to interact with modern RESTful APIs.
# requests will automatically serialize the Python dictionary to JSON
# and set the 'Content-Type' header to 'application/json'.
post_url_json = f'{JSON_PLACEHOLDER_URL}/users' # Creating a new user
json_payload = {
    'name': 'Python User',
    'username': 'pythonuser',
    'email': 'python@example.com',
    'address': {
        'street': 'Kulas Light',
        'suite': 'Apt. 556',
        'city': 'Gwenborough',
        'zipcode': '92998-3874'
    },
    'phone': '1-770-736-8031 x56442',
    'website': 'hildegard.org'
}

print(f"\nSending POST request to: {post_url_json}")
print(f"  With JSON payload (dict): {json_payload}")
try:
    response_json_post = requests.post(post_url_json, json=json_payload)

    if response_json_post.status_code == 201:
        print("  POST Request successful (Status 201 Created)!")
        new_user_data = response_json_post.json()
        print("  Received new user data:")
        print(json.dumps(new_user_data, indent=2))
        print(f"  Content-Type header sent: {response_json_post.request.headers.get('Content-Type')}")
    else:
        print(f"  POST Request failed with status code: {response_json_post.status_code}")
        print(f"  Response content: {response_json_post.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred: {e}")


# --- 2.3. POST Request with File Uploads (`files` argument) ---
print("\n--- 2.3. POST Request with File Uploads (`files` argument) ---")

# For uploading files (e.g., images, documents).
# requests will automatically set the 'Content-Type' header to 'multipart/form-data'.

# Create a dummy file for demonstration
dummy_file_path = "upload_test.txt"
try:
    with open(dummy_file_path, "w") as f:
        f.write("This is a test file to be uploaded.\n")
        f.write("It contains some simple text content.")
    print(f"Created dummy file: {dummy_file_path}")

    # Open the file in binary read mode ('rb')
    with open(dummy_file_path, 'rb') as file_to_upload:
        # The 'files' argument takes a dictionary.
        # Key: the field name expected by the server for the file upload.
        # Value: the file object, or a tuple: ('filename', file_object, 'content_type', {'headers'})
        files_payload = {'my_uploaded_file': file_to_upload}

        print(f"\nSending POST request to {HTTPBIN_URL}/post for file upload...")
        response_file_upload = requests.post(f'{HTTPBIN_URL}/post', files=files_payload)

        if response_file_upload.status_code == 200:
            print("  File upload successful (Status 200 OK)!")
            upload_result = response_file_upload.json()
            print("  Server's response (reflecting uploaded file data):")
            # The 'files' key in httpbin's response shows the uploaded file's content
            print(f"  Uploaded file content: {upload_result.get('files', {}).get('my_uploaded_file')}")
            print(f"  Content-Type header sent: {response_file_upload.request.headers.get('Content-Type')}")
        else:
            print(f"  File upload failed with status code: {response_file_upload.status_code}")
            print(f"  Response content: {response_file_upload.text}")

except FileNotFoundError:
    print(f"  Error: Dummy file '{dummy_file_path}' not found. Cannot run file upload test.")
except requests.exceptions.RequestException as e:
    print(f"  An error occurred during file upload: {e}")
finally:
    # Clean up the dummy file
    if os.path.exists(dummy_file_path):
        os.remove(dummy_file_path)
        print(f"  Cleaned up dummy file: {dummy_file_path}")


# --- 2.4. POST Request with Custom Headers ---
print("\n--- 2.4. POST Request with Custom Headers ---")

# You can always add custom headers. Note that Requests will prioritize its
# automatically set Content-Type header if you use 'json' or 'data' arguments.
post_url_headers = f'{HTTPBIN_URL}/post'
custom_headers = {
    'Authorization': 'Bearer YOUR_ACTUAL_TOKEN_HERE',
    'X-Client-Version': '2.5.0'
}
json_data_for_headers = {'message': 'Hello from Python!'}

print(f"\nSending POST request with custom headers to: {post_url_headers}")
print(f"  Headers: {custom_headers}")
try:
    response_headers_post = requests.post(post_url_headers, json=json_data_for_headers, headers=custom_headers)

    if response_headers_post.status_code == 200:
        print("  POST Request successful!")
        sent_headers_by_server = response_headers_post.json().get('headers', {})
        print("  Headers received by the server:")
        print(json.dumps(sent_headers_by_server, indent=2))
        # Note: 'Authorization' and 'X-Client-Version' should be present
        # 'Content-Type' will be 'application/json' because of 'json' argument
    else:
        print(f"  POST Request failed with status code: {response_headers_post.status_code}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred: {e}")

```

## Sending PUT Requests with Python's `requests` Module: Theory and Code

### 1. Theory: Understanding HTTP PUT Requests

The HTTP **PUT** method is a request method used to send data to a server to either **create or replace** a resource at a specified Uniform Resource Identifier (URI).

#### 1.1 Purpose of PUT

- **Replacing Existing Resources:** The most common use case for PUT is to completely replace an existing resource with new data. When you send a PUT request, you provide the _entire_ updated representation of the resource. If fields are omitted from the payload, they might be removed from the resource on the server (depending on the API's implementation).
- **Creating Resources at a Known URI:** If the client already knows the exact URI where a resource should reside, PUT can be used to create that resource. This is less common than using POST for creation, as POST typically lets the server determine the new resource's URI. However, if the client has control over the URI (e.g., creating a file at a specific path), PUT is appropriate for creation.
- **Idempotent:** PUT requests are **idempotent**. This is a crucial characteristic. It means that making the same PUT request multiple times will have the exact same effect on the server as making it once. If you send the same updated data to the same URI repeatedly, the resource will simply be set to that state each time, without creating duplicates or causing additional side effects.

#### 1.2 PUT vs. POST

Understanding the difference between PUT and POST is key:

| Feature           | POST                                                                        | PUT                                                     |
| :---------------- | :-------------------------------------------------------------------------- | :------------------------------------------------------ |
| **Purpose**       | Create new resource, submit data.                                           | Replace existing resource, or create if URI known.      |
| **Idempotency**   | Generally **Not Idempotent** (repeated requests can have different effects) | **Idempotent** (repeated requests have the same effect) |
| **URI Handling**  | Server typically assigns URI for new resource.                              | Client specifies the exact URI of the resource.         |
| **Data Location** | In the request body.                                                        | In the request body.                                    |

#### 1.3 Key Arguments for `requests.put()`

The `requests.put()` function shares many arguments with `requests.post()`:

- **`url` (required):** The exact URI of the resource you want to update or create. This URL _must_ identify the specific resource, not a collection.
- **`data` (optional):**
  - Used for sending **form-encoded data** (`application/x-www-form-urlencoded`) or raw binary data in the request body.
  - Example: `data={'field1': 'value1', 'field2': 'value2'}`
- **`json` (optional):**
  - Used for sending data as **JSON** (`application/json`). This is the most common and recommended way to send structured data to RESTful APIs for updates.
  - Example: `json={'name': 'Updated Name', 'status': 'active'}`
- **`headers` (optional):** A dictionary of HTTP headers (e.g., `Authorization`). `requests` will automatically set `Content-Type` if `data` or `json` is used.
- **`timeout` (optional):** How many seconds to wait for the server.
- **`auth` (optional):** For authentication.
- **`proxies` (optional):** For proxy configuration.

#### 1.4 The `Response` Object

Similar to other `requests` methods, `requests.put()` returns a `Response` object. Key attributes and methods include:

- **`response.status_code`**: Typically:
  - `200 OK`: If the resource was successfully updated.
  - `201 Created`: If the resource was successfully created (at the specified URI).
  - `204 No Content`: If the resource was successfully updated, but the server has no content to return in the body.
  - `4xx`/`5xx`: For client or server errors.
- **`response.ok`**: `True` for `2xx` status codes.
- **`response.text`**: Response content as a Unicode string.
- **`response.json()`**: Parses JSON response.
- **`response.headers`**: Response headers.
- **`response.raise_for_status()`**: Raises `HTTPError` for 4xx/5xx status codes.

### 2. Code Examples

```python
import requests
import json # For pretty-printing JSON responses

print("--- Python `requests` Module: PUT Request Examples ---")

# Base URL for examples (using JSONPlaceholder, a fake online REST API)
# For PUT, we typically target a specific resource, e.g., /posts/1
BASE_API_URL = 'https://jsonplaceholder.typicode.com'

# --- 2.1. Basic PUT Request with JSON Data (Common Use Case: Full Replacement) ---
print("\n--- 2.1. Basic PUT Request with JSON Data (Common Use Case: Full Replacement) ---")

# Target a specific post by its ID to update it.
# JSONPlaceholder will return the updated resource, but in a real API,
# it might return 204 No Content or just a success message.
post_id_to_update = 1
put_url_json = f'{BASE_API_URL}/posts/{post_id_to_update}'

# The payload should represent the COMPLETE new state of the resource.
# If original post had more fields, and you only send these, the others might be removed.
# For JSONPlaceholder, userId and id are often sent back as part of the structure.
full_update_payload = {
    'id': post_id_to_update, # It's good practice to include the ID in the payload as well
    'title': 'My Fully Updated Post Title',
    'body': 'This is the entirely new content for the post. All old content is replaced.',
    'userId': 1 # Assuming this post belongs to user 1
}

print(f"Sending PUT request to: {put_url_json}")
print(f"  With JSON payload: {full_update_payload}")
try:
    response_put_json = requests.put(put_url_json, json=full_update_payload)

    if response_put_json.status_code == 200: # 200 OK for successful update
        print("  PUT Request successful (Status 200 OK)!")
        updated_resource = response_put_json.json()
        print("  Received updated resource data:")
        print(json.dumps(updated_resource, indent=2))
        print(f"  Content-Type header sent: {response_put_json.request.headers.get('Content-Type')}")
    else:
        print(f"  PUT Request failed with status code: {response_put_json.status_code}")
        print(f"  Response content: {response_put_json.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred: {e}")


# --- 2.2. PUT Request with Form Data (`data` argument) ---
print("\n--- 2.2. PUT Request with Form Data (`data` argument) ---")

# Less common for full replacements in modern APIs, but still possible.
# Content-Type: application/x-www-form-urlencoded will be set automatically.
user_id_to_update = 5
put_url_form = f'{BASE_API_URL}/users/{user_id_to_update}'

form_update_payload = {
    'id': user_id_to_update,
    'name': 'Updated Name Form',
    'username': 'updated.user',
    'email': 'updated@example.com'
    # Note: If the actual user resource has more fields (like address),
    # sending only these would typically erase the others if it's a true PUT (full replacement).
    # JSONPlaceholder might merge them.
}

print(f"\nSending PUT request to: {put_url_form}")
print(f"  With form data: {form_update_payload}")
try:
    response_put_form = requests.put(put_url_form, data=form_update_payload)

    if response_put_form.status_code == 200:
        print("  PUT Request successful (Form Data)!")
        updated_resource = response_put_form.json()
        print("  Received updated resource data:")
        print(json.dumps(updated_resource, indent=2))
        print(f"  Content-Type header sent: {response_put_form.request.headers.get('Content-Type')}")
    else:
        print(f"  PUT Request failed with status code: {response_put_form.status_code}")
        print(f"  Response content: {response_put_form.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred: {e}")


# --- 2.3. PUT Request with Creation at a Known URI (Simulated) ---
print("\n--- 2.3. PUT Request with Creation at a Known URI (Simulated) ---")

# If an API allows clients to specify the ID for creation, PUT can be used.
# JSONPlaceholder will actually *create* a new resource if the ID doesn't exist,
# but it still calls it a "put" operation. In a real scenario, you'd pick an ID
# that you know is available or for which you have authority.
new_resource_id = 999
put_url_create = f'{BASE_API_URL}/posts/{new_resource_id}'
create_payload = {
    'id': new_resource_id,
    'title': 'Post Created by PUT',
    'body': 'This resource was explicitly created at a specific ID via PUT.',
    'userId': 1
}

print(f"\nSending PUT request to create/replace at specific ID: {put_url_create}")
print(f"  With JSON payload: {create_payload}")
try:
    response_put_create = requests.put(put_url_create, json=create_payload)

    # Note: JSONPlaceholder returns 200 OK even for "creation" via PUT.
    # A real API might return 201 Created if it's genuinely new.
    if response_put_create.status_code in [200, 201]:
        print(f"  PUT Request successful (Status {response_put_create.status_code})!")
        created_resource = response_put_create.json()
        print("  Received resource data:")
        print(json.dumps(created_resource, indent=2))
        if response_put_create.status_code == 201:
            print("  (Server explicitly indicated 'Created')")
        else:
            print("  (Server indicated 'OK', implying creation/replacement)")
    else:
        print(f"  PUT Request failed with status code: {response_put_create.status_code}")
        print(f"  Response content: {response_put_create.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred: {e}")


# --- 2.4. Handling PUT Responses: `raise_for_status()` ---
print("\n--- 2.4. Handling PUT Responses: `raise_for_status()` ---")

# Using `raise_for_status()` is a good practice for error checking.
# Let's simulate a bad request to see it in action.
url_bad_put = 'https://httpbin.org/status/400' # Simulates a 400 Bad Request
invalid_payload = {'invalid_field': 'value'}

print(f"\nSending PUT request to simulate 400 Bad Request: {url_bad_put}")
try:
    response_bad_put = requests.put(url_bad_put, json=invalid_payload)
    response_bad_put.raise_for_status() # This will raise an HTTPError
    print("  This line will NOT be reached if HTTPError occurs.")
except requests.exceptions.HTTPError as e:
    print(f"  Caught HTTPError: {e}")
    print(f"  Status Code: {e.response.status_code}, Reason: {e.response.reason}")
    print(f"  Response content: {e.response.text}")
except requests.exceptions.RequestException as e:
    print(f"  Caught a general RequestException: {e}")

```

## Sending DELETE Requests with Python's `requests` Module: Theory and Code

### 1. Theory: Understanding HTTP DELETE Requests

The HTTP **DELETE** method is a request method used to delete the specified resource on the server.

#### 1.1 Purpose of DELETE

- **Removing Resources:** The primary purpose of a DELETE request is to remove a resource identified by its URI.
- **Idempotent:** DELETE requests are **idempotent**. This is a crucial characteristic. If you send the same DELETE request multiple times to the same URI, the effect on the server will be the same as if you sent it once. After the first successful deletion, subsequent identical DELETE requests to the same URI will typically still result in a success status (e.g., 200 OK or 204 No Content), indicating that the resource is no longer there, even if it was already deleted. They will not cause further changes or errors on the server _due to the act of deletion itself_.
- **No Request Body (Typically):** DELETE requests typically do not have a request body, as the resource to be deleted is identified solely by the URL. While HTTP specifications allow a body, it's generally ignored by servers and is not part of RESTful API best practices for DELETE.

#### 1.2 When to Use `requests.delete()`

You use `requests.delete()` whenever your goal is to remove a specific item or resource from a web server via its API. Common scenarios include:

- Deleting a user account.
- Removing a product from an inventory.
- Deleting a file.
- Removing a specific comment or post.

#### 1.3 Key Arguments for `requests.delete()`

The `requests.delete()` function is relatively simple, primarily requiring the URL of the resource to be deleted:

- **`url` (required):** The exact URI of the resource you want to delete. This URL _must_ identify the specific resource, not a collection.
- **`headers` (optional):** A dictionary of HTTP headers (e.g., `Authorization` for authentication).
- **`timeout` (optional):** How many seconds to wait for the server.
- **`auth` (optional):** For authentication.
- **`proxies` (optional):** For proxy configuration.
- **`allow_redirects` (optional):** Default `True`.

#### 1.4 The `Response` Object

Similar to other `requests` methods, `requests.delete()` returns a `Response` object. Key attributes and methods include:

- **`response.status_code`**: Typically:
  - `200 OK`: If the deletion was successful and the server is returning a response body (e.g., a confirmation message).
  - `202 Accepted`: The request has been accepted for processing, but the processing has not been completed. This is common for operations that take time.
  - `204 No Content`: If the deletion was successful and the server has no content to return in the body (this is the most common and often preferred status for DELETE).
  - `404 Not Found`: If the resource to be deleted does not exist.
  - `403 Forbidden`: If the client does not have permission to delete the resource.
  - `4xx`/`5xx`: For other client or server errors.
- **`response.ok`**: `True` for `2xx` status codes.
- **`response.text`**: Response content as a Unicode string (often empty for `204 No Content`).
- **`response.json()`**: Parses JSON response (if content exists and is JSON).
- **`response.headers`**: Response headers.
- **`response.raise_for_status()`**: Raises `HTTPError` for 4xx/5xx status codes.

### 2. Code Examples

```python
import requests
import json # For pretty-printing JSON responses

print("--- Python `requests` Module: DELETE Request Examples ---")

# Base URL for examples (using JSONPlaceholder and httpbin.org)
# JSONPlaceholder allows simulating DELETE, but doesn't actually remove data permanently.
# httpbin.org/delete reflects the request.
JSON_PLACEHOLDER_URL = 'https://jsonplaceholder.typicode.com'
HTTPBIN_URL = 'https://httpbin.org'

# --- 2.1. Basic DELETE Request to a Specific Resource ---
print("\n--- 2.1. Basic DELETE Request to a Specific Resource ---")

# To simulate deleting a post with ID 1
# In a real application, you'd first ensure the resource exists, then delete.
post_id_to_delete = 1
delete_url = f'{JSON_PLACEHOLDER_URL}/posts/{post_id_to_delete}'

print(f"Sending DELETE request to: {delete_url}")
try:
    response_delete = requests.delete(delete_url)

    # JSONPlaceholder typically returns 200 OK for DELETE, even though no content is sent back.
    # A common response for successful deletion in many APIs is 204 No Content.
    if response_delete.status_code == 200 or response_delete.status_code == 204:
        print(f"  DELETE Request successful (Status {response_delete.status_code})!")
        print(f"  Response content (often empty for DELETE): '{response_delete.text}'")
        # If response.status_code is 204, response.text will be empty.
        # If it's 200 with no content, .text might also be empty or contain minimal JSON.
    else:
        print(f"  DELETE Request failed with status code: {response_delete.status_code}")
        print(f"  Response content: {response_delete.text}")

except requests.exceptions.RequestException as e:
    print(f"  An error occurred during the DELETE request: {e}")


# --- 2.2. DELETE Request and Idempotency (Simulated) ---
print("\n--- 2.2. DELETE Request and Idempotency (Simulated) ---")

# Demonstrate idempotency by sending the same DELETE request multiple times.
# httpbin.org/delete simply reflects the request, so it won't simulate actual deletion.
# We'll rely on the status code for an illustrative example.
print("\nSimulating idempotency for DELETE (using httpbin.org/status/200):")
delete_idempotent_url = f'{HTTPBIN_URL}/status/200' # Simulate a successful delete endpoint

print(f"1. First DELETE request to {delete_idempotent_url}")
try:
    response_delete_1 = requests.delete(delete_idempotent_url)
    if response_delete_1.status_code == 200:
        print("  First delete successful (Status 200 OK).")
    else:
        print(f"  First delete failed: {response_delete_1.status_code}")
except requests.exceptions.RequestException as e:
    print(f"  Error on first delete: {e}")

print(f"\n2. Second (identical) DELETE request to {delete_idempotent_url}")
try:
    response_delete_2 = requests.delete(delete_idempotent_url)
    if response_delete_2.status_code == 200:
        print("  Second delete successful (Status 200 OK).")
        print("  (Demonstrates idempotency: same result as first request).")
    else:
        print(f"  Second delete failed: {response_delete_2.status_code}")
except requests.exceptions.RequestException as e:
    print(f"  Error on second delete: {e}")


# --- 2.3. Handling DELETE Responses with `raise_for_status()` ---
print("\n--- 2.3. Handling DELETE Responses with `raise_for_status()` ---")

# `raise_for_status()` is excellent for concise error handling with DELETE requests.

# Example 1: Successful DELETE (200 OK or 204 No Content) - no error raised
url_success_delete = f'{JSON_PLACEHOLDER_URL}/posts/2'
print(f"Sending DELETE request to {url_success_delete} (simulating success)...")
try:
    response_success_delete = requests.delete(url_success_delete)
    response_success_delete.raise_for_status() # No error for 200/204
    print(f"  DELETE request successful (Status {response_success_delete.status_code}), no error raised.")
except requests.exceptions.RequestException as e:
    print(f"  Caught an unexpected RequestException: {e}")


# Example 2: DELETE request to a non-existent resource (404 Not Found) - error raised
url_not_found_delete = f'{JSON_PLACEHOLDER_URL}/posts/999999' # Likely won't exist
print(f"\nSending DELETE request to non-existent resource: {url_not_found_delete}")
try:
    response_not_found_delete = requests.delete(url_not_found_delete)
    response_not_found_delete.raise_for_status() # This will raise an HTTPError
    print("  This line will NOT be reached if an HTTPError occurs.")
except requests.exceptions.HTTPError as e:
    print(f"  Caught HTTPError for status {e.response.status_code}: {e}")
    print(f"  Reason: {e.response.reason}")
    print(f"  Response content: {e.response.text}") # Might contain an error message from server
except requests.exceptions.RequestException as e:
    print(f"  Caught a general RequestException: {e}")


# --- 2.4. DELETE Request with Authentication (Conceptual) ---
print("\n--- 2.4. DELETE Request with Authentication (Conceptual) ---")

# In real-world scenarios, DELETE operations almost always require authentication.
# This example is conceptual as JSONPlaceholder doesn't require auth.
# Use a valid token for actual APIs.
authenticated_delete_url = f'{JSON_PLACEHOLDER_URL}/posts/3' # Still using JSONPlaceholder for URL structure
auth_headers = {
    'Authorization': 'Bearer YOUR_ACTUAL_ACCESS_TOKEN',
    'X-API-KEY': 'YOUR_API_KEY' # Another common auth header
}

print(f"\nSending authenticated DELETE request to: {authenticated_delete_url}")
print(f"  (Using conceptual Authorization headers)")
try:
    response_auth_delete = requests.delete(authenticated_delete_url, headers=auth_headers)
    if response_auth_delete.ok:
        print(f"  Authenticated DELETE successful (Status {response_auth_delete.status_code})!")
    else:
        print(f"  Authenticated DELETE failed with status code: {response_auth_delete.status_code}")
        print(f"  Response: {response_auth_delete.text}")
        if response_auth_delete.status_code == 401:
            print("  (Likely Unauthorized - check your token/credentials)")
        elif response_auth_delete.status_code == 403:
            print("  (Likely Forbidden - you don't have permission to delete this resource)")
except requests.exceptions.RequestException as e:
    print(f"  An error occurred: {e}")

```
