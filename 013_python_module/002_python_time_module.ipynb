### **Python `time` Module: Overview, Concepts, and Theory**

The `time` module in Python provides a variety of time-related functions, such as working with the current time, measuring time intervals, and manipulating time representations. It is one of the most frequently used modules when dealing with time and performance-related tasks.

The `time` module works with Unix timestamps (the number of seconds since January 1, 1970) and provides functions that help you interact with this time format, perform conversions, and control the flow of time in your programs.

---

### **Key Concepts of the `time` Module:**

1. **Working with Time in Seconds:**

   - In many parts of the `time` module, time is represented in seconds (or fractions of a second). For example, `time.time()` returns the current time in seconds since the Unix epoch.

2. **Time Representation:**

   - Time in Python is often represented as a floating-point number where the integer part represents seconds, and the fractional part represents fractions of a second.

3. **Time Formatting and Parsing:**

   - The module provides functions to format time as human-readable strings and to parse strings back into structured time representations.

4. **Time Measurement:**
   - The `time` module is useful for measuring elapsed time, benchmarking code performance, and adding delays in code execution.

---

### **Commonly Used Functions in the `time` Module:**

#### 1. **Getting the Current Time:**

- **`time.time()`**:

  - Returns the current time in seconds since the epoch (Jan 1, 1970).
  - This is useful for measuring time intervals or getting the current timestamp.

  ```python
  import time
  current_time = time.time()
  print(current_time)  # Output: A floating point number representing the current timestamp
  ```

- **`time.localtime([secs])`**:

  - Converts a time in seconds since the epoch to a struct_time in local time. If `secs` is not provided, it uses the current time.

  ```python
  local_time = time.localtime()
  print(local_time)  # Output: struct_time object
  ```

- **`time.gmtime([secs])`**:
  - Converts a time in seconds since the epoch to a struct_time in UTC (Coordinated Universal Time).
  ```python
  utc_time = time.gmtime()
  print(utc_time)  # Output: struct_time object
  ```

#### 2. **Formatting Time:**

- **`time.strftime(format[, t])`**:

  - Converts a `struct_time` to a string as per the provided format string. The `t` argument is optional and defaults to the current time if not provided.
  - Common format codes:
    - `%Y` – Year (4 digits)
    - `%m` – Month (2 digits)
    - `%d` – Day (2 digits)
    - `%H` – Hour (24-hour format)
    - `%M` – Minute (2 digits)
    - `%S` – Second (2 digits)

  ```python
  formatted_time = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime())
  print(formatted_time)  # Output: Formatted string, e.g., '2025-03-25 12:45:30'
  ```

- **`time.asctime([t])`**:

  - Converts a `struct_time` to a string representing the time in a human-readable format.
  - If no argument is passed, it uses the current local time.

  ```python
  ascii_time = time.asctime()
  print(ascii_time)  # Output: 'Mon Mar 25 12:45:30 2025'
  ```

- **`time.ctime([secs])`**:
  - Converts a time expressed in seconds since the epoch to a string representing the local time.
  - Similar to `time.asctime()`, but takes a timestamp as input.
  ```python
  timestamp = time.time()
  print(time.ctime(timestamp))  # Output: 'Mon Mar 25 12:45:30 2025'
  ```

#### 3. **Time Manipulation and Conversion:**

- **`time.mktime(t)`**:

  - Converts a `struct_time` in local time to seconds since the epoch.

  ```python
  local_time = time.localtime()
  timestamp = time.mktime(local_time)
  print(timestamp)  # Output: A timestamp (seconds since epoch)
  ```

- **`time.sleep(seconds)`**:
  - Suspends the execution of the current program for the specified number of seconds. Useful for adding delays in programs.
  ```python
  print("Start")
  time.sleep(2)
  print("End after 2 seconds")  # Output: "Start", then after 2 seconds, "End after 2 seconds"
  ```

#### 4. **Performance Measurement:**

- **`time.perf_counter()`**:

  - Returns the value of a performance counter, which is typically used to measure short durations. It provides the highest available resolution timer to measure time intervals.

  ```python
  start = time.perf_counter()
  # Your code here...
  end = time.perf_counter()
  print(f"Elapsed time: {end - start} seconds")
  ```

- **`time.process_time()`**:
  - Returns the processor time (CPU time) in seconds, which is useful for measuring the CPU time consumed by a program.
  ```python
  start = time.process_time()
  # Code to measure CPU time
  end = time.process_time()
  print(f"CPU Time: {end - start} seconds")
  ```

#### 5. **Working with Time Zones:**

- **`time.tzname`**:

  - Returns a tuple containing the names of the time zones used for local time and daylight saving time.

  ```python
  print(time.tzname)  # Output: ('Standard Time', 'Daylight Saving Time')
  ```

- **`time.altzone`**:
  - The number of seconds west of UTC for the time zone that is used during daylight saving time.
  ```python
  print(time.altzone)  # Output: Number of seconds west of UTC during DST
  ```

### **Practical Use Cases of the `time` Module:**

1. **Measuring Code Performance:**
   - You can use `time.perf_counter()` or `time.process_time()` to measure how long a specific section of code takes to run. This is useful for benchmarking and performance profiling.
2. **Creating Delays:**

   - Using `time.sleep()`, you can add pauses or delays in the program, which is helpful in various scenarios like rate-limiting API calls or creating timed animations in games.

3. **Time Conversion and Formatting:**

   - The `time` module is frequently used to convert Unix timestamps to human-readable formats, such as logging events with timestamps, and formatting them for display.

4. **Scheduling Tasks:**

5. **Working with Time Zones:**
   - The `time` module provides access to time zone information, which is useful when working with applications that need to adjust for different time zones.

### 1. Getting the Current Time (Epoch Time)

The `time.time()` function returns the current time as a floating-point number, representing the number of seconds since the epoch (January 1, 1970, 00:00:00 (UTC)). This is excellent for measuring durations.

```python
import time

print("--- 1. Current Time (Epoch) ---")
current_epoch_time = time.time()
print(f"Current time in seconds since epoch: {current_epoch_time}")
# Example Output: Current time in seconds since epoch: 1716913833.8828125
```

### 2. Pausing Execution (`time.sleep()`)

The `time.sleep(seconds)` function pauses the execution of the current thread for a specified number of seconds. This is useful for rate limiting, waiting for resources, or just observing timed events.

```python
import time

print("\n--- 2. Pausing Execution (`time.sleep()`) ---")
print("Starting a 3-second countdown...")
for i in range(3, 0, -1):
    print(i, end=' ')
    time.sleep(1) # Pause for 1 second
print("Go!")
```

### 3. Measuring Execution Time

You can use `time.time()` at the beginning and end of a code block to measure how long it takes to execute.

```python

start_time = time.time()

elapsed_time = end_time - start_time

```

### 4. Working with Local Time (`time.localtime()`, `time.mktime()`)

`time.localtime()` converts a time in seconds since the epoch to a `struct_time` object in local time. A `struct_time` object is a tuple-like object with named attributes (e.g., `tm_year`, `tm_mon`, `tm_mday`, `tm_hour`, `tm_min`, `tm_sec`, `tm_wday`, `tm_yday`, `tm_isdst`).

`time.mktime()` is the inverse: it converts a `struct_time` (representing local time) back into seconds since the epoch.

```python
import time

print("\n--- 4. Local Time (`time.localtime()`, `time.mktime()`) ---")

# Get current time as struct_time in local timezone
current_local_time = time.localtime()
print(f"Current local time (struct_time): {current_local_time}")
print(f"Year: {current_local_time.tm_year}")
print(f"Month: {current_local_time.tm_mon}")
print(f"Day: {current_local_time.tm_mday}")
print(f"Hour: {current_local_time.tm_hour}")
print(f"Minute: {current_local_time.tm_min}")
print(f"Second: {current_local_time.tm_sec}")
print(f"Weekday (0=Mon, 6=Sun): {current_local_time.tm_wday}")
print(f"Day of year: {current_local_time.tm_yday}")
print(f"DST (0=No, 1=Yes, -1=Unknown): {current_local_time.tm_isdst}")


# Convert struct_time back to epoch time
epoch_from_struct = time.mktime(current_local_time)
print(f"Epoch time from struct_time: {epoch_from_struct}")
print(f"Original epoch time: {time.time()}") # Will be slightly different due to elapsed time
```

### 6. Formatting Time (`time.strftime()`, `time.strptime()`)

`time.strftime(format, t)` converts a `struct_time` object (or a tuple representing it) into a formatted string.
`time.strptime(string, format)` parses a string representing a time according to a format and returns a `struct_time` object.

A list of format codes can be found in the Python documentation (e.g., `%Y` for full year, `%m` for month, `%d` for day, `%H` for hour (24-hour), `%M` for minute, `%S` for second, `%A` for full weekday name, `%B` for full month name).

```python

# Format current local time
current_local_struct = time.localtime()

print(f"Formatted local time: {formatted_time}")
# Example Output: Formatted local time: 2025-05-28 22:30:33 Wednesday

```

### 7. Performance Counter (`time.perf_counter()`)

`time.perf_counter()` returns the value (in fractional seconds) of a high-resolution performance counter that is useful for measuring the **shortest durations**. It includes time elapsed during sleep and is system-wide.

```python
import time

print("\n--- 7. Performance Counter (`time.perf_counter()`) ---")

# Best for measuring small code execution times
start_perf = time.perf_counter()
_ = [x * x for x in range(100_000)] # A simple list comprehension
end_perf = time.perf_counter()
print(f"List comprehension time (perf_counter): {end_perf - start_perf:.6f} seconds")

# Contrast with time.process_time() if you need CPU time only (not elapsed time including sleeps)
start_process = time.process_time()
time.sleep(0.1) # This sleep time will NOT be counted by process_time
_ = [x * x for x in range(100_000)]
end_process = time.process_time()
print(f"List comprehension time (process_time, excludes sleep): {end_process - start_process:.6f} seconds")
```

### Key Takeaways for `time` Module:

- **`time.time()`**: For measuring overall wall-clock time between two points.
- **`time.sleep()`**: To pause program execution.
- **`time.perf_counter()`**: For precise short-duration timing measurements (e.g., benchmarking small code snippets).
- **`time.localtime()`, `time.gmtime()`, `time.strftime()`**: For converting between epoch time and human-readable/formatted time strings, considering local timezone or UTC.

While the `time` module is great for these tasks, for more complex date and time arithmetic, timezone conversions, and general date manipulation, the `datetime` module is usually more powerful and flexible.


````python
import time

print("--- 1. Benchmarking Code Execution ---")

def factorial_iterative(n):
    """Calculates factorial iteratively."""
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res

def factorial_recursive(n):
    """Calculates factorial recursively."""
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial_recursive(n - 1)

test_number = 1000

# Benchmark iterative approach
start_time_iter = time.perf_counter()
result_iter = factorial_iterative(test_number)
end_time_iter = time.perf_counter()
print(f"Iterative Factorial ({test_number}!):")
print(f"  Result (first 50 digits): {str(result_iter)[:50]}...")
print(f"  Time taken: {end_time_iter - start_time_iter:.6f} seconds\n")

# Benchmark recursive approach
# Note: For large numbers, recursion might hit Python's recursion limit
# or be significantly slower due to function call overhead.
try:
    start_time_rec = time.perf_counter()
    result_rec = factorial_recursive(test_number)
    end_time_rec = time.perf_counter()
    print(f"Recursive Factorial ({test_number}!):")
    print(f"  Result (first 50 digits): {str(result_rec)[:50]}...")
    print(f"  Time taken: {end_time_rec - start_time_rec:.6f} seconds")
except RecursionError:
    print(f"Recursive Factorial ({test_number}!) skipped: Recursion limit reached.")
````

````python
import time
import requests # Assuming 'requests' library for API calls

print("\n--- 2. Rate Limiting / Throttling API Calls ---")

API_URL = "https://jsonplaceholder.typicode.com/todos/1" # A free public API
REQUESTS_PER_SECOND = 2
MIN_DELAY_BETWEEN_REQUESTS = 1 / REQUESTS_PER_SECOND

def fetch_data_with_rate_limit(url):
    last_request_time = 0

    for i in range(5): # Make 5 API calls
        current_time = time.time()
        elapsed = current_time - last_request_time

        if elapsed < MIN_DELAY_BETWEEN_REQUESTS:
            sleep_duration = MIN_DELAY_BETWEEN_REQUESTS - elapsed
            print(f"Sleeping for {sleep_duration:.2f}s to respect rate limit...")
            time.sleep(sleep_duration)

        try:
            response = requests.get(url)
            response.raise_for_status() # Raise an exception for bad status codes
            data = response.json()
            print(f"Request {i+1}: Successfully fetched data (Title: {data['title'][:20]}...)")
            last_request_time = time.time() # Update last request time AFTER successful call
        except requests.exceptions.RequestException as e:
            print(f"Request {i+1}: Error fetching data - {e}")
            last_request_time = time.time() # Still update time to prevent rapid retries on error
````

````python
import time
import datetime # Often used alongside time for more robust date/time objects

print("\n--- More Practice with Python's `time` module ---")

# --- 1. Getting Current Time ---
print("\n--- 1. Getting Current Time ---")

# time.time() - Returns the current time in seconds since the epoch (float)
current_epoch_time = time.time()
print(f"Current epoch time (seconds since epoch): {current_epoch_time}")

# time.gmtime() - Get current time as struct_time in UTC (GMT)
current_utc_time_struct = time.gmtime()
print(f"Current UTC time (struct_time): {current_utc_time_struct}")
print(f"UTC Year: {current_utc_time_struct.tm_year}, UTC Hour: {current_utc_time_struct.tm_hour}")

# time.localtime() - Get current time as struct_time in local timezone (already covered, but good to recall)
current_local_time_struct = time.localtime()
print(f"Current Local time (struct_time): {current_local_time_struct}")
print(f"Local Year: {current_local_time_struct.tm_year}, Local Hour: {current_local_time_struct.tm_hour}")

# --- 2. Time Delays (`time.sleep()`) ---
print("\n--- 2. Time Delays (`time.sleep()`) ---")

print("Pausing for 2 seconds...")
start_sleep_time = time.time()
time.sleep(2) # Pauses execution for 2 seconds
end_sleep_time = time.time()
print(f"Resumed after {end_sleep_time - start_sleep_time:.2f} seconds.")

print("Counting down...")
for i in range(3, 0, -1):
    print(f"{i}...")
    time.sleep(1) # Pause for 1 second between counts
print("Blast off!")

# --- 3. Formatting Time (`time.strftime()`) ---
print("\n--- 3. Formatting Time (`time.strftime()`) ---")
# time.strftime(format, t) converts a struct_time or tuple to a string
# If 't' is omitted, time.localtime() is used by default.

# Common format codes:
# %Y: Year with century (e.g., 2025)
# %m: Month as a zero-padded decimal number (01-12)
# %d: Day of month as a zero-padded decimal number (01-31)
# %H: Hour (24-hour clock) as a zero-padded decimal number (00-23)
# %M: Minute as a zero-padded decimal number (00-59)
# %S: Second as a zero-padded decimal number (00-59)
# %w: Weekday as a decimal number (0-6, Sunday is 0)
# %a: Abbreviated weekday name (e.g., Thu)
# %A: Full weekday name (e.g., Thursday)
# %b: Abbreviated month name (e.g., May)
# %B: Full month name (e.g., May)
# %j: Day of year as a zero-padded decimal number (001-366)
# %Z: Time zone name (e.g., Bangladesh Standard Time)
# %z: UTC offset (e.g., +0600)
# %x: Local date representation
# %X: Local time representation
# %c: Local date and time representation

# Using current local time
formatted_date = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime())
print(f"Formatted local date/time: {formatted_date}")

formatted_full = time.strftime("%A, %B %d, %Y %H:%M:%S %Z%z", time.localtime())
print(f"Formatted full local date/time: {formatted_full}")

# Using a specific struct_time (e.g., from gmtime)
specific_struct_time = (2024, 7, 15, 14, 30, 0, 0, 197, 0) # (Year, Month, Day, Hour, Min, Sec, Wday, Yday, Isdst)
formatted_specific = time.strftime("%d %b %Y %I:%M %p UTC", specific_struct_time)
print(f"Formatted specific UTC time: {formatted_specific}")

# Unix-like date format (common in logs/cli)
unix_format = time.strftime("%a %b %d %H:%M:%S %Y", time.localtime())
print(f"Unix-like date: {unix_format}")

# --- 4. Parsing Time (`time.strptime()`) ---
print("\n--- 4. Parsing Time (`time.strptime()`) ---")
# time.strptime(string, format) converts a string to a struct_time object

date_string = "2023-10-26 15:45:30"
format_string = "%Y-%m-%d %H:%M:%S"

try:
    parsed_time_struct = time.strptime(date_string, format_string)
    print(f"Parsed struct_time from string: {parsed_time_struct}")
    print(f"Parsed Year: {parsed_time_struct.tm_year}, Parsed Month: {parsed_time_struct.tm_mon}")
except ValueError as e:
    print(f"Error parsing date string: {e}")

# Example with a different format
another_date_str = "Dec 25, 2024 10:30 AM"
another_format_str = "%b %d, %Y %I:%M %p" # %I for 12-hour, %p for AM/PM
try:
    parsed_another_time = time.strptime(another_date_str, another_format_str)
    print(f"Parsed another struct_time: {parsed_another_time}")
except ValueError as e:
    print(f"Error parsing another date string: {e}")

# --- 5. Converting between Epoch Time and struct_time ---
print("\n--- 5. Converting between Epoch Time and struct_time ---")

# time.mktime() - Converts struct_time (local) to epoch time
# (Already covered, but reinforcing)
local_struct = time.localtime()
epoch_from_local_struct = time.mktime(local_struct)
print(f"Epoch from local struct: {epoch_from_local_struct}")

# time.gmtime() - Converts epoch time to struct_time (UTC)
epoch_time_now = time.time()
utc_struct_from_epoch = time.gmtime(epoch_time_now)
print(f"UTC struct from epoch: {utc_struct_from_epoch}")

# time.localtime() - Converts epoch time to struct_time (Local)
local_struct_from_epoch = time.localtime(epoch_time_now)
print(f"Local struct from epoch: {local_struct_from_epoch}")

# --- 6. Measuring Execution Time ---
print("\n--- 6. Measuring Execution Time ---")

def perform_some_task():
    """A dummy function to simulate a task."""
    total = 0
    for i in range(1_000_000): # One million iterations
        total += i * 2
    return total

start_task_time = time.time() # Get time before task
result = perform_some_task()
end_task_time = time.time()   # Get time after task

elapsed_time = end_task_time - start_task_time
print(f"Task completed. Result: {result}")
print(f"Time taken for task: {elapsed_time:.4f} seconds")

# For more precise measurements in performance-critical code,
# consider `time.perf_counter()` (high-resolution timer, includes sleep time)
# or `time.process_time()` (CPU time, doesn't include sleep time).

print("\n--- Using time.perf_counter() for more precise timing ---")
start_perf_time = time.perf_counter()
time.sleep(0.1) # Simulate a small pause
result_perf = perform_some_task()
end_perf_time = time.perf_counter()
elapsed_perf_time = end_perf_time - start_perf_time
print(f"Task result (perf_counter): {result_perf}")
print(f"Time taken (perf_counter, including sleep): {elapsed_perf_time:.6f} seconds")

# --- 7. Working with struct_time attributes directly ---
print("\n--- 7. Working with struct_time attributes directly ---")

# Accessing attributes
current_t = time.localtime()
print(f"Current Year: {current_t.tm_year}")
print(f"Current Month: {current_t.tm_mon}")
print(f"Current Day of Week (0=Mon, 6=Sun): {current_t.tm_wday}") # tm_wday
print(f"Is Daylight Saving Time in effect? {bool(current_t.tm_isdst)}")

# Example: Get the day of the week name
weekday_names = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
print(f"Today is: {weekday_names[current_t.tm_wday]}")

# --- 8. Using `datetime` for more complex date/time operations (often preferred) ---
print("\n--- 8. Using `datetime` for more complex date/time operations ---")
# While `time` module is fundamental, `datetime` module offers more user-friendly
# objects (datetime, date, time, timedelta) and functionalities.

# Get current datetime object
now_dt = datetime.datetime.now()
print(f"Current datetime object: {now_dt}")

# Format datetime object (similar to strftime)
formatted_dt = now_dt.strftime("%Y-%m-%d %H:%M:%S")
print(f"Formatted datetime: {formatted_dt}")

# Create a specific datetime object
specific_dt = datetime.datetime(2025, 12, 25, 9, 0, 0)
print(f"Specific datetime: {specific_dt}")

# Calculate time difference (timedelta)
time_difference = specific_dt - now_dt
print(f"Time until Christmas 2025: {time_difference}")
print(f"Days: {time_difference.days}")
print(f"Seconds: {time_difference.total_seconds()}")

# Convert struct_time to datetime object
struct_time_to_dt = datetime.datetime(*time.localtime()[:6]) # Only pass year, month, day, hour, min, sec
print(f"Converted struct_time to datetime: {struct_time_to_dt}")
````
