### 1. Unpacking Iterables (Tuple/List Unpacking)

**Theory:** Python allows you to unpack elements from iterables (like tuples and lists) directly into variables. This makes code cleaner than accessing elements by index. The number of variables on the left must match the number of elements in the iterable.

**Code:**

```python
# Basic unpacking
point = (10, 20)
x, y = point
print(f"Point coordinates: x={x}, y={y}") # Output: Point coordinates: x=10, y=20

# Unpacking in loops
data = [('Alice', 30), ('Bob', 24)]
for name, age in data:
    print(f"{name} is {age} years old.")
# Output:
# Alice is 30 years old.
# Bob is 24 years old.

# Swapping variables (without a temp variable)
a = 5
b = 10
a, b = b, a
print(f"Swapped: a={a}, b={b}") # Output: Swapped: a=10, b=5

# Unpacking with the '*' operator (Python 3.x) for arbitrary length
# The '*' collects remaining items into a list.
first, *middle, last = [1, 2, 3, 4, 5, 6]
print(f"First: {first}, Middle: {middle}, Last: {last}") # Output: First: 1, Middle: [2, 3, 4, 5], Last: 6

*start, end = [1, 2, 3]
print(f"Start: {start}, End: {end}") # Output: Start: [1, 2], End: 3

# Function return values
def get_user_info():
    return "Charlie", 35, "charlie@example.com"

name, age, email = get_user_info()
print(f"User info: {name}, {age}, {email}") # Output: User info: Charlie, 35, charlie@example.com
```

### 2. List Comprehensions (and Generator Expressions)

**Theory:** A concise way to create lists (or other iterables) based on existing iterables. They are often more readable and faster than traditional `for` loops for simple transformations or filtering. Generator expressions are similar but return an iterator, saving memory for very large datasets.

**Code:**

```python
# List comprehension: squares of numbers
numbers = [1, 2, 3, 4, 5]
squares = [n**2 for n in numbers]
print(f"Squares: {squares}") # Output: Squares: [1, 4, 9, 16, 25]

# List comprehension with condition (filtering evens)
evens = [n for n in numbers if n % 2 == 0]
print(f"Evens: {evens}") # Output: Evens: [2, 4]

# List comprehension with if-else (conditional assignment)
parity_labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(f"Parity Labels: {parity_labels}") # Output: Parity Labels: ['odd', 'even', 'odd', 'even', 'odd']

# Nested list comprehensions (flattening a 2D list)
matrix = [[1, 2], [3, 4], [5, 6]]
flat_list = [num for row in matrix for num in row]
print(f"Flattened list: {flat_list}") # Output: Flattened list: [1, 2, 3, 4, 5, 6]

# Dictionary comprehension
squares_dict = {n: n**2 for n in numbers}
print(f"Squares Dict: {squares_dict}") # Output: Squares Dict: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Set comprehension
unique_letters = {char for char in "hello world" if char.isalpha()}
print(f"Unique Letters: {unique_letters}") # Output: Unique Letters: {'d', 'o', 'r', 'l', 'e', 'h', 'w'} (order may vary)

# Generator expression (uses parentheses instead of square brackets)
# Doesn't create the full list in memory, yields values one by one.
gen_squares = (n**2 for n in numbers)
print(f"Generator squares (object): {gen_squares}") # Output: Generator squares (object): <generator object <genexpr> at 0x...>
print(f"First two from generator: {next(gen_squares)}, {next(gen_squares)}")
print(f"Remaining as list: {list(gen_squares)}") # Output: First two from generator: 1, 4\nRemaining as list: [9, 16, 25]
```

### 3. `zip()` Function

**Theory:** `zip()` is used to combine elements from multiple iterables (lists, tuples, etc.) into an iterator of tuples. The $i$-th tuple contains the $i$-th element from each of the input iterables. It stops when the shortest iterable is exhausted.

**Code:**

```python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 22]
cities = ["New York", "London", "Paris"]

# Zipping multiple lists
for name, age, city in zip(names, ages, cities):
    print(f"{name} is {age} and lives in {city}.")
# Output:
# Alice is 25 and lives in New York.
# Bob is 30 and lives in London.
# Charlie is 22 and lives in Paris.

# Zipping with different lengths (stops at shortest)
colors = ["red", "green", "blue"]
hex_codes = ["#FF0000", "#00FF00", "#0000FF", "#CCCCCC"]
for color, hex_code in zip(colors, hex_codes):
    print(f"Color: {color}, Hex: {hex_code}")
# Output:
# Color: red, Hex: #FF0000
# Color: green, Hex: #00FF00
# Color: blue, Hex: #0000FF

# Unzipping (using zip(*) - effectively transpose)
zipped_data = [('Alice', 25), ('Bob', 30), ('Charlie', 22)]
names, ages = zip(*zipped_data)
print(f"Names: {list(names)}") # Output: Names: ['Alice', 'Bob', 'Charlie']
print(f"Ages: {list(ages)}")   # Output: Ages: [25, 30, 22]
```

### 4. `enumerate()` Function

**Theory:** When you need both the index and the value while iterating over a sequence, `enumerate()` is more Pythonic and efficient than manually managing an index counter.

**Code:**

```python
fruits = ["apple", "banana", "cherry"]

# Using enumerate
for index, fruit in enumerate(fruits):
    print(f"Item {index}: {fruit}")
# Output:
# Item 0: apple
# Item 1: banana
# Item 2: cherry

# Starting index from a different number
for i, fruit in enumerate(fruits, start=1):
    print(f"Product {i}: {fruit}")
# Output:
# Product 1: apple
# Product 2: banana
# Product 3: cherry
```

### 5. `collections.defaultdict`

**Theory:** A subclass of `dict` that calls a `default_factory` (a function returning a default value) when a `KeyError` is encountered. This avoids the need to check if a key exists before appending to a list or incrementing a counter.

**Code:**

```python
from collections import defaultdict

# Grouping items
data_points = [
    ('fruit', 'apple'), ('color', 'red'), ('fruit', 'banana'),
    ('color', 'green'), ('animal', 'dog'), ('fruit', 'orange')
]

grouped_data = defaultdict(list) # Default factory is list
for category, item in data_points:
    grouped_data[category].append(item)

print(f"Grouped data: {grouped_data}")
# Output: Grouped data: defaultdict(<class 'list'>, {'fruit': ['apple', 'banana', 'orange'], 'color': ['red', 'green'], 'animal': ['dog']})

# Counting occurrences
word_counts = defaultdict(int) # Default factory is int (defaults to 0)
sentence = "hello world hello python"
for word in sentence.split():
    word_counts[word] += 1

print(f"Word counts: {word_counts}")
# Output: Word counts: defaultdict(<class 'int'>, {'hello': 2, 'world': 1, 'python': 1})
```

### 6. `collections.Counter`

**Theory:** A subclass of `dict` designed for counting hashable objects. It's a convenient way to get frequency counts.

**Code:**

```python
from collections import Counter

# Counting elements in a list
my_list = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
counts = Counter(my_list)
print(f"Fruit counts: {counts}") # Output: Fruit counts: Counter({'apple': 3, 'banana': 2, 'orange': 1})

# Counting characters in a string
char_counts = Counter("Mississippi")
print(f"Character counts: {char_counts}") # Output: Character counts: Counter({'i': 4, 's': 4, 'p': 2, 'M': 1})

# Common operations
print(f"Most common 2: {counts.most_common(2)}") # Output: Most common 2: [('apple', 3), ('banana', 2)]
print(f"Count of 'banana': {counts['banana']}")  # Output: Count of 'banana': 2
print(f"Count of 'grape': {counts['grape']}")    # Output: Count of 'grape': 0 (returns 0 for missing)
```

### 7. `f-strings` (Formatted String Literals - Python 3.6+)

**Theory:** A way to embed expressions inside string literals, using curly braces `{}`. They offer a concise and readable way to format strings, often outperforming older methods like `%` formatting or `str.format()`.

**Code:**

```python
name = "Alice"
age = 30
pi = 3.14159265

# Basic f-string
print(f"Hello, {name}! You are {age} years old.") # Output: Hello, Alice! You are 30 years old.

# Expressions inside f-strings
print(f"Next year, {name} will be {age + 1}.") # Output: Next year, Alice will be 31.
print(f"Pi doubled: {pi * 2}")               # Output: Pi doubled: 6.2831853

# Formatting (precision, padding, dates)
print(f"Pi with 2 decimal places: {pi:.2f}") # Output: Pi with 2 decimal places: 3.14
print(f"Age padded: {age:03d}")             # Output: Age padded: 030
print(f"Name centered: {name:^15}")        # Output: Name centered:    Alice

from datetime import datetime
now = datetime.now()
print(f"Current date and time: {now:%Y-%m-%d %H:%M:%S}") # Output: Current date and time: 2025-06-11 22:44:07 (approx)

# Debugging expressions (Python 3.8+)
value = 123
print(f"{value=}") # Output: value=123
```

### 8. `_` (Underscore) for Throwaway Variables

**Theory:** The single underscore `_` is a common convention to indicate a variable that you don't intend to use. It improves code readability by signaling that a value is being unpacked or iterated over, but its specific content is not needed.

**Code:**

```python
# Unpacking, ignoring values
data = ("item_id_123", "Product Name", 100, 25.50)
product_id, product_name, _, price = data # Ignore count and category if not needed
print(f"Product ID: {product_id}, Name: {product_name}, Price: {price}") # Output: Product ID: item_id_123, Name: Product Name, Price: 25.5

# Looping a fixed number of times (when index isn't needed)
for _ in range(3):
    print("Performing operation...")
# Output:
# Performing operation...
# Performing operation...
# Performing operation...

# In tuple unpacking from a function
def get_user_status():
    return "active", "john_doe", "online"

status, _, _ = get_user_status() # Only interested in status
print(f"User status: {status}") # Output: User status: active
```

### 9. `else` Clause with Loops (`for...else`, `while...else`)

**Theory:** The `else` block associated with a `for` or `while` loop executes _only if the loop completes normally_ (i.e., not terminated by a `break` statement). This is handy for "found/not found" scenarios.

**Code:**

```python
# for...else
items = [1, 2, 3, 4, 5]

# Example 1: Loop completes
for item in items:
    if item == 6:
        print("Found 6!")
        break
else:
    print("6 not found in the list.") # This will execute
# Output: 6 not found in the list.

# Example 2: Loop breaks
for item in items:
    if item == 3:
        print("Found 3!")
        break
else:
    print("3 not found in the list.") # This will NOT execute
# Output: Found 3!

# while...else
count = 0
while count < 3:
    print(f"Count is {count}")
    count += 1
    if count == 5: # This condition will never be met
        break
else:
    print("While loop completed without break.") # This will execute
# Output:
# Count is 0
# Count is 1
# Count is 2
# While loop completed without break.
```

### 10. Context Managers (`with` statement)

**Theory:** The `with` statement ensures that resources (like file handles, network connections, locks) are properly acquired and released, even if errors occur. It uses objects that implement the context manager protocol (`__enter__` and `__exit__` methods). This prevents resource leaks.

**Code:**

```python
# Opening and automatically closing files
file_name = "my_data.txt"

# Create a dummy file
with open(file_name, "w") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")

print(f"Reading from {file_name} using with statement:")
with open(file_name, "r") as f:
    content = f.read()
    print(content)
# No need to call f.close() - it's handled automatically

# Example with a potential error (still cleans up)
try:
    with open(file_name, "r") as f:
        data = f.read()
        print("File read successfully.")
        raise ValueError("Simulating an error inside the 'with' block.")
except ValueError as e:
    print(f"Caught an error: {e}")
# The file is still guaranteed to be closed
print("File resource was safely released despite the error.")

# Custom context manager (illustrative)
class MyContext:
    def __enter__(self):
        print("Entering context...")
        return self # Value bound to 'as' variable

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Exiting context...")
        if exc_type:
            print(f"An exception occurred: {exc_type.__name__}: {exc_val}")
            return False # Propagate the exception
        return True # Suppress the exception if True

print("\nUsing custom context manager:")
with MyContext() as mc:
    print("Inside the custom context.")
    # raise TypeError("Oops, an error!") # Uncomment to see exception handling
print("After custom context.")
```

### 11. Walrus Operator (`:=`) (Assignment Expressions - Python 3.8+)

**Theory:** The walrus operator allows you to assign a value to a variable as part of a larger expression. This can lead to more compact and sometimes more readable code by avoiding redundant computations or multiple lines for simple checks.

**Code:**

```python
# Traditional approach
# value = get_some_value()
# if value:
#     print(f"Value is {value}")

# Using walrus operator (assign and check)
def get_some_value(success=True):
    return "Data fetched!" if success else None

print("\n--- Walrus Operator (Assignment Expressions) ---")
if data := get_some_value(): # Assigns 'Data fetched!' to data, then checks if it's truthy
    print(f"Fetched data: {data}") # Output: Fetched data: Data fetched!

# In a loop (e.g., reading chunks from a file)
def read_chunk(file_object, size):
    return file_object.read(size)

content_to_read = "This is a long string to demonstrate reading in chunks."
with open("temp_walrus.txt", "w") as f:
    f.write(content_to_read)

print("Reading file in chunks:")
with open("temp_walrus.txt", "r") as f:
    while chunk := read_chunk(f, 10): # Assigns 10 chars to chunk, then checks if chunk is not empty
        print(f"Read chunk: '{chunk}'")
# Output:
# Read chunk: 'This is a '
# Read chunk: 'long strin'
# Read chunk: 'g to demon'
# Read chunk: 'strate rea'
# Read chunk: 'ding in ch'
# Read chunk: 'unks.'

import os
os.remove("temp_walrus.txt") # Clean up

# Avoiding recomputing values
results = [10, 5, 0, 8, 3]
successful_results = [result for r in results if (result := r * 2) > 10]
# Here, 'result' is computed once and then used in the condition and the list element.
print(f"Original results: {results}")
print(f"Successful results (doubled and >10): {successful_results}") # Output: Successful results (doubled and >10): [20, 16]
```

### 12. `any()` and `all()` Functions

**Theory:** These built-in functions provide a concise way to check conditions across an iterable of boolean values.

- `any(iterable)`: Returns `True` if at least one element in the iterable is true.
- `all(iterable)`: Returns `True` if all elements in the iterable are true.

**Code:**

```python
numbers = [1, 0, 5, -2, 8]
booleans = [True, False, True]

print("\n--- any() and all() ---")
# Check if any number is positive
if any(n > 0 for n in numbers): # Using a generator expression
    print("At least one number is positive.") # Output: At least one number is positive.

# Check if all numbers are positive
if all(n > 0 for n in numbers):
    print("All numbers are positive.")
else:
    print("Not all numbers are positive.") # Output: Not all numbers are positive.

# With a list of booleans
print(f"any({booleans}): {any(booleans)}") # Output: any([True, False, True]): True
print(f"all({booleans}): {all(booleans)}") # Output: all([True, False, True]): False

empty_list = []
print(f"any({empty_list}): {any(empty_list)}") # Output: any([]): False
print(f"all({empty_list}): {all(empty_list)}") # Output: all([]): True (Vacuously true)
```

### 13. Using `_` for Thousands Separator (Numeric Literals - Python 3.6+)

**Theory:** You can use underscores `_` within numeric literals (integers and floats) to improve readability. Python ignores the underscores when interpreting the number.

**Code:**

```python
print("\n--- Underscore in Numeric Literals ---")
population = 1_000_000
large_number = 123_456_789_012_345.678_9
binary_value = 0b1011_0101_0001
hex_value = 0xFF_EE_DD

print(f"Population: {population}") # Output: Population: 1000000
print(f"Large Number: {large_number}") # Output: Large Number: 123456789012345.6789
print(f"Binary Value: {binary_value}") # Output: Binary Value: 2901
print(f"Hex Value: {hex_value}")     # Output: Hex Value: 16776413
```

### 14. `collections.namedtuple`

**Theory:** Provides a way to create simple, lightweight object types. Named tuples are immutable (like regular tuples), but their fields are accessible by name (like dictionaries or objects) instead of just index. This makes the code more readable and self-documenting.

**Code:**

```python
from collections import namedtuple

# Define a Point named tuple
Point = namedtuple('Point', ['x', 'y']) # Or Point = namedtuple('Point', 'x y')

p1 = Point(10, 20)
print(f"\n--- collections.namedtuple ---")
print(f"Point: {p1}")        # Output: Point: Point(x=10, y=20)
print(f"X coordinate: {p1.x}") # Access by name: Output: X coordinate: 10
print(f"Y coordinate: {p1.y}") # Access by name: Output: Y coordinate: 20
print(f"As tuple: {p1[0]}, {p1[1]}") # Still accessible by index

# Immutable nature
try:
    p1.x = 30
except AttributeError as e:
    print(f"Error trying to modify namedtuple: {e}") # Output: Error trying to modify namedtuple: can't set attribute

# Convert to dictionary
print(f"Point as dict: {p1._asdict()}") # Output: Point as dict: {'x': 10, 'y': 20}

# Creating from iterable
p2 = Point._make([5, 15])
print(f"Point from list: {p2}") # Output: Point from list: Point(x=5, y=15)
```
