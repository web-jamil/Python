### **Functions in the `re` Module**

Here are the most commonly used functions in the `re` module:

#### 1. **`re.match(pattern, string)`**

- Tries to match the pattern from the beginning of the string.
- Returns a match object if the pattern is found at the beginning; otherwise, returns `None`.

```python
import re
result = re.match(r'hello', 'hello world')
if result:
    print("Matched!")
```

#### 2. **`re.search(pattern, string)`**

- Searches the string for the first location where the pattern matches.
- Returns a match object if a match is found; otherwise, returns `None`.

```python
result = re.search(r'world', 'hello world')
if result:
    print("Match found!")
```

#### 3. **`re.findall(pattern, string)`**

- Returns all non-overlapping matches of the pattern in the string as a list.
- Useful for finding all occurrences of a pattern in the text.

```python
result = re.findall(r'\d+', 'I have 2 apples and 10 bananas')
print(result)  # ['2', '10']
```

#### 4. **`re.finditer(pattern, string)`**

- Similar to `findall()`, but returns an iterator yielding match objects for each match.
- Useful for accessing more detailed match information, like position and group.

```python
result = re.finditer(r'\d+', 'I have 2 apples and 10 bananas')
for match in result:
    print(match.group())  # '2', '10'
```

#### 5. **`re.sub(pattern, repl, string)`**

- Replaces occurrences of the pattern in the string with the specified replacement (`repl`).

```python
result = re.sub(r'\d+', 'number', 'I have 2 apples and 10 bananas')
print(result)  # 'I have number apples and number bananas'
```

- Splits the string at each match of the pattern.

```python
result = re.split(r'\s+', 'This is a test')
print(result)  # ['This', 'is', 'a', 'test']
```

- Compiles a regular expression pattern into a regex object, which can be used multiple times for matching, searching, etc.
- It improves performance when you need to use the same pattern multiple times.

```python
regex = re.compile(r'\d+')
result = regex.findall('123 abc 456')
print(result)  # ['123', '456']
```

### **Advanced Concepts in Regex**

#### 1. **Groups and Capturing**

- You can group parts of your regular expression using parentheses `()`. You can later refer to these groups in your program.
- Groups are numbered starting from 1 based on their position in the regex.

```python
result = re.search(r'(\d+)\s+(\w+)', '123 apples')
if result:
    print(result.group(1))  # '123'
    print(result.group(2))  # 'apples'
```

#### 2. **Non-Capturing Groups**

- If you want to group part of the pattern without capturing it (i.e., not assigning a group number), use `(?:...)`.

```python
result = re.search(r'(?:\d+)\s+(\w+)', '123 apples')
print(result.group(1))  # 'apples'
```

#### 3. **Lookahead and Lookbehind Assertions**

- **Positive Lookahead** (`(?=...)`): Matches a group only if it’s followed by a certain pattern.
- **Negative Lookahead** (`(?!...)`): Matches a group only if it’s not followed by a certain pattern.
- **Positive Lookbehind** (`(?<=...)`): Matches a group only if it’s preceded by a certain pattern.
- **Negative Lookbehind** (`(?<!...)`): Matches a group only if it’s not preceded by a certain pattern.

```python
# Positive lookahead example
result = re.search(r'\d+(?=\s+apples)', '123 apples and 456 oranges')
print(result.group())  # '123'
```

#### 4. **Flags in Regex**

Flags are used to modify the behavior of the regex matching:

- `re.IGNORECASE` or `re.I`: Case-insensitive matching.
- `re.MULTILINE` or `re.M`: Multiline matching where `^` and `$` match the start/end of each line.
- `re.DOTALL` or `re.S`: Makes the `.` (dot) match newlines as well.
- `re.VERBOSE` or `re.X`: Allows you to write more readable regex patterns with comments and line breaks.

```python
result = re.search(r'^hello', 'Hello world', re.IGNORECASE)
print(result.group())  # 'Hello'
```

Regular expressions (regex or regexp) are powerful tools for pattern matching and manipulation of strings. In Python, the `re` module provides full support for regular expressions.

### What is a Regular Expression?

At its core, a regular expression is a sequence of characters that defines a search pattern. When you search for a pattern in text, you can use these expressions to describe what you're looking for, rather than explicitly listing every possible string.

Think of it like a wild-card search, but much more sophisticated.

### Basic Concepts and Syntax

Here's a breakdown of common regex syntax elements:

**1. Literals:**
Most characters match themselves literally.

- `a`: Matches the character 'a'.
- `123`: Matches the string "123".

**2. Metacharacters (Special Characters):**
These characters have special meanings in regex.

- **`.` (Dot):** Matches any single character (except newline by default, unless `re.DOTALL` flag is used).

  - `a.b`: Matches "aab", "acb", "axb", etc.

- **`^` (Caret):** Matches the start of the string (or start of a line if `re.MULTILINE` is used).

  - `^Hello`: Matches strings that start with "Hello".

- **`$` (Dollar):** Matches the end of the string (or end of a line if `re.MULTILINE` is used).

  - `World$`: Matches strings that end with "World".

- **`*` (Asterisk):** Matches the preceding character zero or more times (greedy).

  - `ab*c`: Matches "ac", "abc", "abbc", "abbbc", etc.

- **`+` (Plus):** Matches the preceding character one or more times (greedy).

  - `ab+c`: Matches "abc", "abbc", "abbbc", etc. (but not "ac").

- **`?` (Question Mark):** Matches the preceding character zero or one time (greedy). Also used for non-greedy matching (see below).

  - `ab?c`: Matches "ac", "abc".

- **`{m}` (Exact Count):** Matches the preceding character exactly `m` times.

  - `a{3}`: Matches "aaa".

- **`{m,n}` (Range Count):** Matches the preceding character at least `m` times and at most `n` times.

  - `a{2,4}`: Matches "aa", "aaa", "aaaa".

- **`{m,}` (Minimum Count):** Matches the preceding character at least `m` times.

  - `a{2,}`: Matches "aa", "aaa", "aaaa", etc.

- **`|` (Alternation/OR):** Matches either the expression before or after the `|`.

  - `cat|dog`: Matches "cat" or "dog".

- **`()` (Grouping):** Groups expressions together and captures the matched text.

  - `(ab)+`: Matches "ab", "abab", "ababab", etc.
  - Used for applying quantifiers to a sequence and for backreferences.

- **`[]` (Character Set):** Matches any one character within the set.

  - `[abc]`: Matches 'a', 'b', or 'c'.
  - `[0-9]`: Matches any digit from 0 to 9.
  - `[a-zA-Z]`: Matches any lowercase or uppercase letter.
  - `[aeiou]`: Matches any vowel.
  - `[^abc]`: Matches any character _not_ 'a', 'b', or 'c' (caret inside `[]` negates the set).

- **`\` (Backslash):** Escapes special characters or introduces special sequences.

  - `\.`: Matches a literal dot.
  - `\*`: Matches a literal asterisk.

**3. Special Sequences (Shorthands):**
These are predefined character sets.

- **`\d`:** Matches any digit (0-9). Equivalent to `[0-9]`.
- **`\D`:** Matches any non-digit character. Equivalent to `[^0-9]`.
- **`\w`:** Matches any word character (alphanumeric + underscore). Equivalent to `[a-zA-Z0-9_]`.
- **`\W`:** Matches any non-word character. Equivalent to `[^a-zA-Z0-9_]`.
- **`\s`:** Matches any whitespace character (space, tab, newline, carriage return, form feed, vertical tab).
- **`\S`:** Matches any non-whitespace character.
- **`\b`:** Matches a word boundary.
  - `\bcat\b`: Matches "cat" when it's a whole word, not part of "catalog".
- **`\B`:** Matches a non-word boundary.
- **`\A`:** Matches the start of the string (regardless of `re.MULTILINE`).
- **`\Z`:** Matches the end of the string (regardless of `re.MULTILINE`).

**4. Greedy vs. Non-Greedy (Lazy) Quantifiers:**
By default, quantifiers (`*`, `+`, `?`, `{m,n}`) are "greedy," meaning they try to match as much as possible. Appending a `?` makes them "non-greedy" or "lazy," matching as little as possible.

- `.*`: Matches everything up to the last occurrence of the next pattern.
- `.*?`: Matches everything up to the first occurrence of the next pattern.

Example: Text = `<b>Hello</b> World <b>Python</b>`

- `<b>.*</b>`: Matches `<b>Hello</b> World <b>Python</b>` (the entire string).
- `<b>.*?</b>`: Matches `<b>Hello</b>` (the first occurrence).

### Python `re` Module Functions

The `re` module provides several functions for working with regular expressions.

**1. `re.compile(pattern, flags=0)`:**
Compiles a regex pattern into a regular expression object. This is useful for performance when you're going to use the same pattern multiple times.

**2. `re.match(pattern, string, flags=0)`:**
Attempts to match the pattern **only at the beginning** of the string. Returns a match object if successful, `None` otherwise.

**3. `re.search(pattern, string, flags=0)`:**
Scans through the string looking for the first location where the pattern produces a match. Returns a match object if successful, `None` otherwise.

**4. `re.findall(pattern, string, flags=0)`:**
Returns a list of all non-overlapping matches in the string. If groups are present, it returns a list of tuples.

**5. `re.finditer(pattern, string, flags=0)`:**
Returns an iterator yielding match objects for all non-overlapping matches. Useful when you need more details about each match (like start/end positions).

**6. `re.sub(pattern, repl, string, count=0, flags=0)`:**
Replaces occurrences of `pattern` in `string` with `repl`. `repl` can be a string or a function. `count` limits the number of replacements.

**7. `re.split(pattern, string, maxsplit=0, flags=0)`:**
Splits the `string` by occurrences of `pattern`. `maxsplit` limits the number of splits.

**Match Object Methods:**
If `re.search()` or `re.match()` finds a match, they return a match object.

- `match.group(0)` or `match.group()`: Returns the entire matched string.
- `match.group(1)`, `match.group(2)`, ...: Returns the string matched by the corresponding capturing group.
- `match.groups()`: Returns a tuple of all captured groups.
- `match.start()`: Returns the starting index of the match.
- `match.end()`: Returns the ending index of the match (exclusive).
- `match.span()`: Returns a tuple `(start, end)` of the match.

**Flags:**
Flags modify the behavior of regex matching. They can be combined using the bitwise OR operator `|`.

- `re.IGNORECASE` or `re.I`: Performs case-insensitive matching.
- `re.MULTILINE` or `re.M`: Makes `^` and `$` match the start/end of each line, not just the start/end of the string.
- `re.DOTALL` or `re.S`: Makes `.` match any character, including newlines.
- `re.VERBOSE` or `re.X`: Allows you to write regex patterns with whitespace and comments for better readability.
- `re.ASCII` or `re.A`: Makes `\w`, `\W`, `\b`, `\B`, `\d`, `\D`, `\s`, `\S` perform ASCII-only matching.
- `re.UNICODE` or `re.U` (default in Python 3): Makes `\w`, `\W`, `\b`, `\B`, `\d`, `\D`, `\s`, `\S` match Unicode characters.

```python
import re

print("--- 1. Basic Matching with re.search() ---")
text = "The quick brown fox."
pattern = "quick"
match = re.search(pattern, text)
if match:
    print(f"'{pattern}' found in '{text}' at index {match.start()} to {match.end()-1}.")
    print(f"Matched string: '{match.group()}'")
else:
    print(f"'{pattern}' not found in '{text}'.")

print("\n--- 2. re.match() vs re.search() ---")
text1 = "Hello World"
text2 = "Say Hello World"

# re.match() only checks at the beginning
match_match = re.match("Hello", text1)
print(f"re.match('Hello', '{text1}'): {match_match.group() if match_match else None}")

match_match_fail = re.match("Hello", text2)
print(f"re.match('Hello', '{text2}'): {match_match_fail.group() if match_match_fail else None}")

# re.search() checks anywhere in the string
match_search = re.search("Hello", text2)
print(f"re.search('Hello', '{text2}'): {match_search.group() if match_search else None}")

print("\n--- 3. Using Metacharacters ---")
# . (any character)
print(f"re.search('c.t', 'cat'): {re.search('c.t', 'cat').group()}")
print(f"re.search('c.t', 'cut'): {re.search('c.t', 'cut').group()}")

# ^ (start of string)
print(f"re.search('^The', 'The start'): {re.search('^The', 'The start').group()}")
print(f"re.search('^The', 'start The'): {re.search('^The', 'start The')}") # None

# $ (end of string)
print(f"re.search('end$', 'The end'): {re.search('end$', 'The end').group()}")
print(f"re.search('end$', 'ending'): {re.search('end$', 'ending')}") # None

# * (zero or more)
print(f"re.search('go*gle', 'ggle'): {re.search('go*gle', 'ggle').group()}")
print(f"re.search('go*gle', 'google'): {re.search('go*gle', 'google').group()}")

# + (one or more)
print(f"re.search('go+gle', 'google'): {re.search('go+gle', 'google').group()}")
print(f"re.search('go+gle', 'ggle'): {re.search('go+gle', 'ggle')}") # None

# ? (zero or one)
print(f"re.search('colou?r', 'color'): {re.search('colou?r', 'color').group()}")
print(f"re.search('colou?r', 'colour'): {re.search('colou?r', 'colour').group()}")

# {} (quantifiers)
print(f"re.search('a{{3}}', 'aaabaaa'): {re.search('a{{3}}', 'aaabaaa').group()}") # Matches 'aaa'
print(f"re.search('x{{2,4}}', 'xxxxxx'): {re.search('x{{2,4}}', 'xxxxxx').group()}") # Matches 'xxxx' (greedy)

# | (OR)
print(f"re.search('apple|orange', 'I like orange juice'): {re.search('apple|orange', 'I like orange juice').group()}")

# () (Grouping and Capturing)
match_groups = re.search(r'(\d+)-(\d+)-(\d+)', 'Phone: 123-456-7890')
if match_groups:
    print(f"Full match: {match_groups.group(0)}")
    print(f"Area code: {match_groups.group(1)}")
    print(f"First 3: {match_groups.group(2)}")
    print(f"Last 4: {match_groups.group(3)}")
    print(f"All groups: {match_groups.groups()}")

# [] (Character Set)
print(f"re.search('[aeiou]', 'rhythm'): {re.search('[aeiou]', 'rhythm')}") # None (no vowels)
print(f"re.search('[aeiou]', 'hello'): {re.search('[aeiou]', 'hello').group()}") # Matches 'e'
print(f"re.search('[^0-9]', '123abc456'): {re.search('[^0-9]', '123abc456').group()}") # Matches 'a'

# \ (Escaping)
print(f"re.search('literal\\.dot', 'This is a literal.dot'): {re.search('literal\\.dot', 'This is a literal.dot').group()}")

print("\n--- 4. Using Special Sequences ---")
# \d (digit)
print(f"re.findall('\\d', '1 fish 2 fish red fish blue fish'): {re.findall(r'\d', '1 fish 2 fish red fish blue fish')}") # ['1', '2']
# Using r'' for raw string to avoid backslash issues

# \w (word character)
print(f"re.findall('\\w+', 'Hello, world! (123)'): {re.findall(r'\w+', 'Hello, world! (123)')}") # ['Hello', 'world', '123']

# \s (whitespace)
print(f"re.findall('\\s', 'Hello  world\n'): {re.findall(r'\s', 'Hello  world\n')}") # [' ', ' ', '\n']

# \b (word boundary)
print(f"re.search('\\bcat\\b', 'The cat sat.'): {re.search(r'\bcat\b', 'The cat sat.').group()}")
print(f"re.search('\\bcat\\b', 'The catalog.'): {re.search(r'\bcat\b', 'The catalog.')}") # None

print("\n--- 5. re.findall() ---")
text = "Emails: user@example.com, another.user@domain.net, info@sub.domain.org"
emails = re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text)
print(f"Found emails: {emails}")

print("\n--- 6. re.finditer() ---")
for match_obj in re.finditer(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text):
    print(f"Email found: {match_obj.group()} at {match_obj.span()}")

print("\n--- 7. re.sub() (Substitution) ---")
# Replace digits with '#'
text_sub = "My phone number is 123-456-7890."
new_text = re.sub(r'\d', '#', text_sub)
print(f"Original: '{text_sub}'")
print(f"Substituted: '{new_text}'")

# Replace first 3 occurrences
text_sub_count = "1 2 3 4 5 6"
new_text_count = re.sub(r'\d', 'X', text_sub_count, count=3)
print(f"Original: '{text_sub_count}'")
print(f"Substituted (count=3): '{new_text_count}'")

# Using a function for replacement (e.g., double each digit)
def double_digit(match):
    return str(int(match.group(0)) * 2)

text_func_sub = "Numbers: 1, 5, 10, 25"
new_text_func_sub = re.sub(r'\d+', double_digit, text_func_sub)
print(f"Original: '{text_func_sub}'")
print(f"Substituted (double digits): '{new_text_func_sub}'")


print("\n--- 8. re.split() ---")
sentence = "Hello   world, this is a test. How are you?"
words = re.split(r'\s+|,|\.', sentence) # Split by one or more spaces, or comma, or dot
# Filter out empty strings from splitting
words = [word for word in words if word]
print(f"Original: '{sentence}'")
print(f"Split words: {words}")

print("\n--- 9. Using Flags ---")
# re.IGNORECASE (re.I)
print(f"re.search('apple', 'APPLE', re.I): {re.search('apple', 'APPLE', re.I).group()}")

# re.MULTILINE (re.M)
multi_line_text = "Line 1\nLine 2\nLine 3"
# Without re.M, '^Line' only matches 'Line 1'
print(f"re.findall('^Line', multi_line_text): {re.findall('^Line', multi_line_text)}")
# With re.M, '^Line' matches 'Line' at the start of each line
print(f"re.findall('^Line', multi_line_text, re.M): {re.findall('^Line', multi_line_text, re.M)}")

# re.DOTALL (re.S)
text_with_newline = "First line.\nSecond line."
# Without re.S, '.' does not match newline
print(f"re.search('line.Second', text_with_newline): {re.search('line.Second', text_with_newline)}") # None
# With re.S, '.' matches newline
print(f"re.search('line.Second', text_with_newline, re.S): {re.search('line.Second', text_with_newline, re.S).group()}")

# re.VERBOSE (re.X) for readability
regex_verbose = re.compile(r"""
    ^                   # Start of the string
    (\d{3})             # Capture 3 digits (e.g., area code)
    -                   # Literal hyphen
    (\d{3})             # Capture another 3 digits
    -                   # Literal hyphen
    (\d{4})             # Capture 4 digits
    $                   # End of the string
""", re.X)
match_verbose = regex_verbose.search("123-456-7890")
if match_verbose:
    print(f"\nVerbose Regex Match: {match_verbose.groups()}")
else:
    print("\nVerbose Regex No Match.")

print("\n--- 10. Compiling Regex for Performance ---")
# Compile the pattern if you use it multiple times
compiled_pattern = re.compile(r'\b\w+\b') # Match whole words

long_text = "This is a very long text to test performance of regular expressions. " * 1000
# Using compiled pattern
import time
start_time = time.time()
re.findall(compiled_pattern, long_text)
end_time = time.time()
print(f"Time with compiled pattern: {end_time - start_time:.6f} seconds")

# Without compiling (pattern is compiled implicitly each time)
start_time = time.time()
re.findall(r'\b\w+\b', long_text)
end_time = time.time()
print(f"Time without compiled pattern: {end_time - start_time:.6f} seconds")

```


Regular expressions are a language within a language, allowing you to define search patterns for strings. **Metacharacters** are the special characters in this language that don't represent themselves literally but instead have a special meaning, allowing for powerful and flexible pattern matching.

Understanding metacharacters is key to mastering regular expressions.

---

### Theory of Metacharacters

Metacharacters can be broadly categorized by their function:

1.  **Anchors:** These do not match any character but assert a position within the string.
2.  **Quantifiers:** These specify how many times the preceding character or group must occur.
3.  **Character Classes/Sets:** These match one of a set of characters.
4.  **Alternation:** This allows for "OR" logic in your pattern.
5.  **Grouping and Capturing:** These combine parts of a pattern and can capture matched text.
6.  **Escaping:** This allows you to treat a metacharacter as a literal character.

When a regex engine processes a string with a pattern, it doesn't just look for literal matches. It interprets these metacharacters to determine valid positions, repetitions, and character types, making the pattern much more expressive than a simple substring search.

---

### Metacharacters: Syntax and Detailed Explanation

Here's a comprehensive breakdown of the most common metacharacters:

#### 1. Anchors

Anchors assert a position. They don't consume any characters.

- **`^` (Caret)**

  - **Syntax:** `^pattern`
  - **Theory:** Matches the position at the beginning of the string.
  - **Behavior with `re.MULTILINE` flag:** If the `re.MULTILINE` (or `re.M` in Python's `re` module) flag is used, `^` also matches the position immediately after each newline character.
  - **Example:**
    - `^Hello`: Matches "Hello World" but not "Say Hello World".
    - With `re.M`, `^Line` in `"Line 1\nLine 2"` would match "Line" at the start of both "Line 1" and "Line 2".

- **`$` (Dollar)**

  - **Syntax:** `pattern$`
  - **Theory:** Matches the position at the end of the string.
  - **Behavior with `re.MULTILINE` flag:** If `re.MULTILINE` is used, `$` also matches the position immediately before each newline character. It also matches before a final newline, if present (e.g., in "abc\n", it matches before the '\n').
  - **Example:**
    - `World$`: Matches "Hello World" but not "Hello World!".
    - With `re.M`, `.$` in `"hello\nworld"` would match `o` and `d` (the last char of each line).

- **`\A` (Start of String)**

  - **Syntax:** `\Apattern`
  - **Theory:** Matches only the absolute beginning of the entire string, regardless of the `re.MULTILINE` flag.
  - **Example:**
    - `\AHello` in `"Hello World\nHello Python"` will only match the first "Hello".

- **`\Z` (End of String)**

  - **Syntax:** `pattern\Z`
  - **Theory:** Matches only the absolute end of the entire string, regardless of the `re.MULTILINE` flag. It matches just before the final newline if the string ends with one.
  - **Example:**
    - `World\Z` in `"Hello World\nExtra"` will not match.
    - `World\Z` in `"Hello World"` will match.
    - `World\Z` in `"Hello World\n"` will match "World" (before the final newline).

- **`\b` (Word Boundary)**

  - **Syntax:** `\bword\b`
  - **Theory:** Matches a position where one side is a "word" character (`\w`) and the other side is a "non-word" character (`\W`) or the beginning/end of the string. It asserts that the pattern is a whole word.
  - **Example:**
    - `\bcat\b`:
      - Matches "The **cat** sat."
      - Does NOT match "The **cat**alog." (because 'c' is preceded by ' ', word boundary is met. 't' is followed by 'a', not a word boundary. Hence the whole match fails)
      - Does NOT match "He scatters."
    - `\bapple`: Matches "apple juice" (matches before 'a') and "An apple" (matches after ' ').
    - `apple\b`: Matches "apple juice" (matches before ' ') and "an apple." (matches before '.').

- **`\B` (Non-Word Boundary)**
  - **Syntax:** `\Bpattern\B`
  - **Theory:** Matches a position where both sides are "word" characters or both sides are "non-word" characters (i.e., not a word boundary). It's the opposite of `\b`.
  - **Example:**
    - `\Bcat\B`:
      - Matches "The **cat**alog." (because 'c' is preceded by 'a', 't' is followed by 'a'. Not word boundaries.)
      - Does NOT match "The cat sat."

#### 2. Quantifiers

Quantifiers specify the number of occurrences of the preceding character or group. By default, they are **greedy**, meaning they try to match as many characters as possible while still allowing the rest of the regex to match.

- **`*` (Asterisk)**

  - **Syntax:** `X*`
  - **Theory:** Matches the preceding element `X` zero or more times.
  - **Example:**
    - `ab*c`: Matches "ac", "abc", "abbc", "abbbc".

- **`+` (Plus)**

  - **Syntax:** `X+`
  - **Theory:** Matches the preceding element `X` one or more times.
  - **Example:**
    - `ab+c`: Matches "abc", "abbc", "abbbc" (but not "ac").

- **`?` (Question Mark)**

  - **Syntax:** `X?`
  - **Theory:** Matches the preceding element `X` zero or one time.
  - **Example:**
    - `colou?r`: Matches "color", "colour".

- **`{n}` (Exact Count)**

  - **Syntax:** `X{n}`
  - **Theory:** Matches the preceding element `X` exactly `n` times.
  - **Example:**
    - `\d{3}`: Matches exactly three digits, like "123".

- **`{n,}` (Minimum Count)**

  - **Syntax:** `X{n,}`
  - **Theory:** Matches the preceding element `X` at least `n` times.
  - **Example:**
    - `a{2,}`: Matches "aa", "aaa", "aaaa", etc.

- **`{n,m}` (Range Count)**

  - **Syntax:** `X{n,m}`
  - **Theory:** Matches the preceding element `X` at least `n` times and at most `m` times.
  - **Example:**
    - `[0-9]{5,7}`: Matches a sequence of 5, 6, or 7 digits.

- **Lazy (Non-Greedy) Quantifiers (adding `?`)**
  - **Syntax:** `X*?`, `X+?`, `X??`, `X{n,}?`, `X{n,m}?`
  - **Theory:** These quantifiers try to match as few characters as possible.
  - **Example:**
    - Text: `<b>Hello</b> World <b>Python</b>`
    - `<b>.*</b>` (greedy): Matches `<b>Hello</b> World <b>Python</b>`
    - `<b>.*?</b>` (lazy): Matches `<b>Hello</b>`

#### 3. Character Classes / Sets

These define a set of characters to match at a single position.

- **`.` (Dot)**

  - **Syntax:** `.`
  - **Theory:** Matches any single character except newline (`\n`) by default.
  - **Behavior with `re.DOTALL` flag (`re.S`):** If this flag is used, `.` will also match the newline character.
  - **Example:**
    - `a.b`: Matches "acb", "a#b", "axb".
    - Without `re.S`, `a.b` does not match "a\nb". With `re.S`, it does.

- **`[]` (Character Set)**

  - **Syntax:** `[characters]`
  - **Theory:** Matches any one character listed inside the square brackets.
  - **Example:**
    - `[abc]`: Matches 'a', 'b', or 'c'.
    - `[0-9]`: Matches any digit (same as `\d`).
    - `[a-zA-Z]`: Matches any lowercase or uppercase letter.
    - `[aeiouAEIOU]`: Matches any vowel (case-sensitive by default).
    - `[a-z_]`: Matches any lowercase letter or an underscore.
    - **Negation:** `[^characters]` matches any character _not_ in the set.
      - `[^0-9]`: Matches any non-digit character (same as `\D`).

- **Special Sequences (Shorthands for Common Character Sets):**
  - `\d`: Matches any digit (0-9). Equivalent to `[0-9]`.
  - `\D`: Matches any non-digit character. Equivalent to `[^0-9]`.
  - `\w`: Matches any "word" character (alphanumeric and underscore). Equivalent to `[a-zA-Z0-9_]`.
  - `\W`: Matches any non-word character. Equivalent to `[^a-zA-Z0-9_]`.
  - `\s`: Matches any whitespace character (space, tab `\t`, newline `\n`, carriage return `\r`, form feed `\f`, vertical tab `\v`).
  - `\S`: Matches any non-whitespace character.

#### 4. Alternation

- **`|` (Pipe / OR operator)**
  - **Syntax:** `expression1|expression2`
  - **Theory:** Matches either `expression1` or `expression2`.
  - **Example:**
    - `cat|dog`: Matches "cat" or "dog".
    - `gr(a|e)y`: Matches "gray" or "grey".

#### 5. Grouping and Capturing

- **`()` (Parentheses)**

  - **Syntax:** `(pattern)`
  - **Theory:**
    1.  **Grouping:** Treats a sequence of characters as a single unit, allowing quantifiers to apply to the whole group.
    2.  **Capturing:** Stores the matched substring of the group for later retrieval (e.g., using `match.group(1)` or backreferences).
  - **Example:**
    - `(ab)+`: Matches "ab", "abab", "ababab".
    - `(\d{3})-(\d{4})`: In "123-4567", `group(1)` would be "123", `group(2)` would be "4567".

- **`(?:pattern)` (Non-Capturing Group)**

  - **Syntax:** `(?:pattern)`
  - **Theory:** Groups characters like `()`, but does _not_ capture the matched text. Useful when you need to group for applying quantifiers or alternation but don't need to extract the content.
  - **Example:**
    - `(?:ab)+`: Matches "ab", "abab", but you can't access `group(1)` for 'ab'.
    - `re.search('(?:\d{3}-){2}\d{4}', '123-456-7890').group()` # Matches the whole phone number

- **`(?P<name>pattern)` (Named Capturing Group)**

  - **Syntax:** `(?P<name>pattern)`
  - **Theory:** A capturing group that can be referenced by a name instead of a number, making code more readable.
  - **Example:**
    - `(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})`
    - You can then access `match.group('year')`, `match.group('month')`, `match.group('day')`.

- **Backreferences (`\n` or `\g<name>`)**
  - **Syntax:** `\1`, `\2`, ..., `\g<name>`
  - **Theory:** Refers to the text previously matched by a capturing group. `\1` refers to the first group, `\2` to the second, etc. `\g<name>` refers to a named group.
  - **Example:**
    - `(\w+)\s+\1`: Matches repeated words like "hello hello" or "world world".
    - `(?P<tag>\w+).*?</(?P=tag)>`: Matches HTML-like tags like `<b>...</b>` or `<p>...</p>`.

#### 6. Escaping

- **`\` (Backslash)**
  - **Syntax:** `\` followed by a metacharacter (e.g., `\.`, `\*`, `\?`)
  - **Theory:** Removes the special meaning of the metacharacter, treating it as a literal character.
  - **Example:**
    - `\.`: Matches a literal dot.
    - `\$`: Matches a literal dollar sign.
    - If you want to match "[www.example.com](https://www.example.com)", you'd use `www\.example\.com`.

### Best Practices

- **Raw Strings (`r''`):** Always use raw strings for regex patterns in Python (e.g., `r'\d+'` instead of `'\d+'`). This prevents Python's string escaping rules from interfering with regex backslashes. For example, `'\n'` means newline in Python strings, but `r'\n'` is treated as a literal backslash followed by 'n' by regex, which is what you want for the regex `\n` metacharacter.
- **Compile Patterns:** For patterns used repeatedly, compile them using `re.compile()` for better performance.
- **Flags:** Use flags like `re.I` for case-insensitivity, `re.M` for multi-line matching, and `re.S` for dot-all matching.
- **Readability with `re.VERBOSE`:** For complex patterns, use `re.VERBOSE` (`re.X`) to allow whitespace and comments within the pattern for better readability.

Understanding and effectively using these metacharacters unlocks the full power of regular expressions for complex text processing tasks.


The `re.match()` function in Python's `re` module is a fundamental tool for regular expression matching, but it has a very specific behavior that distinguishes it from other functions like `re.search()`.

---

### `re.match()`: Theory

**Core Concept:**
`re.match(pattern, string, flags=0)` attempts to match the regular expression `pattern` **only at the beginning of the `string`**. It will succeed only if the pattern matches at index 0 of the string. If the pattern is found anywhere else in the string (not at the beginning), `re.match()` will return `None`.

**Key Characteristics:**

1.  **Anchored to the Start:** This is the most crucial aspect. `re.match()` implicitly assumes that your pattern should start at the very beginning of the target string. It's as if your pattern implicitly has a `^` (start-of-string anchor) at its beginning.

2.  **Returns a Match Object or `None`:**

    - If a match is found at the beginning of the string, it returns a `Match object`. This object contains information about the match (e.g., the matched string, its start and end indices, captured groups).
    - If no match is found at the beginning, it returns `None`.

3.  **No Full String Scan:** Unlike `re.search()`, `re.match()` does not scan the entire string. Its job is done as soon as it determines whether the pattern starts at index 0.

**When to Use `re.match()`:**

- When you need to validate if a string _starts_ with a specific pattern.
- When you're parsing data where you expect a particular format to appear at the very beginning of a line or text block.
- For simple prefix checks.

**Analogy:**
Imagine you have a book, and you're looking for the word "Chapter".

- `re.match()` is like checking _only the very first word_ on the very first page to see if it's "Chapter". If it's "Chapter 1", great\! If it's "Foreword", you immediately give up, even if "Chapter" appears later on that page or on page 100.
- `re.search()` is like scanning the entire book from the beginning, finding the _first_ instance of "Chapter" wherever it may be.

---

### `re.match()`: Syntax and Code Examples

The basic syntax for `re.match()` is:

```python
re.match(pattern, string, flags=0)
```

- `pattern`: The regular expression string to match. It's highly recommended to use raw strings (e.g., `r'pattern'`) to avoid issues with backslashes.
- `string`: The string against which to attempt the match.
- `flags` (optional): Modifiers that change the behavior of the regex matching (e.g., `re.IGNORECASE`, `re.MULTILINE`, `re.DOTALL`).

**Code Examples:**

```python
import re

print("--- Basic Usage of re.match() ---")

# Example 1: Successful match at the beginning
text1 = "Hello World"
pattern1 = r"Hello"
match1 = re.match(pattern1, text1)

if match1:
    print(f"Match 1 successful!")
    print(f"  Pattern: '{pattern1}'")
    print(f"  String:  '{text1}'")
    print(f"  Matched: '{match1.group(0)}'") # group(0) returns the entire matched string
    print(f"  Start Index: {match1.start()}")
    print(f"  End Index:   {match1.end()}")
    print(f"  Span:        {match1.span()}")
else:
    print(f"Match 1 failed (expected success).")

print("\n" + "="*30 + "\n")

# Example 2: Match not at the beginning (re.match fails)
text2 = "Say Hello World"
pattern2 = r"Hello"
match2 = re.match(pattern2, text2)

if match2:
    print(f"Match 2 successful (unexpected).")
else:
    print(f"Match 2 failed as expected!")
    print(f"  Pattern: '{pattern2}'")
    print(f"  String:  '{text2}'")
    print(f"  Reason: '{Hello}' does not appear at the beginning of the string.")

print("\n" + "="*30 + "\n")

# Example 3: Contrast with re.search()
search_match = re.search(pattern2, text2)
if search_match:
    print(f"Using re.search() on '{text2}' with pattern '{pattern2}':")
    print(f"  Found '{search_match.group(0)}' at index {search_match.start()}")
else:
    print(f"re.search() failed (unexpected).")

print("\n" + "="*30 + "\n")

print("--- Using Metacharacters with re.match() ---")

# Example 4: Using '^' explicitly (redundant with re.match, but shows behavior)
text3 = "Start of Line"
pattern3 = r"^Start" # The '^' is effectively implied by re.match
match3 = re.match(pattern3, text3)
print(f"Pattern '{pattern3}' on '{text3}': {match3.group() if match3 else None}")

pattern3_b = r"Start" # Same effect as above
match3_b = re.match(pattern3_b, text3)
print(f"Pattern '{pattern3_b}' on '{text3}' (without '^'): {match3_b.group() if match3_b else None}")

text4 = "Something Start of Line"
match4 = re.match(pattern3, text4)
print(f"Pattern '{pattern3}' on '{text4}': {match4.group() if match4 else None}") # Fails because "Something" is first

print("\n" + "="*30 + "\n")

# Example 5: Matching numbers at the beginning
text5 = "123 Main St."
pattern5 = r"\d+" # One or more digits
match5 = re.match(pattern5, text5)
print(f"Pattern '{pattern5}' on '{text5}': {match5.group() if match5 else None}")

text6 = "Address: 456 Oak Ave."
match6 = re.match(pattern5, text6)
print(f"Pattern '{pattern5}' on '{text6}': {match6.group() if match6 else None}") # Fails, "Address" is first

print("\n" + "="*30 + "\n")

print("--- Using Groups and Flags with re.match() ---")

# Example 6: Capturing groups
text7 = "Date: 2023-10-26"
pattern7 = r"Date: (\d{4})-(\d{2})-(\d{2})"
match7 = re.match(pattern7, text7)

if match7:
    print(f"Matched: {match7.group(0)}")
    print(f"Year: {match7.group(1)}")
    print(f"Month: {match7.group(2)}")
    print(f"Day: {match7.group(3)}")
    print(f"All groups: {match7.groups()}")
else:
    print(f"No match for pattern '{pattern7}' on '{text7}'.")

print("\n" + "="*30 + "\n")

# Example 7: Using re.IGNORECASE flag
text8 = "python programming"
pattern8 = r"Python"
match8 = re.match(pattern8, text8, re.IGNORECASE)

if match8:
    print(f"Case-insensitive match for '{pattern8}' on '{text8}': {match8.group()}")
else:
    print(f"Case-insensitive match failed.")

print("\n" + "="*30 + "\n")

# Example 8: Using re.MULTILINE flag (doesn't affect re.match on '^')
# Note: re.match always works from string start. The ^ metacharacter will only
# match start of string. re.M only affects ^ when used with re.search or re.findall
# to match start of lines, but re.match isn't designed to search for lines.
text9 = "Line1\nLine2\nLine3"
pattern9 = r"^Line2" # This pattern will only ever match if "Line2" is at the start of the entire string
match9 = re.match(pattern9, text9, re.MULTILINE)
print(f"Pattern '{pattern9}' on multi-line text (re.M): {match9.group() if match9 else None}") # Still None, as Line2 is not at string start

# If you want to match "Line2" at the start of a *line*, you'd use re.search and re.M
search_multi_line = re.search(r"^Line2", text9, re.MULTILINE)
print(f"Using re.search(r'^Line2', text, re.M): {search_multi_line.group() if search_multi_line else None}")

```
