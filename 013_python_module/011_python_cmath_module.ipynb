### **Python `cmath` Module: Overview, Concepts, and Theory**

The `cmath` module in Python is part of the standard library and is specifically designed for working with complex numbers. It provides a range of mathematical functions for handling complex numbers, including functions for trigonometry, logarithms, and other complex number operations.

Unlike the `math` module, which only supports real numbers, the `cmath` module supports complex numbers and extends the functionality of mathematical operations to handle complex results.

---

### **Key Concepts of the `cmath` Module:**

1. **Complex Numbers:**

   - A **complex number** is a number that has both a real part and an imaginary part. It is represented as `a + bj`, where:
     - `a` is the real part of the complex number.
     - `b` is the imaginary part of the complex number.
     - `j` or `i` represents the imaginary unit (i.e., `i^2 = -1`).

2. **Complex Arithmetic:**

   - Complex numbers support standard arithmetic operations like addition, subtraction, multiplication, and division. The `cmath` module provides functions for performing these operations and for converting between polar and rectangular forms of complex numbers.

3. **Polar Coordinates:**

   - Complex numbers can be represented in **polar form**, where a complex number `z = a + bj` is written as `r(cos θ + j sin θ)` or `r * e^(jθ)`, where:
     - `r` is the magnitude (absolute value) of the complex number.
     - `θ` is the phase (angle), also known as the argument of the complex number.

4. **Functions in `cmath`:**
   - The `cmath` module includes various functions to calculate magnitudes, phases, trigonometric functions, logarithms, square roots, and other operations specifically for complex numbers.

---

### **Important Functions in the `cmath` Module:**

#### 1. **Complex Number Representation:**

- **`cmath.phase(z)`**:

  - Returns the phase (or argument) of the complex number `z`. The phase is the angle `θ` that the complex number makes with the positive real axis in the complex plane.

  - **Example:**
    ```python
    import cmath
    z = 3 + 4j
    print(cmath.phase(z))  # Output: 0.9272952180016122 (which is atan2(4, 3))
    ```

- **`cmath.polar(z)`**:

  - Converts a complex number `z` from rectangular form (a + bj) to polar form (r, θ), where `r` is the magnitude and `θ` is the phase.

  - **Example:**
    ```python
    z = 3 + 4j
    print(cmath.polar(z))  # Output: (5.0, 0.9272952180016122)
    ```

- **`cmath.rect(r, theta)`**:

  - Converts a complex number from polar form `(r, θ)` to rectangular form `r * (cos θ + j sin θ)`.

  - **Example:**
    ```python
    r, theta = 5, 0.9272952180016122
    print(cmath.rect(r, theta))  # Output: (3+4j)
    ```

#### 2. **Magnitude of a Complex Number:**

- **`cmath.abs(z)`**:

  - Returns the absolute value (magnitude) of the complex number `z`, which is the distance of the complex number from the origin in the complex plane.
  - This is equivalent to `cmath.polar(z)[0]` (the magnitude `r`).

  - **Example:**
    ```python
    z = 3 + 4j
    print(cmath.abs(z))  # Output: 5.0
    ```

#### 3. **Trigonometric Functions:**

- **`cmath.sin(z)`**: Returns the sine of the complex number `z`.

- **`cmath.cos(z)`**: Returns the cosine of the complex number `z`.

- **`cmath.tan(z)`**: Returns the tangent of the complex number `z`.

- **`cmath.asin(z)`**: Returns the inverse sine (arc sine) of `z`.

- **`cmath.acos(z)`**: Returns the inverse cosine (arc cosine) of `z`.

- **`cmath.atan(z)`**: Returns the inverse tangent (arc tangent) of `z`.

  - **Example:**
    ```python
    z = 1 + 2j
    print(cmath.sin(z))  # Output: (3.165778513216168+1.9596010412194936j)
    ```

#### 4. **Exponential and Logarithmic Functions:**

- **`cmath.exp(z)`**: Returns `e` raised to the power of `z`, i.e., `e^z`.

- **`cmath.log(z, base)`**: Returns the logarithm of `z` to the given `base`. If no base is provided, the natural logarithm (base `e`) is returned.

- **`cmath.log10(z)`**: Returns the base-10 logarithm of `z`.

- **`cmath.sqrt(z)`**: Returns the square root of `z`. The result is also a complex number.

  - **Example:**
    ```python
    z = 1 + 2j
    print(cmath.exp(z))  # Output: (-1.1312047117837034+2.471726672004818j)
    ```

#### 5. **Inverse Hyperbolic and Hyperbolic Functions:**

- **`cmath.sinh(z)`**: Returns the hyperbolic sine of `z`.

- **`cmath.cosh(z)`**: Returns the hyperbolic cosine of `z`.

- **`cmath.tanh(z)`**: Returns the hyperbolic tangent of `z`.

- **`cmath.asinh(z)`**: Returns the inverse hyperbolic sine of `z`.

- **`cmath.acosh(z)`**: Returns the inverse hyperbolic cosine of `z`.

- **`cmath.atanh(z)`**: Returns the inverse hyperbolic tangent of `z`.

  - **Example:**
    ```python
    z = 1 + 2j
    print(cmath.sinh(z))  # Output: (3.165778513216168+1.9596010412194936j)
    ```

#### 6. **Constants:**

- **`cmath.pi`**: The mathematical constant `π` (approximately 3.14159).

- **`cmath.e`**: The base of the natural logarithm `e` (approximately 2.71828).

  - **Example:**
    ```python
    print(cmath.pi)  # Output: 3.141592653589793
    print(cmath.e)   # Output: 2.718281828459045
    ```

````python
# --- Python's `cmath` Module: All About Phase and Magnitude in Code ---

# In the realm of complex numbers, a number `z = x + yj` can be represented
# not only in its rectangular (Cartesian) form (`x`, `y`) but also in its
# polar form (magnitude `r` and phase `phi`).
# The `cmath` module provides specific functions for working with these
# polar coordinates.

import cmath
import math # For math.pi and comparison

print("--- 1. Understanding Magnitude (Modulus/Absolute Value) ---")

# 1.1 Definition: The magnitude (or modulus, or absolute value) of a complex number
#    `z = x + yj` is the diwstance from the origin (0,0) to the point (x,y) in the
#    complex plane. It is calculated as `r = sqrt(x^2 + y^2)`.
# 1.2 Python Function: The built-in `abs()` function is used to calculate the magnitude
#    of a complex number. There is no separate `cmath.abs()` function.

z1 = 3 + 4j # Real part = 3, Imaginary part = 4
z2 = -2 + 0j # Purely real (negative)
z3 = 0 - 5j # Purely imaginary (negative)
z4 = 1 + 1j # For common angles

print(f"1.1 For z1 = {z1}:")
magnitude_z1 = abs(z1)
print(f"    Magnitude (abs(z1)): {magnitude_z1:.4f}") # Output: 5.0 (sqrt(3^2 + 4^2) = sqrt(9+16) = sqrt(25))

print(f"\n1.2 For z2 = {z2}:")
magnitude_z2 = abs(z2)
print(f"    Magnitude (abs(z2)): {magnitude_z2:.4f}") # Output: 2.0 (sqrt((-2)^2 + 0^2) = 2)

print(f"\n1.3 For z3 = {z3}:")
magnitude_z3 = abs(z3)
print(f"    Magnitude (abs(z3)): {magnitude_z3:.4f}") # Output: 5.0 (sqrt(0^2 + (-5)^2) = 5)

print(f"\n1.4 For z4 = {z4}:")
magnitude_z4 = abs(z4)
print(f"    Magnitude (abs(z4)): {magnitude_z4:.4f}") # Output: 1.4142 (sqrt(1^2 + 1^2) = sqrt(2))


print("\n--- 2. Understanding Phase (Argument) ---")

# 2.1 Definition: The phase (or argument) of a complex number `z = x + yj` is the
#    angle (in radians) from the positive real axis to the line segment connecting
#    the origin to the point (x,y) in the complex plane.
#    It's calculated using `atan2(y, x)`, which correctly handles all four quadrants.
#    The result is typically in the range `(-pi, pi]`.
# 2.2 Python Function: `cmath.phase(z)` calculates the phase.

z_phase1 = 1 + 1j   # Quadrant 1
z_phase2 = -1 + 1j  # Quadrant 2
z_phase3 = -1 - 1j  # Quadrant 3
z_phase4 = 1 - 1j   # Quadrant 4
z_phase_pos_real = 5 + 0j
z_phase_neg_real = -5 + 0j
z_phase_pos_imag = 0 + 5j
z_phase_neg_imag = 0 - 5j

print(f"2.1 For z_phase1 = {z_phase1}:")
phase_z_phase1 = cmath.phase(z_phase1)
print(f"    Phase (cmath.phase({z_phase1})): {phase_z_phase1:.4f} radians (approx pi/4)") # Output: 0.7854

print(f"\n2.2 For z_phase2 = {z_phase2}:")
phase_z_phase2 = cmath.phase(z_phase2)
print(f"    Phase (cmath.phase({z_phase2})): {phase_z_phase2:.4f} radians (approx 3*pi/4)") # Output: 2.3562

print(f"\n2.3 For z_phase3 = {z_phase3}:")
phase_z_phase3 = cmath.phase(z_phase3)
print(f"    Phase (cmath.phase({z_phase3})): {phase_z_phase3:.4f} radians (approx -3*pi/4)") # Output: -2.3562

print(f"\n2.4 For z_phase4 = {z_phase4}:")
phase_z_phase4 = cmath.phase(z_phase4)
print(f"    Phase (cmath.phase({z_phase4})): {phase_z_phase4:.4f} radians (approx -pi/4)") # Output: -0.7854

print(f"\n2.5 Special cases on axes:")
print(f"    Phase of {z_phase_pos_real}: {cmath.phase(z_phase_pos_real):.4f} (0 radians)")
print(f"    Phase of {z_phase_neg_real}: {cmath.phase(z_phase_neg_real):.4f} (pi radians)")
print(f"    Phase of {z_phase_pos_imag}: {cmath.phase(z_phase_pos_imag):.4f} (pi/2 radians)")
print(f"    Phase of {z_phase_neg_imag}: {cmath.phase(z_phase_neg_imag):.4f} (-pi/2 radians)")


print("\n--- 3. Converting Between Rectangular and Polar Forms ---")

# The `cmath` module provides functions to convert between these representations.

# 3.1 `cmath.polar(z)`: Converts a complex number from rectangular to polar form.
# - Returns a tuple `(r, phi)`, where `r` is the magnitude and `phi` is the phase (in radians).
print(f"3.1 `cmath.polar(z)` (Rectangular to Polar):")
z_rect_to_polar = -3 + 4j # Quadrant 2

magnitude, phase = cmath.polar(z_rect_to_polar)
print(f"    For {z_rect_to_polar}:")
print(f"    Magnitude (r): {magnitude:.4f}") # Output: 5.0
print(f"    Phase (phi): {phase:.4f} radians") # Output: 2.2143 (approx)

# Verify with another common angle (e.g., 1+0j for 0 phase, magnitude 1)
mag_0, phase_0 = cmath.polar(1 + 0j)
print(f"    For 1+0j: Magnitude={mag_0:.4f}, Phase={phase_0:.4f}")

# 3.2 `cmath.rect(r, phi)`: Converts polar coordinates to a complex number (rectangular form).
# - `r`: The magnitude (non-negative real number).
# - `phi`: The phase angle (in radians).
print(f"\n3.2 `cmath.rect(r, phi)` (Polar to Rectangular):")

# Reconstruct z_rect_to_polar using its magnitude and phase
reconstructed_z = cmath.rect(magnitude, phase)
print(f"    From r={magnitude:.4f}, phi={phase:.4f}: {reconstructed_z:.4f}")
# Output: (-3.0000+4.0000j) (approximately the original z_rect_to_polar)

# Create a complex number from polar coordinates
r_val = 10.0
phi_val = math.pi / 6 # 30 degrees
z_from_polar = cmath.rect(r_val, phi_val)
print(f"    From r={r_val}, phi={phi_val:.4f} (30 deg): {z_from_polar:.4f}")
# Output: (8.6603+5.0000j) (10 * cos(30) + 10 * sin(30)j)

# Note: `cmath.rect(r, phi)` is equivalent to `r * (math.cos(phi) + math.sin(phi)*1j)`
# for real `r` and `phi`.
````

````python
# --- Python's `cmath` Module: All About Complex Conjugate in Code ---

# The complex conjugate is a fundamental concept in complex number arithmetic.
# For a complex number `z = a + bj`, its complex conjugate, denoted as `z*` or `z̄`,
# is `a - bj`. In essence, you just flip the sign of the imaginary part.

# In Python, the `cmath` module provides a function for this, but also, the built-in
# `complex` type itself has a `.conjugate()` method.

import cmath

print("--- 1. Understanding the Complex Conjugate ---")

# 1.1 Definition:
# If `z = a + bj` (where `a` is the real part and `b` is the imaginary part),
# then its complex conjugate `z̄ = a - bj`.

# Geometrically: The complex conjugate is the reflection of the complex number
# across the real axis in the complex plane.

# Examples:
# Conjugate of (2 + 3j) is (2 - 3j)
# Conjugate of (5j) is (-5j) (since 5j = 0 + 5j)
# Conjugate of (7) is (7) (since 7 = 7 + 0j)


print("\n--- 2. Using the `.conjugate()` Method of `complex` Type ---")

# This is the most direct and common way to get the complex conjugate in Python.
# Every complex number object has this method.

z1 = 2 + 3j
z2 = -1 - 4j
z3 = 5j
z4 = -7
z5 = complex(0.5, -0.8) # Using constructor

print(f"2.1 For z1 = {z1}:")
conjugate_z1 = z1.conjugate()
print(f"    z1.conjugate(): {conjugate_z1}") # Output: (2-3j)

print(f"\n2.2 For z2 = {z2}:")
conjugate_z2 = z2.conjugate()
print(f"    z2.conjugate(): {conjugate_z2}") # Output: (-1+4j)

print(f"\n2.3 For z3 = {z3} (pure imaginary):")
conjugate_z3 = z3.conjugate()
print(f"    z3.conjugate(): {conjugate_z3}") # Output: -5j

print(f"\n2.4 For z4 = {z4} (pure real):")
conjugate_z4 = z4.conjugate()
print(f"    z4.conjugate(): {conjugate_z4}") # Output: (-7+0j)

print(f"\n2.5 For z5 = {z5}:")
conjugate_z5 = z5.conjugate()
print(f"    z5.conjugate(): {conjugate_z5}") # Output: (0.5+0.8j)


print("\n--- 3. Using `cmath.conjugate()` Function ---")

# The `cmath` module also provides a `conjugate()` function. Its behavior is
# identical to the `.conjugate()` method of the `complex` type.
# This might be useful for consistency if you're already using other `cmath` functions.

z_cmath1 = 6 - 2j
z_cmath2 = -10j

print(f"3.1 For z_cmath1 = {z_cmath1}:")
cmath_conjugate_z1 = cmath.conjugate(z_cmath1)
print(f"    cmath.conjugate(z_cmath1): {cmath_conjugate_z1}") # Output: (6+2j)

print(f"\n3.2 For z_cmath2 = {z_cmath2}:")
cmath_conjugate_z2 = cmath.conjugate(z_cmath2)
print(f"    cmath.conjugate(z_cmath2): {cmath_conjugate_z2}") # Output: 10j


print("\n--- 4. Properties and Applications of Complex Conjugates ---")

# 4.1 Product of a Complex Number and Its Conjugate:
# The product of a complex number and its conjugate is always a non-negative real number,
# equal to the square of its magnitude.
# `z * z̄ = (a + bj)(a - bj) = a^2 - (bj)^2 = a^2 - b^2 j^2 = a^2 + b^2 = |z|^2`
z_product = 3 + 4j
product_result = z_product * z_product.conjugate()
magnitude_squared = abs(z_product)**2
print(f"4.1 For z_product = {z_product}:")
print(f"    z_product * z_product.conjugate() = {product_result}") # Output: (25+0j)
print(f"    abs(z_product)**2 = {magnitude_squared}") # Output: 25.0
print(f"    Are they equal? {product_result.real == magnitude_squared}") # True

# 4.2 Division of Complex Numbers:
# Conjugates are essential for dividing complex numbers. To divide `z1 / z2`,
# you multiply the numerator and denominator by the conjugate of the denominator (`z2̄`).
# This makes the denominator a real number.
# `(a + bj) / (c + dj) = (a + bj)(c - dj) / (c + dj)(c - dj) = (ac + bd + (bc - ad)j) / (c^2 + d^2)`

z_numerator = 2 + 3j
z_denominator = 1 - 2j

# Manual division using conjugate
manual_denominator_real = z_denominator * z_denominator.conjugate()
manual_numerator_new = z_numerator * z_denominator.conjugate()
manual_result = manual_numerator_new / manual_denominator_real
print(f"\n4.2 Complex Division using Conjugate:")
print(f"    ({z_numerator}) / ({z_denominator}) manually: {manual_result}") # Output: (0.8+1.4j)

# Python's built-in division handles this automatically
built_in_result = z_numerator / z_denominator
print(f"    ({z_numerator}) / ({z_denominator}) built-in: {built_in_result}") # Output: (0.8+1.4j)
print(f"    Are results equal? {manual_result == built_in_result}") # True

# 4.3 Extracting Real and Imaginary Parts (less common, but illustrative)
# Real part: `(z + z̄) / 2`
# Imaginary part: `(z - z̄) / (2j)`
z_extract = 4 + 6j
real_part_extracted = (z_extract + z_extract.conjugate()) / 2
imag_part_extracted = (z_extract - z_extract.conjugate()) / (2j)
print(f"\n4.3 Extracting parts using conjugate for {z_extract}:")
print(f"    Real part: {real_part_extracted}") # Output: (4+0j)
print(f"    Imaginary part: {imag_part_extracted}") # Output: (6+0j)
````

````python
# --- Python: All About Hyperbolic Functions in Code ---

# Hyperbolic functions are mathematical functions that are analogous to the
# ordinary trigonometric functions (sine, cosine, tangent) but are defined
# using the hyperbola rather than the circle.

# In Python, these functions are available in two main modules:
# 1.  `math`: For real numbers.
# 2.  `cmath`: For complex numbers.

import math
import cmath

print("--- 1. Basic Hyperbolic Functions (`math` module - for real numbers) ---")

# These functions take a real number as input and return a real number.

# 1.1 `math.sinh(x)`: Hyperbolic Sine of x.
#    Defined as `(e^x - e^-x) / 2`.
print(f"1.1 `math.sinh(x)` (Hyperbolic Sine):")
print(f"    sinh(0): {math.sinh(0):.4f}")        # Output: 0.0
print(f"    sinh(1): {math.sinh(1):.4f}")        # Output: 1.1752
print(f"    sinh(-1): {math.sinh(-1):.4f}")       # Output: -1.1752

# 1.2 `math.cosh(x)`: Hyperbolic Cosine of x.
#    Defined as `(e^x + e^-x) / 2`.
print(f"\n1.2 `math.cosh(x)` (Hyperbolic Cosine):")
print(f"    cosh(0): {math.cosh(0):.4f}")        # Output: 1.0
print(f"    cosh(1): {math.cosh(1):.4f}")        # Output: 1.5431
print(f"    cosh(-1): {math.cosh(-1):.4f}")       # Output: 1.5431 (cosh is an even function)

# 1.3 `math.tanh(x)`: Hyperbolic Tangent of x.
#    Defined as `sinh(x) / cosh(x)`.
print(f"\n1.3 `math.tanh(x)` (Hyperbolic Tangent):")
print(f"    tanh(0): {math.tanh(0):.4f}")        # Output: 0.0
print(f"    tanh(1): {math.tanh(1):.4f}")        # Output: 0.7616
print(f"    tanh(10): {math.tanh(10):.4f}")       # Approaches 1 as x increases
print(f"    tanh(-10): {math.tanh(-10):.4f}")     # Approaches -1 as x decreases


print("\n--- 2. Inverse Hyperbolic Functions (`math` module - for real numbers) ---")

# These functions return the value whose hyperbolic sine, cosine, or tangent is the input.
# They have specific input domains for real-valued outputs.

# 2.1 `math.asinh(x)`: Inverse Hyperbolic Sine of x.
#    - Input `x` can be any real number.
print(f"2.1 `math.asinh(x)` (Inverse Hyperbolic Sine):")
print(f"    asinh(0): {math.asinh(0):.4f}")      # Output: 0.0
print(f"    asinh(1): {math.asinh(1):.4f}")      # Output: 0.8814
print(f"    asinh(-1): {math.asinh(-1):.4f}")     # Output: -0.8814

# 2.2 `math.acosh(x)`: Inverse Hyperbolic Cosine of x.
#    - Input `x` must be greater than or equal to 1.
print(f"\n2.2 `math.acosh(x)` (Inverse Hyperbolic Cosine):")
print(f"    acosh(1): {math.acosh(1):.4f}")      # Output: 0.0
print(f"    acosh(2): {math.acosh(2):.4f}")      # Output: 1.3170

# What happens if input is out of range? (ValueError)
try:
    math.acosh(0.5) # Input < 1
except ValueError as e:
    print(f"    Error for acosh(0.5): {e} - math domain error (input < 1).")


# 2.3 `math.atanh(x)`: Inverse Hyperbolic Tangent of x.
#    - Input `x` must be strictly between -1 and 1 (exclusive).
print(f"\n2.3 `math.atanh(x)` (Inverse Hyperbolic Tangent):")
print(f"    atanh(0): {math.atanh(0):.4f}")      # Output: 0.0
print(f"    atanh(0.5): {math.atanh(0.5):.4f}")    # Output: 0.5493

# What happens if input is out of range? (ValueError)
try:
    math.atanh(1.0) # Input >= 1
except ValueError as e:
    print(f"    Error for atanh(1.0): {e} - math domain error (input not in (-1, 1)).")

try:
    math.atanh(-1.0) # Input <= -1
except ValueError as e:
    print(f"    Error for atanh(-1.0): {e} - math domain error (input not in (-1, 1)).")


print("\n--- 3. Hyperbolic Functions (`cmath` module - for complex numbers) ---")

# The `cmath` module provides versions of these functions that can handle
# and return complex numbers. They have wider domains than their `math` counterparts.

# 3.1 `cmath.sinh(z)`: Complex Hyperbolic Sine.
z_complex_sinh = 0.5 + 1.2j
print(f"3.1 `cmath.sinh(z)` (Complex Hyperbolic Sine):")
print(f"    sinh({z_complex_sinh}): {cmath.sinh(z_complex_sinh):.4f}")

# Example with a real number input (still returns a complex number with 0 imaginary part)
print(f"    sinh(1.0): {cmath.sinh(1.0):.4f}") # Output: (1.1752+0.0000j)

# 3.2 `cmath.cosh(z)`: Complex Hyperbolic Cosine.
z_complex_cosh = -0.3 + 0.9j
print(f"\n3.2 `cmath.cosh(z)` (Complex Hyperbolic Cosine):")
print(f"    cosh({z_complex_cosh}): {cmath.cosh(z_complex_cosh):.4f}")

# 3.3 `cmath.tanh(z)`: Complex Hyperbolic Tangent.
z_complex_tanh = 1.0 + 0.5j
print(f"\n3.3 `cmath.tanh(z)` (Complex Hyperbolic Tangent):")
print(f"    tanh({z_complex_tanh}): {cmath.tanh(z_complex_tanh):.4f}")


print("\n--- 4. Inverse Hyperbolic Functions (`cmath` module - for complex numbers) ---")

# These inverse functions also accept and return complex numbers.

# 4.1 `cmath.asinh(z)`: Complex Inverse Hyperbolic Sine.
z_asinh_complex = 1.5 + 0.5j
print(f"4.1 `cmath.asinh(z)` (Complex Inverse Hyperbolic Sine):")
print(f"    asinh({z_asinh_complex}): {cmath.asinh(z_asinh_complex):.4f}")

# Example where `math.asinh` (for real numbers) accepts this, but this is a complex number
print(f"    asinh(0.5): {cmath.asinh(0.5):.4f}") # Output: (0.4812+0.0000j)

# 4.2 `cmath.acosh(z)`: Complex Inverse Hyperbolic Cosine.
#    - Handles inputs less than 1, returning a complex number.
z_acosh_complex = 0.8 - 0.3j
print(f"\n4.2 `cmath.acosh(z)` (Complex Inverse Hyperbolic Cosine):")
print(f"    acosh({z_acosh_complex}): {cmath.acosh(z_acosh_complex):.4f}")

# Example where `math.acosh` would error (input < 1) - `cmath` handles it
print(f"    acosh(0.5): {cmath.acosh(0.5):.4f}") # Output: (0.0000+1.0472j)

# 4.3 `cmath.atanh(z)`: Complex Inverse Hyperbolic Tangent.
#    - Handles inputs outside (-1, 1), returning a complex number.
z_atanh_complex = 2.0 + 0.1j
print(f"\n4.3 `cmath.atanh(z)` (Complex Inverse Hyperbolic Tangent):")
print(f"    atanh({z_atanh_complex}): {cmath.atanh(z_atanh_complex):.4f}")

# Example where `math.atanh` would error (input >= 1)
print(f"    atanh(1.5): {cmath.atanh(1.5):.4f}") # Output: (0.8047+1.5708j)
````

````python
# --- Python: All About Inverse Hyperbolic Functions in Code ---

# Inverse hyperbolic functions (also known as area hyperbolic functions) are the
# inverse operations of the basic hyperbolic functions (sinh, cosh, tanh).
# They are used to find the value whose hyperbolic sine, cosine, or tangent is the input.

# In Python, these functions are available in two modules:
# 1.  `math`: For real numbers.
# 2.  `cmath`: For complex numbers.

import math
import cmath

print("--- 1. Inverse Hyperbolic Functions (`math` module - for real numbers) ---")

# These functions take a real number as input and return a real number.
# They have specific input domains for their real-valued outputs.

# 1.1 `math.asinh(x)`: Inverse Hyperbolic Sine of x (Area Sine Hyperbolic).
#    - Input `x` can be any real number.
print(f"1.1 `math.asinh(x)` (asinh):")
print(f"    asinh(0): {math.asinh(0):.4f}")        # Output: 0.0
print(f"    asinh(1): {math.asinh(1):.4f}")        # Output: 0.8814
print(f"    asinh(-1): {math.asinh(-1):.4f}")       # Output: -0.8814
print(f"    asinh(10): {math.asinh(10):.4f}")       # Output: 2.9982

# 1.2 `math.acosh(x)`: Inverse Hyperbolic Cosine of x (Area Cosine Hyperbolic).
#    - Input `x` must be greater than or equal to 1.
#    - Result is always non-negative.
print(f"\n1.2 `math.acosh(x)` (acosh):")
print(f"    acosh(1): {math.acosh(1):.4f}")        # Output: 0.0
print(f"    acosh(2): {math.acosh(2):.4f}")        # Output: 1.3170
print(f"    acosh(5): {math.acosh(5):.4f}")        # Output: 2.2924

# What happens if input is out of range? (ValueError)
try:
    math.acosh(0.5) # Input < 1
except ValueError as e:
    print(f"    Error for acosh(0.5): {e} - math domain error (input < 1 for real results).")


# 1.3 `math.atanh(x)`: Inverse Hyperbolic Tangent of x (Area Tangent Hyperbolic).
#    - Input `x` must be strictly between -1 and 1 (exclusive), i.e., `(-1 < x < 1)`.
print(f"\n1.3 `math.atanh(x)` (atanh):")
print(f"    atanh(0): {math.atanh(0):.4f}")        # Output: 0.0
print(f"    atanh(0.5): {math.atanh(0.5):.4f}")      # Output: 0.5493
print(f"    atanh(-0.5): {math.atanh(-0.5):.4f}")     # Output: -0.5493

# What happens if input is out of range? (ValueError)
try:
    math.atanh(1.0) # Input is 1 (not exclusive)
except ValueError as e:
    print(f"    Error for atanh(1.0): {e} - math domain error (input not in (-1, 1)).")

try:
    math.atanh(-1.0) # Input is -1 (not exclusive)
except ValueError as e:
    print(f"    Error for atanh(-1.0): {e} - math domain error (input not in (-1, 1)).")

try:
    math.atanh(2.0) # Input > 1
except ValueError as e:
    print(f"    Error for atanh(2.0): {e} - math domain error (input not in (-1, 1)).")


print("\n--- 2. Inverse Hyperbolic Functions (`cmath` module - for complex numbers) ---")

# The `cmath` module provides versions of these functions that can handle
# and return complex numbers. Their domains are extended to cover the entire
# complex plane.

# 2.1 `cmath.asinh(z)`: Complex Inverse Hyperbolic Sine.
#    - Can take any complex number (or real number) as input.
#    - Returns a complex number.
z_asinh_complex = 1.5 + 0.5j
print(f"2.1 `cmath.asinh(z)` (Complex asinh):")
print(f"    asinh({z_asinh_complex}): {cmath.asinh(z_asinh_complex):.4f}")

# Example where `math.asinh` (for real numbers) accepts the real part,
# `cmath` returns a complex result for the complex number.
print(f"    asinh(0.5): {cmath.asinh(0.5):.4f}") # Output: (0.4812+0.0000j)


# 2.2 `cmath.acosh(z)`: Complex Inverse Hyperbolic Cosine.
#    - Handles inputs that would be out of range for `math.acosh` (i.e., `x < 1`),
#      by returning a complex number.
print(f"\n2.2 `cmath.acosh(z)` (Complex acosh):")
z_acosh_complex = 0.8 - 0.3j
print(f"    acosh({z_acosh_complex}): {cmath.acosh(z_acosh_complex):.4f}")

# Example where `math.acosh` would error (input < 1) - `cmath` handles it
print(f"    acosh(0.5): {cmath.acosh(0.5):.4f}") # Output: (0.0000+1.0472j)


# 2.3 `cmath.atanh(z)`: Complex Inverse Hyperbolic Tangent.
#    - Handles inputs that would be out of range for `math.atanh` (i.e., `x <= -1` or `x >= 1`),
#      by returning a complex number.
print(f"\n2.3 `cmath.atanh(z)` (Complex atanh):")
z_atanh_complex = 2.0 + 0.1j # Example input outside (-1, 1) range for real atanh
print(f"    atanh({z_atanh_complex}): {cmath.atanh(z_atanh_complex):.4f}")

# Example where `math.atanh` would error (input >= 1)
print(f"    atanh(1.5): {cmath.atanh(1.5):.4f}") # Output: (0.8047+1.5708j)

# Example with a pure imaginary number
print(f"    atanh(2j): {cmath.atanh(2j):.4f}") # Output: (0.0000+1.1071j)


print("\n--- 3. Choosing Between `math` and `cmath` ---")

# - **Use `math` module:** When you are working exclusively with **real numbers** and expect
#   real-valued results. The `ValueError` it raises for out-of-domain inputs can be
#   helpful for debugging or validating input constraints.
# - **Use `cmath` module:** When you are working with **complex numbers** or if your
#   calculations might naturally lead to complex results (e.g., trying to find `acosh(0.5)`
#   in a broader context). `cmath` provides a seamless extension to the complex plane.
````

````python
# --- Python: All About Exponential and Logarithmic Functions in Code ---

# Python provides robust support for exponential and logarithmic calculations
# through its `math` module (for real numbers) and `cmath` module (for complex numbers).

import math
import cmath

print("--- 1. Exponential Functions ---")

# These functions typically involve Euler's number 'e' (approximately 2.71828)
# as the base.

# 1.1 `math.exp(x)`: Returns e raised to the power x ($e^x$).
#    - This function is for real-valued inputs and outputs.
print(f"1.1 `math.exp(x)` (e^x for real numbers):")
print(f"    e^0 (math.exp(0)): {math.exp(0)}")   # Output: 1.0
print(f"    e^1 (math.exp(1)): {math.exp(1):.4f}") # Output: 2.7183
print(f"    e^2 (math.exp(2)): {math.exp(2):.4f}") # Output: 7.3891
print(f"    e^-1 (math.exp(-1)): {math.exp(-1):.4f}") # Output: 0.3679

# 1.2 `math.expm1(x)`: Returns $e^x - 1$.
#    - This is a specialized function that provides higher accuracy for small values of `x`
#      compared to calculating `math.exp(x) - 1` directly, due to floating-point precision issues.
#    - For real-valued inputs and outputs.
print(f"\n1.2 `math.expm1(x)` (e^x - 1 for real numbers):")
small_positive_x = 1e-9 # A very small positive number
print(f"    e^({small_positive_x}) - 1 (math.expm1({small_positive_x})): {math.expm1(small_positive_x):.10f}")
# Compare with direct calculation, which might show less precision for very small x
print(f"    (math.exp({small_positive_x}) - 1): {(math.exp(small_positive_x) - 1):.10f}")

# 1.3 `cmath.exp(z)`: Returns e raised to the power z ($e^z$).
#    - This function is designed for complex numbers. It can take a real or complex number
#      as input and always returns a complex number.
print(f"\n1.3 `cmath.exp(z)` (e^z for complex numbers):")
z_complex_exp = 0 + (math.pi / 2) * 1j # Represents 0 + (pi/2)i, which is 1j
print(f"    e^(0 + pi/2 j) (cmath.exp({z_complex_exp})): {cmath.exp(z_complex_exp):.4f}") # Output: (0.0000+1.0000j) (approx 1j)

real_as_complex_input = 2 + 0j
print(f"    e^(2+0j) (cmath.exp({real_as_complex_input})): {cmath.exp(real_as_complex_input):.4f}") # Output: (7.3891+0.0000j) (same as math.exp(2))


print("\n--- 2. Logarithmic Functions ---")

# Logarithmic functions are the inverse of exponential functions.
# `log(x, base)` means "to what power must 'base' be raised to get 'x'".
# For real number logarithms, the input `x` (the argument of the logarithm) must be positive.

# 2.1 `math.log(x, base=e)`: Returns the logarithm of x to the given base.
#    - If `base` is not specified, it calculates the natural logarithm (base `e`, `ln(x)`).
#    - For real-valued inputs and outputs.
print(f"2.1 `math.log(x, base)` (log_base(x) for real numbers):")
print(f"    ln(e^2) (math.log(math.e**2)): {math.log(math.e**2)}") # Output: 2.0
print(f"    ln(1) (math.log(1)): {math.log(1)}") # Output: 0.0
print(f"    log_10(100) (math.log(100, 10)): {math.log(100, 10)}") # Output: 2.0

# 2.2 `math.log1p(x)`: Returns the natural logarithm of (1+x) ($\ln(1+x)$).
#    - Similar to `expm1`, `log1p` provides higher accuracy for small values of `x` (i.e., when `1+x` is close to 1).
#    - For real-valued inputs and outputs.
print(f"\n2.2 `math.log1p(x)` (ln(1+x) for real numbers):")
print(f"    ln(1 + {small_positive_x}) (math.log1p({small_positive_x})): {math.log1p(small_positive_x):.10f}")
print(f"    (math.log(1 + {small_positive_x})): {(math.log(1 + small_positive_x)):.10f}")

# 2.3 `math.log2(x)`: Returns the base-2 logarithm of x ($\log_2 x$).
#    - For real-valued inputs and outputs.
print(f"\n2.3 `math.log2(x)` (log_2(x) for real numbers):")
print(f"    log2(8) (math.log2(8)): {math.log2(8)}") # Output: 3.0
print(f"    log2(1024) (math.log2(1024)): {math.log2(1024)}") # Output: 10.0

# 2.4 `math.log10(x)`: Returns the base-10 logarithm of x ($\log_{10} x$).
#    - For real-valued inputs and outputs.
print(f"\n2.4 `math.log10(x)` (log_10(x) for real numbers):")
print(f"    log10(1000) (math.log10(1000)): {math.log10(1000)}") # Output: 3.0
print(f"    log10(1) (math.log10(1)): {math.log10(1)}") # Output: 0.0

# Domain errors for `math` logarithms (input must be strictly positive)
try:
    math.log(0) # Logarithm of zero is undefined
except ValueError as e:
    print(f"\n    Error for log(0): {e} - math domain error.")
try:
    math.log(-5) # Logarithm of a negative number is undefined in real numbers
except ValueError as e:
    print(f"    Error for log(-5): {e} - math domain error.")

# 2.5 `cmath.log(z, base=e)`: Returns the logarithm of z for complex numbers.
#    - This function handles inputs that would be out of the real domain for `math.log`.
#      It can take zero or negative real numbers, returning a complex result.
print(f"\n2.5 `cmath.log(z, base)` (log for complex numbers):")
z_log_negative_real = -1 + 0j # Represents -1 (a negative real number)
print(f"    ln(-1) (cmath.log({z_log_negative_real})): {cmath.log(z_log_negative_real):.4f}") # Output: (0.0000+3.1416j) (approx j*pi)

z_complex_log_input = 1 + 1j
print(f"    ln(1+1j) (cmath.log({z_complex_log_input})): {cmath.log(z_complex_log_input):.4f}")

# 2.6 `cmath.log10(z)`: Returns the base-10 logarithm of z for complex numbers.
print(f"\n2.6 `cmath.log10(z)` (log_10 for complex numbers):")
print(f"    log10(-1) (cmath.log10({z_log_negative_real})): {cmath.log10(z_log_negative_real):.4f}")


print("\n--- 3. Power Functions ---")

# Power functions raise a base number to an exponent.

# 3.1 Built-in `**` operator: The most common and versatile way to calculate powers.
#    - Works seamlessly with integers, floats, and complex numbers.
#    - If the operation naturally results in a complex number (e.g., square root of a negative number),
#      it automatically returns a complex number.
print(f"3.1 Built-in `**` operator (x ** y):")
print(f"    2 to the power 3 (2 ** 3): {2 ** 3}") # Output: 8
print(f"    4 to the power 0.5 (4 ** 0.5): {4 ** 0.5}") # Output: 2.0 (square root)
print(f"    8 to the power 1/3 (8 ** (1/3)): {8 ** (1/3):.4f}") # Output: 2.0000 (cube root)

# Complex number results from real input
print(f"    -1 to the power 0.5 (-1 ** 0.5): {(-1) ** 0.5}") # Output: (6.123233995736766e-17+1j) (approx 1j)

# Complex base and exponent
complex_base_pow = 2 + 3j
complex_exponent_pow = 0.5 + 0.2j
print(f"    ({complex_base_pow}) ** 2: {complex_base_pow ** 2}") # Output: (-5+12j)
print(f"    ({complex_base_pow}) ** ({complex_exponent_pow}): {complex_base_pow ** complex_exponent_pow:.4f}")


# 3.2 `math.pow(x, y)`: Returns x raised to the power y ($x^y$).
#    - Always converts its arguments to float and returns a float.
#    - Raises `ValueError` for cases that would yield complex results (e.g., negative base with fractional exponent).
print(f"\n3.2 `math.pow(x, y)` (x^y for real numbers):")
print(f"    math.pow(2, 3): {math.pow(2, 3)}") # Output: 8.0
print(f"    math.pow(4, 0.5): {math.pow(4, 0.5)}") # Output: 2.0

try:
    math.pow(-1, 0.5) # Negative base with fractional exponent results in a complex number
except ValueError as e:
    print(f"    Error for math.pow(-1, 0.5): {e} - math domain error.")


# 3.3 `math.sqrt(x)`: Returns the square root of x ($\sqrt{x}$).
#    - A specialized power function (`x**0.5`). Input must be non-negative.
#    - For real numbers.
print(f"\n3.3 `math.sqrt(x)` (Square Root for real numbers):")
print(f"    math.sqrt(25): {math.sqrt(25)}") # Output: 5.0
print(f"    math.sqrt(2): {math.sqrt(2):.4f}") # Output: 1.4142

# 3.4 `cmath.sqrt(z)`: Returns the square root of z for complex numbers.
#    - Handles negative real numbers or complex numbers, always returning a complex result.
print(f"\n3.4 `cmath.sqrt(z)` (Square Root for complex numbers):")
print(f"    cmath.sqrt(-1): {cmath.sqrt(-1)}") # Output: 1j
print(f"    cmath.sqrt(3+4j): {cmath.sqrt(3+4j):.4f}") # Output: (2.0000+1.0000j)


# 3.5 `cmath.pow(z1, z2)`: Returns z1 raised to the power z2 ($z_1^{z_2}$) for complex numbers.
#    - Handles complex bases and exponents, always returns a complex number.
print(f"\n3.5 `cmath.pow(z1, z2)` (x^y for complex numbers):")
print(f"    cmath.pow({complex_base_pow}, {complex_exponent_pow}): {cmath.pow(complex_base_pow, complex_exponent_pow):.4f}")
print(f"    cmath.pow(-1, 0.5): {cmath.pow(-1, 0.5)}") # Output: (0.0000+1.0000j) (approx 1j)


print("\n--- 4. Summary and Key Use Cases ---")

print("Key takeaways for choosing the right function:")
print("- **`math` module functions:** Best for operations exclusively involving real numbers.")
print("  They will raise a `ValueError` if the result would enter the complex domain, which can be useful for strict real number calculations.")
print("- **`cmath` module functions:** Essential when dealing with complex numbers, or when inputs might naturally lead to complex results.")
print("  They seamlessly extend operations to the complex plane.")
print("- **Built-in `**` operator:** This is the most convenient and Pythonic way for general power calculations.")
print("  It's versatile enough to handle both real and complex numbers as bases and exponents, returning a complex number if necessary.")
````

````python
# --- Python: All About Square Root Functions in Code ---

# In Python, you can calculate the square root of a number using several methods,
# depending on whether you're dealing with real numbers, complex numbers, or
# prefer using operators.

import math
import cmath

print("--- 1. Using `math.sqrt()` (for real numbers) ---")

# 1.1 `math.sqrt(x)`: Returns the square root of x ($\sqrt{x}$).
#    - **Input:** Must be a non-negative real number.
#    - **Output:** A float.
#    - **Error:** Raises a `ValueError` if the input is negative.

print(f"1.1 Square root of positive numbers:")
print(f"    sqrt(9): {math.sqrt(9)}")       # Output: 3.0
print(f"    sqrt(2.25): {math.sqrt(2.25)}")  # Output: 1.5
print(f"    sqrt(0): {math.sqrt(0)}")       # Output: 0.0
print(f"    sqrt(2): {math.sqrt(2):.4f}")    # Output: 1.4142

# 1.2 Handling negative input (raises ValueError)
try:
    print(f"\n1.2 Attempting sqrt(-1):")
    math.sqrt(-1)
except ValueError as e:
    print(f"    Error: {e} - `math.sqrt()` raises ValueError for negative input.")

try:
    print(f"\n    Attempting sqrt(-9):")
    math.sqrt(-9)
except ValueError as e:
    print(f"    Error: {e} - `math.sqrt()` raises ValueError for negative input.")


print("\n--- 2. Using `cmath.sqrt()` (for complex numbers) ---")

# 2.1 `cmath.sqrt(z)`: Returns the square root of z ($\sqrt{z}$).
#    - **Input:** Can be any number (real or complex).
#    - **Output:** Always returns a complex number.
#    - **Handling negative real numbers:** Returns a pure imaginary complex number for negative real inputs.

print(f"2.1 Square root of positive real numbers:")
print(f"    sqrt(9) (cmath): {cmath.sqrt(9)}")       # Output: (3+0j)
print(f"    sqrt(2) (cmath): {cmath.sqrt(2):.4f}")    # Output: (1.4142+0.0000j)

print(f"\n2.2 Square root of negative real numbers:")
print(f"    sqrt(-1) (cmath): {cmath.sqrt(-1)}")     # Output: 1j
print(f"    sqrt(-9) (cmath): {cmath.sqrt(-9)}")     # Output: 3j

print(f"\n2.3 Square root of complex numbers:")
complex_num = 3 + 4j # Common complex number example
print(f"    sqrt(3+4j) (cmath): {cmath.sqrt(complex_num):.4f}") # Output: (2.0000+1.0000j)

complex_num_neg = -5 - 12j
print(f"    sqrt(-5-12j) (cmath): {cmath.sqrt(complex_num_neg):.4f}") # Output: (2.0000-3.0000j)


print("\n--- 3. Using the `**` Operator (Power Operator) ---")

# 3.1 `x ** 0.5`: Raising a number to the power of 0.5 is equivalent to taking its square root.
#    - **Input:** Can be any number (integer, float, complex).
#    - **Output:** Returns a float or a complex number, depending on the input and result.
#    - **Behavior for negative inputs:** Returns a complex number if the base is negative.

print(f"3.1 Square root using `** 0.5` for positive numbers:")
print(f"    9 ** 0.5: {9 ** 0.5}")       # Output: 3.0
print(f"    2.25 ** 0.5: {2.25 ** 0.5}")  # Output: 1.5

print(f"\n3.2 Square root using `** 0.5` for negative numbers:")
print(f"    -1 ** 0.5: {(-1) ** 0.5}")     # Output: (6.123233995736766e-17+1j) (approx 1j)
print(f"    -9 ** 0.5: {(-9) ** 0.5}")     # Output: (1.8369701987210297e-16+3j) (approx 3j)

print(f"\n3.3 Square root using `** 0.5` for complex numbers:")
print(f"    (3+4j) ** 0.5: {(3+4j) ** 0.5:.4f}") # Output: (2.0000+1.0000j)


print("\n--- 4. Choosing the Right Square Root Method ---")

# - **`math.sqrt()`:**
#   - **Best for:** When you are *certain* your input is a non-negative real number
#     and you expect a real-valued float result. It's concise and clear about its domain.
#   - **Caution:** Will raise `ValueError` for negative inputs.

# - **`cmath.sqrt()`:**
#   - **Best for:** When you might have **negative or complex inputs**, and you need
#     the square root to be calculated in the complex domain. Always returns a complex number.
#   - **Caution:** Always returns a complex number, even for purely real, positive inputs (e.g., `cmath.sqrt(9)` is `(3+0j)`).

# - **`** 0.5` operator:**
#   - **Best for:** A general-purpose solution that handles both real and complex numbers
#     transparently. It's often the most concise and flexible choice if you don't
#     strictly need `ValueError` for negative real inputs.
#   - **Caution:** For purely real, positive inputs, it returns a float (e.g., `9 ** 0.5` is `3.0`),
#     but for negative inputs, it switches to complex numbers, which can be a subtle behavior
#     if you're always expecting a float.

# Example illustrating output types:
print("\n--- 5. Illustrating Output Types ---")
num_pos = 4
num_neg = -4

print(f"Input: {num_pos}")
print(f"  math.sqrt({num_pos}): {math.sqrt(num_pos)}, Type: {type(math.sqrt(num_pos))}")
print(f"  cmath.sqrt({num_pos}): {cmath.sqrt(num_pos)}, Type: {type(cmath.sqrt(num_pos))}")
print(f"  {num_pos} ** 0.5: {num_pos ** 0.5}, Type: {type(num_pos ** 0.5)}")

print(f"\nInput: {num_neg}")
try:
    math.sqrt(num_neg)
except ValueError as e:
    print(f"  math.sqrt({num_neg}): {e}")
print(f"  cmath.sqrt({num_neg}): {cmath.sqrt(num_neg)}, Type: {type(cmath.sqrt(num_neg))}")
print(f"  {num_neg} ** 0.5: {num_neg ** 0.5}, Type: {type(num_neg ** 0.5)}")
````

````python
# --- Python: All About Predicate Functions in Code ---

# In programming, a "predicate function" (or simply a "predicate") is a function
# that returns a boolean value (True or False) based on its input.
# They are typically used for testing conditions, filtering data, or making decisions
# within control flow statements.

# --- 1. Basic Predicate Functions ---

print("--- 1. Basic Predicate Functions ---")

# Predicates are often named starting with "is_", "has_", "can_", or ending with "?".

# 1.1 `is_even(number)`: Checks if a number is even.
def is_even(number):
    """Returns True if the number is even, False otherwise."""
    return number % 2 == 0

print(f"1.1 Is 4 even? {is_even(4)}")   # Output: True
print(f"    Is 7 even? {is_even(7)}")   # Output: False
print(f"    Is 0 even? {is_even(0)}")   # Output: True

# 1.2 `is_positive(number)`: Checks if a number is positive.
def is_positive(number):
    """Returns True if the number is positive, False otherwise."""
    return number > 0

print(f"\n1.2 Is 10 positive? {is_positive(10)}") # Output: True
print(f"    Is -5 positive? {is_positive(-5)}") # Output: False
print(f"    Is 0 positive? {is_positive(0)}")   # Output: False

# 1.3 `has_vowels(word)`: Checks if a string contains any vowels.
def has_vowels(word):
    """Returns True if the word contains any vowel (a, e, i, o, u), case-insensitive."""
    vowels = "aeiou"
    return any(char in vowels for char in word.lower())

print(f"\n1.3 Does 'hello' have vowels? {has_vowels('hello')}") # Output: True
print(f"    Does 'rhythm' have vowels? {has_vowels('rhythm')}") # Output: False
print(f"    Does 'PYTHON' have vowels? {has_vowels('PYTHON')}") # Output: True


print("\n--- 2. Predicates in Conditional Statements ---")

# Predicate functions are very naturally used within `if`/`elif`/`else` blocks.

def describe_number(number):
    if is_positive(number):
        if is_even(number):
            print(f"2.1 {number} is a positive even number.")
        else:
            print(f"    {number} is a positive odd number.")
    elif number < 0:
        print(f"    {number} is a negative number.")
    else: # number must be 0
        print(f"    {number} is zero (neither positive nor negative, but even).")

describe_number(6)
describe_number(-3)
describe_number(0)
describe_number(9)


print("\n--- 3. Predicates in Filtering Operations ---")

# Predicates are commonly used with `filter()` or list/generator comprehensions
# to select elements from a collection that meet a certain condition.

# 3.1 Using `filter()` with a predicate
numbers_list = [1, -2, 3, -4, 5, 0, 6, -7]
positive_numbers = list(filter(is_positive, numbers_list))
print(f"3.1 Positive numbers from {numbers_list}: {positive_numbers}")

even_numbers = list(filter(is_even, numbers_list))
print(f"    Even numbers from {numbers_list}: {even_numbers}")

# 3.2 Using list comprehensions with a predicate (more common in Python)
names = ["Alice", "Bob", "Charlie", "David", "Eve"]

def starts_with_A(name):
    """Predicate: Returns True if name starts with 'A'."""
    return name.startswith('A')

names_starting_with_A = [name for name in names if starts_with_A(name)]
print(f"\n3.2 Names starting with 'A' from {names}: {names_starting_with_A}")

# Predicate for objects
class User:
    def __init__(self, name, age, is_admin=False):
        self.name = name
        self.age = age
        self.is_admin = is_admin

    def __repr__(self):
        return f"User(name='{self.name}', age={self.age}, admin={self.is_admin})"

def is_adult(user):
    """Predicate: Returns True if user is an adult (age >= 18)."""
    return user.age >= 18

def is_admin_user(user):
    """Predicate: Returns True if user is an admin."""
    return user.is_admin

users = [
    User("Anna", 22, True),
    User("Ben", 16, False),
    User("Chris", 35, False),
    User("Diana", 19, True)
]

adult_users = [user for user in users if is_adult(user)]
print(f"\n3.3 Adult users: {adult_users}")

admin_users = [user for user in users if is_admin_user(user)]
print(f"    Admin users: {admin_users}")


print("\n--- 4. Predicates in `all()` and `any()` ---")

# The built-in `all()` and `any()` functions are perfect for working with predicates.

# 4.1 `all(iterable)`: Returns True if all elements of the iterable are true (or the iterable is empty).
#    - Often used with generator expressions which include a predicate.
numbers_to_check_all = [2, 4, 6, 8]
print(f"4.1 Are all numbers in {numbers_to_check_all} even? {all(is_even(num) for num in numbers_to_check_all)}") # True

numbers_to_check_all_mixed = [2, 4, 5, 8]
print(f"    Are all numbers in {numbers_to_check_all_mixed} even? {all(is_even(num) for num in numbers_to_check_all_mixed)}") # False

# 4.2 `any(iterable)`: Returns True if any element of the iterable is true.
#    - Also often used with generator expressions which include a predicate.
numbers_to_check_any = [1, 3, 5, 6]
print(f"\n4.2 Is any number in {numbers_to_check_any} even? {any(is_even(num) for num in numbers_to_check_any)}") # True

numbers_to_check_any_none = [1, 3, 5, 7]
print(f"    Is any number in {numbers_to_check_any_none} even? {any(is_even(num) for num in numbers_to_check_any_none)}") # False


print("\n--- 5. Lambda Functions as Inline Predicates ---")

# Lambda functions are often used to create simple, anonymous predicate functions on the fly.

# 5.1 Using lambda with `filter()`
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers_list))
print(f"5.1 Odd numbers (lambda with filter): {odd_numbers}")

# 5.2 Using lambda with list comprehensions
positive_and_odd = [num for num in numbers_list if num > 0 and num % 2 != 0]
print(f"5.2 Positive and odd numbers (list comp): {positive_and_odd}")

# More explicit lambda predicate for user age check
young_users = [user for user in users if (lambda u: u.age < 18)(user)]
print(f"    Young users (lambda): {young_users}")


print("\n--- 6. The Role of Predicate Functions ---")

print("Predicate functions enhance code clarity and reusability by:")
print("- Encapsulating conditional logic into named, testable units.")
print("- Making code more readable by describing the *intent* of a check.")
print("- Facilitating functional programming patterns (e.g., with `filter`, `all`, `any`).")
print("- Improving maintainability by centralizing condition definitions.")
````
