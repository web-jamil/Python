### **Important Functions in the `pyjokes` Module:**

#### 1. **`pyjokes.get_joke(language='en', category='general')`**

- **Description:** The `get_joke` function retrieves a random joke from the specified language and category. You can specify the language (e.g., "en" for English) and category (e.g., "general," "programming," "knock-knock") to filter the joke selection.

- **Parameters:**

  - `language`: A string indicating the language of the joke. Default is 'en' (English).
  - `category`: A string indicating the type of joke to retrieve. Default is 'general'. Other valid values are 'programming', 'knock-knock', and 'neutral'.

- **Example Usage:**

  ```python
  import pyjokes

  # Get a random joke in English from the "general" category
  joke = pyjokes.get_joke(language='en', category='general')
  print(joke)

  # Get a random programming joke in Spanish
  joke = pyjokes.get_joke(language='es', category='programming')
  print(joke)
  ```

  **Output:**

  ```text
  Why do programmers prefer dark mode? Because the light attracts bugs.
  ```

#### 2. **`pyjokes.get_jokes(language='en', category='general')`**

- **Description:** This function is similar to `get_joke`, but instead of returning a single joke, it returns a list of jokes from the specified category and language.

- **Parameters:**

  - `language`: The language of the jokes. Default is 'en' (English).
  - `category`: The category of jokes to retrieve. Can be 'general', 'programming', 'knock-knock', or 'neutral'.

- **Example Usage:**

  ```python
  import pyjokes

  # Get a list of 3 random jokes in English from the "programming" category
  jokes = pyjokes.get_jokes(language='en', category='programming')
  for joke in jokes:
      print(joke)
  ```

  **Output:**

  ```text
  Why do programmers prefer dark mode? Because the light attracts bugs.
  Why do Java developers wear glasses? Because they can’t C#.
  What’s a programmer’s favorite hangout place? The Foo Bar.
  ```

#### 3. **`pyjokes.get_joke_list(language='en', category='general')`**

- **Description:** This function is an extension of the `get_jokes()` function and returns a list of jokes in the form of strings. It can be used to retrieve multiple jokes from a specific category or language.

python
import pyjokes
import json
import getpass
from pathlib import Path
from datetime import datetime
from collections import defaultdict
import random
from colorama import Fore, Style, init

# Initialize colorama for colored text

init(autoreset=True)

# File paths

USERS_FILE = "users.json"
JOKE_HISTORY_FILE = "joke_history.json"

# Supported languages and categories

SUPPORTED_LANGUAGES = ['en', 'de', 'es', 'fr', 'gl', 'eu']
SUPPORTED_CATEGORIES = ['neutral', 'chuck', 'all']

def initialize_files():
"""Create JSON files if they don't exist"""
if not Path(USERS_FILE).exists():
with open(USERS_FILE, 'w') as f:
json.dump({"admin": {"password": "admin123", "role": "admin"}}, f)

    if not Path(JOKE_HISTORY_FILE).exists():
        with open(JOKE_HISTORY_FILE, 'w') as f:
            json.dump({}, f)

def register_user():
"""Register a new user with password confirmation"""
username = input(Fore.CYAN + "Enter new username: " + Style.RESET_ALL)

    with open(USERS_FILE, 'r') as f:
        users = json.load(f)

    if username in users:
        print(Fore.RED + "Username already exists!" + Style.RESET_ALL)
        return False

    password = getpass.getpass(Fore.CYAN + "Enter password: " + Style.RESET_ALL)
    confirm = getpass.getpass(Fore.CYAN + "Confirm password: " + Style.RESET_ALL)

    if password != confirm:
        print(Fore.RED + "Passwords don't match!" + Style.RESET_ALL)
        return False

    users[username] = {"password": password, "role": "user"}
    with open(USERS_FILE, 'w') as f:
        json.dump(users, f, indent=4)

    print(Fore.GREEN + "Registration successful!" + Style.RESET_ALL)
    return True

def authenticate():
"""Login with username & password"""
username = input(Fore.CYAN + "Username: " + Style.RESET_ALL)
password = getpass.getpass(Fore.CYAN + "Password: " + Style.RESET_ALL)

    with open(USERS_FILE, 'r') as f:
        users = json.load(f)

    if username not in users or users[username]["password"] != password:
        print(Fore.RED + "Invalid credentials!" + Style.RESET_ALL)
        return None

    print(Fore.GREEN + f"\nWelcome, {username}!" + Style.RESET_ALL)
    return {"username": username, "role": users[username]["role"]}

def log_joke(username, joke, language, category, rating=None):
"""Store joke history with ratings"""
with open(JOKE_HISTORY_FILE, 'r') as f:
history = json.load(f)

    if username not in history:
        history[username] = []

    joke_entry = {
        "joke": joke,
        "language": language,
        "category": category,
        "timestamp": str(datetime.now()),
        "rating": rating
    }
    history[username].append(joke_entry)

    with open(JOKE_HISTORY_FILE, 'w') as f:
        json.dump(history, f, indent=4)

def get_user_favorite_language(username):
"""Determine user's most used language for jokes"""
with open(JOKE_HISTORY_FILE, 'r') as f:
history = json.load(f)

    if username not in history or not history[username]:
        return None

    lang_counts = defaultdict(int)
    for entry in history[username]:
        lang_counts[entry["language"]] += 1

    return max(lang_counts.items(), key=lambda x: x[1])[0]

def rate_joke(username):
"""Let user rate the last joke"""
with open(JOKE_HISTORY_FILE, 'r') as f:
history = json.load(f)

    if username not in history or not history[username]:
        print(Fore.YELLOW + "No jokes to rate yet!" + Style.RESET_ALL)
        return

    last_joke = history[username][-1]
    print(Fore.CYAN + f"\nLast joke: {last_joke['joke']}")

    try:
        rating = int(input(Fore.CYAN + "Rate (1-5): " + Style.RESET_ALL))
        if 1 <= rating <= 5:
            last_joke["rating"] = rating
            with open(JOKE_HISTORY_FILE, 'w') as f:
                json.dump(history, f, indent=4)
            print(Fore.GREEN + "Thanks for rating!" + Style.RESET_ALL)
        else:
            print(Fore.RED + "Rating must be between 1-5!" + Style.RESET_ALL)
    except ValueError:
        print(Fore.RED + "Invalid input!" + Style.RESET_ALL)

def get_personalized_joke(username):
"""Recommend joke based on user's history"""
fav_lang = get_user_favorite_language(username)
if fav_lang:
return pyjokes.get_joke(language=fav_lang)
else:
return pyjokes.get_joke()

def admin_panel():
"""Admin features: View users & joke stats"""
print(Fore.MAGENTA + "\nADMIN PANEL")
print("1. View all users")
print("2. View joke history")
print("3. Exit admin panel")

    choice = input(Fore.CYAN + "\nEnter choice (1-3): " + Style.RESET_ALL)

    if choice == '1':
        with open(USERS_FILE, 'r') as f:
            users = json.load(f)
        print(Fore.YELLOW + "\nRegistered Users:" + Style.RESET_ALL)
        for user, data in users.items():
            print(f"- {user} (Role: {data['role']})")

    elif choice == '2':
        with open(JOKE_HISTORY_FILE, 'r') as f:
            history = json.load(f)
        print(Fore.YELLOW + "\nJoke History:" + Style.RESET_ALL)
        for user, jokes in history.items():
            print(f"\nUser: {user}")
            for joke in jokes[-3:]:  # Show last 3 jokes
                print(f"- {joke['joke']} (Lang: {joke['language']}, Rating: {joke.get('rating', 'N/A')})")

    elif choice == '3':
        return

    else:
        print(Fore.RED + "Invalid choice!" + Style.RESET_ALL)

def joke_game(username, role):
"""Main joke-telling game with multi-language support"""
while True:
print(Fore.BLUE + "\nJOKE TELLER GAME")
print("1. Random joke")
print("2. Chuck Norris joke")
print("3. Joke in specific language")
print("4. Multiple jokes")
print("5. Personalized joke (based on your history)")
print("6. Rate last joke")
if role == "admin":
print("7. Admin Panel")
print("8. Exit")

        choice = input(Fore.CYAN + "\nEnter choice: " + Style.RESET_ALL)

        if choice == '1':
            joke = pyjokes.get_joke()
            print(Fore.GREEN + f"\n{joke}")
            log_joke(username, joke, 'en', 'neutral')

        elif choice == '2':
            joke = pyjokes.get_joke(category='chuck')
            print(Fore.GREEN + f"\n{joke}")
            log_joke(username, joke, 'en', 'chuck')

        elif choice == '3':
            lang = input(Fore.CYAN + f"Choose language ({'/'.join(SUPPORTED_LANGUAGES)}): " + Style.RESET_ALL)
            if lang in SUPPORTED_LANGUAGES:
                joke = pyjokes.get_joke(language=lang)
                print(Fore.GREEN + f"\n{joke}")
                log_joke(username, joke, lang, 'neutral')
            else:
                print(Fore.RED + "Unsupported language!" + Style.RESET_ALL)

        elif choice == '4':
            try:
                count = int(input(Fore.CYAN + "How many jokes? (1-10): " + Style.RESET_ALL))
                count = max(1, min(10, count))
                jokes = [pyjokes.get_joke() for _ in range(count)]
                for i, joke in enumerate(jokes, 1):
                    print(Fore.GREEN + f"\n{i}. {joke}")
                    log_joke(username, joke, 'en', 'neutral')
            except ValueError:
                print(Fore.RED + "Invalid number!" + Style.RESET_ALL)

        elif choice == '5':
            joke = get_personalized_joke(username)
            print(Fore.GREEN + f"\n{joke}")
            log_joke(username, joke, get_user_favorite_language(username) or 'en', 'neutral')

        elif choice == '6':
            rate_joke(username)

        elif choice == '7' and role == "admin":
            admin_panel()

        elif choice == '8':
            print(Fore.YELLOW + "Thanks for playing!" + Style.RESET_ALL)
            break

        else:
            print(Fore.RED + "Invalid choice!" + Style.RESET_ALL)

class JokeTeller:
"""Core joke telling functionality"""
SUPPORTED_LANGUAGES = ['en', 'de', 'es', 'fr', 'gl', 'eu']
SUPPORTED_CATEGORIES = ['neutral', 'chuck', 'all']
MAX_JOKE_LENGTH = 120 # For formatting

    def __init__(self, db_manager):
        self.db = db_manager

    def tell_joke(self, language='en', category='neutral'):
        """Get a joke with specified language and category"""
        try:
            joke = pyjokes.get_joke(language=language, category=category)
            return self._format_joke(joke)
        except:
            return None

    def tell_multiple_jokes(self, count=1, language='en', category='neutral'):
        """Get multiple jokes"""
        jokes = []
        for _ in range(min(10, max(1, count))):
            joke = self.tell_joke(language, category)
            if joke:
                jokes.append(joke)
        return jokes

    def _format_joke(self, joke):
        """Format long jokes with text wrapping"""
        if len(joke) > self.MAX_JOKE_LENGTH:
            return textwrap.fill(joke, width=self.MAX_JOKE_LENGTH)
        return joke

class JokeHistoryManager:
"""Manages joke history and statistics"""
def **init**(self, db_manager):

    def log_joke(self, username, joke, language, category, rating=None):
        """Record a joke in history"""
        history = self.db.read_file(self.db.joke_history_file)
        stats = self.db.read_file(self.db.stats_file)

        if username not in history:
            history[username] = []

        joke_entry = {
            "joke": joke,
            "language": language,
            "category": category,
            "timestamp": str(datetime.now()),
            "rating": rating
        }
        history[username].append(joke_entry)

        # Update statistics
        stats["total_jokes_told"] += 1
        if joke in stats["popular_jokes"]:
            stats["popular_jokes"][joke] += 1
        else:
            stats["popular_jokes"][joke] = 1

        self.db.write_file(self.db.joke_history_file, history)
        self.db.write_file(self.db.stats_file, stats)

    def get_user_history(self, username, limit=5):
        """Get user's joke history"""
        history = self.db.read_file(self.db.joke_history_file)
        return history.get(username, [])[-limit:]

    def get_user_favorite_language(self, username):
        """Determine user's preferred joke language"""
        history = self.db.read_file(self.db.joke_history_file)
        if username not in history or not history[username]:
            return None

        lang_counts = defaultdict(int)
        for entry in history[username]:
            lang_counts[entry["language"]] += 1

        return max(lang_counts.items(), key=lambda x: x[1])[0]

````python
import pyjokes
import json
import getpass
from pathlib import Path
from datetime import datetime
from collections import defaultdict
import random
from colorama import Fore, Style, init

# Initialize colorama for colored text
init(autoreset=True)

# File paths
USERS_FILE = "users.json"
JOKE_HISTORY_FILE = "joke_history.json"

# Supported languages and categories
SUPPORTED_LANGUAGES = ['en', 'de', 'es', 'fr', 'gl', 'eu']
SUPPORTED_CATEGORIES = ['neutral', 'chuck', 'all']

def initialize_files():
    """Create JSON files if they don't exist"""
    if not Path(USERS_FILE).exists():
        with open(USERS_FILE, 'w') as f:
            json.dump({"admin": {"password": "admin123", "role": "admin"}}, f)
    
    if not Path(JOKE_HISTORY_FILE).exists():
        with open(JOKE_HISTORY_FILE, 'w') as f:
            json.dump({}, f)

def register_user():
    """Register a new user with password confirmation"""
    username = input(Fore.CYAN + "Enter new username: " + Style.RESET_ALL)
    
    with open(USERS_FILE, 'r') as f:
        users = json.load(f)
    
    if username in users:
        print(Fore.RED + "Username already exists!" + Style.RESET_ALL)
        return False
    
    password = getpass.getpass(Fore.CYAN + "Enter password: " + Style.RESET_ALL)
    confirm = getpass.getpass(Fore.CYAN + "Confirm password: " + Style.RESET_ALL)
    
    if password != confirm:
        print(Fore.RED + "Passwords don't match!" + Style.RESET_ALL)
        return False
    
    users[username] = {"password": password, "role": "user"}
    with open(USERS_FILE, 'w') as f:
        json.dump(users, f, indent=4)
    
    print(Fore.GREEN + "Registration successful!" + Style.RESET_ALL)
    return True

def authenticate():
    """Login with username & password"""
    username = input(Fore.CYAN + "Username: " + Style.RESET_ALL)
    password = getpass.getpass(Fore.CYAN + "Password: " + Style.RESET_ALL)
    
    with open(USERS_FILE, 'r') as f:
        users = json.load(f)
    
    if username not in users or users[username]["password"] != password:
        print(Fore.RED + "Invalid credentials!" + Style.RESET_ALL)
        return None
    
    print(Fore.GREEN + f"\nWelcome, {username}!" + Style.RESET_ALL)
    return {"username": username, "role": users[username]["role"]}

def log_joke(username, joke, language, category, rating=None):
    """Store joke history with ratings"""
    with open(JOKE_HISTORY_FILE, 'r') as f:
        history = json.load(f)
    
    if username not in history:
        history[username] = []
    
    joke_entry = {
        "joke": joke,
        "language": language,
        "category": category,
        "timestamp": str(datetime.now()),
        "rating": rating
    }
    history[username].append(joke_entry)
    
    with open(JOKE_HISTORY_FILE, 'w') as f:
        json.dump(history, f, indent=4)

def get_user_favorite_language(username):
    """Determine user's most used language for jokes"""
    with open(JOKE_HISTORY_FILE, 'r') as f:
        history = json.load(f)
    
    if username not in history or not history[username]:
        return None
    
    lang_counts = defaultdict(int)
    for entry in history[username]:
        lang_counts[entry["language"]] += 1
    
    return max(lang_counts.items(), key=lambda x: x[1])[0]

def rate_joke(username):
    """Let user rate the last joke"""
    with open(JOKE_HISTORY_FILE, 'r') as f:
        history = json.load(f)
    
    if username not in history or not history[username]:
        print(Fore.YELLOW + "No jokes to rate yet!" + Style.RESET_ALL)
        return
    
    last_joke = history[username][-1]
    print(Fore.CYAN + f"\nLast joke: {last_joke['joke']}")
    
    try:
        rating = int(input(Fore.CYAN + "Rate (1-5): " + Style.RESET_ALL))
        if 1 <= rating <= 5:
            last_joke["rating"] = rating
            with open(JOKE_HISTORY_FILE, 'w') as f:
                json.dump(history, f, indent=4)
            print(Fore.GREEN + "Thanks for rating!" + Style.RESET_ALL)
        else:
            print(Fore.RED + "Rating must be between 1-5!" + Style.RESET_ALL)
    except ValueError:
        print(Fore.RED + "Invalid input!" + Style.RESET_ALL)

def get_personalized_joke(username):
    """Recommend joke based on user's history"""
    fav_lang = get_user_favorite_language(username)
    if fav_lang:
        return pyjokes.get_joke(language=fav_lang)
    else:
        return pyjokes.get_joke()

def admin_panel():
    """Admin features: View users & joke stats"""
    print(Fore.MAGENTA + "\nADMIN PANEL")
    print("1. View all users")
    print("2. View joke history")
    print("3. Exit admin panel")
    
    choice = input(Fore.CYAN + "\nEnter choice (1-3): " + Style.RESET_ALL)
    
    if choice == '1':
        with open(USERS_FILE, 'r') as f:
            users = json.load(f)
        print(Fore.YELLOW + "\nRegistered Users:" + Style.RESET_ALL)
        for user, data in users.items():
            print(f"- {user} (Role: {data['role']})")
    
    elif choice == '2':
        with open(JOKE_HISTORY_FILE, 'r') as f:
            history = json.load(f)
        print(Fore.YELLOW + "\nJoke History:" + Style.RESET_ALL)
        for user, jokes in history.items():
            print(f"\nUser: {user}")
            for joke in jokes[-3:]:  # Show last 3 jokes
                print(f"- {joke['joke']} (Lang: {joke['language']}, Rating: {joke.get('rating', 'N/A')})")
    
    elif choice == '3':
        return
    
    else:
        print(Fore.RED + "Invalid choice!" + Style.RESET_ALL)

def joke_game(username, role):
    """Main joke-telling game with multi-language support"""
    while True:
        print(Fore.BLUE + "\nJOKE TELLER GAME")
        print("1. Random joke")
        print("2. Chuck Norris joke")
        print("3. Joke in specific language")
        print("4. Multiple jokes")
        print("5. Personalized joke (based on your history)")
        print("6. Rate last joke")
        if role == "admin":
            print("7. Admin Panel")
        print("8. Exit")
        
        choice = input(Fore.CYAN + "\nEnter choice: " + Style.RESET_ALL)
        
        if choice == '1':
            joke = pyjokes.get_joke()
            print(Fore.GREEN + f"\n{joke}")
            log_joke(username, joke, 'en', 'neutral')
        
        elif choice == '2':
            joke = pyjokes.get_joke(category='chuck')
            print(Fore.GREEN + f"\n{joke}")
            log_joke(username, joke, 'en', 'chuck')
        
        elif choice == '3':
            lang = input(Fore.CYAN + f"Choose language ({'/'.join(SUPPORTED_LANGUAGES)}): " + Style.RESET_ALL)
            if lang in SUPPORTED_LANGUAGES:
                joke = pyjokes.get_joke(language=lang)
                print(Fore.GREEN + f"\n{joke}")
                log_joke(username, joke, lang, 'neutral')
            else:
                print(Fore.RED + "Unsupported language!" + Style.RESET_ALL)
        
        elif choice == '4':
            try:
                count = int(input(Fore.CYAN + "How many jokes? (1-10): " + Style.RESET_ALL))
                count = max(1, min(10, count))
                jokes = [pyjokes.get_joke() for _ in range(count)]
                for i, joke in enumerate(jokes, 1):
                    print(Fore.GREEN + f"\n{i}. {joke}")
                    log_joke(username, joke, 'en', 'neutral')
            except ValueError:
                print(Fore.RED + "Invalid number!" + Style.RESET_ALL)
        
        elif choice == '5':
            joke = get_personalized_joke(username)
            print(Fore.GREEN + f"\n{joke}")
            log_joke(username, joke, get_user_favorite_language(username) or 'en', 'neutral')
        
        elif choice == '6':
            rate_joke(username)
        
        elif choice == '7' and role == "admin":
            admin_panel()
        
        elif choice == '8':
            print(Fore.YELLOW + "Thanks for playing!" + Style.RESET_ALL)
            break
        
        else:
            print(Fore.RED + "Invalid choice!" + Style.RESET_ALL)

def main():
    """Main program execution"""
    initialize_files()
    
    print(Fore.BLUE + "\n=== JOKE TELLER GAME ===")
    print("1. Login")
    print("2. Register")
    print("3. Exit")
    
    while True:
        option = input(Fore.CYAN + "\nSelect option (1-3): " + Style.RESET_ALL)
        
        if option == '1':
            user_data = authenticate()
            if user_data:
                joke_game(user_data["username"], user_data["role"])
                break
        
        elif option == '2':
            if register_user():
                user_data = authenticate()
                if user_data:
                    joke_game(user_data["username"], user_data["role"])
                    break
        
        elif option == '3':
            print(Fore.YELLOW + "Goodbye!" + Style.RESET_ALL)
            break
        
        else:
            print(Fore.RED + "Invalid option!" + Style.RESET_ALL)
if __name__ =="__main__":
    main()
````
