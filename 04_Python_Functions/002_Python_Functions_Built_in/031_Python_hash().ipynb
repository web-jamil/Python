# Python `hash()`

The **`hash()`** method returns the hash value of an object if it has one.

Hash values are just integers that are used to compare dictionary keys during a dictionary lookup quickly.

Internally, **`hash()`** method calls **`__hash__()`** method of an object which is set by default for any object. We'll look at this later..

**Syntax**:

```python
hash(object)
```


## `hash()` Parameters

**`hash()`** method takes a single parameter:

- **object** - the object whose hash value is to be returned (integer, string, float)


## Return Value from `hash()`

**`hash()`** method returns the hash value of an object if it has one.

If an object has custom **`__hash__()`** method, it truncates the return value to the size of **`Py_ssize_t`**.


````python
# Example 1: How hash() works in Python?
# hash for integer unchanged
print('Hash for 181 is:', hash(181))

# hash for decimal
print('Hash for 181.23 is:',hash(181.23))

# hash for string
print('Hash for Python is:', hash('Python'))
````

````python
# Example 2: hash() for immutable tuple object?
# hash() method only works for immutable objects as tuple.

# tuple of vowels
vowels = ('a', 'e', 'i', 'o', 'u')

print('The hash is:', hash(vowels))
````

## How hash() works for custom objects?

As stated above, hash() method internally calls **`__hash__()`** method. So, any objects can override **`__hash__()`** for custom hash values.

But for correct hash implementation, **`__hash__()`** should always return an integer. And, both **`__eq__()`** and **`__hash__()`** methods have to be implemented.

Below are the cases for correct **`__hash__()`** override.

**Cases for custom hash implementation for objects**

| **`__eq__()`**           | **`__hash__()`**          | **Description**                                                                                                                                                  |
| :----------------------- | :------------------------ | :--------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Defined (by default)** | **Defined (by default)**  | **If left as is, all objects compare unequal (except themselves)**                                                                                               |
| **(If mutable) Defined** | **Should not be defined** | **Implementation of hashable collection requires key's hash value be immutable**                                                                                 |
| **Not defined**          | **Should not be defined** | **If `__eq__()` isn't defined, `__hash__()` should not be defined.**                                                                                             |
| **Defined**              | **Not defined**           | **Class instances will not be usable as hashable collection. `__hash__()` implicity set to `None`. Raises `TypeError` exception if tried to retrieve the hash.** |
| **Defined**              | **Retain from Parent**    | **`__hash__ = <ParentClass>.__hash__`**                                                                                                                          |
| **Defined**              | **Doesn't want to hash**  | **`__hash__ = None`. Raises TypeError exception if tried to retrieve the hash.**                                                                                 |


````python
# Example 3: hash() for Custom Objects by overriding __hash__()

class Person:
    def __init__(self, age, name):
        self.age = age
        self.name = name

    def __eq__(self, other):
        return self.age == other.age and self.name == other.name

    def __hash__(self):
        print('The hash is:')
        return hash((self.age, self.name))

person = Person(23, 'Adam')
print(hash(person))
````

> **Note:** You don't have to implement **`__eq__()`** method for the hash as it is created by default for all objects.


Hash functions are a cornerstone of efficient data management in computer science, and Python leverages them extensively in its core data structures.

## Hash Functions in Python: Theory and Code

### 1. Theory of Hashing

#### What is a Hash Function?

A **hash function** is a mathematical algorithm that takes an input (or 'key') of arbitrary size and transforms it into a fixed-size value, typically an integer. This output is called a **hash value**, **hash code**, **digest**, or simply a **hash**.

#### Purpose

The primary purpose of a hash function in data structures is to quickly map data to a specific location in memory (often an index in an array or a "bucket"). This enables very fast lookups, insertions, and deletions of data.

#### Key Properties of a Good Hash Function:

1.  **Deterministic:** For the same input, the hash function must _always_ produce the exact same hash value. If the input doesn't change, the hash shouldn't change.
2.  **Efficiency:** Computing the hash value should be very fast, ideally a constant-time operation.
3.  **Uniform Distribution (Collision Resistance):** A good hash function distributes hash values uniformly across its output range. This minimizes "collisions," where different inputs produce the same hash value. While collisions are inevitable (many inputs, fewer possible hash outputs), a good hash function makes them rare and manageable.

#### Hashable Objects

In Python, an object is **hashable** if it has a hash value that remains constant throughout its lifetime. Hashable objects must also have an `__eq__` method (for equality comparison), and if two objects are equal (`a == b`), then their hash values _must_ also be equal (`hash(a) == hash(b)`).

#### Use in Data Structures

- **Dictionaries (`dict`):** Dictionary keys _must_ be hashable. When you store a key-value pair, Python uses the key's hash to determine where to store it. When you look up a key, it re-calculates the hash to quickly find the potential location.
- **Sets (`set`):** Elements of a set _must_ be hashable. Sets use hashing to efficiently store unique elements and perform fast membership tests.

### 2. Python's `hash()` Built-in Function

Python provides the built-in `hash()` function to compute the hash value of an object.

**Syntax:** `hash(object)`

**Theory:**

- **Immutable Types are Hashable:** By default, all immutable built-in types are hashable. This includes:
  - `int`, `float`, `complex` (numbers)
  - `str` (strings)
  - `tuple` (tuples, _provided all their elements are also hashable_)
  - `bool` (Booleans - `True` hashes to `1`, `False` to `0`)
  - `NoneType` (`None` is hashable)
  - `frozenset` (an immutable version of `set`)
- **Mutable Types are NOT Hashable:** Mutable built-in types are generally _not_ hashable. This includes:
  - `list`
  - `dict` (dictionary)
  - `set`
  - `bytearray`
    This is because their content can change after creation, which could alter their hash value, breaking the integrity of hash-based data structures. Attempting to hash them will raise a `TypeError`.
- **Hash Randomization:** For security reasons (to prevent certain denial-of-service attacks), the hash values of strings, bytes, and datetime objects are "salted" with a random value at the start of each Python process. This means `hash("hello")` might return a different number each time you run your Python script, but it will be consistent _within that single run_.

**Code Examples:**

```python
import sys

print("--- 2. Python's `hash()` Built-in Function ---")

# 2.1 Hashing Immutable Built-in Types
print("\n2.1 Hashing Immutable Types:")
print(f"Hash of integer 10: {hash(10)}")
print(f"Hash of float 3.14: {hash(3.14)}")
print(f"Hash of string 'Python': {hash('Python')}")
print(f"Hash of tuple (1, 2): {hash((1, 2))}")
print(f"Hash of True: {hash(True)}")
print(f"Hash of False: {hash(False)}")
print(f"Hash of None: {hash(None)}")

# Demonstrate determinism within a single run (but not across runs)
s_val = "example_string"
print(f"\nHash of '{s_val}' (first call): {hash(s_val)}")
print(f"Hash of '{s_val}' (second call): {hash(s_val)}")
s_val_copy = "example_string" # A new string object with the same value
print(f"Hash of '{s_val_copy}' (new object, same value): {hash(s_val_copy)}")


# 2.2 Attempting to Hash Mutable Built-in Types (will raise TypeError)
print("\n2.2 Attempting to Hash Mutable Types (will raise TypeError):")

try:
    my_list = [1, 2, 3]
    print(f"Hash of list {my_list}: {hash(my_list)}")
except TypeError as e:
    print(f"    Error hashing list: {e}")

try:
    my_dict = {"a": 1, "b": 2}
    print(f"Hash of dict {my_dict}: {hash(my_dict)}")
except TypeError as e:
    print(f"    Error hashing dictionary: {e}")

try:
    my_set = {10, 20, 30}
    print(f"Hash of set {my_set}: {hash(my_set)}")
except TypeError as e:
    print(f"    Error hashing set: {e}")


# 2.3 Tuples containing mutable elements are NOT hashable
print("\n2.3 Tuples containing mutable elements are NOT hashable:")
hashable_tuple = (1, "hello", (3, 4)) # All elements are immutable
print(f"Hash of hashable tuple {hashable_tuple}: {hash(hashable_tuple)}")

unhashable_tuple_with_list = (1, [2, 3]) # Contains a mutable list
try:
    print(f"Hash of unhashable tuple {unhashable_tuple_with_list}: {hash(unhashable_tuple_with_list)}")
except TypeError as e:
    print(f"    Error hashing tuple with mutable element: {e}")

unhashable_tuple_with_dict = (1, {"key": "value"}) # Contains a mutable dict
try:
    print(f"Hash of unhashable tuple {unhashable_tuple_with_dict}: {hash(unhashable_tuple_with_dict)}")
except TypeError as e:
    print(f"    Error hashing tuple with mutable element: {e}")
```

### 3. Hashing Custom Objects

**Theory:**
By default, instances of user-defined classes are hashable. Their hash value is based on their unique identity (memory address, `id()`). This means two different instances, even if they have identical attribute values, will have different hash values.

If you want custom objects to be hashable based on their _content_ (i.e., if two objects with the same content should be considered equal and have the same hash), you must override two special methods:

1.  `__eq__(self, other)`: Defines what makes two objects equal.
2.  `__hash__(self)`: Defines how the object's hash value is computed.

**The `__eq__` and `__hash__` Contract:**

- If you implement `__eq__` but _not_ `__hash__`, Python automatically makes your objects unhashable.
- If you implement both, you _must_ ensure that if `a == b` is `True`, then `hash(a)` _must_ equal `hash(b)`.
- The `__hash__` method must return a constant value for the object's lifetime. This implies that all attributes used in `__hash__` must be immutable. If your object's state can change in a way that would affect its hash, it should not be made hashable.

**Code Examples:**

```python
print("\n--- 3. Hashing Custom Objects ---")

# 3.1 Default Hashing for Custom Objects
print("\n3.1 Default Hashing for Custom Objects:")
class MyDefaultObject:
    def __init__(self, value):
        self.value = value

obj1 = MyDefaultObject(10)
obj2 = MyDefaultObject(20)
obj1_copy = MyDefaultObject(10) # Different instance, same value

print(f"ID of obj1: {id(obj1)}, Hash of obj1: {hash(obj1)}")
print(f"ID of obj2: {id(obj2)}, Hash of obj2: {hash(obj2)}")
print(f"ID of obj1_copy: {id(obj1_copy)}, Hash of obj1_copy: {hash(obj1_copy)}")
# Notice obj1 and obj1_copy have different IDs and different default hashes,
# even though their 'value' attribute is the same.

# They are not considered equal by default, so they can both be dictionary keys
default_dict = {obj1: "first", obj1_copy: "second"}
print(f"Default objects in dict: {default_dict}")
print(f"obj1 == obj1_copy: {obj1 == obj1_copy}") # False by default


# 3.2 Making Custom Objects Hashable Based on Content
print("\n3.2 Making Custom Objects Hashable Based on Content:")
class ImmutablePoint:
    def __init__(self, x, y):
        # Attributes used for hashing must be immutable
        self.x = x
        self.y = y

    def __eq__(self, other):
        # Define equality based on content
        if not isinstance(other, ImmutablePoint):
            return NotImplemented
        return self.x == other.x and self.y == other.y

    def __hash__(self):
        # Compute hash based on immutable content (x, y)
        # Hashing a tuple of immutable values is a common pattern
        return hash((self.x, self.y))

p1 = ImmutablePoint(1, 2)
p2 = ImmutablePoint(3, 4)
p1_copy = ImmutablePoint(1, 2) # Different instance, but same content

print(f"Hash of p1 (1, 2): {hash(p1)}")
print(f"Hash of p2 (3, 4): {hash(p2)}")
print(f"Hash of p1_copy (1, 2): {hash(p1_copy)}") # Same hash as p1!

print(f"p1 == p1_copy: {p1 == p1_copy}") # True, as per our __eq__

# Now, they can be used as dictionary keys or set elements based on content equality
content_dict = {p1: "Start Point", p2: "End Point"}
print(f"Dictionary with custom hashable objects: {content_dict}")
print(f"Value for p1_copy (retrieved using equal object): {content_dict[p1_copy]}") # Can retrieve using p1_copy

content_set = {p1, p2, p1_copy} # p1_copy is considered equal to p1, so only one instance is stored
print(f"Set with custom hashable objects: {content_set}") # Only two distinct points


# 3.3 Attempting to Hash a Mutable Custom Object (Incorrect Design)
print("\n3.3 Attempting to Hash a Mutable Custom Object (Incorrect Design):")
class MutableContainer:
    def __init__(self, data):
        self.data = data # 'data' is a list, which is mutable

    def __eq__(self, other):
        return isinstance(other, MutableContainer) and self.data == other.data

    # If __hash__ is not defined, and __eq__ is, Python makes it unhashable by default.
    # If we *tried* to implement __hash__ based on mutable data, it would be problematic:
    # def __hash__(self):
    #     return hash(tuple(self.data)) # This would work initially, but hash changes if self.data changes!

mutable_obj = MutableContainer([1, 2])
try:
    print(f"Hash of mutable_obj: {hash(mutable_obj)}")
except TypeError as e:
    print(f"    Error: {e} - MutableContainer is unhashable (due to mutable content or missing __hash__).")

# If we were to force a __hash__ on a mutable object and use it as a key:
class BadKey:
    def __init__(self, val):
        self.val = val
    def __eq__(self, other):
        return self.val == other.val
    def __hash__(self):
        return hash(self.val) # Hash depends on mutable 'val'

bad_key_instance = BadKey(10)
bad_dict = {bad_key_instance: "original_value"}
print(f"Bad dict initial: {bad_dict}")
print(f"Value for BadKey(10): {bad_dict[BadKey(10)]}") # Works

bad_key_instance.val = 20 # Mutate the key after it's in the dictionary!
print(f"Bad dict after key mutation: {bad_dict}") # Dictionary still holds the old hash/location

# Now, trying to access it with the old value (which is still the key's hash) will fail
# or access the wrong bucket. Trying to access with the new value (new hash) won't find it.
try:
    print(f"Trying to access with old value (BadKey(10)): {bad_dict[BadKey(10)]}")
except KeyError as e:
    print(f"    Error: {e} - Key not found after mutation (old hash).")

try:
    print(f"Trying to access with new value (BadKey(20)): {bad_dict[BadKey(20)]}")
except KeyError as e:
    print(f"    Error: {e} - Key not found after mutation (new hash).")
# This demonstrates why mutable objects cannot be hashable.

```
