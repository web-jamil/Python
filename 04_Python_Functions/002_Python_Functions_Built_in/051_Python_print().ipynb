# Python `print()`

The **`print()`** function prints the given object to the standard output device (screen) or to the text stream file.

**Syntax**:

```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```


## `print()` Parameters

- **objects** - object to the printed. **\*** indicates that there may be more than one object
- **sep** - objects are separated by sep. **Default value:** **`' '`**
- **end** - **`end`** is printed at last
- **file** - must be an object with write(string) method. If omitted it, **`sys.stdout`** will be used which prints objects on the screen.
- **flush** - If **`True`**, the stream is forcibly flushed. **Default value:** **`False`**

> **Note:** **`sep`**, **`end`**, **`file`**, and **`flush`** are keyword arguments. If you want to use **`sep`** argument, you have to use:

```python
print(*objects, sep = 'separator')
```

not

```python
print(*objects, 'separator')
```


## Return Value from `print()`

The **`print()`** function doesn't return any value; returns **`None`**.


````python
# Example 1: How print() works in Python?

print("Python is fun.")

a = 5
# Two objects are passed
print("a =", a)

b = a
# Three objects are passed
print('a =', a, '= b')
````

In the above program, only **`objects`** parameter is passed to **`print()`** function (in all three print statements).

Hence,

- **`' '`** separator is used. Notice, the space between two objects in the output.
- **`end`** parameter **`'\n'`** (newline character) is used. Notice, each print statement displays the output in the new line.
- **`file`** is **`sys.stdout`**. The output is printed on the screen.
- **`flush`** is **`False`**. The stream is not forcibly flushed.


````python
# Example 2: print() with separator and end parameters

a = 5
print("a =", a, sep='00000', end='\n\n\n')
print("a =", a, sep='0', end='')
````

We passed the **`sep`** and **`end`** parameters in the above program.

````python
# Example 3: print() with file parameter

# In Python, you can print objects to the file by specifying the file parameter.

sourceFile = open('python.txt', 'w')
print('Pretty cool, huh!', file = sourceFile)
sourceFile.close()
````

This program tries to open the **python.txt** in writing mode. If this file doesn't exist, **python.txt** file is created and opened in writing mode.

Here, we have passed **`sourceFile`** file object to the **`file`** parameter. The string object 'Pretty cool, huh!' is printed to **python.txt** file (check it in your system).

Finally, the file is closed using **`close()`** method.


The **`print()`** function in Python is one of the most commonly used built-in functions. It is used to display output to the console or other standard output devices. Below is a comprehensive guide to the **`print()`** function, covering everything from basic concepts to advanced topics, including examples and explanations.

---

## **1. What is the `print()` Function?**

- **Definition**: The `print()` function outputs text or other data to the console or standard output.
- **Purpose**: Used for debugging, displaying results, and interacting with users.
- **Syntax**:
  ```python
  print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
  ```
  - `*objects`: The objects to print (can be multiple, separated by commas).
  - `sep`: Separator between objects (default is a space `' '`).
  - `end`: Character to append at the end (default is a newline `'\n'`).
  - `file`: The file-like object to write to (default is `sys.stdout`).
  - `flush`: Whether to forcibly flush the output stream (default is `False`).

---

## **2. Basic Usage of `print()`**

### **2.1. Printing Strings**

- **Purpose**: Print a simple string.
- **Example**:
  ```python
  print("Hello, World!")  # Output: Hello, World!
  ```

### **2.2. Printing Multiple Objects**

- **Purpose**: Print multiple objects separated by spaces (default).
- **Example**:
  ```python
  print("Hello", "World", 42)  # Output: Hello World 42
  ```

### **2.3. Printing with a Custom Separator**

- **Purpose**: Use a custom separator between objects.
- **Example**:
  ```python
  print("Hello", "World", 42, sep="-")  # Output: Hello-World-42
  ```

### **2.4. Printing Without a Newline**

- **Purpose**: Print without adding a newline at the end.
- **Example**:
  ```python
  print("Hello", end=" ")
  print("World")  # Output: Hello World
  ```

---

## **3. Intermediate Usage of `print()`**

### **3.1. Printing to a File**

- **Purpose**: Write output to a file instead of the console.
- **Example**:
  ```python
  with open("output.txt", "w") as f:
      print("Hello, World!", file=f)
  # Writes "Hello, World!" to output.txt
  ```

### **3.2. Flushing the Output**

- **Purpose**: Force the output to be written immediately (useful for real-time logging).
- **Example**:
  ```python
  import time
  print("Loading", end="", flush=True)
  for _ in range(3):
      time.sleep(1)
      print(".", end="", flush=True)
  # Output: Loading... (printed in real-time)
  ```

### **3.3. Printing Formatted Strings**

- **Purpose**: Use formatted strings (f-strings) for dynamic output.
- **Example**:
  ```python
  name = "Alice"
  age = 30
  print(f"{name} is {age} years old.")  # Output: Alice is 30 years old.
  ```

---

## **4. Advanced Usage of `print()`**

### **4.1. Printing with ANSI Escape Codes**

- **Purpose**: Add colors and styles to console output.
- **Example**:
  ```python
  print("\033[1;31;40mRed Text on Black Background\033[0m")
  # Output: Red Text on Black Background (in red color)
  ```

### **4.2. Printing with `pprint` for Pretty Printing**

- **Purpose**: Print complex data structures (e.g., dictionaries, lists) in a readable format.
- **Example**:
  ```python
  from pprint import pprint
  data = {"name": "Alice", "age": 30, "skills": ["Python", "JavaScript"]}
  pprint(data)
  # Output:
  # {'age': 30,
  #  'name': 'Alice',
  #  'skills': ['Python', 'JavaScript']}
  ```

### **4.3. Printing with Logging**

- **Purpose**: Use `print()` in combination with logging for better control over output.
- **Example**:
  ```python
  import logging
  logging.basicConfig(level=logging.DEBUG)
  logging.debug("This is a debug message")
  # Output: DEBUG:root:This is a debug message
  ```

---

## **5. Real-World Examples**

### **5.1. Debugging with `print()`**

- **Issue**: Use `print()` to debug code by displaying variable values.
- **Example**:
  ```python
  x = 10
  y = 20
  print(f"x: {x}, y: {y}")  # Output: x: 10, y: 20
  ```

### **5.2. Progress Indicator**

- **Issue**: Display a progress indicator in a loop.
- **Example**:
  ```python
  import time
  for i in range(10):
      print(f"Progress: {i + 1}/10", end="\r", flush=True)
      time.sleep(1)
  # Output: Progress: 10/10 (updates in place)
  ```

### **5.3. Printing Tables**

- **Issue**: Print data in a tabular format.
- **Example**:
  ```python
  data = [
      ["Alice", 30, "Python"],
      ["Bob", 25, "JavaScript"],
      ["Charlie", 35, "Java"]
  ]
  for row in data:
      print(f"{row[0]:<10} {row[1]:<5} {row[2]:<10}")
  # Output:
  # Alice      30    Python
  # Bob        25    JavaScript
  # Charlie    35    Java
  ```

---

## **6. Common Pitfalls and Best Practices**

### **6.1. Avoid Overusing `print()` for Debugging**

- **Pitfall**: Using `print()` excessively for debugging can clutter the code.
- **Best Practice**: Use a logging library for debugging in production code.
- **Example**:
  ```python
  import logging
  logging.basicConfig(level=logging.DEBUG)
  logging.debug("Debug message")
  ```

### **6.2. Use `sep` and `end` Effectively**

- **Pitfall**: Forgetting to customize `sep` or `end` when needed.
- **Best Practice**: Use `sep` and `end` to control formatting.
- **Example**:
  ```python
  print("Hello", "World", sep=", ", end="!\n")  # Output: Hello, World!
  ```

### **6.3. Handle Large Output Efficiently**

- **Pitfall**: Printing large amounts of data can slow down the program.
- **Best Practice**: Use pagination or write to a file for large outputs.
- **Example**:
  ```python
  with open("large_output.txt", "w") as f:
      for i in range(100000):
          print(f"Line {i}", file=f)
  ```

---

## **7. Summary**

The `print()` function is a powerful tool for displaying output in Python. Key points to remember:

- **Basic Usage**: Print strings, multiple objects, and customize separators.
- **Intermediate Usage**: Print to files, flush output, and use formatted strings.
- **Advanced Usage**: Use ANSI escape codes, `pprint`, and logging.
- **Best Practices**: Avoid overusing `print()` for debugging, customize `sep` and `end`, and handle large outputs efficiently.
