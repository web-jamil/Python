# Python `isinstance()`

The **`isinstance()`** function checks if the object (first argument) is an instance or subclass of classinfo class (second argument).

**Syntax**:

```python
isinstance(object, classinfo)
```


## `isinstance()` Parameters

**`isinstance()`** takes two parameters:

- **object** - **`object`** to be checked
- **classinfo** - class, type, or tuple of classes and types


## Return Value from `isinstance()`

**`isinstance()`** returns:

- **`True`** if the **`object`** is an instance or subclass of a class or any element of the tuple
- **`False`** otherwise

If **`classinfo`** is not a type or tuple of types, a **`TypeError`** exception is raised.


````python
# Example 1: How isinstance() works?

class Foo:
    a = 5

fooInstance = Foo()

print(isinstance(fooInstance, Foo))
print(isinstance(fooInstance, (list-65, tuple)))
print(isinstance(fooInstance, (list, tuple, Foo)))
````

The `isinstance()` function in Python is a built-in function that checks whether an object is an instance of a specified class or a tuple of classes. It is one of the most useful functions in Python for type checking. It can also be very helpful in various scenarios like polymorphism, checking types of objects passed as function arguments, and ensuring that the behavior of code is predictable.

Let’s break down the `isinstance()` function in detail from basic to advanced usage:

---

### 1. **Basic Syntax of `isinstance()`**

The basic syntax of the `isinstance()` function is:

```python
isinstance(object, classinfo)
```

- **object**: The object you want to check.
- **classinfo**: A class, a tuple of classes, or a type.

The function returns `True` if the `object` is an instance of the class (or any subclass of that class) specified by `classinfo`, and `False` otherwise.

---

### 2. **Checking for Single Class**

One of the most common uses of `isinstance()` is to check if an object is an instance of a specific class.

#### Example:

```python
x = 5
print(isinstance(x, int))  # Output: True (5 is an instance of int)

y = "hello"
print(isinstance(y, str))  # Output: True ("hello" is an instance of str)

z = 3.14
print(isinstance(z, float))  # Output: True (3.14 is an instance of float)
```

Here, `isinstance(x, int)` returns `True` because `x` is of type `int`. Similarly, it checks for other types such as `str` and `float`.

---

### 3. **Using `isinstance()` with a Tuple of Classes**

`classinfo` can be a tuple of multiple classes, and `isinstance()` will check if the object is an instance of any of those classes.

#### Example:

```python
x = 10
print(isinstance(x, (int, float, str)))  # Output: True (x is an instance of int)

y = "hello"
print(isinstance(y, (int, float, str)))  # Output: True (y is an instance of str)

z = 3.14
print(isinstance(z, (int, float, str)))  # Output: True (z is an instance of float)
```

In this case, `isinstance()` returns `True` for all three objects because each object is an instance of one of the classes in the tuple `(int, float, str)`.

---

### 4. **Using `isinstance()` with Subclasses**

`isinstance()` also works with subclasses. If a class inherits from another class, `isinstance()` will return `True` for both the subclass and the parent class.

#### Example:

```python
class Animal:
    pass

class Dog(Animal):
    pass

class Cat(Animal):
    pass

dog = Dog()
cat = Cat()

print(isinstance(dog, Animal))  # Output: True (Dog is a subclass of Animal)
print(isinstance(cat, Animal))  # Output: True (Cat is a subclass of Animal)
```

In this example, `Dog` and `Cat` are subclasses of `Animal`, so `isinstance(dog, Animal)` and `isinstance(cat, Animal)` both return `True`.

---

### 5. **Checking for Multiple Inheritance**

Python allows multiple inheritance, and `isinstance()` correctly handles multiple inheritance as well. If an object is an instance of any class in the inheritance chain, `isinstance()` will return `True`.

#### Example:

```python
class Animal:
    pass

class Mammal(Animal):
    pass

class Bird(Animal):
    pass

class Bat(Mammal, Bird):
    pass

bat = Bat()

print(isinstance(bat, Mammal))  # Output: True (Bat is a subclass of Mammal)
print(isinstance(bat, Bird))    # Output: True (Bat is a subclass of Bird)
print(isinstance(bat, Animal))  # Output: True (Bat is a subclass of Animal)
```

Here, `Bat` inherits from both `Mammal` and `Bird`, and `isinstance()` correctly identifies it as an instance of all these classes.

---

### 6. **Using `isinstance()` with Custom Classes**

You can use `isinstance()` to check whether an object is an instance of a custom class.

#### Example:

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, employee_id):
        super().__init__(name)
        self.employee_id = employee_id

employee = Employee("Jane", 101)

print(isinstance(employee, Person))    # Output: True (Employee is a subclass of Person)
print(isinstance(employee, Employee))  # Output: True (employee is an instance of Employee)
```

Here, `employee` is an instance of `Employee`, and `Employee` is a subclass of `Person`, so `isinstance(employee, Person)` returns `True`.

---

### 7. **Using `isinstance()` with Abstract Classes**

You can also use `isinstance()` with abstract classes. An abstract class cannot be instantiated directly, but `isinstance()` can check if an object is an instance of a subclass of an abstract class.

#### Example:

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

circle = Circle(5)

print(isinstance(circle, Shape))  # Output: True (Circle is a subclass of Shape)
```

Here, `Shape` is an abstract class, and `Circle` is a concrete subclass that implements the abstract `area()` method. `isinstance(circle, Shape)` returns `True` because `Circle` is a subclass of `Shape`.

---

### 8. **Checking for Types of Built-In Data Structures**

`isinstance()` is often used to check if an object belongs to one of Python's built-in data types such as `list`, `tuple`, `dict`, `set`, etc.

#### Example:

```python
my_list = [1, 2, 3]
print(isinstance(my_list, list))  # Output: True (my_list is an instance of list)

my_dict = {"name": "John", "age": 30}
print(isinstance(my_dict, dict))  # Output: True (my_dict is an instance of dict)

my_set = {1, 2, 3}
print(isinstance(my_set, set))    # Output: True (my_set is an instance of set)
```

---

### 9. **Avoiding Excessive Use of `isinstance()`**

While `isinstance()` is powerful, excessive use of it can make your code less flexible and more rigid. It's often better to rely on **duck typing** in Python, which is based on the concept "If it walks like a duck and quacks like a duck, it is a duck."

Instead of explicitly checking the type with `isinstance()`, you should focus on whether the object behaves as expected (i.e., has the required methods or attributes).

#### Example of Duck Typing (Preferred Over `isinstance()`):

```python
class Dog:
    def speak(self):
        print("Woof!")

class Cat:
    def speak(self):
        print("Meow!")

def animal_sound(animal):
    animal.speak()

dog = Dog()
cat = Cat()

animal_sound(dog)  # Output: Woof!
animal_sound(cat)  # Output: Meow!
```

Here, we don’t need to check the type of `animal` with `isinstance()`. Instead, we just assume that the object passed to `animal_sound` has the `speak()` method, which is a core tenet of duck typing.

---

### 10. **Performance Considerations**

While `isinstance()` is quite efficient, you should be cautious about overusing it in performance-critical sections of your code. If you’re checking types frequently in loops or complex code, it could introduce unnecessary overhead.

- It supports **polymorphism** and **multiple inheritance**, allowing you to check if an object belongs to any class in its inheritance chain.

The **`isinstance()`** function in Python is used to check if an object is an instance of a specific class or type. It is a powerful tool for type checking and validation, and it supports advanced features like checking against multiple types and inheritance hierarchies. Below is a comprehensive guide to **`isinstance()`** from basic to advanced levels, including examples and use cases.

---

## **Basic Usage of `isinstance()`**

### 1. **Check if an Object is of a Specific Type**

- **Purpose**: Verify if an object belongs to a specific type (e.g., `int`, `str`, `list`).
- **Example**:
  ```python
  num = 42
  print(isinstance(num, int))  # True
  print(isinstance(num, str))  # False
  ```

---

### 2. **Check Against Multiple Types**

- **Purpose**: Check if an object is an instance of any type in a tuple of types.
- **Example**:
  ```python
  value = 3.14
  print(isinstance(value, (int, float)))  # True (value is a float)
  ```

---

### 3. **Check for Custom Classes**

- **Purpose**: Verify if an object is an instance of a user-defined class.
- **Example**:
  ```python
  class MyClass:
      pass
  obj = MyClass()
  print(isinstance(obj, MyClass))  # True
  ```

---

## **Intermediate Usage of `isinstance()`**

### 4. **Check for Inheritance**

- **Purpose**: Verify if an object is an instance of a parent class or subclass.
- **Example**:
  ```python
  class Parent:
      pass
  class Child(Parent):
      pass
  obj = Child()
  print(isinstance(obj, Parent))  # True (Child inherits from Parent)
  ```

---

### 5. **Check for Abstract Base Classes (ABCs)**

- **Purpose**: Verify if an object implements a specific interface or abstract base class.
- **Example**:
  ```python
  from collections.abc import Iterable
  my_list = [1, 2, 3]
  print(isinstance(my_list, Iterable))  # True (list is iterable)
  ```

---

### 6. **Check for Built-in Types**

- **Purpose**: Verify if an object is an instance of built-in types like `list`, `dict`, `set`, etc.
- **Example**:
  ```python
  my_dict = {"key": "value"}
  print(isinstance(my_dict, dict))  # True
  ```

---

### 7. **Check for `None`**

- **Purpose**: Verify if an object is `None`.
- **Example**:
  ```python
  value = None
  print(isinstance(value, type(None)))  # True
  ```

---

## **Advanced Usage of `isinstance()`**

### 8. **Check for Multiple Inheritance**

- **Purpose**: Verify if an object is an instance of a class with multiple inheritance.
- **Example**:
  ```python
  class A:
      pass
  class B:
      pass
  class C(A, B):
      pass
  obj = C()
  print(isinstance(obj, A))  # True
  print(isinstance(obj, B))  # True
  ```

---

### 9. **Check for Generic Types (Type Hints)**

- **Purpose**: Verify if an object matches a generic type (e.g., `List[int]`, `Dict[str, int]`).
- **Example**:
  ```python
  from typing import List, Dict
  my_list = [1, 2, 3]
  print(isinstance(my_list, List))  # True (Python 3.7+)
  print(isinstance(my_list, List[int]))  # False (Cannot check generic types at runtime)
  ```
- **Note**: `isinstance()` cannot check generic type parameters (e.g., `List[int]`). Use static type checkers like `mypy` for this purpose.

---

### 10. **Check for Callable Objects**

- **Purpose**: Verify if an object is callable (e.g., a function, method, or class).
- **Example**:
  ```python
  def my_function():
      pass
  print(isinstance(my_function, type(lambda: None)))  # True (function is callable)
  ```

---

### 11. **Check for Numeric Types**

- **Purpose**: Verify if an object is a numeric type (e.g., `int`, `float`, `complex`).
- **Example**:
  ```python
  import numbers
  value = 3.14
  print(isinstance(value, numbers.Real))  # True (float is a real number)
  ```

---

### 12. **Check for Custom Type Hierarchies**

- **Purpose**: Verify if an object is an instance of a custom type hierarchy.
- **Example**:
  ```python
  class Animal:
      pass
  class Dog(Animal):
      pass
  class Cat(Animal):
      pass
  my_pet = Dog()
  print(isinstance(my_pet, Animal))  # True (Dog inherits from Animal)
  ```

---

### 13. **Check for Protocol Classes (Python 3.8+)**

- **Purpose**: Verify if an object conforms to a protocol (duck typing).
- **Example**:
  ```python
  from typing import Protocol
  class SupportsClose(Protocol):
      def close(self) -> None:
          pass
  class File:
      def close(self) -> None:
          print("File closed")
  my_file = File()
  print(isinstance(my_file, SupportsClose))  # True (File implements close())
  ```

---

### 14. **Check for Enum Types**

- **Purpose**: Verify if an object is an instance of an `enum` type.
- **Example**:
  ```python
  from enum import Enum
  class Color(Enum):
      RED = 1
      GREEN = 2
      BLUE = 3
  my_color = Color.RED
  print(isinstance(my_color, Color))  # True
  ```

---

### 15. **Check for Dataclass Types (Python 3.7+)**

- **Purpose**: Verify if an object is an instance of a `dataclass`.
- **Example**:
  ```python

  @dataclass
  class Point:

      y: int
  p = Point(1, 2)
  print(isinstance(p, Point))  # True
  ```

## **Common Pitfalls and Best Practices**

### 1. **Avoid Overusing `isinstance()`**

- **Issue**: Overusing `isinstance()` can make code less flexible and harder to maintain.
- **Best Practice**: Use duck typing and rely on behavior rather than type checking where possible.

---

### 2. **Prefer `isinstance()` Over `type()`**

- **Issue**: Using `type(obj) == SomeClass` does not account for inheritance.
- **Best Practice**: Use `isinstance()` to handle inheritance properly:
  ```python
  class Parent:
      pass
  class Child(Parent):
      pass
  obj = Child()
  print(type(obj) == Parent)  # False
  print(isinstance(obj, Parent))  # True
  ```

---

### 3. **Use `isinstance()` with Abstract Base Classes (ABCs)**

- **Best Practice**: Use ABCs to define interfaces and check for them using `isinstance()`:
  ```python
  from collections.abc import Sequence
  my_list = [1, 2, 3]
  print(isinstance(my_list, Sequence))  # True
  ```

---

### 4. **Combine `isinstance()` with `try`-`except`**

- **Best Practice**: Use `try`-`except` blocks for behavior-based checks and `isinstance()` for type-based checks:
  ```python
  try:
      result = obj.some_method()
  except AttributeError:
      print("Object does not support some_method")
  ```
