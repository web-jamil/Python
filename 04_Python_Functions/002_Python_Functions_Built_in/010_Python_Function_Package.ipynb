## What are packages?

We don't usually store all of our files on our computer in the same location. We use a well-organized hierarchy of directories for easier access.

Similar files are kept in the same directory, for example, we may keep all the songs in the "**music**" directory. Analogous to this, Python has packages for directories and **[modules](https://github.com/milaan9/04_Python_Functions/blob/main/007_Python_Function_Module.ipynb)** for files.

As our application program grows larger in size with a lot of modules, we place similar modules in one package and different modules in different packages. This makes a project (program) easy to manage and conceptually clear.

Similarly, as a directory can contain subdirectories and files, a Python package can have sub-packages and modules.

A directory must contain a file named **`__init__.py`** in order for Python to consider it as a package. This file can be left empty but we generally place the initialization code for that package in this file.

## Importing module from a package

We can import modules from packages using the dot **`.`** operator.

For example, if we want to import the **`start`** module in the above example, it can be done as follows:

```python
>>>import Game.Level.start
```

Now, if this module contains a **[function](https://github.com/milaan9/04_Python_Functions/blob/main/001_Python_Functions.ipynb)** named **`select_difficulty()`**, we must use the full name to reference it.

```python
>>>Game.Level.start.select_difficulty(2)
```

If this construct seems lengthy, we can import the module without the package prefix as follows:

```python
>>>from Game.Level import start
```

We can now call the function simply as follows:

```python
>>>start.select_difficulty(2)
```

Another way of importing just the required function (or class or variable) from a module within a package would be as follows:

```python
>>>from Game.Level.start import select_difficulty
```

Now we can directly call this function.

```python
>>>select_difficulty(2)
```

Although easier, this method is not recommended. Using the full **[namespace](https://github.com/milaan9/01_Python_Introduction/blob/main/013_Python_Namespace_and_Scope.ipynb)** avoids confusion and prevents two same identifier names from colliding.

### **1. Understanding Python Packages**

- A **module** is simply a file containing Python code, which could define functions, classes, or variables.
- A **package** is a directory that contains multiple Python modules along with a special file called `__init__.py`. This file can be empty or can contain initialization code for the package.

#### **Structure of a Python Package:**

```plaintext
package_name/
    __init__.py
    module1.py
    module2.py
    subpackage/
        __init__.py
        submodule1.py
```

- The `__init__.py` file makes the directory a package.
- Modules inside the package (`module1.py`, `module2.py`) can be imported into your Python code.
- Subpackages can exist within a package as well (like the `subpackage` above).

- **subpackage** is another package within the main package.

### **How to Install Python Packages: A Complete Guide from Basic to Advanced**

Installing Python packages is an essential part of using Python for various tasks, especially in fields like **Data Science**, **Machine Learning**, and **Web Development**. The process can range from basic to advanced depending on your needs, from simple installations using **pip** to managing environments and dependencies using **conda**.

Here's an in-depth guide to installing Python packages, from basic to advanced.

---

### **1. Installing Python Packages Using pip (Basic)**

`pip` is the most commonly used package manager for Python. It allows you to install and manage libraries and dependencies for Python.

#### **Basic Installation with pip:**

1. **Using pip to Install a Package**:

   - Open your terminal or command prompt.
   - Type the following command to install a package (e.g., `numpy`):
     ```bash
     pip install numpy
     ```
   - This will download and install the `numpy` package from **PyPI (Python Package Index)**.

2. **Installing Multiple Packages at Once**:

   - You can install multiple packages by separating their names with a space:
     ```bash
     pip install numpy pandas matplotlib
     ```

3. **Installing a Specific Version of a Package**:

   - If you need a particular version of a package, you can specify the version:
     ```bash
     pip install numpy==1.21.0
     ```

4. **Upgrading an Existing Package**:

   - To upgrade a package to the latest version, you can use the `--upgrade` flag:
     ```bash
     pip install --upgrade numpy
     ```

5. **Listing Installed Packages**:

   - To view all installed packages, use:
     ```bash
     pip list
     ```

6. **Uninstalling a Package**:
   - To uninstall a package, use the following command:
     ```bash
     pip uninstall numpy
     ```

---

### **2. Installing Python Packages Using pip from a Requirements File (Intermediate)**

A **requirements.txt** file is commonly used to manage dependencies for a project. This file lists all the packages and their specific versions needed to run the project.

#### **Steps:**

1. **Creating a `requirements.txt` File**:

   - You can generate a `requirements.txt` file by running:
     ```bash
     pip freeze > requirements.txt
     ```

2. **Installing Packages from `requirements.txt`**:

   - To install all the packages listed in a `requirements.txt` file, use the following command:
     ```bash
     pip install -r requirements.txt
     ```

3. **Freezing Installed Packages**:
   - If you've already installed packages and want to create a `requirements.txt` with the exact versions of all installed packages, run:
     ```bash
     pip freeze > requirements.txt
     ```

---

### **3. Installing Python Packages Using Conda (Advanced)**

**Conda** is a powerful package manager and environment manager, commonly used for data science, machine learning, and scientific computing tasks. Conda handles Python packages as well as non-Python packages (e.g., libraries like `libxml2`, `openblas`, etc.).

#### **Using Conda for Package Management**:

1. **Install a Package Using Conda**:

   - To install a package using conda, type:
     ```bash
     conda install numpy
     ```
   - Conda will automatically check for dependencies and install them.

2. **Installing Specific Version of a Package**:

   - To install a specific version of a package:
     ```bash
     conda install numpy=1.21.0
     ```

3. **Creating a New Conda Environment and Installing Packages**:

   - You can create a new environment with specific packages. For example:
     ```bash
     conda create --name myenv numpy pandas matplotlib
     ```
   - This command creates a new environment named `myenv` and installs the specified packages.

4. **Activating and Deactivating Conda Environments**:

   - To activate a conda environment:
     ```bash
     conda activate myenv
     ```
   - To deactivate the environment:
     ```bash
     conda deactivate
     ```

5. **List Installed Packages in Conda Environment**:
   - To list all the packages installed in your current environment:
     ```bash
     conda list
     ```

---

### **4. Installing Python Packages from Source (Advanced)**

In some cases, you might need to install Python packages from source, especially if the package isn’t available on PyPI or if you want to work with the latest development version.

#### **Steps to Install from Source**:

1. **Clone the Repository**:

   - If the package is hosted on GitHub or another repository, you can clone it:
     ```bash
     git clone https://github.com/somepackage/somepackage.git
     ```

2. **Navigate to the Package Directory**:

   - Change to the directory where the setup script is located:
     ```bash
     cd somepackage
     ```

3. **Install the Package**:

   - Use `pip` to install the package from the source:
     ```bash
     pip install .
     ```
   - The `.` refers to the current directory where the `setup.py` file is located.

4. **Installing in Editable Mode**:
   - If you plan to make changes to the source code, you can install the package in editable mode:
     ```bash
     pip install -e .
     ```

---

### **5. Installing Packages in Jupyter Notebooks (Advanced)**

If you're working in a **Jupyter Notebook**, you can install Python packages directly from the notebook interface.

#### **Steps**:

1. **Install Packages Inside Jupyter Notebook**:
   - Use the following command to install packages:
     ```python
     !pip install numpy
     ```
   - Or, with conda:
     ```python
     !conda install numpy
     ```

### **6. Virtual Environments (Advanced)**

A **virtual environment** allows you to manage dependencies for different projects without conflicts.

#### **Creating and Using Virtual Environments**:

1. **Creating a Virtual Environment with `venv`**:

   - To create a virtual environment, run:
     ```bash
     python -m venv myenv
     ```

2. **Activating the Virtual Environment**:

   - On **Windows**:
     ```bash
     myenv\Scripts\activate
     ```
   - On **Mac/Linux**:
     ```bash
     source myenv/bin/activate
     ```

3. **Deactivating the Virtual Environment**:

   - Simply type:
     ```bash
     deactivate
     ```

4. **Installing Packages in the Virtual Environment**:
   - Once the environment is activated, you can install packages using `pip` or `conda`.

### **7. Advanced Package Management with `pipenv` and `poetry`**

`pipenv` and `poetry` are tools that help manage project dependencies, virtual environments, and the packaging process.

#### **Using Pipenv**:

  ```bash

  ```

- **Creating a Virtual Environment with Pipenv**:

  ```bash
  pipenv install numpy
  ```

- **Activate the Pipenv Shell**:
  ```bash
  pipenv shell
  ```

#### **Using Poetry**:

- **Install Poetry**:

  ```bash
  curl -sSL https://install.python-poetry.org | python3 -
  ```

- **Create a New Poetry Project**:

  ```bash
  poetry new myproject
  ```

- **Install Packages with Poetry**:
  ```bash
  poetry add numpy
  ```

### **8. Installing Packages for Specific Python Versions (Advanced)**

Sometimes, you need to install packages for specific Python versions (e.g., Python 2.x vs. Python 3.x). Here's how you can manage this:

#### **Use `pyenv` to Install Different Python Versions**:

1. **Install `pyenv`** (to manage multiple Python versions):

   - Follow the instructions [here](https://github.com/pyenv/pyenv).

2. **Install Specific Python Version**:

   - Example to install Python 3.8.5:
     ```bash
     pyenv install 3.8.5
     ```

3. **Switch Between Versions**:

   - Set the Python version globally:
     ```bash
     pyenv global 3.8.5
     ```

4. **Installing Packages for a Specific Python Version**:
   - Once the correct Python version is set, you can install packages as usual with `pip`.

**`pip`** is the **package installer for Python**. It is the standard tool for installing, managing, and uninstalling Python packages from the **Python Package Index (PyPI)** and other repositories. Below is a comprehensive guide to `pip`, covering everything from basic usage to advanced features.

---

## **1. Basic Concepts**

### **a. What is `pip`?**

- `pip` is a command-line tool for installing and managing Python packages.
- It connects to PyPI (Python Package Index) to download and install packages.
- It also resolves dependencies automatically.

### **b. Installing `pip`**

- `pip` is included by default with Python 3.4 and later.
- To check if `pip` is installed:
  ```bash
  pip --version
  ```
- If not installed, you can install it using:
  ```bash
  python -m ensurepip --upgrade
  ```

---

## **2. Basic Usage**

### **a. Installing Packages**

- Install a package from PyPI:
  ```bash
  pip install package_name
  ```
- Install a specific version:
  ```bash
  pip install package_name==1.2.3
  ```
- Install multiple packages:
  ```bash
  pip install package1 package2
  ```

### **b. Upgrading Packages**

- Upgrade an installed package to the latest version:
  ```bash
  pip install --upgrade package_name
  ```

### **c. Uninstalling Packages**

- Remove an installed package:
  ```bash
  pip uninstall package_name
  ```

### **d. Listing Installed Packages**

- List all installed packages and their versions:
  ```bash
  pip list
  ```

### **e. Checking Package Information**

- Show details about an installed package:
  ```bash
  pip show package_name
  ```

---

## **3. Intermediate Usage**

### **a. Installing from Requirements Files**

- Use a `requirements.txt` file to install multiple packages at once.
- Example `requirements.txt`:
  ```
  numpy==1.21.0
  pandas==1.3.0
  ```
- Install packages from `requirements.txt`:
  ```bash
  pip install -r requirements.txt
  ```

### **b. Freezing Installed Packages**

- Generate a `requirements.txt` file from installed packages:
  ```bash
  pip freeze > requirements.txt
  ```

### **c. Installing from a Local Directory**

- Install a package from a local directory:
  ```bash
  pip install /path/to/package
  ```

### **d. Installing from a Git Repository**

- Install a package directly from a Git repository:
  ```bash
  pip install git+https://github.com/username/repository.git
  ```

### **e. Installing in Editable Mode**

- Install a package in editable mode (useful for development):
  ```bash
  pip install -e /path/to/package
  ```

---

## **4. Advanced Usage**

### **a. Using Virtual Environments**

- Create a virtual environment:
  ```bash
  python -m venv myenv
  ```
- Activate the virtual environment:
  - On Windows:
    ```bash
    myenv\Scripts\activate
    ```
  - On macOS/Linux:
    ```bash
    source myenv/bin/activate
    ```
- Install packages in the virtual environment:
  ```bash
  pip install package_name
  ```

### **b. Installing from Wheels**

- Install a package from a `.whl` file:
  ```bash
  pip install package_name.whl
  ```

### **c. Using a Custom Index**

- Install a package from a custom PyPI index:
  ```bash
  pip install --index-url https://custom.pypi.org/simple package_name
  ```

### **d. Installing Pre-Release Versions**

- Install a pre-release version of a package:
  ```bash
  pip install --pre package_name
  ```

### **e. Ignoring Installed Packages**

- Reinstall a package even if it is already installed:
  ```bash
  pip install --ignore-installed package_name
  ```

## **5. Dependency Management**

### **a. Resolving Dependencies**

- `pip` automatically resolves and installs dependencies.
- Example:
  ```bash
  pip install pandas  # Installs pandas and its dependencies (e.g., numpy)
  ```

### **b. Checking for Dependency Conflicts**

- Check for dependency conflicts:
  ```bash
  pip check
  ```

### **c. Installing with No Dependencies**

- Install a package without its dependencies:
  ```bash
  pip install --no-deps package_name
  ```

### **a. Configuring `pip`**

- `pip` caches downloaded packages by default.
- Clear the cache:
  ```bash
  pip cache purge
  ```
