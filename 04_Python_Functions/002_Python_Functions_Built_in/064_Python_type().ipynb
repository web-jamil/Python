# Python `type()`

The **`type()`** function either returns the type of the object or returns a new type object based on the arguments passed.

**Syntax**:

```python
type(object)
type(name, bases, dict)
```


## `type()` With a Single Object Parameter

If a single **`object`** is passed to **`type()`**, the function returns its type.


````python
# Example 1: Get Type of an Object

numbers_list = [1, 2]
print(type(numbers_list))

numbers_dict = {1: 'one', 2: 'two'}
print(type(numbers_dict))

class Foo:
    a = 0

foo = Foo()
print(type(foo))
````

If you need to check the type of an object, it is better to use the **[Python isinstance() function](https://github.com/milaan9/04_Python_Functions/blob/main/002_Python_Functions_Built_in/034_Python_isinstance%28%29.ipynb)** instead. It's because the **`isinstance()` function** also checks if the given object is an instance of the subclass.


## `type()` With name, bases and dict Parameters

If three parameters are passed to **`type()`**, it returns a new type object.

The three parameters are:

| Parameter | Description                                                                                                         |
| :-------- | :------------------------------------------------------------------------------------------------------------------ |
| **name**  | **a class name; becomes the `__name__` attribute**                                                                  |
| **name**  | **a tuple that itemizes the base class; becomes the `__bases__` attribute**                                         |
| **name**  | **a dictionary which is the namespace containing definitions for the class body; becomes the `__dict__` attribute** |


````python
# Example 2: Create a type object

o1 = type('X', (object,), dict(a='Foo', b=12))

print(type(o1))
print(vars(o1))

class test:
    a = 'Foo'
    b = 12
  
o2 = type('Y', (test,), dict(a='Foo', b=12))
print(type(o2))
print(vars(o2))
````

In the program, we have used the **[Python vars() function](https://github.com/milaan9/04_Python_Functions/blob/main/002_Python_Functions_Built_in/065_Python_vars%28%29.ipynb)** which returns the **`__dict__`** attribute. **`__dict__`** is used to store object's writable attributes.

You can easily change these attributes if necessary. For example, if you need to change the **`__name__`** attribute of **`o1`** to **`'Z'`**, use:

```python
o1.__name = 'Z'
```

## **1. Basics of `type()`**

### **Checking the Type of an Object**

The `type()` function returns the type of an object.

```python
# Example 1: Integer
x = 42
print(type(x))  # Output: <class 'int'>

# Example 2: String
y = "Hello, World!"
print(type(y))  # Output: <class 'str'>

# Example 3: List
z = [1, 2, 3]
print(type(z))  # Output: <class 'list'>
```

### **Types of Built-in Objects**

Python has several built-in types, and `type()` can be used to identify them.

| Type       | Example            | `type()` Output      |
| ---------- | ------------------ | -------------------- |
| Integer    | `42`               | `<class 'int'>`      |
| Float      | `3.14`             | `<class 'float'>`    |
| String     | `"Hello"`          | `<class 'str'>`      |
| List       | `[1, 2, 3]`        | `<class 'list'>`     |
| Tuple      | `(1, 2, 3)`        | `<class 'tuple'>`    |
| Dictionary | `{"key": "value"}` | `<class 'dict'>`     |
| Set        | `{1, 2, 3}`        | `<class 'set'>`      |
| Boolean    | `True`             | `<class 'bool'>`     |
| None       | `None`             | `<class 'NoneType'>` |

## **2. Advanced Usage of `type()`**

### **Dynamic Type Checking**

You can use `type()` to check the type of an object dynamically.

```python
def check_type(obj):
    if type(obj) == int:
        print("It's an integer!")
    elif type(obj) == str:
        print("It's a string!")
    else:
        print("It's something else!")

check_type(42)       # Output: It's an integer!
check_type("Hello")  # Output: It's a string!
check_type([1, 2, 3])  # Output: It's something else!
```

### **Using `type()` with Custom Classes**

You can use `type()` to check the type of instances of custom classes.

```python
class Dog:
    def __init__(self, name):
        self.name = name

my_dog = Dog("Buddy")
print(type(my_dog))  # Output: <class '__main__.Dog'>
```

### **Creating Classes Dynamically with `type()`**

The `type()` function can also be used to create classes dynamically. This is an advanced feature and is rarely used in everyday programming.

```python
# Dynamically create a class
MyClass = type('MyClass', (), {'x': 42})

# Create an instance of the dynamically created class
obj = MyClass()
print(type(obj))  # Output: <class '__main__.MyClass'>
print(obj.x)      # Output: 42
```

## **3. Type Comparison**

### **`type()` vs `isinstance()`**

- `type()` checks the exact type of an object.
- `isinstance()` checks if an object is an instance of a class or a subclass.

```python
class Animal:
    pass

class Dog(Animal):
    pass

my_dog = Dog()

# Using type()
print(type(my_dog) == Dog)    # Output: True
print(type(my_dog) == Animal) # Output: False

# Using isinstance()
print(isinstance(my_dog, Dog))    # Output: True
print(isinstance(my_dog, Animal)) # Output: True
```

## **4. Practical Use Cases**

### **Debugging**

Use `type()` to debug and understand the type of objects in your code.

```python
def process_data(data):
    print(f"Processing data of type: {type(data)}")
    # Additional processing logic

process_data(42)       # Output: Processing data of type: <class 'int'>
process_data("Hello")  # Output: Processing data of type: <class 'str'>
```

### **Type Validation**

Use `type()` to validate the type of input data.

```python
def add_numbers(a, b):
    if type(a) != int or type(b) != int:
        raise TypeError("Both arguments must be integers")
    return a + b

print(add_numbers(2, 3))      # Output: 5
# print(add_numbers(2, "3"))  # Raises TypeError
```

## **5. Advanced Topics**

### **Metaclasses and `type()`**

In Python, `type()` is not just a function; it is also a metaclass. A metaclass is the type of a class, and it controls how a class is created.

```python
class Meta(type):
    def __new__(cls, name, bases, dct):
        print(f"Creating class {name}")
        return super().__new__(cls, name, bases, dct)

class MyClass(metaclass=Meta):
    pass

# Output: Creating class MyClass
```

### **Type Annotations and `type()`**

Python supports type annotations, which can be used with `type()` for advanced type checking.

```python
from typing import List

def process_items(items: List[int]):
    if type(items) != list:
        raise TypeError("Expected a list")
    for item in items:
        if type(item) != int:
            raise TypeError("All items must be integers")
    print("Processing items...")

process_items([1, 2, 3])  # Output: Processing items...
# process_items([1, "2", 3])  # Raises TypeError
```
