# Python `len()`

The **`len()`** function returns the number of items (length) in an object.

**Syntax**:

```python
len(s)
```


## `len()` Parameters

- **s** - a sequence (string, bytes, tuple, list, or range) or a collection (dictionary, set or frozen set)


## Return Value from `len()`

**`len()`** function returns the number of items of an object.

Failing to pass an argument or passing an invalid argument will raise a **`TypeError`** exception.


````python
# Example 1: How len() works with tuples, lists and range?

testList = []
print(testList, 'length is', len(testList))

testList = [1, 2, 3]
print(testList, 'length is', len(testList))

testTuple = (1, 2, 3)
print(testTuple, 'length is', len(testTuple))

testRange = range(1, 10)
print('Length of', testRange, 'is', len(testRange))
````

Internally, **`len()`** calls object's **`__len__`** method. You can think of **`len()`** as:

```python
def len(s):
    return s.__len__()
```

So, you can assign custom length to the object (if necessary)


````python
# Example 4: How len() works for custom objects?

class Session:
    def __init__(self, number = 0):
        self.number = number
    
    def __len__(self):
        return self.number


# default length is 0
s1 = Session()
print(len(s1))

# giving custom length
s2 = Session(6)
print(len(s2))
````

The **`len()`** function in Python is used to determine the number of items in an object, such as a string, list, tuple, dictionary, or other iterable. It is one of the most commonly used built-in functions in Python. Below is a comprehensive guide to the **`len()`** function, covering everything from basic concepts to advanced topics, including examples and explanations.

---

## **1. What is the `len()` Function?**

- **Definition**: The `len()` function returns the number of items in an object.
- **Purpose**: Used to determine the size or length of an object.
- **Example**:
  ```python
  my_list = [1, 2, 3, 4, 5]
  print(len(my_list))  # Output: 5
  ```

---

## **2. Basic Usage of `len()`**

### **2.1. Strings**

- **Purpose**: Get the number of characters in a string.
- **Example**:
  ```python
  my_string = "Hello, World!"
  print(len(my_string))  # Output: 13
  ```

### **2.2. Lists**

- **Purpose**: Get the number of elements in a list.
- **Example**:
  ```python
  my_list = [1, 2, 3, 4, 5]
  print(len(my_list))  # Output: 5
  ```

### **2.3. Tuples**

- **Purpose**: Get the number of elements in a tuple.
- **Example**:
  ```python
  my_tuple = (1, 2, 3, 4, 5)
  print(len(my_tuple))  # Output: 5
  ```

### **2.4. Dictionaries**

- **Purpose**: Get the number of key-value pairs in a dictionary.
- **Example**:
  ```python
  my_dict = {"a": 1, "b": 2, "c": 3}
  print(len(my_dict))  # Output: 3
  ```

### **2.5. Sets**

- **Purpose**: Get the number of elements in a set.
- **Example**:
  ```python
  my_set = {1, 2, 3, 4, 5}
  print(len(my_set))  # Output: 5
  ```

---

## **3. Intermediate Usage of `len()`**

### **3.1. Nested Data Structures**

- **Purpose**: Get the length of nested data structures.
- **Example**:
  ```python
  nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
  print(len(nested_list))  # Output: 3 (number of sublists)
  ```

### **3.2. Custom Objects**

- **Purpose**: Use `len()` with custom objects by defining the `__len__()` method.
- **Example**:

  ```python
  class MyClass:
      def __len__(self):
          return 10

  obj = MyClass()
  print(len(obj))  # Output: 10
  ```

### **3.3. Empty Objects**

- **Purpose**: Get the length of empty objects.
- **Example**:
  ```python
  empty_list = []
  empty_dict = {}
  print(len(empty_list))  # Output: 0
  print(len(empty_dict))  # Output: 0
  ```

---

## **4. Advanced Usage of `len()`**

### **4.1. Using `len()` with Generators**

- **Issue**: Generators do not have a length because they produce items on-the-fly.
- **Example**:

  ```python
  my_generator = (x for x in range(10))
  # print(len(my_generator))  # Raises TypeError: object of type 'generator' has no len()
  ```

- **Fix**: Convert the generator to a list or use a counter.
  ```python
  my_generator = (x for x in range(10))
  print(len(list(my_generator)))  # Output: 10
  ```

### **4.2. Using `len()` with NumPy Arrays**

- **Purpose**: Get the length of a NumPy array.
- **Example**:
  ```python
  import numpy as np
  arr = np.array([1, 2, 3, 4, 5])
  print(len(arr))  # Output: 5
  ```

### **4.3. Using `len()` with Pandas DataFrames**

- **Purpose**: Get the number of rows in a DataFrame.
- **Example**:
  ```python
  import pandas as pd
  df = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
  print(len(df))  # Output: 3 (number of rows)
  ```

---

## **5. Real-World Examples**

### **5.1. Validating Input Length**

- **Issue**: Ensure user input meets a specific length requirement.
- **Example**:
  ```python
  password = input("Enter a password: ")
  if len(password) < 8:
      print("Password must be at least 8 characters long.")
  else:
      print("Password is valid.")
  ```

### **5.2. Iterating Over a List**

- **Issue**: Use `len()` to iterate over a list using indices.
- **Example**:
  ```python
  my_list = [10, 20, 30, 40, 50]
  for i in range(len(my_list)):
      print(f"Index: {i}, Value: {my_list[i]}")
  ```

### **5.3. Checking for Empty Data Structures**

- **Issue**: Check if a data structure is empty.
- **Example**:
  ```python
  my_list = []
  if len(my_list) == 0:
      print("The list is empty.")
  ```

---

## **6. Common Pitfalls and Best Practices**

### **6.1. Avoid Using `len()` with Infinite Iterators**

- **Pitfall**: Infinite iterators (e.g., `itertools.count()`) do not have a length.
- **Best Practice**: Avoid using `len()` with infinite iterators.
- **Example**:
  ```python
  from itertools import count
  infinite_iterator = count()
  # print(len(infinite_iterator))  # Raises TypeError
  ```

### **6.2. Use `len()` with Caution on Large Data**

- **Pitfall**: Calling `len()` on large data structures can be memory-intensive.
- **Best Practice**: Use generators or lazy evaluation for large datasets.
- **Example**:
  ```python
  large_list = range(10**6)
  print(len(large_list))  # Output: 1000000 (but uses minimal memory)
  ```

### **6.3. Use `len()` for Readability**

- **Best Practice**: Use `len()` to make code more readable and expressive.
- **Example**:
  ```python
  my_list = [1, 2, 3, 4, 5]
  if len(my_list) > 0:
      print("List is not empty.")
  ```

---

## **7. Summary**

The `len()` function is a versatile and essential tool in Python for determining the size or length of objects. Key points to remember:

- **Basic Usage**: Works with strings, lists, tuples, dictionaries, and sets.
- **Intermediate Usage**: Handles nested data structures and custom objects.
- **Advanced Usage**: Works with NumPy arrays, Pandas DataFrames, and generators.
- **Best Practices**: Avoid using `len()` with infinite iterators and large datasets.
