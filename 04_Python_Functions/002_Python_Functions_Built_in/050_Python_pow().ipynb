# Python `pow()`

The **`pow()`** function returns the power of a number.

**Syntax**:

```python
pow(x, y, z)
```


## `pow()` Parameters

The pow() function takes three parameters:

- **x** - a number, the base
- **y** - a number, the exponent
- **z (optional)** - a number, used for modulus

Hence,

- **`pow(x, y)`** is equal to **$x^y$**
- **`pow(x, y, z)`** is equal to **$x^y$`% z`**


````python
# Example 1: Python pow()

# positive x, positive y (x**y)
print(pow(2, 2))    # 4

# negative x, positive y
print(pow(-2, 2))    # 4  

# positive x, negative y
print(pow(2, -2))    # 0.25

# negative x, negative y
print(pow(-2, -2))    # 0.25
````

````python
# Example 2: pow() with three arguments (x**y) % z

x = 7
y = 2
z = 5

print(pow(x, y, z))    # 4
````

Here, 7 powered by 2 equals 49. Then, 49 modulus 5 equals 4.

## **1. Basics of `pow()`**

### **Calculating Powers**

The `pow()` function calculates the power of a number. It takes two arguments:

- `base`: The number to be raised to a power.
- `exponent`: The power to which the base is raised.

```python
# Example 1: Calculate 2^3
result = pow(2, 3)
print(result)  # Output: 8

# Example 2: Calculate 5^2
result = pow(5, 2)
print(result)  # Output: 25
```

### **Using `pow()` with Integers and Floats**

The `pow()` function works with both integers and floating-point numbers.

```python
# Example 1: Integer base and exponent
print(pow(2, 3))  # Output: 8

# Example 2: Floating-point base and exponent
print(pow(2.5, 2))  # Output: 6.25
```

---

## **2. Advanced Usage of `pow()`**

### **Modular Exponentiation**

The `pow()` function can take a third argument, `mod`, to perform **modular exponentiation**. This is useful in cryptography and number theory.

```python
# Example: Calculate (2^3) % 5
result = pow(2, 3, 5)
print(result)  # Output: 3 (since 8 % 5 = 3)
```

### **Negative Exponents**

You can use negative exponents to calculate the reciprocal of a number.

```python
# Example: Calculate 2^-3
result = pow(2, -3)
print(result)  # Output: 0.125 (since 2^-3 = 1 / 2^3 = 1/8)
```

### **Fractional Exponents**

Fractional exponents can be used to calculate roots.

```python
# Example: Calculate the square root of 16
result = pow(16, 0.5)
print(result)  # Output: 4.0

# Example: Calculate the cube root of 27
result = pow(27, 1/3)
print(result)  # Output: 3.0
```

## **3. Practical Use Cases**

### **Cryptographic Calculations**

Modular exponentiation is widely used in cryptographic algorithms like RSA.

```python
# Example: Calculate (base^exponent) % mod for RSA
base = 123
exponent = 17
mod = 3233
result = pow(base, exponent, mod)
print(result)  # Output: 855
```

### **Scientific Computations**

`pow()` is useful for scientific calculations involving exponents.

```python
# Example: Calculate the volume of a sphere (4/3 * π * r^3)
import math
radius = 5
volume = (4/3) * math.pi * pow(radius, 3)
print(volume)  # Output: 523.5987755982989
```

## **4. Advanced Topics**

### **Comparison with `**` Operator\*\*

The `**` operator can also be used to calculate powers, but `pow()` is more versatile because it supports modular exponentiation.

```python
# Example: Using ** operator
result = 2 ** 3
print(result)  # Output: 8

# Example: Using pow()
result = pow(2, 3)
print(result)  # Output: 8
```

### **Custom Power Functions**

You can create custom power functions for specific use cases.

```python
# Example: Custom power function with logging
def custom_pow(base, exponent):
    print(f"Calculating {base}^{exponent}")
    return base ** exponent

result = custom_pow(2, 3)
print(result)  # Output: Calculating 2^3, 8
```

### **Handling Large Exponents**

For very large exponents, `pow()` is more efficient than the `**` operator.

```python
# Example: Calculate 2^1000
result = pow(2, 1000)
print(result)  # Output: A very large number
```
