## Python Exception Handling: From Basic to Advanced

Exception handling is an essential concept in Python programming, allowing developers to handle runtime errors gracefully. It ensures that the program doesn’t crash unexpectedly and provides a mechanism for managing errors that occur during execution.

### **1. Basic Concepts of Exception Handling**

#### 1.1 What is an Exception?

An exception is an event that disrupts the normal flow of the program. These events can be caused by errors like dividing by zero, trying to access a non-existent file, or type mismatches in the code.

#### 1.2 The `try` and `except` Block

At the core of Python's exception handling are `try` and `except` blocks.

- **`try` block**: Code that might cause an exception is placed inside the `try` block.
- **`except` block**: The code inside this block is executed when an exception occurs.

#### **Basic Syntax**:

```python
try:
    # Code that may raise an exception
    x = 10 / 0  # Division by zero error
except ZeroDivisionError as e:
    print(f"Error: {e}")
```

- **`ZeroDivisionError`** is the type of exception raised when dividing by zero.
- The `except` block catches and handles the exception.

### **2. Common Exception Types**

Here are some common built-in exceptions in Python:

1. **`ZeroDivisionError`**: Raised when dividing by zero.
2. **`FileNotFoundError`**: Raised when trying to access a file that does not exist.
3. **`ValueError`**: Raised when an operation or function receives an argument with the right type but an inappropriate value.
4. **`IndexError`**: Raised when trying to access an index that is out of range in a list or other sequence.
5. **`TypeError`**: Raised when an operation or function is applied to an object of inappropriate type.

### **3. Handling Multiple Exceptions**

You can handle multiple exceptions in a single `try-except` block by using multiple `except` clauses.

#### **Example**:

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except ValueError:
    print("Error: Invalid input, please enter an integer!")
```

In this example:

- **`ZeroDivisionError`** is caught when dividing by zero.
- **`ValueError`** is caught when the user input is not an integer.

### **4. Catching All Exceptions with a Generic `except`**

While it’s generally a good practice to catch specific exceptions, you can catch all exceptions with a generic `except` clause:

```python
try:
    # Code that may raise any exception
    num = int(input("Enter a number: "))
    result = 10 / num
except Exception as e:
    print(f"An error occurred: {e}")
```

Here, `Exception` is the base class for all built-in exceptions, so it will catch any exception that occurs.

### **5. Using `else` and `finally` Blocks**

#### 5.1. The `else` Block

The `else` block runs if the `try` block doesn't raise any exceptions. It allows you to run code that should only be executed when no exception occurs.

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
else:
    print(f"Result is: {result}")
```

#### 5.2. The `finally` Block

The `finally` block is executed no matter what, even if an exception occurs or not. It's useful for cleaning up resources, like closing files or releasing network connections.

```python
try:
    file = open("example.txt", "r")
    content = file.read()
except FileNotFoundError:
    print("File not found.")
finally:
    file.close()  # Always closes the file, regardless of whether an exception occurred
```

## **6. Raising Exceptions**

You can raise your own exceptions using the `raise` keyword. This is useful for custom error handling in your code.

#### **Basic Syntax**:

```python
raise Exception("Custom error message")
```

You can raise exceptions with specific types for more control over error handling.

#### **Example**:

```python
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be at least 18.")
    else:
        print("Age is valid.")

try:
    validate_age(16)
except ValueError as e:
    print(f"Error: {e}")
```

In this case, if the age is less than 18, a `ValueError` will be raised with a custom message.

---

## **7. Custom Exceptions**

You can define your own exceptions by creating a new class that inherits from Python's built-in `Exception` class. This allows you to create specific error types for your application.

#### **Example**:

```python
class AgeLimitError(Exception):
    pass

def validate_age(age):
    if age < 18:
        raise AgeLimitError("Age must be at least 18.")

try:
    validate_age(16)
except AgeLimitError as e:
    print(f"Custom error: {e}")
```

### **8. Exception Chaining (`from` Keyword)**

When raising an exception while handling another exception, Python allows you to use the `from` keyword to provide more context about the original exception.

#### **Example**:

```python
try:
    a = 1 / 0
except ZeroDivisionError as e:
    raise ValueError("A custom error occurred") from e
```

Here, a `ZeroDivisionError` is caught, and a new `ValueError` is raised with the original exception attached. This is helpful in debugging because it shows the context of the error.

---

## **9. Exception Handling in Loops**

You can also use exception handling inside loops to catch errors during repeated operations.

#### **Example**:

```python
numbers = [10, 0, 5]

for num in numbers:
    try:
        print(10 / num)
    except ZeroDivisionError:
        print("Cannot divide by zero!")
```

## **10. Best Practices**

Here are some best practices to follow when using exception handling in Python:

1. **Catch Specific Exceptions**: Always try to catch specific exceptions instead of using a generic `except Exception` block. This helps in identifying and debugging issues more accurately.
2. **Avoid Using Bare `except`**: A bare `except` block can catch all exceptions, including those that should be handled elsewhere. It is better to explicitly specify which exceptions you want to catch.

3. **Use `finally` for Cleanup**: The `finally` block is the perfect place to release resources like file handles or network connections.

4. **Don't Use Exceptions for Flow Control**: Exceptions should not be used for normal flow control, such as checking conditions that you expect might occur often. They are meant for exceptional situations, not normal operation.

5. **Document Custom Exceptions**: If you create custom exceptions, document them well, especially their usage and meaning.

---

## **11. Summary**

- **`try` block**: Contains code that may raise an exception.
- **`except` block**: Handles exceptions.
- **`else` block**: Executes when no exception occurs.
- **`finally` block**: Always executes, regardless of whether an exception occurred or not.
- **`raise` keyword**: Used to explicitly raise exceptions, including custom exceptions.
- **Custom Exceptions**: You can define your own exceptions by subclassing the `Exception` class.
- **Exception Chaining**: Use the `from` keyword to chain exceptions and provide better context for errors.

Exception handling in Python is a powerful tool for managing errors and ensuring that your programs run smoothly, even in the face of unexpected issues. By mastering these techniques, you can write more robust, error-resistant code.

```python
file = None
try:
    file = open("data.txt")
    # Process file
except IOError:
    print("File error")
finally:
    if file:
        file.close()  # Always runs
```

Create application-specific errors.

```python
class InvalidEmailError(ValueError):
    """Raised when email format is invalid"""
    def __init__(self, email):
        super().__init__(f"Invalid email: {email}")
        self.email = email

# Usage
email = "bad-email"
if "@" not in email:
    raise InvalidEmailError(email)
```

## **9. Context Managers (`with`)**

Automated resource cleanup.

```python
with open("file.txt") as f:
    content = f.read()
# File automatically closed
```

**Custom context manager:**

```python
class DatabaseConnection:
    def __enter__(self):
        self.conn = connect_db()
        return self.conn

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.conn.close()
        if exc_type:
            logging.error("DB error", exc_info=True)

# Usage
with DatabaseConnection() as db:
    db.execute_query(...)
```

## **10. Exception Groups (Python 3.11+)**

Handle multiple exceptions simultaneously.

```python
try:
    raise ExceptionGroup("group", [ValueError(1), TypeError(2)])
except* ValueError:
    print("Handled ValueError")
except* TypeError:
    print("Handled TypeError")
```

Inspect and modify tracebacks.

```python
import traceback

try:
    # ...
except Exception:
    traceback.print_exc()  # Print to stderr
    tb = traceback.format_exc()  # Get as string
```

## **12. Exception Hooks**

Global exception handling.

```python
import sys

def global_except_hook(exctype, value, traceback):
    print(f"Global handler caught: {exctype.__name__}: {value}")

sys.excepthook = global_except_hook
```

## **13. Common Anti-Patterns**

**❌ Bare except:**

```python
try:
    # ...
except:  # Catches ALL exceptions including KeyboardInterrupt
    pass
```

**❌ Silent catching:**

```python
try:
    # ...
except Exception:
    pass  # Hides all errors
```

**✅ Better:**

```python
try:
    # ...
except SpecificError as e:
    handle_error(e)
```
