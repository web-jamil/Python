# Python `all()`

The **`all()`** method returns **`True`** when all elements in the given iterable are true. If not, it returns **`False`**.

**Syntax**:

```python
all(iterable)
```


## `all()` Parameters

**`any()`** method takes a single parameter:

- **iterable** - any iterable (**[list](https://github.com/milaan9/02_Python_Datatypes/blob/main/003_Python_List.ipynb)**, **[tuple](https://github.com/milaan9/02_Python_Datatypes/blob/main/004_Python_Tuple.ipynb)**, **[dictionary](https://github.com/milaan9/02_Python_Datatypes/blob/main/005_Python_Dictionary.ipynb)**, etc.) which contains the elements


## Return Value from `all()`

**`all()`** method returns:

- **`True`** If all elements in an iterable are true
- **`False`** If any element in an iterable is false

**Truth table for **`all()`\*\*\*\*:

| When                                     | Return Value |
| :--------------------------------------- | :----------- |
| **All values are true**                  | **True**     |
| **All values are false**                 | **False**    |
| **One value is true (others are false)** | **False**    |
| **One value is false (others are true)** | **False**    |
| **Empty Iterable**                       | **True**     |


````python
# Example 1: How all() works for lists?

# all values true
l = [1, 3, 4, 5]
print(all(l))

# all values false
l = [0, False]
print(all(l))

# one false value
l = [1, 3, 4, 0]
print(all(l))

# one true value
l = [0, False, 5]
print(all(l))

# empty iterable
l = []
print(all(l))
````

The **`all()`** method works in a similar way for tuples and sets like lists.


````python
# Example 2: How all() works for strings?

s = "This is good"
print(all(s))

# 0 is False
# '0' is True
s = '000'
print(all(s))

s = ''
print(all(s))
````

````python
# Example 3: Using all() with Python Dictionaries

# In the case of dictionaries, if all keys (not values) are false or the dictionary is empty, 
# `all()` returns `True`. Else, it returns false for all other cases.

s = {0: 'False', 1: 'False'}
print(all(s))

s = {1: 'True', 2: 'True'}
print(all(s))

s = {1: 'True', False: 0}
print(all(s))

s = {}
print(all(s))

# 0 is False
# '0' is True
s = {'0': 'True'}
print(all(s))
````

````python
print("--- Python's Built-in all() Function ---")
print("---------------------------------------\n")

# 1. Basic Usage: Checking if all elements are True
print("1. Basic Usage:")
list1 = [True, True, True]
list2 = [True, False, True]
tuple1 = (1, 2, 3) # Non-zero numbers are considered True
tuple2 = (1, 0, 3) # 0 is considered False
string1 = "hello"  # Non-empty strings are True
string2 = ""       # Empty string is False

print(f"all({list1})  -> {all(list1)}")
print(f"all({list2}) -> {all(list2)}")
print(f"all({tuple1}) -> {all(tuple1)}")
print(f"all({tuple2}) -> {all(tuple2)}")
print(f"all('{string1}') -> {all(string1)}")
print(f"all('{string2}')    -> {all(string2)}\n")

# 2. What is considered 'False' in Python?
#    - False
#    - None
#    - Numeric zero (0, 0.0, 0j)
#    - Empty sequences ('', [], (), {})
#    - Empty sets (set())
print("2. What is considered 'False' for all():")
falsey_values = [False, None, 0, 0.0, 0j, '', [], (), {}, set()]
print(f"all({falsey_values}) -> {all(falsey_values)}")
print(f"all([1, 2, None]) -> {all([1, 2, None])}")
print(f"all(['apple', '', 'banana']) -> {all(['apple', '', 'banana'])}\n")


# 3. Empty Iterable
print("3. Empty Iterable:")
# all() returns True for an empty iterable. This is often
# intuitively explained as "there are no false elements."
print(f"all([])    -> {all([])}")
print(f"all(())    -> {all(())}")
print(f"all('')    -> {all('')}")
print(f"all({{}})   -> {all({})}")
print(f"all(set()) -> {all(set())}\n")


# 4. Using with Generator Expressions
print("4. Using with Generator Expressions:")
# This is memory-efficient for large iterables as it doesn't create
# a full list in memory. It stops checking as soon as it finds a False.
numbers = [1, 2, 3, 4, 5]
is_all_positive = all(x > 0 for x in numbers)
print(f"Are all numbers in {numbers} positive? {is_all_positive}")

numbers_with_zero = [1, 2, 0, 4, 5]
is_all_positive_or_non_zero = all(x > 0 for x in numbers_with_zero)
print(f"Are all numbers in {numbers_with_zero} positive? {is_all_positive_or_non_zero}")

# Example with string check
words = ["hello", "world", "python"]
all_start_with_h = all(word.startswith('h') for word in words)
print(f"Do all words in {words} start with 'h'? {all_start_with_h}")

words2 = ["hello", "howdy", "hi"]
all_start_with_h2 = all(word.startswith('h') for word in words2)
print(f"Do all words in {words2} start with 'h'? {all_start_with_h2}\n")


# 5. Practical Use Cases
print("5. Practical Use Cases:")

# a) Validating user input
def validate_scores(scores):
    # Check if all scores are between 0 and 100
    if all(0 <= score <= 100 for score in scores):
        print(f"Scores {scores} are all valid.")
    else:
        print(f"Scores {scores} contain invalid entries.")

validate_scores([85, 92, 78])
validate_scores([90, -5, 105, 70])

# b) Checking conditions on a list of objects
class Car:
    def __init__(self, brand, is_electric):
        self.brand = brand
        self.is_electric = is_electric

    def __repr__(self):
        return f"Car({self.brand}, Electric:{self.is_electric})"

cars = [
    Car("Tesla", True),
    Car("Nissan", True),
    Car("Ford", False)
]

all_electric = all(car.is_electric for car in cars)
print(f"\nAre all cars {cars} electric? {all_electric}")

electric_cars = [
    Car("Tesla", True),
    Car("Lucid", True)
]
all_electric_2 = all(car.is_electric for car in electric_cars)
print(f"Are all cars {electric_cars} electric? {all_electric_2}\n")

# c) Checking if a dictionary has all required keys
required_keys = {"name", "age", "city"}
data1 = {"name": "Alice", "age": 30, "city": "New York"}
data2 = {"name": "Bob", "age": 25}

has_all_keys1 = all(key in data1 for key in required_keys)
print(f"Does {data1} have all required keys {required_keys}? {has_all_keys1}")

has_all_keys2 = all(key in data2 for key in required_keys)
print(f"Does {data2} have all required keys {required_keys}? {has_all_keys2}\n")

print("--- End of all() Function Demonstration ---")
````

````python
# The all() function returns True if all elements in an iterable are True,
# otherwise, it returns False.

# Example 1: Using all() with a list of boolean values
bool_list = [True, True, False]
print(all(bool_list))  # Output: False

# Example 2: Using all() with a list of integers
num_list = [1, 2, 3, 0]  # 0 is considered False
print(all(num_list))  # Output: False

# Example 3: Using all() with an empty iterable
empty_list = []
print(all(empty_list))  # Output: True (because there are no False values)

# Example 4: Using all() with a list of strings
string_list = ["Hello", "World", ""]
print(all(string_list))  # Output: False (empty string is considered False)

# Example 5: Using all() with a generator expression
numbers = [2, 4, 6, 8]
print(all(n % 2 == 0 for n in numbers))  # Output: True (all numbers are even)
````
