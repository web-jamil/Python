# Python `hex()`

The **`hex()`** function converts an integer number to the corresponding hexadecimal string.

**Syntax**:

```python
hex(x)
```


## `hex()` Parameters

**`hex()`** function takes a single argument.

- **x** - integer number (int object or it has to define **`__index__()`** method that returns an integer)


## Return Value from `hex()`

**`hex()`** function converts an integer to the corresponding hexadecimal number in string form and returns it.

The returned hexadecimal string starts with the prefix **`0x`** indicating it's in hexadecimal form.


````python
# Example 1: How hex() works?

number = 435
print(number, 'in hex =', hex(number))

number = 0
print(number, 'in hex =', hex(number))

number = -34
print(number, 'in hex =', hex(number))

returnType = type(hex(number))
print('Return type from hex() is', returnType)
````

If you need to find a hexadecimal representation of a float, you need to use **`float.hex()`** method.


````python
# Example 2: Hexadecimal representation of a float

number = 2.5
print(number, 'in hex =', float.hex(number))

number = 0.0
print(number, 'in hex =', float.hex(number))

number = 10.5
print(number, 'in hex =', float.hex(number))
````

### **Python `hex()` Function: A Comprehensive Guide (From Basic to Advanced)**

The `hex()` function in Python is used to convert an integer into its **hexadecimal representation** (base 16). Hexadecimal numbers are widely used in programming, especially for representing memory addresses, colors in web development, and for low-level programming tasks.

In this guide, we'll explore everything you need to know about the `hex()` function, from basic usage to more advanced applications and formatting tricks.

---

### **1. Basic Usage of `hex()`**

#### **1.1 Syntax**

```python
hex(number)
```

- **`number`**: The integer you want to convert to its hexadecimal representation. It must be an integer (positive, negative, or zero).
- **Returns**: A string representing the hexadecimal value, prefixed with `0x` to indicate that it's a hexadecimal number.

#### **1.2 Example: Basic Conversion**

```python
n = 255
print(hex(n))  # Output: 0xff
```

**Explanation**:

- The `hex()` function converts the decimal number `255` into its hexadecimal representation, which is `0xff`. The result is returned as a string prefixed with `0x`.

#### **1.3 Example: Negative Numbers**

```python
n = -255
print(hex(n))  # Output: -0xff
```

**Explanation**:

- The function works with negative numbers as well. The result for `-255` is `-0xff`, with the negative sign preserved.

#### **1.4 Example: Zero**

```python
n = 0
print(hex(n))  # Output: 0x0
```

**Explanation**:

- `0` is converted to `0x0` in hexadecimal.

---

### **2. Hexadecimal Representation**

In hexadecimal, each digit represents 4 bits (half a byte), and it uses 16 symbols: `0-9` and `a-f`. The hexadecimal number system is commonly used in computing because it aligns nicely with the binary system (since each hex digit represents 4 binary bits).

#### **2.1 Hexadecimal Conversion Example**

Let's manually convert a decimal number to hexadecimal to understand the process:

- **Decimal 255 → Hexadecimal FF**:
  - 255 ÷ 16 = 15 remainder 15 (F in hex)
  - 15 ÷ 16 = 0 remainder 15 (F in hex)
  - Therefore, `255` in decimal is `FF` in hexadecimal.

```python
n = 255
print(hex(n))  # Output: 0xff
```

---

### **3. Advanced Usage and Features**

#### **3.1 Removing the `0x` Prefix**

By default, the `hex()` function returns a string prefixed with `0x` to denote that the number is in hexadecimal. If you only want the hexadecimal digits without the prefix, you can slice the string:

```python
n = 255
hex_value = hex(n)[2:]  # Remove '0x' prefix
print(hex_value)  # Output: ff
```

**Explanation**:

- The slicing operation `[2:]` removes the first two characters (`0x`) from the string returned by `hex()`, leaving just the hexadecimal digits.

#### **3.2 Converting Back from Hexadecimal to Decimal**

To convert a hexadecimal string back to a decimal number, you can use the `int()` function with base 16:

```python
hex_str = 'ff'  # Hexadecimal representation
decimal_value = int(hex_str, 16)  # Convert hex string to decimal
print(decimal_value)  # Output: 255
```

**Explanation**:

- `int(hex_str, 16)` converts the hexadecimal string `'ff'` back to its decimal equivalent `255`.

#### **3.3 Padding Hexadecimal Numbers**

If you need to format hexadecimal numbers with a fixed width (for example, for consistent display), you can use string formatting.

```python
n = 255
formatted_hex = f"{hex(n)[2:]:>6}"  # Pad with spaces for a width of 6
print(formatted_hex)  # Output: '   ff'
```

**Explanation**:

- The `f"{...:>6}"` syntax ensures the string is right-aligned and padded with spaces to a total width of 6 characters.

Alternatively, you can use `zfill()` to pad with zeroes:

```python
n = 255
padded_hex = hex(n)[2:].zfill(6)  # Pad with zeros to make the length 6
print(padded_hex)  # Output: 0000ff
```

**Explanation**:

- `.zfill(6)` ensures the hexadecimal string has a length of 6 by padding it with leading zeros.

#### **3.4 Handling Uppercase Hexadecimal**

By default, the `hex()` function returns hexadecimal characters in lowercase (e.g., `0xabc`). If you want the hexadecimal string to be in uppercase, you can use the `upper()` method:

```python
n = 255
uppercase_hex = hex(n)[2:].upper()  # Convert to uppercase
print(uppercase_hex)  # Output: FF
```

**Explanation**:

- `upper()` converts all the hexadecimal characters to uppercase (`ff` becomes `FF`).

---

### **4. Common Use Cases for Hexadecimal**

#### **4.1 Memory Addresses**

Hexadecimal is commonly used to represent memory addresses in low-level programming, such as when working with pointers in languages like C. In Python, although you don't deal directly with pointers, hexadecimal can still be useful for debugging or working with system-level tools.

```python
address = 1234567890
print(hex(address))  # Output: 0x499602d2
```

**Explanation**:

- The `hex()` function provides a readable hexadecimal representation of an integer, which can be useful when working with memory addresses in debugging.

#### **4.2 Colors in Web Development**

In web development, colors are often represented in hexadecimal format. For example, the color `#ff5733` corresponds to a certain RGB value.

```python
# Convert RGB values to hexadecimal color
red = 255
green = 87
blue = 51
hex_color = f"#{red:02x}{green:02x}{blue:02x}"
print(hex_color)  # Output: #ff5733
```

**Explanation**:

- The format `"{value:02x}"` ensures that each color component (red, green, and blue) is converted to a two-digit hexadecimal number (with leading zeros if necessary).

#### **4.3 Working with Flags and Bit Masks**

In many low-level applications, flags and bit masks are represented as hexadecimal numbers because it's easier to group bits in sets of 4 (one hex digit per 4 bits).

```python
flag = 0x0F  # 00001111 in binary
mask = 0xF0  # 11110000 in binary

# Use bitwise AND to apply the mask
result = flag & mask
print(hex(result))  # Output: 0x0
```

**Explanation**:

- In this example, the bitwise AND between `0x0F` and `0xF0` results in `0x00`. The hexadecimal format is useful for reading and working with such masks.

### **5. Edge Cases and Considerations**

#### **5.1 Handling Large Numbers**

Python handles arbitrarily large integers, so the `hex()` function will work with large integers. However, hexadecimal strings for very large integers can become quite long:

```python
n = 1234567890123456789012345678901234567890
print(hex(n))  # Output: 0x112210f4b47ac17bbf03948bb4f8aabbf77f2c
```

**Explanation**:

- Python can handle very large integers, and the `hex()` function will provide their hexadecimal representation without any overflow.

#### **5.2 Negative Numbers**

The `hex()` function also works with negative integers, and the result will include a negative sign:

```python
n = -255
print(hex(n))  # Output: -0xff
```

**Explanation**:

- The function properly handles negative integers by returning a string prefixed with `-0x`.

The `hex()` function in Python is a built-in function used to convert an integer into its hexadecimal (base-16) representation. Hexadecimal numbers use digits from `0` to `9` and letters `A` to `F` (or `a` to `f`) to represent values from 10 to 15. This function is widely used in low-level programming, debugging, and working with systems that use hexadecimal representations (e.g., memory addresses, color codes).

Below is a comprehensive guide to the `hex()` function, covering everything from basic usage to advanced applications.

---

## **1. Basic Usage of `hex()`**

### **Syntax**

```python
hex(integer)
```

- **Parameter**: `integer` (can be an integer of any size, positive or negative).
- **Return Value**: A string prefixed with `0x` representing the hexadecimal value of the input integer.

### **Examples**

```python
# Convert integer to hexadecimal
print(hex(10))       # Output: '0xa'
print(hex(255))      # Output: '0xff'
print(hex(-10))      # Output: '-0xa'
```

---

## **2. Working with Different Input Types**

### **Integers**

- The `hex()` function works directly with integers.

```python
print(hex(16))       # Output: '0x10'
print(hex(1024))     # Output: '0x400'
```

### **Floats**

- Floats are not supported and will raise a `TypeError`.

```python
# print(hex(10.5))   # Raises TypeError: 'float' object cannot be interpreted as an integer
```

### **Strings**

- Strings representing integers can be converted to hexadecimal after converting them to integers.

```python
num_str = "255"
print(hex(int(num_str)))  # Output: '0xff'
```

---

## **3. Advanced Usage**

### **Custom Hexadecimal Conversion Without `0x` Prefix**

- If you want the hexadecimal value without the `0x` prefix, you can slice the string.

```python
hex_value = hex(255)
print(hex_value[2:])  # Output: 'ff'
```

### **Hexadecimal Representation of Binary or Octal Numbers**

- Convert binary or octal numbers to integers first, then to hexadecimal.

```python
binary_num = 0b1010   # Binary representation of 10
octal_num = 0o12      # Octal representation of 10

print(hex(binary_num))  # Output: '0xa'
print(hex(octal_num))   # Output: '0xa'
```

### **Hexadecimal Representation of Negative Numbers**

- Negative integers are converted to their hexadecimal representation with a `-0x` prefix.

```python
print(hex(-10))  # Output: '-0xa'
```
