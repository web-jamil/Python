# Python Anonymous/**`lambda`** Function

In this class, you'll learn about the anonymous function, also known as **`lambda`** functions. You'll learn what they are, their syntax and how to use them (with examples).


## What are **`lambda`** functions in Python?

In Python, an anonymous function is a **[function](https://github.com/milaan9/04_Python_Functions/blob/main/001_Python_Functions.ipynb)** that is defined without a name.

While normal functions are defined using the **`def`** keyword in Python, anonymous functions are defined using the **`lambda`** keyword.

In opposite to a normal function, a Python **`lambda`** function is a single expression. But, in a lambda body, we can expand with expressions over multiple lines using parentheses **`()`** or a multiline string **`""" """`**.

For example: **`lambda n:n+n`**

The reason behind the using anonymous function is for instant use, that is, one-time usage and the code is very concise so that there is more readability in the code.

Hence, anonymous functions are also called **`lambda`** functions.

- Lambda forms can take any number of arguments but return just one value in the form of an expression. They cannot contain commands or multiple expressions.

- An anonymous function cannot be a direct call to print because **lambda** requires an expression.

- **`lambda`** functions have their own local namespace and cannot access variables other than those in their parameter list and those in the global namespace.

- Although it appears that lambdas are a one-line version of a function, they are not equivalent to inline statements in C or C++, whose purpose is to stack allocation by passing function, during invocation for performance reasons.

**Syntax:**

```python
lambda argument_list: expression
```


Let’s see an example to print even numbers without a **`lambda`** function and with a **`lambda`** function. See the difference in line of code as well as readability of code.


````python
# Example 1: Program for even numbers without lambda function

def even_numbers(nums):
    even_list = []
    for n in nums:
        if n % 2 == 0:
            even_list.append(n)
    return even_list

num_list = [10, 9, 16, 78, 2, 3, 7, 1]
ans = even_numbers(num_list)
print("Even numbers are:", ans)
````

````python
# Example 1: Program for even number with a lambda function

l = [10, 9, 16, 78, 2, 3, 7, 1]
even_nos = list(filter(lambda x: x % 2 == 0, l))
print("Even numbers are: ", even_nos)
````

````python
# Example 2: Program to show the use of lambda functions

double = lambda x: x * 2

print(double(6))
````

**Explanation:**

In the above program, lambda **`x: x * 2`** is the lambda function. Here **`x`** is the argument and **`x * 2`** is the expression that gets evaluated and returned.

This function has no name. It returns a function object which is assigned to the identifier **`double`**. We can now call it as a normal function. The statement

```python
>>> double = lambda x: x * 2
```

is nearly the same as:

```python
>>> def double(x):
>>>     return x * 2
```


````python
# Example 3: Normal Function definition is here

def square(x):
    return x*x

# anonymous function
sqr = lambda x: x*x

#Calling square function
print("Square of number is",square(10)) #call normal function
print("Square of number is",sqr(2)) #call anonymous function
````

## Use of **`lambda`** Function in python

We use **`lambda`** function when we require a nameless function for a short period of time.

In Python, we generally use it as an argument to a higher-order function (a function that takes in other functions as **[arguments](https://github.com/milaan9/04_Python_Functions/blob/main/004_Python_Function_Arguments.ipynb)**). **`lambda`** function are used along with built-in functions like **`filter()`**, **`map()`**, **`reduce()`** etc.


### `lambda` function use with `filter()`

The **`filter()`** function in Python takes in a function and a list as arguments.

The function is called with all the items in the list and a new list is returned which contains items for which the function evaluates to **`True`**.

Here is an example use of **`filter()`** function to filter out only even numbers from a list.


````python
# Example 1: Program to filter out only the even items from a list

my_list = [1, 5, 4, 6, 8, 11, 3, 12]  # total 8 elements

new_list = list(filter(lambda x: (x%2 == 0), my_list)) # returns the output in form of a list

print("Even numbers are: ", new_list)
````

````python
# Example 2: 

l = [-10, 5, 12, -78, 6, -1, -7, 9]
positive_nos = list(filter(lambda x: x > 0, l))
print("Positive numbers are: ", positive_nos)
````

### `lambda` function with `map()`

The **`map()`** function in Python takes in a function and a list.

The function is called with all the items in the list and a new list is returned which contains items returned by that function for each item.

Here is an example use of **`map()`** function to double all the items in a list.


````python
# Example 1: Program to double each item in a list using map()

my_list = [1, 5, 4, 6, 8, 11, 3, 12]  # total 8 elements
new_list = list(map(lambda x: x * 2, my_list)) # returns the output in form of a list
print("Double values are: ", new_list)
````

````python
# Example 2:

list1 = [2, 3, 4, 8, 9]
list2 = list(map(lambda x: x*x*x, list1))
print("Cube values are:", list2)
````

### `lambda` function with `reduce()`

The **`reduce()`** function is used to minimize sequence elements into a single value by applying the specified condition.

The **`reduce()`** function is present in the **`functools`** module; hence, we need to import it using the import statement before using it.


````python
# Example 1:

from functools import reduce
list1 = [20, 13, 4, 8, 9]
add = reduce(lambda x, y: x+y, list1)
print("Addition of all list elements is : ", add)
````

### Lambda Expressions in Python

A **lambda expression** is a compact and anonymous function that is defined using the keyword `lambda`. It is a concise way to write small functions, and is often used for functions that are simple and can be defined in a single line.

### Syntax:

The general syntax of a lambda function is:

```python
lambda arguments: expression
```

Where:

- **`lambda`**: The keyword to define an anonymous function.
- **`arguments`**: The parameters that are passed to the lambda function (can be multiple or none).
- **`expression`**: The expression that is evaluated and returned when the lambda function is called.

### Examples of Lambda Expressions:

#### 1. Simple Lambda with One Argument

```python
# Lambda function that squares a number
square = lambda x: x ** 2
print(square(5))  # Output: 25
```

#### 2. Lambda with Multiple Arguments

```python
# Lambda function that adds two numbers
add = lambda x, y: x + y
print(add(3, 4))  # Output: 7
```

#### 3. Lambda with No Arguments

```python
# Lambda function that returns a constant value
get_five = lambda: 5
print(get_five())  # Output: 5
```

#### 4. Lambda Inside a Function (Higher-order function)

```python
# Using lambda inside another function
def operate_on_two_numbers(x, y, func):
    return func(x, y)

result = operate_on_two_numbers(5, 3, lambda x, y: x * y)
print(result)  # Output: 15 (multiplication)
```

#### 5. Lambda with Conditional Expression (Ternary Operator)

```python
# Lambda with conditional expression
is_even = lambda x: "Even" if x % 2 == 0 else "Odd"
print(is_even(10))  # Output: Even
print(is_even(7))   # Output: Odd
```

#### 6. Lambda with Default Arguments

```python
# Lambda with default argument values
multiply = lambda x, y=2: x * y
print(multiply(3))    # Output: 6  (Uses the default value for y)
print(multiply(3, 5)) # Output: 15 (Uses provided value for y)
```

### Use Cases for Lambda Expressions:

1. **Sorting**: Lambda expressions are often used with `sorted()` and `sort()` functions.

   ```python
   # Sort a list of tuples based on the second element
   pairs = [(1, 2), (3, 4), (2, 1)]
   pairs_sorted = sorted(pairs, key=lambda x: x[1])
   print(pairs_sorted)  # Output: [(2, 1), (1, 2), (3, 4)]
   ```

2. **Filtering**: You can use `filter()` to apply a lambda function for filtering elements from a collection.

   ```python
   # Filter even numbers from a list
   numbers = [1, 2, 3, 4, 5, 6]
   even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
   print(even_numbers)  # Output: [2, 4, 6]
   ```

3. **Mapping**: Lambda expressions are commonly used with `map()` to apply a function to each item in a collection.

   ```python
   # Square each number in a list
   numbers = [1, 2, 3, 4]
   squares = list(map(lambda x: x ** 2, numbers))
   print(squares)  # Output: [1, 4, 9, 16]
   ```

4. **Reducing**: You can use lambda with `reduce()` (from the `functools` module) to apply a function cumulatively to the items in an iterable.

   ```python
   from functools import reduce
   # Calculate the product of all elements in the list
   numbers = [1, 2, 3, 4]
   product = reduce(lambda x, y: x * y, numbers)
   print(product)  # Output: 24
   ```

### Lambda Functions vs Regular Functions

- **Anonymous**: Lambda functions are anonymous, meaning they are not assigned a name (unless you explicitly assign them).
- **Single Expression**: A lambda function can only contain a single expression. It can't have multiple statements or logic like a regular function.
- **Usefulness**: Lambda functions are useful for simple operations that are needed temporarily, such as when you want to pass a small function as an argument.

### Advantages of Lambda Functions:

1. **Concise**: Lambda functions provide a more concise way to write small functions.
2. **Inline Use**: Lambda functions are used frequently as an argument to higher-order functions like `map()`, `filter()`, and `sorted()`.

### Limitations of Lambda Functions:

1. **Single Expression**: A lambda function can only contain one expression, making it less flexible than regular functions.
2. **Readability**: Overusing lambda functions for complex operations can hurt the readability of code, especially for beginners.

### When to Use Lambda Functions:

1. When you need a short, throwaway function that is used once or twice.
2. When passing a simple function as an argument to functions like `sorted()`, `map()`, `filter()`, etc.
3. When you want to use a function on the fly without having to define it with `def`.

### Conclusion:

Lambda expressions in Python provide a quick and compact way to define simple, one-line functions. They are highly useful in functional programming styles and commonly used in higher-order functions. However, their use should be balanced, as they are only suitable for simple operations and can reduce code readability in more complex scenarios.


Lambda expressions in Python are a concise way to create small, anonymous functions. They are often used for short, throwaway functions that are not needed elsewhere in the code. Lambda expressions are limited to a single expression and cannot include statements or annotations. Below is a comprehensive guide to lambda expressions, from basic to advanced usage.

---

## **1. Basic Syntax**

### **1.1 Simple Lambda Expression**

- **Purpose**: Creates a small, anonymous function.
- **Syntax**: `lambda arguments: expression`
- **Example**:
  ```python
  add = lambda x, y: x + y
  print(add(3, 5))  # Output: 8
  ```
  - The lambda expression `lambda x, y: x + y` takes two arguments `x` and `y` and returns their sum.

---

### **1.2 Lambda with No Arguments**

- **Purpose**: Creates a lambda function that takes no arguments.
- **Syntax**: `lambda: expression`
- **Example**:
  ```python
  greet = lambda: "Hello, World!"
  print(greet())  # Output: Hello, World!
  ```

---

### **1.3 Lambda with Default Arguments**

- **Purpose**: Uses default values for arguments.
- **Syntax**: `lambda arg=value: expression`
- **Example**:
  ```python
  multiply = lambda x, y=2: x * y
  print(multiply(3))  # Output: 6 (uses default y=2)
  print(multiply(3, 4))  # Output: 12
  ```

---

## **2. Intermediate Usage**

### **2.1 Lambda with Conditional Logic**

- **Purpose**: Uses conditional expressions (`if-else`) inside a lambda.
- **Syntax**: `lambda args: value_if_true if condition else value_if_false`
- **Example**:
  ```python
  is_even = lambda x: "Even" if x % 2 == 0 else "Odd"
  print(is_even(4))  # Output: Even
  print(is_even(5))  # Output: Odd
  ```

---

### **2.2 Lambda with Multiple Arguments**

- **Purpose**: Takes multiple arguments and processes them.
- **Syntax**: `lambda arg1, arg2, ...: expression`
- **Example**:
  ```python
  power = lambda x, y: x ** y
  print(power(2, 3))  # Output: 8
  ```

---

### **2.3 Lambda with `map()`**

- **Purpose**: Applies a lambda function to each item in an iterable.
- **Syntax**: `map(lambda args: expression, iterable)`
- **Example**:
  ```python
  numbers = [1, 2, 3, 4]
  squared = map(lambda x: x ** 2, numbers)
  print(list(squared))  # Output: [1, 4, 9, 16]
  ```

---

### **2.4 Lambda with `filter()`**

- **Purpose**: Filters items in an iterable based on a condition.
- **Syntax**: `filter(lambda args: condition, iterable)`
- **Example**:
  ```python
  numbers = [1, 2, 3, 4, 5, 6]
  evens = filter(lambda x: x % 2 == 0, numbers)
  print(list(evens))  # Output: [2, 4, 6]
  ```

---

### **2.5 Lambda with `sorted()`**

- **Purpose**: Sorts an iterable using a lambda as the key function.
- **Syntax**: `sorted(iterable, key=lambda args: expression)`
- **Example**:
  ```python
  names = ["Alice", "Bob", "Charlie", "David"]
  sorted_names = sorted(names, key=lambda x: len(x))
  print(sorted_names)  # Output: ['Bob', 'Alice', 'David', 'Charlie']
  ```

---

## **3. Advanced Usage**

### **3.1 Lambda with `reduce()`**

- **Purpose**: Applies a lambda function cumulatively to items in an iterable.
- **Syntax**: `functools.reduce(lambda arg1, arg2: expression, iterable)`
- **Example**:

  ```python
  from functools import reduce

  numbers = [1, 2, 3, 4]
  product = reduce(lambda x, y: x * y, numbers)
  print(product)  # Output: 24
  ```

---

### **3.2 Lambda with Nested Functions**

- **Purpose**: Uses lambda inside another function.
- **Syntax**: Define a function that returns a lambda.
- **Example**:

  ```python
  def multiplier(n):
      return lambda x: x * n

  double = multiplier(2)
  print(double(5))  # Output: 10
  ```

---

### **3.3 Lambda with `itertools`**

- **Purpose**: Combines lambda with `itertools` functions for advanced iteration.
- **Syntax**: Use `itertools` functions like `takewhile`, `dropwhile`, etc.
- **Example**:

  ```python
  import itertools

  numbers = [1, 2, 3, 4, 5]
  filtered = itertools.takewhile(lambda x: x < 4, numbers)
  print(list(filtered))  # Output: [1, 2, 3]
  ```

---

### **3.4 Lambda with `collections.defaultdict`**

- **Purpose**: Uses lambda to provide a default value for `defaultdict`.
- **Syntax**: `collections.defaultdict(lambda: default_value)`
- **Example**:

  ```python
  from collections import defaultdict

  counter = defaultdict(lambda: 0)
  counter["a"] += 1
  print(counter["a"])  # Output: 1
  print(counter["b"])  # Output: 0 (default value)
  ```

---

### **3.5 Lambda with `functools.partial`**

- **Purpose**: Creates a new function with some arguments pre-filled.
- **Syntax**: `functools.partial(lambda args: expression, *args, **kwargs)`
- **Example**:

  ```python
  from functools import partial

  add = lambda x, y: x + y
  add_five = partial(add, y=5)
  print(add_five(3))  # Output: 8
  ```

---

## **4. Practical Use Cases**

- **Short Functions**: Use lambda for simple, one-off functions.
- **Functional Programming**: Combine with `map`, `filter`, and `reduce`.
- **Sorting and Key Functions**: Use as a key function in `sorted` or `max`/`min`.
- **Callbacks**: Pass lambda as a callback function to other functions.

---

## **5. Limitations of Lambda Expressions**

- **Single Expression**: Lambda can only contain a single expression.
- **No Statements**: Cannot include statements like `if`, `for`, or `while`.
- **No Annotations**: Cannot use type annotations.
- **Readability**: Overuse of lambda can reduce code readability.

---

## **6. Summary**

Lambda expressions are a powerful and concise way to create small, anonymous functions in Python. They are widely used in functional programming, sorting, filtering, and other scenarios where a short function is needed. However, they should be used judiciously to maintain code readability.
