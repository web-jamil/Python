# Python `float()`

The **`float()`** method returns a floating point number from a number or a string.

**Syntax**:

```python
float([x])
```


## `float()` Parameters

The **`float()`** method takes a single parameter:

- **`x` (Optional)** - number or string that needs to be converted to floating point number
  If it's a string, the string should contain decimal points

**Different parameters with **`float()`\*\*\*\*:

| Parameter Type   | Usage                                                                                                                                                                                |
| :--------------- | :----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Float number** | **Use as a floating number**                                                                                                                                                         |
| **Integer**      | **Use as an integer**                                                                                                                                                                |
| **String**       | **Must contain decimal numbers. Leading and trailing whitespaces are removed. Optional use of "`+`", "`-`" signs. Could contain `NaN`, `Infinity`, `inf` (lowercase or uppercase).** |


## Return Value from `float()`

**`float()`** method returns:

- Equivalent floating point number if an argument is passed
- 0.0 if no arguments passed
- **`OverflowError`** exception if the argument is outside the range of Python float


`````python
# Example 1: How float() works in Python?

# for integers
print(float(10))

# for floats
print(float(11.22))

# for string floats
print(float("-13.33"))

# for string floats with whitespaces
print(float("     -24.45\n"))

# string float error
print(float("abc"))
`````

`````python
# Example 2: float() for infinity and Nan(Not a number)?

# for NaN
print(float("nan"))
print(float("NaN"))

# for inf/infinity
print(float("inf"))
print(float("InF"))
print(float("InFiNiTy"))
print(float("infinity"))
`````

Floating-point numbers in Python (and in most programming languages) are represented using the **IEEE 754 standard**, which can introduce certain types of errors due to the way floating-point arithmetic is handled. These errors arise because floating-point numbers are stored in a finite number of bits, leading to precision limitations. Below is a comprehensive guide to **all types of errors in floating-point numbers** in Python, including explanations and examples.

---

## **1. Representation Error**

- **Cause**: Floating-point numbers are stored in binary, and some decimal numbers cannot be represented exactly in binary.
- **Example**:
  ```python
  num = 0.1 + 0.2
  print(num)  # Output: 0.30000000000000004 (not exactly 0.3)
  ```
- **Explanation**: The decimal number `0.1` cannot be represented exactly in binary, leading to a small error in the result.

---

## **2. Rounding Error**

- **Cause**: Floating-point numbers have limited precision, so values are rounded to fit into the available bits.
- **Example**:
  ```python
  num = 1.0 / 3.0
  print(num)  # Output: 0.3333333333333333 (not exactly 1/3)
  ```
- **Explanation**: The result is rounded to fit into the 64-bit floating-point representation.

---

## **3. Overflow Error**

- **Cause**: A floating-point number exceeds the maximum value that can be represented.
- **Example**:
  ```python
  num = 1.8e308  # Maximum value for a 64-bit float
  print(num * 10)  # Output: inf (infinity)
  ```
- **Explanation**: The result exceeds the range of representable floating-point numbers, resulting in `inf` (infinity).

---

## **4. Underflow Error**

- **Cause**: A floating-point number is smaller than the minimum value that can be represented.
- **Example**:
  ```python
  num = 1e-324  # Minimum positive value for a 64-bit float
  print(num / 10)  # Output: 0.0 (underflow)
  ```
- **Explanation**: The result is too small to be represented, so it is rounded to `0.0`.

---

## **5. Loss of Significance (Catastrophic Cancellation)**

- **Cause**: Subtracting two nearly equal floating-point numbers results in a loss of significant digits.
- **Example**:
  ```python
  a = 1.23456789
  b = 1.23456780
  result = a - b
  print(result)  # Output: 8.999999999621423e-09 (loss of precision)
  ```
- **Explanation**: The result has fewer significant digits than the original numbers.

---

## **6. Accumulation of Errors**

- **Cause**: Repeated arithmetic operations can accumulate small errors, leading to significant inaccuracies.
- **Example**:
  ```python
  total = 0.0
  for _ in range(1000000):
      total += 0.1
  print(total)  # Output: 100000.00000133288 (not exactly 100000.0)
  ```
- **Explanation**: The small error in representing `0.1` accumulates over many iterations.

---

## **7. Comparison Errors**

- **Cause**: Floating-point numbers should not be compared directly due to precision issues.
- **Example**:
  ```python
  a = 0.1 + 0.2
  b = 0.3
  print(a == b)  # Output: False (due to representation error)
  ```
- **Fix**: Use a tolerance level for comparisons:
  ```python
  tolerance = 1e-9
  print(abs(a - b) < tolerance)  # Output: True
  ```

---

## **8. Denormal Numbers**

- **Cause**: Very small floating-point numbers are represented as denormal numbers, which have reduced precision.
- **Example**:
  ```python
  num = 1e-323  # Denormal number
  print(num)  # Output: 1e-323 (very small, but not zero)
  ```
- **Explanation**: Denormal numbers allow representation of very small values but with reduced precision.

---

## **9. NaN (Not a Number)**

- **Cause**: Invalid arithmetic operations (e.g., `0/0` or `sqrt(-1)`) result in `NaN`.
- **Example**:
  ```python
  import math
  result = math.sqrt(-1)
  print(result)  # Output: nan
  ```
- **Explanation**: `NaN` represents an undefined or unrepresentable value.

---

## **10. Infinity**

- **Cause**: Arithmetic operations that exceed the range of representable floating-point numbers result in `inf` or `-inf`.
- **Example**:
  ```python
  num = 1e308
  print(num * 10)  # Output: inf
  ```
- **Explanation**: The result exceeds the maximum representable value, so it is represented as `inf`.

---

## **11. Machine Epsilon**

- **Cause**: The smallest number that, when added to `1.0`, produces a result different from `1.0`. It represents the precision of floating-point numbers.
- **Example**:
  ```python
  import sys
  epsilon = sys.float_info.epsilon
  print(epsilon)  # Output: 2.220446049250313e-16 (for 64-bit floats)
  ```
- **Explanation**: Machine epsilon defines the smallest representable difference between two floating-point numbers.

---

## **12. Guard Digits**

- **Cause**: Intermediate calculations use extra digits (guard digits) to reduce rounding errors.
- **Example**:
  ```python
  a = 1.23456789
  b = 1.23456780
  result = a - b
  print(result)  # Output: 8.999999999621423e-09 (guard digits reduce error)
  ```
- **Explanation**: Guard digits improve the accuracy of intermediate results.

---

## **13. Double Rounding**

- **Cause**: Rounding a result twice can introduce additional errors.
- **Example**:
  ```python
  num = 1.005
  rounded = round(num, 2)  # First rounding: 1.00
  print(rounded)  # Output: 1.0 (double rounding error)
  ```
- **Explanation**: Double rounding can lead to unexpected results.

---

## **14. Non-Associativity of Floating-Point Arithmetic**

- **Cause**: Floating-point addition and multiplication are not associative due to rounding errors.
- **Example**:
  ```python
  a = 1e16
  b = -1e16
  c = 1.0
  result1 = (a + b) + c
  result2 = a + (b + c)
  print(result1, result2)  # Output: 1.0 0.0 (different results)
  ```
- **Explanation**: The order of operations affects the result due to rounding errors.

---

## **15. Subnormal Numbers**

- **Cause**: Very small numbers are represented as subnormal numbers, which have reduced precision.
- **Example**:
  ```python
  num = 1e-324  # Subnormal number
  print(num)  # Output: 0.0 (underflow)
  ```
- **Explanation**: Subnormal numbers allow representation of very small values but with reduced precision.

---

Floating-point errors are inherent in the way floating-point numbers are represented and manipulated in computers. Common errors include:

- **Representation errors**: Due to binary representation limitations.
- **Rounding errors**: Due to finite precision.
- **Overflow/underflow**: Due to exceeding the range of representable values.
- **Loss of significance**: Due to subtracting nearly equal numbers.
- **Accumulation of errors**: Due to repeated arithmetic operations.

To mitigate these errors:

- Use **tolerances** for comparisons.
- Avoid **direct equality checks** for floating-point numbers.
- Use libraries like `decimal` or `fractions` for precise arithmetic.

**Representation error** is one of the most fundamental issues in floating-point arithmetic. It occurs because floating-point numbers are stored in a finite number of bits, and not all real numbers can be represented exactly in this format. This error is inherent in the way computers handle floating-point numbers and is a direct consequence of the **IEEE 754 standard**, which defines how floating-point numbers are stored and manipulated.

Below is a comprehensive guide to **representation error**, covering everything from basic concepts to advanced topics, including examples and explanations.

---

## **1. What is Representation Error?**

- **Definition**: Representation error occurs when a real number cannot be represented exactly in the floating-point format used by computers.
- **Cause**: Floating-point numbers are stored in a finite number of bits (e.g., 32-bit or 64-bit), leading to approximations of real numbers.
- **Example**:
  ```python
  num = 0.1
  print(num)  # Output: 0.1 (but internally, it's not exactly 0.1)
  ```

---

## **2. Why Does Representation Error Occur?**

- **Binary Representation**: Computers store numbers in binary (base-2), but many decimal (base-10) numbers cannot be represented exactly in binary.
- **Finite Precision**: Floating-point numbers have a limited number of bits for the mantissa (significand), which limits their precision.
- **Example**:
  ```python
  # Decimal 0.1 in binary is an infinite repeating fraction: 0.0001100110011...
  # It cannot be represented exactly in a finite number of bits.
  print(0.1 + 0.2)  # Output: 0.30000000000000004 (not exactly 0.3)
  ```

---

**NaN** (Not a Number) is a special floating-point value used to represent undefined or unrepresentable numerical results, such as the result of dividing zero by zero or taking the square root of a negative number. In Python, `NaN` is defined in the `math` and `numpy` modules and is part of the IEEE 754 floating-point standard. Below is a comprehensive guide to **NaN** in Python, covering everything from basic concepts to advanced topics, including examples and explanations.

---

## **1. What is NaN?**

- **Definition**: NaN stands for "Not a Number" and is used to represent undefined or unrepresentable numerical results.
- **Cause**: NaN can result from operations like `0/0`, `inf - inf`, or `sqrt(-1)`.
- **Example**:
  ```python
  import math
  result = math.sqrt(-1)
  print(result)  # Output: nan
  ```

---

## **2. Why Does NaN Occur?**

- **Undefined Operations**: Operations like `0/0` or `inf - inf` are mathematically undefined.
- **Invalid Input**: Functions like `sqrt(-1)` or `log(-1)` produce NaN because the input is invalid.
- **Example**:
  ```python
  import math
  result = math.log(-1)
  print(result)  # Output: nan
  ```
