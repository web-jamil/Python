# Python User-defined Functions

In this class, you will find the advantages of using user-defined functions and best practices to follow.


## What are user-defined functions in Python?

Functions that we define ourselves to do certain specific task are referred as user-defined functions. The way in which we define and call **[functions in Python](https://github.com/milaan9/04_Python_Functions/blob/main/001_Python_Functions.ipynb)** are already discussed.

Functions that readily come with Python are called built-in functions. If we use functions written by others in the form of library, it can be termed as library functions.

All the other functions that we write on our own fall under user-defined functions. So, our user-defined function could be a library function to someone else.


## Advantages of user-defined functions

1. User-defined functions help to decompose a large program into small segments which makes program easy to understand, maintain and debug.
2. If repeated code occurs in a program. Function can be used to include those codes and execute when needed by calling that function.
3. Programmars working on large project can divide the workload by making different functions.


````python
# Example 1: illustrate the use of user-defined functions

num1 = input('Enter first number: ')
num2 = input('Enter second number: ')

def my_addition(num1,num2):
    sum = float(num1) + float(num2)
    return sum

#num1 = 9.3
#num2 = 6.9

print("The sum is", my_addition(num1, num2))
````

**Explanation:**

Here, we have defined the function **`my_addition()`** which adds two numbers and returns the result.

This is our user-defined function. We could have multiplied the two numbers inside our function (it's all up to us). But this operation would not be consistent with the name of the function. It would create ambiguity.

It is always a good idea to name functions according to the task they perform.

In the above example, **[print()](https://github.com/milaan9/04_Python_Functions/blob/main/002_Python_Functions_Built_in/051_Python_print%28%29.ipynb)** is a **[built-in function in Python](https://github.com/milaan9/04_Python_Functions/tree/main/002_Python_Functions_Built_in)**.


## User-Defined Functions in Python: From Basic to Advanced

In Python, functions allow you to group a set of related statements together so that they can be executed multiple times, simplifying your code and improving its reusability. **User-defined functions** are functions that you create, as opposed to built-in functions provided by Python.

### **1. Basic Concepts of Functions**

#### 1.1 What is a Function?

A function is a block of code that only runs when it is called. It can take inputs, perform some operations, and return a value.

#### 1.2 Defining a Function

To define a function in Python, you use the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`. The function body follows, indented.

```python
# Basic function definition
def greet():
    print("Hello, World!")
```

#### 1.3 Calling a Function

Once a function is defined, you can call it by using its name followed by parentheses.

```python
greet()  # Output: Hello, World!
```

---

### **2. Parameters and Arguments**

#### 2.1 Function Parameters

Parameters are the values you pass into a function when you call it. They are defined within the parentheses when you define the function.

```python
def greet(name):
    print(f"Hello, {name}!")
```

#### 2.2 Function Arguments

Arguments are the actual values passed to the function when it is called.

```python
greet("Alice")  # Output: Hello, Alice!
```

#### 2.3 Multiple Parameters

You can define multiple parameters in a function.

```python
def add(a, b):
    return a + b

result = add(3, 5)  # Output: 8
```

---

### **3. Return Statement**

The `return` statement allows you to send a result back from a function.

#### 3.1 Returning a Value

```python
def multiply(x, y):
    return x * y

result = multiply(2, 3)  # Output: 6
```

#### 3.2 Returning Multiple Values

You can return multiple values from a function by separating them with commas. These values will be returned as a tuple.

```python
def swap(a, b):
    return b, a

result = swap(5, 10)  # Output: (10, 5)
```

#### 3.3 Returning None

If you don’t use the `return` statement, the function returns `None` by default.

```python
def no_return():
    print("This function doesn't return anything.")

result = no_return()  # Output: This function doesn't return anything.
print(result)  # Output: None
```

---

### **4. Default Parameters**

You can assign default values to parameters. If the caller doesn't provide a value for the parameter, the default value will be used.

```python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet("Alice")  # Output: Hello, Alice!
greet()  # Output: Hello, Guest!
```

---

### **5. Keyword Arguments**

Keyword arguments allow you to pass arguments to a function by explicitly naming the parameter.

```python
def greet(name, age):
    print(f"Hello, {name}. You are {age} years old.")

greet(name="Alice", age=25)  # Output: Hello, Alice. You are 25 years old.
```

---

### **6. Variable-Length Arguments**

Sometimes you don't know beforehand how many arguments a function should accept. Python allows you to handle such cases using variable-length arguments.

#### 6.1 \*args (Non-keyword Variable Arguments)

`*args` allows a function to accept an arbitrary number of positional arguments. It collects them as a tuple.

```python
def add(*numbers):
    return sum(numbers)

print(add(1, 2, 3))  # Output: 6
print(add(5, 10, 15, 20))  # Output: 50
```

#### 6.2 \*\*kwargs (Keyword Variable Arguments)

`**kwargs` allows a function to accept an arbitrary number of keyword arguments. It collects them as a dictionary.

```python
def person_info(**info):
    for key, value in info.items():
        print(f"{key}: {value}")

person_info(name="Alice", age=25, city="New York")
# Output:
# name: Alice
# age: 25
# city: New York
```

#### 6.3 Mixing \*args and \*\*kwargs

You can use both `*args` and `**kwargs` in the same function, but `*args` should come before `**kwargs` in the function definition.

```python
def example(arg1, *args, kwarg1=None, **kwargs):
    print(f"arg1: {arg1}")
    print(f"args: {args}")
    print(f"kwarg1: {kwarg1}")
    print(f"kwargs: {kwargs}")

example(1, 2, 3, 4, kwarg1='Hello', key1='value1', key2='value2')
```

---

### **7. Lambda Functions (Anonymous Functions)**

Lambda functions are short, throwaway functions that are defined using the `lambda` keyword. They can take any number of arguments but can only have one expression.

#### Basic Syntax:

```python
lambda arguments: expression
```

#### Example:

```python
multiply = lambda x, y: x * y
print(multiply(3, 4))  # Output: 12
```

Lambda functions are often used for simple operations or as arguments to higher-order functions like `map()`, `filter()`, and `sorted()`.

---

### **8. Recursion**

Recursion is a technique where a function calls itself to solve a problem. It must have a base case to stop the recursion.

#### Example (Factorial Function):

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))  # Output: 120
```

#### Important Considerations with Recursion:

- **Base Case**: Always define a base case to avoid infinite recursion.
- **Performance**: Recursion can be less efficient and might cause a stack overflow for very deep recursive calls.

---

### **9. Function Scope and Lifetime**

#### 9.1 Local and Global Variables

- **Local Variables**: Variables defined inside a function are local to that function and cannot be accessed outside.
- **Global Variables**: Variables defined outside any function are global and can be accessed from anywhere in the program.

```python
x = 10  # Global variable

def my_function():
    x = 5  # Local variable
    print(f"Local x: {x}")

my_function()  # Output: Local x: 5
print(f"Global x: {x}")  # Output: Global x: 10
```

#### 9.2 `global` Keyword

You can modify global variables inside a function using the `global` keyword.

```python
x = 10

def modify_global():
    global x
    x = 20

modify_global()
print(x)  # Output: 20
```

#### 9.3 `nonlocal` Keyword

The `nonlocal` keyword allows you to modify a variable from the nearest enclosing scope that is not global.

```python
def outer_function():
    x = 10
    def inner_function():
        nonlocal x
        x = 20
    inner_function()
    print(x)

outer_function()  # Output: 20
```

---

### **10. Advanced Concepts**

#### 10.1 Function Annotations

Function annotations provide a way to attach metadata to functions, like type hints.

```python
def add(a: int, b: int) -> int:
    return a + b

print(add(3, 5))  # Output: 8
```

#### 10.2 Closures

A closure occurs when a function captures the variables from its enclosing scope. It allows a function to remember the environment in which it was created.

```python
def outer():
    x = 10
    def inner():
        print(x)
    return inner

closure = outer()
closure()  # Output: 10
```

#### 10.3 Decorators

A decorator is a function that modifies the behavior of another function. It’s often used to add functionality to existing functions without changing their code.

```python
def decorator_function(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@decorator_function
def say_hello():
    print("Hello!")

say_hello()
# Output:
# Before function call
# Hello!
# After function call
```

### **11. Summary**

- **User-defined functions** allow you to group related code, improving reusability and maintainability.
- **Parameters** are used to pass values into functions, while **arguments** are the actual values passed.
- Functions can return values using `return`.
- **Default parameters** allow you to set default values if arguments are not provided.
- **Variable-length arguments** like `*args` and `**kwargs` allow you to pass an arbitrary number of arguments to functions.
- **Lambda functions** are short, anonymous functions for simple tasks.
- **Recursion** is a technique where functions call themselves to solve problems.
- **Function scope** determines where variables can be accessed and modified.
- Advanced concepts like **closures**, **decorators**, and **function annotations** provide powerful tools for writing flexible and efficient code.

Mastering user-defined functions will allow you to write clean, modular, and maintainable Python code.

```python
# ❌ Bad - list is created once at definition time
def add_item(item, lst=[]):
    lst.append(item)
    return lst
```

## **9. Function Decorators**

Modify function behavior:

```python
def log_time(func):
    """Decorator to log execution time"""
    import time
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end-start:.2f}s")
        return result
    return wrapper

@log_time
def slow_function():
    time.sleep(1)

slow_function()  # Output: slow_function took 1.00s
```

## **11. Generators**

Functions that yield values:

```python
def count_up_to(max):
    """Generate numbers up to max"""
    count = 1
    while count <= max:
        yield count
        count += 1

counter = count_up_to(5)
print(list(counter))  # [1, 2, 3, 4, 5]
```

## **16. Best Practices**

1. **Single Responsibility Principle**: Each function should do one thing
2. **Descriptive names**: Use verbs for functions (`calculate_total()`)
3. **Limit parameters**: 3-4 max (use objects if more needed)
4. **Pure functions**: Avoid side effects when possible
5. **Proper documentation**: Always write docstrings
6. **Type hints**: Improve code readability and IDE support
7. **Keep functions short**: Typically <20 lines

**Example of well-designed function:**

```python
def calculate_discounted_price(price: float, discount: float = 0.0) -> float:
    """
    Calculate final price after applying discount

    Args:
        price: Original price (must be positive)
        discount: Percentage discount (0-100)

    Returns:
        Final price after discount

    Raises:
        ValueError: If price is negative or discount invalid
    """
    if price < 0:
        raise ValueError("Price cannot be negative")
    if not 0 <= discount <= 100:
        raise ValueError("Discount must be 0-100")

    return price * (1 - discount/100)
```
