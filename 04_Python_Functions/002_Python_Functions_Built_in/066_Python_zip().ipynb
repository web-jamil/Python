# Python `zip()`

The **`zip()`** function takes iterables (can be zero or more), aggregates them in a tuple, and return it.

**Syntax**:

```python
zip(*iterables)
```


## `zip()` Parameter

| Parameter       | Description                                                                         |
| :-------------- | :---------------------------------------------------------------------------------- |
| **`iterables`** | **can be built-in iterables (like: list, string, dict), or user-defined iterables** |

> Recommended Reading: **[Python Iterators, **iter** and **next**](https://github.com/milaan9/07_Python_Advanced_Topics/blob/main/001_Python_Iterators.ipynb)**


## Return Value from `zip()`

The **`zip()`** function returns an iterator of tuples based on the iterable objects.

- If we do not pass any parameter, **`zip()`** returns an empty iterator
- If a single iterable is passed, **`zip()`** returns an iterator of tuples with each tuple having only one element.
- If multiple iterables are passed, **`zip()`** returns an iterator of tuples with each tuple having elements from all the iterables.

Suppose, two iterables are passed to **`zip()`**; one iterable containing three and other containing five elements. Then, the returned iterator will contain three tuples. It's because iterator stops when the shortest iterable is exhausted.


````python
# Example 1: Python zip()

number_list = [1, 2, 3]
str_list = ['one', 'two', 'three']

# No iterables are passed
result = zip()

# Converting itertor to list
result_list = list(result)
print(result_list)

# Two iterables are passed
result = zip(number_list, str_list)

# Converting itertor to set
result_set = set(result)
print(result_set)
````

````python
# Example 2: Different number of iterable elements

numbersList = [1, 2, 3]
str_list = ['one', 'two']
numbers_tuple = ('ONE', 'TWO', 'THREE', 'FOUR')

# Notice, the size of numbersList and numbers_tuple is different
result = zip(numbersList, numbers_tuple)

# Converting to set
result_set = set(result)
print(result_set)

result = zip(numbersList, str_list, numbers_tuple)

# Converting to set
result_set = set(result)
print(result_set)
````

The **`*`** operator can be used in conjunction with **`zip()`** to unzip the list.

```python
zip(*zippedList)
```


````python
# Example 3: Unzipping the Value Using zip()

coordinate = ['x', 'y', 'z']
value = [3, 4, 5]

result = zip(coordinate, value)
result_list = list(result)
print(result_list)

c, v =  zip(*result_list)
print('c =', c)
print('v =', v)
````

The `zip()` function in Python is a built-in function that is used to **combine multiple iterables** (like lists, tuples, etc.) into a single iterable of tuples. Each tuple contains elements from the input iterables that are at the same position. This is particularly useful for parallel iteration and pairing data.

## **1. Basics of `zip()`**

### **Combining Iterables**

The `zip()` function takes one or more iterables as arguments and returns an iterator of tuples.

```python
# Example 1: Combining two lists
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']
zipped = zip(list1, list2)
print(list(zipped))  # Output: [(1, 'a'), (2, 'b'), (3, 'c')]

# Example 2: Combining three iterables
tuple1 = (1, 2, 3)
tuple2 = ('a', 'b', 'c')
set1 = {10, 20, 30}
zipped = zip(tuple1, tuple2, set1)
print(list(zipped))  # Output: [(1, 'a', 10), (2, 'b', 20), (3, 'c', 30)]
```

### **Working with Lists, Tuples, and Strings**

`zip()` works with any iterable, including strings.

```python
# Example 1: Combining lists and tuples
list1 = [1, 2, 3]
tuple1 = ('a', 'b', 'c')
zipped = zip(list1, tuple1)
print(list(zipped))  # Output: [(1, 'a'), (2, 'b'), (3, 'c')]

# Example 2: Combining strings
str1 = "hello"
str2 = "world"
zipped = zip(str1, str2)
print(list(zipped))  # Output: [('h', 'w'), ('e', 'o'), ('l', 'r'), ('l', 'l'), ('o', 'd')]
```

## **2. Advanced Usage of `zip()`**

### **Unzipping with `zip()`**

You can "unzip" a zipped object by using the `*` operator.

```python
zipped = [(1, 'a'), (2, 'b'), (3, 'c')]
list1, list2 = zip(*zipped)
print(list1)  # Output: (1, 2, 3)
print(list2)  # Output: ('a', 'b', 'c')
```

### **Handling Unequal-Length Iterables**

If the input iterables are of unequal length, `zip()` stops creating tuples when the shortest iterable is exhausted.

```python
list1 = [1, 2, 3]
list2 = ['a', 'b']
zipped = zip(list1, list2)
print(list(zipped))  # Output: [(1, 'a'), (2, 'b')]
```

To handle unequal-length iterables, you can use `itertools.zip_longest()` from the `itertools` module.

```python
from itertools import zip_longest

list1 = [1, 2, 3]
list2 = ['a', 'b']
zipped = zip_longest(list1, list2, fillvalue=None)
print(list(zipped))  # Output: [(1, 'a'), (2, 'b'), (3, None)]
```

### **Using `zip()` with Dictionaries**

You can use `zip()` to combine keys and values into a dictionary.

```python
keys = ['a', 'b', 'c']
values = [1, 2, 3]
my_dict = dict(zip(keys, values))
print(my_dict)  # Output: {'a': 1, 'b': 2, 'c': 3}
```

## **3. Practical Use Cases**

### **Parallel Iteration**

`zip()` is commonly used to iterate over multiple iterables in parallel.

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 90, 95]

for name, score in zip(names, scores):
    print(f"{name} scored {score}")

# Output:
# Alice scored 85
# Bob scored 90
# Charlie scored 95
```

### **Transposing Data**

You can use `zip()` to transpose a matrix (swap rows and columns).

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transposed = list(zip(*matrix))
print(transposed)

# Output:
# [(1, 4, 7), (2, 5, 8), (3, 6, 9)]
```

### **Creating Dictionaries**

`zip()` is often used to create dictionaries from two lists (keys and values).

```python
keys = ['name', 'age', 'city']
values = ['Alice', 25, 'New York']

my_dict = dict(zip(keys, values))
print(my_dict)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

## **4. Advanced Topics**

### **Combining `zip()` with `enumerate()`**

You can combine `zip()` with `enumerate()` to get both the index and the paired elements.

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 90, 95]

for index, (name, score) in enumerate(zip(names, scores)):
    print(f"{index + 1}. {name} scored {score}")

# Output:
# 1. Alice scored 85
# 2. Bob scored 90
# 3. Charlie scored 95
```

### **Using `zip()` with `*` (Unpacking Operator)**

You can use the `*` operator to unpack iterables into `zip()`.

```python
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']
list3 = [10, 20, 30]

zipped = zip(*[list1, list2, list3])
print(list(zipped))  # Output: [(1, 'a', 10), (2, 'b', 20), (3, 'c', 30)]
```

### **Performance Considerations**

- `zip()` is lazy and returns an iterator, so it is memory-efficient.
- For large datasets, prefer `zip()` over manually combining iterables.

````python
print("--- Python's Built-in zip() Function ---")
print("--------------------------------------\n")

# 1. Basic Usage: Combining two lists
print("1. Basic Usage: Combining two lists")
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 22]

# zip() returns an iterator of tuples
zipped_data = zip(names, ages)
print(f"Type of zipped_data: {type(zipped_data)}")

# To see the contents, you need to consume the iterator (e.g., convert to list)
list_of_tuples = list(zipped_data)
print(f"list(zip(names, ages)): {list_of_tuples}\n")

# Once consumed, the iterator is exhausted:
print(f"Trying to list again (will be empty): {list(zipped_data)}\n")

# 2. Combining more than two iterables
print("2. Combining more than two iterables")
colors = ['red', 'green', 'blue']
hex_codes = ['#FF0000', '#00FF00', '#0000FF']
rgb_values = [(255, 0, 0), (0, 255, 0), (0, 0, 255)]

for color, hex_code, rgb in zip(colors, hex_codes, rgb_values):
    print(f"Color: {color}, Hex: {hex_code}, RGB: {rgb}")
print("\n")

# 3. Handling Iterables of Different Lengths (Shortest Iterable Dictates Length)
print("3. Handling Iterables of Different Lengths")
short_list = [1, 2]
long_list = ['a', 'b', 'c', 'd']

# zip() will stop when the shortest iterable is exhausted
zipped_different_lengths = list(zip(short_list, long_list))
print(f"zip({short_list}, {long_list}): {zipped_different_lengths}\n")

# If you need to handle unequal lengths, consider `itertools.zip_longest`

# 4. Unzipping (Reverse of Zipping) using '*' operator
print("4. Unzipping (Reverse of Zipping)")
# This is a common and powerful use case for zip()
# 'stars' operator unpacks the list of tuples into separate arguments for zip()
students = [('Alice', 25), ('Bob', 30), ('Charlie', 22)]
names_unzipped, ages_unzipped = zip(*students)

print(f"Original students data: {students}")
print(f"Unzipped names: {list(names_unzipped)}")
print(f"Unzipped ages: {list(ages_unzipped)}\n")

# Unzipping with more than two columns
combined_data = [('red', '#FF0000', (255, 0, 0)),
                 ('green', '#00FF00', (0, 255, 0))]
unzipped_colors, unzipped_hex, unzipped_rgb = zip(*combined_data)
print(f"Unzipped colors: {list(unzipped_colors)}")
print(f"Unzipped hex codes: {list(unzipped_hex)}")
print(f"Unzipped RGB values: {list(unzipped_rgb)}\n")


# 5. Creating Dictionaries from Zipped Data
print("5. Creating Dictionaries from Zipped Data")
keys = ['name', 'age', 'city']
values = ['David', 40, 'London']

person_dict = dict(zip(keys, values))
print(f"Dictionary from zipped keys and values: {person_dict}\n")

# Example: Mapping student IDs to names
student_ids = [101, 102, 103]
student_names = ['Eve', 'Frank', 'Grace']
student_mapping = dict(zip(student_ids, student_names))
print(f"Student ID to name mapping: {student_mapping}\n")

# 6. Iterating with zip()
print("6. Iterating with zip()")
items = ['pen', 'notebook', 'eraser']
prices = [1.50, 5.00, 0.75]

print("Item prices:")
for item, price in zip(items, prices):
    print(f"{item}: ${price:.2f}")
print("\n")

# 7. Use Cases with Generator Expressions
print("7. Use Cases with Generator Expressions")
temperatures = [25, 27, 28, 26, 29] # Celsius
cities = ["Dhaka", "Chattogram", "Khulna", "Sylhet", "Rajshahi"]

# Display cities and their temperatures, filtering for warm cities
print("Warm cities (over 26°C):")
for city, temp in zip(cities, temperatures):
    if temp > 26:
        print(f"{city}: {temp}°C")
print("\n")

# 8. zip() with `itertools.zip_longest` for unequal lengths
print("8. zip() with itertools.zip_longest (for unequal lengths)")
import itertools

short_names = ['Alice', 'Bob']
long_scores = [90, 85, 78, 92]

# zip_longest fills missing values with `fillvalue` (default is None)
paired_data = list(itertools.zip_longest(short_names, long_scores, fillvalue='N/A'))
print(f"zip_longest({short_names}, {long_scores}): {paired_data}\n")

print("--- End of zip() Function Demonstration ---")
````
