# Python Random Module

You can generate random numbers in Python by using random module.

Python offers **`random`** module that can generate random numbers.

These are pseudo-random number as the sequence of number generated depends on the seed.

If the seeding value is same, the sequence will be the same. For example, if you use 2 as the seeding value, you will always see the following sequence.


````python
import random
random.seed(2)

print(random.random())
print(random.random())
print(random.random())
````

## **2. Basic Functions**

### **a. Generating Random Numbers**

1. **`random.random()`**

   - Generates a random float between 0.0 and 1.0.
   - Example:
     ```python
     print(random.random())  # Output: e.g., 0.3745401188473625
     ```

2. **`random.uniform(a, b)`**

   - Generates a random float between `a` and `b`.
   - Example:
     ```python
     print(random.uniform(1, 10))  # Output: e.g., 5.142342
     ```

3. **`random.randint(a, b)`**

   - Generates a random integer between `a` and `b` (inclusive).
   - Example:
     ```python
     print(random.randint(1, 10))  # Output: e.g., 7
     ```

4. **`random.randrange(start, stop, step)`**
   - Generates a random integer from `range(start, stop, step)`.
   - Example:
     ```python
     print(random.randrange(0, 100, 5))  # Output: e.g., 45
     ```

### **b. Random Choices**

5. **`random.choice(seq)`**

   - Returns a random element from a non-empty sequence (e.g., list, tuple).
   - Example:
     ```python
     print(random.choice(['apple', 'banana', 'cherry']))  # Output: e.g., 'banana'
     ```

6. **`random.choices(population, weights=None, k=1)`**

   - Returns a list of `k` random elements from the population, with optional weights.
   - Example:
     ```python
     print(random.choices(['red', 'green', 'blue'], weights=[10, 1, 1], k=2))  # Output: e.g., ['red', 'red']
     ```

7. **`random.sample(population, k)`**
   - Returns a list of `k` unique elements from the population.
   - Example:
     ```python
     print(random.sample(range(100), 5))  # Output: e.g., [42, 17, 89, 3, 56]
     ```

### **c. Shuffling**

8. **`random.shuffle(seq)`**
   - Shuffles a sequence in place.
   - Example:
     ```python
     cards = ['Ace', 'King', 'Queen', 'Jack']
     random.shuffle(cards)
     print(cards)  # Output: e.g., ['Queen', 'Ace', 'Jack', 'King']
     ```

## **3. Advanced Functions**

### **a. Seeding**

1. **`random.seed(a=None)`**
   - Initializes the random number generator with a seed value.
   - Using the same seed ensures reproducibility of random numbers.
   - Example:
     ```python
     random.seed(42)
     print(random.random())  # Output: 0.6394267984578837
     random.seed(42)
     print(random.random())  # Output: 0.6394267984578837 (same as above)
     ```

### **b. Random Distributions**

2. **`random.gauss(mu, sigma)`**

   - Generates a random float from a Gaussian (normal) distribution with mean `mu` and standard deviation `sigma`.
   - Example:
     ```python
     print(random.gauss(0, 1))  # Output: e.g., -0.234567
     ```

3. **`random.expovariate(lambd)`**

   - Generates a random float from an exponential distribution with parameter `lambd`.
   - Example:
     ```python
     print(random.expovariate(1.5))  # Output: e.g., 0.456789
     ```

4. **`random.betavariate(alpha, beta)`**

   - Generates a random float from a Beta distribution with parameters `alpha` and `beta`.
   - Example:
     ```python
     print(random.betavariate(2, 5))  # Output: e.g., 0.345678
     ```

5. **`random.triangular(low, high, mode)`**
   - Generates a random float from a triangular distribution.
   - Example:
     ```python
     print(random.triangular(1, 10, 5))  # Output: e.g., 4.56789
     ```

The **normal distribution**, also known as the **Gaussian distribution**, is one of the most important probability distributions in statistics and data science. It is characterized by its bell-shaped curve and is widely used in natural and social sciences. Below is a comprehensive guide to the normal distribution, covering everything from basic concepts to advanced techniques.

---

## **1. Basic Concepts**

### **a. What is a Normal Distribution?**

- A **normal distribution** is a continuous probability distribution that is symmetric around its mean.
- It is defined by two parameters:
  1. **Mean (μ):** The center of the distribution.
  2. **Standard Deviation (σ):** The spread or width of the distribution.

### **b. Properties of Normal Distribution**

1. **Bell-Shaped Curve:** The distribution is symmetric and bell-shaped.
2. **Mean = Median = Mode:** All three measures of central tendency are equal.
3. **68-95-99.7 Rule:**
   - ~68% of the data falls within **1 standard deviation** of the mean.
   - ~95% of the data falls within **2 standard deviations** of the mean.
   - ~99.7% of the data falls within **3 standard deviations** of the mean.

### **c. Probability Density Function (PDF)**

- The PDF of a normal distribution is given by:
  \[
  f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}
  \]
- It describes the likelihood of a random variable taking a specific value.

---

## **2. Generating Normal Distribution Data**

### **a. Using `random` Module**

- The `random` module provides the `gauss()` function to generate random numbers from a normal distribution.
- Example:

  std_dev = 1
  random_number = random.gauss(mean, std_dev)
  print(random_number)  # Output: e.g., -0.234567
  ```

  ```

## **1. Uniform Distribution**

### **a. What is a Uniform Distribution?**

- A **uniform distribution** is a probability distribution where all outcomes are equally likely.

### **b. Properties**

- **Continuous Uniform Distribution:**
  - Probability Density Function (PDF): \( f(x) = \frac{1}{b - a} \) for \( a \leq x \leq b \)

## **2. Binomial Distribution**

### **a. What is a Binomial Distribution?**

- A **binomial distribution** models the number of successes in a fixed number of independent Bernoulli trials.
- It is defined by two parameters:
  1. **n:** Number of trials.
  2. **p:** Probability of success in each trial.

### **b. Properties**

- Probability Mass Function (PMF):
  \[
  P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}
  \]
- Mean: \( \mu = np \)

### **c. Generating Binomial Distribution Data**

### **a. What is a Poisson Distribution?**

- A **Poisson distribution** models the number of events occurring in a fixed interval of time or space.
- It is defined by one parameter:
  1. **λ (lambda):** Average rate of occurrence.

### **b. Properties**

- Probability Mass Function (PMF):
  \[
  P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}

- Mean: \( \mu = \lambda \)

### **c. Generating Poisson Distribution Data**

## **4. Exponential Distribution**

### **a. What is an Exponential Distribution?**

- An **exponential distribution** models the time between events in a Poisson process.
- It is defined by one parameter:

### **b. Properties**

- Probability Density Function (PDF):

## **5. Gamma Distribution**

### **a. What is a Gamma Distribution?**

- A **gamma distribution** models the time until \( k \) events occur in a Poisson process.
- It is defined by two parameters:
  1. **k (shape):** Number of events.
  2. **θ (scale):** Scale parameter.

### **b. Properties**

- Probability Density Function (PDF):
  \[
  f(x) = \frac{x^{k - 1} e^{-x / \theta}}{\theta^k \Gamma(k)}
  \]
- Mean: \( \mu = k\theta \)
- Variance: \( \sigma^2 = k\theta^2 \)

### **c. Generating Gamma Distribution Data**

## **6. Beta Distribution**

### **a. What is a Beta Distribution?**

- A **beta distribution** models probabilities and is often used as a prior in Bayesian statistics.
- It is defined by two parameters:
  1. **α (alpha):** Shape parameter.
  2. **β (beta):** Shape parameter.

### **b. Properties**

- Probability Density Function (PDF):
  \[
  f(x) = \frac{x^{\alpha - 1} (1 - x)^{\beta - 1}}{B(\alpha, \beta)}
  \]
- Mean: \( \mu = \frac{\alpha}{\alpha + \beta} \)
- Variance: \( \sigma^2 = \frac{\alpha \beta}{(\alpha + \beta)^2 (\alpha + \beta + 1)} \)

### **c. Generating Beta Distribution Data**
