JSON (JavaScript Object Notation) is a lightweight data interchange format that is easy for humans to read and write, and easy for machines to parse and generate. Python provides a built-in module called `json` to work with JSON data. Below is a comprehensive guide to using JSON in Python, from basic to advanced concepts.

---

## **1. Basic JSON Syntax in Python**

### **1.1 Importing the `json` Module**

```python
import json
```

### **1.2 JSON Data Types**

JSON supports the following data types:

- **Objects**: `{ "key": "value" }` (Python dictionary)
- **Arrays**: `[1, 2, 3]` (Python list)
- **Strings**: `"Hello"` (Python string)
- **Numbers**: `42` or `3.14` (Python `int` or `float`)
- **Booleans**: `true` or `false` (Python `True` or `False`)
- **Null**: `null` (Python `None`)

---

## **2. Encoding (Serialization): Python to JSON**

### **2.1 `json.dumps()`**

Converts a Python object to a JSON-formatted string.

```python
data = {
    "name": "John",
    "age": 30,
    "is_student": False,
    "courses": ["Math", "Science"]
}

json_string = json.dumps(data)
print(json_string)
# Output: {"name": "John", "age": 30, "is_student": false, "courses": ["Math", "Science"]}
```

### **2.2 `json.dump()`**

Writes a Python object to a JSON file.

```python
with open("data.json", "w") as file:
    json.dump(data, file)
```

### **2.3 Customizing Serialization**

- **`indent`**: Pretty-print JSON with indentation.
- **`sort_keys`**: Sort keys alphabetically.
- **`separators`**: Customize separators (e.g., `(",", ": ")`).
- **`default`**: Handle non-serializable objects.

```python
json_string = json.dumps(data, indent=4, sort_keys=True)
print(json_string)
```

---

## **3. Decoding (Deserialization): JSON to Python**

### **3.1 `json.loads()`**

Converts a JSON-formatted string to a Python object.

```python
json_string = '{"name": "John", "age": 30, "is_student": false, "courses": ["Math", "Science"]}'
data = json.loads(json_string)
print(data)
# Output: {'name': 'John', 'age': 30, 'is_student': False, 'courses': ['Math', 'Science']}
```

### **3.2 `json.load()`**

Reads a JSON file and converts it to a Python object.

```python
with open("data.json", "r") as file:
    data = json.load(file)
    print(data)
```

---

## **4. Advanced JSON Handling**

### **4.1 Custom Object Serialization**

If you have custom objects, you can define a function to serialize them.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

def serialize_person(obj):
    if isinstance(obj, Person):
        return {"name": obj.name, "age": obj.age}
    raise TypeError("Object not serializable")

person = Person("Alice", 25)
json_string = json.dumps(person, default=serialize_person)
print(json_string)
# Output: {"name": "Alice", "age": 25}
```

### **4.2 Custom Object Deserialization**

You can use the `object_hook` parameter to deserialize JSON into custom objects.

```python
def deserialize_person(dct):
    if "name" in dct and "age" in dct:
        return Person(dct["name"], dct["age"])
    return dct

json_string = '{"name": "Alice", "age": 25}'
person = json.loads(json_string, object_hook=deserialize_person)
print(person.name, person.age)
# Output: Alice 25
```

### **4.3 Handling Non-ASCII Characters**

Use the `ensure_ascii` parameter to handle non-ASCII characters.

```python
data = {"name": "José"}
json_string = json.dumps(data, ensure_ascii=False)
print(json_string)
# Output: {"name": "José"}
```

### **4.4 Parsing JSON with Comments**

JSON does not natively support comments, but you can preprocess the JSON string to remove them.

```python
import re

json_string = '''
{
    "name": "John", // This is a comment
    "age": 30
}
'''
json_string = re.sub(r'//.*', '', json_string)  # Remove comments
data = json.loads(json_string)
print(data)
# Output: {'name': 'John', 'age': 30}
```

## **5. Error Handling**

### **5.1 `json.JSONDecodeError`**

Handles errors when decoding invalid JSON.

```python
invalid_json = '{"name": "John", "age": 30'  # Missing closing brace
try:
    data = json.loads(invalid_json)
except json.JSONDecodeError as e:
    print(f"Invalid JSON: {e}")
```

### **5.2 Handling Non-Serializable Objects**

If an object is not serializable, you can define a `default` function or use a library like `pickle`.

```python
import datetime

data = {"time": datetime.datetime.now()}
try:
    json_string = json.dumps(data)
except TypeError as e:
    print(f"Error: {e}")
```

## **6. Working with JSON Lines (JSONL)**

JSON Lines is a format where each line is a separate JSON object.

```python
jsonl_data = '''
{"name": "John", "age": 30}
{"name": "Alice", "age": 25}
'''

for line in jsonl_data.strip().split('\n'):
    data = json.loads(line)
    print(data)
```

## **8. JSON Schema Validation**

Use libraries like `jsonschema` to validate JSON data against a schema.

```python
from jsonschema import validate

schema = {
    "type": "object",
    "properties": {
        "name": {"type": "string"},
        "age": {"type": "number"}
    },
    "required": ["name"]
}

data = {"name": "John", "age": 30}
validate(instance=data, schema=schema)  # Raises ValidationError if invalid
```

## **9. JSON Path (Querying JSON)**

Use libraries like `jsonpath-ng` to query JSON data.

```python
from jsonpath_ng import parse

data = {
    "people": [
        {"name": "John", "age": 30},
        {"name": "Alice", "age": 25}
    ]
}

jsonpath_expr = parse('$.people[?(@.age > 26)]')
matches = [match.value for match in jsonpath_expr.find(data)]
print(matches)
# Output: [{'name': 'John', 'age': 30}]
```
