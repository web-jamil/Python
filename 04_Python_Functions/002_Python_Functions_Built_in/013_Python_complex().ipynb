# Python `complex()`

The **`complex()`** method returns a complex number when real and imaginary parts are provided, or it converts a string to a complex number.

**Syntax**:

```python
complex([real[, imag]])
```


## `complex()` Parameters

In general, **`complex()`** method takes two parameters:

- **real** - real part. If **`real`** is omitted, it defaults to 0.
- **imag** - imaginary part. If **`imag`** is omitted, it defaults to 0.

If the first parameter passed to this method is a string, it will be interpreted as a complex number. In this case, the second parameter shouldn't be passed.


## Return Value from `complex()`

As suggested by the name, **`complex()`** method returns a complex number.

If the string passed to this method is not a valid complex number, **`ValueError`** exception is raised.

> **Note**: The string passed to **`complex()`** should be in the form **`real+imagj`** or **`real+imagJ`**


````python
# Example 1: How to create a complex number in Python?

z = complex(2, -3)
print(z)

z = complex(1)
print(z)

z = complex()
print(z)

z = complex('5-9j')
print(z)
````

````python
# Example 2: Create complex Number Without Using complex()

# It's possible to create a complex number without using `complex()` method. 
# For that, you have to put 'j' or 'J' after a number.

a = 2+3j
print('a =',a)
print('Type of a is',type(a))

b = -2j
print('b =',b)
print('Type of b is',type(a))

c = 0j
print('c =',c)
print('Type of c is',type(c))
````

The `complex()` function in Python is a built-in constructor used to create complex numbers. A complex number is a number that consists of a real part and an imaginary part, represented as `a + bj`, where `a` is the real part, `b` is the imaginary part, and `j` (or `J`) is the imaginary unit (√-1). The `complex()` function is essential for working with complex numbers in mathematical computations, engineering, and scientific applications.

Below is a comprehensive guide to the `complex()` constructor, covering everything from basic usage to advanced techniques.

---

## **1. Basic Usage of `complex()`**

### **Syntax**

```python
complex(real, imag)
```

- **Parameters**:
  - `real`: The real part of the complex number (optional, defaults to `0`).
  - `imag`: The imaginary part of the complex number (optional, defaults to `0`).
- **Return Value**: A complex number of the form `real + imag*j`.

### **Examples**

```python
# Create a complex number with real and imaginary parts
c1 = complex(3, 4)
print(c1)  # Output: (3+4j)

# Create a complex number with only the real part
c2 = complex(5)
print(c2)  # Output: (5+0j)

# Create a complex number with only the imaginary part
c3 = complex(0, 7)
print(c3)  # Output: 7j
```

---

## **2. Creating Complex Numbers from Strings**

### **Parsing a String**

- The `complex()` constructor can also parse a string representation of a complex number.

```python
# Parse a string into a complex number
c4 = complex("3+4j")
print(c4)  # Output: (3+4j)

# Parse a string with only the imaginary part
c5 = complex("7j")
print(c5)  # Output: 7j

# Parse a string with only the real part
c6 = complex("5")
print(c6)  # Output: (5+0j)
```

### **String Format Requirements**

- The string must be in a valid format, such as `"a+bj"`, `"a"`, or `"bj"`.
- Spaces are not allowed in the string.

```python
# c7 = complex("3 + 4j")  # Raises ValueError: invalid literal for complex()
```

---

## **3. Advanced Usage**

### **Accessing Real and Imaginary Parts**

- Use the `.real` and `.imag` attributes to access the real and imaginary parts of a complex number.

```python
c = complex(3, 4)
print(c.real)  # Output: 3.0
print(c.imag)  # Output: 4.0
```

### **Complex Number Arithmetic**

- Perform arithmetic operations on complex numbers.

```python
c1 = complex(3, 4)
c2 = complex(1, 2)

# Addition
print(c1 + c2)  # Output: (4+6j)

# Subtraction
print(c1 - c2)  # Output: (2+2j)

# Multiplication
print(c1 * c2)  # Output: (-5+10j)

# Division
print(c1 / c2)  # Output: (2.2-0.4j)
```

### **Complex Conjugate**

- Use the `.conjugate()` method to get the complex conjugate of a number.

```python
c = complex(3, 4)
print(c.conjugate())  # Output: (3-4j)
```

### **Magnitude (Absolute Value)**

- Use the `abs()` function to compute the magnitude of a complex number.

```python
c = complex(3, 4)
print(abs(c))  # Output: 5.0 (sqrt(3^2 + 4^2))
```

---

## **4. Practical Applications**

### **Signal Processing**

- Complex numbers are used to represent signals in frequency domains (e.g., Fourier transforms).

```python
import numpy as np

# Create a complex signal
signal = np.array([complex(1, 2), complex(3, 4), complex(5, 6)])
print(signal)
```

### **Electrical Engineering**

- Complex numbers are used to represent impedance, voltage, and current in AC circuits.

```python
# Impedance calculation
resistance = 3  # Real part
reactance = 4   # Imaginary part
impedance = complex(resistance, reactance)
print(impedance)  # Output: (3+4j)
```

### **Physics and Mathematics**

- Complex numbers are used in quantum mechanics, fluid dynamics, and other fields.

```python
# Quantum state representation
quantum_state = complex(0.707, 0.707)
print(quantum_state)  # Output: (0.707+0.707j)
```

---

## **5. Combining `complex()` with Other Functions**

### **Using `complex()` with `cmath`**

- The `cmath` module provides mathematical functions for complex numbers.

```python
import cmath

c = complex(1, 1)
print(cmath.phase(c))  # Output: 0.7853981633974483 (phase angle in radians)
print(cmath.polar(c))  # Output: (1.4142135623730951, 0.7853981633974483) (magnitude, phase)
```

### **Using `complex()` with `numpy`**

- The `numpy` library supports complex numbers and operations.

```python
import numpy as np

# Create a complex array
arr = np.array([complex(1, 2), complex(3, 4)])
print(arr)
```

## **6. Common Errors and Pitfalls**

### **Invalid String Format**

- Passing an invalid string to `complex()` raises a `ValueError`.

```python
# c = complex("3 + 4j")  # Raises ValueError: invalid literal for complex()
```

### **Mixing Types**

- Ensure that the real and imaginary parts are numeric types.

```python
# c = complex("3", "4")  # Raises TypeError: complex() can't take second arg if first is a string
```
