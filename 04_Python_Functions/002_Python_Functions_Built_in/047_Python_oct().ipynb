# Python `oct()`

The **`oct()`** function takes an integer number and returns its octal representation.

**Syntax**:

```python
oct(x)
```


## `oct()` Parameters

The **`oct()`** function takes a single parameter x.

This parameter could be:

- an integer number (binary, decimal or hexadecimal)
- if not an integer, it should implement **`__index__()`** to return an integer


## Return Value from `oct()`

The **`oct()`** function returns an octal string from the given integer number.


````python
# Example 1: How oct() works in Python?

# decimal to octal
print('oct(10) is:', oct(10))

# binary to octal
print('oct(0b101) is:', oct(0b101))

# hexadecimal to octal
print('oct(0XA) is:', oct(0XA))
````

````python
# Example 2: oct() for custom objects

class Person:
    age = 23

    def __index__(self):
        return self.age

    def __int__(self):
        return self.age

person = Person()
print('The oct is:', oct(person))
````

Here, the Person class implements **`__int__()`** and **`__index__()`**. That's why we can use **`oct()`** on the objects of Person.

> **Note:** For compatibility, it's recommended to implement **`__int__()`** and **`__index__()`** with the same output.


### **Python `oct()` Function: A Comprehensive Guide (From Basic to Advanced)**

The `oct()` function in Python is used to convert an integer into its **octal representation** (base 8). It is a built-in Python function that helps when you need to display or work with numbers in base 8.

In this guide, we'll cover everything from basic usage to advanced tips and tricks related to the `oct()` function.

---

### **1. Basic Usage of `oct()`**

#### **1.1 Syntax**

```python
oct(number)
```

- **`number`**: The integer you want to convert to its octal representation. It can be any integer (positive, negative, or zero).
- **Returns**: A string representing the octal value, prefixed with `0o` to indicate that it's an octal number.

#### **1.2 Example: Basic Conversion**

```python
# Convert a decimal number to octal
n = 64
print(oct(n))  # Output: 0o100
```

**Explanation**:

- The `oct()` function converts `64` into its octal representation, which is `100`. The result is returned as a string prefixed with `0o` to indicate that the number is in octal format.

#### **1.3 Example: Negative Numbers**

```python
# Convert a negative decimal number to octal
n = -64
print(oct(n))  # Output: -0o100
```

**Explanation**:

- The function handles negative numbers as well. The result for `-64` is `-0o100`.

#### **1.4 Example: Zero**

```python
# Convert zero to octal
n = 0
print(oct(n))  # Output: 0o0
```

**Explanation**:

- `0` is converted to `0o0` in octal.

---

### **2. Understanding Octal Representation**

In octal, each digit represents 3 bits of binary data. Here's how conversion works:

- **Decimal 8 → Octal 10**: `8` in decimal is `10` in octal (`8 = 1 * 8^1 + 0 * 8^0`).
- **Decimal 64 → Octal 100**: `64` in decimal is `100` in octal (`64 = 1 * 8^2 + 0 * 8^1 + 0 * 8^0`).

#### **2.1 Octal Representation Example**

Let's convert a decimal number to octal manually to understand the conversion process:

- **Decimal 58** → Divide 58 by 8:
  - `58 ÷ 8 = 7` remainder `2` → LSB (least significant bit)
  - `7 ÷ 8 = 0` remainder `7` → MSB (most significant bit)

Thus, `58` in decimal is `72` in octal.

```python
n = 58
print(oct(n))  # Output: 0o72
```

### **3. Advanced Usage and Features**

#### **3.1 Removing the `0o` Prefix**

If you only want the octal digits without the `0o` prefix, you can slice the string returned by `oct()`:

```python
n = 64
octal_value = oct(n)[2:]  # Remove '0o' prefix
print(octal_value)  # Output: 100
```

**Explanation**:

- The `oct()` function returns a string prefixed with `0o`. By using slicing `[2:]`, we exclude the first two characters (`0o`) and get only the octal digits.

#### **3.2 Converting Back from Octal to Decimal**

To convert from octal back to decimal, you can use the `int()` function with base 8:

```python
octal_str = '100'  # Octal representation
decimal_value = int(octal_str, 8)  # Convert octal string to decimal
print(decimal_value)  # Output: 64
```

**Explanation**:

- `int(octal_str, 8)` converts the octal string `'100'` back to its decimal equivalent `64`.

#### **3.3 Padding Octal Numbers**

If you need to format octal numbers with a fixed width, you can use string formatting techniques. This is especially useful when working with file permissions or other applications where octal numbers need to be standardized.

```python
n = 58
formatted_octal = f"{oct(n)[2:]:>6}"  # Pad with spaces for a width of 6
print(formatted_octal)  # Output: '    72'
```

**Explanation**:

- The `f"{...:>6}"` syntax pads the string to ensure it has a width of 6 characters, aligning the number to the right.

Alternatively, you can use `zfill()` to pad with zeroes:

```python
n = 58
padded_octal = oct(n)[2:].zfill(6)  # Pad with zeros to make the length 6
print(padded_octal)  # Output: 000072
```

**Explanation**:

- `.zfill(6)` ensures the octal string is padded with leading zeros to make the total length 6.

### **4. Common Use Cases of Octal Representation**

#### **4.1 File Permissions**

In Unix-based systems (Linux, macOS), file permissions are often represented in octal format. For example, `755` represents a file permission mode.

```python
# Octal for file permissions
permissions = 0o755  # Read, write, and execute for owner; read and execute for others
print(oct(permissions))  # Output: 0o755
```

**Explanation**:

- In `0o755`, the first digit represents the owner's permissions, and the second and third digits represent the permissions for the group and others. Each digit corresponds to a combination of read, write, and execute permissions in octal.

#### **4.2 Bit Masks**

In some scenarios, you may use octal representations to work with bit masks (e.g., setting, clearing, toggling bits). For example:

```python
# Define a mask for a specific permission (read, write, execute)
mask = 0o444  # Read-only permission for everyone
n = 0o755  # Full permissions

# Apply the mask (e.g., apply read-only mask)
new_permissions = n & mask
print(oct(new_permissions))  # Output: 0o444
```

**Explanation**:

- The mask `0o444` (read-only permission) is applied to `0o755`, resulting in `0o444` (read-only permission for everyone).


**Explanation**:

- Negative integers are handled in the same way, but the result includes the `-0o` prefix.

The `oct()` function in Python is a built-in function used to convert an integer into its octal (base-8) representation. Octal numbers use digits from `0` to `7`. This function is particularly useful in low-level programming, debugging, and working with systems that use octal representations (e.g., file permissions in Unix-like systems).

Below is a comprehensive guide to the `oct()` function, covering everything from basic usage to advanced applications.

---

## **1. Basic Usage of `oct()`**

### **Syntax**

```python
oct(integer)
```

- **Parameter**: `integer` (can be an integer of any size, positive or negative).
- **Return Value**: A string prefixed with `0o` representing the octal value of the input integer.

### **Examples**

```python
# Convert integer to octal
print(oct(10))       # Output: '0o12'
print(oct(255))      # Output: '0o377'
print(oct(-10))      # Output: '-0o12'
```

---

## **2. Working with Different Input Types**

### **Integers**

- The `oct()` function works directly with integers.

```python
print(oct(8))        # Output: '0o10'
print(oct(16))       # Output: '0o20'
```

### **Floats**

- Floats are not supported and will raise a `TypeError`.

```python
# print(oct(10.5))   # Raises TypeError: 'float' object cannot be interpreted as an integer
```

### **Strings**

- Strings representing integers can be converted to octal after converting them to integers.

```python
num_str = "64"
print(oct(int(num_str)))  # Output: '0o100'
```

---

## **3. Advanced Usage**

### **Custom Octal Conversion Without `0o` Prefix**

- If you want the octal value without the `0o` prefix, you can slice the string.

```python

```

### **Octal Representation of Binary or Hexadecimal Numbers**

- Convert binary or hexadecimal numbers to integers first, then to octal.

```python
binary_num = 0b1010   # Binary representation of 10
hex_num = 0xFF        # Hexadecimal representation of 255

print(oct(binary_num))  # Output: '0o12'
print(oct(hex_num))     # Output: '0o377'
```

### **Octal Representation of Negative Numbers**

- Negative integers are converted to their octal representation with a `-0o` prefix.

```python
print(oct(-10))  # Output: '-0o12'
```
