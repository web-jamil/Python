The `raise` keyword in Python is used to initiate an exception, allowing you to create your own error handling mechanisms. It can be used to raise built-in exceptions, custom exceptions, or to re-raise exceptions that have been caught. This guide covers the `raise` keyword from basic to advanced usage.

## Basic Overview

### What is an Exception?

An exception is an error that occurs during the execution of a program. When Python encounters a situation it cannot handle, it raises an exception.

### Using the `raise` Keyword

#### 1. Raising Built-in Exceptions

You can raise built-in exceptions by specifying the exception type. Here’s an example:

```python
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("You cannot divide by zero.")
    return a / b

# Example Usage
try:
    print(divide(10, 0))
except ZeroDivisionError as e:
    print(f"Error: {e}")  # Output: Error: You cannot divide by zero.
```

#### 2. Raising Custom Exceptions

To create and raise custom exceptions, you define a class that inherits from the `Exception` class.

```python
class MyCustomError(Exception):
    pass

def check_value(x):
    if x < 0:
        raise MyCustomError("Value cannot be negative.")

# Example Usage
try:
    check_value(-1)
except MyCustomError as e:
    print(f"Error: {e}")  # Output: Error: Value cannot be negative.
```

## Intermediate Usage

### 1. Raising an Exception with a Message

You can provide a message when raising an exception, which gives more context about the error:

```python
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be at least 18.")

# Example Usage
try:
    validate_age(16)
except ValueError as e:
    print(f"Error: {e}")  # Output: Error: Age must be at least 18.
```

### 2. Rethrowing an Exception

You can catch an exception and re-raise it using `raise` without specifying an exception type. This is useful when you want to perform some action when catching an exception but still want to propagate it:

```python
def check_positive(x):
    try:
        if x < 0:
            raise ValueError("Negative value!")
    except ValueError as e:
        print("Caught an error:", e)
        raise  # Re-raise the caught exception

# Example Usage
try:
    check_positive(-5)
except ValueError as e:
    print(f"Error propagated: {e}")  # Output: Error propagated: Negative value!
```

## Advanced Concepts

### 1. Custom Exception Hierarchies

You can create a hierarchy of exceptions by subclassing:

```python
class MyAppError(Exception):
    pass

class DatabaseError(MyAppError):
    pass

class ValidationError(MyAppError):
    pass

def save_data(data):
    if data is None:
        raise ValidationError("No data provided.")
    # Simulating a database error
    raise DatabaseError("Database connection failed.")

# Example Usage
try:
    save_data(None)
except MyAppError as e:
    print(f"MyAppError occurred: {e}")  # Output: MyAppError occurred: No data provided.
```

### 2. Exception Chaining

When raising a new exception while handling another, you can use `from` to maintain the context of the original exception:

```python
def open_file(filename):
    try:
        with open(filename, 'r') as file:
            return file.read()
    except FileNotFoundError as e:
        raise RuntimeError("File not found. Please check the file name.") from e

# Example Usage
try:
    open_file("non_existent_file.txt")
except RuntimeError as e:
    print(e)  # Output: File not found. Please check the file name.
    print(e.__cause__)  # Output: [Errno 2] No such file or directory: 'non_existent_file.txt'
```

### 3. Combining `try` with `raise`

You can use `raise` within the `try` block of a function to control the flow of exception handling:

```python
def calculate_square_root(number):
    if number < 0:
        raise ValueError("Cannot calculate square root of a negative number.")
    return number ** 0.5

# Example Usage
try:
    print(calculate_square_root(-4))
except ValueError as e:
    print(f"Error: {e}")  # Output: Error: Cannot calculate square root of a negative number.
```

# **Python `raise` Keyword: From Basic to Advanced**

The `raise` keyword in Python is used to **explicitly raise exceptions** in your code. It allows you to trigger errors when certain conditions are met, customize error messages, and propagate exceptions. This guide covers everything from basic usage to advanced techniques.

---

## **1. Basic Usage of `raise`**

### **a. Raising Built-in Exceptions**

You can raise standard Python exceptions (e.g., `ValueError`, `TypeError`, `IndexError`).

**Example:**

```python
x = -5
if x < 0:
    raise ValueError("Negative numbers are not allowed!")
```

**Output:**

```
ValueError: Negative numbers are not allowed!
```

### **b. Raising a Generic Exception**

If no specific exception fits, you can use the base `Exception` class.

**Example:**

```python
if "apple" not in ["banana", "cherry"]:
    raise Exception("Apple not found in the list!")
```

**Output:**

```
Exception: Apple not found in the list!
```

---

## **2. Raising Custom Exceptions**

### **a. Defining a Custom Exception**

You can create your own exception class by inheriting from `Exception`.

**Example:**

```python
class NegativeNumberError(Exception):
    pass

x = -10
if x < 0:
    raise NegativeNumberError("Negative numbers are invalid!")
```

**Output:**

```
NegativeNumberError: Negative numbers are invalid!
```

### **b. Adding Custom Attributes**

You can enhance your exception with extra data.

**Example:**

```python
class InvalidAgeError(Exception):
    def __init__(self, age, message="Age must be >= 18"):
        self.age = age
        self.message = message
        super().__init__(self.message)

age = 15
if age < 18:
    raise InvalidAgeError(age)
```

**Output:**

```
InvalidAgeError: Age must be >= 18
```

---

## **3. Advanced Usage of `raise`**

### **a. Re-raising an Exception**

You can catch an exception and re-raise it (useful in logging or cleanup).

**Example:**

```python
try:
    x = 10 / 0
except ZeroDivisionError as e:
    print("Logging error:", e)
    raise  # Re-raise the same exception
```

**Output:**

```
Logging error: division by zero
ZeroDivisionError: division by zero
```

### **b. Raising a Different Exception**

You can catch one exception and raise another (exception chaining).

**Example:**

```python
try:
    x = 10 / 0
except ZeroDivisionError as e:
    raise ValueError("Invalid division!") from e
```

**Output:**

```
ValueError: Invalid division!
```

### **c. Exception Chaining (`from` Keyword)**

Use `raise ... from` to link exceptions (useful for debugging).

**Example:**

```python
try:
    x = 10 / 0
except ZeroDivisionError as e:
    raise RuntimeError("Math error!") from e
```

**Output:**

```
RuntimeError: Math error!
```

## **4. Best Practices**

### **a. Use Specific Exceptions**

Instead of raising a generic `Exception`, use built-in exceptions (`ValueError`, `TypeError`, etc.) or custom exceptions.

✅ **Good:**

```python
if not isinstance(x, int):
    raise TypeError("Expected an integer!")
```

❌ **Bad:**

```python
if not isinstance(x, int):
    raise Exception("Expected an integer!")
```

### **b. Provide Clear Error Messages**

Always include a descriptive message when raising exceptions.

✅ **Good:**

```python
raise ValueError("Input must be between 1 and 100!")
```

❌ **Bad:**

```python
raise ValueError("Invalid input!")
```

### **c. Use `try-except` for Expected Errors**

If an error is expected (e.g., file not found), handle it gracefully instead of raising it.

**Example:**

```python
try:
    with open("file.txt") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found!")
```

The `raise` keyword in Python is used to trigger (or "raise") exceptions, either predefined or custom exceptions. Raising exceptions in Python allows you to handle errors effectively and control the flow of your program. Below is an in-depth explanation of how the `raise` keyword works, from basic usage to advanced concepts.

### **1. Basic Syntax of `raise`**

The basic syntax for using the `raise` keyword in Python is:

```python
raise ExceptionType("Error message")
```

- **ExceptionType**: This can be any exception class, such as `ValueError`, `TypeError`, or a custom exception you define.
- **Error message**: An optional string that provides information about the exception.

#### Example:

```python
raise ValueError("This is a custom error message")
```

This raises a `ValueError` with the message `"This is a custom error message"`.

### **2. Raising Built-in Exceptions**

Python provides several built-in exceptions that are typically raised in various situations. Here are some of the common built-in exceptions you might raise:

- `TypeError`: Raised when an operation or function is applied to an object of inappropriate type.
- `ValueError`: Raised when a function receives an argument of correct type but inappropriate value.
- `IndexError`: Raised when a sequence subscript is out of range.
- `KeyError`: Raised when a dictionary key is not found.
- `ZeroDivisionError`: Raised when a division or modulo operation is performed with zero.

#### Example:

```python
x = -1
if x < 0:
    raise ValueError("x cannot be negative")
```

In this example, a `ValueError` will be raised if the value of `x` is less than 0.

### **3. Raising Exceptions in Functions**

You can raise exceptions inside functions when certain conditions are met. This can be useful for validation checks.

#### Example:

```python
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("You cannot divide by zero")
    return a / b

print(divide(10, 0))  # This will raise ZeroDivisionError
```

### **4. Raising Custom Exceptions**

You can create your own custom exceptions by subclassing the built-in `Exception` class or any of its subclasses.

#### Example:

```python
class NegativeValueError(Exception):
    def __init__(self, value):
        self.value = value
        self.message = f"Negative value is not allowed: {value}"
        super().__init__(self.message)

def check_positive(x):
    if x < 0:
        raise NegativeValueError(x)
    else:
        return x

print(check_positive(-5))  # Raises a custom exception: NegativeValueError
```

### **5. Raising Exceptions Without Specifying Type**

You can also raise exceptions without specifying the exception type, using `raise` by itself. This is usually done to re-raise the current exception in an `except` block.

#### Example:

```python
try:
    1 / 0
except ZeroDivisionError as e:
    print("An error occurred:", e)
    raise  # Re-raises the current exception
```

In this case, the `raise` keyword re-raises the same exception (`ZeroDivisionError`) that was caught.

### **6. Raising Multiple Exceptions**

Sometimes, you might want to raise multiple exceptions in certain situations. You can use an `if` statement to check for various conditions and raise the corresponding exception.

#### Example:

```python
def validate_number(n):
    if n < 0:
        raise ValueError("Number must be non-negative")
    elif n == 0:
        raise ZeroDivisionError("Number must not be zero")
    else:
        return n

print(validate_number(-5))  # Raises ValueError
print(validate_number(0))   # Raises ZeroDivisionError
```

### **7. `raise` in Exception Handling (try-except blocks)**

Using `raise` within a `try-except` block is often used to handle errors gracefully and re-raise them if necessary.

#### Example:

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
except ZeroDivisionError:
    print("Cannot divide by zero.")
    raise  # Re-raises the exception to terminate the program
except ValueError:
    print("Invalid input! Please enter an integer.")
    raise  # Re-raises the exception to terminate the program
```

In this example, the program catches errors like division by zero or invalid input, prints a message, and then re-raises the exception to stop further execution.

### **8. The `from` Keyword with `raise`**

You can use the `from` keyword in conjunction with `raise` to provide context when raising an exception. This allows you to attach a new exception to an existing exception, which helps in debugging.

#### Example:

```python
try:
    value = int("abc")  # Will raise ValueError
except ValueError as e:
    raise ValueError("Custom error message") from e  # Raise a new ValueError and attach the original exception
```

This creates a chain of exceptions that gives more context about what happened.

### **9. Re-raising Exceptions (without the `from` keyword)**

Sometimes, it is necessary to catch an exception, perform some logic (like logging or cleanup), and then re-raise the exception for higher-level handling.

#### Example:

```python
try:
    raise ValueError("Initial error")
except ValueError as e:
    print(f"Caught an error: {e}")
    raise  # Re-raises the same ValueError
```

### **10. `raise` in Custom Exception Classes**

When defining a custom exception, you can add more context to the error by overriding methods like `__init__`, `__str__`, and others to provide richer error messages.

#### Example:

```python
class CustomError(Exception):
    def __init__(self, message, code):
        self.message = message
        self.code = code
        super().__init__(f"{self.message} (Error Code: {self.code})")

def error_handler():
    raise CustomError("Something went wrong", 1001)

try:
    error_handler()
except CustomError as e:
    print(e)  # Output: Something went wrong (Error Code: 1001)
```

### **11. The `__str__()` Method in Custom Exceptions**

Custom exceptions can also implement the `__str__()` method to control the string representation of the exception, making the exception message more informative.

#### Example:

```python
class NegativeNumberError(Exception):
    def __init__(self, value):
        self.value = value

    def __str__(self):
        return f"NegativeNumberError: The value {self.value} is negative, which is not allowed."

def check_positive(x):
    if x < 0:
        raise NegativeNumberError(x)

try:
    check_positive(-10)
except NegativeNumberError as e:
    print(e)  # Output: NegativeNumberError: The value -10 is negative, which is not allowed.
```

### **Summary**

- **Basic Usage**: Use `raise` to trigger exceptions and handle errors in your code.
- **Built-in Exceptions**: Python provides built-in exceptions such as `ValueError`, `ZeroDivisionError`, etc., that can be raised using `raise`.
- **Custom Exceptions**: You can define your own exceptions by subclassing `Exception` and raise them with `raise`.
- **Re-raising Exceptions**: Use `raise` inside `except` blocks to re-raise exceptions and maintain exception context.
- **Chaining Exceptions**: The `from` keyword can be used to chain exceptions and provide more context about errors.

### **Best Practices**

- Use exceptions to handle errors gracefully and provide useful error messages to users or developers.
- Raise custom exceptions when dealing with specific types of errors to make your code more readable and maintainable.
- Avoid overusing exceptions for control flow, as exceptions should represent errors, not typical program logic.
