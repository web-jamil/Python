# Python `dict()`

The **`dict()`** constructor creates a dictionary in Python.

Different forms of **`dict()`** constructors are:

```python
class dict(**kwarg)
class dict(mapping, **kwarg)
class dict(iterable, **kwarg)
```

> **Note**: <b>`**kwarg`</b> let you take an arbitrary number of keyword arguments.

A keyword argument is an argument preceded by an identifier (eg. **`name=`**). Hence, the keyword argument of the form **`kwarg=value`** is passed to **`dict()`** constructor to create dictionaries.

**`dict()`** doesn't return any value (returns **`None`**).


````python
# Example 1: Create Dictionary Using keyword arguments only

numbers = dict(x=5, y=0)
print('numbers =', numbers)
print(type(numbers))

empty = dict()
print('empty =', empty)
print(type(empty))
````

````python
# Example 2: Create Dictionary Using Iterable

# keyword argument is not passed
numbers1 = dict([('x', 5), ('y', -5)])
print('numbers1 =',numbers1)

# keyword argument is also passed
numbers2 = dict([('x', 5), ('y', -5)], z=8)
print('numbers2 =',numbers2)

# zip() creates an iterable in Python 3
numbers3 = dict(dict(zip(['x', 'y', 'z'], [1, 2, 3])))
print('numbers3 =',numbers3)
````

````python
# Example 3: Create Dictionary Using Mapping

numbers1 = dict({'x': 4, 'y': 5})
print('numbers1 =',numbers1)

# you don't need to use dict() in above code
numbers2 = {'x': 4, 'y': 5}
print('numbers2 =',numbers2)

# keyword argument is also passed
numbers3 = dict({'x': 4, 'y': 5}, z=8)
print('numbers3 =',numbers3)
````

### **Python `dict` Functions: A Comprehensive Guide (From Basic to Advanced)**

In Python, a **dictionary** is an unordered collection of key-value pairs. It's one of the most commonly used data structures due to its fast lookups and versatility. Python provides various built-in functions and methods to interact with dictionaries. In this guide, we'll cover everything you need to know about dictionaries and their functions — from basic operations to advanced usage.

---

### **1. Basic Dictionary Operations**

#### **1.1 Creating a Dictionary**

Dictionaries are created using curly braces `{}`, with key-value pairs separated by a colon `:`.

```python
# Creating a dictionary
my_dict = {"name": "Alice", "age": 25, "city": "New York"}
print(my_dict)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

You can also create dictionaries using the `dict()` constructor:

```python
my_dict = dict(name="Alice", age=25, city="New York")
print(my_dict)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

#### **1.2 Accessing Values**

You can access the value associated with a key using square brackets `[]`:

```python
print(my_dict["name"])  # Output: Alice
```

You can also use the `get()` method, which allows for safer access (i.e., if the key doesn’t exist, it returns `None` or a default value):

```python
print(my_dict.get("age"))  # Output: 25
print(my_dict.get("gender", "Not Provided"))  # Output: Not Provided
```

#### **1.3 Adding or Modifying Items**

To add or modify an entry in a dictionary, assign a value to a key:

```python
my_dict["email"] = "alice@example.com"  # Adds a new key-value pair
my_dict["age"] = 26  # Modifies the value of the existing key
print(my_dict)
```

#### **1.4 Deleting Items**

You can remove key-value pairs using `del`, `pop()`, or `popitem()`:

```python
del my_dict["email"]  # Removes the 'email' key-value pair
print(my_dict)

# Using pop() to remove a specific item and get its value
age = my_dict.pop("age")
print(age)  # Output: 26
print(my_dict)

# Using popitem() to remove and return the last inserted key-value pair
item = my_dict.popitem()
print(item)  # Output: ('city', 'New York')
print(my_dict)
```

---

### **2. Dictionary Methods**

Python dictionaries come with a variety of methods for manipulating and retrieving data.

#### **2.1 `keys()`**

Returns a view object that displays a list of all the keys in the dictionary.

```python
print(my_dict.keys())  # Output: dict_keys(['name'])
```

#### **2.2 `values()`**

Returns a view object that displays a list of all the values in the dictionary.

```python
print(my_dict.values())  # Output: dict_values(['Alice'])
```

#### **2.3 `items()`**

Returns a view object that displays a list of tuple pairs (key, value).

```python
print(my_dict.items())  # Output: dict_items([('name', 'Alice')])
```

#### **2.4 `update()`**

The `update()` method updates the dictionary with the key-value pairs from another dictionary or iterable of key-value pairs.

```python
other_dict = {"age": 30, "email": "alice@example.com"}
my_dict.update(other_dict)
print(my_dict)
```

#### **2.5 `clear()`**

Removes all items from the dictionary, leaving it empty.

```python
my_dict.clear()
print(my_dict)  # Output: {}
```

#### **2.6 `copy()`**

Returns a shallow copy of the dictionary.

```python
new_dict = my_dict.copy()
print(new_dict)  # Output: {}
```

#### **2.7 `fromkeys()`**

Creates a new dictionary with the given keys and assigns a default value to all keys.

```python
keys = ['a', 'b', 'c']
default_value = 0
new_dict = dict.fromkeys(keys, default_value)
print(new_dict)  # Output: {'a': 0, 'b': 0, 'c': 0}
```

---

### **3. Advanced Dictionary Operations**

#### **3.1 Dictionary Comprehensions**

Python allows creating dictionaries through a concise, elegant way using dictionary comprehensions. This is similar to list comprehensions but works with dictionaries.

```python
# Create a dictionary with squares of numbers
squares = {x: x**2 for x in range(1, 6)}
print(squares)  # Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```

#### **3.2 Merging Dictionaries (Python 3.9+)**

Starting from Python 3.9, you can merge two dictionaries using the `|` operator.

```python
dict1 = {"name": "Alice", "age": 25}
dict2 = {"city": "New York", "email": "alice@example.com"}
merged_dict = dict1 | dict2
print(merged_dict)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York', 'email': 'alice@example.com'}
```

#### **3.3 Nested Dictionaries**

Dictionaries can contain other dictionaries as values. These are called **nested dictionaries**.

```python
nested_dict = {
    "person1": {"name": "Alice", "age": 25},
    "person2": {"name": "Bob", "age": 30}
}

# Accessing nested dictionary values
print(nested_dict["person1"]["name"])  # Output: Alice
```

#### **3.4 Using `defaultdict` (from `collections` module)**

A `defaultdict` is a subclass of the dictionary class that provides a default value for missing keys.

```python
from collections import defaultdict

# Creates a defaultdict with default value of 0
dd = defaultdict(int)
dd["apple"] += 1
dd["banana"] += 2
print(dd)  # Output: defaultdict(<class 'int'>, {'apple': 1, 'banana': 2})
```

#### **3.5 Using `Counter` (from `collections` module)**

`Counter` is a specialized dictionary subclass used to count hashable items.

```python
from collections import Counter

fruits = ["apple", "banana", "apple", "orange", "banana", "banana"]
counter = Counter(fruits)
print(counter)  # Output: Counter({'banana': 3, 'apple': 2, 'orange': 1})
```

#### **3.6 Dictionary Views and Iterators**

- **`dict_keys`**: The object returned by `keys()` is a view of the dictionary's keys.
- **`dict_values`**: The object returned by `values()` is a view of the dictionary's values.
- **`dict_items`**: The object returned by `items()` is a view of the dictionary's key-value pairs.

You can convert these views to lists or iterate over them:

```python
# Converting views to lists
keys_list = list(my_dict.keys())
values_list = list(my_dict.values())
items_list = list(my_dict.items())
print(keys_list, values_list, items_list)
```

#### **3.7 Using `setdefault()`**

The `setdefault()` method returns the value of a key if it exists. If the key doesn’t exist, it adds the key with a default value.

```python
my_dict = {"name": "Alice"}
value = my_dict.setdefault("age", 25)
print(value)  # Output: 25
print(my_dict)  # Output: {'name': 'Alice', 'age': 25}
```

#### **3.8 Using `pop()` with Default Value**

The `pop()` method removes a key and returns its value. You can provide a default value to return if the key is missing.

```python
value = my_dict.pop("city", "Not Found")
print(value)  # Output: Not Found
```

---

### **4. Performance Considerations**

- **Time Complexity**: In Python, dictionaries are implemented using hash tables. The average time complexity for dictionary operations like insertion, deletion, and lookup is **O(1)**.
- **Memory**: Dictionaries have a dynamic resizing mechanism to ensure that they remain efficient as they grow. However, dictionaries can be memory-heavy, especially for large data sets due to their hash-based implementation.

---

### **5. Tips and Best Practices**

- **Use Immutable Types as Keys**: In a dictionary, keys must be immutable types (e.g., strings, integers, tuples). This is because dictionary keys must be hashable.
- **Avoid Using Large Mutable Objects as Keys**: Lists or other mutable objects can’t be used as dictionary keys because their hash value can change.
- **Understand Dictionary Views**: The `keys()`, `values()`, and `items()` methods return views that reflect changes made to the dictionary. Use `list()` if you need a snapshot of the dictionary's state.
- **Leverage `defaultdict` and `Counter` for Special Cases**: The `defaultdict` and `Counter` classes from the `collections` module are highly useful for counting and handling missing dictionary keys.

---

### **6. Summary**

- **Basic Operations**: You can create, access, update, and delete key-value pairs in a dictionary using simple methods like `[]`, `get()`, `update()`, `pop()`, and `del`.
- **Dictionary Methods**: Methods like `keys()`, `values()`, and `items()` provide useful ways to access keys, values, and key-value pairs.
- **Advanced Usage**: Use dictionary comprehensions for concise creation, `defaultdict` and `Counter` for specialized use cases, and `fromkeys()` for creating dictionaries with default values.
- **Merging Dictionaries**: Python 3.9 introduced the `|` operator for merging dictionaries, making it easier to combine data.

By mastering Python dictionaries and their functions, you can handle a wide range of data processing tasks efficiently, making them an essential tool in your Python programming toolkit.

### **Python `dict()` Constructor: A Comprehensive Guide (From Basic to Advanced)**

The `dict()` constructor in Python is used to create a dictionary. A dictionary is an unordered collection of key-value pairs, where each key is unique and maps to a specific value. The `dict()` constructor provides various ways to create dictionaries in Python, making it one of the most versatile ways to work with key-value pairs.

This guide will cover the basic usage of the `dict()` constructor, as well as more advanced features and techniques.

---

### **1. Basic Usage of `dict()` Constructor**

#### **1.1 Creating an Empty Dictionary**

You can create an empty dictionary using the `dict()` constructor. This is similar to creating a dictionary with curly braces `{}` but using the `dict()` function.

```python
empty_dict = dict()
print(empty_dict)  # Output: {}
```

---

### **2. Creating Dictionaries Using Key-Value Pairs**

You can create a dictionary by passing key-value pairs to the `dict()` constructor as keyword arguments.

```python
# Creating a dictionary using keyword arguments
my_dict = dict(name="Alice", age=25, city="New York")
print(my_dict)
# Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

**Explanation**:

- `name`, `age`, and `city` are the keys, and `"Alice"`, `25`, and `"New York"` are their corresponding values.

---

### **3. Creating Dictionaries Using Sequences of Key-Value Pairs**

You can create a dictionary by passing an iterable of key-value pairs (like a list or tuple) to the `dict()` constructor.

#### **3.1 Using a List of Tuples**

```python
# Using a list of tuples
key_value_pairs = [("name", "Alice"), ("age", 25), ("city", "New York")]
my_dict = dict(key_value_pairs)
print(my_dict)
# Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

**Explanation**:

- Each tuple contains a key-value pair, and the `dict()` constructor converts this list of tuples into a dictionary.

#### **3.2 Using a List of Lists**

You can also pass a list of lists where each sublist represents a key-value pair.

```python
key_value_pairs = [["name", "Alice"], ["age", 25], ["city", "New York"]]
my_dict = dict(key_value_pairs)
print(my_dict)
# Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

---

### **4. Creating Dictionaries Using the `zip()` Function**

You can use the `zip()` function to create a dictionary from two sequences — one for keys and one for values. The `zip()` function pairs up elements from two iterables into tuples, which can be passed to the `dict()` constructor.

#### **4.1 Using `zip()` with Two Lists**

```python
keys = ["name", "age", "city"]
values = ["Alice", 25, "New York"]
my_dict = dict(zip(keys, values))
print(my_dict)
# Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

**Explanation**:

- `zip(keys, values)` creates an iterator of pairs: `("name", "Alice")`, `("age", 25)`, and `("city", "New York")`.
- The `dict()` constructor then converts these pairs into a dictionary.

---

### **5. Using `dict()` with `fromkeys()`**

The `fromkeys()` method creates a new dictionary from a given sequence of keys, all with the same default value.

#### **5.1 Creating a Dictionary with a Default Value**

```python
keys = ["a", "b", "c"]
default_value = 0
my_dict = dict.fromkeys(keys, default_value)
print(my_dict)
# Output: {'a': 0, 'b': 0, 'c': 0}
```

**Explanation**:

- `fromkeys()` creates a dictionary where each key in the provided sequence (`keys`) maps to the same default value (`0`).

#### **5.2 Creating a Dictionary with `None` as the Default Value**

```python
keys = ["name", "age", "city"]
my_dict = dict.fromkeys(keys)
print(my_dict)
# Output: {'name': None, 'age': None, 'city': None}
```

**Explanation**:

- If no second argument is provided, the default value for each key is `None`.

---

### **6. Using `dict()` with Iterables and Conditional Logic**

You can use the `dict()` constructor in combination with comprehensions to create more complex dictionaries.

#### **6.1 Dictionary Comprehension**

You can use a dictionary comprehension inside the `dict()` constructor to create a dictionary based on conditions or transformations.

```python
# Creating a dictionary of squares
squares = dict((x, x**2) for x in range(1, 6))
print(squares)
# Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```

**Explanation**:

- The comprehension `dict((x, x**2) for x in range(1, 6))` creates a dictionary where the keys are numbers from 1 to 5, and the values are their squares.

#### **6.2 Using `dict()` with `filter()` or `map()`**

You can also combine the `dict()` constructor with functions like `filter()` or `map()` to create dictionaries based on certain conditions or transformations.

```python
# Creating a dictionary of even numbers and their squares
even_squares = dict(map(lambda x: (x, x**2), filter(lambda x: x % 2 == 0, range(1, 6))))
print(even_squares)
# Output: {2: 4, 4: 16}
```

**Explanation**:

- `filter(lambda x: x % 2 == 0, range(1, 6))` filters out the even numbers from 1 to 5.
- `map(lambda x: (x, x**2), ...)` maps each even number to a key-value pair where the value is its square.
- `dict()` converts the result into a dictionary.

### **7. Advanced Usage of `dict()`**

#### **7.1 Creating Nested Dictionaries**

You can create dictionaries that contain other dictionaries as values, which is useful for organizing complex data.

```python
nested_dict = {
    "person1": {"name": "Alice", "age": 25},
    "person2": {"name": "Bob", "age": 30}
}
print(nested_dict)
# Output: {'person1': {'name': 'Alice', 'age': 25}, 'person2': {'name': 'Bob', 'age': 30}}
```

**Explanation**:

- In this example, each key in the outer dictionary maps to another dictionary, which contains personal information about individuals.

#### **7.2 Using `dict()` with `defaultdict` from the `collections` module**

A `defaultdict` is a subclass of the `dict` class that returns a default value for missing keys, instead of raising a `KeyError`.

```python
from collections import defaultdict

# Creating a defaultdict with a default value of 0
dd = defaultdict(int)
dd["apple"] += 1
dd["banana"] += 2
print(dd)
# Output: defaultdict(<class 'int'>, {'apple': 1, 'banana': 2})
```

**Explanation**:

- A `defaultdict(int)` creates a dictionary where each missing key returns a default value of `0` (from `int()`).
- The keys `"apple"` and `"banana"` are updated with values `1` and `2`, respectively.

#### **7.3 Using `OrderedDict` for Ordered Dictionaries**

If you need to maintain the order of key-value pairs, you can use `OrderedDict` from the `collections` module. This ensures that the order of elements remains the same as when they were added.

```python
from collections import OrderedDict

ordered_dict = OrderedDict([("name", "Alice"), ("age", 25), ("city", "New York")])
print(ordered_dict)
# Output: OrderedDict([('name', 'Alice'), ('age', 25), ('city', 'New York')])
```

**Explanation**:

- `OrderedDict` preserves the order in which the key-value pairs are inserted into the dictionary.

---

### **8. Performance Considerations**

- **Time Complexity**: The time complexity of most operations (insertion, lookup, deletion) in a Python dictionary is **O(1)** on average due to the hash table implementation.
- **Memory Usage**: Dictionaries are efficient in terms of memory usage, but large dictionaries can consume significant memory, especially if they contain large numbers of keys or complex values.
- **Iteration**: Iterating over a dictionary (using `for key in dict` or `for key, value in dict.items()`) is also efficient, but the order of iteration depends on the dictionary type. Standard dictionaries (from Python 3.7 onwards) preserve insertion order, while older Python versions did not.
