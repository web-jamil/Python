## 1. Introduction to the `datetime` Module

The `datetime` module is part of Python's standard library and provides:

- Classes for representing dates, times, and datetime objects.
- Methods for formatting, parsing, and performing arithmetic on dates and times.

To use the `datetime` module, import it:

```python
import datetime
```

---

## 2. Key Classes in `datetime`

### `datetime.date`

Represents a date (year, month, day).

#### Creating a Date Object

```python
from datetime import date

# Current date
today = date.today()
print("Today's date:", today)

# Specific date
specific_date = date(2023, 10, 5)
print("Specific date:", specific_date)
```

#### Accessing Date Attributes

```python
print("Year:", today.year)
print("Month:", today.month)
print("Day:", today.day)
```

---

### `datetime.time`

Represents a time (hour, minute, second, microsecond).

#### Creating a Time Object

```python
from datetime import time

# Specific time
t = time(14, 30, 45)  # 2:30:45 PM
print("Time:", t)
```

#### Accessing Time Attributes

```python
print("Hour:", t.hour)
print("Minute:", t.minute)
print("Second:", t.second)
print("Microsecond:", t.microsecond)
```

---

### `datetime.datetime`

Represents both a date and a time.

#### Creating a Datetime Object

```python
from datetime import datetime

# Current datetime
now = datetime.now()
print("Current datetime:", now)

# Specific datetime
dt = datetime(2023, 10, 5, 14, 30, 45)
print("Specific datetime:", dt)
```

#### Accessing Datetime Attributes

```python
print("Year:", dt.year)
print("Month:", dt.month)
print("Day:", dt.day)
print("Hour:", dt.hour)
print("Minute:", dt.minute)
print("Second:", dt.second)
```

---

### `datetime.timedelta`

Represents the difference between two dates or times.

#### Creating a Timedelta Object

```python
from datetime import timedelta

# Difference of 5 days
delta = timedelta(days=5)
print("Timedelta:", delta)
```

#### Arithmetic with Timedelta

```python
future_date = today + timedelta(days=10)
print("Date after 10 days:", future_date)
```

---

### `datetime.tzinfo`

Base class for time zone information. Used with `datetime` objects to handle time zones.

## 3. Common Operations

### Formatting Dates and Times

Use the `strftime()` method to format dates and times as strings.

```python
from datetime import datetime

now = datetime.now()
formatted_date = now.strftime("%Y-%m-%d %H:%M:%S")
print("Formatted datetime:", formatted_date)
```

Common format codes:

- `%Y`: Year (e.g., 2023)
- `%m`: Month (01-12)
- `%d`: Day (01-31)
- `%H`: Hour (00-23)
- `%M`: Minute (00-59)
- `%S`: Second (00-59)

---

### Parsing Strings to Datetime Objects

Use the `strptime()` method to parse strings into `datetime` objects.

```python
from datetime import datetime

date_string = "2023-10-05 14:30:45"
parsed_date = datetime.strptime(date_string, "%Y-%m-%d %H:%M:%S")
print("Parsed datetime:", parsed_date)
```

---

### Arithmetic with Dates and Times

Use `timedelta` for arithmetic operations.

```python
from datetime import datetime, timedelta

now = datetime.now()
future = now + timedelta(days=7)
past = now - timedelta(hours=12)

print("Future datetime:", future)
print("Past datetime:", past)
```

---

### Working with Time Zones

Use the `pytz` library (not part of the standard library) for time zone support.

```python
from datetime import datetime
import pytz

# Localize a datetime object
local_time = datetime.now()
timezone = pytz.timezone("America/New_York")
localized_time = timezone.localize(local_time)
print("Localized time:", localized_time)

# Convert to another time zone
new_timezone = pytz.timezone("Europe/London")
converted_time = localized_time.astimezone(new_timezone)
print("Converted time:", converted_time)
```

## 4. Practical Examples

### Example 1: Calculate Age

```python
from datetime import date

def calculate_age(birthdate):
    today = date.today()
    age = today.year - birthdate.year
    if (today.month, today.day) < (birthdate.month, birthdate.day):
        age -= 1
    return age

birthdate = date(1990, 5, 15)
print("Age:", calculate_age(birthdate))
```

---

### Example 2: Countdown to a Future Date

```python
from datetime import datetime, timedelta

future_date = datetime(2023, 12, 31)
today = datetime.now()
time_left = future_date - today

print("Days left:", time_left.days)
```

---

### Example 3: Format a Datetime Object

```python
from datetime import datetime

now = datetime.now()
formatted = now.strftime("%A, %B %d, %Y %I:%M %p")
print("Formatted datetime:", formatted)
```

---

## 5. Best Practices

- Use `datetime` objects instead of strings for date and time manipulation.
- Always handle time zones explicitly using libraries like `pytz`.
- Use `strftime()` and `strptime()` for formatting and parsing.
- Prefer `timedelta` for date and time arithmetic.

- `date`: Represents a date (year, month, day).
- `time`: Represents a time (hour, minute, second, microsecond).
- `datetime`: Represents both date and time.
- `timedelta`: Represents the difference between two dates or times.
- `tzinfo`: Base class for time zone information.

To use the `datetime` module, import it:

```python
import datetime
```

---

## 2. Key Classes in the `datetime` Module

### `datetime.date`

Represents a date (year, month, day).

#### Creating a `date` Object

```python
from datetime import date

# Current date
today = date.today()
print("Today's date:", today)

# Specific date
specific_date = date(2023, 10, 5)
print("Specific date:", specific_date)
```

#### Common Methods and Attributes

- `year`, `month`, `day`: Access the year, month, and day.
- `weekday()`: Returns the day of the week (0 = Monday, 6 = Sunday).
- `isoweekday()`: Returns the day of the week (1 = Monday, 7 = Sunday).
- `strftime(format)`: Formats the date as a string.

```python
print("Year:", today.year)
print("Month:", today.month)
print("Day:", today.day)
print("Weekday:", today.weekday())
print("ISO Weekday:", today.isoweekday())
print("Formatted date:", today.strftime('%Y-%m-%d'))
```

---

### `datetime.time`

Represents a time (hour, minute, second, microsecond).

#### Creating a `time` Object

```python
from datetime import time

# Specific time
t = time(14, 30, 45)  # 2:30:45 PM
print("Time:", t)
```

#### Common Methods and Attributes

- `hour`, `minute`, `second`, `microsecond`: Access time components.
- `strftime(format)`: Formats the time as a string.

```python
print("Hour:", t.hour)
print("Minute:", t.minute)
print("Second:", t.second)
print("Formatted time:", t.strftime('%H:%M:%S'))
```

---

### `datetime.datetime`

Represents both date and time.

#### Creating a `datetime` Object

```python
from datetime import datetime

# Current date and time
now = datetime.now()
print("Current date and time:", now)

# Specific date and time
specific_datetime = datetime(2023, 10, 5, 14, 30, 45)
print("Specific date and time:", specific_datetime)
```

#### Common Methods and Attributes

- `date()`: Extracts the date part.
- `time()`: Extracts the time part.
- `strftime(format)`: Formats the datetime as a string.

```python
print("Date part:", now.date())
print("Time part:", now.time())
print("Formatted datetime:", now.strftime('%Y-%m-%d %H:%M:%S'))
```

---

### `datetime.timedelta`

Represents the difference between two dates or times.

#### Creating a `timedelta` Object

```python
from datetime import timedelta

# Difference of 5 days
delta = timedelta(days=5)
print("Timedelta:", delta)
```

#### Common Operations

- Add or subtract `timedelta` from `date` or `datetime`.

```python
future_date = now + timedelta(days=10)
print("Date 10 days from now:", future_date)
```

---

### `datetime.tzinfo`

Base class for time zone information.

#### Example: Using `pytz` for Time Zones

```python
from datetime import datetime
import pytz

# Localize a datetime to a specific time zone
local_time = datetime.now()
timezone = pytz.timezone('America/New_York')
localized_time = timezone.localize(local_time)
print("Localized time:", localized_time)
```

## 3. Common Operations

### Formatting Dates and Times

Use `strftime()` to format dates and times.

```python
from datetime import datetime

now = datetime.now()
print("Formatted datetime:", now.strftime('%Y-%m-%d %H:%M:%S'))
```

### Parsing Strings to Dates

Use `strptime()` to parse strings into `datetime` objects.

```python
from datetime import datetime

date_string = "2023-10-05 14:30:45"
parsed_date = datetime.strptime(date_string, '%Y-%m-%d %H:%M:%S')
print("Parsed datetime:", parsed_date)
```

### Arithmetic with Dates and Times

Use `timedelta` for arithmetic operations.

```python
from datetime import datetime, timedelta

now = datetime.now()
future_date = now + timedelta(days=10)
past_date = now - timedelta(days=5)
print("Future date:", future_date)
print("Past date:", past_date)
```

### Working with Time Zones

Use `pytz` or `zoneinfo` (Python 3.9+) for time zone support.

```python
from datetime import datetime
import pytz

# Convert to a different time zone
utc_time = datetime.now(pytz.utc)
print("UTC time:", utc_time)

ny_time = utc_time.astimezone(pytz.timezone('America/New_York'))
print("New York time:", ny_time)
```

## 4. Practical Examples

### Example 1: Calculate Age

```python
from datetime import date

def calculate_age(birthdate):
    today = date.today()
    age = today.year - birthdate.year
    if (today.month, today.day) < (birthdate.month, birthdate.day):
        age -= 1
    return age

birthdate = date(1990, 5, 15)
print("Age:", calculate_age(birthdate))
```

### Example 2: Countdown to a Specific Date

```python
from datetime import datetime, timedelta

target_date = datetime(2023, 12, 31)
today = datetime.now()
time_left = target_date - today
print("Days left:", time_left.days)
```

### Example 3: Format a Date for Display

```python
from datetime import datetime

now = datetime.now()
print("Formatted date:", now.strftime('%A, %B %d, %Y'))
```

```python
from datetime import datetime

dt = datetime(2023, 10, 5, 14, 30, 45)
print(dt)  # Output: 2023-10-05 14:30:45
```

- `now()`: Returns the current local date and time.
- `utcnow()`: Returns the current UTC date and time.
- `combine(date, time)`: Combines a `date` and `time` object.
- `strftime(format)`: Formats the datetime as a string.
- `strptime(string, format)`: Parses a string into a `datetime` object.

#### **Example:**

```python
print(datetime.now())  # Output: Current date and time
print(dt.strftime("%Y-%m-%d %H:%M:%S"))  # Output: 2023-10-05 14:30:45

# Parsing a string
dt_str = "2023-10-05 14:30:45"
dt_parsed = datetime.strptime(dt_str, "%Y-%m-%d %H:%M:%S")
print(dt_parsed)  # Output: 2023-10-05 14:30:45
```
