# Python `abs()`

The **`abs()`** method returns the absolute value of the given number. If the number is a complex number, **`abs()`** returns its magnitude.

**Syntax**:

```python
abs(num)
```


## `abs()` Parameters

**`abs()`** method takes a single argument:

- **num** - a number whose absolute value is to be returned. The number can be:
  - integer
  - floating number
  - complex number


## Return value from `abs()`

**`abs()`** method returns the absolute value of the given number.

- For integers - integer absolute value is returned
- For floating numbers - floating absolute value is returned
- For complex numbers - magnitude of the number is returned


````python
# Example 1: Get absolute value of a number

# random integer
integer = -20
print('Absolute value of -20 is:', abs(integer))

#random floating number
floating = -30.33
print('Absolute value of -30.33 is:', abs(floating))
````

````python
# Example 2: Get magnitude of a complex number

# random complex number
complex = (3 - 4j)
print('Magnitude of 3 - 4j is:', abs(complex))
````

The `abs()` function in Python is a built-in function used to compute the absolute value of a number. The absolute value of a number is its distance from zero on the number line, regardless of its direction. This function works with integers, floats, and complex numbers, making it versatile for various mathematical and programming tasks.

Below is a comprehensive guide to the `abs()` function, covering everything from basic usage to advanced techniques.

---

## **1. Basic Usage of `abs()`**

### **Syntax**

```python
abs(x)
```

- **Parameter**: `x` (a number or an object that implements `__abs__()`).
- **Return Value**: The absolute value of `x`.

### **Examples**

```python
# Absolute value of an integer
print(abs(-5))  # Output: 5

# Absolute value of a float
print(abs(-3.14))  # Output: 3.14

# Absolute value of a complex number
print(abs(3 + 4j))  # Output: 5.0 (magnitude of the complex number)
```

---

## **2. Working with Different Input Types**

### **Integers**

- Computes the absolute value of an integer.

```python
print(abs(-10))  # Output: 10
```

### **Floats**

- Computes the absolute value of a floating-point number.

```python
print(abs(-2.718))  # Output: 2.718
```

### **Complex Numbers**

- Computes the magnitude (Euclidean norm) of a complex number.

```python
print(abs(3 + 4j))  # Output: 5.0 (sqrt(3^2 + 4^2))
```

### **Custom Objects**

- If an object implements the `__abs__()` method, `abs()` can be used on it.

```python
class Distance:
    def __init__(self, value):
        self.value = value

    def __abs__(self):
        return abs(self.value)

d = Distance(-10)
print(abs(d))  # Output: 10
```

---

## **3. Advanced Usage**

### **Absolute Value in Mathematical Computations**

- Use `abs()` in mathematical formulas and algorithms.

```python
# Distance between two points on a number line
point1 = -5
point2 = 3
distance = abs(point1 - point2)
print(distance)  # Output: 8
```

### **Handling Edge Cases**

- `abs()` works seamlessly with edge cases like zero and infinity.

```python
print(abs(0))  # Output: 0
print(abs(float("inf")))  # Output: inf
print(abs(float("-inf")))  # Output: inf
```

### **Absolute Value in Data Analysis**

- Use `abs()` to compute absolute differences in datasets.

```python
data = [1, -2, 3, -4, 5]
absolute_data = [abs(x) for x in data]
print(absolute_data)  # Output: [1, 2, 3, 4, 5]
```

---

## **4. Practical Applications**

### **Error Calculation**

- Use `abs()` to compute the absolute error between predicted and actual values.

```python
actual = 10
predicted = 12
error = abs(actual - predicted)
print(error)  # Output: 2
```

### **Distance Metrics**

- Compute the Manhattan distance between two points.

```python
point1 = (1, 2)
point2 = (4, 6)
manhattan_distance = abs(point1[0] - point2[0]) + abs(point1[1] - point2[1])
print(manhattan_distance)  # Output: 7
```

### **Signal Processing**

- Use `abs()` to compute the magnitude of complex signals.

```python
import numpy as np

signal = np.array([1 + 2j, 3 + 4j, 5 + 6j])
magnitude = np.abs(signal)
print(magnitude)  # Output: [2.236 5.     7.810]
```

---

## **5. Combining `abs()` with Other Functions**

### **Using `abs()` with `map()`**

- Apply `abs()` to all elements in a list.

```python
data = [-1, -2, -3, -4]
absolute_data = list(map(abs, data))
print(absolute_data)  # Output: [1, 2, 3, 4]
```

### **Using `abs()` with `filter()`**

- Filter out negative values from a list.

```python
data = [-1, 2, -3, 4]
positive_data = list(filter(lambda x: abs(x) == x, data))
print(positive_data)  # Output: [2, 4]
```

### **Using `abs()` with `numpy`**

- Compute the absolute value of all elements in a NumPy array.

```python
import numpy as np

arr = np.array([-1, -2, -3, -4])
absolute_arr = np.abs(arr)
print(absolute_arr)  # Output: [1 2 3 4]
```

## **6. Common Errors and Pitfalls**

### **Non-Numeric Input**

- Passing a non-numeric type (e.g., string) to `abs()` raises a `TypeError`.

```python
# print(abs("hello"))  # Raises TypeError: bad operand type for abs(): 'str'
```

### **Custom Objects Without `__abs__()`**

- If a custom object does not implement `__abs__()`, calling `abs()` on it raises a `TypeError`.

```python
class Distance:
    def __init__(self, value):
        self.value = value

d = Distance(-10)
# print(abs(d))  # Raises TypeError: bad operand type for abs(): 'Distance'
```
