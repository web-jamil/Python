# Python `int()`

The **`int()`** method returns an integer object from any number or string.

**Syntax**:

```python
int(x=0, base=10)
```


## `int()` Parameters

**`int()`** method takes two arguments:

- **x** - Number or string to be converted to integer object.
  - The default argument is **`zero`**.
- **base** - Base of the number in x.
  - Can be 0 (code literal) or 2-36.


## Return Value from `int()`

**`int()`** method returns:

- an integer object from the given number or string treats default base as 10
- (No parameters) returns 0
- (If base given) treats the string in the given base (0, 2, 8, 10, 16)


````python
# Example 1: How int() works in Python?

# integer
print("int(123) is:", int(123))

# float
print("int(123.23) is:", int(123.23))

# string
print("int('123') is:", int('123'))
````

````python
# Example 2: How int() works for decimal, octal and hexadecimal?

# binary 0b or 0B
print("For 1010, int is:", int('1010', 2))
print("For 0b1010, int is:", int('0b1010', 2))

# octal 0o or 0O
print("For 12, int is:", int('12', 8))
print("For 0o12, int is:", int('0o12', 8))

# hexadecimal
print("For A, int is:", int('A', 16))
print("For 0xA, int is:", int('0xA', 16))
````

````python
# Example 3: int() for custom objects

# Internally, int() method calls an object's __int__() method.
# So, even if an object isn't a number, you can convert the object into an integer object.
# You can do this by overriding __index__() and __int__() methods of the class to return a number.
# These two methods should return the same value as older versions of Python uses __int__(), 
# while newer uses __index__() method.

class Person:
    age = 23

    def __index__(self):
        return self.age
    
    def __int__(self):
        return self.age

person = Person()
print('int(person) is:', int(person))
````

In Python, errors related to integers can arise in various situations, including improper operations, type mismatches, or invalid input. These errors are typically caught as **exceptions** in Python, which you can handle using `try-except` blocks. Let’s go through the most common **integer-related errors** in Python, along with explanations and examples.

### 1. **TypeError**

A `TypeError` occurs when an operation or function is applied to an object of inappropriate type. This can happen when an operation involving integers is done incorrectly, or when integers are mixed with incompatible types.

#### Example 1: Adding an integer and a string

```python
x = 10
y = "5"
# This will raise a TypeError because you cannot add an integer and a string
result = x + y  # TypeError: unsupported operand type(s) for +: 'int' and 'str'
```

#### Example 2: Trying to multiply a string by a float

```python
x = "Hello"
y = 2.5
# This will raise a TypeError because multiplying a string by a float is not allowed
result = x * y  # TypeError: can't multiply sequence by non-int of type 'float'
```

### 2. **ValueError**

A `ValueError` occurs when a function receives an argument of the right type, but an inappropriate value. This is common when trying to convert non-numeric strings to integers.

#### Example 1: Trying to convert a non-numeric string to an integer

```python
x = "abc"
# This will raise a ValueError because "abc" cannot be converted to an integer
result = int(x)  # ValueError: invalid literal for int() with base 10: 'abc'
```

#### Example 2: Passing an invalid string to `int()` with a base

```python
x = "123abc"
# This will raise a ValueError because the string contains non-numeric characters
result = int(x, 10)  # ValueError: invalid literal for int() with base 10: '123abc'
```

### 3. **ZeroDivisionError**

A `ZeroDivisionError` occurs when you try to divide a number by zero, which is undefined in mathematics.

#### Example:

```python
x = 10
y = 0
# This will raise a ZeroDivisionError because division by zero is not allowed
result = x / y  # ZeroDivisionError: division by zero
```

### 4. **OverflowError**

An `OverflowError` happens when an arithmetic operation exceeds the limits of the data type (in this case, the integer type). However, Python's `int` type is of arbitrary precision, so `OverflowError` is more likely to occur in other contexts like numerical libraries or when working with fixed-size types.

#### Example:

```python
import sys

# This will likely raise an OverflowError on platforms with limited system memory
max_size = sys.maxsize
result = max_size * 2  # OverflowError: Python int too large to convert to C long
```

However, Python `int` doesn't typically throw an `OverflowError` like in other languages (e.g., C, Java). Instead, it can grow to accommodate very large numbers. But you may encounter overflow-related issues in specific libraries or implementations.

### 5. **IndexError** (Related to Integer Indices)

An `IndexError` occurs when trying to access an index that is out of the range of a sequence, such as a list or tuple. While this is not strictly an integer error, it often happens when working with integers as indices.

#### Example:

```python
my_list = [1, 2, 3]
# This will raise an IndexError because there is no index 5 in the list
result = my_list[5]  # IndexError: list index out of range
```

### 6. **TypeError: Integer Division with Floats (Python 2.x behavior)**

In Python 2.x, the `/` operator performs integer division if both operands are integers. In Python 3.x, it always returns a float, and integer division is done using `//`. If you perform an operation expecting integer division behavior in Python 3.x, you might encounter this type of issue.

#### Example (Python 2.x behavior):

```python
# In Python 2.x, this will perform integer division and return 2
x = 5
y = 2
result = x / y  # In Python 2.x, result would be 2, not 2.5
```

In Python 3.x, this is handled correctly using `//` for integer division:

```python
# In Python 3.x, this will return a float
x = 5
y = 2
result = x / y  # Output: 2.5 (float)

# Integer division in Python 3.x
integer_result = x // y  # Output: 2 (integer division)
```

### 7. **AssertionError (Integer-Related Logic)**

An `AssertionError` might be used to validate integer conditions in your code, such as ensuring an integer is within a valid range. If the assertion fails, Python will raise this error.

#### Example:

```python
x = 10
# This will raise an AssertionError because the condition is false
assert x < 5, "x must be less than 5"  # AssertionError: x must be less than 5
```

### 8. **MemoryError (Rare in Python with `int`)**

A `MemoryError` can occur when you attempt to create an extremely large integer object that exceeds available memory. Python's `int` type is dynamically allocated, but in rare cases, an integer can consume excessive memory.

#### Example:

```python
# Generating a very large integer (This will not throw an error in normal circumstances)
large_int = 10 ** 10000
```

In rare circumstances, attempting to work with extremely large numbers might exceed memory limits, and Python will raise a `MemoryError`.

### 9. **TypeError: Unsupported Operand Type(s) for +**

This error can occur if you're trying to perform an operation between an integer and a non-integer type. While this overlaps with `TypeError` discussed earlier, this specific example happens with a variety of operations.

#### Example:

```python
x = 10
y = [1, 2, 3]
# This will raise a TypeError because you cannot add a list and an integer
result = x + y  # TypeError: unsupported operand type(s) for +: 'int' and 'list'
```

### 10. **ValueError: Too Many Values to Unpack**

This error happens when you are trying to unpack a sequence with an unexpected number of elements, which may involve integers.

#### Example:

```python
x, y = [1, 2, 3]
# This will raise a ValueError because there are too many elements to unpack
```

You can handle these errors using `try-except` blocks to prevent your program from crashing and to manage the error flow properly.

An **AssertionError** in Python occurs when an `assert` statement fails. The `assert` statement is used to debug code by ensuring that a certain condition is true. If the condition is false, an `AssertionError` is raised, and the program stops execution (unless the error is caught and handled). Below is a comprehensive guide to **AssertionError**, covering everything from basic concepts to advanced topics, including examples and explanations.

---

## **1. What is an AssertionError?**

- **Definition**: An `AssertionError` is raised when an `assert` statement fails, indicating that a condition expected to be true is actually false.
- **Purpose**: Used for debugging and validating assumptions in code.
- **Example**:
  ```python
  x = 5
  assert x == 10, "x should be 10"  # Raises AssertionError
  ```

---

## **2. Why Use Assertions?**

- **Debugging**: Assertions help catch logical errors during development.
- **Validation**: Ensure that inputs, outputs, or intermediate states meet expected conditions.
- **Documentation**: Assertions act as self-documenting checks in the code.

---

  ```python
      return n * factorial(n - 1)

  ```

## **7. Common Pitfalls and Best Practices**

### **7.1. Do Not Use Assertions for Data Validation**

- **Best Practice**: Use proper error handling (e.g., `if` statements, exceptions) for data validation.
- **Example**:

  ```python

      if b == 0:
          raise ValueError("b cannot be zero")
      return a / b

  result = divide(10, 0)  # Raises ValueError
  ```

### **7.2. Use Assertions for Debugging Only**

- **Pitfall**: Assertions are not a substitute for proper error handling.
- **Best Practice**: Use assertions to catch logical errors during development, not for handling runtime errors.

### **7.3. Provide Clear Error Messages**

- **Pitfall**: Vague error messages make debugging difficult.
- **Best Practice**: Provide clear and descriptive error messages in assertions.
- **Example**:
  ```python

  assert x == 5, f"Expected x to be 5, but got {x}"
  ```

An `AssertionError` is raised when an `assert` statement fails, indicating that a condition expected to be true is actually false. Key points to remember:

In Python, you can assign **maximum** and **minimum** values to variables using built-in constants, functions, and libraries. These values are often used in algorithms, data validation, and mathematical computations. Below is a comprehensive guide to assigning and working with **maximum** and **minimum** values in Python, covering everything from basic concepts to advanced topics.

---

## **1. Maximum and Minimum Values for Integers**

### **1.1. Using `sys.maxsize`**

- **Purpose**: Represents the maximum size of an integer that can be used for indexing (platform-dependent).
- **Example**:
  ```python
  import sys
  max_int = sys.maxsize
  min_int = -sys.maxsize - 1
  print(f"Max integer: {max_int}, Min integer: {min_int}")
  # Output: Max integer: 9223372036854775807, Min integer: -9223372036854775808 (on 64-bit systems)
  ```

### **1.2. Using Arbitrary-Precision Integers**

- **Purpose**: Python integers have arbitrary precision, so there is no strict maximum or minimum value.
- **Example**:
  ```python

  print(large_int)
  ```

---

## **2. Maximum and Minimum Values for Floating-Point Numbers**

### **2.1. Using `sys.float_info`**

- **Purpose**: Provides information about the limits of floating-point numbers.
- **Example**:
  ```python
  import sys
  max_float = sys.float_info.max
  min_float = sys.float_info.min
  ```
