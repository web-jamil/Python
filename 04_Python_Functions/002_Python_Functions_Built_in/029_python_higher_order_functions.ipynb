A **higher-order function** in Python is a function that either:

1. **Takes another function as an argument**, or
2. **Returns a function as its output**.

This allows for more flexible, reusable, and concise code, making functional programming techniques possible in Python.

---

### **Key Features of Higher-Order Functions**

- **Encapsulation**: Helps keep code modular and reusable.
- **Abstraction**: Reduces redundancy by treating functions like variables.
- **Functional Programming**: Enables concepts like **map, filter, and reduce**.

---

### **Examples of Higher-Order Functions in Python**

Here are some common higher-order functions:

#### **1. Passing a Function as an Argument**

```python
def greet(name):
    return f"Hello, {name}!"

def execute_function(func, value):
    return func(value)

result = execute_function(greet, "Jamil")
print(result)  # Output: Hello, Jamil!
```

#### **2. Returning a Function**

```python
def multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

double = multiplier(2)
print(double(5))  # Output: 10
```

#### **3. Using Built-in Higher-Order Functions**

Python provides several built-in functions that demonstrate this concept:

- **`map()`**: Applies a function to each element in an iterable.
- **`filter()`**: Selects elements based on a function.
- **`reduce()`** (from `functools` module): Applies a cumulative operation.

```python
from functools import reduce

# Using map to square each number

print(evens)  # Output: [2, 4]

# Using reduce to compute the product of all numbers
product = reduce(lambda x, y: x * y, numbers)
print(product)  # Output: 120
```

**What are Higher-Order Functions?**

In Python, higher-order functions are functions that can do one or both of the following:

1.  **Accept other functions as arguments.**
2.  **Return other functions as their results.**

Essentially, they treat functions as first-class citizens, meaning you can manipulate them just like any other data type (like integers, strings, or lists). This opens up possibilities for abstraction and creating more flexible and reusable code.

**Why Use Higher-Order Functions?**

- **Abstraction:** They allow you to abstract away common patterns of computation. For example, instead of writing separate loops to process lists in different ways, you can use higher-order functions like `map` or `filter` to express the intent more clearly.
- **Code Reusability:** By accepting functions as arguments, you can create general-purpose functions that can be customized by passing different behaviors.
- **Readability:** When used effectively, higher-order functions can make your code more readable and easier to understand by focusing on _what_ you want to do rather than _how_ to do it.
- **Functional Programming Paradigm:** Higher-order functions are a cornerstone of functional programming, which emphasizes immutability and treating computation as the evaluation of mathematical functions.

**Syntax and Examples:**

Let's look at the syntax and some common examples of higher-order functions in Python:

**1. Accepting Functions as Arguments:**

```python
def greet(func, name):
    """A higher-order function that calls the passed function."""
    greeting = func(name)
    print(greeting)

def say_hello(name):
    return f"Hello, {name}!"

def say_goodbye(name):
    return f"Goodbye, {name}."

greet(say_hello, "Alice")  # Output: Hello, Alice!
greet(say_goodbye, "Bob")  # Output: Goodbye, Bob.
```

In this example:

- `greet` is a higher-order function because it accepts another function (`func`) as an argument.
- `say_hello` and `say_goodbye` are regular functions that are passed as arguments to `greet`.

**2. Returning Functions:**

```python
def create_multiplier(factor):
    """A higher-order function that returns another function."""
    def multiplier(x):
        return x * factor
    return multiplier

double = create_multiplier(2)
triple = create_multiplier(3)

print(double(5))   # Output: 10
print(triple(5))   # Output: 15
```

Here:

- `create_multiplier` is a higher-order function because it returns another function (`multiplier`).
- The returned `multiplier` function "remembers" the `factor` from its enclosing scope (this is a concept called closure).

**Built-in Higher-Order Functions in Python:**

Python has several powerful built-in higher-order functions that you'll encounter frequently:

- **`map(function, iterable, ...)`:** Applies the `function` to each item of the `iterable` (e.g., a list) and returns an iterator of the results.

  ```python
  numbers = [1, 2, 3, 4]
  squared = map(lambda x: x**2, numbers)
  print(list(squared))  # Output: [1, 4, 9, 16]
  ```

- **`filter(function, iterable)`:** Constructs an iterator from elements of the `iterable` for which the `function` returns true.

  ```python
  numbers = [1, 2, 3, 4, 5, 6]
  even_numbers = filter(lambda x: x % 2 == 0, numbers)
  print(list(even_numbers))  # Output: [2, 4, 6]
  ```

- **`reduce(function, iterable[, initializer])`:** Applies a `function` of two arguments cumulatively to the items of an `iterable`, from left to right, so as to reduce the iterable to a single value. You need to import `reduce` from the `functools` module in Python 3.

  ```python
  from functools import reduce

  numbers = [1, 2, 3, 4]
  product = reduce(lambda x, y: x * y, numbers)
  print(product)  # Output: 24
  ```

- **`sorted(iterable, *, key=None, reverse=False)`:** Although technically a built-in function, the `key` argument allows you to pass a function that will be called on each list element prior to making comparisons. This makes it behave like a higher-order function.

  ```python
  students = [
      {"name": "Alice", "grade": "B"},
      {"name": "Bob", "grade": "A"},
      {"name": "Charlie", "grade": "C"},
  ]

  sorted_by_grade = sorted(students, key=lambda student: student["grade"])
  print(sorted_by_grade)
  # Output: [{'name': 'Bob', 'grade': 'A'}, {'name': 'Alice', 'grade': 'B'}, {'name': 'Charlie', 'grade': 'C'}]
  ```

**Using Lambda Functions with Higher-Order Functions:**

Anonymous (lambda) functions are often used in conjunction with higher-order functions to define simple functions inline without needing a formal `def` statement.

```python
numbers = [10, 20, 5, 15]
doubled_and_filtered = filter(lambda x: x > 10, map(lambda y: y * 2, numbers))
print(list(doubled_and_filtered))  # Output: [20, 30]
```

In this example, two lambda functions are used with `map` and `filter` to perform operations concisely.

Employing higher-order functions can lead to several benefits:

## Syntax and Examples

Let's explore the syntax and see some concrete examples of how higher-order functions work in Python.

### 1. Accepting Functions as Arguments

A higher-order function can take one or more functions as arguments. Here's the basic syntax:

```python
def higher_order_function(func, *args, **kwargs):
    # Do something with the 'func' argument
    result = func(*args, **kwargs)
    return result
```

In this syntax:

- `higher_order_function` is our higher-order function.
- `func` is a parameter that is expected to be a function.
- `*args` and `**kwargs` allow the higher-order function to pass arbitrary positional and keyword arguments to the function it receives.

**Example 1: The `map()` Function**

The built-in `map()` function is a classic example of a higher-order function. It takes a function and an iterable as arguments and applies the function to each item in the iterable, returning an iterator of the results.

```python
def square(x):
    return x * x

numbers = [1, 2, 3, 4, 5]
squared_numbers = map(square, numbers)
print(list(squared_numbers))  # Output: [1, 4, 9, 16, 25]
```

In this case, the `map()` function takes the `square` function as an argument and applies it to each element in the `numbers` list.

**Example 2: The `filter()` Function**

Another built-in higher-order function is `filter()`. It takes a function (that returns a boolean) and an iterable. It filters the iterable, keeping only the items for which the function returns `True`.

```python
def is_even(x):
    return x % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = filter(is_even, numbers)
print(list(even_numbers))  # Output: [2, 4, 6]
```

Here, `filter()` uses the `is_even` function to determine which numbers from the `numbers` list should be included in the `even_numbers` result.

**Example 3: A Custom Higher-Order Function**

Let's create our own higher-order function that applies a given operation to a list of numbers:

```python
def apply_operation(numbers, operation):
    results = []
    for number in numbers:
        results.append(operation(number))
    return results

def multiply_by_two(x):
    return x * 2

def add_five(x):
    return x + 5

numbers = [10, 20, 30]
multiplied = apply_operation(numbers, multiply_by_two)
added = apply_operation(numbers, add_five)

print(f"Multiplied by two: {multiplied}")  # Output: Multiplied by two: [20, 40, 60]
print(f"Added five: {added}")          # Output: Added five: [15, 25, 35]
```

In this example, `apply_operation` takes a list of numbers and a function (`operation`) as arguments. It then applies the `operation` function to each number in the list.

### 2. Returning Functions as Results

A higher-order function can also return another function. Here's the general syntax:

```python
def outer_function(parameter):
    def inner_function(x):
        # 'parameter' is accessible in the inner function (closure)
        return x + parameter
    return inner_function

add_ten = outer_function(10)
result = add_ten(5)
print(result)  # Output: 15

add_five = outer_function(5)
result2 = add_five(3)
print(result2) # Output: 8
```

In this case:

- `outer_function` takes a `parameter` and defines an `inner_function`.
- `inner_function` uses the `parameter` from the outer scope (this is called a closure).
- `outer_function` returns the `inner_function` itself (not the result of calling it).

The returned functions (`add_ten` and `add_five`) "remember" the value of `parameter` from their enclosing scope.

**Example: Creating Different Kinds of Multipliers**

```python
def create_multiplier(factor):
    def multiplier(x):
        return x * factor
    return multiplier

multiply_by_3 = create_multiplier(3)
multiply_by_10 = create_multiplier(10)

print(multiply_by_3(7))   # Output: 21
print(multiply_by_10(2))  # Output: 20
```

Here, `create_multiplier` is a higher-order function that returns a new function (`multiplier`). The returned `multiplier` function is specialized to multiply its input by the `factor` that was passed to `create_multiplier`.

## **1. Decorators: Enhancing Functions**

A **decorator** is a higher-order function that modifies or extends the behavior of another function without altering its code. It is commonly used for logging, authentication, caching, and other useful tasks.

### **Syntax for a Decorator**

```python
def decorator_function(original_function):
    def wrapper_function():
        print("Enhancing function behavior!")
        return original_function()
    return wrapper_function

@decorator_function  # Applying the decorator
def greet():
    print("Hello, Jamil!")

greet()
```

### **Output:**

```
Enhancing function behavior!
Hello, Jamil!
```

#### **Example: Timing Function Execution**

```python
import time

def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Execution time: {end_time - start_time:.5f} seconds")
        return result
    return wrapper

@timer_decorator
def compute():
    time.sleep(1)
    print("Function executed!")

compute()
```

### **Key Takeaways**

- **Decorators** enhance a function while keeping its original logic intact.
- **`@decorator_name`** syntax makes applying decorators elegant and readable.

## **2. Currying: Breaking Down Function Parameters**

**Currying** is a technique where a function is **broken down into multiple functions**, each taking one argument at a time. It helps in creating reusable and composable functions.

### **Currying in Python**

```python
def multiply(x):
    def inner(y):
        return x * y
    return inner

double = multiply(2)  # Creates a function that multiplies by 2
print(double(5))  # Output: 10
```

### **Using `functools.partial` for Currying**

Python’s `functools.partial` simplifies currying by pre-fixing arguments:

```python
from functools import partial

def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
print(square(4))  # Output: 16
```

### **Key Takeaways**

- **Currying** transforms multi-parameter functions into a chain of single-parameter functions.
- It enhances **function reusability and composition**.

# Higher-Order Functions in Python

Higher-order functions are functions that can take other functions as arguments and/or return functions as their result. Python supports higher-order functions natively, and they're a fundamental concept in functional programming.

## Basic Syntax

### 1. Functions as Arguments

```python
def apply_operation(func, x, y):
    return func(x, y)

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

result1 = apply_operation(add, 5, 3)       # Returns 8
result2 = apply_operation(multiply, 5, 3)  # Returns 15
```

### 2. Functions as Return Values

```python
def create_greeting(greeting):
    def greet(name):
        return f"{greeting}, {name}!"
    return greet

say_hello = create_greeting("Hello")
say_hi = create_greeting("Hi")

print(say_hello("Alice"))  # Output: "Hello, Alice!"
print(say_hi("Bob"))       # Output: "Hi, Bob!"
```

## Built-in Higher-Order Functions

Python includes several built-in higher-order functions:

### 1. `map(function, iterable)`

Applies a function to all items in an iterable.

```python
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
# squared is [1, 4, 9, 16, 25]
```

### 2. `filter(function, iterable)`

Filters items based on a function that returns True or False.

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
# evens is [2, 4, 6]
```

### 3. `sorted(iterable, key=None, reverse=False)`

Sorts an iterable with an optional key function.

```python
words = ["apple", "banana", "cherry", "date"]
sorted_by_length = sorted(words, key=lambda x: len(x))
# sorted_by_length is ['date', 'apple', 'banana', 'cherry']
```

### 4. `functools.reduce(function, iterable[, initializer])`

Successively applies a function to elements to reduce the iterable to a single value.

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
# product is 120 (1*2*3*4*5)
```

## Decorators (Advanced Higher-Order Functions)

Decorators are functions that modify the behavior of other functions.

### Basic Decorator Syntax

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
# Output:
# Something is happening before the function is called.
# Hello!
# Something is happening after the function is called.
```

### Decorator with Arguments

```python
def repeat(num_times):
    def decorator_repeat(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator_repeat

@repeat(num_times=3)
def greet(name):
    print(f"Hello {name}")

greet("Alice")
# Output:
# Hello Alice
# Hello Alice
# Hello Alice
```

````python
# Defining mathematical operations
def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    return x / y if y != 0 else "Cannot divide by zero"

# Higher-order function that takes a function and two numbers
def calculate(operation, a, b):
    return operation(a, b)

# Testing the higher-order function
print(calculate(add, 10, 5))       # Output: 15
print(calculate(subtract, 10, 5))  # Output: 5
print(calculate(multiply, 10, 5))  # Output: 50
print(calculate(divide, 10, 5))    # Output: 2.0
````
