The **`typing` module** in Python provides support for type hints, which allow you to specify the expected types of variables, function arguments, and return values. Type hints improve code readability, enable better tooling support (e.g., linters, IDEs), and help catch type-related errors early. Below is a comprehensive guide to the `typing` module, covering everything from basic syntax to advanced usage.

---

## **1. Basic Concepts**

### **What are Type Hints?**

- Type hints are annotations that indicate the expected type of a variable, function argument, or return value.
- They are optional and do not affect runtime behavior (Python remains dynamically typed).

### **Why Use Type Hints?**

- **Readability**: Makes code easier to understand.
- **Tooling Support**: Enables better autocompletion, linting, and error checking in IDEs.
- **Documentation**: Acts as a form of documentation for your code.

---

## **2. Basic Syntax**

### **Type Hinting Variables**

- Use a colon (`:`) followed by the type to annotate variables.

#### **Example**

```python
name: str = "Alice"
age: int = 30
is_student: bool = True
```

---

### **Type Hinting Functions**

- Use `->` to specify the return type of a function.

#### **Example**

```python
def greet(name: str) -> str:
    return f"Hello, {name}!"
```

---

### **Type Hinting Collections**

- Use `List`, `Dict`, `Set`, and `Tuple` from the `typing` module to annotate collections.

#### **Example**

```python
from typing import List, Dict, Set, Tuple

names: List[str] = ["Alice", "Bob", "Charlie"]
ages: Dict[str, int] = {"Alice": 30, "Bob": 25}
unique_numbers: Set[int] = {1, 2, 3}
coordinates: Tuple[float, float] = (10.5, 20.5)
```

- Creates a new distinct type.

#### **Example**

```python
from typing import NewType

UserId = NewType('UserId', int)

def get_user(user_id: UserId) -> str:
    return f"User {user_id}"

user_id = UserId(123)
print(get_user(user_id))
```

---

### **5. `Protocol`**

- Represents a structural type (duck typing).

#### **Example**

```python
from typing import Protocol

class Printable(Protocol):
    def print(self) -> None:
        pass

class Document:
    def print(self) -> None:
        print("Printing document")

def print_item(item: Printable) -> None:
    item.print()
```

The **`typing` module** in Python provides support for type hints, which allow you to specify the expected types of variables, function arguments, and return values. Type hints improve code readability, enable better tooling support (e.g., linters, IDEs), and help catch type-related errors early. Below is a comprehensive guide to the `typing` module, covering everything from basic syntax to advanced usage.

---

## **1. Basic Concepts**

### **What are Type Hints?**

- Type hints are annotations that indicate the expected type of a variable, function argument, or return value.
- They are not enforced at runtime but are used by static type checkers (e.g., `mypy`) and IDEs.

### **Why Use Type Hints?**

- **Readability**: Makes code easier to understand.
- **Tooling Support**: Enables better autocompletion, linting, and error detection.
- **Documentation**: Acts as a form of documentation for your code.

## **2. Basic Type Hints**

### **Syntax**

```python
variable: type = value
```

### **Example**

```python
name: str = "Alice"
age: int = 30
is_student: bool = True
```

---

### **Function Annotations**

- You can annotate function arguments and return values.

#### **Syntax**

```python
def function_name(arg1: type, arg2: type) -> return_type:
    pass
```

#### **Example**

```python
def greet(name: str) -> str:
    return f"Hello, {name}!"

print(greet("Alice"))  # Output: Hello, Alice!
```

## **4. Advanced Usage**

### **1. Generics**

- Use `Generic` and `TypeVar` to create reusable generic classes and functions.

#### **Example**

```python
from typing import Generic, TypeVar, List

T = TypeVar('T')

class Box(Generic[T]):
    def __init__(self, item: T):
        self.item = item

    def get_item(self) -> T:
        return self.item

box = Box[int](10)
print(box.get_item())  # Output: 10
```

---

### **2. Protocols (Structural Subtyping)**

- Use `Protocol` to define interfaces or abstract base classes.

#### **Example**

```python
from typing import Protocol

class Printable(Protocol):
    def print(self) -> None:
        pass

class Document:
    def print(self) -> None:
        print("Printing document")

def print_item(item: Printable) -> None:
    item.print()

doc = Document()
print_item(doc)  # Output: Printing document
```

---

### **3. Overloading Functions**

- Use `@overload` to define multiple function signatures for a single function.

#### **Example**

```python
from typing import overload

@overload
def process(data: int) -> int: ...
@overload
def process(data: str) -> str: ...

def process(data):
    if isinstance(data, int):
        return data * 2
    elif isinstance(data, str):
        return data.upper()

print(process(10))    # Output: 20
print(process("abc")) # Output: ABC
```

---

### **4. Annotating Class Methods**

- Use type hints in class methods, including `self` and `cls`.

#### **Example**

```python
class Calculator:
    def add(self, a: int, b: int) -> int:
        return a + b

    @classmethod
    def multiply(cls, a: int, b: int) -> int:
        return a * b
```

---

### **5. Annotating `*args` and `**kwargs`\*\*

- Use `*args` and `**kwargs` with type hints.

#### **Example**

```python
from typing import Tuple, Dict

def process_args(*args: int, **kwargs: str) -> None:
    print("Args:", args)
    print("Kwargs:", kwargs)

process_args(1, 2, 3, name="Alice", city="New York")
```

## **4. Advanced Usage**

### **1. Generics**

- Use `Generic` and `TypeVar` to create reusable, type-safe components.

#### **Example**

```python
from typing import Generic, TypeVar, List

T = TypeVar('T')

class Stack(Generic[T]):
    def __init__(self):
        self.items: List[T] = []

    def push(self, item: T) -> None:
        self.items.append(item)

    def pop(self) -> T:
        return self.items.pop()

stack: Stack[int] = Stack()
stack.push(1)
stack.push(2)
print(stack.pop())  # Output: 2
```

---

### **2. `Protocol`**

- Used to define structural types (duck typing).

#### **Example**

```python
from typing import Protocol

class Printable(Protocol):
    def print(self) -> None:
        pass

class Document:
    def print(self) -> None:
        print("Printing document")

def print_item(item: Printable) -> None:
    item.print()

doc = Document()
print_item(doc)  # Output: Printing document
```

### **3. `Annotated`**

- Used to add metadata to type hints.

#### **Example**

```python
from typing import Annotated

def process_data(data: Annotated[str, "Metadata"]) -> None:
    print(data)

process_data("Hello")  # Output: Hello
```

### **4. `NewType`**

- Used to create distinct types.

#### **Example**

```python
from typing import NewType

UserId = NewType('UserId', int)

def get_user(user_id: UserId) -> None:
    print(f"Fetching user with ID {user_id}")

user_id = UserId(123)
get_user(user_id)  # Output: Fetching user with ID 123
```
