# Python `id()`

The **`id()`** function returns identity (unique integer) of an object.

**Syntax**:

```python
id(object)
```


## `id()` Parameters

**`id()`** function takes a single parameter **`object`**.


## Return Value from `id()`

**`id()`** function returns the identity of the object. This is an integer that is unique for the given object and remains constant during its lifetime.


````python
# Example 1: How id() works?
class Foo:
    b = 5

dummyFoo = Foo()
print('id of dummyFoo =',id(dummyFoo))
````

````python
# More Examples on id()

print('id of 5 =',id(5))

a = 5
print('id of a =',id(a))

b = a
print('id of b =',id(b))

c = 5.0
print('id of c =',id(c))
````

It's important to note that everything in Python is an object, even numbers, and Classes.

Hence, integer **`5`** has a unique id. The id of the integer **`5`** remains constant during the lifetime. Similar is the case for float **`5.5`** and other objects.


The **`id()`** function in Python is used to return the **identity** of an object. This identity is a unique integer that is constant for the object during its lifetime. It is often used to check if two variables refer to the same object in memory. Below is a comprehensive guide to the **`id()`** function, covering its usage from basic to advanced levels.

---

## **Basic Usage of `id()`**

### 1. **Get the Identity of an Object**

- The `id()` function returns a unique integer representing the memory address of the object.
- **Example**:
  ```python
  x = 42
  print(id(x))  # Output: A unique integer (e.g., 140735680123456)
  ```

---

### 2. **Compare Identities of Two Objects**

- Use `id()` to check if two variables refer to the same object.
- **Example**:
  ```python
  a = [1, 2, 3]
  b = a
  print(id(a) == id(b))  # True (a and b refer to the same object)
  ```

---

### 3. **Identity of Immutable vs. Mutable Objects**

- Immutable objects (e.g., integers, strings) may have the same `id` if their values are the same due to Python's memory optimization.
- Mutable objects (e.g., lists, dictionaries) always have unique `id`s.
- **Example**:

  ```python
  x = 10
  y = 10
  print(id(x) == id(y))  # True (immutable objects with the same value may share id)

  a = [1, 2, 3]
  b = [1, 2, 3]
  print(id(a) == id(b))  # False (mutable objects always have unique ids)
  ```

---

## **Intermediate Usage of `id()`**

### 4. **Check if Two Variables Point to the Same Object**

- Use `id()` to verify if two variables are aliases of the same object.
- **Example**:
  ```python
  list1 = [1, 2, 3]
  list2 = list1
  print(id(list1) == id(list2))  # True (both variables point to the same list)
  ```

---

### 5. **Identity of Function Arguments**

- Use `id()` to track how objects are passed to functions (by reference or by value).
- **Example**:

  ```python
  def modify_list(lst):
      print(id(lst))  # Same as the original list's id
      lst.append(4)

  my_list = [1, 2, 3]
  print(id(my_list))  # Original list's id
  modify_list(my_list)
  print(my_list)  # Output: [1, 2, 3, 4]
  ```

---

### 6. **Identity of Class Instances**

- Use `id()` to check the identity of class instances.
- **Example**:

  ```python
  class MyClass:
      pass

  obj1 = MyClass()
  obj2 = MyClass()
  print(id(obj1) == id(obj2))  # False (each instance has a unique id)
  ```

---

### 7. **Identity of Strings**

- Python may reuse memory for identical immutable strings (interning).
- **Example**:
  ```python
  str1 = "hello"
  str2 = "hello"
  print(id(str1) == id(str2))  # True (string interning)
  ```

---

## **Advanced Usage of `id()`**

### 8. **Identity of Integers**

- Small integers (between -5 and 256) are cached in Python, so they share the same `id`.
- **Example**:

  ```python
  a = 10
  b = 10
  print(id(a) == id(b))  # True (small integers are cached)

  c = 1000
  d = 1000
  print(id(c) == id(d))  # False (large integers are not cached)
  ```

---

### 9. **Identity of Custom Objects**

- Use `id()` to track the identity of custom objects and their attributes.
- **Example**:

  ```python
  class Person:
      def __init__(self, name):
          self.name = name

  person1 = Person("Alice")
  person2 = Person("Bob")
  print(id(person1))  # Unique id for person1
  print(id(person2))  # Unique id for person2
  ```

---

### 10. **Identity of `None`**

- `None` is a singleton in Python, so all `None` objects share the same `id`.
- **Example**:
  ```python
  a = None
  b = None
  print(id(a) == id(b))  # True (None is a singleton)
  ```

---

### 11. **Identity of Tuples**

- Tuples are immutable, so identical tuples may share the same `id` due to Python's memory optimization.
- **Example**:
  ```python
  tuple1 = (1, 2, 3)
  tuple2 = (1, 2, 3)
  print(id(tuple1) == id(tuple2))  # True (immutable tuples may share id)
  ```

---

### 12. **Identity of Sets**

- Sets are mutable, so each set has a unique `id`.
- **Example**:
  ```python
  set1 = {1, 2, 3}
  set2 = {1, 2, 3}
  print(id(set1) == id(set2))  # False (mutable sets have unique ids)
  ```

---

### 13. **Identity of Functions**

- Functions are objects in Python, so each function has a unique `id`.
- **Example**:

  ```python
  def my_function():
      pass

  print(id(my_function))  # Unique id for the function
  ```

---

### 14. **Identity of Lambda Functions**

- Lambda functions are also objects with unique `id`s.
- **Example**:
  ```python
  lambda_func = lambda x: x * 2
  print(id(lambda_func))  # Unique id for the lambda function
  ```

---

### 15. **Identity of Class Methods**

- Class methods are bound to the class and have unique `id`s.
- **Example**:

  ```python
  class MyClass:
      def my_method(self):
          pass

  obj = MyClass()
  print(id(obj.my_method))  # Unique id for the bound method
  ```

---

## **Real-World Examples**

### 16. **Check if Two Variables Refer to the Same Object**

```python
list1 = [1, 2, 3]
list2 = list1
print(id(list1) == id(list2))  # True (both variables refer to the same list)
```

### 17. **Track Object Identity in a Loop**

```python
for i in range(3):
    print(id(i))  # Output: Unique ids for each integer
```

### 18. **Identity of Objects in a List**

```python
my_list = [1, "hello", [1, 2, 3]]
for item in my_list:
    print(id(item))  # Output: Unique ids for each item in the list
```

### 19. **Identity of Objects in a Dictionary**

```python
my_dict = {"a": 1, "b": 2, "c": 3}
for key, value in my_dict.items():
    print(f"Key: {key}, ID: {id(key)}")
    print(f"Value: {value}, ID: {id(value)}")
```

### 20. **Identity of Objects in a Set**

```python
my_set = {1, 2, 3}
for item in my_set:
    print(id(item))  # Output: Unique ids for each item in the set
```

- Check if two variables refer to the same object.
- Track object identities in loops, lists, dictionaries, and sets.
- Understand how Python handles memory for immutable and mutable objects.
