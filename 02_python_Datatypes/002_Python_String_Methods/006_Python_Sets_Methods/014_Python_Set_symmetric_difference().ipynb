# Python Set `symmetric_difference()`

The **`symmetric_difference()`** method returns the symmetric difference of two sets.

The symmetric difference of two sets **`A`** and **`B`** is the set of elements that are in either **`A`** or **`B`**, but not in their intersection.

<img src="img/symmetric_difference.png" width="250"/>
</div>

Here, set **`B`** is a superset of set **`A`** and **`A`** is a subset of set **`B`**.


**Syntax**:

```python
A.symmetric_difference(B)
```


````python
# Example 1: Working of symmetric_difference()

A = {'a', 'b', 'c', 'd'}
B = {'c', 'd', 'e' }
C = {}

print(A.symmetric_difference(B))
print(B.symmetric_difference(A))

print(A.symmetric_difference(C))
print(B.symmetric_difference(C))
````

## Symmetric difference using `^` operator

In Python, we can also find the symmetric difference using the **`^`** operator.


````python
A = {'a', 'b', 'c', 'd'}
B = {'c', 'd', 'e' }

print(A ^ B)
print(B ^ A)

print(A ^ A)
print(B ^ B)
````

### Python Set `symmetric_difference()` Method: Concepts and Theory

The **`symmetric_difference()`** method in Python is used to return a set that contains all the elements that are in either of the two sets, but **not in both**. This method is useful when you want to find the unique elements between two sets—i.e., the elements that appear in one set or the other, but not in both.

7. **Edge Cases and Exceptions**
8. **Alternative Method: `^` Operator**

---

### 1. **Definition and Syntax of `symmetric_difference()`**

#### **Syntax**:

```python
set1.symmetric_difference(set2)
```

- **`set1`**: The first set involved in the symmetric difference operation.
- **`set2`**: The second set involved in the symmetric difference operation.

#### **Return Value**:

- **`symmetric_difference()`** returns a new set containing the elements that are in either `set1` or `set2`, but not in both. The result excludes any elements that are shared between the two sets.

---

### 2. **How `symmetric_difference()` Works**

- The **symmetric difference** of two sets contains all the elements that are **unique to each set**. In other words, if an element appears in **both** sets, it will be excluded from the result.
- If you apply **`symmetric_difference()`** to two sets, the resulting set will only include those elements that belong to one set or the other, but not to both.

#### Example:

If you have two sets, `set1 = {1, 2, 3}` and `set2 = {3, 4, 5}`, the **symmetric difference** will return `{1, 2, 4, 5}` because `3` is the common element in both sets and is excluded.

---

### 3. **Example Usage**

#### Example 1: Basic Symmetric Difference of Two Sets

```python
# Define two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Perform symmetric difference operation
result_set = set1.symmetric_difference(set2)

# Output the result
print(result_set)  # Output: {1, 2, 4, 5}
```

- **Explanation**: The elements `1`, `2`, `4`, and `5` are unique to `set1` and `set2`, while the element `3` is present in both sets and is therefore excluded from the result.

#### Example 2: Symmetric Difference with Multiple Sets

```python
# Define three sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set3 = {5, 6, 7}

# Perform symmetric difference between three sets
result_set = set1.symmetric_difference(set2).symmetric_difference(set3)

# Output the result
print(result_set)  # Output: {1, 2, 4, 6, 7}
```

- **Explanation**: The symmetric difference of all three sets will give the elements that are unique across all sets, excluding any common elements.

#### Example 3: Symmetric Difference with an Empty Set

```python
# Define a set and an empty set
set1 = {1, 2, 3}
set_empty = set()

# Perform symmetric difference operation
result_set = set1.symmetric_difference(set_empty)

# Output the result
print(result_set)  # Output: {1, 2, 3}
```

- **Explanation**: The symmetric difference of a set and an empty set is just the original set itself, as there are no elements in the empty set to exclude.

---

### 4. **Return Value of `symmetric_difference()`**

- The **`symmetric_difference()`** method returns a new set that contains all the elements that are unique to either of the sets.
- **Non-destructive**: The original sets are not modified during the operation; instead, a new set is created and returned.

Example:

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1.symmetric_difference(set2)

print(result)  # Output: {1, 2, 4, 5}
print(set1)    # Output: {1, 2, 3}  # original set remains unchanged
print(set2)    # Output: {3, 4, 5}
```

---

### 5. **Time Complexity of `symmetric_difference()`**

- **Time Complexity**: The time complexity of **`symmetric_difference()`** is **O(len(s) + len(t))**, where `s` is the first set and `t` is the second set. This is because the method needs to iterate through both sets to compute the symmetric difference.
  - The complexity is linear with respect to the total number of elements in both sets.

---

### 6. **When to Use `symmetric_difference()`**

#### Use Cases for **`symmetric_difference()`**:

1. **Finding Unique Elements Between Two Sets**: The **`symmetric_difference()`** method is perfect for situations where you need to find the elements that exist in one set but not in the other. This is useful when comparing two groups of items and finding differences.

   - Example:
     ```python
     # Comparing two lists of items and finding unique items
     set_a = {"apple", "banana", "cherry"}
     set_b = {"banana", "cherry", "date"}
     unique_items = set_a.symmetric_difference(set_b)
     print(unique_items)  # Output: {'apple', 'date'}
     ```

2. **Identifying Discrepancies Between Two Datasets**: When working with datasets, symmetric difference helps identify elements that are not shared between two datasets.

   - Example:
     ```python
     students_set1 = {"Alice", "Bob", "Charlie"}
     students_set2 = {"Charlie", "David", "Eve"}
     result = students_set1.symmetric_difference(students_set2)
     print(result)  # Output: {'Alice', 'Bob', 'David', 'Eve'}
     ```

3. **Filter Out Common Elements**: When you want to filter out the elements that are common in both sets and retain only those that are unique to either set, use **`symmetric_difference()`**.

---

### 7. **Edge Cases and Exceptions**

#### Edge Case 1: Symmetric Difference with Empty Set

The symmetric difference of a set with an empty set will be the original set, as there are no elements to exclude from the empty set.

```python
set1 = {1, 2, 3}
empty_set = set()
result = set1.symmetric_difference(empty_set)
print(result)  # Output: {1, 2, 3}
```

#### Edge Case 2: Symmetric Difference of Identical Sets

If two sets are identical, the symmetric difference will be an empty set because there are no unique elements in either set.

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3}
result = set1.symmetric_difference(set2)
print(result)  # Output: set()
```

#### Edge Case 3: Symmetric Difference with Other Iterables

The **`symmetric_difference()`** method works not only with sets but also with other iterables (like lists, tuples, or dictionaries). However, when working with a dictionary, only the keys are considered.

```python
set1 = {1, 2, 3}
list1 = [3, 4, 5]
result = set1.symmetric_difference(list1)
print(result)  # Output: {1, 2, 4, 5}

# Using a dictionary
dict1 = {"a": 1, "b": 2}
result = set1.symmetric_difference(dict1)
print(result)  # Output: {1, 2, 3, 'a', 'b'}
```

---

### 8. **Alternative Method: `^` Operator**

In addition to the **`symmetric_difference()`** method, you can use the **`^`** operator to perform the symmetric difference operation. This operator provides a more concise and readable way to achieve the same result.

#### Example with `^` Operator:

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1 ^ set2
print(result)  # Output: {1, 2, 4, 5}
```

- **Explanation**: The `^` operator is shorthand for the **`symmetric_difference()`** method. It gives the same result as using the method but is often more convenient for simple expressions.

---

### Conclusion

The **`symmetric_difference()`** method is a useful tool in Python for finding the unique elements between two sets. It eliminates the common elements and provides a set of all the elements that appear in only one of the sets. Whether you're comparing datasets, analyzing differences, or simply finding the unique items in two sets, **`symmetric_difference()`** is an essential operation for working with sets in Python.
