# Python Set `union()`

The **`union()`** method returns a new set with distinct elements from all the sets.

The union of two or more sets is the set of all distinct elements present in all the sets. For example:

```python
A = {1, 2}
B = {2, 3, 4}
C = {5}

Then,
A∪B = B∪A = {1, 2, 3, 4}
A∪C = C∪A = {1, 2, 5}
B∪C = C∪B = {2, 3, 4, 5}

A∪B∪C = {1, 2, 3, 4, 5}
```

**Syntax**:

```python
A.union(*other_sets)
```

> **Note:** **`*`** is not part of the syntax. It is used to indicate that the method can take 0 or more arguments.


## Return Value from `union()`

- The **`union()`** method returns a new set with elements from the set and all other sets (passed as an argument).
- If the argument is not passed to **`union()`**, it returns a shallow copy of the set.


````python
# Example 1: Working of union()

A = {'a', 'c', 'd'}
B = {'c', 'd', 2 }
C = {1, 2, 3}

print('A U B =', A.union(B))
print('B U C =', B.union(C))
print('A U B U C =', A.union(B, C))
print('A.union() =', A.union())
````

You can also find the union of sets using the **`|`** operator.


````python
# Example 2: Set Union Using the | Operator

A = {'a', 'c', 'd'}
B = {'c', 'd', 2 }
C = {1, 2, 3}

print('A U B =', A| B)
print('B U C =', B | C)
print('A U B U C =', A | B | C)
````

### Python Set `union()` Method: Concepts and Theory

The **`union()`** method in Python is used to combine elements from two or more sets, returning a new set that contains all unique elements from the sets involved. This method is essential when you want to combine sets while ensuring that there are no duplicate elements, as sets inherently only store unique elements.

In this section, we will cover:

1. **Definition and Syntax of `union()`**
2. **How `union()` Works**
3. **Example Usage**
4. **Return Value of `union()`**
5. **Time Complexity of `union()`**
6. **When to Use `union()`**
7. **Edge Cases and Exceptions**
8. **Alternative Method: `|` Operator**

---

### 1. **Definition and Syntax of `union()`**

#### **Syntax**:

```python
set1.union(set2, set3, ...)
```

- **`set1`**: The first set involved in the union operation.
- **`set2, set3, ...`**: One or more additional sets (or iterables) to be united with `set1`. You can pass any number of sets or iterables.

#### **Return Value**:

- **`union()`** returns a new set that contains all the unique elements from the provided sets or iterables.

---

### 2. **How `union()` Works**

- The **`union()`** method combines all elements from the sets (and other iterables) provided to it.
- **Uniqueness**: The union operation automatically eliminates any duplicate elements, so the resulting set contains only distinct elements.
- **Multiple Sets or Iterables**: You can pass multiple sets or iterables to the **`union()`** method. It will combine them all into a new set.
- **Unordered**: Since sets are unordered collections, the order of elements in the resulting union is not guaranteed.

#### Example:

If you have two sets, `set1 = {1, 2, 3}` and `set2 = {3, 4, 5}`, calling `set1.union(set2)` will return a set containing all unique elements from both sets, i.e., `{1, 2, 3, 4, 5}`.

---

### 3. **Example Usage**

#### Example 1: Basic Union of Two Sets

```python
# Define two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Perform union operation
result_set = set1.union(set2)

# Output the result
print(result_set)  # Output: {1, 2, 3, 4, 5}
```

- **Explanation**: The resulting set contains all unique elements from both `set1` and `set2`. The duplicate `3` is only included once.

#### Example 2: Union of Multiple Sets

```python
# Define three sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set3 = {5, 6, 7}

# Perform union operation
result_set = set1.union(set2, set3)

# Output the result
print(result_set)  # Output: {1, 2, 3, 4, 5, 6, 7}
```

- **Explanation**: The result contains unique elements from all three sets. The duplicates (`3`, `5`) are excluded.

#### Example 3: Union of a Set with an Iterable

```python
# Define a set
set1 = {1, 2, 3}

# Define a list (an iterable)
list1 = [4, 5, 6]

# Perform union operation
result_set = set1.union(list1)

# Output the result
print(result_set)  # Output: {1, 2, 3, 4, 5, 6}
```

- **Explanation**: The list `list1` is treated as an iterable, and its elements are added to `set1`, resulting in a new set with all unique elements.

#### Example 4: Union with an Empty Set

```python
# Define a set and an empty set
set1 = {1, 2, 3}
set2 = set()

# Perform union operation
result_set = set1.union(set2)

# Output the result
print(result_set)  # Output: {1, 2, 3}
```

- **Explanation**: The union of a set and an empty set is the original set itself, as there are no elements in the empty set to add.

---

### 4. **Return Value of `union()`**

- The **`union()`** method returns a new set containing all unique elements from the sets or iterables involved in the union operation.
- **Non-destructive**: The original sets remain unchanged.

Example:

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1.union(set2)

print(result)  # Output: {1, 2, 3, 4, 5}
print(set1)    # Output: {1, 2, 3}  # original set remains unchanged
print(set2)    # Output: {3, 4, 5}
```

---

### 5. **Time Complexity of `union()`**

- **Time Complexity**: The time complexity of **`union()`** is **O(len(s) + len(t))**, where `s` is the first set and `t` is the second set (or other iterables). This is because the method iterates over the elements of both sets and adds them to a new set.
  - If you're performing a union on multiple sets or iterables, the complexity would be proportional to the total number of elements involved.

---

### 6. **When to Use `union()`**

#### Use Cases for **`union()`**:

1. **Combining Sets**: The **`union()`** method is useful when you need to merge two or more sets and eliminate duplicates. This is common in problems involving multiple data sources, where you want a unique set of items.

   - Example:
     ```python
     # Merging sets of students from different classes
     set_class_a = {"John", "Alice", "Tom"}
     set_class_b = {"Tom", "Emma", "Sophia"}
     unique_students = set_class_a.union(set_class_b)
     print(unique_students)  # Output: {'Alice', 'John', 'Sophia', 'Emma', 'Tom'}
     ```

2. **Combining Sets with Other Iterables**: You can combine sets with lists, tuples, or other iterables using **`union()`**. This is useful when your data comes in various formats, and you need to consolidate them into a single set.

   - Example:
     ```python
     set1 = {1, 2, 3}
     list1 = [4, 5, 6]
     result = set1.union(list1)
     print(result)  # Output: {1, 2, 3, 4, 5, 6}
     ```

3. **Eliminating Duplicates Across Multiple Data Sources**: If you have multiple data sources, and you need to remove duplicates while merging them, **`union()`** is a perfect choice.
   - Example:
     ```python
     set1 = {"apple", "banana", "cherry"}
     set2 = {"cherry", "date", "fig"}
     combined = set1.union(set2)
     print(combined)  # Output: {'banana', 'apple', 'cherry', 'date', 'fig'}
     ```

---

### 7. **Edge Cases and Exceptions**

#### Edge Case 1: Union with an Empty Set

If you perform a union operation with an empty set, the resulting set will be the non-empty set.

```python
set1 = {1, 2, 3}
set_empty = set()
result = set1.union(set_empty)
print(result)  # Output: {1, 2, 3}
```

#### Edge Case 2: Union with Identical Sets

If you union two identical sets, the result will be the same set, as no new unique elements are introduced.

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3}
result = set1.union(set2)
print(result)  # Output: {1, 2, 3}
```

#### Edge Case 3: Union with Other Iterables

The **`union()`** method can handle iterables like lists, tuples, and dictionaries. However, when using a dictionary, only the keys will be included in the union operation.

```python
set1 = {1, 2, 3}
dict1 = {"a": 10, "b": 20}
result = set1.union(dict1)
print(result)  # Output: {1, 2, 3, 'a', 'b'}
```

- **Explanation**: Only the keys `'a'` and `'b'` from the dictionary are added to the set.

---

### 8. **Alternative Method: `|` Operator**

In addition to the **`union()`** method, you can use the **`|`** operator to perform a union of sets. The **`|`** operator provides a more concise and readable way to combine sets.

#### Example with `|` Operator:

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1 | set2
print(result)  # Output: {1, 2, 3, 4, 5}
```

- **Explanation**: The `|` operator performs the same union operation as **`union()`**, but is often more convenient for short expressions.
