# Python Set `copy()`

The **`copy()`** method adds a given element to a set. If the element is already present, it doesn't add any element.


A **[set](https://github.com/milaan9/02_Python_Datatypes/blob/main/006_Python_Sets.ipynb)** can be copied using = operator in Python. For example:

```python
numbers = {1, 2, 3, 4}
new_numbers = numbers
```

The problem with copying the set in this way is that if you modify the **`numbers`** set, the **`new_numbers`** set is also modified.


````python
# Example:

numbers = {1, 2, 3, 4}
new_numbers = numbers

new_numbers.add(5)

print('numbers: ', numbers)
print('new_numbers: ', new_numbers)
````

However, if you need the original set to be unchanged when the new set is modified, you can use the **`copy()`** method.

**Syntax**:

```python
set.copy()
```


## `copy()` Parameters

The **`copy()`** method doesn't take any parameters.


## Return Value from `copy()`

The **`copy()`** method returns a shallow copy of the set.


````python
# Example 1: How the copy() method works for sets?

numbers = {1, 2, 3, 4}
new_numbers = numbers.copy()

new_numbers.add(5)

print('numbers: ', numbers)
print('new_numbers: ', new_numbers)
````

### Python Set `copy()` Method: Concepts and Theory

The **`copy()`** method in Python is used to create a shallow copy of a set. A shallow copy means that a new set is created with the same elements as the original set, but it is a completely independent object in memory. Any changes made to the new set will not affect the original set, and vice versa.

### 1. Definition and Syntax

The **`copy()`** method is used to return a shallow copy of the set. This means that a new set is created, and all the elements of the original set are copied over. However, if the elements of the set are mutable objects, the reference to those objects is copied, but the objects themselves are not duplicated (this is the "shallow" part).

#### **Syntax**:

```python
set_copy = set.copy()
```

- **`set`**: The original set from which the copy will be created.
- **`set_copy`**: The new set that is created as a shallow copy.

#### **Return Value**:

- A new set that contains the same elements as the original set.

---

### 2. How `copy()` Works

The **`copy()`** method creates a new set that has the same elements as the original set. The new set is independent of the original set, meaning changes to one set do not affect the other.

#### Key Points:

- **Shallow Copy**: If the set contains references to mutable objects (like lists or dictionaries), the references to those objects are copied, but the objects themselves are not duplicated. Hence, changes to mutable objects inside the set will affect both the original and the copied set.
- **Independent Sets**: If the set only contains immutable objects (e.g., integers, strings, tuples), the sets are truly independent.

#### Example:

```python
# Original Set
original_set = {1, 2, 3}

# Creating a copy of the set
copied_set = original_set.copy()

# Modifying the copied set
copied_set.add(4)

print("Original Set:", original_set)  # Output: {1, 2, 3}
print("Copied Set:", copied_set)      # Output: {1, 2, 3, 4}
```

- **Explanation**: In this case, the copied set is a completely independent object. Modifying the copied set does not affect the original set.

---

### 3. Examples of Using `copy()`

#### Example 1: Creating a Copy of a Set of Integers

```python
# Original Set of integers
original_set = {10, 20, 30, 40}

# Create a shallow copy
new_set = original_set.copy()

# Display the sets
print("Original Set:", original_set)  # Output: {10, 20, 30, 40}
print("New Set:", new_set)            # Output: {10, 20, 30, 40}
```

- **Explanation**: The `copy()` method creates a new set with the same elements, and any changes to one set do not affect the other.

#### Example 2: Copying a Set Containing Mutable Objects

```python
# Set containing mutable elements (lists)
original_set = {(1, 2), (3, 4), (5, 6)}

# Create a shallow copy
copied_set = original_set.copy()

# Display the sets
print("Original Set:", original_set)  # Output: {(1, 2), (3, 4), (5, 6)}
print("Copied Set:", copied_set)      # Output: {(1, 2), (3, 4), (5, 6)}
```

- **Explanation**: Both sets have the same elements. In this case, since the elements are tuples (which are immutable), the sets are truly independent.

#### Example 3: Modifying the Copied Set

```python
# Set of strings
original_set = {"apple", "banana", "cherry"}

# Copy the set
new_set = original_set.copy()

# Add a new element to the copied set
new_set.add("date")

# Display the sets
print("Original Set:", original_set)  # Output: {'apple', 'banana', 'cherry'}
print("Copied Set:", new_set)         # Output: {'apple', 'banana', 'cherry', 'date'}
```

- **Explanation**: The new set is independent of the original set, and modifying it (like adding "date") does not affect the original set.

---

### 4. Shallow Copy vs Deep Copy

- **Shallow Copy**: The **`copy()`** method creates a shallow copy of the set. This means that if the set contains mutable elements (e.g., lists, dictionaries), only the references to these mutable elements are copied. Therefore, changes made to the mutable objects in the copied set will also be reflected in the original set.

- **Deep Copy**: A **deep copy** creates a completely independent copy of the set, including copies of all objects referenced by the set. A deep copy is needed when you want to make a fully independent copy of the set and its contained objects.

Python's **`copy`** module provides a function for deep copying, **`copy.deepcopy()`**, which creates a true deep copy of the set and all objects within it.

#### Example of Shallow Copy:

```python
# Set containing mutable elements (lists)
original_set = {frozenset([1, 2]), frozenset([3, 4])}

# Create a shallow copy
new_set = original_set.copy()

# Modifying the mutable object inside the copied set
new_set = set(frozenset([5, 6]))  # deep copy by constructing new instance directly

# Display the sets
print("Original Set:", original_set)  # Output: original set output
print("New Set:", new_set)      # Output: new modified set
```
