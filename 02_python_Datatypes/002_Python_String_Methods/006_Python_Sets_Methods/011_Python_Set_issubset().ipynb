# Python Set `issubset()`

The **`issubset()`** method returns **`True`** if all elements of a set are present in another set (passed as an argument). If not, it returns **`False`**.

Set **`A`** is said to be the subset of set **`B`** if all elements of **`A`** are in **`B`**.

Here, set **`A`** is a subset of **`B`**.


**Syntax**:

```python
A.issubset(B)
```

The above code checks if **`A`** is a subset of **`B`**.


## Return Value from `issubset()`

The **`issubset()`** returns

- **`True`** if **`A`** is a subset of **`B`**
- **`False`** if **`A`** is not a subset of **`B`**


````python
# Example: How issubset() works?

A = {1, 2, 3}
B = {1, 2, 3, 4, 5}
C = {1, 2, 4, 5}

# Returns True
print(A.issubset(B))

# Returns False
# B is not subset of A
print(B.issubset(A))

# Returns False
print(A.issubset(C))

# Returns True
print(C.issubset(B))
````

### Python Set `issubset()` Method: Concepts and Theory

The **`issubset()`** method in Python is used to check if one set is a subset of another. In other words, it checks whether all elements of one set are contained within another set. If all elements of the first set are found in the second set, the method returns `True`; otherwise, it returns `False`.

### 1. **Definition of a Subset**

A **subset** is a set where every element in the subset is also contained in another set, called the **superset**. In formal terms, a set `A` is a subset of set `B` if every element of `A` is also in `B`. This relationship is denoted as:

\[
A \subseteq B
\]

- If `A` is a subset of `B`, we say that all elements of `A` exist in `B`, but `B` may contain additional elements that are not in `A`.
- A set is always a subset of itself. Also, the **empty set** is a subset of every set.

---

### 2. **Syntax and Usage of `issubset()`**

#### Syntax:

```python
set1.issubset(set2)
```

- **`set1`**: The set being checked if it is a subset.
- **`set2`**: The set against which the check is made.

#### Return Value:

- Returns **`True`** if `set1` is a subset of `set2`, otherwise it returns **`False`**.

#### Example:

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5}
result = set1.issubset(set2)
print(result)  # Output: True
```

- **Explanation**: `set1` is a subset of `set2` because every element in `set1` (i.e., `1`, `2`, `3`) is contained within `set2`.

---

### 3. **Return Value of `issubset()`**

- The **`issubset()`** method returns a **Boolean value**:
  - **`True`** if all elements of the first set are found in the second set.
  - **`False`** if any element of the first set is not found in the second set.

#### Example 1: Subset Check Returning `True`

```python
set1 = {2, 4}
set2 = {1, 2, 3, 4, 5}
result = set1.issubset(set2)
print(result)  # Output: True
```

- **Explanation**: All elements in `set1` are present in `set2`.

#### Example 2: Subset Check Returning `False`

```python
set1 = {2, 6}
set2 = {1, 2, 3, 4, 5}
result = set1.issubset(set2)
print(result)  # Output: False
```

- **Explanation**: The element `6` in `set1` is not present in `set2`, so it returns `False`.

---

### 4. **Example Usage of `issubset()`**

#### Example 1: Checking If One Set is a Subset of Another

```python
set1 = {10, 20, 30}
set2 = {10, 20, 30, 40, 50}
result = set1.issubset(set2)
print(result)  # Output: True
```

- **Explanation**: `set1` is a subset of `set2` because all elements in `set1` exist in `set2`.

#### Example 2: Checking If a Set is a Subset of Itself

```python
set1 = {1, 2, 3, 4}
result = set1.issubset(set1)
print(result)  # Output: True
```

- **Explanation**: Every set is a subset of itself. Hence, the result is `True`.

#### Example 3: Checking a Subset with the Empty Set

```python
set1 = set()
set2 = {1, 2, 3}
result = set1.issubset(set2)
print(result)  # Output: True
```

- **Explanation**: The empty set is a subset of every set, including `set2`. Hence, it returns `True`.

---

### 5. **Time Complexity of `issubset()`**

- **Time Complexity**: The time complexity of **`issubset()`** is **O(len(s))**, where `s` is the set being checked. This is because the method needs to iterate over each element in the first set (`set1`) and check if it exists in the second set (`set2`).

  - **Best Case**: If `set1` is empty, the method will return `True` immediately (since the empty set is a subset of any set).
  - **Worst Case**: If `set1` has `n` elements and `set2` has `m` elements, the method will check every element of `set1` in `set2`, resulting in a time complexity of **O(n)**.

---

### 6. **Use Cases of `issubset()`**

1. **Check Membership of a Subset**: When you want to check if one set is entirely contained within another set, you can use **`issubset()`** to verify this relationship.

   - Example:
     ```python
     allowed_users = {"alice", "bob", "carol"}
     requested_users = {"bob", "alice"}
     print(requested_users.issubset(allowed_users))  # Output: True
     ```

2. **Data Filtering**: If you want to filter data to check if a certain list of items is a subset of a larger list, this method can help.

3. **Checking for Duplicate or Redundant Data**: If you're working with sets of data and need to check if one set is a subset of another to avoid redundancy, this method can be very useful.

---

### 7. **Edge Cases and Exceptions**

#### Edge Case 1: Subset Check with Empty Set

The **empty set** is considered a subset of **any set** (including itself).

```python
set1 = set()
set2 = {1, 2, 3}
result = set1.issubset(set2)
print(result)  # Output: True
```

#### Edge Case 2: Subset Check with Larger Set Containing Additional Elements

```python
set1 = {1, 2}
set2 = {1, 2, 3, 4, 5}
result = set1.issubset(set2)
print(result)  # Output: True
```

- **Explanation**: `set1` is a subset of `set2`, even though `set2` contains additional elements (`3`, `4`, `5`).

#### Edge Case 3: Subset Check with Non-overlapping Sets

```python
set1 = {1, 2}
set2 = {3, 4, 5}
result = set1.issubset(set2)
print(result)  # Output: False
```

- **Explanation**: Since no element in `set1` exists in `set2`, `set1` is not a subset of `set2`.

---

### 8. **Alternative Method: `<=` Operator**

The **`<=`** operator can be used as an alternative to **`issubset()`** to check if one set is a subset of another. It returns `True` if the set on the left is a subset of the set on the right.

#### Example with `<=` Operator:

```python
set1 = {1, 2}
set2 = {1, 2, 3, 4}
result = set1 <= set2
print(result)  # Output: True
```

- **Explanation**: The `<=` operator works similarly to **`issubset()`** but is often used in more compact code.
