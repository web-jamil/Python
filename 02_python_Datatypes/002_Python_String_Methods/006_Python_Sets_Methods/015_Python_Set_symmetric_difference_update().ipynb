# Python Set `symmetric_difference_update()`

The **`symmetric_difference_update()`** method finds the symmetric difference of two sets and updates the set calling it.

The symmetric difference of two sets **`A`** and **`B`** is the set of elements that are in either **`A`** or **`B`**, but not in their intersection.

</div>


**Syntax**:

```python
A.symmetric_difference_update(B)
```


## Return Value from `symmetric_difference_update()`

The **`symmetric_difference_update()`** returns None (returns nothing). Rather, it updates the set calling it.


````python
# Example: Working of symmetric_difference_update()

A = {'a', 'c', 'd'}
B = {'c', 'd', 'e' }

result = A.symmetric_difference_update(B)

print('A =', A)
print('B =', B)
print('result =', result)
````

Here, the set **`A`** is updated with the symmetric difference of set **`A`** and **`B`**. However, the set **`B`** is unchanged.

### Python Set `symmetric_difference_update()` Method: Concepts and Theory

The **`symmetric_difference_update()`** method in Python is used to update the current set by performing a symmetric difference operation with another set or iterable. This method modifies the original set and removes any elements that are common to both sets, adding only the unique elements from either of the sets.

### 1. **Definition and Syntax of `symmetric_difference_update()`**

#### **Syntax**:

```python
set1.symmetric_difference_update(set2)
```

- **`set1`**: The set that will be updated by performing a symmetric difference with `set2`.
- **`set2`**: The second set (or iterable) involved in the symmetric difference operation.

#### **Return Value**:

- **`symmetric_difference_update()`** modifies the original set (`set1`) in place. It does not return a new set, unlike the **`symmetric_difference()`** method.

---

### 2. **How `symmetric_difference_update()` Works**

- **Modification of the Original Set**: Unlike **`symmetric_difference()`**, which creates a new set, **`symmetric_difference_update()`** modifies the set that calls the method.
- **Symmetric Difference Logic**: The method performs the symmetric difference operation by removing any common elements between the two sets and adding the unique elements from both sets.
- **In-Place Update**: This method directly updates the calling set (the set on which it is invoked), which means that the original set is modified.

#### Example:

If you have two sets, `set1 = {1, 2, 3}` and `set2 = {3, 4, 5}`, performing `set1.symmetric_difference_update(set2)` will remove `3` (the common element), and add `4` and `5` (the unique elements), modifying `set1` in place.

---

### 3. **Example Usage**

#### Example 1: Basic Symmetric Difference Update

```python
# Define two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Perform symmetric difference update
set1.symmetric_difference_update(set2)

# Output the updated set
print(set1)  # Output: {1, 2, 4, 5}
```

- **Explanation**: The original set `set1` is updated in place. The element `3` is removed because it is common to both sets, while `4` and `5` are added to `set1`.

#### Example 2: Symmetric Difference Update with Multiple Sets

```python
# Define three sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set3 = {5, 6, 7}

# Perform symmetric difference update between set1, set2, and set3
set1.symmetric_difference_update(set2)
set1.symmetric_difference_update(set3)

# Output the updated set
print(set1)  # Output: {1, 2, 4, 6, 7}
```

- **Explanation**: The symmetric difference update is performed step by step. First, `set1` is updated with `set2`, and then the result is updated with `set3`. Elements that are common in both sets are removed, and unique elements are added.

#### Example 3: Symmetric Difference Update with an Empty Set

```python
# Define a set and an empty set
set1 = {1, 2, 3}
set_empty = set()

# Perform symmetric difference update with an empty set
set1.symmetric_difference_update(set_empty)

# Output the updated set
print(set1)  # Output: {1, 2, 3}
```

- **Explanation**: Since the empty set has no elements, performing a symmetric difference update with an empty set will leave `set1` unchanged.

---

### 4. **Return Value of `symmetric_difference_update()`**

- The **`symmetric_difference_update()`** method modifies the original set in place and does not return any value.
- This is in contrast to **`symmetric_difference()`**, which creates a new set and returns it.

Example:

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1.symmetric_difference_update(set2)

# Output the result
print(result)  # Output: None (as it modifies in place)
print(set1)    # Output: {1, 2, 4, 5} (updated set)
```

- **Explanation**: The method modifies `set1` in place and does not return a value, so `result` will be `None`.

---

### 5. **Time Complexity of `symmetric_difference_update()`**

- **Time Complexity**: The time complexity of **`symmetric_difference_update()`** is **O(len(s) + len(t))**, where `s` is the set on which the method is invoked, and `t` is the set (or iterable) passed as an argument. This is because the method iterates over both sets to compute the symmetric difference and update the calling set.
  - **Space Complexity**: The space complexity is O(1) as the operation modifies the set in place and does not require additional space for storing the result.

---

### 6. **When to Use `symmetric_difference_update()`**

#### Use Cases for **`symmetric_difference_update()`**:

1. **Updating a Set with Unique Elements from Another Set**: When you need to update a set by adding elements that are unique between the two sets, and you want to eliminate the common elements in place.

   - Example:
     ```python
     # Comparing two groups of students and updating one set with students that are not common
     students_set1 = {"Alice", "Bob", "Charlie"}
     students_set2 = {"Charlie", "David", "Eve"}
     students_set1.symmetric_difference_update(students_set2)
     print(students_set1)  # Output: {'Alice', 'Bob', 'David', 'Eve'}
     ```

2. **Handling Large Datasets**: The **`symmetric_difference_update()`** method is particularly useful when you want to update an existing set with the symmetric difference of large datasets efficiently without creating additional sets.

   - Example:
     ```python
     set1 = {1, 2, 3, 4, 5}
     set2 = {3, 4, 5, 6, 7}
     set1.symmetric_difference_update(set2)
     print(set1)  # Output: {1, 2, 6, 7}
     ```

3. **Eliminating Common Data**: When you want to update an existing set to eliminate the common elements between two datasets while retaining the unique elements.
   - Example:
     ```python
     set1 = {1, 2, 3, 4}
     set2 = {3, 4, 5, 6}
     set1.symmetric_difference_update(set2)
     print(set1)  # Output: {1, 2, 5, 6}
     ```

---

### 7. **Edge Cases and Exceptions**

#### Edge Case 1: Symmetric Difference Update with an Empty Set

If you perform a symmetric difference update with an empty set, the original set remains unchanged because there are no elements to update.

```python
set1 = {1, 2, 3}
empty_set = set()
set1.symmetric_difference_update(empty_set)
print(set1)  # Output: {1, 2, 3}
```

#### Edge Case 2: Symmetric Difference Update Between Identical Sets

If the two sets are identical, performing a symmetric difference update will result in an empty set, as all elements are common and thus removed.

```python
set1 = {1, 2, 3}
set2 = {1, 2, 3}
set1.symmetric_difference_update(set2)
print(set1)  # Output: set()
```

#### Edge Case 3: Symmetric Difference Update with Different Types of Iterables

The **`symmetric_difference_update()`** method can handle other iterables like lists or tuples, but it only considers the elements. When working with dictionaries, only the keys are considered.

```python
set1 = {1, 2, 3}
list1 = [3, 4, 5]
set1.symmetric_difference_update(list1)
print(set1)  # Output: {1, 2, 4, 5}

# With a dictionary
dict1 = {"a": 1, "b": 2}
set1.symmetric_difference_update(dict1)
print(set1)  # Output: {1, 2, 4, 5, 'a', 'b'}
```

---

### 8. **Alternative Method: `^=` Operator**

Instead of using **`symmetric_difference_update()`**, you can also use the **`^=`** operator to perform the symmetric difference update in a more concise way. This operator works exactly the same as the method but is shorthand.

#### Example with `^=` Operator:

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set1 ^= set2
print(set1)  # Output: {1, 2, 4, 5}
```

- **Explanation**: The `^=` operator performs the symmetric difference update, modifying `set1` in place by adding the unique elements and removing the common elements.

The **`symmetric_difference_update()`** method in Python is a powerful tool for updating an existing set by removing elements common to both sets and adding unique elements. It modifies the original set in place, which can be highly efficient when working with large datasets. Whether you're comparing datasets, eliminating duplicates, or performing set operations in place, this method is an essential part of working with sets in Python.
