# Python String `isdigit()`

The string **`isdigit()`** method returns **`True`** if all characters in the string are digits. If not, it returns **`False`**.

**Syntax**:

```python
string.isdigit()
```


## `isdigit()` Parameters

The **`isdigit()`** method doesn't take any parameters.


## Return Value from `isdigit()`

The **`isdigit()`** returns:

- **`True`** if all characters in the string are digits.
- **`False`** if at least one character is not digits.


````python
# Example 1: Working of isdigit()

s = "28212"
print(s.isdigit())

# contains alphabets and spaces
s = "Mo3 nicaG el l22er"
print(s.isdigit())
````

**Explanation:**

A digit is a character that has property value:

- **`Numeric_Type = Digit`**
- **`Numeric_Type = Decimal`**

In Python, superscript and subscripts (usually written using unicode) are also considered digit characters. Hence, if the string contains these characters along with decimal characters, **`isdigit()`** returns **`True`**.

The roman numerals, currency numerators and fractions (usually written using unicode) are considered numeric characters but not digits. The **`isdigit()`** returns **`False`** if the string contains these characters.

To check whether a character is a numeric character or not, you can use **[isnumeric()](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/017_Python_String_isnumeric%28%29.ipynb)** method.


````python
# Example 2: String Containing digits and Numeric Characters

s = '23455'
print(s.isdigit())

#s = '²3455'
# subscript is a digit
s = '\u00B23455'
print(s.isdigit())

# s = '½'
# fraction is not a digit
s = '\u00BD'
print(s.isdigit())
````

## **1. Basic Concepts**

### **What is `isdigit()`?**

- `isdigit()` is a string method that checks if all characters in the string are digits.
- It returns `True` if the string is non-empty and all characters are digits; otherwise, it returns `False`.

### **What are Digits?**

- Digits include:
  - Numeric characters (`0-9`).
  - Full-width digits (used in some East Asian scripts).
  - Other Unicode characters that represent digits (e.g., Arabic-Indic digits).

### **Why Use `isdigit()`?**

- **Validation**: Useful for validating user input (e.g., phone numbers, numeric codes).
- **Data Cleaning**: Helps filter out non-digit characters from strings.


The `isdigit()` method in Python is a built-in string method that checks if all the characters in a string are digits (0-9). It returns a boolean value (`True` or `False`). This guide covers everything about the `isdigit()` function, including its syntax, usage, related methods, and advanced applications.

## Basic Overview

### Syntax

```python
str.isdigit()
```

### Returns

- **`True`**: if all characters in the string are digits and there is at least one character.
- **`False`**: if the string is empty or contains any non-digit characters (including spaces, decimal points, or letters).

### Basic Examples

#### 1. Simple Usage

```python
print("12345".isdigit())  # Output: True (all characters are digits)
print("4567".isdigit())   # Output: True
print("123a5".isdigit())  # Output: False (contains a letter)
print("12.34".isdigit())  # Output: False (contains a decimal point)
print("".isdigit())       # Output: False (empty string)
```

#### 2. Unicode Support

The `isdigit()` method can also recognize digits in other numeral systems, such as:

```python
print("²".isdigit())      # Output: True (superscript two)
print("⅗".isdigit())      # Output: True (fraction)
print("一二三".isdigit()) # Output: False (Chinese characters)
```

### Characteristics

- **Does Not Accept Decimal Points**: The method returns `False` for strings that contain decimal points or signs (e.g., `"-"` or `"+"`).
- **Checks All Characters**: If even one character is not a digit, it returns `False`.

### Related String Methods

1. **`isalpha()`**: Checks if all characters in the string are alphabetic.

   ```python
   print("hello".isalpha())  # Output: True
   print("hello123".isalpha()) # Output: False
   ```

2. **`isalnum()`**: Checks if all characters are alphanumeric (letters and digits).

   ```python
   print("abc123".isalnum()) # Output: True
   print("abc 123".isalnum()) # Output: False (contains space)
   ```

3. **`isspace()`**: Checks if all characters in the string are whitespace characters.

   ```python
   print("   ".isspace())    # Output: True
   print("abc".isspace())     # Output: False
   ```

4. **`isnumeric()`**: Similar to `isdigit()`, but also returns `True` for numeric characters in other forms (like fractions).

   ```python
   print("10".isnumeric())    # Output: True
   print("²".isnumeric())      # Output: True
   ```

5. **`isdecimal()`**: Checks if all characters in the string are decimal characters (0-9).
   ```python
   print("12345".isdecimal()) # Output: True
   print("12.34".isdecimal()) # Output: False
   ```

### Advanced Applications

#### 1. Validation of Numeric Input

You can use `isdigit()` to validate user inputs that should only contain digits (like age or ID):

```python
age = input("Enter your age: ")
if age.isdigit():
    print("Valid age.")
else:
    print("Invalid age. Please enter numbers only.")
```

#### 2. Filtering Digits from a List

You can filter out strings in a list to retain only those that contain only digits:

```python
data = ["123", "abc", "45.67", "890"]
digit_strings = [s for s in data if s.isdigit()]
print(digit_strings)  # Output: ['123', '890']
```

#### 3. Counting Numeric Values

You can count how many strings in a list are numeric:

```python
strings = ["123", "abc", "456", "78.9"]
digit_count = sum(1 for s in strings if s.isdigit())
print(digit_count)  # Output: 2 (only "123" and "456" are digits)
```

#### 4. Removing Non-Digit Characters

You can strip a string of any non-digit characters:

```python
mixed_string = "abc123!@#456XYZ"
cleaned_string = ''.join(c for c in mixed_string if c.isdigit())
print(cleaned_string)  # Output: '123456'
```

### Python `isdigit()` Function

The `isdigit()` function in Python is used to check if all characters in a string are digits. This method is a string method, meaning it is applied to a string object.

### Syntax:

```python
str.isdigit()
```

### Parameters:

- The `isdigit()` method does not take any parameters. It only operates on the string it is called upon.

### Return Value:

- **`True`**: If all characters in the string are digits (i.e., characters in the range `0-9`).
- **`False`**: If the string contains any non-digit characters, such as letters, special symbols, or spaces, or if the string is empty.

### Detailed Explanation:

- The `isdigit()` method considers a character as a "digit" if it is a digit in any numbering system. This includes Arabic numerals (`0-9`) and other numerals in different languages.
- The method returns `True` only if **every character** in the string is a digit.
- If the string contains any character that is not a digit (e.g., spaces, letters, punctuation), the method returns `False`.
- The `isdigit()` method can recognize digits from other numeral systems, such as those used in Chinese or Roman numerals.

### Basic Examples:

#### 1. Simple Case with Digits:

```python
text = "12345"
result = text.isdigit()
print(result)  # Output: True
```

- The string `"12345"` consists entirely of digits, so the method returns `True`.

#### 2. String with Letters:

```python
text = "123abc"
result = text.isdigit()
print(result)  # Output: False
```

- The string `"123abc"` contains letters, so `False` is returned.

#### 3. String with Space:

```python
text = "123 45"
result = text.isdigit()
print(result)  # Output: False
```

- The string contains a space, which is not a digit, so it returns `False`.

#### 4. Empty String:

```python
text = ""
result = text.isdigit()
print(result)  # Output: False
```

- An empty string does not contain any digits, so the method returns `False`.

#### 5. String with Special Characters:

```python
text = "123@45"
result = text.isdigit()
print(result)  # Output: False
```

- The presence of the special character `@` means the string is not composed solely of digits, so the result is `False`.

#### 6. Unicode Digits:

```python
text = "٣٤٥"  # Arabic numerals for 345
result = text.isdigit()
print(result)  # Output: True
```

- The string `"٣٤٥"` contains Arabic numerals (which are considered valid digits), so `True` is returned.

#### 7. String with Decimal or Negative Sign:

```python
text = "123.45"
result = text.isdigit()
print(result)  # Output: False
```

- The string `"123.45"` contains a decimal point, which is not a digit, so the result is `False`.

#### 8. String with Mixed Digits and Non-Digits:

```python
text = "12345abc"
result = text.isdigit()
print(result)  # Output: False
```

- Since there are non-digit characters (letters) in the string, the method returns `False`.

### Edge Cases:

1. **Empty String**: An empty string will return `False`, as it does not contain any characters.

   ```python
   text = ""
   result = text.isdigit()
   print(result)  # Output: False
   ```

2. **Unicode Digits**: The method can handle characters from other number systems, such as Arabic or Chinese digits.

   ```python
   text = "٣٤٥"  # Arabic digits for 345
   result = text.isdigit()
   print(result)  # Output: True
   ```

3. **Decimal Numbers**: The method will return `False` for strings containing decimals.

   ```python
   text = "123.45"
   result = text.isdigit()
   print(result)  # Output: False
   ```

4. **Negative Numbers**: The method will return `False` for strings containing negative signs or other non-digit characters.

   ```python
   text = "-123"
   result = text.isdigit()
   print(result)  # Output: False
   ```

5. **Whitespace**: If the string contains spaces, `isdigit()` will return `False`.
   ```python
   text = "123 456"
   result = text.isdigit()
   print(result)  # Output: False
   ```

### Common Use Cases:

1. **Check if User Input is a Number**:
   You can use `isdigit()` to verify if a user's input is a number.

   ```python
   user_input = input("Enter a number: ")
   if user_input.isdigit():
       print("Valid number.")
   else:
       print("Invalid input! Please enter a valid number.")
   ```

2. **Filtering Non-Digit Data**:
   You can use `isdigit()` to filter out non-digit strings from a list.

   ```python
   values = ["123", "abc", "456", "12.34", "789"]
   digits_only = [value for value in values if value.isdigit()]
   print(digits_only)  # Output: ['123', '456', '789']
   ```

3. **Validate Phone Numbers**:
   You can check if a phone number is valid (i.e., consists only of digits).
   ```python
   phone_number = "1234567890"
   if phone_number.isdigit():
       print("Valid phone number.")
   else:
       print("Invalid phone number!")
   ```

### Performance:

- **Time Complexity**: O(n), where `n` is the length of the string, because the method needs to check each character of the string.
- **Space Complexity**: O(1), since `isdigit()` does not require additional space proportional to the size of the string.

### Summary:

The `isdigit()` method is a useful way to check whether all characters in a string are digits. It works with standard Arabic numerals (`0-9`), as well as other numeral systems. The method returns `True` if the string contains only digits, and `False` otherwise. It is commonly used in validating user input and processing numeric data in strings.
