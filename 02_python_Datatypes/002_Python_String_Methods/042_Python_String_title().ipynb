# Python String `title()`

The string **`title()`** method returns a string with first letter of each word capitalized; a title cased string.

**Syntax**:

```python
str.title()
```


## `title()` Parameters

The **`title()`** method doesn't take any parameters.


## Return Value from `title()`

The **`title()`** method returns a title cased version of the string. Meaning, the first character of each word is capitalized (if the first character is a letter).


````python
# Example 1: How Python title() works?

text = 'My favorite number is 99.'
print(text.title())

text = '963 k3l2 *63 fun'
print(text.title())
````

````python
# Example 2: title() with apostrophes

text = "He's an engineer, isn't he?"
print(text.title())
````

> **Note:** **`title()`** capitalizes the first letter after apostrophes as well.

To solve this issue, you can use regex as follows:


````python
# Example 3: Using Regex to Title Case String

import re

def titlecase(s):
    return re.sub(r"[A-Za-z]+('[A-Za-z]+)?",
     lambda mo: mo.group(0)[0].upper() +
     mo.group(0)[1:].lower(),
     s)

text = "He's an engineer, isn't he?"
print(titlecase(text))
````

### Python `title()` Function: From Basic to Advanced

The `title()` method in Python is used to return a **new string** where the first character of each word is capitalized (i.e., converted to uppercase) and all other characters are converted to lowercase. This method is typically used to format a string in **title case**.

### **Syntax:**

```python
string.title()
```

- **`string`**: The string that you want to format.
- **Return Type**: It returns a new string with the first letter of each word capitalized and all other letters in lowercase.

### **How Does `title()` Work?**

- The method capitalizes the first letter of each word and converts all other letters in that word to lowercase.
- A word is considered as a sequence of characters separated by spaces or punctuation.
- The `title()` method doesn’t modify the original string but instead returns a **new string**.

### **Examples:**

#### Basic Examples:

```python
# Example 1: Simple string
s1 = "hello world"
print(s1.title())  # Output: "Hello World" (First letter of each word is capitalized)

# Example 2: String with mixed cases
s2 = "python IS a great LANGUAGE"
print(s2.title())  # Output: "Python Is A Great Language" (Converts to title case)

# Example 3: String with punctuation
s3 = "the quick-brown fox"
print(s3.title())  # Output: "The Quick-Brown Fox" (Hyphenated words are treated as separate words)
```

#### Edge Cases:

1. **Single word string**:
   - The method works even for strings with just one word, converting the first character to uppercase and others to lowercase.

```python
s4 = "hello"
print(s4.title())  # Output: "Hello"
```

2. **String with multiple spaces**:
   - If there are multiple spaces between words, the `title()` method treats each word correctly.

```python
s5 = "hello    world"
print(s5.title())  # Output: "Hello    World"
```

3. **String with non-alphabetic characters**:
   - Punctuation marks (like hyphens or apostrophes) are considered part of words and affect the capitalization.

```python
s6 = "it's a beautiful day"
print(s6.title())  # Output: "It'S A Beautiful Day" (Note the capitalization of 'S' after apostrophe)
```

4. **String with numbers**:
   - Numbers are not affected by the `title()` method since they are not alphabetic characters.

```python
s7 = "my address is 1234 main street"
print(s7.title())  # Output: "My Address Is 1234 Main Street"
```

5. **Empty string**:
   - If the string is empty, the `title()` method will return an empty string.

```python
s8 = ""
print(s8.title())  # Output: ""
```

#### Advanced Examples:

1. **String with mixed punctuation**:
   - Words that contain hyphens or other punctuation will be split and capitalized accordingly.

```python
s9 = "this is a well-known fact"
print(s9.title())  # Output: "This Is A Well-Known Fact"
```

2. **String with contractions**:
   - In case of contractions (like `it's`), `title()` will capitalize the first letter of the first word and may also capitalize letters inside the contraction.

```python
s10 = "it's a nice day"
print(s10.title())  # Output: "It'S A Nice Day"
```

3. **String with mixed case**:
   - The `title()` method also handles mixed-case characters by converting all letters except the first in each word to lowercase.

```python
s11 = "tHIS iS a MiXeD CaSe"
print(s11.title())  # Output: "This Is A Mixed Case"
```

### **Performance Considerations:**

- **Time Complexity**: The time complexity of the `title()` method is **O(n)**, where `n` is the length of the string. This is because the method has to process each character in the string.
- **Space Complexity**: The space complexity is **O(n)** because the method creates a new string to hold the transformed result.

### **When to Use `title()`?**

- **Title Case Formatting**: Use `title()` when you need to format strings as titles or headings where each word's first letter should be capitalized.
- **Word Capitalization**: Useful when handling user input or text processing where each word in a sentence should start with an uppercase letter.

### **Limitations of `title()`**

- The `title()` method does not correctly handle some edge cases like contractions (`it's` -> `It'S`), acronyms (`NASA` -> `Nasa`), and words that include apostrophes or other special characters.
- If you need to handle these cases more precisely (for example, keeping contractions in proper format like `It's` instead of `It'S`), you might need a custom solution or use a third-party library like `titlecase`.

### **Conclusion:**

- The `title()` method is a simple and effective way to convert a string into title case.
- It is useful for formatting titles, headings, or any text where proper capitalization of words is required.
- While it is fast and efficient, be aware of edge cases such as contractions or hyphenated words, which may not be handled perfectly by `title()`. For more nuanced title-case formatting, additional logic may be needed.
