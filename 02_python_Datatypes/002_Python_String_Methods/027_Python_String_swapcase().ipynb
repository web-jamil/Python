# Python String `swapcase()`

The string **`swapcase()`** method converts all uppercase characters to lowercase and all lowercase characters to uppercase characters of the given string, and returns it.

**Syntax**:

```python
string.swapcase()
```

> **Note:** Not necessarily, **`string.swapcase().swapcase() == string`**


## `swapcase()` Parameters

The **`swapcase()`** method doesn't take any parameters.


## Return Value from `swapcase()`

The **`swapcase()`** method returns the string where all uppercase characters are converted to lowercase, and lowercase characters are converted to uppercase.


````python
# Example 1: Swap lowercase to uppercase and vice versa using swapcase()

# example string
string = "THIS SHOULD ALL BE LOWERCASE."
print(string.swapcase())

string = "this should all be uppercase."
print(string.swapcase())

string = "ThIs ShOuLd Be MiXeD cAsEd."
print(string.swapcase())
````

> **Note:** If you want to convert string to lowercase only, use **[lower()](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/025_Python_String_lower%28%29.ipynb)**. Likewise, if you want to convert string to uppercase only, use **[upper()](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/026_Python_String_upper%28%29.ipynb)**.


### Python `swapcase()` Method: From Basic to Advanced

The `swapcase()` method in Python is used to **convert all uppercase letters to lowercase** and all lowercase letters to uppercase in a given string. It does not affect any characters other than alphabetic characters. This method is very useful when you want to invert the case of all characters in a string.

---

### **Syntax:**

```python
string.swapcase()
```

- **`string`**: The string on which the `swapcase()` method is applied.
- This method does not take any arguments, and it returns a **new string** with all the uppercase letters converted to lowercase and vice versa.

### **Return Value:**

- The `swapcase()` method returns a **new string** where the case of all letters is swapped, while leaving non-alphabetical characters (such as digits, punctuation, spaces) unchanged.

---

### **Basic Usage:**

#### Example 1: Basic `swapcase()` Usage

```python
s = "Hello World"
new_string = s.swapcase()
print(f"Original: '{s}'")
print(f"After swapcase: '{new_string}'")
```

**Output:**

```
Original: 'Hello World'
After swapcase: 'hELLO wORLD'
```

- The uppercase letters `"H"` and `"W"` are converted to lowercase, and the lowercase letters `"e"`, `"l"`, `"o"`, etc., are converted to uppercase.

#### Example 2: String with Mixed Case

```python
s = "Python Is Fun"
new_string = s.swapcase()
print(f"Original: '{s}'")
print(f"After swapcase: '{new_string}'")
```

**Output:**

```
Original: 'Python Is Fun'
After swapcase: 'pYTHON iS fUN'
```

- Each alphabetic character's case is flipped, but non-alphabetic characters (if present) remain unaffected.

---

### **Working with Different Characters:**

The `swapcase()` method only affects **alphabetic characters**. Other characters such as spaces, digits, punctuation marks, etc., are **not modified**.

#### Example 3: Non-Alphabetic Characters Unaffected

```python
s = "Hello 123! How are you?"
new_string = s.swapcase()
print(f"Original: '{s}'")
print(f"After swapcase: '{new_string}'")
```

**Output:**

```
Original: 'Hello 123! How are you?'
After swapcase: 'hELLO 123! hOW ARE YOU?'
```

- The numbers and punctuation marks (`123!`, `?`) are left unchanged, while the letters are swapped in case.

---

### **Edge Cases:**

1. **Empty String:**

```python
s = ""
new_string = s.swapcase()
print(f"Original: '{s}'")
print(f"After swapcase: '{new_string}'")
```

**Output:**

```
Original: ''
After swapcase: ''
```

- If the string is empty, `swapcase()` will return an empty string as well.

2. **String with No Alphabetic Characters:**

```python
s = "12345!@#"
new_string = s.swapcase()
print(f"Original: '{s}'")
print(f"After swapcase: '{new_string}'")
```

**Output:**

```
Original: '12345!@#'
After swapcase: '12345!@#'
```

- If the string contains no alphabetic characters, the string remains unchanged, because `swapcase()` only affects alphabetic characters.

3. **String with All Uppercase or All Lowercase Characters:**

```python
s = "HELLO"
new_string = s.swapcase()
print(f"Original: '{s}'")
print(f"After swapcase: '{new_string}'")
```

**Output:**

```
Original: 'HELLO'
After swapcase: 'hello'
```

- All uppercase characters are converted to lowercase.

---

### **Performance Considerations:**

- **Time Complexity**: The time complexity of the `swapcase()` method is **O(n)**, where `n` is the length of the string. This is because the method needs to iterate over each character in the string to change its case.
- **Space Complexity**: The space complexity is also **O(n)** because the method returns a new string, which requires space proportional to the size of the input string.

---

### **Advanced Usage:**

1. **Chaining with Other String Methods**

The `swapcase()` method can be chained with other string methods to perform multiple string operations in a single expression.

#### Example 4: Swapping Case and Removing Whitespace

```python
s = "  Python   is    cool  "
new_string = s.strip().swapcase()
print(f"After strip and swapcase: '{new_string}'")
```

**Output:**

```
After strip and swapcase: 'python   IS    COOL'
```

- In this case, `strip()` removes leading and trailing whitespace before `swapcase()` is applied to the string.

2. **Chaining Multiple Methods with `swapcase()`**

```python
s = "python programming"
new_string = s.replace("python", "Java").swapcase()
print(f"After replace and swapcase: '{new_string}'")
```

**Output:**

```
After replace and swapcase: 'jAVA PROGRAMMING'
```

- You can replace certain substrings in the string and then swap the case of the whole string.

---

### **Summary:**

- The `swapcase()` method in Python is used to convert all **uppercase letters to lowercase** and **all lowercase letters to uppercase**.
- Non-alphabetic characters (spaces, digits, punctuation) remain **unaffected**.
- It is a **string method** that returns a **new string** with the case of all alphabetic characters swapped.

---

#### Key Points:

- **Syntax**: `string.swapcase()`
- **Time Complexity**: **O(n)**, where `n` is the length of the string.
- **Space Complexity**: **O(n)** due to the creation of a new string.
- **Usage**: Typically used when you need to invert the case of all alphabetic characters in a string.
