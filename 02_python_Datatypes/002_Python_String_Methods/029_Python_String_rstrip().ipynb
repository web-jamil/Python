# Python String `rstrip()`

The string **`rstrip()`** method returns a copy of the string with trailing characters removed (based on the string argument passed).

The **`rstrip()`** removes characters from the right based on the argument (a string specifying the set of characters to be removed).

**Syntax**:

```python
string.rstrip([chars])
```


## `rstrip()` Parameters

The **`rstrip()`** method takes one parameter:

- **`chars`** (optional) - a string specifying the set of characters to be removed.

If **`chars`** argument is not provided, all leading whitespaces on the right are removed from the string.


## Return Value from `rstrip()`

The **`rstrip()`** method returns a copy of the string with trailing characters stripped.

All combinations of characters in **`chars`** argument are removed from the right of the string until first mismatch.


````python
# Example 1: Working of rstrip()

random_string = 'this is good    '

# Trailing whitepsace are removed
print(random_string.rstrip())

# Argument doesn't contain 'd'
# No characters are removed.
print(random_string.rstrip('si oo'))

print(random_string.rstrip('sid oo'))

website = 'www.github.com/'
print(website.rstrip('m/.'))
````

````python
# Example 2:

s1=' I Love Python Tutorial '
print(s1.rstrip())

s2='###HelloWorld###'
print(s2.rstrip('#'))
````

### Python `rstrip()` Method: From Basic to Advanced

The `rstrip()` method in Python is used to **remove trailing whitespaces** (or specified characters) from the end of a string. It returns a **new string** with the trailing characters removed. This method is a built-in string method in Python and is part of the string class.

---

### **Syntax:**

```python
string.rstrip([chars])
```

- **`string`**: The string on which the `rstrip()` method is applied.
- **`chars`** (Optional): A string specifying the set of characters to be removed. If no argument is passed, the method will remove trailing whitespace characters (such as spaces, tabs, or newlines) by default.

### **Return Value:**

- The method returns a new string with **trailing characters removed**.
- It does **not modify** the original string, because strings are **immutable** in Python.

---

### **Basic Usage:**

#### Example 1: Removing Trailing Whitespace (Default Behavior)

```python
s = "   hello world   "
new_string = s.rstrip()
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: '   hello world   '
After rstrip: '   hello world'
```

- The `rstrip()` method removes only the **trailing spaces** from the right end of the string.

#### Example 2: Removing Trailing Newline Characters

```python
s = "hello world\n"
new_string = s.rstrip()
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: 'hello world\n'
After rstrip: 'hello world'
```

- The `rstrip()` method also removes trailing newline characters (`\n`), which are commonly used in user input or file reading.

#### Example 3: Removing Trailing Tabs

```python
s = "hello world\t"
new_string = s.rstrip()
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: 'hello world\t'
After rstrip: 'hello world'
```

- It removes **trailing tab characters** as well.

---

### **Removing Specific Characters:**

If you pass a string as the `chars` argument, `rstrip()` will remove **all characters** in that string from the right end of the string.

#### Example 4: Removing Specific Characters

```python
s = "hello world!!!"
new_string = s.rstrip("!")
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: 'hello world!!!'
After rstrip: 'hello world'
```

- The `rstrip("!")` removes **all trailing exclamation marks** from the string, not just a single one.

#### Example 5: Removing Multiple Characters

```python
s = "hello123 world456"
new_string = s.rstrip("456")
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: 'hello123 world456'
After rstrip: 'hello123 world'
```

- The method removes all trailing characters that are part of the set `"456"`. This means that all `4`, `5`, and `6` characters at the end are removed, regardless of their position.

---

### **Edge Cases:**

1. **Empty String:**

```python
s = ""
new_string = s.rstrip()
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: ''
After rstrip: ''
```

- If the string is empty, `rstrip()` returns an empty string.

2. **No Trailing Characters to Remove:**

```python
s = "hello"
new_string = s.rstrip()
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: 'hello'
After rstrip: 'hello'
```

- If there are no trailing spaces, newlines, tabs, or specified characters, `rstrip()` will leave the string unchanged.

3. **Only Trailing Spaces:**

```python
s = "   "
new_string = s.rstrip()
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: '   '
After rstrip: ''
```

- If the string consists only of trailing spaces, `rstrip()` removes them, resulting in an empty string.

4. **Specified Characters at the End:**

```python
s = "abcabcabc"
new_string = s.rstrip("abc")
print(f"Original: '{s}'")
print(f"After rstrip: '{new_string}'")
```

**Output:**

```
Original: 'abcabcabc'
After rstrip: ''
```

- If the string ends with characters that are included in the `chars` argument, they are removed. In this case, the entire string consists of the characters "abc", so they are removed.

---

### **Advanced Usage:**

1. **Chaining String Methods with `rstrip()`:**

You can chain `rstrip()` with other string methods to perform more complex operations on strings.

#### Example 6: Remove Trailing Whitespaces and Convert to Uppercase

```python
s = "   hello world   "
new_string = s.rstrip().upper()
print(f"After rstrip and uppercase: '{new_string}'")
```

**Output:**

```
After rstrip and uppercase: 'HELLO WORLD'
```

- In this case, first, the `rstrip()` method removes the trailing spaces, and then `upper()` converts the resulting string to uppercase.

2. **Remove Trailing Characters in Data Cleaning:**

When dealing with data, such as text files or user input, `rstrip()` can be used to clean up unnecessary whitespace or unwanted trailing characters.

#### Example 7: Clean Up Input String

```python
input_string = "  user_data  "
cleaned_string = input_string.rstrip()
print(f"Cleaned Input: '{cleaned_string}'")
```

**Output:**

```
Cleaned Input: '  user_data'
```

- This cleans up any unnecessary trailing spaces from user input while leaving the leading spaces intact.

3. **Removing Newlines in Multi-line Text:**

When reading data from files, strings may contain trailing newlines. You can use `rstrip()` to clean these up.

#### Example 8: Remove Trailing Newlines from File Input

```python
file_content = "First line\nSecond line\nThird line\n"
cleaned_content = file_content.rstrip("\n")
print(f"Cleaned Content: '{cleaned_content}'")
```

**Output:**

```
Cleaned Content: 'First line\nSecond line\nThird line'
```

- In this example, we specifically remove the trailing newline character from the string that represents file content.

---

### **Performance Considerations:**

- The **time complexity** of `rstrip()` is **O(n)**, where `n` is the length of the string because it has to scan through the string from the end to the beginning to identify and remove the trailing characters.
- The **space complexity** is **O(n)**, as it creates a new string with the trailing characters removed.

---

### **Summary:**

The `rstrip()` method in Python is an essential tool for string manipulation. It removes trailing whitespace or any specified set of characters from the right end of a string. It is commonly used for cleaning up strings, especially when working with input data, text processing, or file operations.

---

#### Key Points:

- **Syntax:** `string.rstrip([chars])`
- **Default Behavior:** Removes trailing whitespace characters (spaces, newlines, tabs).
- **Customization:** You can specify the set of characters to remove.
- **Immutability:** Returns a new string without modifying the original string.
- **Time Complexity:** **O(n)** where `n` is the length of the string.
