# Python String `split()`

The string **`split()`** method breaks up a string at the specified separator and returns a list of strings.

**Syntax**:

```python
str.split([separator [, maxsplit]])
```


## `split()` Parameters

The **`split()`** method takes a maximum of 2 parameters:

- **`separator`** (optional)- It is a delimiter. The string splits at the specified **`separator`**.
  If the **`separator`** is not specified, any whitespace (space, newline etc.) string is a separator.

- **`maxsplit`** (optional) - The **`maxsplit`** defines the maximum number of splits.
  The default value of **`maxsplit`** is -1, meaning, no limit on the number of splits.


## Return Value from `split()`

The **`split()`** method breaks the string at the **`separator`** and returns a list of strings.


````python
# Example 1: How split() works in Python?

text= 'Love thy neighbor'

# splits at space
print(text.split())

grocery = 'Milk, Chicken, Bread'

# splits at ','
print(grocery.split(', '))

# Splitting at ':'
print(grocery.split(':'))
````

````python
# Example 2: How split() works when maxsplit is specified?

grocery = 'Milk, Chicken, Bread, Butter'

# maxsplit: 2
print(grocery.split(', ', 2))

# maxsplit: 1
print(grocery.split(', ', 1))

# maxsplit: 5
print(grocery.split(', ', 5))

# maxsplit: 0
print(grocery.split(', ', 0))
````

> **Note:** If **`maxsplit`** is specified, the list will have the maximum of **`maxsplit+1`** items.


> The **`split()`** function in Python is a built-in string method used to split a string into a list of substrings based on a specified delimiter. It is one of the most commonly used string manipulation methods in Python. Below is a comprehensive guide to the `split()` function, covering everything from basic syntax to advanced usage.

---

## **1. Basic Concepts**

### **What is `split()`?**

- `split()` is a string method that divides a string into a list of substrings based on a specified delimiter.
- By default, it splits the string at whitespace characters (spaces, tabs, newlines).

### **Why Use `split()`?**

- **String Parsing**: Useful for breaking down strings into manageable parts.
- **Data Cleaning**: Helps extract meaningful information from strings.
- **Text Processing**: Essential for tasks like tokenization in natural language processing (NLP).

---

## **2. Basic Syntax**

### **Syntax**

```python
string.split(sep=None, maxsplit=-1)
```

### **Parameters**

1. **`sep` (optional)**:
   - The delimiter to split the string.
   - If not provided or `None`, the string is split at whitespace characters.
2. **`maxsplit` (optional)**:
   - The maximum number of splits to perform.
   - If not provided or `-1`, all possible splits are performed.

### **Return Value**

- Returns a list of substrings.

---

## **3. Basic Usage**

### **Example 1: Default Behavior (Splitting at Whitespace)**

```python
text = "Python is awesome"
result = text.split()
print(result)  # Output: ['Python', 'is', 'awesome']
```

### **Example 2: Splitting at a Specific Delimiter**

```python
text = "apple,banana,cherry"
result = text.split(',')
print(result)  # Output: ['apple', 'banana', 'cherry']
```

### **Example 3: Limiting the Number of Splits**

```python
text = "one two three four"
result = text.split(maxsplit=2)
print(result)  # Output: ['one', 'two', 'three four']
```

---

## **4. Advanced Usage**

### **1. Splitting at Multiple Delimiters**

- Use regular expressions (`re` module) to split a string at multiple delimiters.

#### **Example**

```python
import re

text = "apple,banana;cherry|orange"
result = re.split(r'[,;|]', text)
print(result)  # Output: ['apple', 'banana', 'cherry', 'orange']
```

---

### **2. Handling Empty Strings**

- If the string contains consecutive delimiters, `split()` will include empty strings in the result.

#### **Example**

```python
text = "apple,,banana,,,cherry"
result = text.split(',')
print(result)  # Output: ['apple', '', 'banana', '', '', 'cherry']
```

- To remove empty strings, use a list comprehension.

#### **Example**

```python
text = "apple,,banana,,,cherry"
result = [s for s in text.split(',') if s]
print(result)  # Output: ['apple', 'banana', 'cherry']
```

---

### **3. Splitting Lines in a Multiline String**

- Use `splitlines()` to split a multiline string at line breaks.

#### **Example**

```python
text = "Line 1\nLine 2\nLine 3"
result = text.splitlines()
print(result)  # Output: ['Line 1', 'Line 2', 'Line 3']
```

---

### **4. Splitting and Stripping Whitespace**

- Use a list comprehension with `strip()` to remove leading/trailing whitespace from substrings.

#### **Example**

```python
text = "  apple  , banana  ,  cherry  "
result = [s.strip() for s in text.split(',')]
print(result)  # Output: ['apple', 'banana', 'cherry']
```

---

### **5. Splitting and Converting to Other Data Types**

- Use `map()` or list comprehensions to convert substrings to other data types (e.g., integers).

#### **Example**

```python
text = "10,20,30"
result = list(map(int, text.split(',')))
print(result)  # Output: [10, 20, 30]
```

---

### **6. Splitting with Regular Expressions**

- Use the `re` module for more complex splitting patterns.

#### **Example**

```python
import re

text = "apple123banana456cherry"
result = re.split(r'\d+', text)
print(result)  # Output: ['apple', 'banana', 'cherry']
```

---

### **7. Splitting and Joining**

- Use `split()` and `join()` together to manipulate strings.

#### **Example**

```python
text = "apple,banana,cherry"
result = '-'.join(text.split(','))
print(result)  # Output: apple-banana-cherry
```

---

## **5. Common Pitfalls**

1. **Empty Strings**:

   - Consecutive delimiters result in empty strings in the output list. Handle them appropriately.

2. **Whitespace**:

   - By default, `split()` removes leading/trailing whitespace but includes it in the result if specified as a delimiter.

3. **Performance**:
   - For large strings or frequent splitting operations, consider performance implications.

   - Use `split()` for straightforward splitting tasks.

   - Use list comprehensions to filter out empty strings if necessary.

   - Combine `split()` with `strip()`, `map()`, or `join()` for more advanced string manipulation.

The `split()` function in Python is a built-in string method used to divide a string into a list of substrings based on a specified separator (delimiter). It is one of the most commonly used string manipulation functions in Python.

### Syntax:

```python
str.split(sep=None, maxsplit=-1)
```

- **`sep`** (Optional): The delimiter or separator on which the string should be split. By default, it is `None`, which means the string will be split at any whitespace (spaces, tabs, newlines).
- **`maxsplit`** (Optional): Specifies the maximum number of splits to perform. The default value is `-1`, meaning "no limit" (i.e., split the string as many times as possible). If set to a non-negative integer `n`, the string will be split at most `n` times.

### Return Value:

- Returns a list of strings obtained by splitting the original string.

### Key Points:

1. **Default Behavior**: If no separator (`sep`) is specified, `split()` splits at any whitespace (spaces, tabs, and newlines). Consecutive spaces or whitespace are treated as a single separator.
2. **Limit of Splits (`maxsplit`)**: The number of splits is controlled by the `maxsplit` argument.
3. **Empty String Handling**: When the string is empty, an empty list `[]` is returned.

### Examples:

#### 1. Basic Example (Splitting by Whitespace):

```python
text = "Hello world"
result = text.split()
print(result)  # Output: ['Hello', 'world']
```

#### 2. Split by a Specific Separator:

```python
text = "apple,orange,banana"
result = text.split(',')
print(result)  # Output: ['apple', 'orange', 'banana']
```

#### 3. Split with `maxsplit`:

```python
text = "apple orange banana grape"
result = text.split(' ', 2)
print(result)  # Output: ['apple', 'orange', 'banana grape']
```

In this case, the string is split at most 2 times, so you get 3 elements in the list.

#### 4. Splitting with Multiple Consecutive Delimiters:

```python
text = "apple   orange   banana"
result = text.split()
print(result)  # Output: ['apple', 'orange', 'banana']
```

By default, consecutive whitespace is treated as a single delimiter.

#### 5. Splitting an Empty String:

```python
text = ""
result = text.split()
print(result)  # Output: []
```

An empty string results in an empty list.

#### 6. Specifying a Different Separator (Tab Character):

```python
text = "apple\tbanana\tcherry"
result = text.split('\t')
print(result)  # Output: ['apple', 'banana', 'cherry']
```

#### 7. Handling Multiple Delimiters:

If you need to split a string with multiple delimiters (e.g., comma, space, or semicolon), you can use regular expressions. Python's `re.split()` method is better suited for this.

```python
import re
text = "apple, orange;banana grape"
result = re.split(r'[ ,;]', text)
print(result)  # Output: ['apple', 'orange', 'banana', 'grape']
```

### Advanced Usage:

#### 1. Split on Newline Characters:

If a string contains multiple lines and you want to split it into individual lines:

```python
text = """apple
orange
banana"""
result = text.split('\n')
print(result)  # Output: ['apple', 'orange', 'banana']
```

#### 2. Split and Remove Empty Strings:

If you want to split a string, remove empty strings (if any), and get only non-empty substrings:

```python
text = "apple,,,orange,,,banana"
result = [item for item in text.split(',') if item]
print(result)  # Output: ['apple', 'orange', 'banana']
```

This example splits by commas but removes empty strings.

#### 3. Splitting and Stripping Whitespaces from Each Part:

If you need to split a string by whitespace and strip leading/trailing spaces from each resulting substring, you can use a combination of `split()` and `map()`:

```python
text = "  apple  orange  banana "
result = list(map(str.strip, text.split()))
print(result)  # Output: ['apple', 'orange', 'banana']
```

#### 4. Splitting and Returning a Tuple (First and Rest):

If you want to split a string but need the first part and the rest separately, you can use unpacking:

```python
text = "apple orange banana"
first, *rest = text.split()
print(first)  # Output: 'apple'
print(rest)   # Output: ['orange', 'banana']
```

#### 5. Handling Multiple Spaces:

If your string contains multiple spaces between words, `split()` can still handle it efficiently:

```python
text = "   apple    orange     banana  "
result = text.split()
print(result)  # Output: ['apple', 'orange', 'banana']
```

The `split()` function automatically handles multiple consecutive spaces as a single separator.

### Use Cases:

1. **Splitting CSV-like Data**:
   When you have a string representing a list of items separated by commas, you can easily extract the items using `split()`.

   ```python
   data = "name,age,city"
   columns = data.split(',')
   print(columns)  # Output: ['name', 'age', 'city']
   ```

2. **Processing User Input**:
   When accepting user input, you can split the string to parse multiple values.

   ```python
   user_input = input("Enter numbers separated by space: ")
   numbers = user_input.split()
   print(numbers)
   ```

3. **Extracting Words from Sentences**:
   For extracting individual words from a sentence:
   ```python
   sentence = "This is a test."
   words = sentence.split()
   print(words)  # Output: ['This', 'is', 'a', 'test.']
   ```

### Edge Cases:

- **Empty String**: An empty string will return an empty list.
  ```python
  text = ""
  result = text.split()
  print(result)  # Output: []
  ```
- **Multiple Consecutive Delimiters**: In the case of multiple spaces or delimiters, `split()` will ignore the empty substrings between them when no separator is explicitly given.
  ```python
  text = "apple    orange    banana"
  result = text.split()
  print(result)  # Output: ['apple', 'orange', 'banana']
  ```

### Performance:

- **Time Complexity**: O(n), where `n` is the length of the string, because `split()` iterates over each character in the string.
- **Space Complexity**: O(k), where `k` is the number of resulting substrings after splitting.

### Conclusion:

It can handle basic and advanced scenarios, such as splitting by specific delimiters, limiting the number of splits, removing empty strings, and more. It's one of the essential string methods for working with delimited data or parsing text efficiently.


> The `split()` function in Python is a powerful and versatile method used to divide a string into a list of substrings based on specified delimiters. This guide covers everything from the basic usage to more advanced applications of the `split()` function.

## Basic Overview

> ### Syntax

```python
str.split([sep[, maxsplit]])
```

### Parameters

- **`sep`** (optional): The delimiter string on which to split the string. The default is any whitespace (space, newline, tab).
- **`maxsplit`** (optional): The maximum number of splits to make. The default value of `-1` means "all occurrences".

### Returns

- A list of substrings obtained by splitting the original string.

### Basic Examples

#### 1. Default Behavior

By default, `split()` separates the string at any whitespace and removes extra whitespace:

```python
text = "Python is fun!"
result = text.split()
print(result)  # Output: ['Python', 'is', 'fun!']
```

#### 2. Using a Specific Separator

You can specify a separator to split the string:

```python
csv = "apple,banana,cherry"
result = csv.split(",")
print(result)  # Output: ['apple', 'banana', 'cherry']
```

#### 3. Limiting Splits with `maxsplit`

You can limit the number of splits performed:

```python
text = "one two three four five"
result = text.split(" ", 2)  # Split at most 2 times
print(result)  # Output: ['one', 'two', 'three four five']
```

## Advanced Usage

### 1. Handling Different Separators

You can use any string as a separator, including multiple characters:

```python
data = "2025-03-21"
date_parts = data.split("-")
print(date_parts)  # Output: ['2025', '03', '21']
```

### 2. Splitting on Multiple Delimiters

To split a string using multiple delimiters in a single operation, you can use regular expressions with the `re` module:

```python
import re

multi_data = "apple;banana,orange|grape"
result = re.split(r"[;,\|]", multi_data)
print(result)  # Output: ['apple', 'banana', 'orange', 'grape']
```

### 3. Stripping Trailing Whitespace

If you want to ensure that trailing whitespace is removed from each substring, you can use a combination of `split()` and `strip()`:

```python
text = "  hello   world  "
result = [s.strip() for s in text.split()]
print(result)  # Output: ['hello', 'world']
```

### 4. Joining Splitted Strings

You can combine the functionality of `split()` with `join()` for effective string manipulation:

```python
text = "Python is fun!"
words = text.split()
rejoined_text = ' '.join(words)
print(rejoined_text)  # Output: 'Python is fun!'
```

### 5. Iterating Over Splitted Strings

You can iterate over the resulting list from `split()` directly:

```python
text = "one,two,three"
for word in text.split(","):
    print(word)
```

### 6. Splitting Strings in Data Processing

In data processing tasks, `split()` is frequently used to extract relevant data from lines of text:

```python
data_line = "John,Doe,30,Male"
name, surname, age, gender = data_line.split(",")
print(f"Name: {name}, Surname: {surname}, Age: {age}, Gender: {gender}")
```

### 7. Edge Cases

When using `split()`, consider edge cases, such as:

- An empty string returns an empty list:
  ```python
  print("".split())  # Output: []
  ```
- A string without the separator returns the original string in a list:
  ```python
  print("Python".split(","))  # Output: ['Python']
  ```
