# Python String `rsplit()`

The string **`rsplit()`** method splits string from the right at the specified separator and returns a list of strings.

**Syntax**:

```python
str.rsplit([separator [, maxsplit]])
```


## `rsplit()` Parameters

The **`rsplit()`** method takes a maximum of 2 parameters:

- **`separator`** (optional)- It is a delimiter. **`rsplit()`** method splits string starting from the right at the specified **`separator`**.
  If the **`separator`** is not specified, any whitespace (space, newline etc.) string is a separator.

- **`maxsplit`** (optional) - The **`maxsplit`** defines the maximum number of splits.
  The default value of **`maxsplit`** is -1, meaning, no limit on the number of splits.


## Return Value from `rsplit()`

The **`rsplit()`** method breaks the string at the **`separator`** starting from the right and returns a list of strings.


````python
# Example 1: How rsplit() works in Python?

text= 'Love thy neighbor'

# splits at space
print(text.rsplit())

grocery = 'Milk, Coffee, Bread'

# splits at ','
print(grocery.rsplit(', '))

# Splitting at ':'
print(grocery.rsplit(':'))
````

> **Note:** When **`maxsplit`** is not specified, **`rsplit()`** behaves like **`split()`**.


````python
# Example 2: How split() works when maxsplit is specified?

grocery = 'Milk, Coffee, Bread, Butter'

# maxsplit: 2
print(grocery.rsplit(', ', 2))

# maxsplit: 1
print(grocery.rsplit(', ', 1))

# maxsplit: 5
print(grocery.rsplit(', ', 5))

# maxsplit: 0
print(grocery.rsplit(', ', 0))
````

> **Note:** If **`maxsplit`** is specified, the list will have the maximum of **`maxsplit+1`** items.


> The `rsplit()` function in Python is a built-in string method that is similar to `split()`, but it splits the string from the right side (reverse of the `split()` method, which splits from the left). This function is useful when you want to split a string starting from the end and work your way backwards.

### Syntax:

```python
str.rsplit(sep=None, maxsplit=-1)
```

- **`sep`** (Optional): The delimiter or separator on which the string should be split. By default, `sep` is `None`, which means the string will be split at any whitespace. If you provide a separator, the string will be split at occurrences of this separator.
- **`maxsplit`** (Optional): Specifies the maximum number of splits to perform. The default value is `-1`, meaning "no limit" (i.e., split the string as many times as possible). If set to a non-negative integer `n`, the string will be split at most `n` times.

### Return Value:

- Returns a list of strings obtained by splitting the original string.

### Key Differences between `split()` and `rsplit()`:

1. **Direction of Splitting**:
   - `split()` splits the string from the left (starting at the beginning of the string).
   - `rsplit()` splits the string from the right (starting at the end of the string).
2. **Behavior with `maxsplit`**:
   - Both `split()` and `rsplit()` behave similarly when `maxsplit` is provided. However, since `rsplit()` starts from the right, the `maxsplit` parameter controls how many splits happen from the right side, not the left.

### Examples:

#### 1. Basic Example (Splitting by Whitespace):

```python
text = "apple orange banana"
result = text.rsplit()
print(result)  # Output: ['apple', 'orange', 'banana']
```

- Here, the string is split at whitespace, starting from the right side.

#### 2. Split by a Specific Separator:

```python
text = "apple,orange,banana"
result = text.rsplit(',')
print(result)  # Output: ['apple', 'orange', 'banana']
```

- The string is split at commas starting from the right.

#### 3. Split with `maxsplit`:

```python
text = "apple orange banana grape"
result = text.rsplit(' ', 2)
print(result)  # Output: ['apple orange', 'banana', 'grape']
```

- In this case, the string is split at most 2 times starting from the right. The result contains 3 elements because the string was split at the two rightmost spaces.

#### 4. Handling Multiple Consecutive Delimiters:

```python
text = "apple,,,orange,,,banana"
result = text.rsplit(',')
print(result)  # Output: ['apple', '', '', 'orange', '', '', 'banana']
```

- The string is split at commas from the right. Note that empty strings are included in the result because consecutive commas are treated as separators.

#### 5. Split with No Separator Specified:

```python
text = "apple   orange  banana"
result = text.rsplit()
print(result)  # Output: ['apple', 'orange', 'banana']
```

- `rsplit()` without a separator treats consecutive spaces as a single delimiter.

#### 6. Split and Limit the Number of Splits:

```python
text = "apple orange banana grape"
result = text.rsplit(' ', 1)
print(result)  # Output: ['apple orange banana', 'grape']
```

- The string is split from the right at most 1 time, leaving the first part as `apple orange banana` and the second part as `grape`.

#### 7. Split with an Empty String:

```python
text = ""
result = text.rsplit()
print(result)  # Output: []
```

- An empty string results in an empty list because there is nothing to split.

#### 8. Split with Multiple Spaces:

```python
text = "  apple    orange   banana  "
result = text.rsplit()
print(result)  # Output: ['apple', 'orange', 'banana']
```

- `rsplit()` will handle multiple spaces by splitting at each whitespace and ignoring any extra spaces between words.

### Use Cases:

1. **Splitting from the End**: If you want to split a string starting from the right side (for example, when processing file paths or domain names), you can use `rsplit()`.

   ```python
   path = "/home/user/documents/file.txt"
   directories = path.rsplit('/', 2)
   print(directories)  # Output: ['/home/user', 'documents', 'file.txt']
   ```

2. **Extracting Last Few Elements**: When working with delimited strings and you only need the last few elements, `rsplit()` with a `maxsplit` argument is ideal.

   ```python
   sentence = "The quick brown fox jumps over the lazy dog"
   last_two = sentence.rsplit(' ', 2)
   print(last_two)  # Output: ['The quick brown fox jumps over', 'the', 'lazy dog']
   ```

3. **Handling Strings with Known Delimiters**: In scenarios where you know that the string will have a known number of delimiters from the right, `rsplit()` makes it easier to handle.

   ```python
   csv_string = "value1,value2,value3,value4"
   values = csv_string.rsplit(',', 1)
   print(values)  # Output: ['value1,value2,value3', 'value4']
   ```

4. **Parsing URL or Path Segments**: For parsing the segments of a URL or file path from the right side, `rsplit()` can be used effectively.
   ```python
   url = "www.example.com/images/photo.jpg"
   parts = url.rsplit('/', 1)
   print(parts)  # Output: ['www.example.com/images', 'photo.jpg']
   ```

### Edge Cases:

1. **Multiple Consecutive Separators**: If the separator is repeated multiple times, `rsplit()` will treat consecutive occurrences of the separator as a single one, just like `split()`.

2. **No Separator**: If no separator is provided, `rsplit()` splits by whitespace. Consecutive spaces are treated as a single separator.

3. **Empty String**: If the string is empty, the result will be an empty list `[]`.

4. **Single Character String**: If the string consists of a single character, the result will be a list with that single character.

### Performance:

- **Time Complexity**: O(n), where `n` is the length of the string. This is because the function needs to iterate through the string to split it.
- **Space Complexity**: O(k), where `k` is the number of substrings returned by `rsplit()`. This space is used to store the split parts in a list.

### Conclusion:

The `rsplit()` function in Python is a powerful tool for splitting strings from the right side, especially when you want to limit the number of splits or work with strings that are delimited in a specific manner (such as file paths or URLs). It is a valuable string manipulation tool, particularly for handling cases where the right-most parts of the string are of greater importance.


> The `rsplit()` method in Python is a built-in string method used to split a string into a list of substrings, starting from the right (end) of the string. It is similar to the `split()` method, but while `split()` begins splitting from the left, `rsplit()` starts from the right. This guide covers all aspects of the `rsplit()` method, including its syntax, usage, and advanced applications.

## Basic Overview

### Syntax

```python
str.rsplit([sep[, maxsplit]])
```

### Parameters

- **`sep`** (optional): The delimiter string on which to split the string. The default is any whitespace (space, newline, tab).
- **`maxsplit`** (optional): The maximum number of splits to make from the right. The default value of `-1` means "all occurrences".

### Returns

- A list of substrings obtained by splitting the original string from the right.

### Basic Examples

#### 1. Default Behavior

When no separator is specified, `rsplit()` removes any whitespace:

```python
text = "Python is fun!"
result = text.rsplit()
print(result)  # Output: ['Python', 'is', 'fun!']
```

#### 2. Using a Specific Separator

You can specify a separator to split the string:

```python
csv_data = "apple,banana,cherry"
result = csv_data.rsplit(",")
print(result)  # Output: ['apple', 'banana', 'cherry']
```

#### 3. Limiting Splits with `maxsplit`

You can limit the number of splits performed starting from the right:

```python
text = "one two three four five"
result = text.rsplit(" ", 2)  # Split at most 2 times from the right
print(result)  # Output: ['one two three', 'four', 'five']
```

## Advanced Usage

### 1. Handling Different Delimiters

You can use any string as a separator, just like with `split()`:

```python
data = "2025-03-21"
date_parts = data.rsplit("-")
print(date_parts)  # Output: ['2025', '03', '21']
```

### 2. Stripping Trailing Whitespace

If you want to ensure that trailing whitespace is removed from the resulting substrings, use a combination of `rsplit()` and `strip()`:

```python
text = "  hello   world  "
result = [s.strip() for s in text.rsplit()]
print(result)  # Output: ['hello', 'world']
```

### 3. Joining Split Strings

You can combine `rsplit()` with `join()` for effective string manipulation:

```python
text = "Python is fun!"
words = text.rsplit()
rejoined_text = ' '.join(words)
print(rejoined_text)  # Output: 'Python is fun!'
```

### 4. Iterating Over Split Strings

You can iterate over the resulting list from `rsplit()` directly:

```python
text = "one,two,three"
for word in text.rsplit(","):
    print(word)
```

### 5. Extracting File Extensions

`rsplit()` is particularly useful for extracting file extensions:

```python
file_name = "example.docx"
name, ext = file_name.rsplit(".", 1)  # Split at the last dot
print(f"Name: {name}, Extension: {ext}")  # Output: Name: example, Extension: docx
```

### 6. Edge Cases

When using `rsplit()`, consider the following edge cases:

- An empty string returns an empty list:
  ```python
  print("".rsplit())  # Output: []
  ```
- A string without the separator returns the original string in a list:
  ```python
  print("Python".rsplit(","))  # Output: ['Python']
  ```

### 7. Working with Multi-Line Strings

`rsplit()` can be used effectively to handle multi-line strings:

```python
multi_line = "first line\nsecond line\nthird line"
lines = multi_line.rsplit("\n")
print(lines)  # Output: ['first line', 'second line', 'third line']
```

## **1. Basic Concepts**

### **What is `rsplit()`?**

- `rsplit()` is a string method that splits a string into a list of substrings, starting from the right (end) of the string.
- It is similar to `split()`, but the splitting starts from the end.

### **Why Use `rsplit()`?**

- **Right-to-Left Splitting**: Useful when you need to split a string from the end.
- **Limited Splits**: Allows you to limit the number of splits starting from the right.
- **String Parsing**: Helps extract meaningful information from the end of a string.

---

## **2. Basic Syntax**

### **Syntax**

```python
string.rsplit(sep=None, maxsplit=-1)
```

### **Parameters**

1. **`sep` (optional)**:
   - The delimiter to split the string.
   - If not provided or `None`, the string is split at whitespace characters.
2. **`maxsplit` (optional)**:
   - The maximum number of splits to perform.
   - If not provided or `-1`, all possible splits are performed.

### **Return Value**

- Returns a list of substrings.

---

## **3. Basic Usage**

### **Example 1: Default Behavior (Splitting at Whitespace)**

```python
text = "Python is awesome"
result = text.rsplit()
print(result)  # Output: ['Python', 'is', 'awesome']
```

### **Example 2: Splitting at a Specific Delimiter**

```python

result = text.rsplit(',')

```

### **Example 3: Limiting the Number of Splits**

```python

result = text.rsplit(maxsplit=2)
```

## **4. Advanced Usage**

### **1. Splitting at Multiple Delimiters**

- Use regular expressions (`re` module) to split a string at multiple delimiters from the right.

#### **Example**

```python

text = "apple,banana;cherry|orange"
result = re.split(r'[,;|]', text)
print(result)  # Output: ['apple', 'banana', 'cherry', 'orange']
```

---

### **2. Handling Empty Strings**

- If the string contains consecutive delimiters, `rsplit()` will include empty strings in the result.

#### **Example**

```python

result = text.rsplit(',')
print(result)  # Output: ['apple', '', 'banana', '', '', 'cherry']
```

- To remove empty strings, use a list comprehension.

#### **Example**

```python
text = "apple,,banana,,,cherry"
result = [s for s in text.rsplit(',') if s]
print(result)  # Output: ['apple', 'banana', 'cherry']
```

---

### **3. Splitting Lines in a Multiline String**

- Use `rsplitlines()` to split a multiline string at line breaks from the right.

#### **Example**

```python

result = text.rsplitlines()

```

---

### **4. Splitting and Stripping Whitespace**

- Use a list comprehension with `strip()` to remove leading/trailing whitespace from substrings.

#### **Example**

```python
text = "  apple  , banana  ,  cherry  "
result = [s.strip() for s in text.rsplit(',')]
print(result)  # Output: ['apple', 'banana', 'cherry']
```

---

### **5. Splitting and Converting to Other Data Types**

- Use `map()` or list comprehensions to convert substrings to other data types (e.g., integers).

#### **Example**

```python

result = list(map(int, text.rsplit(',')))
```

- Use the `re` module for more complex splitting patterns from the right.

#### **Example**
