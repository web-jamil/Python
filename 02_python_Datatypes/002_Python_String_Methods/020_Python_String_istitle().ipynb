# Python String `istitle()`

The string **`istitle()`** method returns **`True`** if the string is a titlecased string. If not, it returns **`False`**.

**Syntax**:

```python
string.istitle()
```


## `istitle()` Parameters

The **`istitle()`** method doesn't take any parameters.


## Return Value from `istitle()`

The **`istitle()`** returns:

- **`True`** if the string is a titlecased string.
- **`False`** if the string is not a titlecased string or an empty string.


````python
# Example 1: Working of istitle()

s = 'Python Is Good.'
print(s.istitle())

s = 'Python is good'
print(s.istitle())

s = 'This Is @ Symbol.'
print(s.istitle())

s = '99 Is A Number'
print(s.istitle())

s = 'PYTHON'
print(s.istitle())
````

````python
# Example 2: How to use istitle()?

s = 'I Love Python.'
if s.istitle() == True:
    print('Titlecased String')
else:
    print('Not a Titlecased String')
  
s = 'PYthon'
if s.istitle() == True:
    print('Titlecased String')
else:
    print('Not a Titlecased String')
````

````python
# Example 3:

s1='I Love Python Tutorial'

print(s1.istitle())
s2='I love python 3'
print(s2.istitle())
````

### Python `istitle()` Function: From Basic to Advanced

The `istitle()` method in Python is used to check if a string follows **title case**. Title case means that the first letter of each word in the string is uppercase, and the rest of the letters in each word are lowercase.

### **Syntax:**

```python
string.istitle()
```

- **`string`**: The string that you want to check.
- **Return Type**:
  - **`True`**: If the string is in title case (i.e., each word starts with an uppercase letter, and all other letters in the word are lowercase).
  - **`False`**: If the string is not in title case.

### **How Does `istitle()` Work?**

- The method returns **`True`** if the string follows the title case rule, which means:
  - The first character of each word is uppercase.
  - All other characters in the word are lowercase.
- It ignores non-alphabetic characters (like punctuation or spaces) and checks only alphabetic characters.
- The method returns **`False`** if any word in the string does not follow the title case pattern.

### **Examples:**

#### Basic Examples:

```python
# Example 1: String in title case
s1 = "The Quick Brown Fox"
print(s1.istitle())  # Output: True (Each word starts with an uppercase letter)

# Example 2: String not in title case
s2 = "The quick brown fox"
print(s2.istitle())  # Output: False (The second word starts with a lowercase 'q')

# Example 3: String with mixed case
s3 = "Hello World!"
print(s3.istitle())  # Output: True (Each word starts with uppercase 'H' and 'W')
```

#### Edge Cases:

1. **String with Multiple Words**:

   - If the string has multiple words, each word should follow title case for the method to return `True`.

   ```python
   s4 = "This Is A Title"
   print(s4.istitle())  # Output: True
   ```

2. **String with Punctuation**:

   - Punctuation marks or symbols are ignored by the `istitle()` method. It only checks alphabetic characters.

   ```python
   s5 = "Hello, World!"
   print(s5.istitle())  # Output: True (Punctuation doesn't affect the result)
   ```

3. **Single Word**:

   - If the string has only one word, `istitle()` will return `True` if the word starts with an uppercase letter and all other characters are lowercase.

   ```python
   s6 = "Python"
   print(s6.istitle())  # Output: True (Starts with uppercase 'P')

   s7 = "python"
   print(s7.istitle())  # Output: False (Starts with lowercase 'p')
   ```

4. **Empty String**:

   - An empty string will always return `True` because it trivially follows the title case rule (as there are no words to violate it).

   ```python
   s8 = ""
   print(s8.istitle())  # Output: True (Empty string is considered valid)
   ```

#### More Advanced Examples:

```python
# Example 1: String with punctuation and digits
s9 = "The Great Wall of  China!"
print(s9.istitle())  # Output: True (Words are in title case, punctuation is ignored)

# Example 2: Mixed case words and punctuation
s10 = "A Quick Brown fox Jumped Over the Lazy Dog!"
print(s10.istitle())  # Output: True (Each word starts with uppercase letter)

# Example 3: Title case with hyphenated words
s11 = "A-B-C Is Great!"
print(s11.istitle())  # Output: True (Hyphenated words are treated correctly)

# Example 4: String with lowercase letters in words
s12 = "The Quick Brown fox"
print(s12.istitle())  # Output: False (The word "fox" is not in title case)

# Example 5: Title with capitalized numbers (works for digit words as well)
s13 = "1234 Is A Number"
print(s13.istitle())  # Output: True (Numbers at the beginning of a word are treated like letters)
```

### **Performance Considerations:**

- **Time Complexity**: The time complexity of the `istitle()` method is **O(n)**, where `n` is the length of the string. This is because the method has to check each character in the string to determine whether it follows the title case rule.
- **Space Complexity**: The space complexity is **O(1)** because the method does not require additional memory or space based on the length of the string.

### **Summary of the `istitle()` Method:**

- **Returns `True`**: If each word in the string starts with an uppercase letter, and the rest of the letters in the word are lowercase.
- **Returns `False`**: If any word in the string doesn’t follow the title case rule (e.g., if any word starts with a lowercase letter or contains uppercase letters in the middle).
- **Ignores non-alphabetic characters** (such as punctuation or numbers) when determining if the string follows title case.
- It is especially useful for checking if a string is formatted according to title case conventions (like titles of books, articles, etc.).

This method is particularly helpful when working with text formatting and ensuring that strings follow proper capitalization conventions.
