# Python `ord()` Function: Complete Guide

## What is the `ord()` function?

The `ord()` function in Python is a built-in function that returns the Unicode code point (integer representation) of a single Unicode character.

## Syntax

```python
ord(character)
```

### Parameters:

- `character`: A single Unicode character (string of length 1)

### Return Value:

- Returns an integer representing the Unicode code point of the character

## Basic Examples

```python
print(ord('A'))    # Output: 65 (Unicode for uppercase 'A')
print(ord('a'))    # Output: 97 (Unicode for lowercase 'a')
print(ord('0'))    # Output: 48 (Unicode for zero)
print(ord('€'))    # Output: 8364 (Euro symbol)
print(ord(' '))    # Output: 32 (Space character)
print(ord('\n'))   # Output: 10 (Newline character)
```

## Key Characteristics

1. **Single Character Only**:

   - `ord()` only works with strings of length 1
   - Raises `TypeError` if given a string with more than one character

   ```python
   ord('AB')  # Raises TypeError: ord() expected a character, but string of length 2 found
   ```

2. **Unicode Support**:

   - Works with any Unicode character, not just ASCII
   - Supports emojis, special symbols, etc.

   ```python
   print(ord('😊'))  # Output: 128522 (Smiling face emoji)
   ```

3. **Inverse Function**:

   - The inverse operation is performed by `chr()`
   - `chr(ord('x'))` returns 'x'

   ```python
   char = 'B'
   print(chr(ord(char)))  # Output: 'B'
   ```

## Practical Applications

### 1. Character to ASCII/Unicode Conversion

```python
# Convert characters to their code points
letters = 'Python'
for char in letters:
    print(f"{char}: {ord(char)}")
```

### 2. Simple Encryption

```python
# Caesar cipher (shift by 1)
message = "hello"
encrypted = ''.join([chr(ord(c) + 1) for c in message])
print(encrypted)  # Output: 'ifmmp'
```

### 3. Character Validation

```python
# Check if character is uppercase
def is_uppercase(char):
    return 65 <= ord(char) <= 90

print(is_uppercase('A'))  # True
print(is_uppercase('a'))  # False
```

### 4. Sorting by Unicode Values

```python
words = ['apple', 'Banana', 'cherry']
sorted_words = sorted(words, key=lambda x: ord(x[0].lower()))
print(sorted_words)  # ['apple', 'Banana', 'cherry']
```

## Common Errors

1. **Empty String**:

   ```python
   ord('')  # TypeError: ord() expected a character, but string of length 0 found
   ```

2. **Non-String Input**:

   ```python
   ord(65)  # TypeError: ord() expected string of length 1, but int found
   ```

3. **Multiple Characters**:
   ```python
   ord('abc')  # TypeError: ord() expected a character, but string of length 3 found
   ```

## Comparison with Similar Functions

| Function  | Purpose                                            | Input                  | Output                 |
| --------- | -------------------------------------------------- | ---------------------- | ---------------------- |
| `ord()`   | Character → Unicode code point                     | Single character (str) | Integer                |
| `chr()`   | Unicode code point → character                     | Integer                | Single character (str) |
| `ascii()` | Returns string containing printable representation | Any object             | String                 |

## Advanced Usage

### 1. Working with Different Encodings

```python
# Get UTF-8 bytes from Unicode code point
code_point = ord('ñ')  # 241
byte_representation = code_point.to_bytes(2, byteorder='big')
print(byte_representation)  # b'\x00\xf1'
```

### 2. Creating Character Ranges

```python
# Generate all lowercase letters
lowercase_letters = [chr(i) for i in range(ord('a'), ord('z')+1)]
print(lowercase_letters)
```

### 3. Unicode Normalization

```python
# Handling accented characters
e_acute = 'é'
print(ord(e_acute))  # 233 (for precomposed é)
```

```python
# Efficient processing of many characters
text = "Hello World"
code_points = [ord(c) for c in text]
```

the `ord()` function in Python. It's a fundamental function that deals with character encoding.

**What `ord()` Does:**

The `ord()` function in Python takes a single character (a string of length 1) as input and returns an integer representing the Unicode code point of that character.

Think of it as asking Python: "What is the numerical representation of this character in the Unicode standard?"

**Unicode Code Point:**

Unicode is a universal character encoding standard that assigns a unique number (code point) to virtually every character in every known writing system. These code points are usually represented in hexadecimal with a "U+" prefix (e.g., U+0041 for 'A'). The `ord()` function gives you the integer equivalent of this code point.

**Syntax:**

```python
ord(c)
```

**Parameters:**

- `c`: A string representing a single Unicode character. If you pass a string with more than one character, you'll get a `TypeError`.

**Return Value:**

- An integer representing the Unicode code point of the character `c`.

**Examples:**

Let's see `ord()` in action with various characters:

```python
# Uppercase letters
print(ord('A'))   # Output: 65
print(ord('B'))   # Output: 66
print(ord('Z'))   # Output: 90

# Lowercase letters
print(ord('a'))   # Output: 97
print(ord('b'))   # Output: 98
print(ord('z'))   # Output: 122

# Digits
print(ord('0'))   # Output: 48
print(ord('1'))   # Output: 49
print(ord('9'))   # Output: 57

# Symbols
print(ord(' '))   # Output: 32 (space)
print(ord('!'))   # Output: 33
print(ord('@'))   # Output: 64
print(ord('#'))   # Output: 35

# Unicode characters (beyond basic ASCII)
print(ord('€'))   # Output: 8364 (Euro sign)
print(ord('你好')) # Output: 20320 (first character of "你好" - "hello" in Chinese)
print(ord('你好'[1])) # Output: 22909 (second character of "你好")
print(ord('🐍'))   # Output: 128013 (snake emoji)

# Trying to use ord() with a string of length > 1 will raise a TypeError
# print(ord("AB"))  # This will cause: TypeError: ord() expected a character, but string of length 2 found
```

**Relationship with `chr()`:**

The `ord()` function is the inverse of the `chr()` function.

- `ord(character)` gives you the Unicode code point (integer) of a character.
- `chr(integer)` gives you the character (string of length 1) corresponding to a Unicode code point.

You can think of them as converting between the visual representation (character) and the numerical representation (Unicode code point).

```python
char = 'G'
code_point = ord(char)
print(f"The Unicode code point of '{char}' is: {code_point}")  # Output: The Unicode code point of 'G' is: 71

new_char = chr(code_point)
print(f"The character for the code point {code_point} is: '{new_char}'")  # Output: The character for the code point 71 is: 'G'
```

The `ord()` function is useful in various scenarios, including:

- **Cryptography:** In some basic cryptographic algorithms, you might work with the numerical values of characters.
- **Data Processing:** When reading or writing text data in specific formats or protocols, you might need to interact with Unicode code points.
- **Implementing Custom String Functions:** If you're building your own string manipulation functions, `ord()` can be helpful for analyzing individual characters.

In essence, `ord()` provides a way to peek under the hood and see the numerical identity of a character within the vast Unicode standard, allowing for more fine-grained control and understanding of text data in Python.


The `TypeError` in Python is one of the built-in exceptions that indicates an operation or function has been applied to an object of an inappropriate type. Understanding `TypeError` is important for debugging and writing robust Python code. Here’s a detailed look at `TypeError`, including its common causes and how to handle it.

### What is TypeError?

`TypeError` is raised when:

- An operation or function receives an argument of the right type but an inappropriate value.
- An operation is attempted on an object of a type that doesn’t support it.

### Common Causes of TypeError

1. **Invalid Operation on Incompatible Types**:

   ```python
   result = '1' + 2  # TypeError: can only concatenate str (not "int") to str
   ```

2. **Incorrect Argument Type Passed to a Function**:

   ```python
   def multiply(a, b):
       return a * b

   multiply(5, '2')  # TypeError: can't multiply sequence by non-int of type 'str'
   ```

3. **Using an Object in a Context Where a Different Type is Expected**:

   ```python
   my_list = [1, 2, 3]
   my_list[0] = 'a'  # Works fine
   my_list[1:2] = 'b'  # TypeError: can only assign an iterable
   ```

4. **Method Calls on Incompatible Types**:

   ```python
   'hello'.index(1)  # TypeError: string index must be integers
   ```

5. **Misusing Built-in Functions**:
   ```python
   numbers = [1, 2, 3]
   print(max(1, 2, 3, numbers))  # TypeError: '>' not supported between instances of 'list' and 'int'
   ```

### Handling TypeError

You can handle `TypeError` using `try` and `except` blocks. This allows your program to continue running even if a `TypeError` occurs.

#### Example:

```python
try:
    result = 'hello' + 5
except TypeError as e:
    print(f"Caught a TypeError: {e}")
```

To avoid encountering `TypeError`, consider the following practices:

   Before performing operations, you can check the types of variables.

   ```python
   if isinstance(variable, int):
       # Perform integer operations
   ```

2. **Using Type Annotations**:
   Type annotations in function definitions can help clarify expected argument types.

   ```python
   def add(a: int, b: int) -> int:
       return a + b
   ```

3. **Careful Function Use**:
   Ensure that functions are called with the appropriate types of arguments.

The `TypeError` related to the `ord()` function in Python occurs when the function receives an argument that is not a single character string. Understanding the context of this error and how to handle it is important for effective programming. Here’s a detailed overview:

### Understanding `ord()` and `TypeError`

The `ord()` function is designed to return the Unicode code point of a single character. If you provide it with an argument that does not meet its requirements, a `TypeError` will be raised.

### When Does `TypeError` Occur with `ord()`?

1. **Input Length Is Greater Than 1**:
   If you pass a string with more than one character, Python raises a `TypeError`.

   ```python
   ord('AB')  # Raises TypeError: ord() expected a unicode character, but string of length 2 found
   ```

2. **Input Is Not a String**:
   Passing a non-string type, like an integer, list, or None, will also result in a `TypeError`.

   ```python
   ord(65)    # Raises TypeError: ord() expected a unicode character, but int found
   ord([])    # Raises TypeError: ord() expected a unicode character, but list found
   ord(None)  # Raises TypeError: ord() expected a unicode character, but NoneType found
   ```

3. **Empty String Input**:
   If you attempt to pass an empty string, Python raises an error since there is no character to convert.
   ```python
   ord('')  # Raises TypeError: ord() expected a unicode character, but string of length 0 found
   ```

### Structure of TypeError for `ord()`

The error message generally follows this format:

```
TypeError: ord() expected a unicode character, but <type> found
```

- `<type>` will represent the type of the argument you passed, indicating what went wrong.

### Handling TypeError with `ord()`

To handle potential `TypeError` exceptions when using `ord()`, you can use a `try` and `except` block. This is useful for validation before calling the function.

#### Example:

```python
def safe_ord(character):
    try:
        return ord(character)
    except TypeError as e:
        print(f"Error: {e}")

# Tests
print(safe_ord('A'))     # Output: 65
print(safe_ord('AB'))    # Output: Error: ord() expected a unicode character, but string of length 2 found
print(safe_ord(65))      # Output: Error: ord() expected a unicode character, but int found
print(safe_ord(None))    # Output: Error: ord() expected a unicode character, but NoneType found
print(safe_ord(''))      # Output: Error: ord() expected a unicode character, but string of length 0 found
```

To prevent `TypeError` when using `ord()`, follow these practices:
