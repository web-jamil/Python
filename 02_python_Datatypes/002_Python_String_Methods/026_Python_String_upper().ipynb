# Python String `upper()`

The string **`upper()`** method converts all lowercase characters in a string into uppercase characters and returns it.

**Syntax**:

```python
string.upper()
```


## `upper()` Parameters

The **`upper()`** method doesn't take any parameters.


## Return Value from `upper()`

The **`upper()`** method returns the uppercased string from the given string. It converts all lowercase characters to uppercase.

If no lowercase characters exist, it returns the original string.


````python
# Example 1: Convert a string to uppercase

# example string
string = "this should be uppercase!"
print(string.upper())

# string with numbers
# all alphabets whould be lowercase
string = "Th!s Sh0uLd B3 uPp3rCas3!"
print(string.upper())
````

````python
# Example 2: How upper() is used in a program?

# first string
firstString = "python is awesome!"

# second string
secondString = "PyThOn Is AwEsOmE!"

if(firstString.upper() == secondString.upper()):
    print("The strings are same.")
else:
    print("The strings are not same.")
````

### Python `upper()` Method: From Basic to Advanced

The `upper()` method in Python is used to **convert all characters** in a string to **uppercase**. It is a built-in method of Python's string class.

### **Syntax:**

```python
string.upper()
```

- **`string`**: The string on which the `upper()` method is called.

### **Return Value:**

- The method returns a new string with all the characters converted to **uppercase**. If the string already contains uppercase letters or if the string doesn't contain any alphabetic characters, the method will leave those characters unchanged.
- It does **not modify** the original string because strings in Python are **immutable**.

---

### **Basic Usage:**

#### Example 1: Converting Lowercase to Uppercase

```python
s = "hello world"
new_string = s.upper()
print(new_string)  # Output: "HELLO WORLD"
```

- Here, all the lowercase characters in the string `"hello world"` are converted to uppercase, resulting in `"HELLO WORLD"`.

#### Example 2: Converting Mixed Case to Uppercase

```python
s = "HeLLo WoRLd"
new_string = s.upper()
print(new_string)  # Output: "HELLO WORLD"
```

- The `upper()` method converts both uppercase and lowercase letters in the string to uppercase. It ignores non-alphabetic characters (such as spaces).

#### Example 3: Already Uppercase String

```python
s = "HELLO"
new_string = s.upper()
print(new_string)  # Output: "HELLO"
```

- If the string is already in uppercase, the method returns the string unchanged.

---

### **How `upper()` Works:**

- The `upper()` method **converts all the alphabetic characters** in the string to uppercase and leaves **non-alphabetic characters** (like digits, punctuation, spaces, etc.) unaffected.
- The string must be **Unicode-compliant**, meaning it can handle any characters in the Unicode standard.

#### Example 4: Non-Alphabetic Characters Are Unchanged

```python
s = "abc 123 #$%"
new_string = s.upper()
print(new_string)  # Output: "ABC 123 #$%"
```

- Notice that spaces, numbers, and special characters `#$%` remain unchanged, while only the alphabetic characters are converted to uppercase.

---

### **Edge Cases:**

1. **Empty String:**

```python
s = ""
new_string = s.upper()
print(new_string)  # Output: ""
```

- If the string is empty, the `upper()` method simply returns an empty string.

2. **String with No Alphabetic Characters:**

```python
s = "12345!@#"
new_string = s.upper()
print(new_string)  # Output: "12345!@#"
```

- If the string contains no alphabetic characters, the string remains unchanged.

3. **Unicode Characters:**

The `upper()` method works with Unicode characters as well, and it will convert them to uppercase if possible.

#### Example 5: Unicode Character Conversion

```python
s = "öçş"
new_string = s.upper()
print(new_string)  # Output: "ÖÇŞ"
```

- The method correctly converts Unicode characters to uppercase if the uppercase equivalent is defined in Unicode.

---

### **Advanced Usage:**

1. **Chaining String Methods with `upper()`:**

You can chain `upper()` with other string methods to perform complex operations in one step.

#### Example 6: Remove Leading/Trailing Spaces and Convert to Uppercase

```python
s = "  hello world  "
new_string = s.strip().upper()
print(new_string)  # Output: "HELLO WORLD"
```

- Here, `strip()` removes any leading or trailing spaces, and `upper()` converts the string to uppercase.

2. **Comparing Uppercase Strings:**

You can use `upper()` in comparisons to make case-insensitive string comparisons.

#### Example 7: Case-Insensitive Comparison

```python
s1 = "hello"
s2 = "HELLO"
if s1.upper() == s2.upper():
    print("The strings are equal (case-insensitive).")
```

- By converting both strings to uppercase, we ensure that the comparison is case-insensitive.

3. **Using `upper()` in Data Preprocessing:**

When processing text, `upper()` is useful for normalizing strings, making them all uppercase for consistency.

#### Example 8: Normalizing User Input

```python
user_input = "pyThOn"
normalized_input = user_input.upper()
print(normalized_input)  # Output: "PYTHON"
```

- This is often used in cases where the case of the input doesn't matter (e.g., when checking for commands or matching words).

4. **Using `upper()` with Lists or Other Iterable Structures:**

You can apply the `upper()` method to each string in a list or another iterable structure using list comprehension.

#### Example 9: Applying `upper()` to Each Word in a List

```python
words = ["hello", "world", "python"]
upper_words = [word.upper() for word in words]
print(upper_words)  # Output: ['HELLO', 'WORLD', 'PYTHON']
```

- This converts each word in the list to uppercase.

---

### **Performance Considerations:**

- The **time complexity** of `upper()` is **O(n)**, where `n` is the length of the string, because it needs to iterate over each character in the string to check if it's lowercase.
- The **space complexity** is **O(n)**, as the method creates a new string and returns it with the updated characters.

---

### **Summary:**

The `upper()` method is a simple yet powerful tool for string manipulation in Python. It is used to convert all alphabetic characters in a string to uppercase while leaving non-alphabetic characters untouched. Whether used for basic text transformations or advanced data preprocessing, the `upper()` method is essential for handling text data that requires uniformity in case.

---

#### Key Points:

- **Syntax:** `string.upper()`
- **Return:** A new string with all alphabetic characters converted to uppercase.
- **Doesn't modify:** The original string (because strings are immutable in Python).
- **Works on:** Alphabetic characters, while leaving non-alphabetic characters unchanged.
