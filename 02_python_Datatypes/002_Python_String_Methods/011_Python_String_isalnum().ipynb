# Python String `isalnum()`

The string **`isalnum()`** method returns **`True`** if all characters in the string are alphanumeric (either alphabets or numbers). If not, it returns **`False`**.

**Syntax**:

```python
string.isalnum()
```


## `isalnum()` Parameters

The **`isalnum()`** method doesn't take any parameters.


## Return Value from `isalnum()`

The **`isalnum()`** returns:

- **`True`** if all characters in the string are alphanumeric
- **`False`** if at least one character is not alphanumeric


````python
# Example 1: Working of isalnum()

name = "M234onica"
print(name.isalnum())

# contains whitespace
name = "M3onica Gell22er "
print(name.isalnum())

name = "Mo3nicaGell22er"
print(name.isalnum())

name = "133"
print(name.isalnum())
````

````python
# Example 2: Working of isalnum()

name = "M0n1caG3ll3r"

if name.isalnum() == True:
    print("All characters of string (name) are alphanumeric.")
else:
    print("All characters are not alphanumeric.")
````

````python
# Example 3:

s="I Love Python 3 Tutorial"
print(s.isalnum())

s="Python" # No space
print(s.isalnum())
````

The `isalnum()` method in Python is a string method that checks whether all characters in a string are alphanumeric (i.e., the string contains only letters and numbers) and whether the string is not empty.

### Syntax:

```python
string.isalnum()
```

- **string**: The string on which the method is called.

### Return Value:

- **`True`** if all characters in the string are alphanumeric and the string is not empty.
- **`False`** if there is at least one character that is not alphanumeric (i.e., a space, punctuation, etc.), or if the string is empty.

### Definition of Alphanumeric:

- **Letters**: Any uppercase (`A-Z`) or lowercase (`a-z`) English letter.
- **Numbers**: Any digit (`0-9`).

### Important Points:

1. The string **must not be empty** for `isalnum()` to return `True`. Even if it contains only numbers or letters, it will still return `True` if it’s non-empty.
2. Spaces, punctuation marks, and special characters will cause `isalnum()` to return `False`.
3. The method checks all characters in the string, so if even one character is non-alphanumeric, the result will be `False`.

### Example Usages:

#### 1. Basic Examples:

```python
# Example 1: Alphanumeric String
text = "Hello123"
print(text.isalnum())  # Output: True (letters and numbers)

# Example 2: Non-Alphanumeric String (contains space)
text = "Hello 123"
print(text.isalnum())  # Output: False (contains space)

# Example 3: Non-Alphanumeric String (contains a special character)
text = "Hello@123"
print(text.isalnum())  # Output: False (contains a special character '@')

# Example 4: String with only numbers
text = "123456"
print(text.isalnum())  # Output: True (only numbers)

# Example 5: Empty String
text = ""
print(text.isalnum())  # Output: False (empty string)
```

#### 2. Edge Cases:

```python
# Edge Case 1: String with letters and punctuation
text = "Hello,World"
print(text.isalnum())  # Output: False (contains a comma)

# Edge Case 2: String with whitespace
text = "Hello World"
print(text.isalnum())  # Output: False (contains space between words)
```

#### 3. Mixed Data (letters and numbers):

```python
# Example 6: Valid alphanumeric string (mix of letters and digits)
text = "abc123XYZ"
print(text.isalnum())  # Output: True (letters and numbers)

# Example 7: String with mixed characters but spaces
text = "abc 123 XYZ"
print(text.isalnum())  # Output: False (contains spaces)
```

#### 4. Non-alphanumeric (Special Characters):

```python
# Example 8: String with only special characters
text = "!@#"
print(text.isalnum())  # Output: False (only special characters, no alphanumeric)
```

### Use Cases:

1. **Validation of User Input**:

   - This method is often used to validate user inputs (e.g., in forms) where the input should only consist of letters and numbers.

   ```python
   username = input("Enter your username: ")
   if username.isalnum():
       print("Valid username!")
   else:
       print("Invalid username. It should contain only letters and numbers.")
   ```

2. **String Filtering**:
   - If you want to check whether a string contains only alphanumeric characters (i.e., no spaces or special characters), `isalnum()` is very useful.
   ```python
   text = "hello123"
   if text.isalnum():
       print("This string is valid for a code identifier.")
   ```

### Performance:

- The time complexity of the `isalnum()` method is **O(n)**, where `n` is the length of the string, since it checks every character in the string.

### Conclusion:

- **`isalnum()`** is a built-in method in Python for determining if a string is composed entirely of alphanumeric characters (letters and numbers).
- It is useful in scenarios where you need to validate strings to ensure they don't contain special characters or spaces, especially when handling user inputs or parsing data that needs to be in a certain format.

The `isalnum()` method in Python is a built-in string method that checks whether all characters in a string are alphanumeric (i.e., either letters or digits) and returns a boolean value (`True` or `False`). Below is a comprehensive overview of the `isalnum()` function, including its usage, related functions, and some advanced examples.

### Basic Overview

#### Syntax

```python
str.isalnum()
```

#### Returns

- **`True`**: if all characters in the string are alphanumeric and there is at least one character.
- **`False`**: if the string is empty or contains characters that are not letters or digits.

### Characteristics

1. **Alphanumeric Characters**: It includes both uppercase and lowercase letters (A-Z, a-z) and digits (0-9).
2. **Whitespace and Punctuation**: Spaces, punctuation marks, and special characters (like `@`, `#`, and `!`) will cause the method to return `False`.
3. **Empty Strings**: An empty string (`''`) will return `False`.

### Usage Examples

#### 1. Basic Examples

```python
print("Hello123".isalnum())  # True
print("Hello 123".isalnum()) # False (contains space)
print("12345".isalnum())      # True
print("Hello@World".isalnum())# False (contains special character)
print("".isalnum())           # False (empty string)
```

#### 2. UTF-8 Characters

The `isalnum()` method also supports Unicode characters, meaning it can recognize alphabetic characters from different languages.

```python
print("México".isalnum())     # True
print("Café".isalnum())       # True
print("Café 2".isalnum())     # False (contains space)
```

### Related String Methods

1. **`isalpha()`**: Checks if all characters in the string are alphabetic.

   ```python
   print("Hello".isalpha())  # True
   print("Hello123".isalpha()) # False
   ```

2. **`isdigit()`**: Checks if all characters in the string are digits.

   ```python
   print("123".isdigit())    # True
   print("123abc".isdigit())  # False
   ```

3. **`isnumeric()`**: Similar to `isdigit()`, but also returns `True` for digits in other forms (like fractions).

   ```python
   print("²".isnumeric())     # True (superscript 2)
   ```

4. **`isspace()`**: Checks if all characters in the string are whitespace characters.

   ```python
   print("   ".isspace())    # True
   ```

5. **`isupper()`, `islower()`**: Checks if all characters in the string are upper or lower case letters, respectively.
   ```python
   print("HELLO".isupper())   # True
   print("hello".islower())    # True
   ```

### Advanced Applications

#### 1. Validation

You can use `isalnum()` for validating inputs where only alphanumeric characters are allowed. For example, a username check:

```python
username = "User123"
if username.isalnum():
    print("Valid username")
else:
    print("Invalid username")
```

#### 2. Filtering Alphanumeric Strings

You can utilize `isalnum()` to filter out non-alphanumeric characters from a list of strings.

```python
strings = ["Hello123", "Test@", "Valid456", "!Invalid", "123"]
alphanumeric_strings = [s for s in strings if s.isalnum()]
print(alphanumeric_strings)  # ['Hello123', 'Valid456', '123']
```

#### 3. Removing Non-Alphanumeric Characters

If you have a string and you want to remove all non-alphanumeric characters, here’s how you could do it:

```python
input_string = "Hello, World! 2021."
cleaned_string = ''.join(c for c in input_string if c.isalnum())
print(cleaned_string)  # Output: HelloWorld2021
```
