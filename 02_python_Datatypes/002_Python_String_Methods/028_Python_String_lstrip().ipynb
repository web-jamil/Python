# Python String `lstrip()`

The string **`lstrip()`** method returns a copy of the string with leading characters removed (based on the string argument passed).

The **`lstrip()`** removes characters from the left based on the argument (a string specifying the set of characters to be removed).

**Syntax**:

```python
string.lstrip([chars])
```


## `lstrip()` Parameters

The **`lstrip()`** method takes one parameter:

- **`chars`** (optional) - a string specifying the set of characters to be removed.

If **`chars`** argument is not provided, all leading whitespaces are removed from the string.


## Return Value from `lstrip()`

The **`lstrip()`** method returns a copy of the string with leading characters stripped.

All combinations of characters in the **`chars`** argument are removed from the left of the string until the first mismatch.


````python
# Example 1: Working of lstrip()

random_string = '   this is good '

# Leading whitepsace are removed
print(random_string.lstrip())

# Argument doesn't contain space
# No characters are removed.
print(random_string.lstrip('sti'))

print(random_string.lstrip('s ti'))

website = 'https://www.github.com/' 
print(website.lstrip('htps:/.'))
````

### Python `lstrip()` Function: From Basic to Advanced

The `lstrip()` method in Python is a built-in string function that is used to **remove leading whitespaces** (or characters) from a string. It returns a new string with the leading whitespace or characters removed from the left side of the string.

### **Syntax:**

```python
string.lstrip([chars])
```

- **`string`**: The string on which `lstrip()` is called.
- **`chars`** _(optional)_: A string specifying the set of characters to be removed. If no argument is provided, it removes **leading whitespaces** by default.

### **Return Value:**

- The method returns a **new string** where the leading characters (whitespaces by default, or any characters specified in `chars`) have been removed. It does **not modify** the original string, as strings are immutable in Python.

---

### **Basic Examples:**

#### Example 1: Removing Leading Whitespaces

```python
s = "   Hello World"
new_string = s.lstrip()
print(new_string)  # Output: "Hello World"
```

- In this example, the `lstrip()` method removes the leading spaces from the string `"   Hello World"`.

#### Example 2: String Without Leading Whitespace

```python
s = "Hello World"
new_string = s.lstrip()
print(new_string)  # Output: "Hello World"
```

- If the string doesn't have any leading whitespace, the `lstrip()` method returns the string as it is.

#### Example 3: Removing Leading Specific Characters

You can specify which characters you want to remove by passing a string to the `chars` parameter.

```python
s = "xxxyyyHello World"
new_string = s.lstrip('xy')
print(new_string)  # Output: "Hello World"
```

- In this case, both `'x'` and `'y'` are removed from the left side of the string.

---

### **How It Works:**

- The `lstrip()` method removes characters from the **beginning (left side)** of the string, and it stops removing when it encounters a character that is **not in the set of characters specified** (or when it reaches the end of the string).
- If no characters are specified in `chars`, it removes **whitespace** characters like spaces, tabs, and newlines.

#### Example 4: Removing Multiple Leading Characters

```python
s = "   1234Hello"
new_string = s.lstrip(' 123')
print(new_string)  # Output: "Hello"
```

- Here, the characters `' '`, `'1'`, `'2'`, and `'3'` are removed from the left side of the string.

#### Example 5: Removing All Leading Digits

```python
s = "123abc456"
new_string = s.lstrip('0123456789')
print(new_string)  # Output: "abc456"
```

- The `lstrip()` method removes all the leading digits from the string.

---

### **Advanced Usage and Considerations:**

1. **Handling Empty String:**

If the string consists entirely of characters to be removed (whitespace by default), it returns an empty string.

```python
s = "    "
new_string = s.lstrip()
print(new_string)  # Output: ""
```

- In this case, all the spaces are removed, leaving an empty string.

2. **Unicode Characters:**

The `lstrip()` method also works with Unicode characters, not just ASCII characters.

#### Example 6: Unicode Character Handling

```python
s = "   \u2003Hello World"  # The string starts with a Unicode space character.
new_string = s.lstrip()
print(new_string)  # Output: "Hello World"
```

- The Unicode character `\u2003` is considered a whitespace, and `lstrip()` removes it as well.

3. **Using `lstrip()` in Data Preprocessing:**

In text preprocessing, `lstrip()` is useful to clean strings by removing unnecessary leading spaces that might appear in input data (e.g., when reading from files or user inputs).

#### Example 7: Cleaning Text Input

```python
user_input = "    Some user input"
clean_input = user_input.lstrip()
print(clean_input)  # Output: "Some user input"
```

4. **Combination with Other String Methods:**

You can chain `lstrip()` with other string methods for more complex string manipulations.

#### Example 8: Combining `lstrip()` with `strip()`

```python
s = "   !!Python!!"
cleaned_string = s.lstrip().strip("! ")
print(cleaned_string)  # Output: "Python"
```

- First, `lstrip()` removes leading spaces, and then `strip()` removes the leading and trailing `!` and spaces.

5. **Performance Consideration:**

- The **time complexity** of `lstrip()` is **O(n)**, where `n` is the length of the string. This is because it iterates over each character from the left side until it encounters a character that is not part of the specified set of characters.
- The **space complexity** is **O(n)**, as the method returns a new string with the leading characters removed.

6. **String Comparison After `lstrip()`:**

You can use `lstrip()` for case-insensitive or leading space-insensitive string comparisons.

#### Example 9: Case-Insensitive Comparison

```python
s1 = "   Python is Awesome"
s2 = "python is awesome"

# Use lstrip to remove leading spaces, then compare in a case-insensitive way
if s1.lstrip().lower() == s2.lstrip().lower():
    print("The strings are equal (ignoring leading spaces and case).")
```

- This compares two strings after removing leading spaces and converting both strings to lowercase for a case-insensitive comparison.

---

### **Edge Cases:**

1. **String with No Leading Characters:**

```python
s = "Python"
new_string = s.lstrip("X")
print(new_string)  # Output: "Python"
```

- If no characters to be removed exist in the string, `lstrip()` returns the string as is.

2. **String with Only Leading Characters to Remove:**

```python
s = "####Hello"
new_string = s.lstrip("#")
print(new_string)  # Output: "Hello"
```

- When only specific characters like `#` are specified, they are removed from the left side, leaving the rest of the string intact.

3. **Whitespace Handling with Tabs and Newlines:**

```python
s = "\t\n  Hello"
new_string = s.lstrip()
print(new_string)  # Output: "Hello"
```

- `lstrip()` also removes tabs (`\t`), newlines (`\n`), and other forms of whitespace by default.

---

### **Conclusion:**

The `lstrip()` method is a simple yet useful tool for removing leading whitespace or specific characters from a string. It is often used in text processing and cleaning tasks to normalize strings before further analysis. The method is versatile and can be used for various character removal tasks, whether dealing with spaces, punctuation, or even custom characters.
