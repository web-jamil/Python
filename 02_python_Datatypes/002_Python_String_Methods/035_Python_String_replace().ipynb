# Python String `replace()`

The string **`replace()`** method returns a copy of the string where all occurrences of a substring is replaced with another substring.

**Syntax**:

```python
str.replace(old, new [, count])
```


## `replace()` Parameters

The **`replace()`** method can take maximum of 3 parameters:

- **`old`** - old substring you want to replace
- **`new`** - new substring which will replace the old substring
- **`count`** (optional) - the number of times you want to replace the **`old`** substring with the **`new`** substring

> **Note:** If **`count`** is not specified, the **`replace()`** method replaces all occurrences of the **`old`** substring with the **`new`** substring.


## Return Value from `replace()`

The **`replace()`** method returns a copy of the string where the **`old`** substring is replaced with the **`new`** substring. The original string is unchanged.

If the **`old`** substring is not found, it returns the copy of the original string.


````python
# Example 1: Using replace()

song = 'cold, cold heart'

# replacing 'cold' with 'hurt'
print(song.replace('cold', 'hurt'))

song = 'Let it be, let it be, let it be, let it be'

# replacing only two occurences of 'let'
print(song.replace('let', "don't let", 2))
````

````python
# Example 2: Using replace()

song = 'cold, cold heart'
replaced_song = song.replace('o', 'e')

# The original string is unchanged
print('Original string:', song)

print('Replaced string:', replaced_song)

song = 'let it be, let it be, let it be'

# maximum of 0 substring is replaced
# returns copy of the original string
print(song.replace('let', 'so', 0))
````

````python
# Example 3:

s1='I Love Python Tutorial, I Love Life, I Love Coding'
print(s1.replace('Love','Like'))
print(s1.replace('I','He',2))
````

### Python `replace()` Function: From Basic to Advanced

The `replace()` function in Python is a string method used to replace all occurrences of a specified substring with another substring. It returns a new string with the replacements made, leaving the original string unchanged.

### **Syntax:**

```python
string.replace(old, new, count)
```

- **`old`**: The substring that you want to replace.
- **`new`**: The substring that will replace the old substring.
- **`count`** (optional): The number of times to replace the old substring with the new one. If not specified, it replaces all occurrences.

### **Return Type:**

- The function returns a **new string** with all occurrences (or the first `count` occurrences) of `old` replaced by `new`.

### **Basic Examples:**

#### Example 1: Replace all occurrences of a substring

```python
s = "Hello, world! Hello, everyone!"
new_string = s.replace("Hello", "Hi")
print(new_string)  # Output: "Hi, world! Hi, everyone!"
```

#### Example 2: Replace only the first occurrence

```python
s = "apple apple apple"
new_string = s.replace("apple", "orange", 1)
print(new_string)  # Output: "orange apple apple"
```

#### Example 3: No occurrences to replace

```python
s = "apple banana cherry"
new_string = s.replace("grape", "orange")
print(new_string)  # Output: "apple banana cherry" (no change)
```

### **Handling Empty Strings:**

- **Empty `old` string**: If you pass an empty string as `old`, it will replace it at every position between characters (essentially inserting `new` between every character).

```python
s = "abc"
new_string = s.replace("", "-")
print(new_string)  # Output: "-a-b-c-" (inserts hyphen between each character)
```

- **Empty `new` string**: If you pass an empty string as `new`, it will remove all occurrences of `old`.

```python
s = "hello world"
new_string = s.replace("o", "")
print(new_string)  # Output: "hell wrld"
```

### **Edge Cases:**

#### Example 4: Replacing the substring with itself

```python
s = "apple apple apple"
new_string = s.replace("apple", "apple")
print(new_string)  # Output: "apple apple apple" (no change, as replacing with the same string)
```

#### Example 5: Replacing with an empty string

```python
s = "hello world"
new_string = s.replace("world", "")
print(new_string)  # Output: "hello "
```

### **Advanced Examples:**

#### Example 6: Replacing with special characters

```python
s = "The cost is $20"
new_string = s.replace("$", "USD ")
print(new_string)  # Output: "The cost is USD 20"
```

#### Example 7: Using count to limit the number of replacements

```python
s = "one, two, one, three, one"
new_string = s.replace("one", "zero", 2)
print(new_string)  # Output: "zero, two, zero, three, one" (only first 2 occurrences replaced)
```

#### Example 8: Case-sensitive replacement

```python
s = "Hello hello HELLO"
new_string = s.replace("hello", "hi")
print(new_string)  # Output: "Hello hi HELLO" (case-sensitive replacement)
```

#### Example 9: Replace multiple substrings (advanced usage)

If you want to replace multiple substrings at once, you can use `replace()` multiple times:

```python
s = "apple orange banana apple grape"
s = s.replace("apple", "kiwi").replace("orange", "mango").replace("banana", "pear")
print(s)  # Output: "kiwi mango pear kiwi grape"
```

### **Performance Considerations:**

- **Time Complexity**: The time complexity of `replace()` is **O(n)**, where `n` is the length of the string. This is because Python needs to iterate through the string to find and replace occurrences.
- **Space Complexity**: The space complexity is **O(n)** because a new string is created with the replacements made.

### **When to Use `replace()`?**

- **Text Preprocessing**: Use `replace()` to clean or preprocess strings by removing or substituting certain substrings.
- **String Manipulation**: It’s handy when you need to replace parts of a string, such as formatting output, updating URLs, or performing text-based transformations.

### **Limitations of `replace()`**

- **No In-place Modification**: `replace()` does not modify the original string. It returns a new string. Python strings are **immutable**, so if you want to modify the string, you'll need to assign the result to a new variable or the same variable.
- **Case-sensitive**: `replace()` is case-sensitive by default. If you need a case-insensitive replacement, you'll have to handle it manually or use regular expressions (`re.sub()`).

### **Alternatives for More Advanced Replacements:**

- **Regular Expressions (`re.sub()`)**: If you need more advanced pattern-based replacements, you can use the `re.sub()` function from the `re` module. This allows you to replace based on patterns (e.g., replacing all digits or replacing case-insensitive matches).

Example:

```python
import re

s = "Hello 123, Hello 456"
new_string = re.sub(r"\d+", "number", s)
print(new_string)  # Output: "Hello number, Hello number"
```

### **Conclusion:**

- The `replace()` method is a powerful, simple way to perform string replacements in Python.
- It supports replacing substrings with other substrings and offers optional control over the number of replacements.
- It's efficient for straightforward string manipulations, but for more complex patterns or case-insensitive replacements, you may want to explore regular expressions (`re.sub()`).
