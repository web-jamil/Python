# Python String `strip()`

The string **`strip()`** method returns a copy of the string by removing both the leading and the trailing characters (based on the string argument passed).

The **`strip()`** method removes characters from both left and right based on the argument (a string specifying the set of characters to be removed).

**Syntax**:

```python
string.strip([chars])
```


## `strip()` Parameters

The **`strip()`** method takes one parameter:

- **`chars`** (optional) - a string specifying the set of characters to be removed from the left and right part of the string.

If **`chars`** argument is not provided, all leading and trailing whitespaces are removed from the string.


## Return Value from `strip()`

The **`strip()`** method returns a copy of the string with both leading and trailing characters stripped.


## Working of the `strip()` method

- When the character of the string in the left mismatches with all the characters in the **`chars`** argument, it stops removing the leading characters.
- Similarly, when the character of the string in the right mismatches with all the characters in the **`chars`** argument, it stops removing the trailing characters.


````python
# Example 1: Working of the strip() method

string = '  xoxo love xoxo   '

# Leading and trailing whitespaces are removed
print(string.strip())

# All <whitespace>,x,o,e characters in the left
# and right of string are removed
print(string.strip(' xoe'))

# Argument doesn't contain space
# No characters are removed.
print(string.strip('stx'))

string = 'android is awesome'
print(string.strip('an'))
````

**Explanation:**

Here, we can see that the first expression **`string.strip()`** without any arguments removed the whitespaces from the left and right of **`string`**.

**`string.strip(' xoe')`** removed all whitespace, **`x`**, **`o`**, and **`e`** that lead or trailed the string.

Since **`string`** has whitespace at the beginning and end, the expression **`string.strip('stx')`** does not change the string. **`x`** is not removed since it is at the middle of the string (whitespaces lead and trail the string)

**`string.strip('an')`** removed **`an`** leading the string.


The `strip()` function in Python is a string method used to remove leading and trailing whitespace or specified characters from a string. It is commonly used for cleaning up user input or manipulating strings. Below is a comprehensive guide to the `strip()` function, from basic to advanced usage.

---

### **1. Basic Usage of `strip()`**

The `strip()` method removes leading and trailing whitespace (spaces, tabs, and newlines) by default.

#### Syntax:

```python
string.strip([chars])
```

- `chars` (optional): A string specifying the characters to remove. If omitted, it removes whitespace.

#### Example:

```python
text = "   Hello, World!   "
cleaned_text = text.strip()
print(cleaned_text)  # Output: "Hello, World!"
```

---

### **2. Removing Specific Characters**

You can specify a set of characters to remove from the beginning and end of the string.

#### Example:

```python
text = "xxxyHello, World!yyyx"
cleaned_text = text.strip("xy")
print(cleaned_text)  # Output: "Hello, World!"
```

Here, `strip("xy")` removes all `x` and `y` characters from the beginning and end of the string.

---

### **3. `lstrip()` and `rstrip()`**

- `lstrip()`: Removes leading (left) whitespace or specified characters.
- `rstrip()`: Removes trailing (right) whitespace or specified characters.

#### Example:

```python
text = "   Hello, World!   "
left_stripped = text.lstrip()
right_stripped = text.rstrip()

print(left_stripped)  # Output: "Hello, World!   "
print(right_stripped) # Output: "   Hello, World!"
```

#### Removing Specific Characters:

```python
text = "xxxyHello, World!yyyx"
left_stripped = text.lstrip("xy")
right_stripped = text.rstrip("xy")

print(left_stripped)  # Output: "Hello, World!yyyx"
print(right_stripped) # Output: "xxxyHello, World!"
```

---

### **4. Advanced Usage**

#### **a. Combining `strip()` with Other String Methods**

You can chain `strip()` with other string methods for more complex operations.

#### Example:

```python
text = "   Hello, World!   "
cleaned_text = text.strip().upper()
print(cleaned_text)  # Output: "HELLO, WORLD!"
```

#### **b. Using `strip()` with User Input**

`strip()` is often used to clean up user input.

#### Example:

```python
user_input = input("Enter your name: ").strip()
print(f"Hello, {user_input}!")
```

#### **c. Removing Multiple Characters**

You can specify multiple characters to remove.

#### Example:

```python
text = "!!!Hello, World!???"
cleaned_text = text.strip("!?")
print(cleaned_text)  # Output: "Hello, World"
```

#### **d. Using `strip()` with Unicode Characters**

`strip()` works with Unicode whitespace and characters.

#### Example:

```python
text = "　　Hello, World!　　"  # Uses full-width spaces
cleaned_text = text.strip()
print(cleaned_text)  # Output: "Hello, World!"
```

---

### **5. Common Pitfalls**

#### **a. `strip()` Does Not Modify the Original String**

Strings in Python are immutable, so `strip()` returns a new string.

#### Example:

```python
text = "   Hello, World!   "
cleaned_text = text.strip()
print(text)         # Output: "   Hello, World!   "
print(cleaned_text) # Output: "Hello, World!"
```

#### **b. `strip()` Only Removes Leading and Trailing Characters**

It does not remove characters in the middle of the string.

#### Example:

```python
text = "Hello,   World!"
cleaned_text = text.strip()
print(cleaned_text)  # Output: "Hello,   World!"
```

---

### **6. Performance Considerations**

- `strip()` is efficient for small to medium-sized strings.
- For very large strings or performance-critical applications, consider using regular expressions or other optimizations.

---

### **7. Alternatives to `strip()`**

- **Regular Expressions (`re` module):** For more complex patterns.

  ```python
  import re
  text = "   Hello, World!   "
  cleaned_text = re.sub(r'^\s+|\s+$', '', text)
  print(cleaned_text)  # Output: "Hello, World!"
  ```

- **String Slicing:** For specific use cases.
  ```python
  text = "xxxyHello, World!yyyx"
  cleaned_text = text[3:-3]
  print(cleaned_text)  # Output: "Hello, World!"
  ```

---

### **8. Summary of `strip()` Functions**

| Method     | Description                                    |
| ---------- | ---------------------------------------------- |
| `strip()`  | Removes leading and trailing whitespace/chars. |
| `lstrip()` | Removes leading whitespace/chars.              |
| `rstrip()` | Removes trailing whitespace/chars.             |
