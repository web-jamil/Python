# Python String `rindex()`

The string **`rindex()`** method returns the highest index of the substring inside the string (if found). If the substring is not found, it raises an exception.

**Syntax**:

```python
str.rindex(sub[, start[, end]] )
```


## `rindex()` Parameters

The **`rindex()`** method takes three parameters:

- **`sub`** - It's the substring to be searched in the **`str`** string.
- **`start`** and **`end`** (optional) - substring is searched within **`str[start:end]`**


## Return Value from `rindex()`

- If substring exists inside the string, it returns the highest index in the string where the substring is found.
- If substring doesn't exist inside the string, it raises a **`ValueError`** exception.

**`rindex()`** method is similar to **[rfind() method for strings](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/036_Python_String_rfind%28%29.ipynb)**.

The only difference is that **`rfind()`** returns **`-1`** if the substring is not found, whereas **`rindex()`** throws an exception.


````python
# Example 1: rindex() With No start and end Argument

quote = 'Let it be, let it be, let it be'

result = quote.rindex('let it')
print("Substring 'let it':", result)
  
result = quote.rindex('small')
print("Substring 'small ':", result)
````

> **Note:** Index in Python starts from 0 and not 1.


````python
# Example 2: rindex() With start and end Arguments

quote = 'Do small things with great love'

# Substring is searched in ' small things with great love' 
print(quote.rindex('t', 2))

# Substring is searched in 'll things with'
print(quote.rindex('th', 6, 20))

# Substring is searched in 'hings with great lov'
print(quote.rindex('o small ', 10, -1))
````

The `rindex()` method in Python is used to find the highest index of a specified substring within a string. If the substring is found, `rindex()` returns the index of its last occurrence. However, if the substring is not found, it raises a `ValueError`. This method is particularly useful when you need to locate the last position of a substring in a string.

### Syntax

```python
str.rindex(substring, start=0, end=len(string))
```

### Parameters

- **substring**: The substring whose last index you want to find.
- **start** (optional): The starting index from which to search the string. Default is `0`.
- **end** (optional): The ending index (exclusive) to search in the string. Default is `len(string)`.

### Return Value

- The method returns the highest index of the specified substring within the string.
- If the substring is not found, a `ValueError` is raised.

### Examples of Usage

1. **Basic Usage**:

   ```python
   text = "Hello, World! Hello!"
   result = text.rindex("Hello")
   print(result)  # Output: 14 (index of the last occurrence)
   ```

2. **Searching for a Substring Not Found**:

   ```python
   text = "Python programming"
   try:
       index = text.rindex("Java")
   except ValueError as e:
       print(e)  # Output: substring not found
   ```

3. **Using Start and End Parameters**:

   ```python
   text = "banana"
   result = text.rindex("a", 0, 5)  # Searching from index 0 to 5
   print(result)  # Output: 3 (last 'a' before index 5)
   ```

4. **Finding the Last Occurrence of a Character**:

   ```python
   text = "racecar"
   index = text.rindex("c")
   print(index)  # Output: 5 (last occurrence of 'c')
   ```

5. **Handling Edge Cases**:
   If you try to search within an empty string, it will raise a `ValueError`:
   ```python
   try:
       text = ""
       index = text.rindex("test")
   except ValueError as e:
       print(e)  # Output: substring not found
   ```

### Differences from `rfind()`

- **`rindex()`**: Raises a `ValueError` if the substring is not found.
- **`rfind()`**: Returns `-1` if the substring is not found, which allows for easier checks of presence without using exceptions.

### Summary

The `rindex()` method is useful for finding the last occurrence of a substring within a string in Python. Here’s a quick summary:

- **Function**: `str.rindex(substring, start=0, end=len(string))`

  - `substring`: The string to search for.

Alright, let's explore the `rindex()` method in Python strings. It's used to find the highest index (rightmost occurrence) of a substring within a string.

**What `rindex()` Does:**

The `rindex()` method searches for a specified substring within a string and returns the starting index of the last occurrence of that substring.

The key characteristic of `rindex()` is that if the substring is _not_ found, it raises a `ValueError`. This is the main difference between `rindex()` and the `rfind()` method, which returns -1 if the substring is not found.

**Syntax:**

```python
string.rindex(sub[, start[, end]])
```

**Parameters:**

- `sub`: The substring to search for. This is a mandatory argument.
- `start` (optional): An integer indicating the starting position within the string where the search should begin. The default value is 0 (the beginning of the string).
- `end` (optional): An integer indicating the ending position within the string where the search should end. The search is performed within the slice `string[start:end]`. The default value is the length of the string (the end of the string).

**Return Value:**

- An integer representing the starting index of the last occurrence of the substring `sub` within the specified range of the string.

**Raises:**

- `ValueError`: If the substring `sub` is not found within the specified range.

**Examples:**

Let's see `rindex()` in action:

```python
text = "banana"

# Find the last occurrence of "a"
index1 = text.rindex("a")
print(f"Last index of 'a' in '{text}': {index1}")  # Output: Last index of 'a' in 'banana': 5

# Find the last occurrence of "an"
index2 = text.rindex("an")
print(f"Last index of 'an' in '{text}': {index2}")  # Output: Last index of 'an' in 'banana': 3

# Find the last occurrence of "na"
index3 = text.rindex("na")
print(f"Last index of 'na' in '{text}': {index3}")  # Output: Last index of 'na' in 'banana': 4

# Searching within a specific range [start:end]
index4 = text.rindex("a", 0, 4)  # Search "a" in "bana"
print(f"Last index of 'a' in '{text}[0:4]': {index4}")  # Output: Last index of 'a' in 'banana'[0:4]: 3

# Searching for a substring that doesn't exist will raise a ValueError
try:
    index5 = text.rindex("x")
    print(f"Last index of 'x' in '{text}': {index5}")
except ValueError:
    print(f"'x' not found in '{text}' (ValueError raised)")  # Output: 'x' not found in 'banana' (ValueError raised)

# More complex example with start and end
sentence = "the quick brown fox jumps over the lazy brown dog"
last_the = sentence.rindex("the")
print(f"Last index of 'the' in '{sentence}': {last_the}")  # Output: Last index of 'the' in 'the quick brown fox jumps over the lazy brown dog': 31

last_brown = sentence.rindex("brown")
print(f"Last index of 'brown' in '{sentence}': {last_brown}")  # Output: Last index of 'brown' in 'the quick brown fox jumps over the lazy brown dog': 40

last_brown_limited = sentence.rindex("brown", 0, 30)  # Search "brown" before "jumps"
print(f"Last index of 'brown' in '{sentence}[0:30]': {last_brown_limited}")  # Output: Last index of 'brown' in 'the quick brown fox jumps over the lazy brown dog'[0:30]: 10

try:
    last_fox_limited = sentence.rindex("fox", 35)  # Search "fox" after "over"
    print(f"Last index of 'fox' in '{sentence}[35:]': {last_fox_limited}")
except ValueError:
    print(f"'fox' not found in '{sentence}[35:]' (ValueError raised)")  # Output: 'fox' not found in 'the quick brown fox jumps over the lazy brown dog'[35:] (ValueError raised)
```

**Key Characteristics:**

- **Finds the Rightmost Occurrence:** The primary purpose is to locate the last instance of the substring.
- **Raises `ValueError` if Not Found:** This is the defining difference from `rfind()`. If the substring isn't present, your program will encounter an error that needs to be handled (using `try-except` blocks).
- **Optional `start` and `end` Parameters:** You can limit the search to a specific portion (slice) of the string. The returned index is relative to the beginning of the original string, not the `start` index.

**Comparison with `rfind()`:**

- `rindex()`: Raises a `ValueError` if the substring is not found.
- `rfind()`: Returns -1 if the substring is not found.

The choice between `rindex()` and `rfind()` depends on how you want to handle the case where the substring is absent. If the absence of the substring is an exceptional condition that should halt or signal an error, `rindex()` is appropriate. If it's a normal possibility that your program should handle gracefully (e.g., by proceeding with alternative logic), `rfind()` is the better choice.

**Use Cases:**

- **Finding the Last Delimiter:** Useful for tasks like extracting the filename from a full path (finding the last `/`) or the extension from a filename (finding the last `.`).
- **Error Checking and Validation:** If the presence of a specific substring is mandatory at or before a certain point in the string, `rindex()` can be used to verify this and raise an error if it's missing.
- **Parsing Data:** When processing structured text where the last occurrence of a separator or marker has special significance.
- **String Manipulation from the End:** Performing operations based on the position of the last instance of a particular substring.

In summary, `rindex()` provides a way to locate the rightmost occurrence of a substring within a string, with the important characteristic of raising a `ValueError` if the substring is not found, making it suitable for scenarios where the presence of the substring is expected.


# Python `rindex()` String Method: Complete Guide

## What is the `rindex()` method?

The `rindex()` method is a string method that returns the **highest index** (last occurrence) of a substring in a string. It's similar to `rfind()` but raises an exception if the substring is not found.

## Syntax

```python
str.rindex(sub[, start[, end]])
```

### Parameters:

- `sub` - substring to search for (required)
- `start` - optional starting index (default 0)
- `end` - optional ending index (default end of string)

### Return Value:

- Returns the highest index where the substring is found
- Raises `ValueError` if substring is not found

## Key Characteristics

1. **Right-to-Left Search**: Finds the last occurrence of the substring
2. **Case Sensitive**: Searches are case-sensitive
3. **Optional Bounds**: Can specify search range with start/end
4. **Exception on Failure**: Raises ValueError if not found (unlike rfind())

## Basic Examples

### Simple Usage

```python
text = "hello world, hello python"
print(text.rindex("hello"))  # 13 (second occurrence)
```

### With Start/End Parameters

```python
text = "banana"
print(text.rindex("a", 0, 4))  # 3 (last 'a' in 'bana')
```

### Substring Not Found

```python
text = "hello"
try:
    print(text.rindex("x"))
except ValueError as e:
    print("Substring not found")  # This will execute
```

## Comparison with Similar Methods

| Method     | Search Direction | Returns if Not Found | Case Sensitive |
| ---------- | ---------------- | -------------------- | -------------- |
| `rindex()` | Right-to-left    | Raises ValueError    | Yes            |
| `rfind()`  | Right-to-left    | Returns -1           | Yes            |
| `index()`  | Left-to-right    | Raises ValueError    | Yes            |
| `find()`   | Left-to-right    | Returns -1           | Yes            |

## Practical Applications

### 1. Finding File Extensions

```python
filename = "archive.tar.gz"
dot_pos = filename.rindex('.')
extension = filename[dot_pos+1:]
print(extension)  # "gz"
```

### 2. Parsing Paths

```python
path = "/home/user/documents/file.txt"
last_slash = path.rindex('/')
filename = path[last_slash+1:]
print(filename)  # "file.txt"
```

### 3. Extracting Domain from Email

```python
email = "user@sub.domain.com"
at_pos = email.rindex('@')
domain = email[at_pos+1:]
print(domain)  # "sub.domain.com"
```

## Advanced Usage

### 1. Finding Nested Parentheses

```python
expression = "(a + (b * c) + d)"
close_pos = expression.rindex(')')
open_pos = expression.rindex('(', 0, close_pos)
inner = expression[open_pos+1:close_pos]
print(inner)  # "b * c"
```

### 2. Finding Last Word

```python
text = "The quick brown fox"
last_space = text.rindex(' ')
last_word = text[last_space+1:]
print(last_word)  # "fox"
```

### 3. With Slicing

```python
text = "abcXYZdefXYZghi"
# Find last XYZ before position 10
pos = text.rindex("XYZ", 0, 10)
print(pos)  # 6
```

## Common Errors

1. **Substring Not Found**:

   ```python
   "hello".rindex("x")  # ValueError: substring not found
   ```

2. **Empty Substring**:

   ```python
   "hello".rindex("")  # Returns length of string (5)
   ```

3. **Invalid Indices**:
   ```python
   "hello".rindex("l", 10)  # ValueError: substring not found
   ```

## Performance Considerations

- Time complexity is O(n) in worst case
- More efficient than using `find()` with loops to find last occurrence
- For repeated searches, consider compiling a regular expression

## Best Practices

1. **Error Handling**:

   ```python
   try:
       pos = text.rindex(sub)
   except ValueError:
       # Handle not found case
   ```

2. **Case Insensitive Search**:

   ```python
   text.lower().rindex(sub.lower())
   ```

3. **Combining with Slicing**:
   ```python
   text[text.rindex(':')+1:]  # Get everything after last colon
   ```


### Parameters

1. **substring** (required): The string you want to search for in the original string.
2. **start** (optional): The starting index from where the search should begin (default is 0).

### Return Value

### Exceptions

### Examples

```python
text = "hello world, hello universe"
index = text.rindex("hello")

# Output: 13
```

The last occurrence of `"hello"` starts at index 13.

```python
text = "hello world, hello universe"
```

Here, the search is limited to the substring `"hello world"`, and the last occurrence within this range is at index 0.

```python
text = "hello world, hello universe"
```

A `ValueError` is raised because `"Python"` is not present in the string.
