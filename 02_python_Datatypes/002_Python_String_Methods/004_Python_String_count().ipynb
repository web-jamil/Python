# Python String `count()`

The **`count()`** method returns the number of occurrences of a substring in the given string.

In simple words, **`count()`** method searches the substring in the given **[string](http://localhost:8888/notebooks/01_Learn_Python4Data/02_Python_Datatypes/002_Python_String.ipynb)** and returns how many times the substring is present in it.

It also takes optional parameters **`start`** and **`end`** to specify the starting and ending positions in the string respectively.

**Syntax**:

```python
string.count(substring, start=..., end=...)
```


## `count()` Parameters

The **`count()`** method only requires a single parameter for execution. However, it also has two optional parameters:

- **substring** - string whose count is to be found.
- **start (Optional)** - starting index within the string where search starts.
- **end (Optional)** - ending index within the string where search ends.

> **Note:** Index in Python starts from 0, not 1.


## Return Value from `count()`

The **`count()`** method returns the number of occurrences of the substring in the given string.


````python
# Example 1: Count number of occurrences of a given substring

# define string
string = "Python is awesome, isn't it?"
substring = "is"

count = string.count(substring)

# print count
print("The count is:", count)
````

````python
# Example 2: Count number of occurrences of a given substring using start and end

# define string
string = "Python is awesome, isn't it?"
substring = "i"

# count after first 'i' and before the last 'i'
count = string.count(substring, 8, 25)

# print count
print("The count is:", count)
````

**Explanation:** Here, the counting starts after the first **`i`** has been encountered, i.e. **`7th`** index position.

And, it ends before the last **`i`**, i.e. **`25th`** index position.


````python
# Example 3:

s="I Love Python Tutorial"
print(s.count('o'))      # 3
print(s.count('o',5))    # 2
print(s.count('t'))      # 2
print(s.count('t',2,9))  # 0
````

The **`count()`** method in Python is a built-in string method that returns the number of non-overlapping occurrences of a substring in a given string. It is a useful method for counting specific patterns or characters within a string.

Below is a comprehensive guide to the `count()` method, covering everything from basic syntax to advanced usage.

---

## **1. Basic Concepts**

### **What is `count()`?**

- `count()` is a string method that counts the number of occurrences of a substring in a string.
- It returns an integer representing the count of non-overlapping occurrences.

### **Why Use `count()`?**

- **Pattern Counting**: Useful for counting specific patterns or characters in a string.
- **Data Analysis**: Helps in analyzing text data by counting occurrences of specific words or phrases.
- **Validation**: Can be used to validate the presence of certain substrings in a string.

---

## **2. Basic Syntax**

### **Syntax**

```python
string.count(substring, start, end)
```

### **Parameters**

1. **`substring`**:
   - The substring to search for in the string.
2. **`start` (optional)**:
   - The starting index within the string where the search begins.
   - Default is `0`.
3. **`end` (optional)**:
   - The ending index within the string where the search ends.
   - Default is the end of the string.

### **Return Value**

- Returns the number of non-overlapping occurrences of the substring in the string.

---

## **3. Basic Usage**

### **Example 1: Counting a Substring**

```python
text = "hello world hello"
count = text.count("hello")
print(count)  # Output: 2
```

### **Example 2: Counting a Single Character**

```python
text = "hello world"
count = text.count("l")
print(count)  # Output: 3
```

### **Example 3: Counting with Start and End Indices**

```python
text = "hello world hello"
count = text.count("hello", 6, 18)
print(count)  # Output: 1
```

---

## **4. Advanced Usage**

### **1. Counting Overlapping Substrings**

- By default, `count()` counts non-overlapping occurrences. To count overlapping substrings, use a loop or regular expressions.

#### **Example**

```python
text = "abababab"
substring = "aba"
count = sum(1 for i in range(len(text)) if text.startswith(substring, i))
print(count)  # Output: 3
```

---

### **2. Counting Multiple Substrings**

- Use a loop or list comprehension to count multiple substrings.

#### **Example**

```python
text = "hello world hello"
substrings = ["hello", "world"]
counts = {s: text.count(s) for s in substrings}
print(counts)  # Output: {'hello': 2, 'world': 1}
```

---

### **3. Case-Insensitive Counting**

- Convert the string and substring to the same case (e.g., lowercase) to perform a case-insensitive count.

#### **Example**

```python
text = "Hello World Hello"
substring = "hello"
count = text.lower().count(substring.lower())
print(count)  # Output: 2
```

---

### **4. Counting Words in a Text**

- Split the text into words and use `count()` to count specific words.

#### **Example**

```python
text = "hello world hello"
words = text.split()
count = words.count("hello")
print(count)  # Output: 2
```

---

### **5. Counting with Regular Expressions**

- Use the `re` module to count substrings with more complex patterns.

#### **Example**

```python
import re

text = "hello world hello"
pattern = r"hello"
count = len(re.findall(pattern, text))
print(count)  # Output: 2
```

---

### **6. Counting in a List of Strings**

- Use a loop or list comprehension to count occurrences in a list of strings.

#### **Example**

```python
texts = ["hello world", "hello python", "hello"]
substring = "hello"
count = sum(text.count(substring) for text in texts)
print(count)  # Output: 3
```

---

## **5. Common Pitfalls**

1. **Overlapping Substrings**:

   - `count()` does not count overlapping substrings by default. Use a loop or regular expressions for overlapping counts.

2. **Case Sensitivity**:

   - `count()` is case-sensitive. Convert the string and substring to the same case for case-insensitive counting.

3. **Empty Substrings**:
   - Counting an empty substring returns the length of the string plus one.

#### **Example**

```python
text = "hello"
count = text.count("")
print(count)  # Output: 6
```

### Python `count()` Method

The `count()` method in Python is used to **count the occurrences of a substring** in a string or the occurrences of an element in a list. It is a built-in method available for both strings and lists.

### 1. `count()` for Strings

#### Syntax:

```python
str.count(substring, start=0, end=len(str))
```

#### Parameters:

- **`substring`**: The substring whose occurrences you want to count.
- **`start`** (optional): The starting index from where to begin the search (default is 0).
- **`end`** (optional): The ending index until where to search (default is the length of the string).

#### Return Value:

- Returns the **number of non-overlapping occurrences** of the `substring` in the string.

### Key Points:

- The method is case-sensitive (i.e., it treats lowercase and uppercase characters as different).
- The search is **non-overlapping**. For example, in the string `"abab"`, the substring `"ab"` will appear twice, not three times.
- If the substring is not found, it returns 0.
- The `start` and `end` parameters allow for searching a specific portion of the string.

### Examples:

#### 1. Basic Example:

```python
text = "hello world"
result = text.count("o")
print(result)  # Output: 2
```

- The letter `"o"` occurs 2 times in the string `"hello world"`.

#### 2. Using `start` and `end` Parameters:

```python
text = "hello world, hello"
result = text.count("hello", 0, 12)
print(result)  # Output: 1
```

- Here, the substring `"hello"` is only counted once because we limit the search to the first 12 characters (from index 0 to index 12).

#### 3. Case Sensitivity:

```python
text = "Hello World"
result = text.count("hello")
print(result)  # Output: 0
```

- The method is case-sensitive, so `"hello"` is not the same as `"Hello"`.

#### 4. Substring Not Found:

```python
text = "hello world"
result = text.count("x")
print(result)  # Output: 0
```

- Since the substring `"x"` is not found in the string, the result is 0.

#### 5. Count Occurrences in a Substring:

```python
text = "apple apple apple"
result = text.count("apple")
print(result)  # Output: 3
```

- The substring `"apple"` occurs 3 times in the string.

---

### 2. `count()` for Lists

The `count()` method is also available for lists to count the occurrences of an element in the list.

#### Syntax:

```python
list.count(element)
```

#### Parameters:

- **`element`**: The element whose occurrences you want to count in the list.

#### Return Value:

- Returns the number of times the `element` appears in the list.

### Key Points:

- The method counts the occurrences of **exact matches** of the element.
- If the element is not found, it returns 0.
- The method can count **duplicates** in the list.

### Examples:

#### 1. Basic Example:

```python
numbers = [1, 2, 3, 1, 1, 4]
result = numbers.count(1)
print(result)  # Output: 3
```

- The number `1` occurs 3 times in the list `[1, 2, 3, 1, 1, 4]`.

#### 2. Count String in List:

```python
words = ["apple", "banana", "apple", "cherry"]
result = words.count("apple")
print(result)  # Output: 2
```

- The string `"apple"` occurs 2 times in the list.

#### 3. Count in an Empty List:

```python
numbers = []
result = numbers.count(5)
print(result)  # Output: 0
```

- Since the list is empty, it returns 0.

#### 4. Count an Element Not in the List:

```python
numbers = [1, 2, 3, 4]
result = numbers.count(5)
print(result)  # Output: 0
```

- The number `5` is not present in the list, so it returns 0.

---

### Performance Considerations:

- **Time Complexity for String `count()`**: O(n), where `n` is the length of the string. The method needs to scan the entire string to find all occurrences of the substring.
- **Time Complexity for List `count()`**: O(n), where `n` is the length of the list. The method needs to check each element of the list to see if it matches the given element.

### Summary:

- The method is case-sensitive for strings and works on exact matches for list elements.
