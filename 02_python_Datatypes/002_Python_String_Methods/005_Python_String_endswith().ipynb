# Python String `endswith()`

The **`endswith()`** method returns True if a string ends with the specified suffix. If not, it returns False.

**Syntax**:

```python
str.endswith(suffix[, start[, end]])
```


## `endswith()` Parameters

The **`endswith()`** method takes three parameters:

- **suffix** - String or tuple of suffixes to be checked.
- **start (Optional)** - Beginning position where **suffix** is to be checked within the string.
- **end (Optional)** - Ending position where **suffix** is to be checked within the string.


## Return Value from `endswith()`

The **`endswith()`** method returns a boolean.

- It returns **True** if strings ends with the specified suffix.
- It returns **False** if string doesn't end with the specified suffix.


````python
# Example 1: endswith() Without start and end Parameters

text = "Python is easy to learn."

result = text.endswith('to learn')
# returns False
print(result)

result = text.endswith('to learn.')
# returns True
print(result)

result = text.endswith('Python is easy to learn.')
# returns True
print(result)
````

````python
# Example 2: endswith() With start and end Parameters

text = "Python programming is easy to learn."

# start parameter: 7
# "programming is easy to learn." string is searched
result = text.endswith('learn.', 7)
print(result)

# Both start and end is provided
# start: 7, end: 26
# "programming is easy" string is searched

result = text.endswith('is', 7, 26)
# Returns False
print(result)

result = text.endswith('easy', 7, 26)
# returns True
print(result)
````

## Passing Tuple to `endswith()`

It's possible to pass a tuple suffixes to the **`endswith()`** method in Python.

If the string ends with any item of the tuple, **`endswith()`** returns **`True`**. If not, it returns **`False`**.


````python
# Example 3: endswith() With Tuple Suffix

text = "programming is easy"
result = text.endswith(('programming', 'python'))

# prints False
print(result)

result = text.endswith(('python', 'easy', 'java'))

#prints True
print(result)

# With start and end parameter
# 'programming is' string is checked
result = text.endswith(('is', 'an'), 0, 14)

# prints True
print(result)
````

If you need to check if a string starts with the specified prefix, you can use **[startswith() method in Python](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/041_Python_String_startswith%28%29.ipynb)**.


### Python `endswith()` Method: From Basic to Advanced

The `endswith()` method in Python is used to check whether a string ends with a specified suffix. It is a built-in method of the `str` class and returns `True` if the string ends with the specified suffix, and `False` otherwise.

---

### **Syntax:**

```python
string.endswith(suffix[, start[, end]])
```

#### Parameters:

- **`suffix`** (required): A string or a tuple of strings to check if the string ends with.
- **`start`** (optional): The position in the string where the search should start. Default is 0.
- **`end`** (optional): The position in the string where the search should end. Default is the end of the string.

#### Return Value:

- Returns `True` if the string ends with the specified `suffix` (or one of the suffixes in the tuple).
- Returns `False` if it does not.

### **Basic Usage:**

#### Example 1: Simple usage with a string suffix

```python
s = "Hello World"
print(s.endswith("World"))  # True
print(s.endswith("world"))  # False
```

**Explanation**: The string ends with `"World"`, but not with `"world"` because Python is **case-sensitive**.

---

#### Example 2: Checking for a suffix with multiple options

```python
s = "example.txt"
print(s.endswith((".txt", ".md")))  # True
print(s.endswith((".jpg", ".png")))  # False
```

**Explanation**: The method returns `True` when the string ends with either `.txt` or `.md`, and `False` for other extensions like `.jpg`.

---

### **Advanced Usage:**

#### Example 3: Using `start` and `end` parameters

```python
s = "Welcome to Python programming"
print(s.endswith("Python", 0, 15))  # True
print(s.endswith("programming", 7, 21))  # True
```

**Explanation**: The `start` and `end` parameters allow checking if a substring within a given range ends with a specific suffix. Here, the substring `"Welcome to Python"` ends with `"Python"`, and `"Python programming"` ends with `"programming"`.

---

#### Example 4: Using `endswith()` with an empty string

```python
s = ""
print(s.endswith(""))  # True
```

**Explanation**: An empty string always ends with another empty string, so `endswith()` will return `True`.

---

#### Example 5: Checking multiple suffixes with a tuple

```python
s = "file_name.csv"
print(s.endswith(("txt", "csv", "xml")))  # True
print(s.endswith(("pdf", "docx")))  # False
```

**Explanation**: The string `"file_name.csv"` ends with `"csv"`, which is present in the tuple `("txt", "csv", "xml")`.

---

### **Edge Cases:**

1. **Empty String:**
   - An empty string always returns `True` for `endswith("")`.

```python
s = ""
print(s.endswith(""))  # True
```

2. **Case Sensitivity:**
   - The method is case-sensitive, so `"Hello".endswith("hello")` will return `False`.

```python
s = "Hello"
print(s.endswith("hello"))  # False
```

3. **Start and End Indices:**
   - Using the `start` and `end` parameters effectively allows you to check a portion of the string. For example:

```python
s = "abcabc"
print(s.endswith("abc", 0, 6))  # True
print(s.endswith("abc", 3, 6))  # True
```

In the above case, checking from `start = 3` to `end = 6` results in the string `"abc"` which ends with the suffix `"abc"`.

---

### **Performance Considerations:**

- **Time Complexity**: The time complexity of `endswith()` is **O(m)**, where `m` is the length of the `suffix`. It compares the suffix from the end of the string to determine if the string ends with the given suffix.
- **Space Complexity**: The space complexity is **O(1)** as no extra space is used except for the comparison.

---

### **Summary:**

- The `endswith()` method checks if a string ends with the specified suffix.
- **Syntax**: `string.endswith(suffix[, start[, end]])`
  - `suffix`: The string or tuple of strings you want to check for.
  - `start`: Optional. The starting position to check within the string.
  - `end`: Optional. The ending position to check within the string.
- Returns `True` if the string ends with the specified suffix, otherwise `False`.

---

#### Key Points:

- **Case-sensitive**: The method is case-sensitive, so "abc".endswith("ABC") will return `False`.
- **Empty String**: An empty string always ends with another empty string.
- **Checking Multiple Suffixes**: You can pass a tuple to check against multiple suffixes at once.
