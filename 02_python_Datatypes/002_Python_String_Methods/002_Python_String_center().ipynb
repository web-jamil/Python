# Python String `center()`

The **`center()`** method returns a string which is padded with the specified character.

**Syntax**:

```python
string.center(width[, fillchar])
```


## `center()` Parameters

The **`center()`** method takes two arguments:

- **width** - length of the string with padded characters
- **fillchar** (optional) - padding character

The **`fillchar`** argument is optional. If it's not provided, space is taken as default argument.


## Return Value from `center()`

The **`center()`** method returns a string padded with specified **`fillchar`**. It doesn't modify the original string.


````python
# Example 1: center() Method With Default fillchar

string = "Python is awesome"

new_string = string.center(24)

print("Centered String: ", new_string)
````

````python
# Example 2: center() Method With * fillchar

string = "Python is awesome"

new_string = string.center(24, '*')

print("Centered String: ", new_string)
````

### Python String `center()` Method: Concepts and Theory

The **`center()`** method in Python is used to **center-align** a string within a specified width by padding the string with a specific character. This method is often used to format strings for display purposes, especially when you need the text to appear centered in a given space.

### 1. **Definition of the `center()` Method**

The **`center()`** method takes a string and **pads** it with characters (default is spaces) on both sides so that the original string is **centered** in the specified width. If the width is smaller than the string's length, the method returns the original string.

---

### 2. **Syntax and Parameters**

#### Syntax:

```python
string.center(width, fillchar=' ')
```

- **`width`**: The total width of the string after padding. This is the required total length of the returned string.
- **`fillchar`** (optional): The character used for padding. The default is a space `' '`.

#### Description of Parameters:

- **`width`**: The total width you want the string to be. If the `width` is less than the length of the original string, the method simply returns the original string without any padding.
- **`fillchar`**: The character you want to use for padding. By default, this is a space `' '`, but you can provide any other character (such as `'-'`, `'0'`, etc.).

---

### 3. **Return Value of `center()`**

- The method returns a new string that is the **original string** centered in a **new string** of the specified width.
- If the specified width is smaller than the length of the original string, the method returns the original string unchanged.
- The padding is applied **equally** to both sides of the original string. If the padding is odd, the extra padding is added to the **right side**.

#### Example 1: Padding with Default Spaces

```python
text = "hello"
result = text.center(10)
print(result)  # Output: '  hello   '
```

- **Explanation**: The string `"hello"` is centered within a string of width `10`. Since `"hello"` has a length of `5`, the remaining `5` characters are distributed equally on both sides as spaces.

#### Example 2: Padding with Custom Character

```python
text = "hello"
result = text.center(10, '-')
print(result)  # Output: '--hello---'
```

- **Explanation**: The string `"hello"` is centered within a string of width `10`. The remaining `5` characters are filled with the character `'-'`.

#### Example 3: Width Smaller Than String Length

```python
text = "hello"
result = text.center(3)
print(result)  # Output: 'hello'
```

- **Explanation**: Since the width `3` is smaller than the length of the string `"hello"` (which is `5`), the method simply returns the original string without any modification.

---

### 4. **Example Usage of `center()`**

#### Example 1: Centering a String with Spaces

```python
text = "Python"
result = text.center(20)
print(f"'{result}'")
```

**Output:**

```
'       Python        '
```

- **Explanation**: The string `"Python"` is centered in a string of width `20`, with spaces added equally on both sides.

#### Example 2: Centering a String with a Custom Character

```python
text = "Welcome"
result = text.center(20, '*')
print(f"'{result}'")
```

**Output:**

```
'*****Welcome*****'
```

- **Explanation**: The string `"Welcome"` is centered in a string of width `20`, with `'*'` used for padding.

#### Example 3: Centering with Odd Padding

```python
text = "Data"
result = text.center(9, '0')
print(f"'{result}'")
```

**Output:**

```
'000Data000'
```

- **Explanation**: The string `"Data"` is centered in a string of width `9`, with `'0'` used for padding. The remaining characters are split equally on both sides, and the extra `'0'` is added to the right side.

---

### 5. **Time Complexity of `center()`**

The time complexity of the **`center()`** method is **O(n)**, where `n` is the specified width. This is because the method needs to create a new string of the given width and insert padding characters accordingly.

- **Best Case**: If the width is smaller than or equal to the string length, no padding is needed, so the complexity is **O(1)**.
- **Worst Case**: If the width is larger than the string length, the method needs to create a new string of the specified width, so the complexity is **O(n)**, where `n` is the width.

### 6. **Use Cases of `center()`**

1. **Text Formatting for Output**:

   - When you need to center text in a fixed-width area, like in console output or text-based UI designs.

   Example: Centering headings or titles for display:

   ```python
   title = "Report"
   print(title.center(30, '*'))
   # Output: '***************Report***************'
   ```

2. **Creating a Neat Alignment for Columns**:

   - In reports or tabular data, `center()` can be used to center-align column headers or values.

   Example:

   ```python
   header = "Name"
   print(header.center(15, '-'))
   # Output: '------Name------'
   ```

3. **Formating Strings for a Table-like Display**:

   - When you are creating text-based tables, `center()` can be used to ensure that the content is well aligned in the middle of each column.

   Example:

   ```python
   data = ["Name", "Age", "City"]
   for item in data:
       print(item.center(15, '-'))
   ```

   Output:

   ```
   ----Name-----
   ----Age------
   ----City-----
   ```

4. **Creating a Styled Heading**:

   - Centering a heading or title within a design.

   Example:

   ```python
   heading = "Python Tutorial"
   print(heading.center(50, '='))
   # Output: '=================Python Tutorial================='
   ```

---

### 7. **Edge Cases and Exceptions**

1. **Width Less Than String Length**:

   - If the width provided is less than the length of the original string, the method will not truncate the string. It will simply return the original string.

   ```python
   text = "longtext"
   result = text.center(5)
   print(result)  # Output: 'longtext'
   ```

2. **Empty String**:

   - An empty string will be padded with the specified character until it reaches the desired width.

   ```python
   text = ""
   result = text.center(10, '-')
   print(result)  # Output: '----------'
   ```

3. **Custom Padding Character**:

   - Any character can be used as padding, but if the padding character is longer than one character, the method will only use the first character for padding.

   ```python
   text = "abc"
   result = text.center(10, "xy")
   print(result)  # Output: 'xyabcxyxy'
   ```

---

### 8. **Alternatives to `center()`**

1. **Using `str.format()` for Padding**:

   - You can use **`str.format()`** to create a string with specified width and padding.

   Example:

   ```python
   text = "hello"
   result = "{:^10}".format(text)  # Center-alignment with spaces
   print(result)  # Output: '  hello   '
   ```

2. **Using F-Strings** (Python 3.6+):

   - F-strings also support alignment and padding.

   Example:

   ```python
   text = "world"
   result = f"{text:^10}"  # Center-alignment with spaces
   print(result)  # Output: '  world   '
   ```

3. **Manual Padding**:

   - You can manually create a padded string using string concatenation or multiplication.

   Example:

   ```python
   text = "example"
   width = 20
   padding = (width - len(text)) // 2
   result = ' ' * padding + text + ' ' * (width - len(text) - padding)
   print(result)  # Output: '    example     '
   ```
