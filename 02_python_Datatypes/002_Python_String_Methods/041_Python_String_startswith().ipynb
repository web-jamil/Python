# Python String `startswith()`

The string **`startswith()`** method returns **`True`** if a string starts with the specified prefix(string). If not, it returns **`False`**.

**Syntax**:

```python
str.startswith(prefix[, start[, end]])
```


## `startswith()` Parameters

The **`startswith()`** method takes a maximum of three parameters:

- **`prefix`** - String or tuple of strings to be checked
- **`start`** (optional) - Beginning position where **`prefix`** is to be checked within the string.
- **`end`** (optional) - Ending position where **`prefix`** is to be checked within the string.


## Return Value from `startswith()`

The **`startswith()`** method returns a boolean.

- It returns **`True`** if the string starts with the specified prefix.
- It returns **`False`** if the string doesn't start with the specified prefix.


````python
# Example 1: startswith() Without start and end Parameters

text = "Python is easy to learn."

result = text.startswith('is easy')
# returns False
print(result)

result = text.startswith('Python is ')
# returns True
print(result)

result = text.startswith('Python is easy to learn.')
# returns True
print(result)
````

````python
# Example 2: startswith() With start and end Parameters

text = "Python programming is easy."

# start parameter: 7
# 'programming is easy.' string is searched
result = text.startswith('programming is', 7)
print(result)

# start: 7, end: 18
# 'programming' string is searched
result = text.startswith('programming is', 7, 18)
print(result)

result = text.startswith('program', 7, 18)
print(result)
````

## Passing Tuple to `startswith()`

It's possible to pass a tuple of prefixes to **`startswith()`** method in Python.

If the string starts with any item of the tuple, **`startswith()`** returns **`True`**. If not, it returns **`False`**


````python
# Example 3: startswith() With Tuple Prefix

text = "programming is easy"
result = text.startswith(('python', 'programming'))

# prints True
print(result)

result = text.startswith(('is', 'easy', 'java'))

# prints False
print(result)

# With start and end parameter
# 'is easy' string is checked
result = text.startswith(('programming', 'easy'), 12, 19)

# prints False
print(result)
````

### Python `startswith()` Method: From Basic to Advanced

The `startswith()` method in Python is used to check if a string starts with a specified prefix. It is a built-in method of the `str` class.

---

### **Syntax:**

```python
string.startswith(prefix[, start[, end]])
```

#### Parameters:

- **`prefix`** (required): A string or a tuple of strings to check if the string starts with.
- **`start`** (optional): The position in the string where the search should start. Default is 0.
- **`end`** (optional): The position in the string where the search should end. Default is the end of the string.

#### Return Value:

- Returns `True` if the string starts with the specified `prefix` (or one of the prefixes in the tuple).
- Returns `False` if it does not.

### **Basic Usage:**

#### Example 1: Simple usage with a string prefix

```python
s = "Hello World"
print(s.startswith("Hello"))  # True
print(s.startswith("World"))  # False
```

**Explanation**: The string starts with `"Hello"`, but not with `"World"` because Python is **case-sensitive**.

---

#### Example 2: Checking for a prefix with multiple options

```python
s = "example.txt"
print(s.startswith(("text", "example")))  # True
print(s.startswith(("image", "video")))  # False
```

**Explanation**: The method returns `True` when the string starts with either `"example"` or `"text"`. It returns `False` for `"image"` and `"video"`.

---

### **Advanced Usage:**

#### Example 3: Using `start` and `end` parameters

```python
s = "Welcome to Python programming"
print(s.startswith("Welcome", 0, 7))  # True
print(s.startswith("Python", 11, 17))  # True
```

**Explanation**: The `start` and `end` parameters allow checking if a substring within a given range starts with a specific prefix. In this case, `"Welcome"` starts at position 0 and ends at position 7, while `"Python"` starts at position 11 and ends at position 17.

---

#### Example 4: Using `startswith()` with an empty string

```python
s = ""
print(s.startswith(""))  # True
```

**Explanation**: An empty string always starts with another empty string, so `startswith()` will return `True`.

---

#### Example 5: Checking multiple prefixes with a tuple

```python
s = "file_name.csv"
print(s.startswith(("file", "document", "image")))  # True
print(s.startswith(("document", "image")))  # False
```

**Explanation**: The string `"file_name.csv"` starts with `"file"`, which is present in the tuple `("file", "document", "image")`, so it returns `True`. However, it doesn't start with `"document"` or `"image"`, so the second check returns `False`.

---

### **Edge Cases:**

1. **Empty String:**
   - An empty string always returns `True` for `startswith("")`.

```python
s = ""
print(s.startswith(""))  # True
```

2. **Case Sensitivity:**
   - The method is case-sensitive, so `"Hello".startswith("hello")` will return `False`.

```python
s = "Hello"
print(s.startswith("hello"))  # False
```

3. **Start and End Indices:**
   - The `start` and `end` parameters allow you to check a portion of the string. For example:

```python
s = "abcabc"
print(s.startswith("abc", 0, 3))  # True
print(s.startswith("abc", 3, 6))  # True
```

In this example, checking from `start = 0` to `end = 3` returns `True`, as `"abc"` is the prefix of the substring. Similarly, checking from `start = 3` to `end = 6` also returns `True`.

---

### **Performance Considerations:**

- **Time Complexity**: The time complexity of `startswith()` is **O(k)**, where `k` is the length of the `prefix`. It compares the prefix from the start of the string to check if the string begins with the specified prefix.
- **Space Complexity**: The space complexity is **O(1)** as no extra space is used except for the comparison.

---

### **Summary:**

- The `startswith()` method checks if a string starts with the specified prefix.
- **Syntax**: `string.startswith(prefix[, start[, end]])`
  - `prefix`: The string or tuple of strings you want to check for.
  - `start`: Optional. The starting position to check within the string.
  - `end`: Optional. The ending position to check within the string.
- Returns `True` if the string starts with the specified prefix, otherwise `False`.

---

#### Key Points:

- **Case-sensitive**: The method is case-sensitive, so `"abc".startswith("ABC")` will return `False`.
- **Empty String**: An empty string always starts with another empty string.
- **Checking Multiple Prefixes**: You can pass a tuple to check against multiple prefixes at once.
