The `chr()` function in Python is a built-in function that converts an integer representing a Unicode code point into its corresponding character. Here’s an in-depth look at the `chr()` function, including syntax, usage, and examples.

### Syntax

```python
chr(integer)
```

### Parameters

- **integer**: An integer representing a valid Unicode code point (ranging from 0 to 1,114,111, or `0x10FFFF` in hexadecimal).

### Return Value

- The function returns a string representing the character corresponding to the given Unicode code point.

### Common Uses

1. **Convert Unicode Code Points to Characters**:

   ```python
   print(chr(65))  # Output: 'A'
   print(chr(97))  # Output: 'a'
   ```

2. **Using with ASCII Values**:

   ```python
   print(chr(48))  # Output: '0'
   ```

3. **Non-ASCII Characters**:

   ```python
   print(chr(241))  # Output: 'ñ'
   print(chr(937))  # Output: 'Ω'
   ```

4. **Special Characters**:

   ```python
   print(chr(10))   # Output: Newline (creates a line break)
   print(chr(32))   # Output: ' ' (space)
   ```

5. **Unicode Characters**:
   ```python
   print(chr(128522))  # Output: '😊'
   ```

### Error Handling

If an input integer is outside the valid range of Unicode code points, a `ValueError` is raised:

```python
try:
    print(chr(-1))          # Raises ValueError
except ValueError as e:
    print(e)  # Output: chr() arg not in range(0x110000)

try:
    print(chr(1_114_112))   # Raises ValueError
except ValueError as e:
    print(e)  # Output: chr() arg not in range(0x110000)
```

**Relationship with `ord()`:**

As mentioned earlier, `chr()` is the inverse of `ord()`:

- `ord(character)` → Unicode code point (integer)
- `chr(integer)` → Unicode character (string of length 1)

They allow you to move back and forth between the character representation and its numerical Unicode value.

```python
character = 'π'
code = ord(character)
print(f"The Unicode code point of '{character}' is: {code}")  # Output: The Unicode code point of 'π' is: 960

back_to_char = chr(code)
print(f"The character for the code point {code} is: '{back_to_char}'")  # Output: The character for the code point 960 is: 'π'
```

# Python `chr()` Function: Complete Guide

## What is the `chr()` function?

The `chr()` function is Python's built-in function that returns the string representing a character whose Unicode code point is the integer provided. It performs the inverse operation of `ord()`.

## Syntax

```python
chr(integer)
```

### Parameters:

- `integer`: An integer representing a Unicode code point (range 0 to 1,114,111 or 0x10FFFF in base 16)

### Return Value:

- Returns a string representing the character corresponding to the Unicode code point

## Basic Examples

```python
print(chr(65))     # Output: 'A'
print(chr(97))     # Output: 'a'
print(chr(8364))   # Output: '€' (Euro symbol)
print(chr(128522)) # Output: '😊' (Smiling face emoji)
print(chr(10))     # Output: '\n' (Newline character)
```

## Key Characteristics

1. **Valid Range**:

   - Accepts integers from 0 to 1,114,111 (0x10FFFF)
   - Raises `ValueError` if outside this range

   ```python
   chr(-1)  # ValueError: chr() arg not in range(0x110000)
   chr(1114112)  # ValueError: chr() arg not in range(0x110000)
   ```

2. **Unicode Support**:

   - Supports all valid Unicode characters
   - Works with emojis, special symbols, and non-Latin scripts

   ```python
   print(chr(947))    # 'γ' (Greek gamma)
   print(chr(12354))   # 'あ' (Japanese hiragana)
   ```

3. **Inverse Function**:
   - `ord()` is the inverse of `chr()`
   - `ord(chr(65))` returns 65

## Practical Applications

### 1. Generating Character Sequences

```python
# Generate alphabet
alphabet = [chr(i) for i in range(ord('a'), ord('z')+1)]
print(alphabet)  # ['a', 'b', 'c', ..., 'z']
```

### 2. ASCII Art Generation

```python
# Simple box drawing
top = chr(9484) + chr(9472)*10 + chr(9488)
middle = chr(9474) + ' '*10 + chr(9474)
bottom = chr(9492) + chr(9472)*10 + chr(9496)
print(top)
print(middle)
print(bottom)
```

### 3. Special Character Insertion

```python
# Add copyright symbol
copyright_notice = f"Copyright {chr(169)} 2023"
print(copyright_notice)  # "Copyright © 2023"
```

### 4. Password Generation

```python
import random
# Generate random password with special characters
password = ''.join(chr(random.randint(33, 126)) for _ in range(12))
print(password)
```

## Common Errors

   ```python
   chr(-1)  # ValueError

   ```

2. **Non-Integer Input**:
   ```python
   chr('A')  # TypeError: an integer is required
   ```

## Advanced Usage

### 1. Working with Surrogate Pairs

```python
# Characters beyond BMP (Basic Multilingual Plane)
high_char = chr(0x1F600)  # 😀
print(high_char)  # Works in Python 3
```

### 2. Creating Custom Character Maps

```python
# Create substitution cipher
cipher_map = {chr(i): chr(i+1) for i in range(ord('a'), ord('z'))}
cipher_map['z'] = 'a'  # Wrap around
print(cipher_map['a'])  # 'b'
```

### 3. Hexadecimal/Octal Conversion

```python
# Convert hex to character
hex_value = '61'
char = chr(int(hex_value, 16))
print(char)  # 'a'
```

## Security Considerations

When using `chr()` with user input:

- Always validate the integer range
- Be cautious with control characters (0-31)
- Consider sanitizing output for display purposes

```python
# Safe character conversion
def safe_chr(num):
    try:
        c = chr(num)
        return c if c.isprintable() else f'\\x{num:02x}'
    except ValueError:
        return '[INVALID]'
```
