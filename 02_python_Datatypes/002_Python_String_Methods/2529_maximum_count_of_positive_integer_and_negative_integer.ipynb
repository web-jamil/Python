**2529. Maximum Count of Positive Integer and Negative Integer**

# **Problem Statement**

Given an array `nums` sorted in **non-decreasing order**, return the maximum count between the number of **positive** integers and **negative** integers.

- `0` is **neither** positive nor negative.

### **Example 1**

#### **Input:**

```python
nums = [-2,-1,-1,1,2,3]
```

#### **Output:**

```
3
```

#### **Explanation:**

- Negative numbers: `[-2, -1, -1]` → count = **3**
- Positive numbers: `[1, 2, 3]` → count = **3**
- The maximum count is **3**.

---

### **Example 2**

#### **Input:**

```python
nums = [-3,-2,-1,0,0,1,2]
```

#### **Output:**

```
3
```

#### **Explanation:**

- Negative numbers: `[-3, -2, -1]` → count = **3**
- Positive numbers: `[1, 2]` → count = **2**
- The maximum count is **3**.

---

### **Example 3**

#### **Input:**

```python
nums = [5,20,66,1314]
```

#### **Output:**

```
4
```

#### **Explanation:**

- Negative numbers: `[]` → count = **0**
- Positive numbers: `[5, 20, 66, 1314]` → count = **4**
- The maximum count is **4**.

---

## **Constraints**

- `1 <= nums.length <= 2000`
- `-2000 <= nums[i] <= 2000`
- `nums` is **sorted in non-decreasing order**.

---

## **Solution Approach**

### **Brute Force (O(n))**

- Count the number of positive and negative integers **using a loop**.
- Return the maximum of both counts.

### **Optimized Approach (O(log n)) - Binary Search**

- Since the array is sorted:
  - **Find the first positive number's index** (using binary search).
  - **Find the first non-negative number's index** (using binary search).
  - Compute `neg_count` and `pos_count` using indices.

---

## **Code Implementation (O(log n) Approach)**

```python
from bisect import bisect_left, bisect_right

def maximumCount(nums):
    neg_count = bisect_left(nums, 0)   # First occurrence of 0 (negative count)
    pos_count = len(nums) - bisect_right(nums, 0)  # Count of positives
    return max(neg_count, pos_count)

# Test cases
print(maximumCount([-2,-1,-1,1,2,3]))  # Output: 3
print(maximumCount([-3,-2,-1,0,0,1,2]))  # Output: 3
print(maximumCount([5,20,66,1314]))  # Output: 4
```

---

## **Complexity Analysis**

| Approach                                | Time Complexity | Space Complexity |
| --------------------------------------- | --------------- | ---------------- |
| **Brute Force (Loop)**                  | `O(n)`          | `O(1)`           |
| **Binary Search (`bisect_left/right`)** | `O(log n)`      | `O(1)`           |

### **Why is Binary Search O(log n)?**

- `bisect_left(nums, 0)` finds the first non-negative number’s index in **O(log n)**.
- `bisect_right(nums, 0)` finds the first positive number’s index in **O(log n)**.
- Both operations combined are still **O(log n)**.

---

## **Follow-up**

Can this be solved in `O(log n)` time?  
✅ Yes! The **binary search approach** achieves `O(log n)` complexity. 🚀


````python
def maximumCount(nums):
    neg_count = 0  # Counter for negative numbers
    pos_count = 0  # Counter for positive numbers

    for num in nums:
        if num < 0:
            neg_count += 1
        elif num > 0:
            pos_count += 1

    return max(neg_count, pos_count)

# Test Cases
test_cases = [
    ([-2,-1,-1,1,2,3], 3),  # Equal pos and neg count
    ([-3,-2,-1,0,0,1,2], 3),  # Contains zeros
    ([5,20,66,1314], 4),  # All positive numbers
    ([-5,-4,-3,-2,-1], 5),  # All negative numbers
    ([0, 0, 0, 0], 0),  # All zeros
    ([1, 2, 3, 4, 5], 5),  # Only positive numbers
    ([-10, -5, -2, 0, 3, 7, 8], 3),  # Mixed case
    ([1000], 1),  # Single positive number
    ([-1000], 1),  # Single negative number
    ([0], 0),  # Single zero
    ([1, -1], 1),  # Equal count of positives and negatives
]

# Running tests
for i, (nums, expected) in enumerate(test_cases):
    result = maximumCount(nums)
    print(f"Test {i+1}: {nums} → Output: {result}, Expected: {expected}, {'✅' if result == expected else '❌'}")
````

Here’s the **Binary Search Approach** **without using the `bisect` module** to solve the problem in **O(log n)** time complexity.

---

## **Approach**

Since the array is sorted in **non-decreasing order**, we can use **binary search** to efficiently find:

1. The **first non-negative number's index** (this gives the count of negatives).
2. The **first positive number's index** (this gives the count of positives).

We then return the maximum of these two counts.

---

## **Binary Search Function**

We write a helper function **`binary_search_first(nums, target)`** to find:

- The **first index** where `nums[i] >= target`.

---

## **Python Code (Binary Search - O(log n))**

```python
def binary_search_first(nums, target):
    """Finds the first index where nums[index] >= target."""
    left, right = 0, len(nums)  # Set search bounds
    while left < right:
        mid = (left + right) // 2
        if nums[mid] >= target:
            right = mid  # Look for an earlier occurrence
        else:
            left = mid + 1  # Move right
    return left  # First index where nums[index] >= target

def maximumCount(nums):
    # Find the first occurrence of 0 (neg_count)
    neg_count = binary_search_first(nums, 0)
    # Find the first occurrence of a positive number (pos_count)
    pos_count = len(nums) - binary_search_first(nums, 1)

    return max(neg_count, pos_count)

# Test Cases
test_cases = [
    ([-2,-1,-1,1,2,3], 3),  # Equal pos and neg count
    ([-3,-2,-1,0,0,1,2], 3),  # Contains zeros
    ([5,20,66,1314], 4),  # All positive numbers
    ([-5,-4,-3,-2,-1], 5),  # All negative numbers
    ([0, 0, 0, 0], 0),  # All zeros
    ([1, 2, 3, 4, 5], 5),  # Only positive numbers
    ([-10, -5, -2, 0, 3, 7, 8], 3),  # Mixed case
    ([1000], 1),  # Single positive number
    ([-1000], 1),  # Single negative number
    ([0], 0),  # Single zero
    ([1, -1], 1),  # Equal count of positives and negatives
]

# Running tests
for i, (nums, expected) in enumerate(test_cases):
    result = maximumCount(nums)
    print(f"Test {i+1}: {nums} → Output: {result}, Expected: {expected}, {'✅' if result == expected else '❌'}")
```

---

## **Complexity Analysis**

| Approach                                  | Time Complexity | Space Complexity |
| ----------------------------------------- | --------------- | ---------------- |
| **Binary Search (`binary_search_first`)** | `O(log n)`      | `O(1)`           |

- **Time Complexity:** `O(log n)`
  - We perform **two binary searches**, each taking `O(log n)`, so the total remains **O(log n)**.
- **Space Complexity:** `O(1)`
  - No extra space is used except a few integer variables.

---

## **Why Does This Work?**

1. **Finding negatives**:

   - `binary_search_first(nums, 0)` finds the **first index where `nums[i] >= 0`**.
   - The count of negatives is simply **this index** (`neg_count`).

2. **Finding positives**:
   - `binary_search_first(nums, 1)` finds the **first index where `nums[i] >= 1`**.
   - The count of positives is **everything from this index to the end** (`pos_count = len(nums) - this index`).

---

## **Edge Cases Covered**

✅ All negative numbers  
✅ All positive numbers  
✅ Contains zeros only  
✅ Mix of negatives, zeros, and positives  
✅ Single element cases  
✅ Maximum length of `2000` elements

---

| Approach                     | Time Complexity | Best Use Case                 |

| **Brute Force (O(n))**       | `O(n)`          | Small arrays (≤1000 elements) |
| **Binary Search (O(log n))** | `O(log n)`      | Large arrays (≥1000 elements) |
