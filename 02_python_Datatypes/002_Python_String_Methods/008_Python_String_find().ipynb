# Python String `find()`

The **`find()`** method returns the index of first occurrence of the substring (if found). If not found, it returns **-1**.

**Syntax**:

```python
str.find(sub[, start[, end]] )
```


## `find()` Parameters

The **`find()`** method takes maximum of three parameters:

- **sub** - It is the substring to be searched in the **`str`** string.
- **start** and **end** (optional) - The range **`str[start:end]`** within which substring is searched.


## Return Value from `find()`

The **`find()`** method returns an integer value:

- If the substring exists inside the string, it returns the index of the first occurence of the substring.
- If substring doesn't exist inside the string, it returns **-1**.

````python
# Example 1: find() With No start and end Argument

quote = 'Let it be, let it be, let it be'

# first occurance of 'let it'(case sensitive)
result = quote.find('let it')
print("Substring 'let it':", result)

# find returns -1 if substring not found
result = quote.find('small')
print("Substring 'small ':", result)

# How to use find()
if (quote.find('be,') != -1):
    print("Contains substring 'be,'")
else:
    print("Doesn't contain substring")
````

````python
# Example 2: find() With start and end Arguments

quote = 'Do small things with great love'

# Substring is searched in 'hings with great love'
print(quote.find('small things', 10))

# Substring is searched in ' small things with great love' 
print(quote.find('small things', 2))

# Substring is searched in 'hings with great lov'
print(quote.find('o small ', 10, -1))

# Substring is searched in 'll things with'
print(quote.find('things ', 6, 20))
````

### Python String `find()` Method: Concepts and Theory

The **`find()`** method in Python is used to search for a specified substring within a string and return the index of its first occurrence. If the substring is not found, it returns **`-1`**.

This method is useful for checking if a string contains a specific substring and locating its position in the string.

### 1. **Definition of the `find()` Method**

The **`find()`** method is used to search for a specified substring in a given string. It returns the **index of the first occurrence** of the substring within the string. If the substring is not found, it returns **`-1`**.

---

### 2. **Syntax and Parameters**

#### Syntax:

```python
string.find(substring, start=0, end=len(string))
```

- **`substring`**: The substring you want to search for.
- **`start`** (optional): The index from where to start the search. Default is `0`.
- **`end`** (optional): The index where the search ends. Default is the length of the string (`len(string)`).

#### Description of Parameters:

- **`substring`**: This is the substring you want to find within the string.
- **`start`**: This optional parameter allows you to specify a starting index from where the search begins. By default, it starts at index `0` (the beginning of the string).
- **`end`**: This optional parameter allows you to specify the ending index up to which the search will occur. By default, the search will go up to the end of the string.

---

### 3. **Return Value of `find()`**

- **If the substring is found**: The method returns the **index** of the first occurrence of the substring in the string.
- **If the substring is not found**: The method returns **`-1`**.

#### Example 1: Substring Found

```python
text = "hello world"
index = text.find("world")
print(index)  # Output: 6
```

- **Explanation**: The substring "world" starts at index `6` in the string `text`.

#### Example 2: Substring Not Found

```python
text = "hello world"
index = text.find("python")
print(index)  # Output: -1
```

- **Explanation**: The substring "python" is not found in the string, so `find()` returns `-1`.

---

### 4. **Example Usage of `find()`**

#### Example 1: Basic Usage

```python
text = "The quick brown fox jumps over the lazy dog"
result = text.find("fox")
print(result)  # Output: 16
```

- **Explanation**: The substring "fox" starts at index `16` in the string `text`.

#### Example 2: Using `start` and `end` Parameters

```python
text = "The quick brown fox jumps over the lazy dog"
result = text.find("fox", 10, 30)
print(result)  # Output: 16
```

- **Explanation**: The search for "fox" starts from index `10` and ends at index `30`. The substring "fox" is still found starting at index `16`.

#### Example 3: Searching for a Substring That Doesn't Exist

```python
text = "The quick brown fox"
result = text.find("cat")
print(result)  # Output: -1
```

- **Explanation**: The substring "cat" is not found in the string, so `find()` returns `-1`.

---

### 5. **Time Complexity of `find()`**

The time complexity of the **`find()`** method is **O(n)**, where `n` is the length of the string. This is because the method may need to scan the entire string in the worst case to locate the substring.

- **Best Case**: If the substring is found early in the string, the time complexity is **O(k)**, where `k` is the position of the substring.
- **Worst Case**: The time complexity is **O(n)**, where `n` is the length of the string (when the substring is not found or is located at the end of the string).

---

### 6. **Use Cases of `find()`**

1. **Checking for Substring Presence**: **`find()`** is often used to check if a substring exists within a string. If the return value is `-1`, the substring is not present.

   ```python
   text = "apple pie"
   if text.find("pie") != -1:
       print("Substring found!")
   else:
       print("Substring not found.")
   ```

2. **Extracting Substrings**: If you want to extract part of a string based on the position of a certain substring, you can use **`find()`** to locate the index and then slice the string.

   ```python
   text = "apple pie"
   start_index = text.find("pie")
   if start_index != -1:
       print(text[start_index:])  # Output: pie
   ```

3. **Validating User Input**: You can use **`find()`** to check if the input string contains certain words or phrases. For example, checking for a specific domain in an email address or validating a specific code.

4. **Finding Position for Further Processing**: **`find()`** is useful when you need to process parts of the string after locating a specific substring.

---

### 7. **Edge Cases and Exceptions**

1. **Empty String**: If the string is empty, **`find()`** will always return `-1`, regardless of the substring.

   ```python
   text = ""
   result = text.find("anything")
   print(result)  # Output: -1
   ```

2. **Substring is an Empty String**: If the substring is an empty string, **`find()`** returns `0` because an empty string is always found at the beginning of any string.

   ```python
   text = "hello"
   result = text.find("")
   print(result)  # Output: 0
   ```

3. **Case Sensitivity**: **`find()`** is case-sensitive, so the case of the substring matters.

   ```python
   text = "Hello World"
   result = text.find("hello")
   print(result)  # Output: -1
   ```

   - **Explanation**: "hello" (lowercase) is not found in "Hello World" because the method is case-sensitive.

4. **Finding Multiple Occurrences**: **`find()`** only returns the **first** occurrence of the substring. If you want to find multiple occurrences, you would need to call `find()` repeatedly, updating the `start` position each time.

   ```python
   text = "hello hello hello"
   first_occurrence = text.find("hello")
   second_occurrence = text.find("hello", first_occurrence + 1)
   print(first_occurrence, second_occurrence)  # Output: 0 6
   ```

---

### 8. **Alternatives to `find()`**

1. **`index()` Method**:

   - **`index()`** is similar to **`find()`**, but it raises a **`ValueError`** if the substring is not found.
   - Example:

   ```python
   text = "apple pie"
   result = text.index("pie")
   print(result)  # Output: 6
   ```

2. **Regular Expressions (`re` module)**:

   - You can use the **`re`** module for more advanced searching patterns (such as wildcards or case-insensitive searches).
   - Example:

   ```python
   import re
   text = "apple pie"
   result = re.search("pie", text)
   if result:
       print(result.start())  # Output: 6
   ```

3. **`in` Operator**:
   - The **`in`** operator checks if a substring exists within a string and returns `True` or `False`. This is a simpler alternative if you only need to check for existence.
   - Example:
   ```python
   text = "apple pie"
   if "pie" in text:
       print("Substring found!")  # Output: Substring found!
   ```

---

### Conclusion

The **`find()`** method in Python is a simple and powerful way to search for substrings within a string. It is useful for determining the position of the first occurrence of a substring and is widely used in string processing tasks. The method provides options for specifying search ranges and handles edge cases like empty strings and case sensitivity. While it has a time complexity of **O(n)**, it is efficient enough for most use cases. If you need to locate the first occurrence of a substring, **`find()`** is a go-to method, and for more complex substring search needs, you can consider using regular expressions or other methods.
