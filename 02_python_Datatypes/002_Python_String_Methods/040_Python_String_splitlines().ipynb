# Python String `splitlines()`

The string **`splitlines()`** method splits the string at line breaks and returns a list of lines in the string.

**Syntax**:

```python
str.splitlines([keepends])
```


## `splitlines()` Parameters

The **`splitlines()`** method takes maximum of 1 parameter:

- **`keepends`** (optional)- If **`keepends`** is provided and **`True`**, line breaks are also included in items of the list.

By default, the line breaks are not included.


## Return Value from `splitlines()`

The **`splitlines()`** method returns a list of lines in the string.

If there are not line break characters, it returns a list with a single item (a single line).


| Representation         | Description                     |
| :--------------------- | :------------------------------ |
| **`\n`**               | **Line Feed**                   |
| **`\r`**               | **Carriage Return**             |
| **`\r\n`**             | **Carriage Return + Line Feed** |
| **`\v`** or **`\x0b`** | **Line Tabulation**             |
| **`\f`** or **`\x0c`** | **Form Feed**                   |
| **`\x1c`**             | **File Separator**              |
| **`\x1d`**             | **Group Separator**             |
| **`\x1e`**             | **Record Separator**            |
| **`\x85`**             | **Next Line (C1 Control Code)** |
| **`\u2028`**           | **Line Separator**              |
| **`\u2029`**           | **Paragraph Separator**         |


````python
# Example 1: How splitlines() works?

grocery = 'Milk\nCoffee\r\nBread\rButter'

print(grocery.splitlines())
print(grocery.splitlines(True))

grocery = 'Milk Coffee Bread Butter'
print(grocery.splitlines())
````

# Python `splitlines()` Method: Complete Guide with Examples

## What is the `splitlines()` method?

The `splitlines()` method splits a string at line boundaries and returns a list of lines in the string. It recognizes all common line ending conventions across different platforms.

## Syntax

```python
str.splitlines([keepends])
```

### Parameters:

- `keepends` (optional):
  - If `True`, line breaks are included in the resulting list
  - If `False` (default), line breaks are not included

### Return Value:

- Returns a list of the lines in the string, breaking at line boundaries

## Supported Line Boundaries

The method recognizes all of these as line boundaries:

- `\n` - Line Feed (Unix/Mac)
- `\r` - Carriage Return (old Mac)
- `\r\n` - Carriage Return + Line Feed (Windows)
- `\v` or `\x0b` - Vertical Tab
- `\f` or `\x0c` - Form Feed
- `\x1c` - File Separator
- `\x1d` - Group Separator
- `\x1e` - Record Separator
- `\x85` - Next Line (C1 control code)
- `\u2028` - Line Separator
- `\u2029` - Paragraph Separator

## Basic Examples

### Simple Usage

```python
text = "First line\nSecond line\r\nThird line"
lines = text.splitlines()
print(lines)
# Output: ['First line', 'Second line', 'Third line']
```

### With keepends=True

```python
text = "Line 1\nLine 2\r\nLine 3"
lines = text.splitlines(True)
print(lines)
# Output: ['Line 1\n', 'Line 2\r\n', 'Line 3']
```

### Mixed Line Endings

```python
text = "Unix\nWindows\r\nOld Mac\rEnd"
print(text.splitlines())
# Output: ['Unix', 'Windows', 'Old Mac', 'End']
```

## Practical Applications

### 1. Processing Text Files

```python
with open('data.txt', 'r') as file:
    lines = file.read().splitlines()  # Better than readlines() as it removes \n
    for line in lines:
        print(f"Processed: {line}")
```

### 2. Counting Lines

```python
text = """This is line 1
This is line 2
This is line 3"""
line_count = len(text.splitlines())
print(f"Total lines: {line_count}")  # Output: Total lines: 3
```

### 3. Parsing Multi-line Strings

```python
config = """host=localhost
port=5432
user=admin"""
settings = {}
for line in config.splitlines():
    if '=' in line:
        key, value = line.split('=', 1)
        settings[key] = value
print(settings)
# Output: {'host': 'localhost', 'port': '5432', 'user': 'admin'}
```

## Comparison with Similar Methods

| Method         | Splits At           | Returns | Line Breaks | Empty Lines |
| -------------- | ------------------- | ------- | ----------- | ----------- |
| `splitlines()` | All line boundaries | List    | Optional    | Preserved   |
| `split('\n')`  | Only `\n`           | List    | Never       | Preserved   |
| `readlines()`  | All line boundaries | List    | Always      | Preserved   |

### 3. Reconstructing Original String

```python

```

## Common Mistakes

1. **Confusing with `split('\n')`**:

   ```python
   text = "Line1\r\nLine2"
   print(text.split('\n'))      # ['Line1\r', 'Line2'] - doesn't handle \r\n properly
   print(text.splitlines())     # ['Line1', 'Line2'] - handles all line endings
   ```

2. **Assuming Universal Newlines**:
   ```python
   # When reading files, better to use splitlines() than split('\n')
   with open('file.txt', 'r') as f:
       # Prefer:
       lines = f.read().splitlines()
       # Over:
       lines = f.read().split('\n')
   ```

## Performance Considerations

- More efficient than manually splitting on multiple line ending types
- For very large files, consider iterating line by line instead:
  ```python
  with open('bigfile.txt') as f:
      for line in f:
          line = line.rstrip('\r\n')  # Remove line endings
          # process line
  ```

## Best Practices

1. **Use for Cross-Platform Compatibility**:

   - Handles all line ending styles uniformly
   - Essential when processing text from different operating systems

2. **Prefer over `readlines()`**:

   ```python
   # Instead of:
   with open('file.txt') as f:
       lines = f.readlines()  # Keeps \n at end
       lines = [line.rstrip('\n') for line in lines]

   # Better:
   with open('file.txt') as f:
       lines = f.read().splitlines()  # Cleaner
   ```

3. **Combine with Other Methods**:
   ```python
   # Clean processing of multi-line string
   data = "\n".join(line.strip() for line in text.splitlines() if line.strip())
   ```

The `splitlines()` method in Python is used to split a string into a list of lines. It is particularly useful for handling strings that contain newline characters (`\n`, `\r`, or combinations like `\r\n`). The method breaks the string at line boundaries and returns the individual lines as elements of a list.

---

### Syntax

```python
string.splitlines(keepends=False)
```

### Parameters

- **keepends** (optional): A boolean value.
  - `False` (default): Does not include line-ending characters (e.g., `\n`, `\r`) in the resulting list.
  - `True`: Keeps the line-ending characters at the end of each line.

### Return Value

- Returns a list of strings, where each string represents a line in the original string.

---

### Examples

#### Example 1: Basic Usage

```python
text = "Hello world\nPython is fun\nLet's learn"
lines = text.splitlines()
print(lines)
# Output: ['Hello world', 'Python is fun', "Let's learn"]
```

Here, the method splits the string at each newline character (`\n`), and the line-ending characters are not included in the result.

---

#### Example 2: Using `keepends=True`

```python
text = "Hello world\nPython is fun\r\nLet's learn"
lines = text.splitlines(keepends=True)
print(lines)
# Output: ['Hello world\n', 'Python is fun\r\n', "Let's learn"]
```

In this case, the line-ending characters (`\n`, `\r\n`) are preserved in the output.

---

#### Example 3: Handling Different Line Endings

The `splitlines()` method automatically handles different types of newline characters.

```python
text = "Line 1\rLine 2\nLine 3\r\nLine 4"
lines = text.splitlines()
print(lines)
# Output: ['Line 1', 'Line 2', 'Line 3', 'Line 4']
```

Here, the method splits the string at `\r`, `\n`, and `\r\n`.

---

#### Example 4: Empty Lines

If the string contains empty lines (e.g., consecutive newline characters), they are retained as empty strings in the result.

```python
text = "Line 1\n\nLine 2\n"
lines = text.splitlines()
print(lines)
# Output: ['Line 1', '', 'Line 2']
```

---

### Summary

The `splitlines()` method is a powerful way to process multiline strings. Key features include:

- Automatically handling various newline characters (`\n`, `\r`, `\r\n`).
- Providing control over whether to retain line-ending characters using `keepends`.
- Ensuring that empty lines are preserved.


the `splitlines()` method in Python strings. It's designed specifically to break a string into a list of lines, handling various line boundaries.

**What `splitlines()` Does:**

The `splitlines()` method splits a string into a list of substrings based on line boundaries. It recognizes several common ways that lines are terminated, making it more versatile than simply splitting by a single newline character (`\n`).

**Syntax:**

```python
string.splitlines([keepends])
```

**Parameters:**

- `keepends` (optional): A boolean value.
  - If `True`, the line breaks are included in the resulting list of lines.
  - If `False` (the default), the line breaks are not included.

**Return Value:**

- A list of strings, where each string is a line from the original string.

**Line Boundaries Recognized by `splitlines()`:**

`splitlines()` handles the following line break characters:

- `\n` (Line Feed) - Unix, Linux, macOS
- `\r` (Carriage Return) - Older macOS
- `\r\n` (Carriage Return + Line Feed) - Windows
- `\v` or `\x0b` (Line Tabulation)
- `\f` or `\x0c` (Form Feed)
- `\x1c` (File Separator)
- `\x1d` (Group Separator)
- `\x1e` (Record Separator)
- `\x85` (Next Line (NEL))
- `\u2028` (Line Separator)
- `\u2029` (Paragraph Separator)

**Examples:**

Let's see `splitlines()` in action with different line endings and the `keepends` parameter:

```python
text = "Line 1\nLine 2\rLine 3\r\nLine 4"

# Default behavior (keepends=False)
lines1 = text.splitlines()
print(f"splitlines() (default):\n{lines1}")
# Output:
# splitlines() (default):
# ['Line 1', 'Line 2', 'Line 3', 'Line 4']

# keepends=True
lines2 = text.splitlines(keepends=True)
print(f"\nsplitlines(keepends=True):\n{lines2}")
# Output:
# splitlines(keepends=True):
# ['Line 1\n', 'Line 2\r', 'Line 3\r\n', 'Line 4']

text2 = "Single line"
lines3 = text2.splitlines()
print(f"\nsplitlines() on a single line: {lines3}")
# Output:
# splitlines() on a single line: ['Single line']

text3 = "Line with\vvertical tab"
lines4 = text3.splitlines()
print(f"\nsplitlines() with vertical tab: {lines4}")
# Output:
# splitlines() with vertical tab: ['Line with', 'vertical tab']

text4 = "Line with\fform feed"
lines5 = text4.splitlines()
print(f"\nsplitlines() with form feed: {lines5}")
# Output:
# splitlines() with form feed: ['Line with', 'form feed']

text5 = "Line with\u2028line separator"
lines6 = text5.splitlines()
print(f"\nsplitlines() with line separator: {lines6}")
# Output:
# splitlines() with line separator: ['Line with', 'line separator']

text6 = "Line with\u2029paragraph separator"
lines7 = text6.splitlines()
print(f"\nsplitlines() with paragraph separator: {lines7}")
# Output:
# splitlines() with paragraph separator: ['Line with', 'paragraph separator']

text7 = "Line ending with a newline\n"
lines8 = text7.splitlines()
print(f"\nsplitlines() with trailing newline (default): {lines8}")
# Output:
# splitlines() with trailing newline (default): ['Line ending with a newline']

lines9 = text7.splitlines(keepends=True)
print(f"\nsplitlines() with trailing newline (keepends=True): {lines9}")
# Output:
# splitlines() with trailing newline (keepends=True): ['Line ending with a newline\n']

text8 = "\nFirst line\n\nThird line"
lines10 = text8.splitlines()
print(f"\nsplitlines() with empty lines: {lines10}")
# Output:
# splitlines() with empty lines: ['', 'First line', '', 'Third line']

lines11 = text8.splitlines(keepends=True)
print(f"\nsplitlines() with empty lines (keepends=True): {lines11}")
# Output:
# splitlines() with empty lines (keepends=True): ['\n', 'First line\n', '\n', 'Third line']
```

**Key Characteristics:**

- **Handles Multiple Line Endings:** This is the primary advantage over a simple `split('\n')`. It correctly identifies and splits lines regardless of the operating system or the specific line break character used.
- **Optional `keepends` Parameter:** Provides flexibility in whether to retain the line break characters in the resulting list. This can be useful if you need to reconstruct the original string or process lines while preserving their terminators.
- **Empty Lines:** Empty lines (resulting from consecutive line breaks) are included as empty strings in the output list.
- **Trailing Line Breaks:** If the string ends with a line break, `splitlines()` (with the default `keepends=False`) will not include an extra empty string at the end of the list. However, with `keepends=True`, the trailing line break will be included in the last element.
- **Unicode Support:** `splitlines()` correctly handles Unicode line separator characters.

**Use Cases:**

`splitlines()` is particularly useful when you need to:

- Process text files that might have been created on different operating systems with varying line endings.
- Iterate over the lines of a multi-line string.
- Count the number of lines in a string.
- Prepare text data for further analysis or processing, where each line needs to be treated as a separate unit.
- Implement text editors or similar applications that need to handle line breaks consistently.

In summary, `splitlines()` is a robust and convenient method for breaking a string into a list of lines, intelligently handling various line break conventions and offering the option to keep the line terminators. It's often the preferred way to split multi-line text in Python.
