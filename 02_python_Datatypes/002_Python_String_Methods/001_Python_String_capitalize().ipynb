# Python String `capitalize()`

In Python, the **`capitalize()`** method converts first character of a string to uppercase letter and lowercases all other characters, if any.

**Syntax**:

```python
string.capitalize()
```


## `capitalize()` Parameters

The **`capitalize()`** function doesn't take any parameter.


## Return Value from `capitalize()`

The **`capitalize()`** function returns a string with the first letter capitalized and all other characters lowercased. It doesn't modify the original string.


````python
# Example 1: Capitalize a Sentence

string = "python is AWesome."

capitalized_string = string.capitalize()

print('Old String: ', string)
print('Capitalized String:', capitalized_string)
````

````python
# Example 2: Non-alphabetic First Character

string = "+ is an operator."

new_string = string.capitalize()

print('Old String:', string)
print('New String:', new_string)
````

The **`capitalize()`** method in Python is a built-in string method that returns a copy of the string with the first character capitalized and the rest of the characters in lowercase. This method is useful for formatting strings, especially when you want to ensure that the first character of a string is capitalized, such as in titles or sentences.

## **1. Basic Concepts**

### **What is `capitalize()`?**

- `capitalize()` is a string method that capitalizes the first character of a string and makes the rest of the characters lowercase.
- It does not modify the original string but returns a new string with the changes.

### **Why Use `capitalize()`?**

- **Formatting**: Ensures the first character of a string is capitalized.
- **Consistency**: Useful for standardizing the format of strings (e.g., names, titles).
- **Readability**: Improves the readability of text by following capitalization rules.

### **Syntax**

## **4. Advanced Usage**

### **1. Capitalizing Each Word in a String**

- Use the `title()` method to capitalize the first letter of each word in a string.

#### **Example**

```python
text = "hello world"
result = text.title()
print(result)  # Output: Hello World
```

---

### **2. Capitalizing Sentences**

- Use the `capitalize()` method in combination with splitting and joining to capitalize each sentence in a string.

#### **Example**

```python
text = "hello. world. python is fun."
sentences = text.split('. ')
capitalized_sentences = [s.capitalize() for s in sentences]
result = '. '.join(capitalized_sentences)
print(result)  # Output: Hello. World. Python is fun.
```

---

### **3. Handling Leading Whitespace**

- `capitalize()` does not handle leading whitespace. Use `strip()` to remove leading whitespace before capitalizing.

#### **Example**

```python
text = "   hello world"
result = text.strip().capitalize()
print(result)  # Output: Hello world
```

---

### **4. Custom Capitalization Function**

- Create a custom function to capitalize specific parts of a string.

#### **Example**

```python
def custom_capitalize(text, indices):
    text_list = list(text)
    for i in indices:
        if i < len(text_list):
            text_list[i] = text_list[i].upper()
    return ''.join(text_list)

text = "hello world"
result = custom_capitalize(text, [0, 6])
print(result)  # Output: Hello World
```

---

### **5. Combining with Other String Methods**

- Combine `capitalize()` with other string methods like `lower()`, `upper()`, and `strip()` for more advanced formatting.

#### **Example**

```python
text = "  HELLO WORLD  "
result = text.strip().lower().capitalize()

```

### Python `capitalize()` Method

The `capitalize()` method is a built-in string method in Python that returns a copy of the string with the **first character capitalized** and all other characters converted to **lowercase**.

### Syntax:

```python
str.capitalize()
```

### Parameters:

- The `capitalize()` method does not take any parameters.

### Return Value:

- **`str.capitalize()`** returns a new string with the first character capitalized and the rest of the characters in lowercase.

### Key Points:

- It only affects the **first character** of the string (making it uppercase) and **changes all other characters to lowercase**.
- If the first character is already in uppercase, it will leave it unchanged, but the remaining characters will still be converted to lowercase.
- If the string is empty, it returns an empty string.
- This method does not modify the original string because strings in Python are **immutable**. It returns a new string.

### Examples:

#### 1. Basic Example:

```python
text = "hello"
capitalized_text = text.capitalize()
print(capitalized_text)  # Output: "Hello"
```

- The first character 'h' is capitalized, and the rest of the string is converted to lowercase.

#### 2. String with Uppercase Letters:

```python
text = "HELLO"
capitalized_text = text.capitalize()
print(capitalized_text)  # Output: "Hello"
```

- The string is converted to lowercase except for the first character, which becomes uppercase.

#### 3. String with Mixed Case:

```python
text = "hElLo WoRLd"
capitalized_text = text.capitalize()
print(capitalized_text)  # Output: "Hello world"
```

- The first letter 'h' is capitalized, and the rest of the string is converted to lowercase.

#### 4. Empty String:

```python
text = ""
capitalized_text = text.capitalize()
print(capitalized_text)  # Output: ""
```

- Since the string is empty, the result is also an empty string.

#### 5. String with Numbers and Special Characters:

```python
text = "123abc"
capitalized_text = text.capitalize()
print(capitalized_text)  # Output: "123abc"
```

- Since the first character is a number, it remains unchanged, and the rest of the string remains as it is.

#### 6. String with Leading Spaces:

```python
text = "  hello world"
capitalized_text = text.capitalize()
print(capitalized_text)  # Output: "  hello world"
```

- The leading spaces remain, and the first non-space character is capitalized. The rest of the string is converted to lowercase.

#### 7. String with Non-Alphabet Characters:

```python
text = "1234. hello"
capitalized_text = text.capitalize()
print(capitalized_text)  # Output: "1234. hello"
```

- Non-alphabet characters like numbers and punctuation remain unchanged. Only the first alphabetic character 'h' is capitalized.

### Special Case Considerations:

1. **Empty String**: If the string is empty, `capitalize()` will return an empty string.
2. **Leading/Trailing Spaces**: The method does not remove leading or trailing spaces. It only affects the first alphabetic character.
3. **Non-Alphabetic Strings**: If the first character is a non-alphabetic character (like a number or punctuation), it remains unchanged, and the string is not affected.
4. **Mixed Case**: The method ensures that only the first character is capitalized, and all other characters are converted to lowercase.

### Performance:

- **Time Complexity**: O(n), where `n` is the length of the string, as it needs to traverse through the string to convert the characters.
- **Space Complexity**: O(n), as a new string is created with the transformed characters.

### Common Use Cases:

1. **Title Case Conversion**:
   The `capitalize()` method can be useful when you want to ensure the first letter of a string (such as a name or a title) is capitalized, regardless of the input case.

   Example:

   ```python
   name = "alice"
   print(name.capitalize())  # Output: "Alice"
   ```

2. **Normalizing Input**:
   You can use `capitalize()` to normalize user input where you want to ensure that the first letter is capitalized and the rest of the string is lowercase (e.g., formatting names).

   Example:

   ```python
   input_name = input("Enter your name: ").strip()
   formatted_name = input_name.capitalize()
   print(f"Hello, {formatted_name}!")  # Output: "Hello, Alice!"
   ```

3. **String Formatting**:
   You can use `capitalize()` in formatting text, like creating properly formatted sentences.

   Example:

   ```python
   sentence = "this is a test sentence."
   print(sentence.capitalize())  # Output: "This is a test sentence."
   ```

4. **Capitalizing Words**:
   While `capitalize()` only affects the first letter of the string, it can be useful in word capitalizations. However, for capitalizing the first letter of every word in a string, the `title()` method would be more appropriate.

   Example:

   ```python
   sentence = "hello world, welcome to python."
   print(sentence.capitalize())  # Output: "Hello world, welcome to python."
   ```

### Alternative Methods for Capitalization:

- **`title()`**: Capitalizes the first letter of each word in the string.

  ```python
  sentence = "hello world"
  print(sentence.title())  # Output: "Hello World"
  ```

- **`upper()`**: Converts all characters in the string to uppercase.

  ```python
  sentence = "hello"
  print(sentence.upper())  # Output: "HELLO"
  ```

- **`lower()`**: Converts all characters in the string to lowercase.
  ```python
  sentence = "HELLO"
  print(sentence.lower())  # Output: "hello"
  ```

### Summary:

The `capitalize()` method is a simple way to ensure that the first letter of a string is capitalized and all other letters are converted to lowercase. It is widely used for formatting text, normalizing input, and working with names or titles. However, it may not be suitable for all capitalizing use cases (e.g., when you need to capitalize the first letter of each word, use `title()` instead).
