# Python String `rpartition()`

The string **`rpartition()`** method splits the string at the last occurrence of the argument string and returns a tuple containing the part the before separator, argument string and the part after the separator.

**Syntax**:

```python
string.rpartition(separator)
```


## `rpartition()` Parameters

The **`rpartition()`** method takes a string parameter **`separator`** that separates the string at the last occurrence of it.


## Return Value from `rpartition()`

The **`rpartition()`** method returns a 3-tuple containing:

- the part before the separator, separator parameter, and the part after the separator if the separator parameter is found in the string
- two empty strings, followed by the string itself if the separator parameter is not found


````python
# Example 1: How rpartition() works?

string = "Python is fun"

# 'is' separator is found
print(string.rpartition('is '))

# 'not' separator is not found
print(string.rpartition('not '))

string = "Python is fun, isn't it"

# splits at last occurence of 'is'
print(string.rpartition('is'))
````

The `rpartition()` method in Python is a string function used to split a string into three parts: the substring before the specified separator, the separator itself, and the substring after the separator. Unlike `partition()`, which splits at the first occurrence of the separator, `rpartition()` splits the string at the **last occurrence** of the separator.

### Syntax

```python
string.rpartition(separator)
```

### Parameters

- **separator**: A string representing the delimiter to partition the input string. It is required.

### Return Value

- **Tuple**: The `rpartition()` method returns a tuple containing three elements:
  1. The part of the string before the last occurrence of the separator.
  2. The separator itself.
  3. The part of the string after the separator.

If the separator is not found, the method returns a tuple containing:

1. An empty string (`''`) for the part before the separator.
2. An empty string (`''`) for the separator.
3. The original string as the third element.

### Example 1: Separator Found

```python
text = "apple, banana, cherry, dragonfruit"
result = text.rpartition(", ")
print(result)
# Output: ('apple, banana, cherry', ', ', 'dragonfruit')
```

Here, the string is split at the **last occurrence** of `", "`.

### Example 2: Separator Not Found

```python
text = "apple banana cherry dragonfruit"
result = text.rpartition(", ")
print(result)
# Output: ('', '', 'apple banana cherry dragonfruit')
```

Since the separator `", "` is not found in the string, the first and second elements of the tuple are empty, while the third element contains the original string.

### Usage Notes

- **Efficient for Reverse Splitting**: Use `rpartition()` when you need to split a string based on the last occurrence of a separator.
- **Consistent Output**: Even if the separator is not found, it ensures that a tuple with three elements is returned.

# Python `rpartition()` String Method: Complete Guide

## What is the `rpartition()` method?

The `rpartition()` method splits a string at the _last occurrence_ of a specified separator and returns a 3-tuple containing:

1. The part before the separator
2. The separator itself
3. The part after the separator

It's the "right-side" version of `partition()` - searching from the end of the string instead of the beginning.

## Syntax

```python
str.rpartition(separator)
```

### Parameters:

- `separator`: The string to split on (required)

### Return Value:

- Always returns a 3-tuple with these elements:
  1. Text before the last separator
  2. The separator itself
  3. Text after the last separator

## Key Characteristics

1. **Last Match Only**: Splits only at the last occurrence
2. **Always Returns 3 Elements**: Even if separator isn't found
3. **Search Right-to-Left**: Scans the string from the end
4. **Non-Destructive**: Original string remains unchanged

## Basic Examples

### Simple rpartition

```python
s = "Python:Programming:Language"
result = s.rpartition(':')
print(result)  # ('Python:Programming', ':', 'Language')
```

### Separator Not Found

```python
s = "Hello World"
result = s.rpartition(',')
print(result)  # ('', '', 'Hello World')  # Note different order vs partition()
```

### Multiple Occurrences

```python
s = "apple,banana,cherry"
result = s.rpartition(',')
print(result)  # ('apple,banana', ',', 'cherry')
```

## Practical Applications

### 1. File Extension Extraction (More Robust)

```python
filename = "archive.tar.gz"
name, sep, ext = filename.rpartition('.')
print(f"Name: {name}, Extension: {ext}")  # Name: archive.tar, Extension: gz
```

### 2. Getting Domain from Email

```python
email = "user.name@subdomain.example.com"
local, sep, domain = email.rpartition('@')
print(f"Domain: {domain}")  # Domain: subdomain.example.com
```

### 3. Splitting Rightmost Path Component

```python
path = "/home/user/documents/file.txt"
parent, sep, filename = path.rpartition('/')
print(f"Parent: {parent}, File: {filename}")  # Parent: /home/user/documents, File: file.txt
```

## Comparison with `partition()`

| Method         | Search Direction | Splits At        | Example Result                            |
| -------------- | ---------------- | ---------------- | ----------------------------------------- |
| `partition()`  | Left-to-right    | First occurrence | `('Python', ':', 'Programming:Language')` |
| `rpartition()` | Right-to-left    | Last occurrence  | `('Python:Programming', ':', 'Language')` |

## Advanced Usage

### 1. Handling Multiple Extensions

```python
filename = "project.backup.tar.gz"
# Get the last extension
name, sep, last_ext = filename.rpartition('.')
# Get the "main" extension
base, sep, ext = name.rpartition('.')
print(f"Base: {base}, Extensions: {ext}.{last_ext}")
```

### 2. Parsing Version Numbers

```python
version = "2.7.18"
main, sep, patch = version.rpartition('.')
print(f"Main: {main}, Patch: {patch}")  # Main: 2.7, Patch: 18
```

### 3. Safe Division of Strings

```python
s = "left=right=value"
# Get the rightmost assignment
left, sep, right = s.rpartition('=')
print(f"Left: {left}, Right: {right}")  # Left: left=right, Right: value
```

## Common Errors

1. **Empty Separator**:

   ```python
   "text".rpartition('')  # ValueError: empty separator
   ```

2. **Non-String Separator**:
   ```python
   "text".rpartition(123)  # TypeError: must be str, not int
   ```

## Performance Notes

- Slightly more efficient than using `split()` and joining when you only need the last split
- Better choice than `rsplit(maxsplit=1)` when you need the separator itself
- Time complexity is O(n) where n is the string length

## Real-World Examples

### 1. Parsing Log Timestamps

```python

print(f"Timestamp: {date_time}, Message: {message}")
```

### 2. URL Processing


### 3. Config File Parsing

```python

print(f"Key: {key.strip()}, Value: {final_value.strip()}")
```

The `rpartition()` method is particularly useful when:

- You need to split at the last occurrence of a separator
- You're working with hierarchical data (paths, domains, etc.)
- You need to preserve the separator in the output
- You want guaranteed tuple output with 3 elements

the `rpartition()` method in Python strings. It's very similar to `partition()`, but with a crucial difference: it splits the string based on the _last_ occurrence of the specified separator.

**What `rpartition()` Does:**

The `rpartition()` method searches for the last occurrence of a specified separator within a string. If the separator is found, it splits the string into a tuple containing three elements:

1.  The part of the string _before_ the last occurrence of the separator.
2.  The separator itself.
3.  The part of the string _after_ the last occurrence of the separator.

If the separator is not found in the string, `rpartition()` returns a tuple containing:

1.  An empty string.
2.  Another empty string.
3.  The original string.

**Syntax:**

```python
string.rpartition(sep)
```

**Parameters:**

- `sep`: The separator string to search for. This is a mandatory argument.

**Return Value:**

- A tuple of three strings: `(head, sep, tail)`.
  - `head`: The part of the string before the last occurrence of `sep`.
  - `sep`: The separator string itself (if found).
  - `tail`: The part of the string after the last occurrence of `sep`.

**Examples:**

Let's see `rpartition()` in action:

```python
text = "Hello,World!"

# rpartitioning by a comma
result1 = text.rpartition(",")
print(f"rpartitioning by ',': {result1}")  # Output: rpartitioning by ',': ('Hello', ',', 'World!') (same as partition as there's only one)

# rpartitioning by a space
result2 = text.rpartition(" ")
print(f"rpartitioning by ' ': {result2}")  # Output: rpartitioning by ' ': ('', '', 'Hello,World!') (space not found)

text2 = "apple#banana#cherry"
# rpartitioning by '#'
result3 = text2.rpartition("#")
print(f"rpartitioning by '#': {result3}")  # Output: rpartitioning by '#': ('apple#banana', '#', 'cherry') (splits at the last '#')

# rpartitioning by a substring
text3 = "www.example.com"
result4 = text3.rpartition(".")
print(f"rpartitioning by '.': {result4}")  # Output: rpartitioning by '.': ('www.example', '.', 'com') (splits at the last '.')

# Separator at the beginning
text4 = "/home/user/document.txt"
result5 = text4.rpartition("/")
print(f"rpartitioning by '/': {result5}")  # Output: rpartitioning by '/': ('', '/', 'home/user/document.txt') (last '/' is at the beginning if string starts with it)

# Separator at the end
text5 = "file.txt/"
result6 = text5.rpartition("/")
print(f"rpartitioning by '/': {result6}")  # Output: rpartitioning by '/': ('file.txt', '/', '') (last '/' is at the end)

# Separator not found
text6 = "no-separator-here"
result7 = text6.rpartition("-")
print(f"rpartitioning by '-': {result7}")  # Output: rpartitioning by '-': ('no', '-', 'separator-here') (same as partition as there's only one)

result8 = text6.rpartition("@")
print(f"rpartitioning by '@': {result8}")  # Output: rpartitioning by '@': ('', '', 'no-separator-here')
```

**Key Characteristics:**

- **Splits at the Last Occurrence:** The crucial difference from `partition()` is that `rpartition()` splits the string at the _last_ instance of the specified separator.
- **Returns a Tuple:** The result is always a tuple containing exactly three strings.
- **Includes the Separator:** Like `partition()`, `rpartition()` includes the separator as the second element of the resulting tuple.
- **Handles Substrings as Separators:** The `sep` argument can be a substring of any length.
- **Behavior When Separator Not Found:** If the separator is not found, `rpartition()` returns `('', '', original_string)`, whereas `partition()` returns `(original_string, '', '')`. This is a key distinction.

**Comparison with `partition()`:**

The primary difference lies in where the split occurs when the separator appears multiple times:

- **`partition(sep)`:** Splits at the _first_ occurrence of `sep`.
- **`rpartition(sep)`:** Splits at the _last_ occurrence of `sep`.

**Use Cases:**

`rpartition()` is particularly useful when you need to:

- Separate a string based on the last occurrence of a delimiter, such as extracting the filename from a path or the extension from a filename.
- Process strings where the relevant separator is the last one.

For example, you might use `rpartition(".")` to reliably get the filename and extension from a path, as the last dot usually separates them.

```python
file_path = "/home/user/documents/report.pdf"
base, sep, extension = file_path.rpartition(".")
print(f"Base: {base}, Separator: {sep}, Extension: {extension}")
# Output: Base: /home/user/documents/report, Separator: ., Extension: pdf
```

In summary, `rpartition()` provides a convenient and predictable way to split a string into three parts based on the last instance of a separator, making it a valuable tool for specific string manipulation tasks, especially when dealing with hierarchical or delimited data where the last separator holds significance.
