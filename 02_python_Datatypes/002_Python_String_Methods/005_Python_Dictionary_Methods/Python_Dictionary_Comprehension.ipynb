## What is Dictionary Comprehension in Python?

Dictionary comprehension is an elegant and concise way to create dictionaries.


````python
# Example 1: Dictionary Comprehension

square_dict = dict()
for num in range(1, 11):
    square_dict[num] = num*num
print(square_dict)
````

Now, let's create the dictionary in the above program using dictionary comprehension.


````python
# dictionary comprehension example
square_dict = {num: num*num for num in range(1, 11)}
print(square_dict)
````

In both programs, we have created a dictionary **`square_dict`** with **number-square key/value pair**.

However, using dictionary comprehension allowed us to **create a dictionary in a single line**.


````python
list_x=[1,2,3,4,5]
result=list_x.remove(3)
print(result)
````

## Conditionals in Dictionary Comprehension

We can further customize dictionary comprehension by adding conditions to it. Let's look at an example.


````python
# Example 4: If Conditional Dictionary Comprehension
    
original_dict = {'Allan': 36, 'Bill': 48, 'Cory': 57, 'Dave': 33}

even_dict = {k: v for (k, v) in original_dict.items() if v % 2 == 0}
print(even_dict)
````

**Explanation:**

As we can see, only the items with even value have been added, because of the **`if`** clause in the dictionary comprehension.


````python
# Example 5: Multiple if Conditional Dictionary Comprehension

original_dict = {'Allan': 36, 'Bill': 48, 'Cory': 57, 'Dave': 33}

new_dict = {k: v for (k, v) in original_dict.items() if v % 2 != 0 if v < 40}
print(new_dict)
````

**Explanation:**

In this case, only the items with an odd value of less than 40 have been added to the new dictionary.

It is because of the multiple **`if`** clauses in the dictionary comprehension. They are equivalent to **`and`** operation where both conditions have to be true.


````python
# Example 6: if-else Conditional Dictionary Comprehension

original_dict = {'Allan': 36, 'Bill': 48, 'Cory': 57, 'Dave': 33}

new_dict_1 = {k: ('old' if v > 40 else 'young')
    for (k, v) in original_dict.items()}
print(new_dict_1)
````

**Explanation:**

In this case, a new dictionary is created via dictionary comprehension.
The items with a value of 40 or more have the value of 'old' while others have the value of 'young'.


## Nested Dictionary Comprehension

Like how a Dictionary is a collection of **key-value** pairs, Python nested dictionaries are an unordered collection of one or two dictionaries. These can be represented as shown below.

**Syntax:**

```python
nested_dict = { 'dict1': {'key1': 'value1'}, 'dict2': {'key2': 'value2'}}
```

Here, nested_dict is a nested dictionary which has two dictionaries with keys **`dict1`** and **`dict2`**. The values of those two keys (**`dict1`**, **`dict2`**) are in-turn dictionaries.

In Python, Nested dictionaries can be created by placing the comma-separated dictionaries enclosed within curly brackets. For example:


````python
# Example 7: creating a nested dictionary

nes_dict = {'dict1': {'Color': 'Red', 'Shape': 'Square'}, 'dict2': {'Color': 'Pink', 'Shape': 'Round'}}
print (nes_dict)
````

````python
# Example 8: Nested Dictionary with Two Dictionary Comprehensions

dictionary = {k1: {k2: k1 * k2 for k2 in range(1, 6)} for k1 in range(2, 5)}
print(dictionary)
````

**Explanation:**

As you can see, we have constructed a multiplication table in a nested dictionary, for numbers from 2 to 4.

Whenever nested dictionary comprehension is used, Python first starts from the outer loop and then goes to the inner one.

So, the above code would be equivalent to:


````python
# Example 8:

dictionary = dict()
for k1 in range(11, 16):
    dictionary[k1] = {k2: k1*k2 for k2 in range(1, 6)}
print(dictionary)
````

It can further be unfolded:


````python
# Example 8:

dictionary = dict()
for k1 in range(11, 16):
    dictionary[k1] = dict()
    for k2 in range(1, 6):
        dictionary[k1][k2] = k1*k2
print(dictionary)
````

All these three programs give us the same output.

### How to Delete Elements from a Nested Dictionary

The key-value pairs in Python nested dictionaries can be deleted using the **`del()`** method. This method can be used to delete either the entire dictionary or a particular key-value pair from a nested dictionary.


````python
# Example 11: deleting the entire nested dictionary

nes_dict = {'dict1': {'Color': 'Red', 'Shape': 'Square'}, 'dict2': {'Color': 'Pink', 'Shape': 'Round'}}
del nes_dict['dict1']
print(nes_dict)

# deleting the a key-value from nested dictionary
nes_dict = {'dict1': {'Color': 'Red', 'Shape': 'Square'}, 'dict2': {'Color': 'Pink', 'Shape': 'Round'}}
del nes_dict['dict1']
del nes_dict['dict2']['Shape']
print (nes_dict)
````

**Dictionary comprehension** in Python is a concise and efficient way to create or modify dictionaries using a single line of code. It allows you to build dictionaries by transforming or filtering existing data, such as lists, other dictionaries, or ranges, and applying conditions or operations to the keys and values.

### 1. Definition and Syntax

Dictionary comprehension is a compact way of creating dictionaries by performing operations on the keys and values of an existing iterable or by applying conditions to them. It is similar to list comprehension, but instead of creating lists, dictionary comprehension creates dictionaries.

#### **Syntax**:

```python
{key_expression: value_expression for item in iterable if condition}
```

- **`key_expression`**: Expression that defines the key for the dictionary.
- **`value_expression`**: Expression that defines the value corresponding to the key.
- **`iterable`**: An iterable object (such as a list, tuple, range, or even another dictionary) to iterate over.
- **`condition` (optional)**: A condition that filters which items are included in the dictionary.

---

### 2. Basic Structure of Dictionary Comprehension

The basic structure of dictionary comprehension involves iterating over an iterable and applying an expression to generate the key-value pairs in the dictionary.

#### Example 1: Basic Dictionary Comprehension

```python
# Creating a dictionary with numbers and their squares
squares = {x: x**2 for x in range(5)}

print(squares)
```

- **Output**: `{0: 0, 1: 1, 2: 4, 3: 9, 4: 16}`

- **Explanation**: In this example, the dictionary comprehension creates a dictionary where the keys are the integers from `0` to `4`, and the values are their respective squares.

---

### 3. Examples of Dictionary Comprehension

#### Example 1: Creating a Dictionary from Two Lists

```python
# Two lists: one for keys and one for values
keys = ['a', 'b', 'c']
values = [1, 2, 3]

# Creating a dictionary using dictionary comprehension
my_dict = {keys[i]: values[i] for i in range(len(keys))}

print(my_dict)
```

- **Output**: `{'a': 1, 'b': 2, 'c': 3}`

- **Explanation**: This example uses two lists, `keys` and `values`, to create a dictionary where each key corresponds to a value from the second list.

#### Example 2: Filtering Items Using Conditions

```python
# Dictionary comprehension with a condition
numbers = [1, 2, 3, 4, 5, 6]

# Create a dictionary of even numbers
even_numbers = {x: x**2 for x in numbers if x % 2 == 0}

print(even_numbers)
```

- **Output**: `{2: 4, 4: 16, 6: 36}`

- **Explanation**: The comprehension only includes items where `x` is even, filtering out odd numbers from the dictionary.

#### Example 3: Swapping Keys and Values

```python
# Original dictionary
original_dict = {'a': 1, 'b': 2, 'c': 3}

# Swap the keys and values using dictionary comprehension
swapped_dict = {v: k for k, v in original_dict.items()}

print(swapped_dict)
```

- **Output**: `{1: 'a', 2: 'b', 3: 'c'}`

- **Explanation**: In this example, the dictionary comprehension swaps the keys and values of the original dictionary.

---

### 4. Using Conditions with Dictionary Comprehension

You can use an `if` condition in dictionary comprehension to **filter the items** that are included in the resulting dictionary. This allows you to control which elements from the iterable are added based on certain criteria.

#### Example: Filtering Keys Based on Condition

```python
# Creating a dictionary from a list of numbers with a condition
numbers = [1, 2, 3, 4, 5, 6]

# Only include even numbers in the dictionary
even_numbers = {x: x**2 for x in numbers if x % 2 == 0}

print(even_numbers)
```

- **Output**: `{2: 4, 4: 16, 6: 36}`

- **Explanation**: The comprehension adds only the even numbers from the list to the dictionary, along with their squares.

---

### 5. Working with Nested Dictionary Comprehension

Just like list comprehensions, dictionary comprehensions can be **nested**. This means that you can generate dictionaries inside dictionaries.

#### Example: Nested Dictionary Comprehension

```python
# Creating a nested dictionary where each key is a number and the value is a dictionary of squares and cubes
nested_dict = {x: {'square': x**2, 'cube': x**3} for x in range(1, 4)}

print(nested_dict)
```

- **Output**: `{1: {'square': 1, 'cube': 1}, 2: {'square': 4, 'cube': 8}, 3: {'square': 9, 'cube': 27}}`

- **Explanation**: In this example, each key (1, 2, 3) in the dictionary points to a dictionary containing the square and cube of that key.

### 6. Advantages of Using Dictionary Comprehension

1. **Conciseness**: Dictionary comprehension allows you to create dictionaries in a single line of code, making your code more compact and readable.
2. **Efficiency**: Comprehensions are often more efficient than using loops to build dictionaries because they are optimized for performance in Python.
3. **Readable**: Comprehensions can make the code more expressive and easier to understand at a glance.
4. **Flexibility**: You can combine transformations, filtering, and nested operations within a single comprehension.

---

### 7. Pitfalls and Best Practices

#### Pitfalls:

- **Readability**: Although comprehensions are concise, overly complex dictionary comprehensions can reduce code readability. It's important to balance readability with conciseness.
- **Performance**: While comprehensions are usually efficient, they can sometimes be slower than traditional loops if used in a very large dataset with complex operations.

#### Best Practices:

- **Keep it Simple**: Use dictionary comprehension when it improves clarity. If the comprehension becomes too complex or nested, consider using a regular for loop.
- **Limit Conditions**: Use conditions in comprehensions to filter the data in a meaningful way but avoid overly complicated conditions.
- **Documentation**: If you are working with nested or complex comprehensions, add comments to explain the logic behind them.

---

### 8. Time Complexity

The time complexity of dictionary comprehension is primarily determined by the number of iterations over the input data. Let's break it down:

- **Iterating over an iterable**: The time complexity of iterating over an iterable is **O(n)**, where `n` is the number of elements in the iterable.
- **Creating key-value pairs**: The time complexity for creating key-value pairs and inserting them into the dictionary is **O(1)** for each insertion, leading to an overall time complexity of **O(n)**.

Thus, the time complexity of a dictionary comprehension is **O(n)**, where `n` is the number of elements in the iterable.

```python
# Combine two lists into a dictionary
person = {keys[i]: values[i] for i in range(len(keys))}

print(person)
```

- **Output**: `{'name': 'Alice', 'age': 30, 'city': 'New York'}`

- **Explanation**: The comprehension combines the two lists (`keys` and `values`) into a dictionary.

### Conclusion

Dictionary comprehension in Python provides a powerful, concise, and efficient way to create and manipulate dictionaries. By using a single line of code, you can transform, filter, and modify data based on conditions, making it a versatile tool in any Python programmer's toolbox.

#### Key Takeaways:

- Dictionary comprehension allows for the creation or modification of dictionaries in a single line of code.
- It supports conditions, transformations, and even nested comprehensions.
- It offers efficiency, but should be used judiciously to avoid sacrificing readability.
- Time complexity for dictionary comprehension is generally **O(n)**, where `n` is the number of elements being processed.

By mastering dictionary comprehension, you can write more Pythonic and optimized code for a variety of use cases.
