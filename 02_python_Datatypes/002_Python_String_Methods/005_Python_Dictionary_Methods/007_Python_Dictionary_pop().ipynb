# Python Dictionary `pop()`

The **`pop()`** method removes and returns an element from a dictionary having the given key.

**Syntax**:

```python
dictionary.pop(key[, default])
```


## `pop()` Parameters

The **`pop()`** method takes two parameters:

- **key** - key which is to be searched for removal
- **default** - value which is to be returned when the key is not in the dictionary


## Return Value from `pop()`

The **`pop()`** method returns:

- If **`key`** is found - removed/popped element from the dictionary
- If **`key`** is not found - value specified as the second argument (default)
- If **`key`** is not found and default argument is not specified - **`KeyError`** exception is raised


````python
# Example 1: Pop an element from the dictionary

# random sales dictionary
sales = { 'apple': 2, 'orange': 3, 'grapes': 4 }

element = sales.pop('apple')
print('The popped element is:', element)
print('The dictionary is:', sales)
````

````python
# Example 2: Pop an element not present from the dictionary

# random sales dictionary
sales = { 'apple': 2, 'orange': 3, 'grapes': 4 }

element = sales.pop('guava')
````

````python
# Example 3: Pop an element not present from the dictionary, provided a default value

# random sales dictionary
sales = { 'apple': 2, 'orange': 3, 'grapes': 4 }

element = sales.pop('guava', 'banana')
print('The popped element is:', element)
print('The dictionary is:', sales)
````

### Python Dictionary `pop()` Method: Concepts and Theory

The **`pop()`** method is a built-in function in Python used to remove a specified key-value pair from a dictionary. This method not only removes the item but also **returns the value** associated with the key that was removed. It is a powerful and commonly used function for managing and modifying dictionaries.

---

### 1. Definition and Syntax

The **`pop()`** method is used to remove a specified key-value pair from a dictionary by using the given key.

#### **Syntax**:

```python
dict.pop(key, default)
```

- **`key`**: The key that you want to remove from the dictionary.
- **`default`** (optional): A value to return if the key is not found in the dictionary. If this parameter is not provided and the key is missing, a `KeyError` will be raised.

#### **Return Value**:

- The method returns the **value** associated with the key that was removed.
- If the key is not found and no default value is provided, a **`KeyError`** is raised.

---

### 2. How `pop()` Works

The `pop()` method removes the key-value pair from the dictionary, based on the provided key, and returns the corresponding value.

#### Key Points:

- The **key-value pair** is completely removed from the dictionary.
- If you provide a **non-existent key** and do not specify a default, a `KeyError` is raised.
- If you provide a **default value** and the key does not exist, the default value will be returned instead of raising an error.

---

### 3. Examples of Using `pop()`

#### Example 1: Removing an Existing Key

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Remove 'age' key and return the associated value
removed_value = person.pop('age')

print(removed_value)  # Output: 30
print(person)         # Output: {'name': 'Alice', 'city': 'New York'}
```

- **Explanation**: The `pop()` method removes the `'age'` key from the dictionary and returns its value `30`.

#### Example 2: Handling Non-Existent Key with Default Value

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30}

# Attempt to remove a key that doesn't exist and provide a default value
removed_value = person.pop('city', 'Not Available')

print(removed_value)  # Output: Not Available
print(person)         # Output: {'name': 'Alice', 'age': 30}
```

- **Explanation**: Since `'city'` is not found in the dictionary, the default value `'Not Available'` is returned instead of raising an error.

#### Example 3: Attempting to Remove a Non-Existent Key Without Default

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30}

# Attempt to remove a key that doesn't exist without a default value
removed_value = person.pop('city')  # This will raise KeyError
```

- **Explanation**: Since `'city'` is not a valid key and no default value is provided, Python raises a `KeyError`.

---

### 4. Handling Missing Keys

One of the challenges of using `pop()` is handling the case where the key doesn't exist in the dictionary. You can handle this in two ways:

#### **With the Default Value**:

As shown in the previous example, providing a `default` value to `pop()` ensures that the method will not raise an error when the key is missing. Instead, it will return the default value you specify.

#### **Without the Default Value**:

If you don't provide a default value and the key is missing, Python will raise a `KeyError`. You can handle this exception using a `try-except` block if you want to manage the error explicitly.

Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30}

# Handling KeyError using try-except
try:
    removed_value = person.pop('city')
except KeyError:
    print("Key not found.")
```

**Output**:

```
Key not found.
```

- **Explanation**: By using `try-except`, you can manage the situation where the key might not exist without letting the program crash.

---

### 5. Return Value of `pop()`

The **`pop()`** method returns the value associated with the key that was removed from the dictionary. This allows you to capture and use the value in your program, if needed.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Capture the removed value
removed_value = person.pop('name')

print(removed_value)  # Output: Alice
```

- **Explanation**: The method removes the `'name'` key and returns its value `'Alice'`.

---

### 6. Advanced Usage of `pop()`

The `pop()` method can also be used in more advanced scenarios, such as:

#### **Removing Multiple Keys Using `pop()` in a Loop**

You may want to remove multiple keys from a dictionary dynamically. You can use the `pop()` method inside a loop to do this.

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# List of keys to remove
keys_to_remove = ['name', 'age']

# Removing keys dynamically
removed_items = {}
for key in keys_to_remove:
    removed_items[key] = person.pop(key, 'Key Not Found')

print(removed_items)  # Output: {'name': 'Alice', 'age': 30}
print(person)         # Output: {'city': 'New York'}
```

- **Explanation**: The loop removes multiple keys (`'name'` and `'age'`) from the dictionary, and the removed items are stored in the `removed_items` dictionary.

#### **Using `pop()` to Rearrange or Modify Data**

You can use `pop()` when you need to rearrange or modify a dictionary, such as when you want to extract and process the key-value pairs:

```python
# Example dictionary
student_grades = {'Alice': 85, 'Bob': 90, 'Charlie': 88}

# Extract the highest score
top_student = max(student_grades, key=student_grades.get)
top_score = student_grades.pop(top_student)

print(f"Top student: {top_student}, Score: {top_score}")  # Output: Top student: Bob, Score: 90
```

- **Explanation**: The `pop()` method is used to remove and store the value of the student with the highest score.

---

### 7. Comparison of `pop()` with Other Methods

Let's compare the `pop()` method with other methods that are used for removing items from a dictionary.

#### **`pop()` vs `del`**

- **`pop()`**: Removes a key-value pair and returns the value. If the key doesn't exist, it raises a `KeyError` unless a default value is provided.
- **`del`**: Removes a key-value pair from a dictionary but does not return the value. It raises a `KeyError` if the key is not found.

Example:

```python
# Using pop()
person = {'name': 'Alice', 'age': 30}
removed_value = person.pop('age')  # Output: 30
print(removed_value)

# Using del
person = {'name': 'Alice', 'age': 30}
del person['age']  # No return value, just removes the key-value pair
```

#### **`pop()` vs `popitem()`**

- **`pop()`**: Removes a key-value pair by specifying the key.
- **`popitem()`**: Removes and returns a random key-value pair (useful in Python 3.7 and later where dictionaries are ordered).

Example:

```python
# Using pop()
person = {'name': 'Alice', 'age': 30}
removed_value = person.pop('age')  # Removes 'age' and returns its value

# Using popitem()
person = {'name': 'Alice', 'age': 30}
removed_item = person.popitem()  # Removes and returns an arbitrary key-value pair
print(removed_item)  # Output could be ('name', 'Alice') or ('age', 30)
```

---

### 8. Time Complexity

- **Time Complexity** of `pop()` is **O(1)** on average. This means that removing a key-value pair from a dictionary is done in constant time, regardless of the size of the dictionary.

---

### 9. Best Practices and Pitfalls

#### **Best Practices**:

- **Always handle potential `KeyError`**: If you're unsure whether the key exists in the dictionary, either use the `default` parameter or handle `KeyError` using a `try-except` block.
- **Use `pop()` when you need the value**: If you need both the key and the value, or you want to perform some operation with the removed value, `pop()` is a good choice.

#### **Pitfalls**:

- **Not handling missing keys**: If you don't provide a `default` value and the key doesn't exist, Python will raise a `KeyError`, which might be unexpected if you're not prepared for it.
- **Using `pop()` when you don't need the returned value**: If you only need to remove the key and don't care about the value, `del` might be a more suitable choice.

---

### Conclusion

The **`pop()`** method is an essential tool for removing key-value pairs from dictionaries. It offers flexibility by allowing you to retrieve the value of the removed key and handle missing keys gracefully with a default value. Understanding how to use it effectively is crucial for managing dictionaries in Python.

#### Key Takeaways:

- You can provide a `default` value to handle missing keys.
- It's an **O(1)** operation, meaning it’s very efficient.
