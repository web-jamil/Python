# Python Dictionary `setdefault()`

The **`setdefault()`** method returns the value of a key (if the key is in dictionary). If not, it inserts key with a value to the dictionary.

**Syntax**:

```python
dict.setdefault(key[, default_value])
```


## `setdefault()` Parameters

The **`setdefault()`** method takes a maximum of two parameters:

- **key** - the key to be searched in the dictionary
- **default_value** (optional) - **`key`** with a value **`default_value`** is inserted to the dictionary if the key is not in the dictionary.

If not provided, the **`default_value`** will be **`None`**.


## Return Value from `setdefault()`

The **`setdefault()`** method returns:

- value of the **`key`** if it is in the dictionary
- None if the key is not in the dictionary and **`default_value`** is not specified
- **`default_value`** if **`key`** is not in the dictionary and **`default_value`** is specified


````python
# Example 1: How setdefault() works when key is in the dictionary?

person = {'name': 'Phill', 'age': 22}

age = person.setdefault('age')
print('person = ',person)
print('Age = ',age)
````

````python
# Example 2: How setdefault() works when key is not in the dictionary?

person = {'name': 'Phill'}

# key is not in the dictionary
salary = person.setdefault('salary')
print('person = ',person)
print('salary = ',salary)

# key is not in the dictionary
# default_value is provided
age = person.setdefault('age', 22)
print('person = ',person)
print('age = ',age)
````

### Python Dictionary `setdefault()` Method: Concepts and Theory

The **`setdefault()`** method in Python is a built-in function used with dictionaries. It provides a convenient way to **get the value of a specified key** if it exists, and if it doesn't exist, it **inserts the key with a default value**.

The `setdefault()` method is particularly useful when you want to ensure a key is present in the dictionary and have it initialized to a specific value if it's missing, all in one operation.

### 1. Definition and Syntax

The **`setdefault()`** method is used to retrieve the value of a key from a dictionary. If the key does not exist, the method inserts the key with the specified default value.

#### **Syntax**:

```python
dict.setdefault(key, default=None)
```

- **`key`**: The key for which you want to get the value.
- **`default`** (optional): The value to insert if the key doesn't exist in the dictionary. If not provided, the default value will be `None`.

#### **Return Value**:

- The method returns the **value** of the key if the key exists in the dictionary.
- If the key doesn't exist, it **inserts** the key with the default value and then returns the default value.

---

### 2. How `setdefault()` Works

- If the specified **key** exists in the dictionary, `setdefault()` returns the **current value** associated with that key.
- If the key does not exist in the dictionary, `setdefault()` inserts the key with the **default value** and returns that value.
- If the default value is not provided, it will insert the key with a value of **`None`**.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30}

# Using setdefault() to get the value of 'age'
age_value = person.setdefault('age', 25)

print(age_value)   # Output: 30
print(person)      # Output: {'name': 'Alice', 'age': 30}
```

In this example, the key `'age'` already exists, so `setdefault()` returns its current value (`30`) without changing it.

#### Example 2: Adding a Key if It Doesn't Exist

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30}

# Using setdefault() for a non-existent key
city_value = person.setdefault('city', 'New York')

print(city_value)   # Output: New York
print(person)       # Output: {'name': 'Alice', 'age': 30, 'city': 'New York'}
```

In this case, `'city'` is not found in the dictionary, so the key `'city'` is inserted with the default value `'New York'`, and that value is returned.

---

### 3. Examples of Using `setdefault()`

#### Example 1: Using `setdefault()` to Set Default Values for Missing Keys

```python
# Example dictionary
inventory = {'apples': 10, 'bananas': 5}

# Set default for a non-existing key
inventory.setdefault('oranges', 8)

print(inventory)  # Output: {'apples': 10, 'bananas': 5, 'oranges': 8}
```

- **Explanation**: The `setdefault()` method ensures that if the key `'oranges'` is not present in the dictionary, it will be added with the default value `8`.

#### Example 2: Using `setdefault()` Without a Default Value

```python
# Example dictionary
inventory = {'apples': 10, 'bananas': 5}

# Using setdefault() without providing a default value
inventory.setdefault('bananas')

print(inventory)  # Output: {'apples': 10, 'bananas': 5}
```

- **Explanation**: In this case, `setdefault()` does not change the dictionary because `'bananas'` is already present, so it simply returns its existing value (`5`).

#### Example 3: Modifying a Key's Value with `setdefault()`

```python
# Example dictionary
sales = {'item1': 100, 'item2': 200}

# Using setdefault() to ensure a key exists, and then updating its value
sales.setdefault('item3', 0)
sales['item3'] += 50  # Update the value of item3

print(sales)  # Output: {'item1': 100, 'item2': 200, 'item3': 50}
```

- **Explanation**: The `setdefault()` method ensures the key `'item3'` exists in the dictionary with the value `0`. After that, you can modify the value of `'item3'`.

---

### 4. Return Value of `setdefault()`

The **return value** of `setdefault()` is the **value** associated with the given key:

- If the key exists, it returns the **current value** of that key.
- If the key doesn't exist, it **inserts the key with the default value** and then returns that value.

---

### 5. Default Value Parameter

The **`default`** parameter is optional. If you don't provide it, the dictionary key will be set to **`None`** by default.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30}

# Using setdefault() without a default value
result = person.setdefault('city')

print(result)     # Output: None
print(person)     # Output: {'name': 'Alice', 'age': 30, 'city': None}
```

- **Explanation**: Since `'city'` does not exist in the dictionary, it is added with the default value `None`.

---

### 6. Use Cases for `setdefault()`

#### 6.1 **Initializing Values for Missing Keys**

`setdefault()` is commonly used when you want to ensure that a dictionary has a key and initialize it with a default value if it doesn't exist. This is useful when you're building a dictionary incrementally.

Example: Counting occurrences of words in a list:

```python
words = ['apple', 'banana', 'apple', 'orange', 'banana', 'banana']

word_count = {}
for word in words:
    word_count.setdefault(word, 0)
    word_count[word] += 1

print(word_count)  # Output: {'apple': 2, 'banana': 3, 'orange': 1}
```

- **Explanation**: The `setdefault()` method ensures that each word has an initial count of `0` if it does not exist in the dictionary. After that, the count is incremented for each occurrence of the word.

#### 6.2 **Working with Nested Dictionaries**

`setdefault()` is also useful when working with nested dictionaries, as it can initialize inner dictionaries for missing keys.

Example: Creating a nested dictionary of students' scores:

```python
scores = {}

# Set scores for different subjects
scores.setdefault('Alice', {}).setdefault('math', 95)
scores.setdefault('Alice', {}).setdefault('science', 90)
scores.setdefault('Bob', {}).setdefault('math', 85)

print(scores)  # Output: {'Alice': {'math': 95, 'science': 90}, 'Bob': {'math': 85}}
```

- **Explanation**: The `setdefault()` method is used to ensure that the inner dictionary for each student is initialized before setting the scores for each subject.

---

### 7. Comparison with Other Methods (`get()` and `pop()`)

#### **`setdefault()` vs `get()`**:

- **`get()`**: Retrieves the value for the specified key if it exists, and returns `None` (or a specified default value) if the key does not exist. It does not modify the dictionary.
- **`setdefault()`**: Similar to `get()`, but if the key does not exist, it inserts the key with the specified default value and returns that value.

Example:

```python
# Example dictionary
person = {'name': 'Alice'}

# Using get()
name = person.get('name', 'Unknown')  # Output: 'Alice'

# Using setdefault()
name = person.setdefault('name', 'Unknown')  # Output: 'Alice'
```

#### **`setdefault()` vs `pop()`**:

- **`pop()`**: Removes a specified key-value pair and returns the value of the key.
- **`setdefault()`**: Retrieves the value if the key exists or inserts the key with a default value if it doesn't exist, but **does not remove the key**.

Example:

```python
# Using pop()
person = {'name': 'Alice'}
removed_value = person.pop('name')  # Removes 'name' and returns its value

# Using setdefault()
person = {'name': 'Alice'}
value = person.setdefault('name', 'Unknown')  # Does not remove 'name'
```

---

### 8. Time Complexity

- **Time Complexity**: The time complexity of the `setdefault()` method is **O(1)** on average. This means that the operation is executed in constant time, regardless of the size of the dictionary.

---

### 9. Best Practices and Pitfalls

#### Best Practices:

- **Use `setdefault()` for default initialization**: It's a useful tool when you need to ensure a key exists in the dictionary with a default value.
- **Use `setdefault()` when building nested dictionaries**: It helps initialize inner dictionaries without having to check if the key exists first.

#### Pitfalls:

- **Unnecessary Insertions**: If you use `setdefault()` on a key that already exists, it can inadvertently set the value to a default if you're not careful.

Example:

```python
person = {'name': 'Alice', 'age': 30}
person.setdefault('age', 25)  # This will not modify 'age' as it's already there.
```

- **Default value initialization**: If you don't want to modify the dictionary unless necessary, consider using `get()` instead of `setdefault()`.

#### Key Takeaways:
