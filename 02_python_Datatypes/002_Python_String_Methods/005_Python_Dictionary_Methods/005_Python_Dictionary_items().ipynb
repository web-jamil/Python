# Python Dictionary `items()`

The **`items()`** method returns a view object that displays a list of dictionary's (key, value) tuple pairs.

**Syntax**:

```python
dictionary.items()
```

> **Note:** **`items()`** method is similar to dictionary's **`viewitems()`** method in Python 2.7.


## `items()` Parameters

The **`items()`** method doesn't take any parameters.


## Return Value from `items()`

The **`items()`** method returns a view object that displays a list of a given dictionary's (key, value) tuple pair.


````python
# Example 1: Get all items of a dictionary with items()

# random sales dictionary
sales = { 'apple': 2, 'orange': 3, 'grapes': 4 }

print(sales.items())
````

````python
# Example 2: How items() works when a dictionary is modified?

# random sales dictionary
sales = { 'apple': 2, 'orange': 3, 'grapes': 4 }

items = sales.items()
print('Original items:', items)

# delete an item from dictionary
del[sales['apple']]
print('Updated items:', items)
````

**Note:** The view object **`items`** doesn't itself return a list of sales items but it returns a view of **`sales`**'s (key, value) pair.

If the list is updated at any time, the changes are reflected on the view object itself, as shown in the above program.


### **Python Dictionary `get()` Method: Concepts and Theory**

The **`get()`** method in Python is one of the most commonly used functions to access values from a dictionary. It is a safer way to retrieve values associated with a specific key without raising an error if the key does not exist. If the key is not found, it returns a default value (which can be specified by the user), or `None` if no default is provided.

Here, we will cover the following aspects of the `get()` method:

1. **Definition and Syntax**
2. **How `get()` Works**
3. **Examples of `get()` in Action**
4. **Use Cases for `get()`**
5. **Differences between `get()` and `[]` (Indexing)**
6. **Time Complexity**
7. **Common Pitfalls and Best Practices**

---

### **1. Definition and Syntax**

The **`get()`** method allows you to retrieve the value for a given key in a dictionary. If the key is not found, it returns the specified default value (which is `None` by default).

#### **Syntax**:

```python
dict.get(key, default=None)
```

- **`key`**: The key whose value you want to retrieve.
- **`default`**: The value to return if the key is not found. This is optional and defaults to `None`.

#### **Return Value**:

- If the key is found, the method returns the corresponding value.
- If the key is not found and a **default value** is provided, it returns the default value.
- If the key is not found and no **default value** is provided, it returns `None`.

---

### **2. How `get()` Works**

- The `get()` method is used to safely retrieve the value associated with a given key in a dictionary.
- If the key exists, the method returns its value.
- If the key does not exist, it either returns the default value (if provided), or `None` (if no default is provided), without raising an exception.
- The `get()` method is often preferred over direct indexing (`[]`), especially when you are unsure if the key exists in the dictionary.

---

### **3. Examples of `get()` in Action**

#### **Example 1: Using `get()` to Retrieve an Existing Key**

```python
# Example dictionary
student_scores = {'Alice': 85, 'Bob': 90, 'Charlie': 78}

# Use get() to retrieve value for 'Bob'
score = student_scores.get('Bob')

print(score)  # Output: 90
```

- **Explanation**: Since the key `'Bob'` exists in the dictionary, the method returns the value `90`.

#### **Example 2: Using `get()` with a Non-Existent Key**

```python
# Example dictionary
student_scores = {'Alice': 85, 'Bob': 90, 'Charlie': 78}

# Attempt to get the score for a non-existent student
score = student_scores.get('David')

print(score)  # Output: None
```

- **Explanation**: The key `'David'` does not exist in the dictionary. Since no default value is specified, `None` is returned.

#### **Example 3: Using `get()` with a Default Value**

```python
# Example dictionary
student_scores = {'Alice': 85, 'Bob': 90, 'Charlie': 78}

# Attempt to get the score for a non-existent student, with a default value
score = student_scores.get('David', 'Not Found')

print(score)  # Output: Not Found
```

- **Explanation**: The key `'David'` does not exist, so the method returns the specified default value `'Not Found'`.

#### **Example 4: Using `get()` with None as Default**

```python
# Example dictionary
student_scores = {'Alice': 85, 'Bob': 90, 'Charlie': 78}

# Attempt to get a non-existent key, and explicitly return None
score = student_scores.get('David', None)

print(score)  # Output: None
```

- **Explanation**: The key `'David'` is not found, and since `None` is explicitly passed as the default value, `None` is returned.

---

### **4. Use Cases for `get()`**

The **`get()`** method is particularly useful when you want to avoid exceptions that occur when trying to access a key that might not exist in the dictionary. It is frequently used in scenarios where dictionary keys may or may not exist, and a fallback value is needed when the key is absent.

#### **4.1 Safe Dictionary Lookups**

When you're working with data where you cannot be sure if all the necessary keys exist, `get()` prevents the program from crashing.

Example:

```python
# Dictionary with user data
user_data = {'name': 'Alice', 'age': 25, 'city': 'New York'}

# Accessing keys safely
name = user_data.get('name', 'Unknown')  # Returns 'Alice'
email = user_data.get('email', 'No email provided')  # Returns 'No email provided'

print(name)   # Output: Alice
print(email)  # Output: No email provided
```

#### **4.2 Handling Missing Keys Gracefully**

In many cases, it's important to handle missing keys without throwing errors. For example, in configuration or data loading tasks, missing information can be handled by providing a default value.

Example:

```python
# Example of missing configurations
config = {'host': 'localhost', 'port': 8080}

# Get the configuration value for 'host'
host = config.get('host', 'default_host')

# Get the configuration value for 'user' (missing in config)
user = config.get('user', 'guest')

print(host)  # Output: localhost
print(user)  # Output: guest
```

#### **4.3 Default Dictionary Values in Loops**

You can use `get()` to set default values during loops, such as incrementing counters or collecting data in a dictionary.

Example:

```python
# List of items bought by customers
items_bought = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']

# Dictionary to store item count
item_count = {}

# Using get() to count occurrences of items
for item in items_bought:
    item_count[item] = item_count.get(item, 0) + 1

print(item_count)  # Output: {'apple': 3, 'banana': 2, 'orange': 1}
```

### **5. Differences between `get()` and `[]` (Indexing)**

The **`get()`** method and **dictionary indexing (`[]`)** both allow you to retrieve the value associated with a specific key, but they differ in behavior when the key is not found.

#### **Key Differences**:

1. **Handling Missing Keys**:

   - **`get()`**: Returns `None` (or a specified default value) if the key is not found.
   - **`[]` (Indexing)**: Raises a **`KeyError`** if the key is not found.

2. **Default Value**:
   - **`get()`**: Allows you to specify a default value to return when the key is not found.
   - **`[]` (Indexing)**: Does not allow specifying a default value.

#### **Example 1: Indexing**

```python
student_scores = {'Alice': 85, 'Bob': 90}

# Accessing an existing key
print(student_scores['Alice'])  # Output: 85

# Attempting to access a missing key will raise an error
# print(student_scores['David'])  # Raises KeyError: 'David'
```

#### **Example 2: Using `get()`**

```python
student_scores = {'Alice': 85, 'Bob': 90}

# Accessing an existing key
print(student_scores.get('Alice'))  # Output: 85

# Attempting to access a missing key returns None
print(student_scores.get('David'))  # Output: None

# Providing a default value for a missing key
print(student_scores.get('David', 'Not Found'))  # Output: Not Found
```

### **6. Time Complexity of `get()`**

The **time complexity** of the **`get()`** method is **O(1)** on average. This is because dictionaries in Python are implemented as hash tables, and retrieving a value based on a key is typically an **O(1)** operation. However, in rare cases (e.g., hash collisions), the operation may take longer, but it generally remains constant time.

- **Time Complexity**: **O(1)**
- **Space Complexity**: **O(1)** (constant space usage)

---

### **7. Common Pitfalls and Best Practices**

#### **7.1 Using `get()` Instead of Direct Indexing for Missing Keys**

When you're unsure whether a key exists in the dictionary, using `get()` is a safer approach compared to using direct indexing with `[]`, which would raise a `KeyError` for missing keys.

#### **7.2 Using Mutable Default Values**

When using `get()` with a default value, avoid using mutable objects (like lists or dictionaries) as the default value, because all keys will reference the same object. This can lead to unintended side effects when modifying the value associated with one key.

Example of a potential issue:

```python
# Mutable default value (list)
my_dict = {}
my_dict.get('key1', []).append(1)
print(my_dict)  # Output: {'key1': [1]}
```

To avoid this, always use immutable objects or a function that returns a fresh object for each key.

Example:

```python
# Use a default function for a fresh list
my_dict = {}
my_dict.get('key1', lambda: []).append(1)
print(my_dict)  # Output: {'key1': [1]}
```

---

### **Conclusion**

The **`get()`** method is a powerful and safe way to retrieve values from a dictionary. It prevents errors from occurring when the key does not exist, making it ideal for situations where the presence of a key is uncertain. It also allows for the specification of a default return value when the key is not found, adding flexibility to your code.

#### **Key Takeaways**:

- **Safely retrieve values** from a dictionary without causing a `KeyError`.
- **Provide a default value** to return if the key does not exist.
- **Time complexity** is **O(1)** on average.
- Use `get()` instead of direct indexing (`[]`) when the existence of the key is uncertain.
