# Python Dictionary `fromkeys()`

The **`fromkeys()`** method creates a new dictionary from the given sequence of elements with a value provided by the user.

**Syntax**:

```python
dictionary.fromkeys(sequence[, value])
```


## `fromkeys()` Parameters

The **`fromkeys()`** method takes two parameters:

- **sequence** - sequence of elements which is to be used as keys for the new dictionary
- **value (Optional)** - value which is set to each each element of the dictionary


## Return Value from `fromkeys()`

The **`fromkeys()`** method returns a new dictionary with the given sequence of elements as the keys of the dictionary.

If the value argument is set, each element of the newly created dictionary is set to the provided value.


````python
# Example 1: Create a dictionary from a sequence of keys

# vowels keys
keys = {'a', 'e', 'i', 'o', 'u' }

vowels = dict.fromkeys(keys)
print(vowels)
````

````python
# Example 2: Create a dictionary from a sequence of keys with value

# vowels keys
keys = {'a', 'e', 'i', 'o', 'u' }
value = 'vowel'

vowels = dict.fromkeys(keys, value)
print(vowels)
````

````python
# Example 3: Create a dictionary from mutable object list

# vowels keys
keys = {'a', 'e', 'i', 'o', 'u' }
value = [1]

vowels = dict.fromkeys(keys, value)
print(vowels)

# updating the value
value.append(2)
print(vowels)
````

**Note:** If **`value`** is a mutable object (whose value can be modified) like **[list](https://github.com/milaan9/02_Python_Datatypes/blob/main/003_Python_List.ipynb)**, **[dictionary](https://github.com/milaan9/02_Python_Datatypes/blob/main/005_Python_Dictionary.ipynb)**, etc., when the mutable object is modified, each element of the sequence also gets updated.

This is because each element is assigned a reference to the same object (points to the same object in the memory).

To avoid this issue, we use dictionary comprehension.


````python
# Example 4:

# vowels keys
keys = {'a', 'e', 'i', 'o', 'u' }
value = [1]

vowels = { key : list(value) for key in keys }
# you can also use { key : value[:] for key in keys }
print(vowels)

# updating the value
value.append(2)
print(vowels)
````

**Note:** Here, for each key in **`keys`**, a new list from **`value`** is created and assigned to it.

In essence, **`value`** isn't assigned to the element but a new list is created from it, which is then assigned to each element in the dictionary.


### **Python Dictionary `fromkeys()` Method: Concepts and Theory**

The **`fromkeys()`** method in Python is a built-in function that allows you to create a new dictionary from a sequence of keys, and it assigns a specified value to each of those keys. This method is often used when you need to create a dictionary where the keys are already known, and you want to initialize them with the same value.

Here, we will cover the following aspects of the `fromkeys()` method:

1. **Definition and Syntax**
2. **How `fromkeys()` Works**
3. **Examples of `fromkeys()` in Action**
4. **Use Cases for `fromkeys()`**
5. **Time Complexity**
6. **Common Pitfalls and Best Practices**

---

### **1. Definition and Syntax**

The **`fromkeys()`** method creates a new dictionary from the provided **iterable of keys**, assigning each key the same **default value**. If no value is provided, it assigns `None` as the default value.

#### **Syntax**:

```python
dict.fromkeys(iterable, value=None)
```

- **`iterable`**: An iterable (such as a list, tuple, or string) containing the keys for the new dictionary.
- **`value`**: The value to assign to each key (optional). If not provided, the default value will be `None`.

#### **Return Value**:

- This method returns a **new dictionary** with the specified keys and their corresponding values.

---

### **2. How `fromkeys()` Works**

- The `fromkeys()` method takes an iterable (like a list or tuple) as input and creates a dictionary where the elements of the iterable become the keys.
- Each key in the new dictionary is initialized with the specified **value** (or `None` if no value is provided).
- The dictionary returned by `fromkeys()` is a new dictionary and is independent of the original iterable.

---

### **3. Examples of `fromkeys()` in Action**

#### **Example 1: Creating a Dictionary with Default Values**

```python
# Create a dictionary with keys from a list and a default value of None
keys = ['name', 'age', 'city']
new_dict = dict.fromkeys(keys)

print(new_dict)  # Output: {'name': None, 'age': None, 'city': None}
```

- **Explanation**: In this example, the keys `['name', 'age', 'city']` are used to create a new dictionary. Since no value is provided, all keys are initialized with the default value `None`.

#### **Example 2: Creating a Dictionary with a Specified Value**

```python
# Create a dictionary with keys from a list and a default value of 0
keys = ['a', 'b', 'c']
new_dict = dict.fromkeys(keys, 0)

print(new_dict)  # Output: {'a': 0, 'b': 0, 'c': 0}
```

- **Explanation**: Here, the keys `['a', 'b', 'c']` are used to create a new dictionary. Each key is initialized with the value `0`.

#### **Example 3: Using a Tuple as the Iterable**

```python
# Create a dictionary with keys from a tuple and a default value of 'Unknown'
keys = ('name', 'age', 'city')
new_dict = dict.fromkeys(keys, 'Unknown')

print(new_dict)  # Output: {'name': 'Unknown', 'age': 'Unknown', 'city': 'Unknown'}
```

- **Explanation**: In this case, we used a tuple as the iterable. The dictionary is created with the keys `'name'`, `'age'`, and `'city'`, all initialized to the value `'Unknown'`.

#### **Example 4: Creating a Dictionary from a String**

```python
# Create a dictionary with keys from a string and a default value of False
keys = "abc"
new_dict = dict.fromkeys(keys, False)

print(new_dict)  # Output: {'a': False, 'b': False, 'c': False}
```

- **Explanation**: When you pass a string as the iterable, each character in the string becomes a key in the new dictionary, all initialized to the value `False`.

---

### **4. Use Cases for `fromkeys()`**

The **`fromkeys()`** method is useful when you need to create a dictionary with a known set of keys, and you want to initialize them with a default value. Some typical use cases include:

#### **4.1 Initializing a Dictionary for Counting**

You might use `fromkeys()` when you want to create a dictionary to store counts or frequencies of items, initializing the counts to zero.

Example:

```python
items = ['apple', 'banana', 'orange']
count_dict = dict.fromkeys(items, 0)

# Now you can update the dictionary as you count occurrences
count_dict['apple'] += 1
count_dict['banana'] += 2

print(count_dict)  # Output: {'apple': 1, 'banana': 2, 'orange': 0}
```

#### **4.2 Creating a Dictionary with Default Values for Multiple Keys**

If you have a list of categories or attributes, and you want to create a dictionary where each category starts with a default value, `fromkeys()` is very useful.

Example:

```python
categories = ['math', 'science', 'history']
student_scores = dict.fromkeys(categories, 0)

# Assign some scores
student_scores['math'] = 90
student_scores['science'] = 85

print(student_scores)  # Output: {'math': 90, 'science': 85, 'history': 0}
```

#### **4.3 Initializing a Dictionary with a Common Value**

When you need to initialize a dictionary with keys from a given iterable and set a common value for all the keys, `fromkeys()` is an ideal solution.

Example:

```python
students = ['Alice', 'Bob', 'Charlie']
attendance = dict.fromkeys(students, False)

print(attendance)  # Output: {'Alice': False, 'Bob': False, 'Charlie': False}
```

---

### **5. Time Complexity of `fromkeys()`**

- **Time Complexity**: The **`fromkeys()`** method has **O(n)** time complexity, where **n** is the number of keys in the iterable. This is because it needs to iterate through the iterable to create the new dictionary.
- **Space Complexity**: The space complexity is also **O(n)**, as the method creates a new dictionary of size equal to the number of keys in the iterable.

---

### **6. Common Pitfalls and Best Practices**

#### **6.1 Mutable Default Values**

One common pitfall when using `fromkeys()` is providing a **mutable object** (like a list or dictionary) as the default value. Since dictionaries in Python store references to objects, all keys in the dictionary will refer to the **same object**, which can lead to unintended side effects.

Example of a potential problem:

```python
# Mutable default value (list)
keys = ['a', 'b', 'c']
new_dict = dict.fromkeys(keys, [])

# Modify one of the lists
new_dict['a'].append(1)

print(new_dict)  # Output: {'a': [1], 'b': [1], 'c': [1]}
```

- **Explanation**: All keys in the dictionary point to the same list object. When you modify the list for one key, the change affects all the other keys as well.
- **Solution**: If you need each key to have a separate list (or any other mutable object), you should use a **default value factory**, such as a `lambda` function or `list()` to ensure a new object is created for each key.

Example:

```python
# Using a default factory for mutable objects (separate lists for each key)
new_dict = dict.fromkeys(keys, lambda: [])

# Modify one of the lists
new_dict['a'].append(1)

print(new_dict)  # Output: {'a': [1], 'b': [], 'c': []}
```

- **Explanation**: By using `lambda: []`, we ensure that each key in the dictionary has its own separate list.

#### **6.2 Using `fromkeys()` for Nested Structures**

If your dictionary contains nested structures, be cautious with `fromkeys()` as it can lead to unintended sharing of references. Always use a factory function or explicitly initialize each nested object if needed.

Example of a common mistake:

```python
nested_dict = dict.fromkeys(['key1', 'key2'], [])
nested_dict['key1'].append(10)
print(nested_dict)  # Output: {'key1': [10], 'key2': [10]}
```

- **Solution**: Use `lambda` to ensure independent lists for each key.

```python
nested_dict = dict.fromkeys(['key1', 'key2'], lambda: [])
nested_dict['key1'].append(10)
print(nested_dict)  # Output: {'key1': [10], 'key2': []}
```

---

### **Conclusion**

The **`fromkeys()`** method is a powerful tool for creating dictionaries with a predefined set of keys, all initialized to the same value. It simplifies the process of initializing dictionaries, especially when the set of keys is known in advance. However, you need to be cautious when using mutable values as defaults, as they can lead to unexpected behavior due to shared references.

#### **Key Takeaways**:

- **Creates a dictionary** with specified keys and a default value (or `None`).
- **Time complexity** is **O(n)** where n is the number of keys in the iterable.
- **Space complexity** is **O(n)** as a new dictionary is created.
- **Be cautious with mutable default values**, as they can lead to shared references between dictionary keys.
