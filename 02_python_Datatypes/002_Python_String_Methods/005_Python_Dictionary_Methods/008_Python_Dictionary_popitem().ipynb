# Python Dictionary `popitem()`

The **`popitem()`** method removes and returns the last element (key, value) pair inserted into the dictionary.

**Syntax**:

```python
dictionary.popitem()
```


## `popitem()` Parameters

The **`popitem()`** method doesn't take any parameters.


## Return Value from `popitem()`

The **`popitem()`** method removes and returns the **(key, value)** pair from the dictionary in the **Last In, First Out (LIFO)** order.

- Returns the latest inserted element **(key,value)** pair from the dictionary.
- Removes the returned element pair from the dictionary.

> **Note:** Before Python 3.7, the **`popitem()`** method returned and removed an arbitrary element (key, value) pair from the dictionary.


````python
# Example: Working of popitem() method

person = {'name': 'Phill', 'age': 22, 'salary': 3500.0}

# ('salary', 3500.0) is inserted at the last, so it is removed.
result = person.popitem()

print('Return Value = ', result)
print('person = ', person)

# inserting a new element pair
person['profession'] = 'Plumber'

# now ('profession', 'Plumber') is the latest element
result = person.popitem()

print('Return Value = ', result)
print('person = ', person)
````

> **Note:** The **`popitem()`** method raises a **`KeyError`** error if the dictionary is empty.


### Python Dictionary `popitem()` Method: Concepts and Theory

The **`popitem()`** method in Python is a built-in function used to remove and return a key-value pair from a dictionary. Unlike the `pop()` method, which requires you to specify a key, **`popitem()`** removes and returns an arbitrary key-value pair. This method is useful when you need to remove items from a dictionary in an unordered fashion, or when you don't care about which specific key-value pair you remove.

### 1. Definition and Syntax

The **`popitem()`** method is used to remove and return an arbitrary key-value pair from the dictionary. The method removes the **last inserted** key-value pair in Python 3.7 and later (since dictionaries are ordered by insertion), but in older versions (pre-Python 3.7), it removes a random item.

#### **Syntax**:

```python
dict.popitem()
```

- **No parameters**: The `popitem()` method doesn't require any parameters. It operates on the dictionary as a whole and removes an arbitrary key-value pair.

#### **Return Value**:

- The method returns a tuple, where the first element is the **key** and the second element is the **value**.
- If the dictionary is empty, it raises a **`KeyError`**.

---

### 2. How `popitem()` Works

- **Python 3.7 and Later**: The `popitem()` method removes and returns the **last inserted** key-value pair (i.e., the last pair added to the dictionary, based on insertion order).
- **Before Python 3.7**: Dictionaries were unordered, so `popitem()` would remove and return an **arbitrary** key-value pair.

### Example:

```python
# Python 3.7 and later example
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Remove and return the last inserted key-value pair
removed_item = person.popitem()

print(removed_item)  # Output: ('city', 'New York')
print(person)        # Output: {'name': 'Alice', 'age': 30}
```

- **Explanation**: In Python 3.7 and later, the `popitem()` method removes and returns the **last inserted** key-value pair (in this case, `'city': 'New York'`).

---

### 3. Examples of Using `popitem()`

#### Example 1: Removing the Last Inserted Item

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Using popitem() to remove and return the last item
last_item = person.popitem()

print(last_item)  # Output: ('city', 'New York')
print(person)     # Output: {'name': 'Alice', 'age': 30}
```

- **Explanation**: The `popitem()` method removes and returns the last inserted key-value pair. In this case, `'city': 'New York'` was the last inserted pair.

#### Example 2: Removing Items from an Empty Dictionary

If you call `popitem()` on an empty dictionary, it raises a `KeyError`.

```python
# Empty dictionary
empty_dict = {}

# Attempting to pop an item from an empty dictionary
try:
    empty_dict.popitem()
except KeyError:
    print("The dictionary is empty!")
```

**Output**:

```
The dictionary is empty!
```

- **Explanation**: When you try to use `popitem()` on an empty dictionary, a `KeyError` is raised, as there are no key-value pairs to remove.

---

### 4. Return Value of `popitem()`

The `popitem()` method returns a **tuple** containing the **key** and the **value** of the removed key-value pair.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Using popitem() to remove the last inserted item and return its key and value
key_value_pair = person.popitem()

print(key_value_pair)  # Output: ('city', 'New York')
```

- **Explanation**: The `popitem()` method returns a tuple, where the first element is the key and the second element is the corresponding value of the removed key-value pair.

---

### 5. Handling Empty Dictionaries

If the dictionary is empty and you call `popitem()`, a **`KeyError`** is raised because there are no items to remove. You can handle this by using a `try-except` block or by checking if the dictionary is empty before calling `popitem()`.

#### Example:

```python
# Empty dictionary
my_dict = {}

# Safe usage of popitem()
if my_dict:
    removed_item = my_dict.popitem()
    print(f"Removed item: {removed_item}")
else:
    print("The dictionary is empty, no items to pop.")
```

**Output**:

```
The dictionary is empty, no items to pop.
```

- **Explanation**: Before calling `popitem()`, the code checks if the dictionary is non-empty. If it is, the method is called; otherwise, a message is printed indicating that the dictionary is empty.

---

### 6. Use Cases for `popitem()`

#### 6.1 **Removing the Last Inserted Item (Ordered Dictionaries)**

In Python 3.7 and later, dictionaries are ordered by insertion, so you can use `popitem()` to remove the **last inserted item**. This can be useful when you need to pop items from a dictionary in reverse order of insertion.

#### Example: Reversing a Dictionary

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Remove items in reverse order using popitem()
while person:
    last_item = person.popitem()
    print(last_item)
```

**Output**:

```
('city', 'New York')
('age', 30)
('name', 'Alice')
```

- **Explanation**: This loop removes and prints the last inserted key-value pair repeatedly, which effectively reverses the order in which items were inserted into the dictionary.

#### 6.2 **Building a New Dictionary from Another**

You can use `popitem()` in a loop to extract and move items from one dictionary to another.

```python
# Example dictionary
old_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Creating a new dictionary by extracting items from old_dict
new_dict = {}

while old_dict:
    item = old_dict.popitem()
    new_dict[item[0]] = item[1]

print(new_dict)  # Output: {'city': 'New York', 'age': 30, 'name': 'Alice'}
```

- **Explanation**: The `popitem()` method is used to move items from `old_dict` to `new_dict`, preserving the order in which the items were removed.

---

### 7. Comparison with `pop()`

While both `pop()` and `popitem()` remove items from a dictionary, there are key differences between the two methods:

#### **`pop()`**:

- Requires you to specify a **key** to remove and return the associated value.
- Allows you to specify a **default** value if the key is not found, avoiding a `KeyError`.

#### **`popitem()`**:

- Removes and returns an **arbitrary** key-value pair (or the last inserted pair in Python 3.7+).
- Does not allow specifying a key or a default value.
- Raises a `KeyError` if the dictionary is empty.

#### Example Comparison:

```python
# Using pop()
person = {'name': 'Alice', 'age': 30}
removed_value = person.pop('age')  # Removes 'age' key
print(removed_value)  # Output: 30
print(person)          # Output: {'name': 'Alice'}

# Using popitem()
person = {'name': 'Alice', 'age': 30}
removed_item = person.popitem()  # Removes and returns the last inserted item
print(removed_item)  # Output: ('age', 30)
print(person)         # Output: {'name': 'Alice'}
```

---

### 8. Time Complexity

- **Time Complexity** of `popitem()` is **O(1)**. This means that removing the last item from the dictionary (or an arbitrary item in older versions of Python) is done in constant time, regardless of the dictionary's size.

---

### 9. Best Practices and Pitfalls

#### Best Practices:

- **Avoid using `popitem()` on empty dictionaries**: Always check if the dictionary is non-empty before using `popitem()`, or use a `try-except` block to handle a `KeyError`.
- **Use `popitem()` when order matters**: In Python 3.7+, if you need to remove items in the order they were inserted, `popitem()` is a good choice.

#### Pitfalls:

- **`popitem()` and empty dictionaries**: Calling `popitem()` on an empty dictionary will raise a `KeyError`. It's important to handle this error appropriately or ensure the dictionary is not empty before calling the method.
- **Uncertainty in older versions of Python**: In versions prior to Python 3.7, dictionaries were unordered, so `popitem()` would remove an arbitrary key-value pair, which could cause unexpected behavior if the order of removal matters.
