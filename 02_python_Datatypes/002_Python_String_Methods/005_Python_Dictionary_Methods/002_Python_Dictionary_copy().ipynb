# Python Dictionary `copy()`

The **`copy()`** method returns a shallow copy of the dictionary.

**Syntax**:

```python
dict.copy()
```


## `copy()` Parameters

The **`copy()`** method doesn't take any parameters.


## Return Value from `copy()`

The **`copy()`** method method returns a shallow copy of the dictionary. It doesn't modify the original dictionary.


````python
# Example 1: How copy works for dictionaries?

original = {1:'one', 2:'two'}
new = original.copy()

print('Orignal: ', original)
print('New: ', new)
````

## Difference in Using `copy()` method, and `=` Operator to Copy Dictionaries

When **`copy()`** method is used, a new dictionary is created which is filled with a copy of the references from the original dictionary.

When **`=`** operator is used, a new reference to the original dictionary is created.


````python
# Example 2: Using = Operator to Copy Dictionaries

original = {1:'one', 2:'two'}
new = original

# removing all elements from the list
new.clear()

print('new: ', new)
print('original: ', original)
````

**Note:** Here, when **`new`** dictionary is cleared, **`original`** dictionary is also cleared.


````python
# Example 3: Using copy() to Copy Dictionaries

original = {1:'one', 2:'two'}
new = original.copy()

# removing all elements from the list
new.clear()

print('new: ', new)
print('original: ', original)
````

**Note:** Here, when **`new`** dictionary is cleared, **`original`** dictionary remains unchanged.


### **Python Dictionary `copy()` Method: Concepts and Theory**

The **`copy()`** method in Python is a built-in function for dictionaries that creates a **shallow copy** of the dictionary. This method is commonly used when you want to create a new dictionary that is a copy of the original dictionary, but modifying the new dictionary does not affect the original dictionary. However, the copy is a **shallow** copy, meaning that nested objects (like lists or other dictionaries) are not copied deeply.

Here, we will cover the following aspects of the `copy()` method:

1. **Definition and Syntax**
2. **How `copy()` Works**
3. **Examples of `copy()` in Action**
4. **Shallow Copy vs Deep Copy**
5. **Use Cases for `copy()`**
6. **Time Complexity**
7. **Common Pitfalls and Best Practices**

---

### **1. Definition and Syntax**

The **`copy()`** method creates a shallow copy of the dictionary, meaning it copies the dictionary itself, but the nested objects inside the dictionary are still references to the original objects. This method does not affect the original dictionary.

#### **Syntax**:

```python
dict.copy()
```

- **`dict`**: The dictionary object from which you want to create a copy.

#### **Return Value**:

- The method returns a **new dictionary** that contains the same keys and values as the original dictionary.
- The new dictionary is independent of the original dictionary, but if the dictionary contains mutable objects (such as lists), those objects will be shared between the original and copied dictionaries.

---

### **2. How `copy()` Works**

- When you call `copy()`, the method creates a new dictionary with the same key-value pairs as the original dictionary.
- If the original dictionary contains mutable objects (like lists or other dictionaries), the copied dictionary will contain references to these objects.
- However, the copied dictionary itself is a **new object**, so changes to the dictionary itself (adding/removing keys, for example) will not affect the original.

---

### **3. Examples of `copy()` in Action**

#### **Example 1: Simple Copy**

```python
# Original dictionary
original_dict = {'name': 'Alice', 'age': 25, 'city': 'New York'}

# Create a copy of the original dictionary
copied_dict = original_dict.copy()

# Modify the copied dictionary
copied_dict['age'] = 26

# Output the dictionaries
print("Original Dictionary:", original_dict)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
print("Copied Dictionary:", copied_dict)      # Output: {'name': 'Alice', 'age': 26, 'city': 'New York'}
```

- **Explanation**: In this example, the `copy()` method creates a new dictionary that is independent of the original. Changing the value of `age` in the copied dictionary does not affect the original dictionary.

#### **Example 2: Copying a Dictionary with Nested Objects**

```python
# Original dictionary with nested list
original_dict = {'name': 'Alice', 'grades': [95, 88, 92]}

# Create a copy of the original dictionary
copied_dict = original_dict.copy()

# Modify the nested object (the list)
copied_dict['grades'][0] = 100

# Output the dictionaries
print("Original Dictionary:", original_dict)  # Output: {'name': 'Alice', 'grades': [100, 88, 92]}
print("Copied Dictionary:", copied_dict)      # Output: {'name': 'Alice', 'grades': [100, 88, 92]}
```

- **Explanation**: Although the `copied_dict` is a new dictionary, both the original and copied dictionaries share the same list in the `grades` key. Modifying the list in the copied dictionary also affects the original dictionary because lists are mutable objects and the copy is shallow.

---

### **4. Shallow Copy vs Deep Copy**

The **`copy()`** method creates a **shallow copy** of the dictionary, which means:

- The new dictionary is a new object, but the values are **references** to the original objects (if those values are mutable, such as lists or dictionaries).
- A **deep copy** would create new copies of the nested objects as well, rather than just copying the references.

#### **Shallow Copy**:

A shallow copy only copies the top-level objects and leaves the nested objects as references. This can be problematic if you modify a nested object in the copied dictionary, as changes will reflect in the original dictionary too.

Example of shallow copy:

```python
import copy

original_dict = {'a': [1, 2, 3], 'b': 10}
shallow_copy_dict = original_dict.copy()

shallow_copy_dict['a'][0] = 100
print("Original:", original_dict)    # Output: {'a': [100, 2, 3], 'b': 10}
print("Shallow Copy:", shallow_copy_dict)  # Output: {'a': [100, 2, 3], 'b': 10}
```

#### **Deep Copy**:

A deep copy creates new copies of both the dictionary and the nested objects.

To create a **deep copy**, you use the `copy` module's `deepcopy()` function:

```python
import copy

original_dict = {'a': [1, 2, 3], 'b': 10}
deep_copy_dict = copy.deepcopy(original_dict)

deep_copy_dict['a'][0] = 100
print("Original:", original_dict)    # Output: {'a': [1, 2, 3], 'b': 10}
print("Deep Copy:", deep_copy_dict)  # Output: {'a': [100, 2, 3], 'b': 10}
```

- **Explanation**: In the case of a deep copy, the change to the `a` list in the deep copy does not affect the original dictionary.

---

### **5. Use Cases for `copy()`**

The **`copy()`** method is useful in several situations:

#### **5.1 Avoiding Unintentional Modifications**

When you need to work with a copy of a dictionary to avoid altering the original, the `copy()` method is handy. For example, if you're going to experiment with modifying a dictionary without affecting the original data.

Example:

```python
original_data = {'a': 1, 'b': 2}
new_data = original_data.copy()
new_data['a'] = 10  # Modify the copied dictionary

print(original_data)  # {'a': 1, 'b': 2}
print(new_data)       # {'a': 10, 'b': 2}
```

#### **5.2 Working with Data Structures that Need to Be Independent**

When dealing with dictionaries that store data structures (like lists or other dictionaries), using `copy()` is beneficial when you need to ensure the copied dictionary is independent but still contains the same data structure references.

Example:

```python
my_dict = {'item1': [10, 20], 'item2': [30, 40]}
my_dict_copy = my_dict.copy()
my_dict_copy['item1'].append(50)

print(my_dict)          # {'item1': [10, 20, 50], 'item2': [30, 40]}
print(my_dict_copy)     # {'item1': [10, 20, 50], 'item2': [30, 40]}
```

#### **5.3 Temporary Changes Without Affecting Original**

You might want to temporarily modify a dictionary while keeping the original intact. The `copy()` method allows you to experiment without modifying the original dictionary.

Example:

```python
config = {'setting1': True, 'setting2': False}
config_copy = config.copy()
config_copy['setting1'] = False

print("Original Config:", config)       # {'setting1': True, 'setting2': False}
print("Modified Config:", config_copy)  # {'setting1': False, 'setting2': False}
```

---

### **6. Time Complexity of `copy()`**

- **Time Complexity**: The **`copy()`** method has **O(n)** time complexity, where **n** is the number of key-value pairs in the dictionary. This is because the method must copy each key-value pair individually.
- **Space Complexity**: The space complexity is **O(n)** because a new dictionary of the same size is created.

---

### **7. Common Pitfalls and Best Practices**

#### **7.1 Shared References in Shallow Copy**

Remember that the **`copy()`** method only performs a shallow copy. If the dictionary contains mutable objects (such as lists), those objects will not be duplicated but rather referenced. Modifying these objects in the copied dictionary will affect the original dictionary as well.

To avoid this, use `deepcopy()` if you need a true independent copy, especially when dealing with nested mutable objects.

#### **7.2 Handling Nested Dictionaries**

If your dictionary contains nested dictionaries and you modify the nested dictionary in the copied version, it will affect the original dictionary as well because the nested dictionaries are still references.

Example:

```python
original_dict = {'outer': {'inner': 'value'}}
copied_dict = original_dict.copy()

# Modify the inner dictionary in the copied dict
copied_dict['outer']['inner'] = 'new value'

print("Original Dictionary:", original_dict)  # Output: {'outer': {'inner': 'new value'}}
print("Copied Dictionary:", copied_dict)      # Output: {'outer': {'inner': 'new value'}}
```

In this case, you might want to use `deepcopy()` to avoid this issue.

---

### **Conclusion**

The **`copy()`** method is a useful tool for creating a shallow copy of a dictionary. It allows you to create a new dictionary with the same key-value pairs while ensuring that changes to the dictionary itself do not affect the original dictionary. However, be mindful of shallow copying when the dictionary contains mutable objects, as this can lead to unintended side effects. For deep copies, use the `deepcopy()` method from the `copy` module.

#### **Key Takeaways**:

- **Shallow copy**: The `copy()` method creates a shallow copy, meaning nested objects are shared between the original and copied dictionaries.
- **Deep copy**: For deep copying, use `copy.deepcopy()`.
- **Time complexity**: **O(n)**, where n is the number of key-value pairs in the dictionary.
- **Space complexity**: **O(n)** because a new dictionary is created.
