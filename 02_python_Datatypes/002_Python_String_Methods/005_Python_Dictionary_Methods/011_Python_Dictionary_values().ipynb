# Python Dictionary `values()`

The **`values()`** method returns a view object that displays a list of all the values in the dictionary.

**Syntax**:

```python
dictionary.values()
```


## `values()` Parameters

The **`values()`** method doesn't take any parameters.


## Return Value from `values()`

The **`values()`** method returns a view object that displays a list of all values in a given dictionary.


````python
# Example 1: Get all values from the dictionary

# random sales dictionary
sales = { 'apple': 2, 'orange': 3, 'grapes': 4 }

print(sales.values())
````

````python
# Example 2: How values() works when dictionary is modified?

# random sales dictionary
sales = { 'apple': 2, 'orange': 3, 'grapes': 4 }

values = sales.values()
print('Original items:', values)

# delete an item from dictionary
del[sales['apple']]
print('Updated items:', values)
````

**Explanation:**

The view object **`values`** doesn't itself return a list of **`sales`** item values but it returns a view of all values of the dictionary.

If the list is updated at any time, the changes are reflected on the view object itself, as shown in the above program.

### 1. Definition and Syntax

The **`values()`** method is used to return a **view object** that provides access to all the values in the dictionary. A view object is a dynamic object that reflects changes to the dictionary. If the dictionary is updated, the view object will be updated as well.

#### **Syntax**:

```python
dict.values()
```

- **`dict`**: The dictionary whose values you want to retrieve.

#### **Return Value**:

- The method returns a **view object** of the dictionary values.
- The view object reflects changes to the dictionary. If the dictionary is updated, the view object will reflect those updates as well.

---

### 2. How `values()` Works

- The `values()` method returns a **view** of the dictionary's values.
- It does not return a list of values but provides a view that can be used like an iterable object (e.g., in a `for` loop).
- A **view object** is a dynamic window into the dictionary's values, so any changes to the dictionary will be reflected in the view object.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Using values() to get the values of the dictionary
values_view = person.values()

print(values_view)  # Output: dict_values(['Alice', 30, 'New York'])
```

- **Explanation**: The `values()` method returns a **view object** of the dictionary values. In this case, it is a `dict_values` object that holds the values `'Alice'`, `30`, and `'New York'`.

---

### 3. Examples of Using `values()`

#### Example 1: Basic Usage

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Get the dictionary values
values = person.values()

# Print the values
print(values)  # Output: dict_values(['Alice', 30, 'New York'])
```

- **Explanation**: This example retrieves the values of the dictionary `person`, which are `'Alice'`, `30`, and `'New York'`.

#### Example 2: Iterating Over Values

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Iterate over the values of the dictionary
for value in person.values():
    print(value)
```

- **Output**:

  ```
  Alice
  30
  New York
  ```

- **Explanation**: The `for` loop iterates over the values of the dictionary, printing each value.

#### Example 3: Checking if a Value Exists

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Check if a value exists in the dictionary
if 30 in person.values():
    print("Age is in the dictionary.")
else:
    print("Age is not in the dictionary.")
```

- **Output**: `Age is in the dictionary.`
- **Explanation**: The `values()` method is used with the `in` keyword to check if the value `30` exists in the dictionary.

---

### 4. Return Value of `values()`

The **`values()`** method returns a **view object** of the dictionary's values. This view object reflects changes made to the dictionary. It behaves like a set in the sense that it only contains the unique values of the dictionary.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Get the values of the dictionary
values = person.values()

# Add a new item to the dictionary
person['gender'] = 'Female'

print(values)  # Output: dict_values(['Alice', 30, 'New York', 'Female'])
```

- **Explanation**: After adding the new key `'gender'` with the value `'Female'`, the `values()` view object is updated to reflect this change.

---

### 5. Iterating Over Values

One common use case for `values()` is to iterate over the dictionary values.

#### Example 1: Basic Iteration

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Iterate over the values using a for loop
for value in person.values():
    print(value)
```

- **Output**:

  ```
  Alice
  30
  New York
  ```

- **Explanation**: You can iterate over the `dict_values` object returned by `values()` just like any other iterable object, such as a list or a set.

#### Example 2: Applying Functions to Values

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Use values() to print the uppercase form of each value (if it's a string)
for value in person.values():
    if isinstance(value, str):
        print(value.upper())
```

- **Output**:

  ```
  ALICE
  NEW YORK
  ```

- **Explanation**: The `values()` method is used in conjunction with a loop to apply a function (in this case, converting the string values to uppercase).

---

### 6. Converting Values to a List

Since `values()` returns a **view object**, it can be converted to a list if you need a **list of values**.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Convert the values to a list
values_list = list(person.values())

print(values_list)  # Output: ['Alice', 30, 'New York']
```

- **Explanation**: The `list()` function is used to convert the `dict_values` view object into a list of values. This can be useful if you need a static collection of the values that won't change as the dictionary is updated.

---

### 7. Checking if a Value Exists

You can check if a specific value exists in the dictionary by using the `in` keyword in combination with `values()`.

#### Example:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}

# Check if a value exists in the dictionary
if 'Alice' in person.values():
    print("Alice is in the dictionary.")
else:
    print("Alice is not in the dictionary.")
```

- **Output**: `Alice is in the dictionary.`
- **Explanation**: This example uses `values()` with the `in` keyword to check if the string `'Alice'` exists as a value in the dictionary.

---

### 8. Time Complexity

The time complexity of the **`values()`** method is **O(1)** because it simply returns a view object of the dictionary values, which is done in constant time. However, the **time complexity of iterating over the values** would be **O(n)**, where `n` is the number of key-value pairs in the dictionary.

---

### 9. Use Cases for `values()`

#### 9.1 **Extracting All Values for Processing**

If you need to perform operations on all the values in a dictionary (for example, calculating the sum of numeric values), `values()` is the method to use.

Example: Summing all the numeric values in a dictionary:

```python
# Example dictionary
scores = {'math': 90, 'english': 85, 'science': 88}

# Sum the values
total_score = sum(scores.values())

print(f"Total score: {total_score}")  # Output: Total score: 263
```

- **Explanation**: The `values()` method is used to extract all the values from the dictionary, and the `sum()` function is used to calculate the total.

#### 9.2 **Identifying Unique Values**

If you need to identify unique values in a dictionary, you can convert the `values()` view to a set (since sets don't allow duplicates).

Example: Finding unique values in a dictionary:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30, 'city': 'New York', 'gender': 'Female'}

# Find unique values
unique_values = set(person.values())

print(unique_values)  # Output: {'Alice', 'Female', 30, 'New York'}
```

- **Explanation**: Converting the values from the dictionary to a set removes any duplicate values, allowing you to work with only the unique values.

---

### 10. Best Practices and Pitfalls

#### Best Practices:

- **Use `values()` for iterating**: It's a good choice when you need to process all the values of a dictionary.
- **Convert to a list when needed**: If you need a static copy of the dictionary's values (e

.g., for sorting or indexing), convert the view object to a list.

#### Pitfalls:

- **Dynamic view**: The `values()` method returns a view object, which can be surprising if you expect a static list of values. The view will change if the dictionary is modified.

Example of this behavior:

```python
# Example dictionary
person = {'name': 'Alice', 'age': 30}

# Get the values view
values_view = person.values()

# Modify the dictionary
person['city'] = 'New York'

# The view object reflects the change
print(values_view)  # Output: dict_values(['Alice', 30, 'New York'])
```

- **Explanation**: Since the `values()` view object is dynamic, any changes to the dictionary are reflected in the view.
