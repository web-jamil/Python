# Python String `index()`

The string **`index()`** method returns the index of a substring inside the **[string](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String.ipynb)** (if found). If the substring is not found, it raises an exception.

**Syntax**:

```python
str.index(sub[, start[, end]] )
```


## `index()` Parameters

The **`index()`** method takes three parameters:

- **sub** - substring to be searched in the string **`str`**.
- **start** and **`end`** (optional) - substring is searched within **`str[start:end]`**.


## Return Value from `index()`

- If substring exists inside the string, it returns the lowest index in the string where substring is found.
- If substring doesn't exist inside the string, it raises a **ValueError** exception.

The **`index()`** method is similar to **[find() method for strings](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/008_Python_String_find%28%29.ipynb)**.

The only difference is that **`find()`** method returns -1 if the substring is not found, whereas **`index()`** throws an exception.


````python
# Example 1: index() With Substring argument Only

sentence = 'Python programming is fun.'

result = sentence.index('is fun')
print("Substring 'is fun':", result)

result = sentence.index('Java')
print("Substring 'Java':", result)
````

> **Note:** Index in Python starts from 0 and not 1. So the occurrence is 19 and not 20.


````python
# Example 2: index() With start and end Arguments

sentence = 'Python programming is fun.'

# Substring is searched in 'gramming is fun.'
print(sentence.index('ing', 10))

# Substring is searched in 'gramming is '
print(sentence.index('g is', 10, -4))

# Substring is searched in 'programming'
print(sentence.index('fun', 7, 18))
````

````python
# Example 3:

s="I Love Python Tutorial"
print(s.index('i'))
print(s.index('i',2))
print(s.index('Love'))
print(s.index('Love',7))
print(s.index('i',0,10))
````

### **Python `index()` Method: All About with Theory**

In Python, the `index()` method is a built-in method used to find the **first occurrence** of a value in a sequence (such as a list, string, or tuple). It returns the **index** of the first occurrence of the specified value. If the value is not found, it raises a `ValueError`.

### **Syntax of the `index()` Method**

```python
sequence.index(value, start, end)
```

#### **Parameters:**

1. **`value`** (required):
   - The value you are looking for in the sequence. This can be any type of object (string, integer, etc.).
2. **`start`** (optional):
   - The index to start the search. The default value is `0` (the beginning of the sequence).
3. **`end`** (optional):
   - The index at which to end the search. The default value is the length of the sequence, meaning it will search until the end of the sequence.

#### **Return Value:**

- The method returns the **index** of the first occurrence of the `value` in the sequence.
- If the `value` is not found in the specified range, it raises a **`ValueError`**.

---

### **Examples of Using the `index()` Method**

#### **1. Basic Example with a List**

Let's first see how the `index()` method works on a **list**.

```python
# Example 1: Using index() on a list
my_list = [10, 20, 30, 40, 50]
index_of_30 = my_list.index(30)
print(index_of_30)
```

**Output**:

```
2
```

Explanation:

- The `index()` method searches for `30` in the list `[10, 20, 30, 40, 50]` and returns the index `2`, since `30` is the third element (0-based indexing).

#### **2. Handling `ValueError` When the Value is Not Found**

If the value you are searching for does not exist in the sequence, Python will raise a `ValueError`.

```python
# Example 2: Handling ValueError when the value is not found
my_list = [10, 20, 30, 40, 50]
try:
    index_of_60 = my_list.index(60)
except ValueError as e:
    print(f"Error: {e}")
```

**Output**:

```
Error: 60 is not in list
```

Explanation:

- The value `60` is not present in the list, so a `ValueError` is raised, and the error message is printed.

#### **3. Using `index()` with a String**

The `index()` method also works with **strings** to find the position of a substring.

```python
# Example 3: Using index() on a string
my_string = "Hello, world!"
index_of_o = my_string.index('o')
print(index_of_o)
```

**Output**:

```
4
```

Explanation:

- The `index()` method finds the first occurrence of the letter `'o'` in the string `'Hello, world!'` at index `4` (0-based indexing).

#### **4. Using `start` and `end` Parameters**

You can specify a **range** in which to search using the `start` and `end` parameters. This can help narrow down the search to a specific part of the sequence.

```python
# Example 4: Using start and end parameters with a list
my_list = [10, 20, 30, 40, 50, 30]
index_of_30 = my_list.index(30, 2)  # Start searching from index 2
print(index_of_30)
```

**Output**:

```
2
```

Explanation:

- The search for `30` starts from index `2`. The first occurrence of `30` is at index `2`, so that is returned.

Here’s another example where we use both `start` and `end` parameters:

```python
# Example 5: Using start and end parameters with a string
my_string = "Hello, world!"
index_of_l = my_string.index('l', 3, 8)  # Searching between indices 3 and 8
print(index_of_l)
```

**Output**:

```
3
```

Explanation:

- The search for `'l'` starts at index `3` and ends at index `8`. The first occurrence of `'l'` in this range is at index `3`.

#### **5. Index of a Tuple**

The `index()` method can also be used on **tuples**.

```python
# Example 6: Using index() on a tuple
my_tuple = (1, 2, 3, 4, 5)
index_of_4 = my_tuple.index(4)
print(index_of_4)
```

**Output**:

```
3
```

Explanation:

- The `index()` method finds `4` in the tuple `(1, 2, 3, 4, 5)` and returns its index, which is `3`.

#### **6. Searching for Multiple Occurrences**

If you want to find multiple occurrences of a value in a sequence, you will have to call `index()` in a loop, adjusting the `start` parameter for each subsequent search.

```python
# Example 7: Finding multiple occurrences of a value in a list
my_list = [10, 20, 30, 40, 20, 50, 20]
start = 0
while True:
    try:
        # Search for 20 starting from 'start' index
        index_of_20 = my_list.index(20, start)
        print(f"20 found at index: {index_of_20}")
        start = index_of_20 + 1  # Start the next search from the next index
    except ValueError:
        break  # Exit the loop when no more 20's are found
```

**Output**:

```
20 found at index: 1
20 found at index: 4
20 found at index: 6
```

Explanation:

- We use a `while` loop and call `index()` repeatedly to find all occurrences of `20` in the list. After each search, we adjust the `start` parameter so the search starts from the next index.

---

### **Important Notes:**

1. **Zero-Indexed:**
   - Python uses **zero-based indexing**, meaning the index of the first element in the sequence is `0`, the second element is `1`, and so on.
2. **Raises `ValueError` if not found:**

   - If the `value` does not exist in the sequence and no `start` or `end` limits match the search, Python raises a `ValueError`. This is important to handle gracefully in your code (e.g., using `try-except` blocks).

3. **Efficiency Consideration:**

   - The `index()` method stops at the **first occurrence** of the value. If there are multiple occurrences, only the first one is returned.
   - The time complexity of the `index()` method is **O(n)**, where `n` is the length of the sequence, as it may need to check each element in the sequence until it finds the desired value.

4. **Using `index()` vs `find()` in Strings:**
   - In strings, the `find()` method is similar to `index()`, but the key difference is that `find()` returns `-1` if the value is not found, while `index()` raises a `ValueError`.

---

### **Comparison Between `index()` and `find()` (For Strings)**

| Method    | Returns When Found            | Returns When Not Found | Common Use Case                                                      |
| --------- | ----------------------------- | ---------------------- | -------------------------------------------------------------------- |
| `index()` | Index of the first occurrence | Raises `ValueError`    | When you want to ensure the value exists in the sequence.            |
| `find()`  | Index of the first occurrence | `-1`                   | When you want to check if the value exists without raising an error. |

---

### **Conclusion**

- The `index()` method is used to find the index of the **first occurrence** of a value in a sequence (such as a list, tuple, or string).
- It takes optional `start` and `end` parameters to limit the search to a specific range of the sequence.
- If the value is not found, it raises a `ValueError`.
- It is helpful when you want to locate the first occurrence of a value, and you can handle cases where the value doesn't exist by using `try-except` blocks or conditional checks.


### **ValueError in Python's `index()` Method: All About**

In Python, the `index()` method is used to find the **first occurrence** of a specified value in a sequence (such as a list, string, or tuple). If the value is **not found** in the sequence, Python raises a **`ValueError`**.

### **Understanding `ValueError` in the `index()` Method**

The `ValueError` is raised by the `index()` method when it cannot find the specified value in the given sequence. The error indicates that the search was unsuccessful, and it provides a message that explains why the error occurred.

#### **Syntax of `index()` Method**

```python
sequence.index(value, start, end)
```

- **`value`**: The value you want to search for in the sequence.
- **`start`** (optional): The index at which to start the search. Default is `0`.
- **`end`** (optional): The index at which to end the search. Default is the length of the sequence.

If the `value` you specify does not exist in the sequence (or the specified `start` and `end` range), a `ValueError` will be raised.

---

### **Example of `ValueError` When the Value is Not Found**

Here's an example of the `ValueError` being raised when searching for an element that is not present in the sequence.

```python
# Example 1: ValueError raised when value is not found
my_list = [1, 2, 3, 4, 5]

# Trying to find the index of a value that doesn't exist in the list
try:
    index_of_6 = my_list.index(6)
except ValueError as e:
    print(f"Error: {e}")
```

**Output**:

```
Error: 6 is not in list
```

Explanation:

- In this example, the list `my_list` contains the values `[1, 2, 3, 4, 5]`, but we are trying to find the index of the number `6`, which is not present.
- The `index()` method raises a `ValueError` and the error message is printed: `"6 is not in list"`.

---

### **Detailed Explanation of `ValueError`**

When the `index()` method fails to find the specified `value` in the sequence, it raises a `ValueError` exception with a message that typically follows this format:

```python
ValueError: <value> is not in <sequence>
```

For example, if you were searching for the string `'apple'` in a list, and `'apple'` was not found, the message would be:

```python
ValueError: 'apple' is not in list
```

#### **Key Points:**

1. **The Search is Unsuccessful:**

   - The `ValueError` is raised **only** if the value is **not found** in the sequence. If the value is found, the `index()` method will return the index of the first occurrence.

2. **The Range Affects the Search:**
   - The optional `start` and `end` parameters **do not prevent the `ValueError` from being raised**. If the value is not found within the specified range, the method will still raise a `ValueError`.

---

### **Handling `ValueError` Using `try-except` Block**

When using the `index()` method, it's common to use a `try-except` block to handle the potential `ValueError` gracefully. This prevents your program from crashing if the value is not found, allowing you to provide an alternate behavior or a user-friendly message.

```python
# Example 2: Using try-except to handle ValueError
my_tuple = ('apple', 'banana', 'cherry')

try:
    index_of_orange = my_tuple.index('orange')
except ValueError as e:
    print(f"ValueError: {e}")
```

**Output**:

```
ValueError: 'orange' is not in tuple
```

Explanation:

- In this case, the tuple `('apple', 'banana', 'cherry')` does not contain the string `'orange'`. The `index()` method raises a `ValueError`, and the error is caught by the `except` block, printing the message `"ValueError: 'orange' is not in tuple"`.

---

### **Avoiding `ValueError` by Checking First**

If you want to avoid the `ValueError` entirely, you can first check whether the value exists in the sequence using the `in` operator before calling `index()`.

```python
# Example 3: Checking before calling index()
my_list = [1, 2, 3, 4, 5]

if 6 in my_list:
    index_of_6 = my_list.index(6)
else:
    print("6 is not in the list")
```

**Output**:

```
6 is not in the list
```

Explanation:

- By using the `in` operator (`if 6 in my_list`), we check if `6` exists in the list before calling `index()`. This prevents the `ValueError` from being raised, as we know that the value is not present in the list.

---

### **Using `index()` with `start` and `end` Parameters**

You can also use the optional `start` and `end` parameters to restrict the search to a specific section of the sequence. However, even with these parameters, if the value is not found within the given range, a `ValueError` will still be raised.

```python
# Example 4: Using start and end parameters
my_list = [1, 2, 3, 4, 5]

# Searching within a specific range (indices 2 to 4)
try:
    index_of_3 = my_list.index(3, 2, 5)  # 3 is within the range 2 to 4
    print(f"Found 3 at index: {index_of_3}")
except ValueError as e:
    print(f"Error: {e}")

# Searching for a value outside of the range (should raise ValueError)
try:
    index_of_3_out_of_range = my_list.index(3, 4, 5)  # 3 is not within the range 4 to 5
except ValueError as e:
    print(f"Error: {e}")
```

**Output**:

```
Found 3 at index: 2
Error: 3 is not in list
```

Explanation:

- In the first case, we successfully find `3` within the range from index `2` to `5`, so the index `2` is returned.
- In the second case, the `index()` method raises a `ValueError` because `3` is not found in the range from index `4` to `5`.

---

---

### **Best Practices to Avoid `ValueError`**

1. **Use `in` before `index()`:**

   - Always check if the value exists in the sequence using the `in` operator before calling `index()`. This helps avoid the exception.

   ```python
   if value in sequence:
       index_of_value = sequence.index(value)
   else:
       print(f"{value} not found")
   ```

2. **Use `try-except` Block:**

   - Use a `try-except` block to gracefully handle the `ValueError` when you're unsure if the value is present.

   ```python
   try:
       index_of_value = sequence.index(value)
   except ValueError:
       print(f"{value} not found in sequence")
   ```

3. **Use `.find()` for Strings:**

   - If you're working with strings and want to avoid exceptions, use the `.find()` method, which returns `-1` when the value is not found instead of raising an exception.

   ```python
   index_of_char = my_string.find('z')
   if index_of_char == -1:
       print("Character not found")
   ```

---

### **Conclusion**

- **`ValueError`** in Python's `index()` method is raised when the specified value cannot be found in the sequence.
- This error is an indication that the value you're searching for does not exist in the list, tuple, or string (or the specified range).
- To handle this gracefully, you can use a `try-except` block or check the value's existence with the `in` operator before calling `index()`.
