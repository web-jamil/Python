# Python List `index()`

The **`index()`** method returns the index of the specified element in the list.

**Syntax**:

```python
list.index(element, start, end)
```


## `index()` Parameters

The list **`index()`** method can take a maximum of three arguments:

- **element** - the element to be searched
- **start (optional)** - start searching from this index
- **end (optional)** - search the element up to this index


## Return Value from `index()`

- The **`index()`** method returns the index of the given element in the list.
- If the element is not found, a **`ValueError`** exception is raised.

> **Note:** The **`index()`** method only returns the first occurrence of the matching element.


````python
# Example 1: Find the index of the element

# vowels list
vowels = ['a', 'e', 'i', 'o', 'i', 'u']

# index of 'e' in vowels
index = vowels.index('e')
print('The index of e:', index)

# element 'i' is searched
# index of the first 'i' is returned
index = vowels.index('i')

print('The index of i:', index)
````

````python
# Example 2: Index of the Element not Present in the List

# vowels list
vowels = ['a', 'e', 'i', 'o', 'u']

# index of'p' is vowels
index = vowels.index('p')
print('The index of p:', index)
````

````python
# Example 3: Working of index() With Start and End Parameters

# alphabets list
alphabets = ['a', 'e', 'i', 'o', 'g', 'l', 'i', 'u']

# index of 'i' in alphabets
index = alphabets.index('e')   # 2
print('The index of e:', index)

# 'i' after the 4th index is searched
index = alphabets.index('i', 4)   # 6
print('The index of i:', index)

# 'i' between 3rd and 5th index is searched
index = alphabets.index('i', 3, 5)   # Error!
print('The index of i:', index)
````

### **Python List Index Functions: Concepts and Theory**

In Python, a **list** is a built-in data type that holds an ordered collection of items, which can be of any data type. Lists are mutable, meaning that their elements can be changed, added, or removed. One of the core features of lists is the ability to access elements by their **index**.

In this guide, we'll explore the **index-related functions** for lists, such as `list.index()`, `list.count()`, and how to use indices for list manipulation effectively.

---

### **1. Accessing List Elements Using Indices**

Lists in Python are indexed collections, meaning that each element in the list has a unique index, starting from `0` for the first element and incrementing by `1` for each subsequent element.

For example:

```python
my_list = ['apple', 'banana', 'cherry', 'date']
```

- **`my_list[0]`** would give `'apple'`
- **`my_list[1]`** would give `'banana'`
- **`my_list[-1]`** would give `'date'` (negative indexing starts from the end)

### **2. List Index Functions**

Python provides a few list methods specifically for working with indices. The most important ones are:

#### **2.1. `list.index(x[, start[, end]])`**

This function is used to find the **first occurrence** of an element in the list. It returns the **index** of the first occurrence of the element `x` within the list, optionally within a specified range from index `start` to index `end`.

- **Syntax**:

  ```python
  list.index(x[, start[, end]])
  ```

- **Parameters**:

  - `x`: The element to find in the list.
  - `start`: The index to start searching from (optional).
  - `end`: The index to stop searching at (optional).

- **Return value**: The index of the first occurrence of `x`.

#### **Example**:

```python
my_list = ['apple', 'banana', 'cherry', 'banana', 'date']
index_of_banana = my_list.index('banana')
print(index_of_banana)  # Output: 1
```

#### **Important Notes**:

- If the element `x` is **not found** in the list, Python will raise a `ValueError`.

  ```python
  my_list = ['apple', 'banana', 'cherry']
  my_list.index('orange')  # Raises ValueError
  ```

- You can specify a **start** and **end** range to search within a sublist.

  ```python
  my_list = ['apple', 'banana', 'cherry', 'banana', 'date']
  index_of_banana = my_list.index('banana', 2)  # Starts searching from index 2
  print(index_of_banana)  # Output: 3
  ```

---

#### **2.2. `list.count(x)`**

The **`count()`** method is used to count how many times an element appears in the list. It returns an **integer** value indicating the frequency of the element `x` in the list.

- **Syntax**:

  ```python
  list.count(x)
  ```

- **Parameters**:

  - `x`: The element whose occurrences you want to count in the list.

- **Return value**: The number of occurrences of the element `x`.

#### **Example**:

```python
my_list = ['apple', 'banana', 'cherry', 'banana', 'date']
banana_count = my_list.count('banana')
print(banana_count)  # Output: 2
```

- This method helps you find how often a particular element occurs in a list.

---

#### **2.3. `list.index()` vs `list.count()`**

- **`list.index(x)`**: Returns the **first occurrence** index of the element `x` in the list.
- **`list.count(x)`**: Returns the **number of occurrences** of the element `x` in the list.

You may combine both methods if you need both the count and the first index of the element.

For example, to find the first index of an element and how many times it appears:

```python
my_list = ['apple', 'banana', 'cherry', 'banana', 'date']
element = 'banana'

index = my_list.index(element)  # First occurrence
count = my_list.count(element)  # Number of occurrences

print(f"First index of '{element}': {index}")
print(f"Total count of '{element}': {count}")
```

---

### **3. Negative Indexing and Slicing**

Negative indices in Python allow you to reference elements starting from the end of the list. This is especially useful for accessing elements without knowing the exact length of the list.

- **`list[-1]`** refers to the last element.
- **`list[-2]`** refers to the second-to-last element, and so on.

#### **Example**:

```python
my_list = ['apple', 'banana', 'cherry', 'date']
print(my_list[-1])  # Output: 'date' (last element)
print(my_list[-2])  # Output: 'cherry' (second-to-last element)
```

---

### **4. List Slicing**

List slicing allows you to extract a part of the list by specifying a start and end index.

- **Syntax**: `list[start:end]`

  - `start`: The index of the first element to include (inclusive).
  - `end`: The index to stop at (exclusive).

- You can also use **negative indices** with slicing.

#### **Example**:

```python
my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']

# Slicing to get elements from index 1 to 3 (exclusive)
sub_list = my_list[1:4]
print(sub_list)  # Output: ['banana', 'cherry', 'date']

# Using negative indexing
sub_list = my_list[-3:-1]
print(sub_list)  # Output: ['cherry', 'date']
```

- **Step size in slicing**: You can also include a step size to skip elements.

  ```python
  my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']
  print(my_list[::2])  # Output: ['apple', 'cherry', 'elderberry']
  ```

- **Negative step**: Slicing can also use a negative step to reverse the list.

  ```python
  my_list = ['apple', 'banana', 'cherry', 'date', 'elderberry']
  print(my_list[::-1])  # Output: ['elderberry', 'date', 'cherry', 'banana', 'apple']
  ```

---

### **5. Modifying List Elements Using Indices**

Python allows you to modify list elements using their indices.

#### **Example**:

```python
my_list = ['apple', 'banana', 'cherry']
my_list[1] = 'blueberry'
print(my_list)  # Output: ['apple', 'blueberry', 'cherry']
```

- You can also modify multiple elements using slicing.

```python
my_list[1:3] = ['blackberry', 'date']
print(my_list)  # Output: ['apple', 'blackberry', 'date']
```

---

### **6. Advanced Indexing: `enumerate()` and List Comprehensions**

Sometimes, you may want to access both the **index and the value** of elements during iteration. The `enumerate()` function is perfect for this.

#### **`enumerate()` Function**:

- **Syntax**: `enumerate(iterable, start=0)`
  - `iterable`: The list or other iterable to loop through.
  - `start`: The starting index (default is 0).

#### **Example**:

```python
my_list = ['apple', 'banana', 'cherry']
for idx, value in enumerate(my_list):
    print(f"Index {idx}: {value}")
```

**Output**:

```
Index 0: apple
Index 1: banana
Index 2: cherry
```

This is useful when you need both the index and the value while iterating over a list.

### **7. Practical Use Cases**

#### **7.1 Finding First Occurrence and Counting Occurrences**

You can find the **first occurrence** of an element and **count** how many times it appears using the `index()` and `count()` methods.

```python
my_list = ['apple', 'banana', 'cherry', 'banana', 'date']
first_index = my_list.index('banana')
count = my_list.count('banana')

print(f"First occurrence of 'banana' is at index {first_index}")
print(f"'banana' appears {count} times in the list.")
```

#### **7.2 Checking for an Element's Existence Using `in`**

You can check if an element exists in a list without using `index()` or `count()` by using the `in` keyword.

```python
if 'banana' in my_list:
    print("'banana' is present in the list")
```
