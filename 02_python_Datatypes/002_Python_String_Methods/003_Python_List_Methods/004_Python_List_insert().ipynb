# Python List `insert()`

The list **`insert()`** method inserts an element to the list at the specified index.

**Syntax**:

```python
list.insert(i, elem)
```

Here, **`elem`** is inserted to the list at the **$i_{th}$** index. All the elements after **`elem`** are shifted to the right.


## `insert()` Parameters

The **`insert()`** method takes two parameters:

- **index** - the index where the element needs to be inserted
- **element** - this is the element to be inserted in the list

> **Notes:**

- If **`index`** is 0, the element is inserted at the beginning of the list.
- If **`index`** is 3, the element is inserted after the 3rd element. Its position will be 4th.


## Return Value from `insert()`

The **`insert()`** method doesn't return anything; returns None. It only updates the current list.


````python
# Example 1: Inserting an Element to the List

# vowel list
vowel = ['a', 'e', 'i', 'u']

# 'o' is inserted at index 3
# the position of 'o' will be 4th
vowel.insert(3, 'o')

print('Updated List:', vowel)
````

````python
# Example 2: Inserting a Tuple (as an Element) to the List

mixed_list = [{1, 2}, [5, 6, 7]]

# number tuple
number_tuple = (3, 4)

# inserting a tuple to the list
mixed_list.insert(1, number_tuple)

print('Updated List:', mixed_list)
````

### **Python List `insert()` Method: Concepts and Theory**

The **`insert()`** method in Python is a powerful built-in function used to **insert an element at a specific position** in a list. This provides flexibility when you need to add an element at a particular index in a list, rather than simply appending it to the end (using **`append()`**) or extending the list with multiple elements (using **`extend()`**).

Let’s dive deep into the **`insert()`** method, its syntax, use cases, and comparisons with other list manipulation methods.

---

### **1. Understanding the `insert()` Method**

The **`insert()`** method allows you to **insert a single element** at a **specific index** in a list, shifting the rest of the elements to the right.

#### **Syntax**:

```python
list.insert(index, element)
```

- **`index`**: The position where the new element should be inserted. The index is **zero-based**, meaning `0` refers to the first element, `1` refers to the second, and so on. Negative indices are also supported (where `-1` refers to the last element).
- **`element`**: The item that you want to insert at the specified index. This can be any valid Python data type (e.g., integer, string, list, tuple, etc.).

#### **Return Value**:

- The **`insert()`** method modifies the original list in place and **does not return anything** (`None`). The element is inserted, and the rest of the list is adjusted accordingly.

#### **Example**:

```python
my_list = [10, 20, 30, 40]
my_list.insert(2, 25)  # Insert 25 at index 2
print(my_list)  # Output: [10, 20, 25, 30, 40]
```

Here, the number `25` is inserted between `20` and `30` at index `2`, shifting the elements after it.

---

### **2. Key Characteristics of `insert()`**

#### **2.1 Index-Based Insertion**

The most important feature of **`insert()`** is the ability to specify exactly where you want to place the new element. The index determines the position.

- **Example**: Insert at the start of the list:

```python
my_list = [10, 20, 30]
my_list.insert(0, 5)  # Insert 5 at the start (index 0)
print(my_list)  # Output: [5, 10, 20, 30]
```

- **Example**: Insert at the end of the list:

```python
my_list = [10, 20, 30]
my_list.insert(len(my_list), 40)  # Insert 40 at the end of the list
print(my_list)  # Output: [10, 20, 30, 40]
```

#### **2.2 Negative Indexing**

You can also use **negative indexing** in **`insert()`**, where `-1` refers to the last element, `-2` refers to the second-to-last element, and so on.

- **Example**: Insert at the second-to-last position:

```python
my_list = [10, 20, 30, 40]
my_list.insert(-1, 35)  # Insert 35 before the last element (index -1)
print(my_list)  # Output: [10, 20, 30, 35, 40]
```

#### **2.3 Shifting of Elements**

When you use **`insert()`**, the elements that are after the specified index are **shifted one position to the right**. This allows you to preserve the relative order of elements in the list.

- **Example**:

```python
my_list = [1, 2, 3, 4, 5]
my_list.insert(3, 10)  # Insert 10 at index 3 (before 4)
print(my_list)  # Output: [1, 2, 3, 10, 4, 5]
```

In this case, elements at indices `3` and higher (`4`, `5`) are shifted one place to the right to make space for the inserted element.

---

### **3. Edge Cases in `insert()`**

#### **3.1 Inserting at an Index Beyond the List Size**

If you try to insert an element at an index that is **larger than the current length** of the list, the new element is **added at the end** of the list, just like **`append()`**.

- **Example**:

```python
my_list = [1, 2, 3]
my_list.insert(10, 4)  # Index 10 is beyond the list size, so 4 is appended at the end
print(my_list)  # Output: [1, 2, 3, 4]
```

#### **3.2 Inserting at a Negative Index Out of Range**

When using negative indices that exceed the list's size, Python will treat the index as the **start of the list** (i.e., `-1` will point to the last element).

- **Example**:

```python
my_list = [10, 20, 30]
my_list.insert(-5, 5)  # Invalid negative index, so element is inserted at the start
print(my_list)  # Output: [5, 10, 20, 30]
```

---

### **4. Performance Considerations**

The **`insert()`** method is a relatively **efficient operation** for small lists. However, as the list grows larger, the time complexity of **insertion** increases due to the need to **shift elements** to the right after inserting the new item.

- **Time Complexity**:

  - Inserting at the beginning or middle of a list has a time complexity of **O(n)**, where **n** is the number of elements after the insertion point. This is because all the elements need to be shifted to the right.
  - Inserting at the end of the list (when the index is equal to the list's length) has a time complexity of **O(1)**, as no elements need to be shifted.

- **Space Complexity**: The space complexity remains **O(1)**, as the list itself is modified in-place.

---

### **5. `insert()` vs Other List Methods**

While **`insert()`** is used for inserting an element at a specific index, other list methods like **`append()`**, **`extend()`**, and **`pop()`** are used for different purposes.

#### **5.1 `append()`**:

- **Purpose**: Adds a single element to the end of the list.
- **Comparison**: Unlike `insert()`, **`append()`** always adds elements at the **end** of the list, not at a specific index.

```python
my_list = [1, 2, 3]
my_list.append(4)  # Adds 4 at the end
print(my_list)  # Output: [1, 2, 3, 4]
```

#### **5.2 `extend()`**:

- **Purpose**: Adds multiple elements (from an iterable) to the end of the list.
- **Comparison**: **`extend()`** is used when you want to add multiple elements at the **end** of the list, whereas **`insert()`** allows you to add a single element at a **specific index**.

```python
my_list = [1, 2, 3]
my_list.extend([4, 5, 6])  # Adds each element of the list
print(my_list)  # Output: [1, 2, 3, 4, 5, 6]
```

#### **5.3 `pop()`**:

- **Purpose**: Removes an element at a specified index and returns it. If no index is provided, it removes and returns the last element.
- **Comparison**: **`pop()`** removes an element, while **`insert()`** adds an element at the specified index.

```python
my_list = [10, 20, 30]
removed_element = my_list.pop(1)  # Removes the element at index 1 (20)
print(my_list)  # Output: [10, 30]
```

### **6. Practical Use Cases for `insert()`**

#### **6.1 Insert at a Specific Position**

You may need to insert an element at a specific position in the list, such as when adding an item in sorted order or when constructing a list from user input.

- **Example**: Insert an element in sorted order:

```python
sorted_list = [1, 3, 5, 7]
sorted_list.insert(1, 2)  # Insert 2 at the correct position to maintain order
print(sorted_list)  # Output: [1, 2, 3, 5, 7]
```

#### **6.2 Inserting a Placeholder Value**

In some cases, you may need to insert a **placeholder** value into a list, such as initializing an empty spot in a list or adding a specific value at a particular index.

- **Example**: Insert a placeholder value:

```python
my_list = ['apple', 'banana', 'cherry']
my_list.insert(1, 'orange')  # Insert 'orange' between 'apple' and 'banana'
print(my_list)  # Output: ['apple', 'orange', 'banana', 'cherry']
```

#### **6.3 Building a List Dynamically**

You can use **`insert()`** when building a list dynamically, especially when certain items need to be inserted at specific positions during iteration.

```python
my_list = []
my_list.insert(0, 'start')  # Insert 'start' at the beginning
my_list.insert(1, 'middle')  # Insert 'middle' after 'start'
my_list.insert(2, 'end')  # Insert 'end' after 'middle'
print(my_list)  # Output: ['start', 'middle', 'end']
```

---

### **7. Common Errors and Edge Cases**

#### **7.1 Index Out of Range**

If you provide an index that is out of range (either too large or too small), **`insert()`** behaves in a predictable way. A larger index will append the element, and a smaller index (for negative numbers) will insert at the start.

```python
my_list = [1, 2, 3]
my_list.insert(10, 4)  # Index 10 is out of range, so 4 is added to the end
print(my_list)  # Output: [1, 2, 3, 4]
```

#### **7.2 Inserting Invalid Data Types**

You can insert any valid Python data type, but you must ensure that the data type you insert fits the context of your program.

```python
my_list = [1, 2, 3]
my_list.insert(1, {'key': 'value'})  # Inserting a dictionary
print(my_list)  # Output: [1, {'key': 'value'}, 2, 3]
```

It is particularly useful when you need fine-grained control over where an item is added in a list. By understanding its syntax, behavior, and use cases, you can effectively leverage **`insert()`** for dynamic list manipulation, solving a variety of problems involving ordered data.
