# Python List `copy()`

The **`copy()`** method returns a shallow copy of the list.

A **[list](https://github.com/milaan9/02_Python_Datatypes/blob/main/003_Python_List.ipynb)** can be copied using the **`=`** operator. For example,


````python
old_list = [1, 2, 3]
new_list = old_list
print('old_list:', old_list)
print('new_list:', new_list)
````

The problem with copying lists in this way is that if you modify **`new_list`**, **`old_list`** is also modified. It is because the new list is referencing or pointing to the same **`old_list`** object.


````python
old_list = [1, 2, 3]
new_list = old_list

# add an element to list
new_list.append('a')

print('New List:', new_list)
print('Old List:', old_list)
````

However, if you need the original list unchanged when the new list is modified, you can use the **`copy()`** method.

> Related tutorial: **[Python Shallow Copy Vs Deep Copy](https://github.com/milaan9/02_Python_Datatypes/blob/main/003_Python_List_Methods/Python_Shallow_Copy_and_Deep_Copy.ipynb)**


**Syntax**:

```python
new_list = list.copy()
```


## `copy()` Parameters

The **`copy()`** method doesn't take any parameters.


## Return Value from `copy()`

The **`copy()`** method returns a new list. It doesn't modify the original list.


````python
# Example 1: Copying a List

# mixed list
my_list = ['cat', 0, 6.7]

# copying a list
new_list = my_list.copy()

print('Copied List:', new_list)
````

If you modify the **`new_list`** in the above example, **`my_list`** will not be modified.


````python
# Copy List Using Slicing Syntax

# shallow copy using the slicing syntax

# mixed list
list = ['cat', 0, 6.7]

# copying a list using slicing
new_list = list[:]

# Adding an element to the new list
new_list.append('dog')

# Printing new and old list
print('Old List:', list)
print('New List:', new_list)
````

### **Python List `copy()` Method: Concepts and Theory**

The **`copy()`** method in Python is a built-in function that allows you to create a **shallow copy** of a list. A shallow copy means a new list is created with the same elements as the original list, but the elements themselves are references to the same objects as in the original list.

In this guide, we will explore the **`copy()`** method in detail, covering its syntax, behavior, how it differs from deep copying, practical use cases, and performance considerations.

---

### **1. Understanding the `copy()` Method**

The **`copy()`** method is used to create a **new list** that is a shallow copy of the original list. This new list will have the same elements as the original, but if the elements are mutable (like other lists or dictionaries), changes made to the elements in the new list will also affect the original list.

#### **Syntax**:

```python
new_list = original_list.copy()
```

- **`original_list`**: The list you want to copy.
- **`new_list`**: The new list created as a shallow copy of the original list.

#### **Return Value**:

- The **`copy()`** method returns a new list that is a shallow copy of the original list.
- It does not return **`None`**. Instead, the new list is returned with the same elements as the original.

#### **Example**:

```python
original_list = [1, 2, 3, 4]
new_list = original_list.copy()
print(new_list)  # Output: [1, 2, 3, 4]
```

In this example, **`new_list`** is a shallow copy of **`original_list`**. Both lists contain the same values, but they are distinct objects in memory.

---

### **2. Shallow Copy vs. Deep Copy**

#### **2.1 Shallow Copy**

A **shallow copy** creates a new list with the same references to the objects as the original list. This means that if the elements of the list are mutable (e.g., lists, dictionaries, sets), modifying the elements inside the new list will also modify the corresponding elements in the original list.

- **Example** (shallow copy with mutable elements):

```python
original_list = [[1, 2], [3, 4]]
new_list = original_list.copy()
new_list[0][0] = 99  # Modify the first element of the inner list
print(original_list)  # Output: [[99, 2], [3, 4]]
print(new_list)       # Output: [[99, 2], [3, 4]]
```

In this case, both **`original_list`** and **`new_list`** point to the same inner lists. Modifying the inner list through **`new_list`** also changes it in **`original_list`**.

#### **2.2 Deep Copy**

A **deep copy**, on the other hand, creates a completely new list with new instances of the objects inside the list. Therefore, modifications to the new list or its elements do not affect the original list at all.

To make a deep copy of a list, you can use the `copy` module's **`deepcopy()`** function.

- **Example** (deep copy with mutable elements):

```python
import copy
original_list = [[1, 2], [3, 4]]
new_list = copy.deepcopy(original_list)
new_list[0][0] = 99
print(original_list)  # Output: [[1, 2], [3, 4]]
print(new_list)       # Output: [[99, 2], [3, 4]]
```

In this case, **`original_list`** and **`new_list`** are completely independent of each other, even the nested elements.

---

### **3. Key Characteristics of `copy()`**

#### **3.1 Shallow Copy**

The **`copy()`** method performs a shallow copy, meaning:

- A new list is created, but the elements in the list are still references to the objects in the original list.
- If the elements are mutable (like lists or dictionaries), changes to those elements will be reflected in both the original and the copied list.

- **Example** (shallow copy of a list with mutable elements):

```python
original_list = [[1, 2], [3, 4]]
new_list = original_list.copy()
new_list[1][1] = 99
print(original_list)  # Output: [[1, 2], [3, 99]]
print(new_list)       # Output: [[1, 2], [3, 99]]
```

#### **3.2 Creates a New List**

The **`copy()`** method creates a new list, but it does not create new copies of the objects contained in the original list. This means that only the outer list is copied; the inner objects (if mutable) are not copied but referenced.

- **Example**:

```python
original_list = [1, 2, 3]
new_list = original_list.copy()
print(original_list is new_list)  # Output: False (they are different objects)
```

In this example, **`original_list`** and **`new_list`** are different objects, but they contain the same elements.

#### **3.3 Does Not Affect Original List**

The **`copy()`** method does not affect the original list. After performing a copy, both the original and new lists can be modified independently, unless the elements are mutable objects, in which case changes to the elements are shared between the original and new lists.

- **Example** (modifying the outer list):

```python
original_list = [1, 2, 3]
new_list = original_list.copy()
new_list.append(4)
print(original_list)  # Output: [1, 2, 3]
print(new_list)       # Output: [1, 2, 3, 4]
```

In this case, the outer lists are independent, and the change to **`new_list`** does not affect **`original_list`**.

---

### **4. Performance Considerations**

The **`copy()`** method performs the operation in **O(n)** time complexity, where **n** is the number of elements in the list. This is because it needs to copy each reference in the list.

- **Time Complexity**: **O(n)**, where **n** is the number of elements in the list.
- **Space Complexity**: **O(n)**, because a new list is created, and the original list is not modified.

Since **`copy()`** is a shallow copy, it is faster than performing a deep copy, especially when the elements of the list are immutable. However, if the elements are mutable and you need to ensure complete independence between the original and the new list, you should use **`deepcopy()`** from the `copy` module.

---

### **5. Use Cases for `copy()`**

The **`copy()`** method is useful in many practical scenarios where you need to create a new list that is independent of the original but shares the same elements. Here are some common use cases:

#### **5.1 Avoiding Accidental Modifications**

If you want to preserve the original list while modifying a new list, you can use **`copy()`** to create a new list.

- **Example** (avoiding accidental modifications):

```python
original_list = [1, 2, 3, 4]
new_list = original_list.copy()
new_list[0] = 99
print(original_list)  # Output: [1, 2, 3, 4]
print(new_list)       # Output: [99, 2, 3, 4]
```

By copying the list, you ensure that the original list is not affected by changes made to the new list.

#### **5.2 Working with Mutable Elements**

If you have a list containing mutable objects (like other lists) and want to create a new list without modifying the original list, you can use **`copy()`**. However, remember that this is a shallow copy, so any changes to the mutable elements inside the new list will affect the original list.

- **Example** (working with mutable elements):

```python
original_list = [[1, 2], [3, 4]]
new_list = original_list.copy()
new_list[0][0] = 99
print(original_list)  # Output: [[99, 2], [3, 4]]
print(new_list)       # Output: [[99, 2], [3, 4]]
```

If you need complete independence of the lists, use **`deepcopy()`** instead of **`copy()`**.

#### **5.3 When Working with Large Lists**

When working with large lists, using **`copy()`** allows you to efficiently create a new list without modifying the original list. Since the copying operation is shallow, the overhead is relatively low, especially for lists containing immutable objects.

---

### **6. Common Errors and Edge Cases**

#### **6.1 Attempting to Copy a Non-List Object**

Since the **`copy()`** method is a list method, it cannot be used on other data types like strings, sets, or tuples. Attempting to call **`copy()`** on a non-list will result in an **`AttributeError`**.

- **Example** (using `copy()` on a non-list):

```python
my_tuple = (1, 2, 3)
my_tuple.copy()  # Raises AttributeError: 'tuple' object has no attribute 'copy'
```

#### **6.2 Modifying the Inner List Elements (Shallow Copy)**

When making a shallow copy, any modifications to mutable elements inside the new list (if the list contains nested lists or dictionaries) will affect the original list.

- **Example** (modifying mutable inner elements):

```python
original_list = [[1, 2], [3, 4]]
new_list = original_list.copy()
new_list[0][0] = 99
print(original_list)  # Output: [[99, 2], [3, 4]]
print(new_list)       # Output: [[99, 2], [3, 4]]
```

To avoid this, use **`deepcopy()`** if you need complete independence between the original and new lists.

---

### **Conclusion**

The **`copy()`** method in Python is a simple and efficient way to create a shallow copy of a list. It is useful when you want to create a new list that shares the same elements as the original but is independent in terms of the outer list. However, be mindful of the fact that **`copy()`** creates references to the same objects in the original list, so modifying mutable elements will affect both the original and the copied lists.

For situations where you need a completely independent copy, especially when dealing with nested lists or other mutable objects, you should use the **`deepcopy()`** method from the `copy` module.
