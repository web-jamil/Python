# Python List `sort()`

The **`sort()`** method sorts the elements of a given list in a specific ascending or descending order.

**Syntax**:

```python
list.sort(key=..., reverse=...)
```

Alternatively, you can also use Python's built-in **[sorted()](https://github.com/milaan9/04_Python_Functions/blob/main/002_Python_Functions_Built_in/060_Python_sorted%28%29.ipynb)** function for the same purpose.

```python
sorted(list, key=..., reverse=...)
```

> **Note:** The simplest difference between **`sort()`** and **`sorted()`** is: **`sort()`** changes the list directly and doesn't return any value, while **`sorted()`** doesn't change the list and returns the sorted list.


## `sort()` Parameters

By default, **`sort()`** doesn't require any extra parameters. However, it has two optional parameters:

- **reverse** - If **`True`**, the sorted list is reversed (or sorted in Descending order)
- **key** - function that serves as a key for the sort comparison


## Return Value from `sort()`

The **`sort()`** method doesn't return any value. Rather, it changes the original list.

If you want a function to return the sorted list rather than change the original list, use **`sorted()`**.


````python
# Example 1: Sort a given list

# vowels list
vowels = ['e', 'a', 'u', 'o', 'i']

# sort the vowels
vowels.sort()

# print vowels
print('Sorted list:', vowels)
````

## Sort in Descending order

The **`sort()`** method accepts a **`reverse`** parameter as an optional argument.

Setting **`reverse = True`** sorts the list in the descending order.

```python
list.sort(reverse=True)
```

Alternately for **`sorted()`**, you can use the following code.

```python
sorted(list, reverse=True)
```


````python
# Example 2: Sort the list in Descending order

# vowels list
vowels = ['e', 'a', 'u', 'o', 'i']

# sort the vowels
vowels.sort(reverse=True)

# print vowels
print('Sorted list (in Descending):', vowels)
````

## Sort with custom function using key

If you want your own implementation for sorting, the **`sort()`** method also accepts a **`key`** function as an optional parameter.

Based on the results of the key function, you can sort the given list.

```python
list.sort(key=len)
```

Alternatively for sorted:

```python
sorted(list, key=len)
```

Here, **`len`** is the Python's in-built function to count the length of an element.

The list is sorted based on the length of each element, from lowest count to highest.


We know that a tuple is sorted using its first parameter by default. Let's look at how to customize the **`sort()`** method to sort using the second element.


````python
# Example 3: Sort the list using key

# take second element for sort
def takeSecond(elem):
    return elem[1]

# random list
random = [(2, 2), (3, 4), (4, 1), (1, 3)]

# sort list with key
random.sort(key=takeSecond)

# print list
print('Sorted list:', random)
````

Let's take another example. Suppose we have a list of information about the employees of an office where each element is a dictionary.

We can sort the list in the following way:


````python
# sorting using custom key
employees = [
    {'Name': 'Alan Turing', 'age': 25, 'salary': 10000},
    {'Name': 'Sharon Lin', 'age': 30, 'salary': 8000},
    {'Name': 'John Hopkins', 'age': 18, 'salary': 1000},
    {'Name': 'Mikhail Tal', 'age': 40, 'salary': 15000},
]

# custom functions to get employee info
def get_name(employee):
    return employee.get('Name')


def get_age(employee):
    return employee.get('age')


def get_salary(employee):
    return employee.get('salary')


# sort by name (Ascending order)
employees.sort(key=get_name)
print(employees, end='\n\n')

# sort by Age (Ascending order)
employees.sort(key=get_age)
print(employees, end='\n\n')

# sort by salary (Descending order)
employees.sort(key=get_salary, reverse=True)
print(employees, end='\n\n')
````

Here, for the first case, our custom function returns the name of each employee. Since the name is a **`string`**, Python by default sorts it using the alphabetical order.

For the second case, age (**`int`**) is returned and is sorted in ascending order.

For the third case, the function returns the salary (**`int`**), and is sorted in the descending order using **`reverse = True`**.

It is a good practice to use the lambda function when the function can be summarized in one line. So, we can also write the above program as:


````python
# sorting using custom key
employees = [
    {'Name': 'Alan Turing', 'age': 25, 'salary': 10000},
    {'Name': 'Sharon Lin', 'age': 30, 'salary': 8000},
    {'Name': 'John Hopkins', 'age': 18, 'salary': 1000},
    {'Name': 'Mikhail Tal', 'age': 40, 'salary': 15000},
]

# sort by name (Ascending order)
employees.sort(key=lambda x: x.get('Name'))
print(employees, end='\n\n')

# sort by Age (Ascending order)
employees.sort(key=lambda x: x.get('age'))
print(employees, end='\n\n')

# sort by salary (Descending order)
employees.sort(key=lambda x: x.get('salary'), reverse=True)
print(employees, end='\n\n')
````

### **Python List `sort()` Method: Concepts and Theory**

The **`sort()`** method is a built-in Python function used to arrange the elements of a list in a specific order. The order can either be ascending (default) or descending, and the sorting can be customized using a key function.

In this guide, we will explore the **`sort()`** method in Python in detail, covering its syntax, behavior, key characteristics, performance, edge cases, and how it differs from similar methods like `sorted()`.

---

### **1. Understanding the `sort()` Method**

The **`sort()`** method modifies a list **in place** by arranging the elements in a specific order. It does not create a new list but sorts the original list. The sorting is done based on either the natural order of the elements (for example, numerical or lexicographical) or custom order defined by a sorting key.

#### **Syntax**:

```python
list.sort(reverse=False, key=None)
```

- **`reverse`** (optional): A boolean value that determines the sort order.
  - **`False`** (default): Sorts in ascending order.
  - **`True`**: Sorts in descending order.
- **`key`** (optional): A function that takes one argument and returns a value to be used for sorting the list. This allows custom sorting criteria.

#### **Return Value**:

- The **`sort()`** method does not return any value. It sorts the list in place and returns **`None`**.

#### **Example**:

```python
my_list = [3, 1, 4, 1, 5, 9, 2]
my_list.sort()
print(my_list)  # Output: [1, 1, 2, 3, 4, 5, 9]
```

In this example, the list is sorted in ascending order (default behavior).

---

### **2. Key Characteristics of `sort()`**

#### **2.1 In-Place Sorting**

The **`sort()`** method modifies the list **in place**. This means that the original list is rearranged, and no new list is created. If you need to preserve the original list and create a sorted version, you should use **`sorted()`**, which returns a new list.

- **Example** (modifies the original list):

```python
my_list = [4, 2, 3, 1]
my_list.sort()
print(my_list)  # Output: [1, 2, 3, 4] (list is modified in place)
```

#### **2.2 Sorting Order**

By default, the **`sort()`** method arranges the elements in **ascending order**. You can reverse the order by passing **`reverse=True`**.

- **Example** (sorting in descending order):

```python
my_list = [3, 1, 4, 1, 5, 9, 2]
my_list.sort(reverse=True)
print(my_list)  # Output: [9, 5, 4, 3, 2, 1, 1]
```

#### **2.3 Using the `key` Argument**

The **`key`** argument allows you to customize the sorting logic by specifying a function to be applied to each element before sorting. This is useful when you need to sort objects like strings, tuples, or dictionaries based on custom criteria.

- **Example** (sorting by length of strings):

```python
my_list = ['apple', 'banana', 'kiwi', 'grape']
my_list.sort(key=len)
print(my_list)  # Output: ['kiwi', 'apple', 'grape', 'banana']
```

In this example, the list of strings is sorted based on their length, not alphabetically.

#### **2.4 Stability of Sort**

Python’s **`sort()`** method is **stable**, meaning that when multiple records have the same key, their original order is preserved.

- **Example** (stable sorting):

```python
my_list = [('apple', 5), ('banana', 3), ('cherry', 5), ('date', 3)]
my_list.sort(key=lambda x: x[1])
print(my_list)  # Output: [('banana', 3), ('date', 3), ('apple', 5), ('cherry', 5)]
```

In this example, elements with the same value (5 or 3) maintain their original relative order after sorting.

---

### **3. Performance Considerations**

The **`sort()`** method uses an algorithm called **Timsort** (a hybrid sorting algorithm derived from merge sort and insertion sort). This algorithm provides excellent performance for real-world data.

- **Time Complexity**:

  - **Best Case**: **O(n)**, when the list is already sorted.
  - **Average Case**: **O(n log n)**.
  - **Worst Case**: **O(n log n)**.

- **Space Complexity**: **O(n)** (Timsort is a stable sort that requires additional memory).

Timsort is highly efficient for most practical use cases, which is why it is the default sorting algorithm in Python.

---

### **4. `sort()` vs `sorted()`**

Although both **`sort()`** and **`sorted()`** are used to sort lists, they have important differences:

| Feature           | `sort()`                                  | `sorted()`                                                          |
| ----------------- | ----------------------------------------- | ------------------------------------------------------------------- |
| **Return Type**   | **None** (modifies the list in place)     | A new sorted list (does not modify the original list)               |
| **Original List** | Modified                                  | Remains unchanged                                                   |
| **Use Case**      | When you want to modify the list in place | When you want a new sorted list without modifying the original list |

#### **Example** (using `sorted()`):

```python
my_list = [3, 1, 4, 1, 5, 9, 2]
new_sorted_list = sorted(my_list)
print(my_list)  # Output: [3, 1, 4, 1, 5, 9, 2] (original list is unchanged)
print(new_sorted_list)  # Output: [1, 1, 2, 3, 4, 5, 9]
```

#### **Key Differences**:

- **`sort()`** modifies the original list and returns `None`.
- **`sorted()`** creates a new sorted list without modifying the original.

---

### **5. Edge Cases and Error Handling**

#### **5.1 Sorting an Empty List**

An empty list remains unchanged when sorted. Calling **`sort()`** on an empty list will not raise any error, and the list will remain empty.

- **Example** (empty list):

```python
my_list = []
my_list.sort()
print(my_list)  # Output: []
```

#### **5.2 Sorting a List with Mixed Data Types**

If the list contains elements of mixed data types (such as numbers and strings), **`sort()`** will raise a **`TypeError`** because Python cannot compare incompatible data types by default.

- **Example** (mixed data types):

```python
my_list = [3, 'banana', 1, 'apple']
my_list.sort()  # Raises TypeError: '<' not supported between instances of 'str' and 'int'
```

To sort lists with mixed data types, you may need to provide a custom `key` function that defines how to handle comparisons.

- **Example** (sorting by string representation):

```python
my_list = [3, 'banana', 1, 'apple']
my_list.sort(key=str)  # Sorts by string representation of elements
print(my_list)  # Output: [1, 3, 'apple', 'banana']
```

---

### **6. Practical Use Cases for `sort()`**

The **`sort()`** method is useful in many scenarios, especially when dealing with numerical or lexicographical data. Here are some practical examples:

#### **6.1 Sorting Numbers**

You can use **`sort()`** to sort lists of numbers in ascending or descending order.

- **Example** (sorting numbers in ascending order):

```python
numbers = [10, 2, 5, 7, 3]
numbers.sort()
print(numbers)  # Output: [2, 3, 5, 7, 10]
```

#### **6.2 Sorting Strings Alphabetically**

Strings are sorted in lexicographical (alphabetical) order by default.

- **Example** (sorting strings):

```python
words = ['apple', 'banana', 'cherry', 'date']
words.sort()
print(words)  # Output: ['apple', 'banana', 'cherry', 'date']
```

#### **6.3 Sorting Objects with Custom Criteria**

You can sort objects (such as dictionaries, tuples, or custom objects) using the `key` argument to define a sorting criterion.

- **Example** (sorting tuples by the second element):

```python
my_list = [(1, 'apple'), (3, 'banana'), (2, 'cherry')]
my_list.sort(key=lambda x: x[1])  # Sort by the second element (fruit names)
print(my_list)  # Output: [(1, 'apple'), (3, 'banana'), (2, 'cherry')]
```

#### **6.4 Sorting Complex Data Structures**

If you have a list of dictionaries or custom objects, you can sort them based on specific fields or attributes.

- **Example** (sorting a list of dictionaries by age):

```python
people = [{'name': 'Alice', 'age': 25}, {'name': 'Bob', 'age': 30}, {'name': 'Charlie', 'age': 20}]
people.sort(key=lambda x: x['age'])
print(people)  # Output: [{'name': 'Charlie', 'age': 20}, {'name': 'Alice', 'age': 25}, {'name': 'Bob', 'age': 30}]
```

---

### **7. Common Errors and Edge Cases**

#### **7.1 Attempting to Sort Non-List Objects**

Since **`sort()`** is a list method, it can only be applied to lists. If you try to use it on a non-list object, Python will raise an **`AttributeError`**.

- **Example** (non-list object):

```python
my_string = "Hello"
my_string.sort()  # Raises AttributeError: 'str' object has no attribute 'sort'
```

#### **7.2 Sorting a List of None Values**

If the list contains `None` values, the **`sort()`** method will not raise an error, but `None` will be placed at the beginning (in ascending order). This behavior might need special handling in some cases.

- **Example** (sorting with `None` values):

```python
my_list = [3, None, 2, None, 1]
my_list.sort()
print(my_list)  # Output: [None, None, 1, 2, 3]
```

### **Conclusion**

The **`sort()`** method is a versatile and powerful function for sorting lists in Python. It offers the ability to sort in ascending or descending order, customize sorting using the `key` argument, and is highly efficient for most practical use cases due to its underlying Timsort algorithm.

It is important to remember that **`sort()`** modifies the list \*\*

in place**, so if you need to preserve the original list, use the **`sorted()`** function instead. By understanding the various parameters and performance characteristics, you can make efficient and effective use of the **`sort()`\*\* method in your Python programs.
