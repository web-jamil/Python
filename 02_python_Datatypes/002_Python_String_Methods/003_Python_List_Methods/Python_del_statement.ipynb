# Python del Statement

In this class, you will learn to use the **`del`** keyword with the help of examples.

**Syntax**:

```python
del obj_name
```

Here, **`del`** is a Python keyword. And, **`obj_name`** can be variables, user-defined objects, lists, items within lists, dictionaries etc.


````python
# Example 1: Delete an user-defined object

class MyClass:
    a = 10
    def func(self):
        print('Hello')

print(MyClass)   # Output: <class '__main__.MyClass'>
````

````python
del MyClass     # deleting MyClass

print(MyClass)  # Error: MyClass is not defined
````

In the program, we have deleted **`MyClass`** using **`del MyClass`** statement.


````python
# Example 2: Delete variable, list, and dictionary

my_var = 5
my_tuple = ('Arthur', 33)
my_dict = {'name': 'Arthur', 'age': 33}

del my_var
del my_tuple
del my_dict
````

````python
# Example 3: Remove items, slices from a list

# The `del` statement can be used to delete an item at a given index. 
# Also, it can be used to remove slices from a list.

my_list = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# deleting the third item
del my_list[2]

# Output: [1, 2, 4, 5, 6, 7, 8, 9]
print(my_list)

# deleting items from 2nd to 4th
del my_list[1:4]

# Output: [1, 6, 7, 8, 9]
print(my_list)

# deleting all elements
del my_list[:]

# Output: []
print(my_list)
````

````python
# Error: my_var is not defined
print(my_var)
````

````python
# Error: my_tuple is not defined
print(my_tuple)
````

````python
# Error: my_dict is not defined
print(my_dict)
````

````python
# Example 4: Remove a key:value pair from a dictionary

person = { 'name': 'Arthur',
  'age': 33,
  'profession': 'Programmer'
}

del person['profession']
print(person)   # Output: {'name': 'Arthur', 'age': 33}
````

You can't delete items of tuples and strings. It's because tuples and strings are immutables; objects that can't be changed after its creation.


````python
my_tuple = (1, 2, 3)

del my_tuple[1]    # Error: 'tuple' object doesn't support item deletion
````

However, you can delete an entire tuple or string.


````python
my_tuple = (1, 2, 3)

del my_tuple    # deleting tuple
my_tuple
````

### **Python `del` Statement: Concepts and Theory**

The **`del`** statement in Python is a powerful built-in keyword used to **delete** objects, variables, elements from lists, dictionaries, or even entire slices of data. It does not return any value and directly affects the object in memory. The **`del`** statement can be used for deleting variables, list elements, or keys from dictionaries, as well as deleting entire objects or slices of sequences.

In this guide, we will explore the **`del`** statement in detail, covering its syntax, behavior, use cases, and performance considerations.

---

### **1. Syntax and Usage of `del`**

The general syntax for the **`del`** statement is:

```python
del object
```

The **`object`** can be a variable, list item, dictionary key, slice of a sequence, or even the entire object. Here are some variations of the **`del`** statement:

- **Deleting a variable**:

  ```python
  del variable_name
  ```

- **Deleting an item from a list by index**:

  ```python
  del list[index]
  ```

- **Deleting multiple items (slice) from a list**:

  ```python
  del list[start:end]
  ```

- **Deleting a key-value pair from a dictionary**:

  ```python
  del dictionary[key]
  ```

- **Deleting an entire list or object**:
  ```python
  del list
  ```

---

### **2. Deleting Variables**

The **`del`** statement can be used to delete variables. After deletion, the variable no longer exists in the current scope.

#### **Example** (deleting a variable):

```python
x = 10
del x  # Deleting variable x
# print(x)  # Raises NameError: name 'x' is not defined
```

In this example, the variable **`x`** is deleted, and trying to access **`x`** after deletion will raise a **`NameError`**, as **`x`** is no longer defined.

#### **Use Case**:

You may want to free memory or remove unnecessary variables, especially in larger programs where resources are critical, or when working with dynamic data.

---

### **3. Deleting List Elements**

You can use **`del`** to delete individual elements from a list using their index, or delete slices of the list to remove a range of elements.

#### **3.1 Deleting a Single Element by Index**

You can delete an element from a list by specifying its index:

- **Example** (deleting an element by index):

```python
my_list = [1, 2, 3, 4]
del my_list[1]  # Deleting element at index 1
print(my_list)  # Output: [1, 3, 4]
```

In this example, the element **`2`** at index **`1`** is deleted, and the resulting list contains **`[1, 3, 4]**.

#### **3.2 Deleting a Slice of a List**

You can also delete multiple elements in a list by specifying a slice:

- **Example** (deleting a slice of elements):

```python
my_list = [1, 2, 3, 4, 5]
del my_list[1:4]  # Deleting elements from index 1 to index 3 (exclusive)
print(my_list)  # Output: [1, 5]
```

In this example, the slice **`[2, 3, 4]`** is deleted from the list, leaving **`[1, 5]`**.

#### **Use Case**:

Deleting list elements with **`del`** is useful when you want to remove elements dynamically without creating a new list or when you need to remove specific items from the list based on their index or range.

---

### **4. Deleting Keys from Dictionaries**

The **`del`** statement can also be used to remove a key-value pair from a dictionary. If the key is not found, a **`KeyError`** is raised.

#### **Example** (deleting a dictionary key):

```python
my_dict = {'a': 1, 'b': 2, 'c': 3}
del my_dict['b']  # Deleting key 'b'
print(my_dict)  # Output: {'a': 1, 'c': 3}
```

In this example, the key **`'b'`** and its associated value **`2`** are deleted, leaving **`{'a': 1, 'c': 3}`**.

#### **Use Case**:

You can use **`del`** to remove keys from dictionaries when certain conditions are met (e.g., when processing data), or when you need to clean up the dictionary.

---

### **5. Deleting Entire Objects**

The **`del`** statement can be used to delete entire objects or variables. This removes the reference to the object, and if no other references to the object exist, the object will be garbage collected.

#### **Example** (deleting an entire object):

```python
my_list = [1, 2, 3, 4]
del my_list  # Deletes the entire list object
# print(my_list)  # Raises NameError: name 'my_list' is not defined
```

Here, the entire list **`my_list`** is deleted. After deletion, trying to access **`my_list`** results in a **`NameError`**, as the list is no longer defined.

#### **Use Case**:

This is useful for clearing memory or explicitly removing references to an object when you no longer need it, especially in long-running programs or when working with large datasets.

---

### **6. Deleting Slices of Sequences**

You can use the **`del`** statement to remove slices from other sequences (like **strings** or **tuples**), but it only works on mutable sequences (like lists). Attempting to delete a slice from an immutable sequence like a **string** or **tuple** will raise a **`TypeError`**.

#### **Example** (deleting a slice from a list):

```python
my_list = [1, 2, 3, 4, 5]
del my_list[2:4]  # Deleting slice from index 2 to 3 (inclusive)
print(my_list)  # Output: [1, 2, 5]
```

In this case, the slice **`[3, 4]`** is deleted, leaving **`[1, 2, 5]`**.

#### **Example** (attempting to delete a slice from an immutable sequence like a string):

```python
my_string = "Hello"
del my_string[1:4]  # Raises TypeError: 'str' object doesn't support item deletion
```

Since **`my_string`** is a string, which is immutable, attempting to delete a slice from it raises an error.

---

### **7. Performance Considerations**

#### **7.1 Time Complexity**

- **Deleting a variable**: Deleting a variable using **`del`** is an **O(1)** operation, as it simply removes the reference to the variable.
- **Deleting list elements**: Deleting a single element or a slice of elements in a list involves shifting elements, which is generally an **O(n)** operation where **n** is the number of elements being shifted.
- **Deleting a dictionary key**: Deleting a key-value pair from a dictionary is generally an **O(1)** operation, as dictionaries use a hash table for fast lookups.

#### **7.2 Memory Considerations**

The **`del`** statement removes references to objects in memory, which helps reduce memory usage. However, the object will only be **garbage collected** if no other references to the object exist. If other references to the object are still in use, the object will persist in memory.

---

### **8. Common Errors and Edge Cases**

#### **8.1 `NameError` When Deleting a Non-Existent Variable**

If you attempt to delete a variable that has not been defined, Python will raise a **`NameError`**.

- **Example**:

```python
del non_existent_var  # Raises NameError: name 'non_existent_var' is not defined
```

#### **8.2 `KeyError` When Deleting a Non-Existent Key in a Dictionary**

If you try to delete a key that does not exist in a dictionary, Python will raise a **`KeyError`**.

- **Example**:

```python
my_dict = {'a': 1, 'b': 2}
del my_dict['c']  # Raises KeyError: 'c'
```

#### **8.3 `TypeError` for Deleting from Immutable Sequences**

You cannot use **`del`** to delete elements or slices from immutable sequences such as **strings** or **tuples**.

- **Example** (attempting to delete from a string):

```python
my_string = "hello"
del my_string[1]  # Raises TypeError: 'str' object doesn't support item deletion
```

---

### **9. Use Cases for `del`**

The **`del`** statement is a versatile tool with many practical applications in Python programming. Here are some common use cases:

- **Clearing variables and freeing memory**: When working with large datasets or in long-running applications, you can use **`del`** to remove variables that are no longer needed and free memory.
- **Deleting items dynamically from a list or dictionary**: **`del`** is useful when you need to remove specific elements or keys dynamically, such as in data processing tasks.

- **Deleting slices from lists**: **`del`** allows you to remove a range of elements from a list efficiently, which can be helpful in certain algorithms and manipulations.

---

### **Conclusion**

The **`del`** statement in Python is a powerful tool for deleting variables, list elements, dictionary keys, and even entire objects. It helps manage memory efficiently by removing references to objects. It also provides a way to remove specific elements or slices from mutable sequences like lists.

While it is a very useful statement, it requires careful handling—especially when dealing with mutable objects—because it can result in unexpected side effects if you're not careful with the references.
