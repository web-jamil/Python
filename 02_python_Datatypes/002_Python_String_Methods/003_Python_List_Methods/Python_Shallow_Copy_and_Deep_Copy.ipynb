# Python Shallow Copy and Deep Copy

In this class, you’ll learn about shallow copy and deep copy in Python with the help of examples.


## Copy an Object in Python

In Python, we use **`=`** operator to create a copy of an object. You may think that this creates a new object; it doesn't. It only creates a new variable that shares the reference of the original object.

Let's take an example where we create a list named **`old_list`** and pass an object reference to **`new_list`** using **`=`** operator.


````python
# Example 1: Copy using = operator

old_list = [[1, 2, 3], [4, 5, 6], [7, 8, 'a']]
new_list = old_list

new_list[2][2] = 9

print('Old List:', old_list)
print('ID of Old List:', id(old_list))

print('New List:', new_list)
print('ID of New List:', id(new_list))
````

As you can see from the output both variables **`old_list`** and **`new_list`** shares the same id i.e **`140673303268168`**.

So, if you want to modify any values in **`new_list`** or **`old_list`**, the change is visible in both.


Essentially, sometimes you may want to have the original values unchanged and only modify the new values or vice versa. In Python, there are two ways to create copies:

1. Shallow Copy
2. Deep Copy

To make these copy work, we use the **`copy`** module.


## Copy Module

We use the **`copy()`** module of Python for shallow and deep copy operations. Suppose, you need to copy the compound list say **`x`**. For example:

```python
import copy
copy.copy(x)
copy.deepcopy(x)
```

Here, the **`copy()`** return a shallow copy of **`x`**. Similarly, **`deepcopy()`** return a deep copy of **`x`**.


## Shallow Copy

A shallow copy creates a new object which stores the reference of the original elements.

So, a shallow copy doesn't create a copy of nested objects, instead it just copies the reference of nested objects. This means, a copy process does not recurse or create copies of nested objects itself.


````python
# Example 2: Create a copy using shallow copy

import copy

old_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
new_list = copy.copy(old_list)

print("Old list:", old_list)
print("New list:", new_list)
````

In above program, we created a nested list and then shallow copy it using **`copy()`** method.

This means it will create new and independent object with same content. To verify this, we print the both **`old_list`** and **`new_list`**.

To confirm that **`new_list`** is different from **`old_list`**, we try to add new nested object to original and check it.


````python
# Example 3: Adding [4, 4, 4] to old_list, using shallow copy

import copy

old_list = [[1, 1, 1], [2, 2, 2], [3, 3, 3]]
new_list = copy.copy(old_list)

old_list.append([4, 4, 4])

print("Old list:", old_list)
print("New list:", new_list)
````

In the above program, we created a shallow copy of **`old_list`**. The **`new_list`** contains references to original nested objects stored in **`old_list`**. Then we add the new list i.e **`[4, 4, 4]`** into **`old_list`**. This new sublist was not copied in **`new_list`**.

However, when you change any nested objects in **`old_list`**, the changes appear in **`new_list`**.


````python
# Example 4: Adding new nested object using Shallow copy

import copy

old_list = [[1, 1, 1], [2, 2, 2], [3, 3, 3]]
new_list = copy.copy(old_list)

old_list[1][1] = 'AA'

print("Old list:", old_list)
print("New list:", new_list)
````

In the above program, we made changes to **`old_list`** i.e **`old_list[1][1] = 'AA'`**. Both sublists of **`old_list`** and **`new_list`** at index **`[1][1]`** were modified. This is because, both lists share the reference of same nested objects.


## Deep Copy

A deep copy creates a new object and recursively adds the copies of nested objects present in the original elements.

Let’s continue with example 2. However, we are going to create deep copy using **`deepcopy()`** function present in copy module. The deep copy creates independent **`copy`** of original object and all its nested objects.


````python
# Example 5: Copying a list using deepcopy()

import copy

old_list = [[1, 1, 1], [2, 2, 2], [3, 3, 3]]
new_list = copy.deepcopy(old_list)

print("Old list:", old_list)
print("New list:", new_list)
````

In the above program, we use **`deepcopy()`** function to create copy which looks similar.

However, if you make changes to any nested objects in original object **`old_list`**, you’ll see no changes to the copy **`new_list`**.


````python
# Example 6: Adding a new nested object in the list using Deep copy

import copy

old_list = [[1, 1, 1], [2, 2, 2], [3, 3, 3]]
new_list = copy.deepcopy(old_list)

old_list[1][0] = 'BB'

print("Old list:", old_list)
print("New list:", new_list)
````

In the above program, when we assign a new value to **`old_list`**, we can see only the **`old_list`** is modified. This means, both the **`old_list`** and the **`new_list`** are independent. This is because the **`old_list`** was recursively copied, which is true for all its nested objects.


### **Shallow Copy and Deep Copy in Python: Concepts and Theory**

In Python, **shallow copy** and **deep copy** refer to two different ways of copying objects. The key difference between the two lies in how the objects within the copied structure (such as lists or dictionaries) are handled—whether they are copied independently or simply referenced. Both are crucial concepts when working with complex data structures in Python, especially when dealing with mutable objects.

Let's break down these two concepts and explore their differences, use cases, and practical examples.

---

### **1. Shallow Copy**

A **shallow copy** creates a new object, but it does **not** create copies of the objects contained within the original object. Instead, it copies references to the original objects. As a result, if the copied object contains references to mutable objects (like lists, dictionaries, or sets), modifications to those mutable objects will affect both the original and the shallow copy.

In simpler terms, a **shallow copy** creates a new object, but it does not recursively copy all the objects inside it. It only copies the references to those objects.

#### **Key Points About Shallow Copy:**

- Creates a new object but doesn't recursively copy objects inside the original.
- Mutable objects within the copied structure still reference the same objects as in the original.
- Useful when you want to create a new object but share the references to the inner objects.

#### **How to Perform a Shallow Copy:**

You can create a shallow copy using the following methods:

1. **Using `copy()` method** (for lists, dictionaries, and sets).
2. **Using `copy` module** with the `copy()` function.

---

#### **Shallow Copy Example Using `copy()` Method**

```python
original_list = [1, 2, [3, 4]]
shallow_copy = original_list.copy()

# Modify the inner list of the shallow copy
shallow_copy[2][0] = 99

print("Original List:", original_list)   # Output: [1, 2, [99, 4]]
print("Shallow Copy:", shallow_copy)     # Output: [1, 2, [99, 4]]
```

Here, **`original_list`** and **`shallow_copy`** are distinct objects, but both share the same reference to the inner list **`[3, 4]`**. Thus, modifying the inner list in **`shallow_copy`** affects **`original_list`**.

#### **Shallow Copy Example Using `copy()` from `copy` Module**

```python
import copy

original_list = [1, 2, [3, 4]]
shallow_copy = copy.copy(original_list)

shallow_copy[2][0] = 99

print("Original List:", original_list)   # Output: [1, 2, [99, 4]]
print("Shallow Copy:", shallow_copy)     # Output: [1, 2, [99, 4]]
```

The **`copy.copy()`** method also creates a shallow copy of the list, and the inner list is still shared between both the original and the copied list.

---

### **2. Deep Copy**

A **deep copy** creates a new object and recursively copies all objects found within the original object. This means that **not only the top-level object is copied**, but all objects contained within it (including any nested objects) are also copied independently. Therefore, changes to the deep copy will not affect the original object and vice versa.

In simpler terms, a **deep copy** duplicates the original object and all of its contents, creating entirely independent copies.

#### **Key Points About Deep Copy:**

- Creates a new object, and recursively copies all objects within the original object.
- The copied object is completely independent of the original object.
- Useful when you need to modify the copy without affecting the original.

#### **How to Perform a Deep Copy:**

To perform a deep copy in Python, you can use the **`deepcopy()`** function from the **`copy`** module.

---

#### **Deep Copy Example Using `deepcopy()` from `copy` Module**

```python
import copy

original_list = [1, 2, [3, 4]]
deep_copy = copy.deepcopy(original_list)

# Modify the inner list of the deep copy
deep_copy[2][0] = 99

print("Original List:", original_list)   # Output: [1, 2, [3, 4]]
print("Deep Copy:", deep_copy)           # Output: [1, 2, [99, 4]]
```

In this example, **`original_list`** and **`deep_copy`** are completely independent. Modifying the inner list in **`deep_copy`** does **not** affect **`original_list`**, demonstrating the true independence provided by a deep copy.

---

### **3. Key Differences Between Shallow Copy and Deep Copy**

| **Aspect**          | **Shallow Copy**                                                                         | **Deep Copy**                                                                            |
| ------------------- | ---------------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------------- |
| **Object Creation** | Creates a new object, but **does not recursively copy** objects inside it.               | Creates a new object, and recursively copies all objects inside it.                      |
| **Inner Objects**   | Inner objects (e.g., lists, dictionaries) are shared between the original and the copy.  | Inner objects are copied independently, so changes to them don't affect the original.    |
| **Use Case**        | Used when you want a new container but are fine with shared references to inner objects. | Used when you need complete independence from the original object and its contents.      |
| **Performance**     | Faster than deep copy because it only copies the references to inner objects.            | Slower than shallow copy because it recursively copies every object inside the original. |
| **Memory**          | Uses less memory than deep copy because it doesn't create copies of nested objects.      | Uses more memory because it copies every object, including nested ones.                  |
| **Python Method**   | **`copy()`** for lists, dictionaries, sets or **`copy.copy()`** from the `copy` module.  | **`copy.deepcopy()`** from the `copy` module.                                            |

---

### **4. Use Cases for Shallow and Deep Copy**

#### **4.1 Shallow Copy Use Case**

- **When you want to create a new container** that contains references to the same objects as the original.
- **Memory efficiency**: If you have a large data structure with nested objects, and you don't need to change the inner objects, a shallow copy will be more memory-efficient.

For example, when you have a list of immutable objects (e.g., strings or numbers), a shallow copy is often sufficient because modifying one of the objects will not affect the others.

#### **4.2 Deep Copy Use Case**

- **When you want to create a completely independent copy** of the original object, including all nested objects.
- **Modifying the deep copy** will not affect the original object or any of its inner objects.

A deep copy is typically needed when you're working with mutable nested data structures, such as lists of lists, dictionaries of lists, etc., and you need to modify the copy without affecting the original structure.

---

### **5. Performance Considerations**

- **Shallow Copy**:
  - **Time Complexity**: Creating a shallow copy of a list or dictionary takes **O(n)** time, where **n** is the number of elements in the list. This is because only the top-level object is copied, and each element is simply referenced.
  - **Space Complexity**: It requires **O(n)** space for the new object, plus the space for the objects it references (which are shared between the original and the copy).
- **Deep Copy**:
  - **Time Complexity**: A deep copy takes **O(n)** time for each level of the object. Since it needs to recursively copy all inner objects, its performance can degrade for large or deeply nested structures.
  - **Space Complexity**: It requires **O(n)** space for the new object, plus the space for all recursively copied objects.

#### **Performance Comparison Example:**

```python
import copy
import time

# Original list with nested lists
original = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Measure shallow copy performance
start = time.time()
shallow_copy = copy.copy(original)
end = time.time()
print("Shallow Copy Time:", end - start)

# Measure deep copy performance
start = time.time()
deep_copy = copy.deepcopy(original)
end = time.time()
print("Deep Copy Time:", end - start)
```

In this example, you'll notice that the **deep copy** takes longer than the **shallow copy**, particularly as the data structure becomes larger or more complex.

---

### **6. Common Pitfalls and Issues**

1. **Unintended Changes**: When working with a shallow copy, you may accidentally modify the original object’s inner data if the inner objects are mutable. Always be mindful of this behavior.

   - **Example** (unintended change in a shallow copy):

   ```python
   original = [1, 2, [3, 4]]
   shallow = original.copy()
   shallow[2][0] = 99  # This will also affect the original list.
   print(original)  # Output: [1, 2, [99, 4]]
   ```

2. **Misuse of Deep Copy**: Using deep copy unnecessarily can result in performance issues when you are working with very large objects or when the structure is not deeply nested.

3. **Circular References**: Both shallow and deep copy methods may have issues with circular references (i.e., objects referencing themselves). In such cases, **`deepcopy()`** handles this correctly, but shallow copies can create unintended issues.

---

### **Conclusion**

- **Shallow Copy**: Best used when you need a new object but are fine with shared references to the inner objects. It is faster and more memory-efficient but can lead to unexpected modifications in nested objects.
- **Deep Copy**: Best used when you need a completely independent copy of the original object, including all nested objects. It is slower and uses more memory but provides complete independence between the original and copied objects.

Understanding when to use shallow or deep copying is essential for managing mutable objects and nested data structures in Python, particularly when you need to manipulate data without affecting the original source.
