The **Ford-Fulkerson Algorithm** is a classic and fundamental algorithm in graph theory used to solve the **maximum flow problem**. It determines the maximum amount of flow that can be sent from a **source node (S)** to a **sink node (T)** in a flow network while respecting the capacity constraints of the edges. Below is a detailed explanation of the Ford-Fulkerson Algorithm, covering everything from basic concepts to implementation and optimization.

---

### 1. **Basic Concepts**

#### a) **Flow Network**

- A **directed graph** where each edge has:
  - **Capacity (C)**: The maximum flow the edge can carry.
  - **Flow (F)**: The current flow through the edge.
- Contains:
  - **Source (S)**: The node where the flow originates.
  - **Sink (T)**: The node where the flow ends.
- **Constraints**:
  - **Capacity Constraint**: Flow on an edge cannot exceed its capacity (F ≤ C).
  - **Flow Conservation**: The flow entering a node (except S and T) must equal the flow leaving it.

#### b) **Residual Graph**

- A graph that represents the remaining capacity of edges after a flow has been sent.
- For each edge in the original graph:
  - A **forward edge** with capacity = original capacity - current flow.
  - A **backward edge** with capacity = current flow (to allow reversing flow).

#### c) **Augmenting Path**

- A path from the source to the sink in the residual graph where additional flow can be sent.
- The flow along this path is limited by the **minimum residual capacity** (bottleneck) on the path.

#### d) **Maximum Flow**

- The maximum amount of flow that can be sent from the source to the sink without violating capacity constraints.

---

### 2. **Ford-Fulkerson Algorithm**

#### a) **Steps**

1. Initialize the flow in all edges to 0.
2. While there is an augmenting path in the residual graph:
   - Find the minimum residual capacity (bottleneck) along the path.
   - Update the flow along the path.
   - Update the residual graph.
3. Terminate when no more augmenting paths exist.

#### b) **Pseudocode**

```
Ford-Fulkerson(Graph G, Source S, Sink T):
    Initialize flow F = 0
    Create residual graph R from G
    While there is an augmenting path P in R:
        Find bottleneck capacity b in P
        Update flow F += b
        Update residual capacities in R
    Return F
```

#### c) **Example**

Consider the following flow network:

```
    S --16--> A --12--> T
     \       /         /
     13    10        20
       \   /         /
        B --14--> C
```

- Step 1: Initialize flow to 0.
- Step 2: Find an augmenting path (e.g., S → A → T) with bottleneck capacity 12.
- Step 3: Update flow and residual graph.
- Repeat until no more augmenting paths exist.

### 3. **Implementation**

#### Python Implementation

```python
class Graph:
    def __init__(self, graph):
        self.graph = graph  # Residual graph
        self.ROW = len(graph)

    def BFS(self, s, t, parent):
        visited = [False] * self.ROW
        queue = []
        queue.append(s)
        visited[s] = True

        while queue:
            u = queue.pop(0)
            for ind, val in enumerate(self.graph[u]):
                if visited[ind] == False and val > 0:
                    queue.append(ind)
                    visited[ind] = True
                    parent[ind] = u
                    if ind == t:
                        return True
        return False

    def FordFulkerson(self, source, sink):
        parent = [-1] * self.ROW
        max_flow = 0

        while self.BFS(source, sink, parent):
            path_flow = float("Inf")
            s = sink
            while s != source:
                path_flow = min(path_flow, self.graph[parent[s]][s])
                s = parent[s]

            max_flow += path_flow

            v = sink
            while v != source:
                u = parent[v]
                self.graph[u][v] -= path_flow
                self.graph[v][u] += path_flow
                v = parent[v]

        return max_flow

# Example usage
graph = [
    [0, 16, 13, 0, 0, 0],
    [0, 0, 10, 12, 0, 0],
    [0, 4, 0, 0, 14, 0],
    [0, 0, 9, 0, 0, 20],
    [0, 0, 0, 7, 0, 4],
    [0, 0, 0, 0, 0, 0]
]

g = Graph(graph)
source = 0
sink = 5
print("Max Flow:", g.FordFulkerson(source, sink))
```

### 4. **Advanced Concepts**

#### a) **Time Complexity**

- **Worst Case**: O(E \* max_flow), where E is the number of edges and max_flow is the maximum flow value.
- **Drawback**: The algorithm can be inefficient if the capacities are large or the augmenting paths are not chosen wisely.

#### b) **Edmonds-Karp Optimization**

- Uses **BFS** to find the shortest augmenting path in each iteration.
- **Time Complexity**: O(V \* E²), where V is the number of vertices and E is the number of edges.

#### c) **Max-Flow Min-Cut Theorem**

- States that the maximum flow in a network is equal to the capacity of the minimum cut.
- A **cut** is a partition of the graph into two disjoint sets (S and T) such that the source is in S and the sink is in T.

#### d) **Applications**

- **Traffic Routing**: Optimize traffic flow in road networks.
- **Data Networks**: Maximize data transfer in computer networks.
- **Bipartite Matching**: Solve matching problems using flow algorithms.
