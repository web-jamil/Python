# Python String `isdecimal()`

The string **`isdecimal()`** method returns **`True`** if all characters in the string are decimal characters. If not, it returns **`False`**.

**Syntax**:

```python
string.isdecimal()
```


## `isdecimal()` Parameters

The **`isdecimal()`** method doesn't take any parameters.


## Return Value from `isdecimal()`

The **`isdecimal()`** returns:

- **`True`** if all characters in the string are decimal characters.
- **`False`** if at least one character is not decimal characters.


````python
# Example 1: Working of isdecimal()

s = "28212"
print(s.isdecimal())

# contains alphabets
s = "32ladk3"
print(s.isdecimal())

# contains alphabets and spaces
s = "Mo3 nicaG el l22er"
print(s.isdecimal())
````

**Explanation:**

The superscript and subscripts are considered digit characters but not decimals. If the string contains these characters (usually written using unicode), **`isdecimal()`** returns **`False`**.

Similarly, roman numerals, currency numerators and fractions are considered numeric numbers (usually written using unicode) but not decimals. The **`isdecimal()`** also returns **`False`** in this case.

There are two methods **`isdigit()`** and **`isnumeric()`** that checks whether the string contains digit characters and numeric characters respectively.

Learn more about **[isdigit()](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/014_Python_String_isdigit%28%29.ipynb)** and **[isnumeric()](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String_Methods/017_Python_String_isnumeric%28%29.ipynb)** methods.


````python
# Example 2: String Containing digits and Numeric Characters

s = '23455'
print(s.isdecimal())

#s = '²3455'
s = '\u00B23455'
print(s.isdecimal())

# s = '½'
s = '\u00BD'
print(s.isdecimal())
````

The `isdecimal()` method in Python is a built-in string method that checks if all characters in a string are decimal characters (0-9). It returns a boolean value (`True` or `False`). This guide provides a comprehensive overview of the `isdecimal()` function, covering its syntax, usage, related methods, and advanced applications.

## Basic Overview

### Syntax

```python
str.isdecimal()
```

### Returns

- **`True`**: if all characters in the string are decimal characters and there is at least one character.
- **`False`**: if the string contains any non-decimal characters (such as letters, punctuation, spaces, or special characters) or is empty.

### Basic Examples

#### 1. Simple Usage

Here are some straightforward examples of using `isdecimal()`:

```python
print("12345".isdecimal())  # Output: True (all characters are decimals)
print("6789".isdecimal())    # Output: True
print("12.34".isdecimal())   # Output: False (contains a decimal point)
print("abc".isdecimal())      # Output: False (contains letters)
print("123abc".isdecimal())   # Output: False (contains letters)
print("".isdecimal())         # Output: False (empty string)
```

#### 2. Unicode Support

`isdecimal()` can recognize certain Unicode characters that represent decimal digits. For example:

```python
print("²".isdecimal())      # Output: True (superscript two, Unicode decimal)
print("3.14".isdecimal())    # Output: False (contains a decimal point)
print("𝟏𝟐𝟑".isdecimal())      # Output: True (full-width digits)
```

### Characteristics

- **Strict on Decimals**: Unlike `isdigit()`, which can accept more forms (like fractions), `isdecimal()` strictly checks for decimal characters (0-9).
- **Checks All Characters**: If even one character is not a decimal digit, it will return `False`.

### Related String Methods

1. **`isdigit()`**: Checks if all characters are digits. This includes decimal characters and other numeric forms (like fractions).

   ```python
   print("123".isdigit())       # Output: True
   print("3.14".isdigit())      # Output: False (contains a decimal point)
   print("²".isdigit())         # Output: True (superscript two)
   ```

2. **`isnumeric()`**: Checks if all characters are numeric, including integers, fractions, and certain numeric symbols.

   ```python
   print("123".isnumeric())    # Output: True
   print("²".isnumeric())       # Output: True (superscript)
   print("⅔".isnumeric())       # Output: True (fraction)
   ```

3. **`isalpha()`**: Checks if all characters are alphabetic (letters).

   ```python
   print("abc".isalpha())      # Output: True
   print("abc123".isalpha())   # Output: False
   ```

4. **`isalnum()`**: Checks if all characters are alphanumeric (letters and digits).

   ```python
   print("abc123".isalnum())   # Output: True
   print("abc 123".isalnum())   # Output: False (contains space)
   ```

5. **`isidentifier()`**: Checks if the string is a valid identifier (e.g., for variable names).
   ```python
   print("variable_name".isidentifier())  # Output: True
   print("1variable".isidentifier())      # Output: False (cannot start with a digit)
   ```

### Advanced Applications

#### 1. Validation of Numeric Input

You can use `isdecimal()` to validate user inputs where only decimal numbers are allowed, such as ages or quantities:

```python
age = input("Enter your age: ")
if age.isdecimal():
    print("Valid age.")
else:
    print("Invalid age. Please enter whole numbers only.")
```

#### 2. Filtering Decimal Strings from a List

You can filter elements in a list to retain only those that are entirely decimal:

```python
data = ["123", "abc", "45.67", "890", "12a34"]
decimal_numbers = [s for s in data if s.isdecimal()]
print(decimal_numbers)  # Output: ['123', '890']
```

#### 3. Counting Decimal Characters

You can count how many strings in a list are decimal:

```python
strings = ["123", "abc", "456", "78.9"]
decimal_count = sum(1 for s in strings if s.isdecimal())
print
```

### Python `isdecimal()` Function

The `isdecimal()` method is a built-in string method in Python. It checks whether all the characters in a string are decimal characters. Unlike the `isdigit()` method, which recognizes a broader range of numeric characters (including digits from other numeral systems), `isdecimal()` specifically checks if the characters are "decimal digits" in the sense of the Unicode character set.

### Syntax:

```python
str.isdecimal()
```

### Parameters:

- The `isdecimal()` method does not take any parameters. It works only on the string it is called upon.

### Return Value:

- **`True`**: If all the characters in the string are decimal digits (i.e., characters in the range `0-9`).
- **`False`**: If the string contains any non-decimal characters, such as letters, special symbols, spaces, or other non-decimal characters, or if the string is empty.

### Key Differences Between `isdigit()` and `isdecimal()`:

- **`isdecimal()`**: Only returns `True` if the characters are decimal digits in the range of `0-9`.
- **`isdigit()`**: Returns `True` for any character that is considered a digit in Unicode, including numbers from other languages or numeral systems (e.g., Roman numerals or Arabic numerals).

### Detailed Explanation:

- **Decimal Digits**: In Unicode, "decimal digits" refers to characters in the range `0-9` which are typically used for numbers. Other number-like characters in different languages, such as digits in other numeral systems (like Arabic or Chinese numerals), are **not** considered decimal digits by `isdecimal()`.

### Basic Examples:

#### 1. Simple Case with Decimal Digits:

```python
text = "12345"
result = text.isdecimal()
print(result)  # Output: True
```

- The string `"12345"` consists only of decimal digits (standard `0-9`), so the method returns `True`.

#### 2. String with Letters:

```python
text = "123abc"
result = text.isdecimal()
print(result)  # Output: False
```

- The string `"123abc"` contains letters (`abc`), which are not decimal digits, so `False` is returned.

#### 3. String with Space:

```python
text = "123 45"
result = text.isdecimal()
print(result)  # Output: False
```

- The presence of a space makes the string invalid for `isdecimal()`, so it returns `False`.

#### 4. Empty String:

```python
text = ""
result = text.isdecimal()
print(result)  # Output: False
```

- An empty string does not contain any characters, so it returns `False`.

#### 5. String with Special Characters:

```python
text = "123@45"
result = text.isdecimal()
print(result)  # Output: False
```

- The `@` character is not a decimal digit, so the method returns `False`.

#### 6. String with Decimal Point:

```python
text = "123.45"
result = text.isdecimal()
print(result)  # Output: False
```

- The presence of a decimal point (`.`) makes the string invalid for `isdecimal()`, and it returns `False`.

#### 7. String with Unicode Decimal Digits:

```python
text = "٣٤٥"  # Arabic numerals for 345
result = text.isdecimal()
print(result)  # Output: True
```

- The string `"٣٤٥"` consists of Arabic decimal digits, which are recognized by `isdecimal()`, so it returns `True`.

#### 8. String with Other Numeral Systems (Not Decimal):

```python
text = "Ⅻ"  # Roman numeral for 12
result = text.isdecimal()
print(result)  # Output: False
```

- Roman numerals like `"Ⅻ"` are not considered decimal digits by `isdecimal()`, so it returns `False`.

### Edge Cases:

1. **Empty String**: Returns `False` because there are no characters to check.

   ```python
   text = ""
   result = text.isdecimal()
   print(result)  # Output: False
   ```

2. **Decimal Numbers with a Dot**: The method will return `False` for any string that includes a decimal point (`.`).

   ```python
   text = "123.45"
   result = text.isdecimal()
   print(result)  # Output: False
   ```

3. **String with Special Symbols**: Any non-decimal character will cause the method to return `False`.

   ```python
   text = "123@456"
   result = text.isdecimal()
   print(result)  # Output: False
   ```

4. **Unicode Digits**: Strings with digits from other languages or numeral systems that are **not** considered decimal digits will return `False`.

   ```python
   text = "Ⅻ"  # Roman numerals for 12
   result = text.isdecimal()
   print(result)  # Output: False
   ```

5. **Unicode Decimal Digits**: Arabic numerals or other numeral systems that represent numbers will return `True`.
   ```python
   text = "٣٤٥"  # Arabic numerals for 345
   result = text.isdecimal()
   print(result)  # Output: True
   ```

### Common Use Cases:

1. **Validate User Input**:
   You can use `isdecimal()` to validate user input and check whether the input is a number (without any symbols, letters, or decimal points).

   ```python
   user_input = input("Enter a number: ")
   if user_input.isdecimal():
       print("Valid number.")
   else:
       print("Invalid input! Please enter a valid number.")
   ```

2. **Filter Only Decimal Digits from List**:
   You can filter out strings that contain only decimal digits from a list of strings.

   ```python
   values = ["123", "abc", "456", "12.34", "789"]
   digits_only = [value for value in values if value.isdecimal()]
   print(digits_only)  # Output: ['123', '456', '789']
   ```

3. **Validate Phone Numbers or Zip Codes**:
   If you need to validate a phone number or a zip code, you can check if it contains only decimal digits.

   ```python
   phone_number = "1234567890"
   if phone_number.isdecimal():
       print("Valid phone number.")
   else:
       print("Invalid phone number!")
   ```

4. **Extracting Decimal Numbers from Text**:
   You might want to extract decimal-only numbers from a string or text, which can be done using list comprehension.
   ```python
   text = "I have 100 apples and 200 oranges."
   numbers = [word for word in text.split() if word.isdecimal()]
   print(numbers)  # Output: ['100', '200']
   ```

### Performance:

- **Time Complexity**: O(n), where `n` is the length of the string, since each character in the string is checked to ensure it's a decimal digit.
- **Space Complexity**: O(1), as no additional space is used besides the input string.

### Summary:

The `isdecimal()` method is useful when you need to verify that a string contains only decimal digits. This method is more restrictive than `isdigit()` because it specifically checks for decimal digits in the range `0-9`. It's commonly used for validating user input, filtering decimal-only numbers, and ensuring that strings represent valid numeric data in certain contexts.
