# Python `if` statement

In control statements, The **`if`** statement is the simplest form. It takes a condition and evaluates to either **`True`** or **`False`**.

If the condition is **`True`**, then the **`True`** block of code will be executed, and if the condition is **`False`**, then the block of code is skipped, and The controller moves to the next line.


### Syntax:

```python
if(condition):
    statement 1
    statement 2
    statement n
```

- An **`if`** statement should be preceded by the keyword **`if`** and ended with a colon **`:`**. The condition of an if statement can be written with or without round brackets **`()`**.

- In Python, the body of the **`if`** statement is indicated by the indentation. The body starts with an indentation and the first unindented line marks the end.

- Python interprets non-zero values as **`True`**. **`None`** and **`0`** are interpreted as **`False`**.

**Example: of `if` statement:**

```python
>>> grade = 70
>>> if grade >= 65:
>>>     print("Passing grade")

Passing grade
```

With this code, we have the variable grade and are giving it the integer value of 70. We are then using the **`if`** statement to evaluate whether or not the variable grade is **greater than or equal `>=`** to 65. If it does meet this **condition**, we are telling the program to print out the string Passing grade.

Again, if we change the grade to 50 or < 65 number, we will receive no output.


````python
# Example 1:
grade = 70
if grade >= 65:
    print("Passing grade")
````

````python
# Example 2: If the number is positive, we print an appropriate message

num = 3
if (num > 0):  # if condition is TRUE: enter the body of if
    print(num, "is a positive number.")
print("This is always printed.")

num = -1
if num > 0:  # if condition is FALSE: do not enter the body of if
    print(num, "is a negative number.")
print("This is also always printed.")
````

**Explanation**:

In the above example, **`num > 0`** is the test expression.

The body of **`if`** is executed only if this evaluates to **`True`**.

When the variable **`num`** is equal to 3, test expression is true and statements inside the body of **`if`** are executed.

If the variable **`num`** is equal to -1, test expression is false and statements inside the body of **`if`** are skipped.

The **[print()](https://github.com/milaan9/04_Python_Functions/blob/main/002_Python_Functions_Built_in/051_Python_print%28%29.ipynb)** statement falls outside of the **`if`** block (unindented). Hence, it is executed regardless of the test expression.


````python
# Example 3: Calculate the square of a number if it greater than 6

number = 9
if number > 6:
    # Calculate square
    print(number * number)
print('Next lines of code')
````

````python
# Example 4: Check if num1 is less than num2

num1, num2 = 6, 9
if(num1 < num2):
    print("num1 is less than num2")
````

````python
x = 12
if x > 10:
    print("Hello")
````

````python
if 5 != 3 * 6:
    print ("Hooray!")
````

````python
if 5 == 15/3:
    print ("Horray!")
````

# Python `if` Statements - Complete Guide

Python's `if` statements are fundamental for controlling program flow. Here's a comprehensive overview of all `if` statement syntaxes and variations:

## 1. Basic `if` Statement

```python
if condition:
    # code to execute if condition is True
```

Example:

```python
x = 10
if x > 5:
    print("x is greater than 5")
```

## 2. `if-else` Statement

```python
if condition:
    # code if True
else:
    # code if False
```

Example:

```python
age = 17
if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
```

## 3. `if-elif-else` Statement

```python
if condition1:
    # code if condition1 is True
elif condition2:
    # code if condition2 is True
elif condition3:
    # code if condition3 is True
else:
    # code if all conditions are False
```

Example:

```python
score = 85
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'
print(f"Your grade is {grade}")
```

## 4. Nested `if` Statements

```python
if condition1:
    if condition2:
        # code if both conditions are True
    else:
        # code if condition1 is True but condition2 is False
else:
    # code if condition1 is False
```

Example:

```python
num = 15
if num > 10:
    print("Number is greater than 10")
    if num % 2 == 0:
        print("And it's even")
    else:
        print("And it's odd")
else:
    print("Number is 10 or less")
```

## 5. Ternary Operator (Conditional Expression)

```python
value_if_true if condition else value_if_false
```

Example:

```python
x = 10
result = "Even" if x % 2 == 0 else "Odd"
print(result)  # Output: "Even"
```

## 6. Short-circuit Evaluation with `and`/`or`

```python
if condition1 and condition2:
    # both must be True
if condition1 or condition2:
    # either must be True
```

Example:

```python
username = "admin"
password = "1234"
if username == "admin" and password == "1234":
    print("Access granted")
else:
    print("Access denied")
```

## 7. `if` with Membership Operators (`in`, `not in`)

```python
if element in collection:
    # code if element exists in collection
if element not in collection:
    # code if element doesn't exist in collection
```

Example:

```python
fruits = ['apple', 'banana', 'orange']
if 'banana' in fruits:
    print("Banana is available")
```

## 8. `if` with Identity Operators (`is`, `is not`)

```python
if x is y:
    # checks if same object in memory
if x is not y:
    # checks if not same object
```

Example:

```python
a = [1, 2, 3]
b = a
if a is b:
    print("Same object")
```

## 9. Chained Comparisons

```python
if lower < value < upper:
    # checks if value is between lower and upper
```

Example:

```python
temperature = 25
if 20 <= temperature <= 30:
    print("Comfortable temperature")
```

## 10. Truthy and Falsy Values

Python evaluates non-boolean values in conditions:

- Falsy: `False`, `0`, `""`, `[]`, `{}`, `None`
- Truthy: Everything else

Example:

```python
name = ""
if name:
    print(f"Hello, {name}")
else:
    print("No name provided")  # This will execute
```

## 11. `if` with Walrus Operator (Python 3.8+)

```python
if (result := some_function()) > 10:
    # assigns and checks in one line
```

Example:

```python
if (user_input := input("Enter a number: ")).isdigit():
    print(f"You entered the number {user_input}")
```

## 12. `if` with Exception Handling

```python
try:
    # some operation
except SomeError:
    # handle error
else:
    # execute if no exception
finally:
    # always execute
```

Example:

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print(f"Result is {result}")
```

## Best Practices:

1. Keep conditions simple and readable
2. Avoid deeply nested `if` statements
3. Use parentheses for complex conditions to clarify precedence
4. Consider using guard clauses for early returns
5. Prefer `if not` for negative conditions with falsy values

These are all the fundamental ways to use `if` statements in Python. The syntax is flexible and can be combined in various ways to handle complex decision-making in your programs.


### Python `if` Statements: All Syntaxes and Concepts

The `if` statement in Python is used for conditional execution. It allows you to check whether an expression or condition is `True` or `False` and then execute a block of code accordingly. Python supports several ways to structure `if` statements, ranging from simple conditionals to more complex ones.

Below is a detailed guide on Python `if` statements, including all possible syntaxes and variations.

---

### **1. Basic `if` Statement**

The simplest form of the `if` statement checks a condition and executes a block of code only if the condition evaluates to `True`.

#### Syntax:

```python
if condition:
    # Code block executed if condition is True
```

#### Example:

```python
x = 10
if x > 5:
    print("x is greater than 5")
```

**Output:**

```
x is greater than 5
```

---

### **2. `if-else` Statement**

The `if-else` statement executes one block of code if the condition is `True`, and a different block if the condition is `False`.

#### Syntax:

```python
if condition:
    # Code block executed if condition is True
else:
    # Code block executed if condition is False
```

#### Example:

```python
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is not greater than 5
```

---

### **3. `if-elif-else` Statement**

The `if-elif-else` statement allows you to check multiple conditions. If the first condition is `False`, the program will check the next `elif` condition, and so on. If all conditions fail, the `else` block will be executed.

#### Syntax:

```python
if condition1:
    # Code block executed if condition1 is True
elif condition2:
    # Code block executed if condition2 is True
elif condition3:
    # Code block executed if condition3 is True
else:
    # Code block executed if no conditions are True
```

#### Example:

```python
x = 8
if x < 5:
    print("x is less than 5")
elif x == 8:
    print("x is equal to 8")
else:
    print("x is greater than 5")
```

**Output:**

```
x is equal to 8
```

---

### **4. Nested `if` Statements**

You can nest `if` statements within each other, which means putting an `if` statement inside another `if` block. This is useful when you need to check multiple conditions in a hierarchical manner.

#### Syntax:

```python
if condition1:
    if condition2:
        # Code block executed if both condition1 and condition2 are True
    else:
        # Code block executed if condition1 is True and condition2 is False
else:
    # Code block executed if condition1 is False
```

#### Example:

```python
x = 7
y = 3
if x > 5:
    if y < 5:
        print("x is greater than 5 and y is less than 5")
    else:
        print("x is greater than 5 and y is not less than 5")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is greater than 5 and y is less than 5
```

---

### **5. `if` with Multiple Conditions (Logical Operators)**

You can combine multiple conditions using logical operators like `and`, `or`, and `not`. These allow you to create more complex conditions.

#### `and` Operator: Both conditions must be `True`.

```python
if condition1 and condition2:
    # Code block executed if both condition1 and condition2 are True
```

#### `or` Operator: At least one condition must be `True`.

```python
if condition1 or condition2:
    # Code block executed if at least one of condition1 or condition2 is True
```

#### `not` Operator: Inverts the truth value of the condition.

```python
if not condition:
    # Code block executed if condition is False
```

#### Example:

```python
x = 8
y = 3
if x > 5 and y < 5:
    print("x is greater than 5 and y is less than 5")

if x > 5 or y > 5:
    print("At least one condition is True")

if not x < 5:
    print("x is not less than 5")
```

**Output:**

```
x is greater than 5 and y is less than 5
At least one condition is True
x is not less than 5
```

---

### **6. `if` Statements with Comparison Operators**

You can use comparison operators (`==`, `!=`, `>`, `<`, `>=`, `<=`) within the `if` statement to compare variables.

#### Example:

```python
x = 10
y = 20
if x == y:
    print("x is equal to y")
elif x != y:
    print("x is not equal to y")
if x > y:
    print("x is greater than y")
else:
    print("x is less than or equal to y")
```

**Output:**

```
x is not equal to y
x is less than or equal to y
```

---

### **7. `if` with `in` and `not in`**

The `in` and `not in` operators are used to check if a value is present or absent in a sequence (like a list, string, or tuple).

#### Example:

```python
my_list = [1, 2, 3, 4, 5]
if 3 in my_list:
    print("3 is in the list")

if 10 not in my_list:
    print("10 is not in the list")
```

**Output:**

```
3 is in the list
10 is not in the list
```

---

### **8. `if` with `is` and `is not`**

The `is` operator checks whether two objects refer to the same object in memory (identity comparison), and `is not` checks the opposite.

#### Example:

```python
x = [1, 2, 3]
y = x
z = [1, 2, 3]

if x is y:
    print("x and y refer to the same object")

if x is not z:
    print("x and z do not refer to the same object")
```

**Output:**

```
x and y refer to the same object
x and z do not refer to the same object
```

---

### **9. Ternary (Conditional) Operator**

Python has a shorthand method for `if-else` called the **ternary operator** (also known as a conditional expression), which allows you to write an `if-else` in a single line.

#### Syntax:

```python
value_if_true if condition else value_if_false
```

#### Example:

```python
x = 10
result = "Greater than 5" if x > 5 else "Less than or equal to 5"
print(result)
```

**Output:**

```
Greater than 5
```

---

### **10. `if` with `try-except` (Exception Handling)**

You can use `try-except` blocks to catch exceptions within `if` statements. This is often used for handling errors that may arise from invalid operations or input.

#### Example:

```python
try:
    x = int(input("Enter a number: "))
    if x > 0:
        print("Positive number")
    elif x < 0:
        print("Negative number")
    else:
        print("Zero")
except ValueError:
    print("Invalid input! Please enter a valid integer.")
```

---

### **Summary**

Here is a quick summary of the main types of `if` statements and their syntaxes:

- **Basic `if`**: Executes code if a condition is `True`.
- **`if-else`**: Executes code if the condition is `True`, otherwise executes alternative code.
- **`if-elif-else`**: Checks multiple conditions and executes corresponding code for the first `True` condition.
- **Nested `if`**: Places one `if` statement inside another for more complex logic.
- **Logical Operators (`and`, `or`, `not`)**: Combine multiple conditions.
- **Comparison Operators (`==`, `!=`, `<`, `>`, `<=`, `>=`)**: Compare values.
- **`in` / `not in`**: Checks for membership in a sequence.
- **`is` / `is not`**: Checks object identity.
- **Ternary Operator**: A compact `if-else` expression.
- **`try-except`**: Used to handle exceptions within `if` statements.

Each of these constructs allows you to control the flow of your program based on different conditions, making your code more powerful and flexible.
