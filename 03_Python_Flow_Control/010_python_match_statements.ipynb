## Python `match` Keyword: From Basic to Advanced

The `match` keyword was introduced in Python 3.10 as part of the **Structural Pattern Matching** feature. It provides a more powerful and flexible way to match patterns in data structures, allowing developers to write cleaner and more readable code.

Pattern matching is similar to a switch-case statement in other languages but with more advanced features and capabilities. In Python, `match` allows you to match values or patterns against specific cases and execute corresponding blocks of code based on the matched patterns.

Below is a comprehensive guide to understanding the `match` keyword in Python from basic to advanced concepts.

---

## **1. Basic Concepts**

### **1.1. Syntax of `match` Statement**

The basic syntax of the `match` statement is as follows:

```python
match expression:
    case pattern_1:
        # Block of code to execute if pattern_1 matches
    case pattern_2:
        # Block of code to execute if pattern_2 matches
    case _:
        # Default case, executes if no pattern matches
```

- **expression**: The value or object being tested.
- **pattern**: The condition or structure you are matching against.
- **case \_**: The default case, often used to catch unmatched patterns (similar to `else` in an `if-else` block).

---

## **2. Basic Examples of `match`**

### **2.1. Matching Basic Values**

The `match` statement can be used to match simple values, like integers or strings.

#### **Example**:

```python
def greet(person):
    match person:
        case "Alice":
            return "Hello, Alice!"
        case "Bob":
            return "Hello, Bob!"
        case _:
            return "Hello, stranger!"

print(greet("Alice"))  # Output: Hello, Alice!
print(greet("Charlie"))  # Output: Hello, stranger!
```

Here, the value of `person` is compared to each case. If a match is found, the corresponding block of code is executed.

---

## **3. Matching Data Structures (Lists, Tuples, Dictionaries)**

### **3.1. Matching Lists**

You can match lists with a specific number of elements and types using `match`.

#### **Example**:

```python
def process_data(data):
    match data:
        case [x, y, z]:
            return f"Matched a 3-element list: {x}, {y}, {z}"
        case [x, y]:
            return f"Matched a 2-element list: {x}, {y}"
        case _:
            return "No match"

print(process_data([1, 2, 3]))  # Output: Matched a 3-element list: 1, 2, 3
print(process_data([1, 2]))     # Output: Matched a 2-element list: 1, 2
```

### **3.2. Matching Tuples**

Similar to lists, tuples can also be matched based on the number and types of elements.

#### **Example**:

```python
def process_tuple(data):
    match data:
        case (x, y, z):
            return f"Matched a tuple: {x}, {y}, {z}"
        case (x, y):
            return f"Matched a 2-element tuple: {x}, {y}"
        case _:
            return "No match"

print(process_tuple((1, 2, 3)))  # Output: Matched a tuple: 1, 2, 3
print(process_tuple((1, 2)))     # Output: Matched a 2-element tuple: 1, 2
```

### **3.3. Matching Dictionaries**

You can also match dictionaries based on their keys and values.

#### **Example**:

```python
def process_dict(data):
    match data:
        case {"name": name, "age": age}:
            return f"Name: {name}, Age: {age}"
        case {"name": name}:
            return f"Name: {name}, Age unknown"
        case _:
            return "No match"

print(process_dict({"name": "Alice", "age": 30}))  # Output: Name: Alice, Age: 30
print(process_dict({"name": "Bob"}))  # Output: Name: Bob, Age unknown
```

Here, the `match` statement extracts values associated with keys directly within the case statement.

---

## **4. Advanced Matching**

### **4.1. Wildcard Pattern (`_`)**

The `_` symbol is used as a wildcard to match anything. It’s often used in the final case to handle unmatched patterns.

#### **Example**:

```python
def describe_number(number):
    match number:
        case 1:
            return "One"
        case 2:
            return "Two"
        case _:
            return "Not 1 or 2"

print(describe_number(1))  # Output: One
print(describe_number(3))  # Output: Not 1 or 2
```

### **4.2. Matching with Conditions (Guard Clauses)**

You can use the `if` clause within a `case` to add conditions that must be true for the match to succeed.

#### **Example**:

```python
def check_age(age):
    match age:
        case age if age < 18:
            return "Minor"
        case age if age >= 18:
            return "Adult"
        case _:
            return "Unknown"

print(check_age(17))  # Output: Minor
print(check_age(25))  # Output: Adult
```

### **4.3. Matching Class Instances**

You can match objects of a class and check their attributes or methods.

#### **Example**:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

def greet_person(person):
    match person:
        case Person(name="Alice", age=30):
            return "Hello Alice, welcome back!"
        case Person(name, age):
            return f"Hello {name}, age {age}"
        case _:
            return "Who are you?"

alice = Person("Alice", 30)
bob = Person("Bob", 25)

print(greet_person(alice))  # Output: Hello Alice, welcome back!
print(greet_person(bob))    # Output: Hello Bob, age 25
```

This example shows how to match an instance of a class and access its attributes in the case clause.

### **4.4. Nested Patterns**

You can use nested patterns to match more complex data structures, like lists of lists, or tuples of tuples.

#### **Example**:

```python
def process_nested(data):
    match data:
        case [[x, y], [a, b]]:
            return f"Matched nested lists: {x}, {y}, {a}, {b}"
        case _:
            return "No match"

print(process_nested([[1, 2], [3, 4]]))  # Output: Matched nested lists: 1, 2, 3, 4
```

### **4.5. Matching Sequences (Using `*`)**

In pattern matching, the `*` operator can be used to match multiple items in a sequence (list, tuple, etc.).

#### **Example**:

```python
def process_sequence(data):
    match data:
        case [x, y, *rest]:
            return f"First two items: {x}, {y}, remaining: {rest}"
        case _:
            return "No match"

print(process_sequence([1, 2, 3, 4, 5]))  # Output: First two items: 1, 2, remaining: [3, 4, 5]
```

---

## **5. Use Cases for `match`**

### **5.1. Parsing Structured Data**

Pattern matching is useful when you need to extract or process structured data such as JSON-like objects, lists, and tuples.

#### **Example**:

```python
def process_message(message):
    match message:
        case {"type": "text", "content": content}:
            return f"Text message: {content}"
        case {"type": "image", "url": url}:
            return f"Image message: {url}"
        case _:
            return "Unknown message type"

message = {"type": "text", "content": "Hello, world!"}
print(process_message(message))  # Output: Text message: Hello, world!
```

---

## **6. Summary**

- **Basic Structure**: `match expression: case pattern_1:`, which allows you to match different patterns against an expression.
- **Wildcards (`_`)**: Used as a catch-all in case no pattern matches.
- **Guard Clauses**: Use `if` within a `case` to add additional conditions.
- **Class Matching**: Match instances of a class and its attributes.
- **Multiple Patterns**: Match multiple patterns using `|`.
- **Sequences**: Use `*` to capture multiple elements in sequences like lists or tuples.

Pattern matching with `match` in Python 3.10+ enhances the language's ability to work with complex data structures and provides more readable, concise, and powerful code compared to traditional `if-else` statements. Understanding and applying these features will make your code more efficient and maintainable.

## **1. Basic `match-case` Syntax**

The `match` statement compares a value against multiple patterns.

```python
match subject:
    case pattern1:
        action1
    case pattern2:
        action2
    case _:
        default_action
```

**Example:**

```python
status = 404
match status:
    case 200:
        print("Success")
    case 404:
        print("Not found")
    case _:
        print("Unknown status")
```

---

## **2. Matching Literal Values**

Match against specific constants (numbers, strings, etc.)

```python
direction = "north"
match direction:
    case "north":
        print("Going up")
    case "south":
        print("Going down")
```

---

## **3. Wildcard Pattern**

`_` acts as a catch-all (like `else` in `if` statements)

```python
match value:
    case 1:
        print("One")
    case _:
        print("Not one")
```

---

## **4. Matching Multiple Patterns**

Combine patterns with `|` (OR)

```python
match command:
    case "quit" | "exit":
        print("Exiting program")
    case "save":
        print("Saving...")
```

---

## **5. Type Matching**

Check types while capturing variables

```python
def handle_value(x):
    match x:
        case int():
            print(f"Got integer: {x}")
        case str():
            print(f"Got string: {x}")
```

---

## **6. Value Capture**

Extract values during matching

```python
point = (3, 5)
match point:
    case (x, y):
        print(f"X: {x}, Y: {y}")
```

---

## **7. Guard Conditions**

Add conditional logic with `if`

```python
match point:
    case (x, y) if x == y:
        print("On diagonal")
    case (x, y):
        print("Regular point")
```

---

## **8. Sequence Patterns**

Match lists, tuples with flexible patterns

```python
colors = ["red", "green", "blue"]
match colors:
    case ["red", *rest]:
        print(f"Starts with red, then {rest}")
    case [_, "green", _]:
        print("Green in middle")
```

---

## **9. Mapping Patterns**

Match dictionaries by keys

```python
config = {"mode": "debug", "level": 3}
match config:
    case {"mode": "debug", "level": lvl}:
        print(f"Debug level {lvl}")
    case {"mode": "production"}:
        print("Production mode")
```

---

## **10. Class Patterns**

Match object attributes

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(3, 5)
match p:
    case Point(x=0, y=0):
        print("Origin")
    case Point(x=x, y=y):
        print(f"Point at ({x}, {y})")
```

---

## **11. Nested Patterns**

Combine patterns recursively

```python
match points:
    case [Point(0, 0), Point(x, y)]:
        print(f"Line from origin to ({x}, {y})")
```

---

## **12. OR Patterns**

Alternative patterns with `|`

```python
match value:
    case 0 | 1 | 2:
        print("Small number")
    case int(n) if n > 100:
        print("Large number")
```

## **13. Performance Considerations**

- `match` is generally as fast as equivalent `if-elif` chains
- Order patterns from most specific to most general
- Prefer simple patterns over complex nested ones when possible

---

## **14. Common Pitfalls**

1. **Forgetting wildcard `_`**: Unmatched cases raise errors
2. **Variable capture confusion**:
   ```python
   case [x, x]:  # Both elements must be equal
   ```
3. **Overusing nested patterns**: Can become hard to read

---

## **15. When to Use vs `if-elif-else`**

**Use `match` when:**

- Checking multiple literal values
- Destructuring complex data types
- Pattern matching gives cleaner code

**Use `if` when:**

- Conditions are complex boolean expressions
- Checking ranges (`if 0 < x < 10`)
- Need compatibility with Python <3.10

---

1. **Order patterns from specific to general**
2. **Use guards for additional filtering**
3. **Extract complex patterns into helper functions**
4. **Add comments for non-obvious patterns**
5. **Prefer simple patterns over clever nested ones**
