### **Type Coercion and Type Conversion in Python: From Basic to Advanced**

In Python, **Type Coercion** and **Type Conversion** are both mechanisms that deal with the automatic or explicit conversion of data from one type to another. While they are related, they have distinct differences in terms of how and when the conversion happens.

---

### **1. Type Coercion**

**Type Coercion** is the automatic conversion of one data type to another, performed by Python itself, usually during operations that involve different types of data.

#### **1.1. How Type Coercion Works**

Python will automatically **coerce** types to ensure that operations between mixed data types are valid. For example:

- **Integer + Float**: Python will automatically convert the integer to a float before performing the addition.
- **String + Integer (with explicit conversion)**: Python won’t coerce these types automatically but will raise an error.

#### **1.2. Examples of Type Coercion**

1. **Integer and Float**:
   Python automatically converts an integer to a float when performing operations involving a float.

```python
a = 5      # integer
b = 3.2    # float
result = a + b
print(result)  # Output: 8.2 (integer `a` is coerced into a float before addition)
```

2. **String to Integer (Coercion isn’t done automatically)**:
   Python **does not** coerce strings to numbers implicitly. Instead, an explicit conversion is required.

```python
string_number = "100"
# You can't add a string and an integer directly without converting
# print(string_number + 5)  # This will throw an error

# Instead, you need to explicitly convert the string to an integer:
result = int(string_number) + 5
print(result)  # Output: 105
```

#### **1.3. Implicit Coercion in Arithmetic Operations**

```python
x = 10        # int
y = 20.5      # float

result = x + y  # Python automatically coerces `x` into a float to make it work with `y`
print(result)  # Output: 30.5
```

Here, Python coerces `x` (an integer) to a float to ensure the addition with `y` (a float) works seamlessly.

---

### **2. Type Conversion**

**Type Conversion**, also known as **Type Casting**, is the explicit conversion of one data type to another. Unlike **coercion**, which happens automatically, type conversion needs to be done manually by the programmer using conversion functions.

#### **2.1. Types of Type Conversion**

1. **Implicit Type Conversion**: When Python automatically converts a lower precision type to a higher precision type.

   - Examples: Converting `int` to `float`.

2. **Explicit Type Conversion**: When the programmer explicitly converts from one type to another using built-in functions like `int()`, `float()`, `str()`, etc.

#### **2.2. Common Type Conversion Functions**

1. **`int()`**: Converts a value to an integer.

   ```python
   x = 3.7
   print(int(x))  # Output: 3 (floating point `x` is truncated)
   ```

2. **`float()`**: Converts a value to a floating-point number.

   ```python
   x = "3.5"
   print(float(x))  # Output: 3.5 (string `x` is converted to float)
   ```

3. **`str()`**: Converts a value to a string.

   ```python
   x = 5
   print(str(x))  # Output: "5"
   ```

4. **`list()`**: Converts a value into a list.

   ```python
   x = "hello"
   print(list(x))  # Output: ['h', 'e', 'l', 'l', 'o']
   ```

5. **`tuple()`**: Converts a value into a tuple.

   ```python
   x = [1, 2, 3]
   print(tuple(x))  # Output: (1, 2, 3)
   ```

6. **`set()`**: Converts a value into a set.
   ```python
   x = [1, 2, 2, 3]
   print(set(x))  # Output: {1, 2, 3} (duplicate values are removed)
   ```

#### **2.3. Explicit Type Conversion Examples**

1. **Convert Float to Integer**:

   ```python
   a = 3.75
   b = int(a)  # Converts to 3 (the decimal part is truncated)
   print(b)     # Output: 3
   ```

2. **Convert Integer to String**:

   ```python
   a = 100
   b = str(a)  # Converts to "100"
   print(b)    # Output: "100"
   ```

3. **Convert String to Float**:
   ```python
   a = "10.5"
   b = float(a)  # Converts the string to a float
   print(b)       # Output: 10.5
   ```

---

### **3. Implicit vs Explicit Type Conversion**

- **Implicit Type Conversion**: This is automatic conversion done by Python to maintain compatibility during operations. The user has no control over it.
  - **Example**: Integer to float during an arithmetic operation.

```python
a = 2
b = 3.6
result = a + b
print(result)  # Output: 5.6 (Implicit conversion of `a` to float)
```

- **Explicit Type Conversion**: The programmer manually casts one data type to another using built-in functions like `int()`, `float()`, `str()`, etc.

```python
a = "50"
b = int(a)  # Explicit conversion from string to integer
print(b)  # Output: 50
```

---

### **4. Advanced Type Conversion**

#### **4.1. Type Conversion in Containers (Lists, Tuples, Sets, etc.)**

Python allows conversion between various container types (e.g., list, tuple, set). These conversions can help in manipulating the data structure according to your requirements.

- **List to Tuple**

  ```python
  lst = [1, 2, 3]
  tpl = tuple(lst)
  print(tpl)  # Output: (1, 2, 3)
  ```

- **Tuple to List**

  ```python
  tpl = (4, 5, 6)
  lst = list(tpl)
  print(lst)  # Output: [4, 5, 6]
  ```

- **List to Set**
  ```python
  lst = [1, 2, 3, 3]
  s = set(lst)
  print(s)  # Output: {1, 2, 3} (duplicates removed)
  ```

#### **4.2. Converting Between Numeric Types**

You can convert between `int`, `float`, and `complex` types in Python. However, you need to be aware of the loss of information during conversion.

- **Complex to Float** (not possible directly, unless the complex number is purely real):

  ```python
  c = 3 + 5j
  # Python does not allow direct conversion of complex to float:
  # result = float(c)  # This will raise an error
  ```

- **Float to Integer**: When converting from a float to an integer, the decimal part is truncated.
  ```python
  x = 3.9
  result = int(x)  # Output: 3 (decimal part discarded)
  ```

---

### **5. Common Errors in Type Conversion**

1. **ValueError during Conversion**:
   Converting a non-convertible value can lead to a `ValueError`.

```python
# Example: trying to convert a non-numeric string to an integer
string_value = "hello"
try:
    result = int(string_value)
except ValueError:
    print("Cannot convert to int!")  # Output: Cannot convert to int!
```

2. **TypeError due to Invalid Type Coercion**:
   Trying to perform operations on incompatible data types will raise a `TypeError`.

```python
# Trying to add an integer to a string without conversion
x = 5
y = "Hello"
try:
    result = x + y  # This will raise a TypeError
except TypeError:
    print("Cannot add integer and string directly!")  # Output: Cannot add integer and string directly!
```

---

### **6. Conclusion**

- **Type Coercion** is the automatic conversion that Python does when performing operations involving mixed data types.
- **Type Conversion** is the manual process of converting one type to another using functions like `int()`, `float()`, `str()`, etc.
- Understanding **Implicit** and **Explicit** type conversion is essential for working with mixed data types in Python.
- **Advanced** conversions include working with data structures (lists, tuples, sets), and handling errors like `ValueError` and `TypeError` during conversions.

Type coercion and type conversion are critical in Python to maintain data consistency and ensure that operations between different data types do not cause runtime errors.

## **8. Custom Type Conversion**

### **`__int__`, `__float__`, `__bool__` Methods**

```python
class Measurement:
    def __init__(self, value):
        self.value = value

    def __int__(self):
        return int(self.value)

    def __float__(self):
        return float(self.value)

m = Measurement(3.9)
int(m)    # 3
float(m)  # 3.9
```

### **`__str__` vs `__repr__`**

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

p = Point(3, 4)
print(str(p))   # (3, 4)
print(repr(p))  # Point(x=3, y=4)
```

## **9. Protocol-Based Conversion**

Python 3.8+ style with `typing.Protocol`:

```python
from typing import Protocol, runtime_checkable

@runtime_checkable
class ConvertibleToFloat(Protocol):
    def __float__(self) -> float: ...

def process(value: ConvertibleToFloat) -> float:
    return float(value)

class Temperature:
    def __float__(self):
        return 98.6

process(Temperature())  # 98.6
```

### **Numeric Tower Rules**

Python automatically promotes types:
`int` → `float` → `complex`

```python
3 + 4.5       # 7.5 (int→float)
2.5 + 3j      # (2.5+3j) (float→complex)
```

### **Special Cases**

## **11. JSON Serialization/Deserialization**

### **Standard Types**

```python
import json
data = {'name': 'Alice', 'age': 30, 'scores': [95, 88]}
json_str = json.dumps(data)  # Serialize
loaded = json.loads(json_str)  # Deserialize
```

### **Custom Encoders**

```python
from datetime import datetime
from json import JSONEncoder

class CustomEncoder(JSONEncoder):
    def default(self, obj):
        if isinstance(obj, datetime):
            return obj.isoformat()
        return super().default(obj)

json.dumps({'now': datetime.now()}, cls=CustomEncoder)
```

## **13. Performance Considerations**

- **Explicit Conversion** is generally cheap
- **Implicit Coercion** has minimal overhead
- **Repeated Conversion** in loops should be avoided:

  ```python
  # Bad
  for i in range(10000):
      x = float(i) * 1.5  # Repeated conversion

  # Good
  x = 1.5
  for i in range(10000):
      result = i * x
  ```
