### Python `if-else` Statements: From Basic to Advanced

In Python, conditional statements (`if`, `elif`, `else`) are used to control the flow of the program based on certain conditions. These statements allow you to execute specific blocks of code depending on whether a condition is true or false. Understanding these statements from basic to advanced concepts is essential for making decisions within your code.

---

## **1. Basic Concepts of `if-else` Statements**

### **1.1. `if` Statement**

The `if` statement evaluates a condition (expression) and executes the code block that follows it if the condition is `True`.

#### **Syntax**:

```python
if condition:
    # Execute this block of code if condition is true
```

#### **Example**:

```python
x = 5
if x > 3:
    print("x is greater than 3")
```

**Output**:

```
x is greater than 3
```

### **1.2. `if-else` Statement**

The `if-else` statement allows you to specify what to do when the condition is `True` and what to do when it is `False`.

#### **Syntax**:

```python
if condition:
    # Execute this block of code if condition is true
else:
    # Execute this block of code if condition is false
```

#### **Example**:

```python
x = 2
if x > 3:
    print("x is greater than 3")
else:
    print("x is not greater than 3")
```

**Output**:

```
x is not greater than 3
```

---

## **2. `if-elif-else` Statement**

The `elif` (else if) statement allows you to check multiple conditions in sequence. If the first `if` condition is `False`, Python checks the `elif` conditions, and if none of them are `True`, it executes the `else` block.

#### **Syntax**:

```python
if condition1:
    # Execute this block of code if condition1 is true
elif condition2:
    # Execute this block of code if condition2 is true
else:
    # Execute this block of code if none of the conditions are true
```

#### **Example**:

```python
x = 7
if x < 5:
    print("x is less than 5")
elif x == 7:
    print("x is equal to 7")
else:
    print("x is greater than 5 but not 7")
```

**Output**:

```
x is equal to 7
```

---

## **3. Nested `if-else` Statements**

A nested `if` statement occurs when an `if` or `else` statement is placed inside another `if` or `else` statement. This allows you to evaluate multiple conditions in a hierarchical manner.

#### **Syntax**:

```python
if condition1:
    if condition2:
        # Execute this block if both conditions are true
    else:
        # Execute this block if condition1 is true but condition2 is false
else:
    # Execute this block if condition1 is false
```

#### **Example**:

```python
x = 7
y = 10
if x > 5:
    if y < 15:
        print("x is greater than 5 and y is less than 15")
    else:
        print("x is greater than 5 but y is not less than 15")
else:
    print("x is not greater than 5")
```

**Output**:

```
x is greater than 5 and y is less than 15
```

---

## **4. Ternary Conditional Operator (Conditional Expressions)**

Python provides a shorthand syntax for `if-else` statements called the **ternary operator**. It allows you to assign values based on a condition in a single line.

#### **Syntax**:

```python
value = true_value if condition else false_value
```

#### **Example**:

```python
x = 5
result = "Even" if x % 2 == 0 else "Odd"
print(result)
```

**Output**:

```
Odd
```

This syntax is especially useful when you want to assign values based on a condition in a concise way.

## **5. Multiple Conditions with `if-else`**

You can combine multiple conditions using logical operators like `and`, `or`, and `not` to evaluate more complex conditions in an `if` or `elif` block.

### **5.1. Using `and`**

- The `and` operator returns `True` if both conditions are `True`.

#### **Example**:

```python
x = 7
y = 10
if x > 5 and y < 15:
    print("Both conditions are true")
else:
    print("At least one condition is false")
```

**Output**:

```
Both conditions are true
```

### **5.2. Using `or`**

- The `or` operator returns `True` if at least one of the conditions is `True`.

#### **Example**:

```python
x = 4
y = 10
if x < 5 or y > 15:
    print("At least one condition is true")
else:
    print("Both conditions are false")
```

**Output**:

```
At least one condition is true
```

### **5.3. Using `not`**

- The `not` operator negates the condition, returning `True` if the condition is `False`.

#### **Example**:

```python
x = 5
if not x > 10:
    print("x is not greater than 10")
else:
    print("x is greater than 10")
```

**Output**:

```
x is not greater than 10
```

Understanding edge cases and how Python handles them is essential. Some common edge cases are:

- **Zero**: The number `0` is considered `False` in Python.
- **`None`**: The special value `None` evaluates to `False`.

#### **Example with Edge Cases**:

```python

y = None

    print("x is truthy")
else:
    print("x is falsy")  # This will print "x is falsy"

if y:
    print("y is truthy")
else:
    print("y is falsy")  # This will print "y is falsy"
```

**Output**:

```
x is falsy
y is falsy
```

### Python `if-else` Statements: From Basic to Advanced

The `if-else` statement in Python is one of the fundamental control flow statements used to execute code conditionally. It evaluates a condition and determines which code block to execute based on whether the condition is `True` or `False`.

Here is a complete guide that explains all aspects of `if-else` statements in Python, ranging from the basic syntax to more advanced usages.

---

### **1. Basic `if-else` Syntax**

The `if-else` statement consists of a condition and two code blocks: one that executes if the condition is `True` and one that executes if the condition is `False`.

#### Syntax:

```python
if condition:
    # Code block executed if the condition is True
else:
    # Code block executed if the condition is False
```

#### Example:

```python
x = 10
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is greater than 5
```

---

### **2. `if-else` with Multiple Conditions (Using `elif`)**

If you have more than two conditions to check, you can use `elif` (else if) between the `if` and `else` blocks to check additional conditions.

#### Syntax:

```python
if condition1:
    # Code block executed if condition1 is True
elif condition2:
    # Code block executed if condition2 is True
else:
    # Code block executed if none of the conditions are True
```

#### Example:

```python
x = 10
if x < 5:
    print("x is less than 5")
elif x == 10:
    print("x is equal to 10")
else:
    print("x is greater than 5 but not 10")
```

**Output:**

```
x is equal to 10
```

---

### **3. Nested `if-else` Statements**

A nested `if-else` statement is an `if` or `else` block inside another `if` or `else` block. This is useful when you need to check additional conditions after the first condition is satisfied.

#### Syntax:

```python
if condition1:
    if condition2:
        # Code block executed if both condition1 and condition2 are True
    else:
        # Code block executed if condition1 is True and condition2 is False
else:
    # Code block executed if condition1 is False
```

#### Example:

```python
x = 10
y = 5
if x > 5:
    if y < 5:
        print("x is greater than 5 and y is less than 5")
    else:
        print("x is greater than 5 but y is not less than 5")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is greater than 5 but y is not less than 5
```

---

### **4. Using Logical Operators in `if-else`**

You can combine multiple conditions using logical operators like `and`, `or`, and `not`. These operators allow you to create more complex conditional expressions.

#### Syntax:

- `and`: Both conditions must be `True`.
- `or`: At least one condition must be `True`.
- `not`: Negates the condition (i.e., `True` becomes `False` and vice versa).

#### Example:

```python
x = 10
y = 5
if x > 5 and y < 10:
    print("x is greater than 5 and y is less than 10")
else:
    print("Either x is less than or equal to 5 or y is greater than or equal to 10")
```

**Output:**

```
x is greater than 5 and y is less than 10
```

---

### **5. Ternary Operator (Conditional Expression)**

Python supports a shorthand method for `if-else` statements known as the ternary operator (conditional expression). It allows you to write simple `if-else` conditions in a single line.

#### Syntax:

```python
value_if_true if condition else value_if_false
```

#### Example:

```python
x = 10
result = "Greater than 5" if x > 5 else "Less than or equal to 5"
print(result)
```

**Output:**

```
Greater than 5
```

---

### **6. Using `in` and `not in` in `if-else`**

Python provides the `in` and `not in` operators to check whether a value exists within a sequence (like a list, string, or tuple).

#### Syntax:

```python
if value in sequence:
    # Code block executed if value is found in the sequence
else:
    # Code block executed if value is not found in the sequence
```

#### Example:

```python
my_list = [1, 2, 3, 4, 5]
if 3 in my_list:
    print("3 is in the list")
else:
    print("3 is not in the list")
```

**Output:**

```
3 is in the list
```

---

### **7. `is` and `is not` in `if-else`**

The `is` operator checks for object identity (whether two variables point to the same object in memory), and `is not` checks for the opposite.

#### Example:

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

if a is b:
    print("a and b refer to the same object")
else:
    print("a and b do not refer to the same object")

if a is not c:
    print("a and c do not refer to the same object")
```

**Output:**

```
a and b refer to the same object
a and c do not refer to the same object
```

---

### **8. Using `try-except` with `if-else`**

You can handle exceptions inside `if-else` statements using `try-except` blocks. This is useful when you're dealing with user input or operations that may raise exceptions.

#### Example:

```python
try:
    x = int(input("Enter a number: "))
    if x > 0:
        print("Positive number")
    elif x < 0:
        print("Negative number")
    else:
        print("Zero")
except ValueError:
    print("Invalid input! Please enter a valid number.")
```

### **9. Combining Multiple `if-else` Conditions with `elif`**

You can chain multiple `if`, `elif`, and `else` conditions to evaluate a sequence of conditions in a particular order.

#### Example:

```python

if age < 13:
    print("Child")
elif age >= 13 and age < 18:
    print("Teenager")
elif age >= 18 and age < 60:
    print("Adult")
else:
    print("Senior Citizen")
```

**Output:**

```
Adult
```

### **10. Using `else` with `for` and `while` Loops**

You can combine `else` with loops. In the case of loops, the `else` block will only be executed if the loop completes without encountering a `break` statement.

#### Example (with `for` loop):

```python

    if i == 3:
        break
else:
    print("Loop completed without breaking")
```


### **Summary of `if-else` Statements**

- **`if` statement**: Executes code if a condition is `True`.
- **`if-else` statement**: Executes one block of code if the condition is `True`, and another if the condition is `False`.
- **`if-elif-else`**: Checks multiple conditions and executes the code corresponding to the first `True` condition.
- **Ternary Operator**: A shorthand for `if-else`, written in one line.
- **Logical Operators**: Use `and`, `or`, and `not` to combine multiple conditions.
- **`in` and `not in`**: Check membership in a sequence (list, tuple, string).
- **`is` and `is not`**: Check if two variables refer to the same object.
- **`try-except`**: Used to handle exceptions inside `if-else` conditions.
- **`else` with loops**: Executes if a loop finishes normally without `break`.

By understanding these different variations of the `if-else` statement, you can write more flexible and efficient Python programs to handle a wide range of scenarios.

## **12. `if-else` with Exception Handling**

Use `try-except-else-finally` for error handling.

### **Syntax**

```python

    # handle error
else:
    # execute if no error

    # always execute
```

### Python `if-else` Statements: From Basic to Advanced

In Python, conditional statements (`if`, `elif`, `else`) are used to control the flow of the program based on certain conditions. These statements allow you to execute specific blocks of code depending on whether a condition is true or false. Understanding these statements from basic to advanced concepts is essential for making decisions within your code.

---

## **1. Basic Concepts of `if-else` Statements**

### **1.1. `if` Statement**

The `if` statement evaluates a condition (expression) and executes the code block that follows it if the condition is `True`.

#### **Syntax**:

```python
if condition:
    # Execute this block of code if condition is true
```

#### **Example**:

```python
x = 5
if x > 3:
    print("x is greater than 3")
```

**Output**:

```
x is greater than 3
```

### **1.2. `if-else` Statement**

The `if-else` statement allows you to specify what to do when the condition is `True` and what to do when it is `False`.

#### **Syntax**:

```python
if condition:
    # Execute this block of code if condition is true
else:
    # Execute this block of code if condition is false
```

#### **Example**:

```python
x = 2
if x > 3:
    print("x is greater than 3")
else:
    print("x is not greater than 3")
```

**Output**:

```
x is not greater than 3
```

---

## **2. `if-elif-else` Statement**

The `elif` (else if) statement allows you to check multiple conditions in sequence. If the first `if` condition is `False`, Python checks the `elif` conditions, and if none of them are `True`, it executes the `else` block.

#### **Syntax**:

```python
if condition1:
    # Execute this block of code if condition1 is true
elif condition2:
    # Execute this block of code if condition2 is true
else:
    # Execute this block of code if none of the conditions are true
```

#### **Example**:

```python
x = 7
if x < 5:
    print("x is less than 5")
elif x == 7:
    print("x is equal to 7")
else:
    print("x is greater than 5 but not 7")
```

**Output**:

```
x is equal to 7
```

---

## **3. Nested `if-else` Statements**

A nested `if` statement occurs when an `if` or `else` statement is placed inside another `if` or `else` statement. This allows you to evaluate multiple conditions in a hierarchical manner.

#### **Syntax**:

```python
if condition1:
    if condition2:
        # Execute this block if both conditions are true
    else:
        # Execute this block if condition1 is true but condition2 is false
else:
    # Execute this block if condition1 is false
```

#### **Example**:

```python
x = 7
y = 10
if x > 5:
    if y < 15:
        print("x is greater than 5 and y is less than 15")
    else:
        print("x is greater than 5 but y is not less than 15")
else:
    print("x is not greater than 5")
```

**Output**:

```
x is greater than 5 and y is less than 15
```

---

## **4. Ternary Conditional Operator (Conditional Expressions)**

Python provides a shorthand syntax for `if-else` statements called the **ternary operator**. It allows you to assign values based on a condition in a single line.

#### **Syntax**:

```python
value = true_value if condition else false_value
```

#### **Example**:

```python
x = 5
result = "Even" if x % 2 == 0 else "Odd"
print(result)
```

**Output**:

```
Odd
```

This syntax is especially useful when you want to assign values based on a condition in a concise way.

---

## **5. Multiple Conditions with `if-else`**

You can combine multiple conditions using logical operators like `and`, `or`, and `not` to evaluate more complex conditions in an `if` or `elif` block.

### **5.1. Using `and`**

- The `and` operator returns `True` if both conditions are `True`.

#### **Example**:

```python
x = 7
y = 10
if x > 5 and y < 15:
    print("Both conditions are true")
else:
    print("At least one condition is false")
```

**Output**:

```
Both conditions are true
```

### **5.2. Using `or`**

- The `or` operator returns `True` if at least one of the conditions is `True`.

#### **Example**:

```python
x = 4
y = 10
if x < 5 or y > 15:
    print("At least one condition is true")
else:
    print("Both conditions are false")
```

**Output**:

```
At least one condition is true
```

### **5.3. Using `not`**

- The `not` operator negates the condition, returning `True` if the condition is `False`.

#### **Example**:

```python
x = 5
if not x > 10:
    print("x is not greater than 10")
else:
    print("x is greater than 10")
```

**Output**:

```
x is not greater than 10
```

---

## **6. `if-else` with Comparison Operators**

Python allows you to use comparison operators in conditional expressions. Some commonly used comparison operators are:

- `==`: Equals
- `!=`: Not equals
- `>`: Greater than
- `<`: Less than
- `>=`: Greater than or equal to
- `<=`: Less than or equal to

### **Example**:

```python
x = 10
if x >= 5 and x <= 15:
    print("x is between 5 and 15")
else:
    print("x is outside the range")
```

**Output**:

```
x is between 5 and 15
```

## **7. Edge Cases with `if-else` Statements**

Understanding edge cases and how Python handles them is essential. Some common edge cases are:

- **Empty Strings**: An empty string (`""`) evaluates to `False` in a condition.
- **Zero**: The number `0` is considered `False` in Python.
- **`None`**: The special value `None` evaluates to `False`.

#### **Example with Edge Cases**:

```python
x = ""
y = None
if x:
    print("x is truthy")
else:
    print("x is falsy")  # This will print "x is falsy"

if y:
    print("y is truthy")
else:
    print("y is falsy")  # This will print "y is falsy"
```

**Output**:

```
x is falsy
y is falsy
```

## **8. Advanced Use-Cases of `if-else`**

### **8.1. Using `if-else` with Functions**

You can use `if-else` to handle various conditions in a function and return different outputs based on those conditions.

#### **Example**:

```python
def check_number(x):
    if x > 0:
        return "Positive"
    elif x < 0:
        return "Negative"
    else:
        return "Zero"

print(check_number(10))  # Output: Positive
print(check_number(-5))  # Output: Negative
print(check_number(0))   # Output: Zero
```

- **Nested `if`**: Placing `if` statements inside other `if` blocks.
- **Ternary operator**: A shorthand way of writing `if-else` expressions.
- **Logical operators (`and`, `or`, `not`)**: Combine conditions.
- **Comparison operators**: Compare values for conditions.
