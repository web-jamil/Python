# Python `if-elif-else` statement

So far, we have presented a Boolean option for conditional statements, with each **[if statement](https://github.com/milaan9/03_Python_Flow_Control/blob/main/001_Python_if_statement.ipynb)** evaluating to either **`True`** or **`False`**. In Python, the **`if-elif-else`** condition statement has an **`elif`** keyword used to chain multiple conditions one after another.


### Syntax:

```python
if condition-1:
     statement 1
elif condition-2:
     stetement 2
elif condition-3:
     stetement 3
     ...
else:
     statement
```

1. The **`elif`** is short for else if. It allows us to check for multiple expressions.

2. If the condition for **`if`** is **`False`**, it checks the condition of the next **`elif`** block and so on.

3. If all the conditions are **`False`**, the body of **`else`** is executed.

4. Only one block among the several **`if-elif-else`** blocks is executed according to the condition.

5. The **`if`** block can have only one **`else`** block. But it can have multiple **`elif`** blocks.

````python
# Example 1:

'''In this program, we check if the number is positive or negative or zero and 
display an appropriate message'''

num = 0

# Try these two variations as well:
# num = 0
# num = -4.5

if num > 0:
    print("Positive number")
elif num == 0:
    print("Zero")
else:
    print("Negative number")
````

If **`num`** is negative, **`Negative number`** is printed.

## **1. Basic `if` Statement**

Executes code when a condition is `True`.

### Syntax

```python
if condition:
    # code to execute
```

### Example

```python
age = 20
if age >= 18:
    print("You're an adult")
```

**Output:** `You're an adult`

---

## **2. `if-else` Statement**

Provides an alternative path when the condition is `False`.

### Syntax

```python
if condition:
    # code if True
else:
    # code if False
```

### Example

```python
num = 7
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

**Output:** `Odd`

---

## **3. `if-elif-else` Chain**

Handles multiple conditions sequentially.

### Syntax

```python
if condition1:
    # code
elif condition2:
    # code
elif condition3:
    # code
else:
    # default case
```

### Example

```python
score = 82
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"Grade: {grade}")
```

**Output:** `Grade: B`

---

## **4. Nested Conditionals**

`if` statements inside other `if` statements.

### Example

```python
num = 15
if num > 10:
    print(">10")
    if num % 3 == 0:
        print("Divisible by 3")
    else:
        print("Not divisible by 3")
else:
    print("<=10")
```

**Output:**

```
>10
Divisible by 3
```

## **5. Ternary Operator**

Compact one-line conditional expression.


### Example

```python

```

### Python `if`, `elif`, and `else` Statements: From Basic to Advanced

The `if`, `elif`, and `else` statements in Python are used for conditional execution. These control flow statements allow you to execute different blocks of code based on conditions. Let's explore these concepts in detail from basic to advanced.

---

### **1. Basic Syntax of `if`, `elif`, and `else`**

#### 1.1 `if` Statement

The `if` statement checks a condition, and if it is `True`, the block of code inside it is executed.

#### Syntax:

```python
if condition:
    # Code to execute if condition is True
```

#### Example:

```python
x = 10
if x > 5:
    print("x is greater than 5")
```

**Output:**

```
x is greater than 5
```

---

#### 1.2 `if-else` Statement

The `if-else` statement checks a condition. If it is `True`, the code inside the `if` block is executed; otherwise, the code inside the `else` block is executed.

#### Syntax:

```python
if condition:
    # Code to execute if condition is True
else:
    # Code to execute if condition is False
```

#### Example:

```python
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is not greater than 5
```

---

#### 1.3 `if-elif-else` Statement

When you need to check multiple conditions, you can use `elif` (short for "else if"). This allows you to check additional conditions if the first one is `False`.

#### Syntax:

```python
if condition1:
    # Code to execute if condition1 is True
elif condition2:
    # Code to execute if condition1 is False and condition2 is True
else:
    # Code to execute if both conditions are False
```

#### Example:

```python
x = 10
if x < 5:
    print("x is less than 5")
elif x == 10:
    print("x is equal to 10")
else:
    print("x is greater than 5 but not 10")
```

**Output:**

```
x is equal to 10
```

---

### **2. Nested `if-else` Statements**

You can place an `if` statement inside another `if` or `else` block, creating nested conditions. This allows for more complex conditional checks.

#### Syntax:

```python
if condition1:
    if condition2:
        # Code to execute if both condition1 and condition2 are True
    else:
        # Code to execute if condition1 is True and condition2 is False
else:
    # Code to execute if condition1 is False
```

#### Example:

```python
x = 10
y = 5
if x > 5:
    if y < 5:
        print("x is greater than 5 and y is less than 5")
    else:
        print("x is greater than 5 but y is not less than 5")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is greater than 5 but y is not less than 5
```

---

### **3. Logical Operators in `if-else` Statements**

Logical operators like `and`, `or`, and `not` allow you to combine multiple conditions in a single `if` statement.

#### 3.1 Using `and`

The `and` operator returns `True` if both conditions are `True`.

#### Example:

```python
x = 10
y = 5
if x > 5 and y < 10:
    print("Both conditions are True")
else:
    print("At least one condition is False")
```

**Output:**

```
Both conditions are True
```

#### 3.2 Using `or`

The `or` operator returns `True` if at least one of the conditions is `True`.

#### Example:

```python
x = 10
y = 15
if x > 5 or y < 10:
    print("At least one condition is True")
else:
    print("Both conditions are False")
```

**Output:**

```
At least one condition is True
```

#### 3.3 Using `not`

The `not` operator reverses the Boolean value of a condition.

#### Example:

```python
x = 10
if not(x < 5):
    print("x is not less than 5")
else:
    print("x is less than 5")
```

**Output:**

```
x is not less than 5
```

---

### **4. Ternary Operator (Conditional Expression)**

Python supports a shorthand method for `if-else` known as the ternary operator (or conditional expression). This allows you to write simple `if-else` conditions in a single line.

#### Syntax:

```python
value_if_true if condition else value_if_false
```

#### Example:

```python
x = 10
result = "Greater than 5" if x > 5 else "Less than or equal to 5"
print(result)
```

**Output:**

```
Greater than 5
```

---

### **5. Using `in` and `not in` for Membership Tests**

The `in` and `not in` operators allow you to check if a value exists in a sequence (such as a list, tuple, or string).

#### Syntax:

```python
if value in sequence:
    # Code to execute if value is in sequence
else:
    # Code to execute if value is not in sequence
```

#### Example:

```python
my_list = [1, 2, 3, 4, 5]
if 3 in my_list:
    print("3 is in the list")
else:
    print("3 is not in the list")
```

**Output:**

```
3 is in the list
```

---

### **6. Using `is` and `is not` for Identity Tests**

The `is` and `is not` operators check whether two variables refer to the same object in memory, not just whether their values are equal.

#### Example:

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

if a is b:
    print("a and b refer to the same object")
else:
    print("a and b do not refer to the same object")

if a is not c:
    print("a and c do not refer to the same object")
```

**Output:**

```
a and b refer to the same object
a and c do not refer to the same object
```

---

### **7. Combining `if-elif-else` with Loops**

The `if-elif-else` statements are often used within loops to control the flow based on certain conditions. You can combine `if` statements inside `for` or `while` loops for more complex logic.

#### Example with `for` loop:

```python
for x in range(5):
    if x == 0:
        print("x is zero")
    elif x % 2 == 0:
        print(f"x is even: {x}")
    else:
        print(f"x is odd: {x}")
```

**Output:**

```
x is zero
x is even: 2
x is even: 4
```

---

### **8. Using `else` with Loops**

In Python, the `else` block can be used after a `for` or `while` loop. The `else` part will execute only if the loop completes without encountering a `break`.

#### Example with `while` loop:

```python
x = 0
while x < 5:
    if x == 3:
        break
    x += 1
else:
    print("Loop completed without breaking")
```

**Output:**

```
(no output since the loop breaks when `x == 3`)
```

---

### **9. Chaining Multiple `elif` Conditions**

You can use multiple `elif` statements to check more than two conditions in a specific order. Python will execute the first `True` condition and skip the rest.

#### Example:

```python
x = 15
if x < 10:
    print("x is less than 10")
elif x >= 10 and x < 20:
    print("x is between 10 and 20")
elif x >= 20 and x < 30:
    print("x is between 20 and 30")
else:
    print("x is 30 or greater")
```

**Output:**

```
x is between 10 and 20
```

---

### **Summary of `if`, `elif`, and `else`**

- **`if`**: Tests a condition and executes the block of code if the condition is `True`.
- **`else`**: Executes the block of code if the `if` condition is `False`.
- **`elif`**: Checks additional conditions if the previous `if` or `elif` conditions are `False`.
- **Logical operators (`and`, `or`, `not`)**: Combine multiple conditions.
- **Ternary operator**: A concise way to write `if-else` in a single line.
- **`in` and `not in`**: Check if a value is in a sequence (list, string, etc.).
- **`is` and `is not`**: Test if two variables refer to the same object.
- **Chaining multiple `elif`**:

Allows you to check multiple conditions in a specific order.

Understanding these conditional statements allows you to write more dynamic, efficient, and flexible Python code, handling a wide range of situations based on varying conditions.
