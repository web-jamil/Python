# Python Nested `if` statement

We can have a nested-**[if-else](https://github.com/milaan9/03_Python_Flow_Control/blob/main/002_Python_if_else_statement.ipynb)** or nested-**[if-elif-else](https://github.com/milaan9/03_Python_Flow_Control/blob/main/003_Python_if_elif_else_statement%20.ipynb)** statement inside another **`if-else`** statement. This is called **nesting** in computer programming. The nested if statements is useful when we want to make a series of decisions.

Any number of these statements can be nested inside one another. Indentation is the only way to figure out the level of nesting. They can get confusing, so they must be avoided unless necessary.

We can use nested if statements for situations where we want to check for a **secondary condition** if the first condition executes as **`True`**.


### Syntax:

#### Example 1:

```python
if conditon_outer:
    if condition_inner:
        statement of nested if
    else:
        statement of nested if else:
    statement ot outer if
else:
    Outer else
statement outside if block
```

#### Example 2:

```python
if expression1:
    statement(s)
    if expression2:
        statement(s)
    elif expression3:
        statement(s)
    elif expression4:
        statement(s)
    else:
        statement(s)
else:
    statement(s)
```


````python
# Example 1:

a=10
if a>=20:  # Condition FALSE
    print ("Condition is True")
else:  # Code will go to ELSE body
    if a>=15:  # Condition FALSE
        print ("Checking second value")
    else:  # Code will go to ELSE body
        print ("All Conditions are false")
````

````python
# Example 2:

x = 10
y = 12
if x > y:
    print( "x>y")
elif x < y:
    print( "x<y")
    if x==10:
        print ("x=10")
    else:
        print ("invalid")
else:
    print ("x=y")
````

````python
# Example 3:

num1 = 0
if (num1 != 0):  # For zero condition is FALSE
    if(num1 > 0):  
        print("num1 is a positive number")
    else:  
        print("num1 is a negative number")
else:  # For zero condition is TRUE
    print("num1 is neither positive nor negative")
````

````python
# Example 4:

'''In this program, we input a number check if the number is 
positive or negative or zero and display an appropriate message. 
This time we use nested if statement'''

num = float(input("Enter a number: "))
if num >= 0:
    if num == 0:
        print("Zero")
    else:
        print("Positive number")
else:
    print("Negative number")
````

````python
# Example 5:

def number_arithmetic(num1, num2):
    if num1 >= num2:
        if num1 == num2:
            print(f'{num1} and {num2} are equal')
        else:
            print(f'{num1} is greater than {num2}')
    else:
        print(f'{num1} is smaller than {num2}')

number_arithmetic(96, 66)
# Output 96 is greater than 66
number_arithmetic(96, 96)
# Output 56 and 56 are equal
````

## **1. Basic Nested `if`**

An `if` statement inside another `if`.

### **Syntax**

```python
if condition1:
    if condition2:
        # code if both conditions are True
```

### **Example**

```python
age = 20
if age >= 18:
    if age <= 30:
        print("You're a young adult")
```

**Output:** `You're a young adult`

---

## **2. Nested `if-else`**

Combining `if-else` inside another `if`.

### **Syntax**

```python
if condition1:
    if condition2:
        # code if condition1 and condition2 are True
    else:
        # code if condition1 is True but condition2 is False
else:
    # code if condition1 is False
```

### **Example**

```python
num = 15
if num > 10:
    if num % 2 == 0:
        print(">10 and even")
    else:
        print(">10 and odd")
else:
    print("<=10")
```

**Output:** `>10 and odd`

---

## **3. Multi-level Nesting**

Multiple levels of nested `if` statements.

### **Example**

```python
x = 25
if x > 10:
    if x < 30:
        if x % 5 == 0:
            print("Between 10-30 and divisible by 5")
        else:
            print("Between 10-30 but not divisible by 5")
    else:
        print(">=30")
else:
    print("<=10")
```

**Output:** `Between 10-30 and divisible by 5`

## **4. Nested `if` with Logical Operators**

Using `and`, `or` to simplify nested conditions.

### **Example (Before)**

```python
if x > 10:
    if y > 5:
        print("Both conditions met")
```

### **Example (After - Simplified)**


## **5. Nested Ternary Operations**

Nested conditional expressions in one line.

### **Syntax**

```python
result = value1 if condition1 else (value2 if condition2 else value3)
```

### **Example**

```python
age = 22
status = "Adult" if age >= 18 else ("Teen" if age >= 13 else "Child")
print(status)  # Output: Adult
```

## **7. Nested `if` in Functions**

Using nested conditions in functions.

### **Example**

```python
def check_triangle(a, b, c):
    if a + b > c:
        if a + c > b:
            if b + c > a:
```

## **8. Avoiding Deep Nesting**

Deep nesting reduces readability. Solutions:

### **A. Use Early Returns**

```python
def process_data(data):
    if not data:
        return "No data"

    if not validate(data):
        return "Invalid data"

    # Main processing
    return "Success"
```

### **B. Combine Conditions**

```python
if (condition1 and
    condition2 and
    condition3):
    # do something
```

### **C. Use Functions**

```python
def all_conditions_met(x, y, z):
    return (x > 10 and
            y < 20 and
            z % 2 == 0)

if all_conditions_met(15, 10, 4):
    print("All conditions met")
```

### Nested `if` Statements in Python: From Basic to Advanced

In Python, **nested `if` statements** refer to placing one `if` statement inside another. This allows you to check multiple conditions in a structured way. Nested `if` statements are commonly used for more complex decision-making logic where one condition is dependent on another.

---

## **1. Basic Syntax of Nested `if` Statements**

### Syntax:

```python
if condition1:
    # Code to execute if condition1 is True
    if condition2:
        # Code to execute if condition2 is also True
    else:
        # Code to execute if condition2 is False
else:
    # Code to execute if condition1 is False
```

### Example:

```python
x = 10
y = 5

if x > 5:
    if y < 10:
        print("x is greater than 5 and y is less than 10")
    else:
        print("x is greater than 5 but y is not less than 10")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is greater than 5 and y is less than 10
```

---

## **2. Nested `if-else` Statements**

You can have both `if` and `else` inside the nested block to handle both conditions when the first `if` condition is `True`.

### Example:

```python
x = 5
y = 8

if x < 10:
    if y < 10:
        print("Both x and y are less than 10")
    else:
        print("x is less than 10 but y is not")
else:
    print("x is not less than 10")
```

**Output:**

```
Both x and y are less than 10
```

---

## **3. Multiple Levels of Nesting**

You can nest `if` statements to more than two levels. The general principle is the same: the inner `if` or `else` is checked only if the outer condition is `True`.

### Example (Multiple Levels of Nesting):

```python
x = 15
y = 10
z = 5

if x > 10:
    if y > 10:
        print("x is greater than 10 and y is greater than 10")
    else:
        if z > 5:
            print("x is greater than 10, but y is not and z is greater than 5")
        else:
            print("x is greater than 10, but y is not and z is not greater than 5")
else:
    print("x is not greater than 10")
```

**Output:**

```
x is greater than 10, but y is not and z is greater than 5
```

---

## **4. Using Logical Operators Inside Nested `if`**

You can combine logical operators like `and`, `or`, and `not` inside nested `if` conditions to check multiple conditions.

### Example with `and` and `or`:

```python
x = 12
y = 8
z = 5

if x > 10:
    if y > 5 and z < 10:
        print("x is greater than 10, y is greater than 5 and z is less than 10")
    else:
        print("x is greater than 10, but either y is not greater than 5 or z is not less than 10")
else:
    print("x is not greater than 10")
```

**Output:**

```
x is greater than 10, y is greater than 5 and z is less than 10
```

---

## **5. Using `elif` with Nested `if` Statements**

You can also use `elif` inside nested `if` blocks. This allows for more efficient condition checking without creating too many levels of nesting.

### Example:

```python
x = 10
y = 20

if x > 5:
    if y < 25:
        print("x is greater than 5 and y is less than 25")
    elif y == 20:
        print("x is greater than 5 and y is equal to 20")
    else:
        print("x is greater than 5, but y is not less than 25")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is greater than 5 and y is equal to 20
```

---

## **6. Using `else` with Nested `if` Statements**

Just like `if`, you can combine `else` with nested `if` statements. The `else` block will execute if none of the `if` conditions inside the nested block are `True`.

### Example:

```python
x = 8
y = 10

if x > 5:
    if y > 15:
        print("x is greater than 5 and y is greater than 15")
    else:
        print("x is greater than 5, but y is not greater than 15")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is greater than 5, but y is not greater than 15
```

---

## **7. Nested `if` Inside a Loop**

You can combine loops and nested `if` statements to create more complex decision-making processes.

### Example with `for` Loop:

```python
numbers = [10, 15, 20, 5, 30]
for number in numbers:
    if number > 10:
        if number % 2 == 0:
            print(f"{number} is greater than 10 and is even")
        else:
            print(f"{number} is greater than 10 but is odd")
    else:
        print(f"{number} is less than or equal to 10")
```

**Output:**

```
15 is greater than 10 but is odd
20 is greater than 10 and is even
5 is less than or equal to 10
30 is greater than 10 and is even
```

---

## **8. Performance Considerations**

### Too many nested `if` statements can lead to a decrease in readability and maintainability. Here are some tips to improve your nested `if` statements:

- **Avoid deep nesting:** Try to avoid nesting `if` statements more than 2-3 levels deep. It can quickly become difficult to maintain and understand.
- **Use functions:** If you have deeply nested `if` blocks, consider refactoring parts of your code into separate functions to improve readability.
- **Use logical operators:** Combine conditions using `and`/`or` to simplify nested `if` statements.

## **9. Real-World Example of Nested `if`**

Let’s say we have a scenario where we need to classify a person’s age group based on their age and gender. We'll use nested `if` to categorize them.

### Example:

```python

    if gender == 'male':
        print("Child, Male")
    else:

    if gender == 'male':
        print("Adult, Male")
    else:
        print("Adult, Female")

    if gender == 'male':
        print("Senior, Male")
    else:
        print("Senior, Female")
```

## **Summary of Nested `if` Statements**

1. **Basic Structure**: You can place `if` statements inside another `if` block, allowing you to check multiple conditions in a structured way.
2. **`if-else`**: You can have `else` within a nested block to handle the `False` case for inner conditions.
3. **Logical Operators**: Combining `and`, `or`, and `not` with `if` conditions helps create more compact nested conditions.
4. **`elif` and `else`**: These help manage multiple possible conditions and handle cases when the initial conditions are `False`.
5. **Loops and Nesting**: You can use loops such as `for` and `while` to repeat the logic inside nested `if` statements.
6. **Performance**: Be mindful of performance and readability when nesting too many levels.

Nested `if` statements are a powerful tool for handling complex decision logic, but they should be used wisely to keep your code clean and maintainable.
