## Python Control Flow Statements: From Basic to Advanced

Control flow statements allow the program to make decisions and execute different code based on conditions. They are crucial for implementing logic in a program. In Python, the primary control flow statements include conditional statements, loops, and exception handling. Here's an overview of control flow in Python, from basic to advanced.

---

### **1. Conditional Statements**

Conditional statements allow the program to decide which block of code to execute based on a condition (True or False).

#### 1.1. **`if` statement**

The `if` statement is used to execute a block of code only if a condition is True.

```python
x = 10
if x > 5:
    print("x is greater than 5")
```

**Output:**

```
x is greater than 5
```

#### 1.2. **`if-else` statement**

The `if-else` statement allows two different blocks of code to execute: one if the condition is True, and another if it's False.

```python
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")
```

**Output:**

```
x is not greater than 5
```

#### 1.3. **`if-elif-else` statement**

The `if-elif-else` statement allows you to check multiple conditions in sequence.

```python
x = 7
if x > 10:
    print("x is greater than 10")
elif x == 7:
    print("x is equal to 7")
else:
    print("x is less than 7")
```

**Output:**

```
x is equal to 7
```

---

### **2. Looping Statements**

Loops allow you to repeatedly execute a block of code multiple times. Python supports two types of loops: `for` and `while`.

#### 2.1. **`for` loop**

The `for` loop is used to iterate over a sequence (e.g., list, tuple, string, range).

```python
for i in range(5):
    print(i)
```

**Output:**

```
0
1
2
3
4
```

In the above example, the loop iterates over the range from 0 to 4 and prints each value.

#### 2.2. **`while` loop**

The `while` loop is used to execute a block of code as long as a condition is True.

```python
x = 0
while x < 5:
    print(x)
    x += 1
```

**Output:**

```
0
1
2
3
4
```

In this case, the loop continues as long as `x` is less than 5.

---

### **3. Control Flow Modifiers**

#### 3.1. **`break` Statement**

The `break` statement is used to exit a loop prematurely when a certain condition is met.

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

**Output:**

```
0
1
2
3
4
```

The loop exits when `i` reaches 5 due to the `break` statement.

#### 3.2. **`continue` Statement**

The `continue` statement is used to skip the current iteration of a loop and continue with the next iteration.

```python
for i in range(5):
    if i == 3:
        continue
    print(i)
```

**Output:**

```
0
1
2
4
```

When `i` is 3, the `continue` statement skips the printing of that iteration and goes to the next one.

#### 3.3. **`pass` Statement**

The `pass` statement is a null operation. It serves as a placeholder for future code.

```python
if True:
    pass  # No action, placeholder for future code
else:
    print("This won't execute")
```

The `pass` statement doesn't do anything but is syntactically necessary when a statement is required but no action is needed.

---

### **4. Exception Handling (Advanced Control Flow)**

Exception handling allows you to handle runtime errors (exceptions) gracefully and continue program execution.

#### 4.1. **`try-except` Block**

The `try-except` block lets you catch exceptions and handle them without terminating the program.

```python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

**Output:**

```
Cannot divide by zero
```

#### 4.2. **`try-except-else` Block**

The `else` block is executed if no exception occurs in the `try` block.

```python
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Division successful")
```

**Output:**

```
Division successful
```

#### 4.3. **`try-except-finally` Block**

The `finally` block always executes, regardless of whether an exception was raised or not.

```python
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("This will always execute")
```

**Output:**

```
This will always execute
```

#### 4.4. **Raising Exceptions**

You can manually raise exceptions using the `raise` keyword.

```python
x = -1
if x < 0:
    raise ValueError("x cannot be negative")
```

This will raise a `ValueError` and stop the program with a message.

---

### **5. Advanced Control Flow Concepts**

#### 5.1. **Lambda Functions (Anonymous Functions)**

Lambda functions allow you to define short, unnamed functions in a single line. They're commonly used with control flow functions like `filter()`, `map()`, and `reduce()`.

```python
square = lambda x: x ** 2
print(square(5))  # Output: 25
```

#### 5.2. **List Comprehensions**

List comprehensions provide a concise way to create lists. They can also include conditional logic for filtering elements.

```python
squares = [x ** 2 for x in range(5)]
print(squares)  # Output: [0, 1, 4, 9, 16]
```

You can also add `if` conditions to filter items:

```python
even_squares = [x ** 2 for x in range(5) if x % 2 == 0]
print(even_squares)  # Output: [0, 4, 16]
```

#### 5.3. **Generators and `yield`**

Generators are special types of iterators that allow you to iterate over data lazily. The `yield` keyword is used to create a generator.

```python
def count_up_to(max):
    count = 1
    while count <= max:
        yield count
        count += 1

counter = count_up_to(5)
for num in counter:
    print(num)
```

**Output:**

```
1
2
3
4
5
```

The `yield` statement generates a sequence of values without holding everything in memory at once.

---

### **6. Conditional Expressions (Ternary Operator)**

Python has a one-line conditional expression, commonly referred to as the ternary operator, which allows conditional logic in a single expression.

```python
x = 10
result = "Positive" if x > 0 else "Non-positive"
print(result)  # Output: Positive
```

The syntax is:

```python
result = value_if_true if condition else value_if_false
```

### **7. Summary of Python Control Flow Statements**

| **Control Flow Statement** | **Description**                                               |
| -------------------------- | ------------------------------------------------------------- |
| `if`                       | Conditional statement to execute code if condition is True    |
| `else`                     | Executes code when the condition is False                     |
| `elif`                     | Checks another condition if the previous ones are False       |
| `for`                      | Loop that iterates over a sequence (list, range, etc.)        |
| `while`                    | Loop that continues as long as a condition is True            |
| `break`                    | Exits a loop prematurely                                      |
| `continue`                 | Skips the current iteration of the loop                       |
| `pass`                     | Placeholder for empty code block                              |
| `try-except`               | Exception handling to catch and handle errors                 |
| `try-except-else`          | `else` block runs if no exception occurs                      |
| `try-except-finally`       | `finally` block runs regardless of exceptions                 |
| `raise`                    | Used to explicitly raise an exception                         |
| `lambda`                   | Anonymous function for short, one-line function calls         |
| List Comprehension         | Concise way to create lists, with optional filtering          |
| Generators & `yield`       | Create lazy iterators to avoid memory usage in large datasets |
| Ternary Expression         | A shorthand for simple if-else conditions                     |

## Basic Control Flow

### Conditional Statements (`if`, `elif`, `else`)

```python
# Basic if statement
x = 10
if x > 5:
    print("x is greater than 5")

# if-else
if x > 15:
    print("x is greater than 15")
else:
    print("x is 15 or less")

# if-elif-else
if x > 20:
    print("x is greater than 20")
elif x > 10:
    print("x is greater than 10 but 20 or less")
else:
    print("x is 10 or less")
```

### Loops (`while`, `for`)

```python
# while loop
count = 0
while count < 5:
    print(count)
    count += 1

# for loop with range
for i in range(5):
    print(i)

# for loop with sequence
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)

# Iterating over dictionaries
person = {'name': 'John', 'age': 30}
for key, value in person.items():
    print(f"{key}: {value}")
```

### Loop Control (`break`, `continue`, `pass`)

```python
# break - exits the loop
for i in range(10):
    if i == 5:
        break
    print(i)  # Prints 0-4

# continue - skips to next iteration
for i in range(5):
    if i == 2:
        continue
    print(i)  # Prints 0,1,3,4

# pass - does nothing (placeholder)
for i in range(5):
    if i == 3:
        pass  # TODO: implement later
    print(i)
```

## Intermediate Concepts

### Nested Control Structures

```python
# Nested if statements
x = 15
if x > 10:
    if x < 20:
        print("x is between 10 and 20")

# Nested loops
for i in range(3):
    for j in range(2):
        print(f"({i}, {j})")

# Loop with conditional
numbers = [1, 2, 3, 4, 5, 6]
for num in numbers:
    if num % 2 == 0:
        print(f"{num} is even")
    else:
        print(f"{num} is odd")
```

### The `else` Clause in Loops

```python
# else with for loop
for i in range(3):
    print(i)
else:
    print("Loop completed successfully")  # Executes if loop completes without break

# else with while loop
n = 5
while n > 0:
    print(n)
    n -= 1
else:
    print("Countdown finished")

# else NOT executed when break occurs
for i in range(3):
    if i == 1:
        break
    print(i)
else:
    print("This won't print due to break")
```

### Ternary Conditional Expressions

```python
# Traditional if-else
x = 10
if x > 5:
    result = "Greater"
else:
    result = "Smaller"

# Ternary equivalent
result = "Greater" if x > 5 else "Smaller"

# Can be nested (though less readable)
y = 15
classification = "High" if y > 20 else ("Medium" if y > 10 else "Low")
```

### Iterators and Generators

```python
# Iterator protocol
my_list = [1, 2, 3]
iterator = iter(my_list)
print(next(iterator))  # 1
print(next(iterator))  # 2

# Generator function
def count_up_to(max):
    count = 1
    while count <= max:
        yield count  # Pauses here and returns count
        count += 1

counter = count_up_to(3)
print(next(counter))  # 1
print(next(counter))  # 2

# Generator expression
squares = (x*x for x in range(5))
print(list(squares))  # [0, 1, 4, 9, 16]
```

## Advanced Concepts

### Context Managers (`with`)

```python
# File handling with context manager
with open('file.txt', 'r') as file:
    content = file.read()
# File automatically closed here

# Custom context manager class
class ManagedFile:
    def __init__(self, filename):
        self.filename = filename

    def __enter__(self):
        self.file = open(self.filename, 'r')
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.file:
            self.file.close()

with ManagedFile('file.txt') as f:
    content = f.read()

# Using contextlib for simpler context managers
from contextlib import contextmanager

@contextmanager
def managed_file(filename):
    try:
        f = open(filename, 'r')
        yield f
    finally:
        f.close()

with managed_file('file.txt') as f:
    content = f.read()
```

### Exception Handling (`try`, `except`, `finally`)

```python
# Basic exception handling
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")

# Multiple exceptions
try:
    # code that might raise exceptions
    value = int("not a number")
except (ValueError, TypeError) as e:
    print(f"Error: {e}")

# Else clause (executes if no exception)
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Division by zero")
else:
    print(f"Result is {result}")

# Finally clause (always executes)
try:
    file = open('file.txt', 'r')
    content = file.read()
except FileNotFoundError:
    print("File not found")
finally:
    file.close() if 'file' in locals() else None
    print("Cleanup complete")

# Raising exceptions
def validate_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

# Custom exceptions
class MyCustomError(Exception):
    pass

try:
    raise MyCustomError("Something went wrong")
except MyCustomError as e:
    print(f"Caught custom error: {e}")
```

### Pattern Matching (Python 3.10+)

```python
# Basic pattern matching
status = 404

match status:
    case 200:
        print("OK")
    case 400:
        print("Bad request")
    case 404:
        print("Not found")
    case _:  # Default case
        print("Unknown status code")

# Complex pattern matching
def handle_command(command):
    match command.split():
        case ["quit"]:
            print("Quitting...")
        case ["go", direction]:
            print(f"Going {direction}")
        case ["drop", *items]:
            print(f"Dropping items: {', '.join(items)}")
        case _:
            print("Unknown command")

handle_command("go north")  # Going north
handle_command("drop sword shield")  # Dropping items: sword, shield
```

### Coroutines and `async/await`

```python
import asyncio

# Basic coroutine
async def greet(name):
    print(f"Hello, {name}")
    await asyncio.sleep(1)  # Simulate I/O operation
    print(f"Goodbye, {name}")

# Running coroutines
async def main():
    await greet("Alice")
    await greet("Bob")

asyncio.run(main())

# Concurrent execution with tasks
async def concurrent_main():
    task1 = asyncio.create_task(greet("Alice"))
    task2 = asyncio.create_task(greet("Bob"))
    await task1
    await task2

asyncio.run(concurrent_main())

# Asynchronous generator
async def async_counter(max):
    for i in range(max):
        yield i
        await asyncio.sleep(0.1)

async def consume_async_gen():
    async for number in async_counter(5):
        print(number)

asyncio.run(consume_async_gen())
```
