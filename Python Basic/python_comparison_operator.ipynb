Python **comparison operators** are used to compare two values and return a **Boolean result** (`True` or `False`). They are essential for **conditional statements** and **looping constructs**.

---

## **1. Basic Comparison Operators**

Python provides **six** fundamental comparison operators:

| Operator | Meaning                  | Example           |
| -------- | ------------------------ | ----------------- |
| `==`     | Equal to                 | `5 == 5` → `True` |
| `!=`     | Not equal to             | `5 != 3` → `True` |
| `>`      | Greater than             | `5 > 3` → `True`  |
| `<`      | Less than                | `3 < 5` → `True`  |
| `>=`     | Greater than or equal to | `5 >= 5` → `True` |
| `<=`     | Less than or equal to    | `3 <= 5` → `True` |

### **Example Usage**

```python
x = 10
y = 20

print(x == y)  # False
print(x != y)  # True
print(x > y)   # False
print(x < y)   # True
print(x >= y)  # False
print(x <= y)  # True
```

Python allows comparisons between **numbers, strings, and other objects**.

### **Numbers**

```python

```

### **Strings (Lexicographical Order)**

```python

```

### **Lists and Tuples**

```python
print([1, 2, 3] == [1, 2, 3])  # True
print((1, 2) < (1, 3))  # True (compares element-wise)
```

### 1. Basic Comparison Operators

These are the most common and straightforward comparison operators.

| Operator | Meaning                  | Example    | Result |
| :------- | :----------------------- | :--------- | :----- |
| `==`     | Equal to                 | `5 == 5`   | `True` |
| `!=`     | Not equal to             | `5 != 10`  | `True` |
| `>`      | Greater than             | `10 > 5`   | `True` |
| `<`      | Less than                | `5 < 10`   | `True` |
| `>=`     | Greater than or equal to | `10 >= 10` | `True` |
| `<=`     | Less than or equal to    | `5 <= 5`   | `True` |

**Code Examples:**

```python
# Equality (==) and Inequality (!=)
print(f"5 == 5: {5 == 5}")       # True
print(f"5 == 10: {5 == 10}")     # False
print(f"'hello' == 'hello': {'hello' == 'hello'}") # True
print(f"'hello' == 'Hello': {'hello' == 'Hello'}") # False (case-sensitive)
print(f"5 != 10: {5 != 10}")     # True
print(f"5 != 5: {5 != 5}")       # False

# Greater than (>) and Less than (<)
print(f"10 > 5: {10 > 5}")       # True
print(f"5 > 10: {5 > 10}")       # False
print(f"5 < 10: {5 < 10}")       # True
print(f"10 < 5: {10 < 5}")       # False

# Greater than or equal to (>=) and Less than or equal to (<=)
print(f"10 >= 10: {10 >= 10}")   # True
print(f"10 >= 5: {10 >= 5}")     # True
print(f"5 >= 10: {5 >= 10}")     # False
print(f"5 <= 5: {5 <= 5}")       # True
print(f"5 <= 10: {5 <= 10}")     # True
print(f"10 <= 5: {10 <= 5}")     # False
```

### 2. Chaining Comparison Operators (Advanced)

Python allows you to chain comparison operators, which is a very concise and readable way to express multiple comparisons.

The expression `a < b < c` is equivalent to `(a < b) and (b < c)`. This makes your code cleaner than using explicit `and` operators.

**Syntax:** `value1 operator1 value2 operator2 value3 ...`

**Code Example:**

```python
age = 25

# Chaining to check if age is between 18 and 30 (inclusive)
print(f"18 <= age <= 30: {18 <= age <= 30}") # True

age = 15
print(f"18 <= age <= 30: {18 <= age <= 30}") # False (because 18 <= 15 is False)

temperature = 28

# Check if temperature is comfortable (between 20 and 30, but not exactly 25)
print(f"20 < temperature < 30 and temperature != 25: {20 < temperature < 30 and temperature != 25}") # True

# This is equivalent to:
print(f"(20 < temperature) and (temperature < 30) and (temperature != 25): {(20 < temperature) and (temperature < 30) and (temperature != 25)}") # True
```

### 3. Comparison with Different Data Types

How comparison operators behave depends on the data types being compared.

#### 3.1. Numbers (Integers, Floats)

Numerical comparisons are straightforward, based on their mathematical values.

```python
print(f"10 == 10.0: {10 == 10.0}") # True (values are equal, types don't prevent comparison)
print(f"10 > 9.99: {10 > 9.99}")   # True
print(f"10 < float('inf'): {10 < float('inf')}") # True (infinity)
print(f"0 == -0.0: {0 == -0.0}")   # True
print(f"float('nan') == float('nan'): {float('nan') == float('nan')}") # False (NaN is never equal to anything, including itself)
print(f"float('nan') != float('nan'): {float('nan') != float('nan')}") # True
```

**Important:** Be cautious with floating-point comparisons due to potential precision issues. Instead of `==`, sometimes checking if the absolute difference is very small is preferred: `abs(a - b) < epsilon`.

#### 3.2. Strings (Lexicographical Comparison)

Strings are compared character by character based on their Unicode (or ASCII) values. This is called **lexicographical comparison**.

```python
print(f"'apple' == 'apple': {'apple' == 'apple'}")     # True
print(f"'apple' == 'Apple': {'apple' == 'Apple'}")     # False (case-sensitive)
print(f"'apple' < 'banana': {'apple' < 'banana'}")     # True (a comes before b)
print(f"'banana' > 'apple': {'banana' > 'apple'}")     # True
print(f"'cat' < 'dog': {'cat' < 'dog'}")             # True
print(f"'cat' < 'catalog': {'cat' < 'catalog'}")     # True (shorter string is "less" if it's a prefix of the longer)
print(f"'Z' < 'a': {'Z' < 'a'}")                     # True (Uppercase letters have lower Unicode values than lowercase)
print(f"'10' > '2': {'10' > '2'}")                   # False (Lexicographical, compares '1' with '2')
```

#### 3.3. Lists, Tuples, and other Sequences (Lexicographical Comparison)

Sequences (lists, tuples, etc.) are also compared lexicographically, element by element.

1.  Python compares the first elements of each sequence.
2.  If they are different, that difference determines the result.
3.  If they are equal, it moves to the next elements and compares them.
4.  This continues until a difference is found or one sequence is exhausted.
5.  If one sequence is a prefix of another (e.g., `[1, 2]` vs `[1, 2, 3]`), the shorter sequence is considered "less than" the longer one.

```python
print(f"[1, 2, 3] == [1, 2, 3]: {[1, 2, 3] == [1, 2, 3]}") # True
print(f"[1, 2, 3] != [1, 2, 4]: {[1, 2, 3] != [1, 2, 4]}") # True
print(f"[1, 5, 3] > [1, 2, 10]: {[1, 5, 3] > [1, 2, 10]}") # True (because 5 > 2)
print(f"(1, 2) < (1, 2, 3): {(1, 2) < (1, 2, 3)}")       # True (shorter prefix is less)
print(f"[1, 2] == (1, 2): {[1, 2] == (1, 2)}")           # False (different types, even if elements are same)
```

**Note:** Python 2 allowed comparing sequences of different types (e.g., `[1] < "hello"`), but this is explicitly disallowed in Python 3 and will raise a `TypeError`. You can only compare sequences of the same type or where one type can be implicitly converted (like `int` and `float`).

#### 3.4. Dictionaries and Sets

- **Dictionaries (`dict`)**: Can only be compared for equality (`==`) and inequality (`!=`). They are equal if and only if they have the same key-value pairs (regardless of insertion order). Other comparisons (`<`, `>`, etc.) are not supported and will raise a `TypeError`.

  ```python
  print(f"{{'a': 1, 'b': 2}} == {{'b': 2, 'a': 1}}: {{'a': 1, 'b': 2}} == {{'b': 2, 'a': 1}}") # True (order doesn't matter)
  print(f"{{'a': 1}} == {{'a': 2}}: {{'a': 1}} == {{'a': 2}}") # False
  # print({'a': 1} < {'b': 2}) # TypeError
  ```

- **Sets (`set`)**: Support equality and inequality, as well as subset/superset relationships using comparison operators.

  - `==`: True if both sets have the same elements.
  - `!=`: True if sets do not have the same elements.
  - `<=`: Subset (A is a subset of B, A `subseteq` B).
  - `<`: Proper subset (A is a subset of B, but A `!=` B).
  - `>=`: Superset (A is a superset of B, A `supseteq` B).
  - `>`: Proper superset (A is a superset of B, but A `!=` B).

  ```python
  set1 = {1, 2, 3}
  set2 = {3, 2, 1}
  set3 = {1, 2, 3, 4}
  set4 = {1, 5}

  print(f"{set1} == {set2}: {set1 == set2}")   # True
  print(f"{set1} <= {set3}: {set1 <= set3}")   # True (set1 is a subset of set3)
  print(f"{set1} < {set3}: {set1 < set3}")     # True (set1 is a proper subset of set3)
  print(f"{set3} >= {set1}: {set3 >= set1}")   # True (set3 is a superset of set1)
  print(f"{set3} > {set1}: {set3 > set1}")     # True (set3 is a proper superset of set1)
  print(f"{set1} <= {set4}: {set1 <= set4}")   # False
  ```

#### 3.5. `None`

`None` can only be compared for equality or inequality. It is equal only to itself.

```python
x = None
y = 10
print(f"x == None: {x == None}") # True
print(f"x != y: {x != y}")     # True
# print(x < y) # TypeError
```

**Best Practice:** When checking for `None`, it's idiomatic in Python to use `is` or `is not` for identity comparison, as `None` is a singleton object.
`x is None` is generally preferred over `x == None`.

#### 3.6. Custom Objects (Magic Methods) - Advanced

For custom classes, comparison operators work by calling special "magic methods" (also known as "dunder methods" due to their double underscores). If these methods are not defined, Python falls back to object identity comparison (`is`) for `==` and `!=`, and raises `TypeError` for ordering comparisons.

- `__eq__(self, other)`: Implements `==`
- `__ne__(self, other)`: Implements `!=`
- `__lt__(self, other)`: Implements `<`
- `__le__(self, other)`: Implements `<=`
- `__gt__(self, other)`: Implements `>`
- `__ge__(self, other)`: Implements `>=`

**Code Example:**

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Define equality: two points are equal if their x and y coordinates are the same
    def __eq__(self, other):
        if not isinstance(other, Point):
            return NotImplemented # Or raise TypeError
        return self.x == other.x and self.y == other.y

    # __ne__ is automatically derived from __eq__ if not defined,
    # but you can define it explicitly if needed.
    # def __ne__(self, other):
    #     return not self.__eq__(other)

    # Define less than: compare based on x, then y if x's are equal
    def __lt__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        if self.x == other.x:
            return self.y < other.y
        return self.x < other.x

    # Using functools.total_ordering to automatically generate >=, >, <=
    # based on __eq__ and __lt__
    from functools import total_ordering
    @total_ordering
    class Point:
        def __init__(self, x, y):
            self.x = x
            self.y = y

        def __eq__(self, other):
            if not isinstance(other, Point):
                return NotImplemented
            return self.x == other.x and self.y == other.y

        def __lt__(self, other):
            if not isinstance(other, Point):
                return NotImplemented
            if self.x == other.x:
                return self.y < other.y
            return self.x < other.x

        def __repr__(self): # For better printing of objects
            return f"Point({self.x}, {self.y})"

p1 = Point(1, 2)
p2 = Point(1, 2)
p3 = Point(3, 4)
p4 = Point(1, 3)

print("\n--- Custom Object Comparisons ---")
print(f"p1 == p2: {p1 == p2}") # True (due to __eq__)
print(f"p1 == p3: {p1 == p3}") # False
print(f"p1 != p3: {p1 != p3}") # True (derived from __eq__)

print(f"p1 < p3: {p1 < p3}")   # True (1 < 3)
print(f"p1 < p4: {p1 < p4}")   # True (x's are equal, then 2 < 3)
print(f"p3 > p1: {p3 > p1}")   # True (derived from __lt__ by @total_ordering)
print(f"p1 <= p2: {p1 <= p2}") # True (derived from __lt__ and __eq__ by @total_ordering)

# Comparing with non-Point objects (will rely on NotImplemented)
print(f"p1 == 'hello': {p1 == 'hello'}") # False (or error if not handled gracefully)
```

### 4. Identity Operators (`is` and `is not`) - Advanced

While `==` checks for **value equality**, `is` checks for **object identity**. That is, whether two variables refer to the _exact same object in memory_.

| Operator | Meaning                | Example      | Result |
| :------- | :--------------------- | :----------- | :----- |
| `is`     | Is the same object     | `x is y`     | `True` |
| `is not` | Is not the same object | `x is not y` | `True` |

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1 # list3 refers to the same object as list1

print("\n--- Identity vs. Value Equality ---")
print(f"list1 == list2: {list1 == list2}")     # True (values are equal)
print(f"list1 is list2: {list1 is list2}")     # False (they are separate objects in memory)

print(f"list1 == list3: {list1 == list3}")     # True (values are equal)
print(f"list1 is list3: {list1 is list3}")     # True (they are the exact same object)

# Small integers and short strings are often "interned" by Python for efficiency.
# This can sometimes make `is` behave like `==` for these types, but it's not guaranteed.
a = 5
b = 5
print(f"a == b: {a == b}") # True
print(f"a is b: {a is b}") # True (Python often interns small integers)

s1 = "hello"
s2 = "hello"
print(f"s1 == s2: {s1 == s2}") # True
print(f"s1 is s2: {s1 is s2}") # True (Python often interns short strings)

s3 = "a very long string that is unlikely to be interned"
s4 = "a very long string that is unlikely to be interned"
print(f"s3 == s4: {s3 == s4}") # True
print(f"s3 is s4: {s3 is s4}") # False (likely separate objects)
```

**Rule of Thumb:**

- Use `==` when you want to compare the _values_ of two objects.
- Use `is` when you want to compare if two variables refer to the _exact same object_ in memory.
- Always use `is None` and `is not None` when checking if a variable holds the `None` value.

```python
dict1 = {'a': 1, 'b': 2}
```

### 3. Identity Operators (`is`, `is not`)

These operators compare the **identity** of two objects, meaning if they refer to the _exact same object in memory_. This is different from `==` which compares _values_.

| Operator | Description                                             | Example (Returns True) |
| :------- | :------------------------------------------------------ | :--------------------- |
| `is`     | `True` if both operands refer to the same object        | `x is y`               |
| `is not` | `True` if both operands do not refer to the same object | `x is not y`           |

**Examples:**

```python
list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a # list_c refers to the same object as list_a

print(f"list_a == list_b: {list_a == list_b}") # True (values are equal)
print(f"list_a is list_b: {list_a is list_b}") # False (they are different objects in memory)
print(f"list_a is list_c: {list_a is list_c}") # True (they are the exact same object)

# Small integers and strings are often "interned" by Python for optimization,
# meaning identical values might point to the same object. This is an implementation detail.
num1 = 10
num2 = 10
num3 = 1000
num4 = 1000

print(f"num1 is num2: {num1 is num2}") # True (for small integers, often interned)
print(f"num3 is num4: {num3 is num4}") # False (larger integers typically not interned)

str_x = "hello"
str_y = "hello"
str_z = "hel" + "lo"

print(f"str_x is str_y: {str_x is str_y}") # True (strings can be interned)
print(f"str_x is str_z: {str_x is str_z}") # True (for compile-time constants)
```

**When to use `is` vs. `==`:**

- Use `==` when you want to compare the _values_ of two objects.
- Use `is` when you want to check if two variables refer to the _exact same object_. This is commonly used for singletons like `None`: `if x is None:`.

### 4. Membership Operators (`in`, `not in`)

These operators test whether a value is a **member** of a sequence (string, list, tuple), set, or the keys of a dictionary.

| Operator | Description                                         | Example (Returns True) |
| :------- | :-------------------------------------------------- | :--------------------- |
| `in`     | `True` if value is found in the sequence/collection | `'a' in 'apple'`       |
| `not in` | `True` if value is not found                        | `'z' not in 'apple'`   |

**Examples:**

```python
my_list = [10, 20, 30, 40, 50]
print(f"30 in my_list: {30 in my_list}")         # True
print(f"60 in my_list: {60 in my_list}")         # False
print(f"100 not in my_list: {100 not in my_list}") # True

my_string = "Python Programming"
print(f"'Python' in my_string: {'Python' in my_string}")       # True
print(f"'java' in my_string: {'java' in my_string}")         # False
print(f"'prog' not in my_string: {'prog' not in my_string}") # False

my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(f"'name' in my_dict: {'name' in my_dict}")     # True (checks keys)
print(f"'Alice' in my_dict: {'Alice' in my_dict}")   # False (does not check values directly)
print(f"'country' not in my_dict: {'country' not in my_dict}") # True
```

### 5. Chaining Comparison Operators

Python allows you to chain multiple comparisons in a single expression, which is more concise and often more readable than using `and` with separate comparisons.

- `a < b < c` is equivalent to `(a < b) and (b < c)`.
- The expression is evaluated from left to right.

**Example:**

```python
score = 85

# Chained comparison
if 80 <= score < 90:

```

This chaining works for all comparison operators.

### 6. Comparison of Different Data Types

- **Numeric Types**: `int`, `float`, `complex` can generally be compared with each other. `complex` numbers, however, do not support `>, <, >=, <=` as they don't have a natural ordering.

  ```python
  print(f"5 == 5.0: {5 == 5.0}")     # True
  print(f"5 < 5.1: {5 < 5.1}")       # True
  # print(f"1 + 2j > 3 + 4j") # TypeError: '>' not supported between instances of 'complex' and 'complex'
  ```

- **Incomparable Types**: Attempting to compare objects of completely different and unrelated types (e.g., an integer and a string) will result in a `TypeError`.

  ```python
  # print(f"10 > 'hello'") # TypeError: '>' not supported between instances of 'int' and 'str'
  # print(f"[1,2] == 'abc'") # False (different types, but == and != are often defined for type checking)
  # print(f"[1,2] < 'abc'") # TypeError: '<' not supported between instances of 'list' and 'str'
  ```

  While `==` and `!=` often return `False` for different types (e.g., `1 == '1'` is `False`), ordering comparisons (`<`, `>`, `<=`, `>=`) will raise a `TypeError` unless a specific comparison mechanism is defined.

### 7. Custom Object Comparison (`__eq__`, `__ne__`, etc.)

By default, custom objects inherit comparison behavior from their base classes (usually `object`). For `==` and `!=`, this typically means comparing identity unless overridden. For ordering comparisons, it generally results in `TypeError`.

To enable comparison operators for your custom classes, you define special "magic methods" (also called dunder methods):

- `__eq__(self, other)`: Implements `self == other`
- `__ne__(self, other)`: Implements `self != other` (often defined as `not self == other`)
- `__lt__(self, other)`: Implements `self < other`
- `__le__(self, other)`: Implements `self <= other`
- `__gt__(self, other)`: Implements `self > other` (often defined using `__lt__`)
- `__ge__(self, other)`: Implements `self >= other` (often defined using `__le__`)

**Example:**

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        # Check if 'other' is also a Point instance
        if not isinstance(other, Point):
            return NotImplemented # Or raise TypeError
        return self.x == other.x and self.y == other.y

    def __ne__(self, other):
        return not self == other # Often simply negate __eq__

    def __lt__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        # Compare points based on their distance from origin (example logic)
        return (self.x**2 + self.y**2) < (other.x**2 + other.y**2)

    def __le__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return (self.x**2 + self.y**2) <= (other.x**2 + other.y**2)

    def __repr__(self): # For clear printing
        return f"Point({self.x}, {self.y})"

p1 = Point(1, 1)
p2 = Point(2, 2)
p3 = Point(1, 1)
p4 = Point(0, 0)

print(f"p1 == p3: {p1 == p3}")    # True (due to __eq__)
print(f"p1 != p2: {p1 != p2}")    # True (due to __ne__)
print(f"p1 < p2: {p1 < p2}")      # True (1^2+1^2=2 < 2^2+2^2=8)
print(f"p1 <= p3: {p1 <= p3}")    # True
print(f"p2 > p1: {p2 > p1}")      # True (Python implicitly uses __lt__)
print(f"p1 < 5: {p1 < 5}")        # False (NotImplemented, then TypeError if not handled)
```

**`@functools.total_ordering` Decorator (Advanced):**
If you define `__eq__` and _at least one_ of `__lt__`, `__le__`, `__gt__`, or `__ge__`, the `functools.total_ordering` decorator can automatically fill in the remaining comparison methods for you.

```python
from functools import total_ordering

@total_ordering
class AnotherPoint:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        if not isinstance(other, AnotherPoint):
            return NotImplemented
        return self.x == other.x and self.y == other.y

    def __lt__(self, other): # Only need to define __lt__ (or another single ordering method)
        if not isinstance(other, AnotherPoint):
            return NotImplemented
        return (self.x**2 + self.y**2) < (other.x**2 + other.y**2)

    def __repr__(self):
        return f"AnotherPoint({self.x}, {self.y})"

ap1 = AnotherPoint(1, 1)
ap2 = AnotherPoint(2, 2)
ap3 = AnotherPoint(1, 1)

print(f"ap1 == ap3: {ap1 == ap3}") # True
print(f"ap1 < ap2: {ap1 < ap2}")   # True
print(f"ap2 > ap1: {ap2 > ap1}")   # True (generated by @total_ordering)
print(f"ap1 <= ap3: {ap1 <= ap3}") # True (generated by @total_ordering)
```

### 8. Boolean Context (Truthiness)

Comparison operators always return boolean values (`True` or `False`). These boolean values are fundamental to Python's "truthiness" concept.

- `True` evaluates to `True` in a boolean context (e.g., `if True:`).
- `False` evaluates to `False` in a boolean context (e.g., `if False:`).

```python
age = 18
if age >= 18: # The comparison (age >= 18) evaluates to True
    print("Adult")

is_active = True
if is_active: # Directly uses the boolean result of the comparison or variable
    print("User is active.")
```

Comparison operators typically have lower precedence than arithmetic operators but higher precedence than logical operators (`and`, `or`, `not`).

```python
# Precedence: arithmetic > comparison > logical
# (5 + 3) > 7 and 10 < (20 - 5)
print(8 > 7 and 10 < 15) # True and True -> True
```

Understanding Python's comparison operators is vital for writing effective and logical code, allowing you to build robust conditional logic and manipulate data efficiently.

## Operator Overloading

Classes can define their own comparison behavior by overriding special methods:

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    # Less than
    def __lt__(self, other):
        return self.pages < other.pages

    # Less than or equal
    def __le__(self, other):
        return self.pages <= other.pages

    # Equal
    def __eq__(self, other):
        return self.pages == other.pages and self.title == other.title

    # Not equal
    def __ne__(self, other):
        return not self.__eq__(other)

    # Greater than
    def __gt__(self, other):
        return self.pages > other.pages

    # Greater than or equal
    def __ge__(self, other):
        return self.pages >= other.pages

book1 = Book("Python 101", 300)
book2 = Book("Advanced Python", 500)

print(book1 < book2)   # True
print(book1 == book2)  # False
```

## Custom Comparisons

### Using `functools.total_ordering`

For classes where you want to implement only some comparison methods:

```python
from functools import total_ordering

@total_ordering
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def __eq__(self, other):
        return self.grade == other.grade

    def __lt__(self, other):
        return self.grade < other.grade

alice = Student("Alice", 85)
bob = Student("Bob", 90)

print(alice <= bob)  # True (uses __lt__ and __eq__)
print(alice > bob)   # False
```

## The `is` Operator vs `==`

- `is` checks for identity (same object in memory)
- `==` checks for equality (same value)

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

a is b  # True (same object)
a is c  # False (different objects)
a == c  # True (same content)

# Special case for small integers and strings (due to Python's optimization)
x = 256
y = 256
x is y  # True (cached integers)

x = 257
y = 257
x is y  # False (unless in the same line or function)
```

## Floating-Point Comparisons

Floating-point arithmetic can be imprecise, so direct equality checks are often unreliable:

```python
# Problematic
0.1 + 0.2 == 0.3  # False!

# Better approaches
# 1. Use a tolerance (epsilon)
def almost_equal(a, b, epsilon=1e-9):
    return abs(a - b) < epsilon

# 2. Use math.isclose (Python 3.5+)
from math import isclose
isclose(0.1 + 0.2, 0.3)  # True
```

## Short-Circuiting in Comparisons

Python evaluates comparisons left-to-right and stops early when possible:

```python
def expensive_call():

    return False
```

```python
class Version:

        self.minor = minor
        self.patch = patch

        return (self.major, self.minor, self.patch)

    def __eq__(self, other):
        return self._key() == other._key()

    def __hash__(self):
        return hash(self._key())

    # Other comparison methods would follow similar pattern
```

## Best Practices

1. **Use `==` for value comparison** and `is` only when you need to check identity
2. **Be careful with floating-point comparisons** - use `math.isclose()` or a tolerance
3. **Implement all six comparison methods** (`__lt__`, `__le__`, `__eq__`, `__ne__`, `__gt__`, `__ge__`) for consistency, or use `@total_ordering`
4. **Make comparisons transitive** - if a == b and b == c, then a == c should be true
5. **Consider performance** - complex comparison operations can slow down sorting
6. **Document your comparison logic** - especially for custom classes
7. **Use built-in functions** like `sorted()`, `max()`, `min()` with `key` parameters when possible
8. **Be consistent with `__hash__`** - objects that compare equal should hash to the same value
9. **Watch for type mismatches** - Python 3 is stricter about comparing different types
10. **Leverage chained comparisons** for cleaner code when appropriate

This comprehensive guide covers Python comparison operators from basic usage to advanced techniques, providing a solid foundation for writing effective comparison logic in your Python code.


````python
# --- Python Comparison Operators: All About in Code ---

print("--- 1. Basic Comparison Operators (==, !=, >, <, >=, <=) ---")

# 1.1 Numbers
a = 10
b = 20
c = 10
d = 10.0

print(f"a == c  (10 == 10): {a == c}")
print(f"a != b  (10 != 20): {a != b}")
print(f"b > a   (20 > 10): {b > a}")
print(f"a < b   (10 < 20): {a < b}")
print(f"a >= c  (10 >= 10): {a >= c}")
print(f"b <= c  (20 <= 10): {b <= c}")
print(f"a == d  (10 == 10.0): {a == d}") # Compares value, not type
print(f"a < d   (10 < 10.0): {a < d}")   # False (10 is not strictly less than 10.0)

print("\n--- 1.2 Strings (Lexicographical Comparison) ---")
str1 = "apple"
str2 = "banana"
str3 = "apple"
str4 = "Apple" # 'A' has lower ASCII/Unicode value than 'a'

print(f"str1 == str3 ('apple' == 'apple'): {str1 == str3}")
print(f"str1 != str2 ('apple' != 'banana'): {str1 != str2}")
print(f"str1 < str2  ('apple' < 'banana'): {str1 < str2}")
print(f"str1 > str4  ('apple' > 'Apple'): {str1 > str4}")

print("\n--- 1.3 Lists and Tuples (Element-by-Element Comparison) ---")
list_a = [1, 2, 3]
list_b = [1, 2, 4]
list_c = [1, 2, 3]
list_d = [1, 2]

print(f"list_a == list_c ([1,2,3] == [1,2,3]): {list_a == list_c}")
print(f"list_a < list_b  ([1,2,3] < [1,2,4]): {list_a < list_b}") # 3 < 4
print(f"list_a > list_d  ([1,2,3] > [1,2]): {list_a > list_d}")   # Longer sequence is greater if prefix matches

tuple_x = (1, 5)
tuple_y = (1, 2, 3)
print(f"tuple_x > tuple_y ((1,5) > (1,2,3)): {tuple_x > tuple_y}") # 5 > 2

print("\n--- 1.4 Dictionaries and Sets ---")
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 2, 'a': 1}
dict3 = {'a': 1, 'b': 3}

print(f"dict1 == dict2 (order irrelevant for equality): {dict1 == dict2}")
print(f"dict1 != dict3: {dict1 != dict3}")
# print(dict1 < dict2) # TypeError: '<' not supported for dictionaries

set1 = {1, 2, 3}
set2 = {1, 2, 3, 4}
set3 = {1, 2}

print(f"set1 == {1,2,3}: {set1 == {1,2,3}}")
print(f"set1 < set2 (set1 is proper subset of set2): {set1 < set2}")
print(f"set2 >= set1 (set2 is superset of set1): {set2 >= set1}")
print(f"set1 > set3 (set1 is proper superset of set3): {set1 > set3}")


print("\n--- 2. Identity Operators (is, is not) ---")
# Compare if two variables refer to the exact same object in memory

list_obj1 = [1, 2, 3]
list_obj2 = [1, 2, 3]
list_obj3 = list_obj1 # list_obj3 now points to the same object as list_obj1

print(f"list_obj1 == list_obj2: {list_obj1 == list_obj2}") # True (values are equal)
print(f"list_obj1 is list_obj2: {list_obj1 is list_obj2}") # False (different objects in memory)
print(f"list_obj1 is list_obj3: {list_obj1 is list_obj3}") # True (same object in memory)

# Integer/string interning (optimization by Python for small/common values)
num_small1 = 5
num_small2 = 5
num_large1 = 1000
num_large2 = 1000

print(f"num_small1 is num_small2: {num_small1 is num_small2}") # Often True
print(f"num_large1 is num_large2: {num_large1 is num_large2}") # Often False

str_lit1 = "python"
str_lit2 = "python"
str_lit3 = "py" + "thon" # Evaluated at compile time to "python"
str_var = "py"
str_concat = str_var + "thon" # Evaluated at runtime

print(f"str_lit1 is str_lit2: {str_lit1 is str_lit2}") # Often True
print(f"str_lit1 is str_lit3: {str_lit1 is str_lit3}") # Often True
print(f"str_lit1 is str_concat: {str_lit1 is str_concat}") # Often False

print(f"None is None: {None is None}") # Always True (None is a singleton)


print("\n--- 3. Membership Operators (in, not in) ---")
# Check if a value is present in a sequence or collection

my_fruits = ["apple", "banana", "cherry", "orange"]
my_sentence = "The quick brown fox jumps over the lazy dog."

print(f"'banana' in my_fruits: {'banana' in my_fruits}")
print(f"'grape' in my_fruits: {'grape' in my_fruits}")
print(f"'kiwi' not in my_fruits: {'kiwi' not in my_fruits}")

print(f"'fox' in my_sentence: {'fox' in my_sentence}")
print(f"'cat' not in my_sentence: {'cat' not in my_sentence}")

my_config = {'user': 'admin', 'role': 'superuser', 'status': 'active'}
print(f"'user' in my_config (checks keys): {'user' in my_config}")
print(f"'admin' in my_config (does not check values): {'admin' in my_config}") # False


print("\n--- 4. Chaining Comparison Operators ---")
# Python allows comparing multiple values in a chain (e.g., a < b < c)
# This is equivalent to (a < b) and (b < c)

temp = 25
if 20 < temp < 30:
    print(f"Temperature ({temp}) is moderate.")

age = 15
if 10 <= age <= 18:
    print(f"Age ({age}) is within the teenage range.")

x, y, z = 10, 20, 30
print(f"x < y < z ({x} < {y} < {z}): {x < y < z}")

a, b, c = 5, 5, 5
print(f"a == b == c ({a} == {b} == {c}): {a == b == c}")


print("\n--- 5. Comparison with Incomparable Types (TypeError) ---")
# Direct comparison operators for different, unrelated types usually raise TypeError
try:
    print(f"10 > 'hello': {10 > 'hello'}")
except TypeError as e:
    print(f"TypeError caught: {e}")

try:
    print(f"[1, 2] < 'abc': {[1, 2] < 'abc'}")
except TypeError as e:
    print(f"TypeError caught: {e}")

# Note: == and != often return False for different types rather than raising an error
print(f"10 == '10': {10 == '10'}") # False


print("\n--- 6. Custom Object Comparison (Magic Methods) ---")
# You can define how your custom objects are compared by implementing dunder methods.

from functools import total_ordering # For convenience

@total_ordering # This decorator helps by filling in missing comparison methods
class MyCoordinate:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        # Define equality: two coordinates are equal if their x and y are the same
        if not isinstance(other, MyCoordinate):
            return NotImplemented # Allow Python to try other comparison or raise error
        return self.x == other.x and self.y == other.y

    # With @total_ordering, if you define __eq__ and __lt__, it can generate __ne__, __le__, __gt__, __ge__
    def __lt__(self, other):
        # Define less than: e.g., based on the sum of x and y
        if not isinstance(other, MyCoordinate):
            return NotImplemented
        return (self.x + self.y) < (other.x + other.y)

    def __repr__(self): # For better printing of the object
        return f"MyCoordinate({self.x}, {self.y})"

c1 = MyCoordinate(1, 2)
c2 = MyCoordinate(3, 4)
c3 = MyCoordinate(1, 2)
c4 = MyCoordinate(0, 5) # Sum = 5, same as c1.x+c1.y

print(f"c1 == c3 ({c1} == {c3}): {c1 == c3}")    # True (due to __eq__)
print(f"c1 != c2 ({c1} != {c2}): {c1 != c2}")    # True (due to __ne__ generated by @total_ordering)
print(f"c1 < c2  ({c1} < {c2}): {c1 < c2}")      # True (3 < 7, due to __lt__)
print(f"c1 <= c4 ({c1} <= {c4}): {c1 <= c4}")    # True (3 <= 5, due to __le__ generated by @total_ordering)
print(f"c2 > c1  ({c2} > {c1}): {c2 > c1}")      # True (generated by @total_ordering)
print(f"c4 >= c1 ({c4} >= {c1}): {c4 >= c1}")    # True (generated by @total_ordering)

# Comparison with a non-MyCoordinate object
print(f"c1 == 'hello': {c1 == 'hello'}") # False (NotImplemented, then default __eq__ from object which returns False)
print(f"c1 < 10: {c1 < 10}")             # False (NotImplemented, then TypeError when Python tries other types)
````
