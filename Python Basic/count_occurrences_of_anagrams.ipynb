# Count Occurrences of Anagrams

**Last Updated**: 17 Mar, 2025

Given a word `pattern` and a string `text` consisting of lowercase characters, the task is to return the count of substrings in `text` which are **anagrams** of the `pattern`.

---

## Examples

### Example 1:

- **Input**:  
  `text = "forxxorfxdofr"`  
  `pattern = "for"`

- **Output**:  
  `3`

- **Explanation**:  
  Anagrams present are `"for"`, `"orf"`, and `"ofr"`.  
  Each appears in the text once and hence the count is **3**.

---

### Example 2:

- **Input**:  
  `text = "aabaabaa"`  
  `pattern = "aaba"`

- **Output**:  
  `4`

- **Explanation**:  
  Anagrams present are `"aaba"` and `"abaa"`.  
  Each appears **twice** in the text and hence the count is **4**.

---

## Notes:

An anagram is a word formed by rearranging the letters of another word.  
This problem can be efficiently solved using the **sliding window technique** combined with **hashing** (like a frequency map).


Let's explore all possible approaches to solve the "Count Occurrences of Anagrams" problem. We are given a `text` and a `pattern`, and we need to find how many substrings in `text` are anagrams of `pattern`.

**Understanding Anagrams:**

Two strings are anagrams of each other if they contain the same characters with the same frequencies, regardless of the order.

**All Possible Approaches:**

Here are several approaches, ranging from straightforward to more optimized:

**1. Brute Force (Generating and Comparing):**

- **Idea:** Generate all possible substrings of `text` that have the same length as `pattern`. For each substring, check if it's an anagram of `pattern`.
- **Steps:**
  1.  Iterate through all possible starting positions of a substring in `text` (from index 0 to `len(text) - len(pattern)`).
  2.  For each starting position, extract a substring of `text` with the same length as `pattern`.
  3.  Compare this substring with `pattern` to check if they are anagrams. A simple way to check for anagrams is to sort both the substring and the pattern and then compare the sorted strings.
  4.  If they are anagrams, increment the count.
- **Time Complexity:**
  - Generating substrings takes O(`len(text) * len(pattern)`).
  - Sorting each substring takes O(`len(pattern) * log(len(pattern))`).
  - Overall: O(`len(text) * len(pattern) * log(len(pattern))`).
- **Space Complexity:** O(`len(pattern)`) for storing the substring and sorted strings temporarily.

```python
def count_anagrams_brute_force(text, pattern):
    n = len(text)
    m = len(pattern)
    count = 0
    for i in range(n - m + 1):
        substring = text[i : i + m]
        if sorted(substring) == sorted(pattern):
            count += 1
    return count
```

**2. Character Count Comparison (Using Dictionaries/Hash Maps):**

- **Idea:** For each substring of `text` with the same length as `pattern`, compare the character counts of the substring with the character counts of `pattern`.
- **Steps:**
  1.  Create a character count map (dictionary or hash map) for the `pattern`. Store the frequency of each character in the `pattern`.
  2.  Iterate through all possible starting positions of a substring in `text` (from index 0 to `len(text) - len(pattern)`).
  3.  For each starting position, extract a substring of `text` with the same length as `pattern`.
  4.  Create a character count map for the current substring.
  5.  Compare the character count map of the substring with the character count map of the `pattern`. If they are identical, increment the count.
- **Time Complexity:**
  - Creating the pattern's character count map takes O(`len(pattern)`).
  - Iterating through substrings takes O(`len(text) * len(pattern)`).
  - Creating the character count map for each substring takes O(`len(pattern)`).
  - Comparing the maps takes O(number of unique characters), which is at most O(26) for lowercase English letters.
  - Overall: O(`len(pattern) + len(text) * len(pattern)`).
- **Space Complexity:** O(number of unique characters) to store the character count maps.

```python
from collections import Counter

def count_anagrams_char_count(text, pattern):
    n = len(text)
    m = len(pattern)
    count = 0
    pattern_counts = Counter(pattern)
    for i in range(n - m + 1):
        substring = text[i : i + m]
        substring_counts = Counter(substring)
        if substring_counts == pattern_counts:
            count += 1
    return count
```

**3. Sliding Window with Character Count Comparison:**

- **Idea:** Optimize the character count comparison by using a sliding window of size `len(pattern)` over the `text`. Maintain the character count of the current window and efficiently update it as the window slides.
- **Steps:**
  1.  Create a character count map for the `pattern`.
  2.  Initialize a character count map for the first window of `text` (of size `len(pattern)`).
  3.  Compare the character count map of the initial window with the pattern's map. If they are equal, increment the count.
  4.  Slide the window one character at a time from left to right.
  5.  For each slide:
      - Decrement the count of the character that is leaving the window.
      - Increment the count of the character that is entering the window.
      - After each update, compare the current window's character count map with the pattern's map. If they are equal, increment the count.
- **Time Complexity:**
  - Creating the pattern's character count map takes O(`len(pattern)`).
  - Processing the first window takes O(`len(pattern)`).
  - Sliding the window takes O(`len(text) - len(pattern)`).
  - Updating the character count map for each slide takes O(1) (constant time for incrementing and decrementing).
  - Comparing the maps takes O(number of unique characters), which is at most O(26).
  - Overall: O(`len(pattern) + len(text)`).
- **Space Complexity:** O(number of unique characters) to store the character count maps.

```python
from collections import Counter

def count_anagrams_sliding_window(text, pattern):
    n = len(text)
    m = len(pattern)
    if m > n:
        return 0

    pattern_counts = Counter(pattern)
    window_counts = Counter(text[:m])
    count = 0

    if window_counts == pattern_counts:
        count += 1

    for i in range(1, n - m + 1):
        left_char = text[i - 1]
        right_char = text[i + m - 1]

        window_counts[left_char] -= 1
        if window_counts[left_char] == 0:
            del window_counts[left_char]

        window_counts[right_char] += 1

        if window_counts == pattern_counts:
            count += 1

    return count
```

**4. Using Character Frequency Arrays (Optimization of Sliding Window):**

- **Idea:** Instead of using dictionaries for character counts, use fixed-size arrays (e.g., of size 26 for lowercase English letters) to store character frequencies. This can be slightly more efficient for comparisons.
- **Steps:**
  1.  Create a frequency array for the `pattern`.
  2.  Initialize a frequency array for the first window of `text`.
  3.  Compare the frequency arrays.
  4.  Slide the window, updating the frequency array by decrementing the count of the leaving character and incrementing the count of the entering character.
  5.  Compare the frequency arrays after each slide.
- **Time Complexity:** O(`len(pattern) + len(text)`).
- **Space Complexity:** O(number of unique characters), which is constant (e.g., 26 for lowercase English letters).

```python
def count_anagrams_sliding_window_array(text, pattern):
    n = len(text)
    m = len(pattern)
    if m > n:
        return 0

    pattern_freq = [0] * 26
    window_freq = [0] * 26
    count = 0

    for char in pattern:
        pattern_freq[ord(char) - ord('a')] += 1

    for i in range(m):
        window_freq[ord(text[i]) - ord('a')] += 1

    if window_freq == pattern_freq:
        count += 1

    for i in range(1, n - m + 1):
        left_char = text[i - 1]
        right_char = text[i + m - 1]

        window_freq[ord(left_char) - ord('a')] -= 1
        window_freq[ord(right_char) - ord('a')] += 1

        if window_freq == pattern_freq:
            count += 1

    return count
```

**5. Using Hashing (Rolling Hash - More Complex):**

- **Idea:** Calculate a hash value for the `pattern`. Then, for each substring of `text` with the same length, calculate its hash value efficiently using a rolling hash technique. If the hash values match, there's a high probability that the substring is an anagram (hash collisions are possible but can be minimized).
- **Steps:**
  1.  Choose a suitable hash function for strings (e.g., based on character values and a prime modulus).
  2.  Calculate the hash value of the `pattern`.
  3.  Calculate the hash value of the first substring of `text` with the same length as `pattern`.
  4.  Compare the hash values. If they match, check if they are actual anagrams (to avoid collisions).
  5.  Use a rolling hash technique to efficiently calculate the hash value of the next substring by updating the hash of the current window instead of recalculating from scratch.
  6.  For each window, compare the hash value and perform a full anagram check if the hashes match.
- **Time Complexity:**
  - Calculating the initial hash values takes O(`len(pattern)`).
  - Rolling the hash takes O(1) per step, so O(`len(text) - len(pattern)`) for the entire text.
  - Full anagram check in case of hash collision can take O(`len(pattern)`). The number of collisions should ideally be low.
  - Average case: O(`len(pattern) + len(text)`).
  - Worst case (many collisions): O(`len(pattern) * len(text)`).
- **Space Complexity:** O(1) or O(number of unique characters) depending on the hash function.

**Considerations for Choosing an Approach:**

- **Simplicity:** The brute force and character count comparison (without sliding window) are easier to understand and implement.
- **Efficiency:** The sliding window approaches (with dictionaries or arrays) offer the best time complexity of O(`len(pattern) + len(text)`).
- **Potential Optimizations:** Using frequency arrays instead of dictionaries can provide a slight performance improvement due to the fixed size and direct indexing.
- **Hashing Complexity:** While hashing can be efficient on average, the implementation of a good rolling hash function and handling collisions correctly can be more complex.

For most practical purposes, the **sliding window approach with character count comparison (using either dictionaries or frequency arrays)** is the recommended method due to its balance of efficiency and relative ease of implementation.

The choice between dictionaries (`Counter`) and frequency arrays depends on the constraints of the problem (e.g., the character set). For lowercase English letters, a fixed-size array of 26 is often slightly more performant.

```python
from collections import Counter

class AnagramCounter:
    """
    A class to find and count anagrams of a pattern within a text.
    """
    def __init__(self, pattern):
        """
        Initializes the AnagramCounter with the pattern to search for.

        Args:
            pattern: The word to find anagrams of.
        """
        if not isinstance(pattern, str):
            raise TypeError("Pattern must be a string.")
        self.pattern = pattern
        self.pattern_len = len(pattern)
        self.pattern_counts = Counter(pattern)

    def count_anagrams(self, text):
        """
        Counts the number of anagrams of the pattern present in the text.

        Args:
            text: The string to search within.

        Returns:
            The count of anagrams.
        """
        if not isinstance(text, str):
            raise TypeError("Text must be a string.")
        n = len(text)
        m = self.pattern_len
        count = 0

        if m == 0:
            return 0

        if m > n:
            return 0

        for i in range(n - m + 1):
            substring = text[i : i + m]
            substring_counts = Counter(substring)
            if substring_counts == self.pattern_counts:
                count += 1

        return count
```

```python
from collections import Counter

def count_anagrams_sliding_window(text, pattern):
    """
    Counts the number of anagrams of pattern found in text using the sliding window
    technique with character count comparison.

    Args:
        text: The string to search within.
        pattern: The word to find anagrams of.

    Returns:
        The count of anagrams.
    """
    n = len(text)
    m = len(pattern)
    if m > n:
        return 0  # Edge case: Pattern longer than text

    pattern_counts = Counter(pattern)
    window_counts = Counter(text[:m])  # Initialize counts for the first window
    count = 0

    if window_counts == pattern_counts:
        count += 1

    for i in range(1, n - m + 1):
        left_char = text[i - 1]
        right_char = text[i + m - 1]

        # Update window counts by removing the leftmost character
        window_counts[left_char] -= 1
        if window_counts[left_char] == 0:
            del window_counts[left_char]

        # Update window counts by adding the rightmost character
        window_counts[right_char] += 1

        # Check if the current window is an anagram
        if window_counts == pattern_counts:
            count += 1

    return count
```

**Explanation of the Sliding Window with Character Count Comparison Approach:**

1.  **`count_anagrams_sliding_window(text, pattern)` Function:**

    - Takes two string arguments: `text` and `pattern`.
    - Gets the lengths of `text` (`n`) and `pattern` (`m`).
    - **Edge Case Handling:**
      - If `m > n`, no anagrams are possible, so it returns 0.
    - Creates a `Counter` (`pattern_counts`) to store the frequency of each character in the `pattern`.
    - Creates a `Counter` (`window_counts`) to store the frequency of characters in the initial window of `text` (the first `m` characters).
    - Initializes a `count` variable to 0.
    - **Initial Window Check:** If the character counts of the initial window match the `pattern_counts`, increment `count`.
    - **Sliding the Window:**
      - Iterates through the `text` starting from the second possible window (index 1) up to the last possible starting position (`n - m`).
      - In each iteration:
        - `left_char`: The character that is leaving the left side of the window (the character at index `i - 1`).
        - `right_char`: The character that is entering the right side of the window (the character at index `i + m - 1`).
        - **Updating `window_counts`:**
          - Decrement the count of `left_char` in `window_counts`. If the count becomes 0, remove the character from `window_counts` to keep it clean.
          - Increment the count of `right_char` in `window_counts`.
        - **Checking for Anagram:** After updating the window counts, compare `window_counts` with `pattern_counts`. If they are equal, increment `count`.
    - Returns the final `count` of anagrams.

2.  **Edge Cases:**

    - **Empty Pattern:** If `pattern` is empty (`m == 0`), the initial `window_counts` (which would also be empty for an empty initial window) will match `pattern_counts`, and the loop won't run, correctly returning 1 (as an empty string is an anagram of an empty string).
    - **Empty Text:** If `text` is empty (`n == 0`), and `pattern` is not empty, `m > n` will be true, and 0 will be returned. If both are empty, the initial check will pass, returning 1.
    - **Pattern Longer Than Text:** The initial check `if m > n:` handles this, returning 0.
    - **No Anagrams:** If no window's character counts ever match the `pattern_counts`, the `count` will remain 0.

  - Creating the initial `window_counts` takes O(`len(pattern)`).

This sliding window approach is significantly more efficient for larger texts compared to the brute force and the simple character count comparison without a sliding window, as it avoids redundant counting of characters in overlapping substrings.

```python
from collections import Counter

class Solution:
    def all_zero(self, count):
        return all(c == 0 for c in count.values())

    def search(self, pat: str, txt: str) -> int:
        k = len(pat)
        pattern_counts = Counter(pat)
        window_counts = Counter()
        n = len(txt)
        result = 0
        i = 0
        j = 0

        while j < n:
            window_counts[txt[j]] += 1

            if j - i + 1 == k:
                if window_counts == pattern_counts:
                    result += 1

                window_counts[txt[i]] -= 1
                if window_counts[txt[i]] == 0:
                    del window_counts[txt[i]]
                i += 1
            j += 1
        return result
```

**Key Concepts and Differences from C++:**

- **`collections.Counter`:** Python's `Counter` is a powerful tool for efficiently counting the frequency of items in a sequence (like a string). It acts similarly to a hash map (or `std::unordered_map` in C++) for character counts.
- **`all(c == 0 for c in count.values())`:** This is a Pythonic way to check if all values in a dictionary (or `Counter`) are zero. It's more concise than iterating through the values like in the C++ `allZero` function.
- **No Explicit Character Array:** Python strings are inherently iterable, so we directly iterate through characters without needing to convert to a character array.
- **Dictionary/Hash Map Behavior:** Python's `Counter` (and dictionaries in general) handle the absence of a key gracefully. When you try to access a key that doesn't exist, it typically returns 0 (for `Counter` when incrementing) or raises a `KeyError` if you try to get its value directly. The code handles this implicitly.
- **Sliding Window Logic:** The core logic of the sliding window remains the same: maintain a window of size `k` and efficiently update the character counts as the window slides.

This Python code provides a clear and efficient implementation of the sliding window approach to count anagrams, leveraging Python's built-in data structures and features. The time complexity remains O(n) where n is the length of the text, and the space complexity is O(1) because the size of the `Counter` objects is limited by the size of the alphabet (constant).


````python
class Solution:
    def all_zero(self, count):
        for val in count.values():
            if val != 0:
                return False
        return True

    def search(self, pat: str, txt: str) -> int:
        k = len(pat)
        pattern_counts = {}
        for char in pat:
            pattern_counts[char] = pattern_counts.get(char, 0) + 1

        window_counts = {}
        n = len(txt)
        result = 0
        i = 0
        j = 0

        while j < n:
            window_counts[txt[j]] = window_counts.get(txt[j], 0) + 1

            if j - i + 1 == k:
                if window_counts == pattern_counts:
                    result += 1

                window_counts[txt[i]] = window_counts.get(txt[i], 0) - 1
                if window_counts[txt[i]] == 0:
                    del window_counts[txt[i]]
                i += 1
            j += 1
        return result

# Example Usage:
if __name__ == "__main__":
    solution = Solution()
    text1 = "forxxorfxdofr"
    pattern1 = "for"
    print(f"Count of anagrams of '{pattern1}' in '{text1}': {solution.search(pattern1, text1)}")

    text2 = "aabaabaa"
    pattern2 = "aaba"
    print(f"Count of anagrams of '{pattern2}' in '{text2}': {solution.search(pattern2, text2)}")

    text3 = "abab"
    pattern3 = "ab"
    print(f"Count of anagrams of '{pattern3}' in '{text3}': {solution.search(pattern3, text3)}")

    text4 = "aaaa"
    pattern4 = "aa"
    print(f"Count of anagrams of '{pattern4}' in '{text4}': {solution.search(pattern4, text4)}")

    text5 = "abcabcabc"
    pattern5 = "cba"
    print(f"Count of anagrams of '{pattern5}' in '{text5}': {solution.search(pattern5, text5)}")
````

```python
class Solution:
    def all_zero(self, count):
        for val in count.values():
            if val != 0:
                return False
        return True

    def search(self, pat: str, txt: str) -> int:
        k = len(pat)
        pattern_counts = {}
        for char in pat:
            pattern_counts[char] = pattern_counts.get(char, 0) + 1

        window_counts = {}
        n = len(txt)
        result = 0
        i = 0
        j = 0

        while j < n:
            window_counts[txt[j]] = window_counts.get(txt[j], 0) + 1

            if j - i + 1 == k:
                if window_counts == pattern_counts:
                    result += 1

                window_counts[txt[i]] = window_counts.get(txt[i], 0) - 1
                if window_counts[txt[i]] == 0:
                    del window_counts[txt[i]]
                i += 1
            j += 1
        return result
```
