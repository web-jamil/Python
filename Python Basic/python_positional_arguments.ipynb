````python
# 📌 Positional arguments in Python functions

# Basic function using positional arguments
def multiply(a, b):
    return a * b

print(multiply(4, 5))   # Output: 20


# 🧭 Arguments must be passed in defined order
def display_info(name, age, city):
    print(f"{name} is {age} years old and lives in {city}.")

display_info("Jamil", 27, "Khulna")


# 🚫 Incorrect ordering leads to incorrect behavior or errors
# display_info(27, "Jamil", "Khulna")   # Logical error
# display_info("Jamil", 27)             # ❌ TypeError: missing 1 required positional argument


# 🔍 Combining positional and default values
def greet_user(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet_user("Jamil")             # Output: Hello, Jamil!
greet_user("Jamil", "Welcome")  # Output: Welcome, Jamil!


# 🏗️ Positional arguments in class methods

class Point:
    def __init__(self, x, y):  # x and y are positional
        self.x = x
        self.y = y

    def move(self, dx, dy):    # dx and dy are positional
        self.x += dx
        self.y += dy

    def location(self):
        return (self.x, self.y)

p = Point(3, 4)
p.move(2, -1)
print(p.location())   # Output: (5, 3)


# 🔧 Mix of positional and keyword in class constructors

class Signal:
    def __init__(self, frequency, amplitude=1.0):
        self.frequency = frequency
        self.amplitude = amplitude

    def summary(self):
        return f"Freq={self.frequency}Hz, Amp={self.amplitude}"

sig1 = Signal(440)
sig2 = Signal(1000, amplitude=0.5)

print(sig1.summary())   # Output: Freq=440Hz, Amp=1.0
print(sig2.summary())   # Output: Freq=1000Hz, Amp=0.5
````

````python
# 1. Functions with Positional Arguments

# A function that expects three positional arguments: 'name', 'age', and 'city'.
def create_profile(name, age, city):
    # The values passed during the call are assigned to the parameters
    # based on their order (position).
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")

# Calling the function using positional arguments.
# The value "Alice" goes to 'name', 30 goes to 'age', and "New York" goes to 'city'.
create_profile("Alice", 30, "New York")
# Output:
# Name: Alice
# Age: 30
# City: New York

# Incorrect order of positional arguments can lead to errors or unexpected behavior.
# Here, "30" is assigned to 'name', and "Alice" is assigned to 'age'.
# create_profile(30, "Alice", "New York") # This would be a logical error.

# You can mix positional and keyword arguments, but positional must come first.
create_profile("Bob", 25, city="Los Angeles")
# 'Bob' goes to 'name' (positional).
# '25' goes to 'age' (positional).
# 'city="Los Angeles"' is a keyword argument.
# Output:
# Name: Bob
# Age: 25
# City: Los Angeles


# 2. Arbitrary Positional Arguments (`*args`)

# The `*args` syntax allows a function to accept a variable number of positional arguments.
# These arguments are collected into a tuple.
def calculate_sum(*numbers):
    # 'numbers' will be a tuple containing all the arguments passed.
    total = sum(numbers)
    print(f"Numbers provided: {numbers}")
    print(f"Sum: {total}")

# Calling the function with different numbers of arguments.
calculate_sum(1, 2, 3)
# Output:
# Numbers provided: (1, 2, 3)
# Sum: 6

calculate_sum(10, 20, 30, 40, 50)
# Output:
# Numbers provided: (10, 20, 30, 40, 50)
# Sum: 150


# 3. Class Constructors (`__init__`)

# The constructor method of a class also uses positional arguments in the same way.
class Dog:
    # `__init__` takes positional arguments 'name' and 'age'.
    # The 'self' argument is a reference to the instance itself and is handled automatically.
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def bark(self):
        print(f"{self.name} says 'Woof!'")

# Creating an instance of the class by passing positional arguments.
# "Fido" goes to 'name', and 4 goes to 'age'.
dog1 = Dog("Fido", 4)
dog1.bark()
# Output: Fido says 'Woof!'

# Another instance with different positional arguments.
dog2 = Dog("Buddy", 7)
dog2.bark()
# Output: Buddy says 'Woof!'

# Incorrect number of positional arguments raises a TypeError.
# dog3 = Dog("Max") # This would fail because 'age' is missing.
````

In Python, arguments passed to a function or method can be categorized based on how they are matched to the parameters in the function's definition. A **positional argument** is an argument whose value is determined by its position or order in the function call.

Let's break down everything about positional arguments in both functions and classes.

### 1\. Positional Arguments in Functions

This is the most common way to pass arguments. The position of the argument you pass in the function call must match the position of the parameter in the function's definition.

#### Basic Positional Arguments

```python
def describe_pet(animal_type, pet_name):
    """Prints information about a pet."""
    print(f"I have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name}.")

# The order of arguments matters.
# 'dog' is the first argument, so it matches 'animal_type'.
# 'Fido' is the second argument, so it matches 'pet_name'.
describe_pet('dog', 'Fido')
# Output:
# I have a dog.
# My dog's name is Fido.

# If we swap the positions, the meaning changes.
describe_pet('Fido', 'dog')
# Output:
# I have a Fido.
# My Fido's name is dog. (This is probably not what we intended)
```

#### Mixing Positional and Keyword Arguments

You can pass arguments by position and then by keyword, but not the other way around. Once you've used a keyword argument, all subsequent arguments must also be keywords.

```python
def introduce_user(name, age, city):
    print(f"Hello, my name is {name}.")
    print(f"I am {age} years old and I live in {city}.")

# Valid: Positional arguments first, then keyword.
introduce_user("Alice", 30, city="New York")

# Invalid: Positional argument after a keyword argument.
# introduce_user("Alice", age=30, "New York")  # This would raise a SyntaxError
```

### 2\. Positional-Only Arguments (`/`)

Introduced in Python 3.8, the forward slash (`/`) is a special parameter that enforces that all arguments before it **must** be passed by position. They cannot be passed as keyword arguments. This is useful for APIs where argument names might change, or to enforce a specific calling style.

```python
def create_rectangle(width, height, /):
    """Calculates the area of a rectangle.
    width and height must be passed by position."""
    return width * height

# Valid calls:
area = create_rectangle(10, 5)  # Passed by position
print(f"Area is {area}") # Output: Area is 50

# Invalid calls:
# area = create_rectangle(width=10, height=5)
# print(f"This would raise a TypeError: 'width' parameter is positional-only")
```

You can also use it to define a mix of positional-only, regular, and keyword-only arguments.

```python
def full_info(first_name, last_name, /, age=None, *, country):
    """
    first_name and last_name: Must be passed positionally.
    age: Can be passed positionally or by keyword (has a default value).
    country: Must be passed by keyword.
    """
    print(f"Name: {first_name} {last_name}, Age: {age}, Country: {country}")

# Valid call:
full_info("John", "Doe", 30, country="USA")
# Output: Name: John Doe, Age: 30, Country: USA

# Another valid call: 'age' passed as keyword
full_info("Jane", "Doe", country="Canada", age=25)
# Output: Name: Jane Doe, Age: 25, Country: Canada

# Invalid call: Trying to use a keyword for a positional-only argument
# full_info(first_name="John", last_name="Doe", country="USA")
# This would raise a TypeError
```

### 3\. Variable Positional Arguments (`*args`)

Sometimes you want a function to accept an arbitrary number of positional arguments. The `*args` syntax allows you to do this. The asterisk (`*`) is the key part; `args` is just a common naming convention.

When a function is called with multiple arguments that aren't assigned to a named parameter, Python collects them into a single **tuple**.

```python
def calculate_sum(initial_value, *numbers):
    """Calculates the sum of an initial value and an arbitrary number of other numbers."""
    total = initial_value
    for num in numbers:
        total += num
    return total

# The arguments 10, 20, 30 are collected into the 'numbers' tuple.
total_sum = calculate_sum(5, 10, 20, 30)
print(f"Total sum is {total_sum}") # Output: Total sum is 65

# The 'numbers' tuple is empty if no additional arguments are passed.
total_sum_2 = calculate_sum(100)
print(f"Total sum is {total_sum_2}") # Output: Total sum is 100
```

### 4\. Positional Arguments in Classes (`__init__`)

The `__init__` method, or constructor, of a class is a function just like any other. Arguments passed when you create an instance of a class are passed to its `__init__` method, following all the same rules for positional and keyword arguments.

The first parameter, `self`, is a special positional argument that refers to the instance being created and is handled automatically by Python.

```python
class Car:
    def __init__(self, make, model, year):
        """Initializes a new Car object."""
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"Car: {self.make} {self.model}, Year: {self.year}")

# The arguments 'Ford', 'Mustang', and 1969 are positional arguments.
my_car = Car('Ford', 'Mustang', 1969)
my_car.display_info()
# Output: Car: Ford Mustang, Year: 1969

# You can also pass keyword arguments to the constructor.
your_car = Car(make='Tesla', model='Model 3', year=2022)
your_car.display_info()
# Output: Car: Tesla Model 3, Year: 2022
```

### Summary Table

| Argument Type           | How It's Defined                        | How It's Called                           | Example                         |
| :---------------------- | :-------------------------------------- | :---------------------------------------- | :------------------------------ |
| **Positional**          | By its order in the function signature. | By its order in the function call.        | `def f(a, b)` -\> `f(1, 2)`     |
| **Positional-Only**     | Before a `/` in the signature.          | Only by its order. Can't use keywords.    | `def f(a, /)` -\> `f(1)`        |
| **Variable Positional** | With a `*` prefix (e.g., `*args`).      | By passing any number of extra arguments. | `def f(*args)` -\> `f(1, 2, 3)` |
| **Keyword**             | After a positional or a `*` prefix.     | By explicitly naming the parameter.       | `def f(a)` -\> `f(a=1)`         |


````python
# --- Unlimited Positional Arguments in Functions ---

def sum_all(*numbers):
    """
    This function accepts an unlimited number of positional arguments.
    The arguments are collected into a tuple named 'numbers'.
    """
    print(f"Inside the function, 'numbers' is of type: {type(numbers)}")
    print(f"Arguments received: {numbers}")

    total = 0
    for num in numbers:
        total += num
    return total

# Calling the function with different numbers of arguments
result_1 = sum_all(10, 20, 30)
print(f"Function call with 3 arguments: {result_1}\n")

result_2 = sum_all(5, 5)
print(f"Function call with 2 arguments: {result_2}\n")

result_3 = sum_all()
print(f"Function call with 0 arguments: {result_3}\n")


def print_info(name, *details):
    """
    *args can be used alongside other arguments.
    'name' is a required positional argument, and all others go into 'details'.
    """
    print(f"Name: {name}")
    print(f"Additional details (a tuple): {details}")

print_info("Alice", 30, "Software Engineer", "New York")
print("\n" + "="*50 + "\n")


# --- Unlimited Positional Arguments in Classes ---

class Collection:
    """
    Demonstrates using *args in a class constructor (__init__).
    """
    def __init__(self, owner, *items):
        self.owner = owner
        # The items are stored as a tuple
        self.items = items
        print(f"Created a new Collection for {self.owner} with items: {self.items}")

    def get_number_of_items(self):
        return len(self.items)

    def display_items(self):
        print(f"Items in the collection: {self.items}")


# Creating class instances with different numbers of items
my_collection = Collection("Bob", "book", "pen", "notebook")
print(f"Number of items in Bob's collection: {my_collection.get_number_of_items()}")
my_collection.display_items()
print("-" * 20)

empty_collection = Collection("Charlie")
print(f"Number of items in Charlie's collection: {empty_collection.get_number_of_items()}")
empty_collection.display_items()
print("-" * 20)


class AdvancedCalculator:
    """
    Demonstrates using *args in a regular class method.
    """
    def __init__(self, initial_value=0):
        self.result = initial_value

    def add_all(self, *operands):
        """Adds an unlimited number of operands to the result."""
        print(f"Adding operands: {operands}")
        for op in operands:
            self.result += op
        print(f"Current result: {self.result}")


calc = AdvancedCalculator(100)
calc.add_all(10, 20, 5)
calc.add_all(100)
````
