Python functions accept **positional arguments** and **keyword arguments**, each serving different purposes. Let's break them down from **basic to advanced** concepts.

---

## **1. Positional Arguments**

Positional arguments are passed **in order**, meaning their position in the function call determines which parameter they correspond to.

### **Example:**

```python
def greet(name, age):
    print(f"Hello, my name is {name} and I am {age} years old.")

greet("Alice", 25)  # Correct usage
greet(25, "Alice")  # Incorrect order, leads to wrong output
```

- The first argument is assigned to `name`, and the second to `age`.
- Changing the order results in incorrect output.

---

## **2. Keyword Arguments**

Keyword arguments are passed using **parameter names**, allowing flexibility in order.

### **Example:**

```python
def greet(name, age):
    print(f"Hello, my name is {name} and I am {age} years old.")

greet(age=25, name="Alice")  # Order doesn't matter
```

- Using `name="Alice"` and `age=25` ensures correct assignment.
- This improves readability and avoids confusion.

---

## **3. Default Arguments**

Default arguments allow parameters to have **predefined values**, making them optional.

### **Example:**

```python
def greet(name, age=30):  # Default age is 30
    print(f"Hello, my name is {name} and I am {age} years old.")

greet("Bob")  # Uses default age (30)
greet("Charlie", 40)  # Overrides default age
```

- If `age` is not provided, it defaults to `30`.

---

## **4. Mixing Positional and Keyword Arguments**

You can mix both types, but **positional arguments must come first**.

### **Example:**

```python
def greet(name, age):
    print(f"Hello, my name is {name} and I am {age} years old.")

greet("Alice", age=25)  # Correct
greet(age=25, "Alice")  # ❌ Incorrect (SyntaxError)
```

- **Rule:** Once a keyword argument is used, all subsequent arguments must also be keyword arguments.

---

## **5. Arbitrary Arguments (`*args` and `**kwargs`)\*\*

### **`*args` (Variable-Length Positional Arguments)**

Allows passing **multiple positional arguments**.

```python
def add_numbers(*args):
    return sum(args)

print(add_numbers(1, 2, 3, 4))  # Output: 10
```

- `*args` collects all extra positional arguments into a tuple.

### **`**kwargs` (Variable-Length Keyword Arguments)\*\*

Allows passing **multiple keyword arguments**.

```python
def describe_person(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

describe_person(name="Alice", age=25, city="New York")
```

- `**kwargs` collects all extra keyword arguments into a dictionary.

---

## **6. Positional-Only and Keyword-Only Arguments**

Python allows restricting arguments to **positional-only** or **keyword-only**.

### **Positional-Only (`/` separator)**

```python
def greet(name, age, /):
    print(f"Hello, my name is {name} and I am {age} years old.")

greet("Alice", 25)  # ✅ Allowed
greet(name="Alice", age=25)  # ❌ Error (must be positional)
```

- The `/` **forces** `name` and `age` to be **positional-only**.

### **Keyword-Only (`*` separator)**

```python
def greet(*, name, age):
    print(f"Hello, my name is {name} and I am {age} years old.")

greet(name="Alice", age=25)  # ✅ Allowed
greet("Alice", 25)  # ❌ Error (must be keyword)
```

- The `*` **forces** `name` and `age` to be **keyword-only**.

---

## **7. Best Practices**

✅ Use **positional arguments** for required values.  
✅ Use **keyword arguments** for optional values.  
✅ Use **default arguments** to avoid unnecessary inputs.  
✅ Use `*args` and `**kwargs` for flexible functions.  
✅ Use `/` and `*` to enforce argument restrictions.

## Basic Concepts

In Python, there are two main ways to pass arguments to functions:

- **Positional arguments**: Passed based on their position/order
- **Keyword arguments**: Passed with an explicit parameter name

```python
def greet(name, message):
    print(f"{message}, {name}!")
```

## Positional Arguments

- Matched to parameters based on order
- Must be provided unless default value exists
- Must come before keyword arguments in a call

```python
def power(base, exponent):
    return base ** exponent

print(power(2, 3))  # 8 (2^3)
print(power(3, 2))  # 9 (3^2)
```

## Keyword Arguments

- Identified by parameter name
- Order doesn't matter
- Make code more readable

```python
def create_user(name, age, email):
    print(f"Creating user: {name}, {age}, {email}")

# All keyword arguments
create_user(name="Alice", age=25, email="alice@example.com")

create_user("Bob", email="bob@example.com", age=30)
```

- Parameters can have default values
- Become optional when calling
- Must come after non-default parameters

```python
def greet(name, message="Hello"):
```

```python
def repeat(num_times):

        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(num_times=3)
def greet(name):

greet("Alice")
# Hello Alice
# Hello Alice
# Hello Alice
```

```python
from functools import partial

def power(base, exponent):

square = partial(power, exponent=2)

```

### Argument forwarding

```python
def trace(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__} with {args}, {kwargs}")
        return func(*args, **kwargs)
    return wrapper

@trace
def greet(name, message="Hello"):
    print(f"{message}, {name}!")

greet("Alice", message="Hi")
# Calling greet with ('Alice',), {'message': 'Hi'}
# Hi, Alice!
```

## Positional and Keyword Arguments in Python

In Python, when you call a function, the arguments you pass can be classified based on how they are matched to the parameters defined in the function signature:

- **Positional Arguments:** Arguments whose meaning is determined by their position in the function call.
- **Keyword Arguments:** Arguments whose meaning is determined by an explicit keyword (parameter name) during the function call.

---

### 1. The Basics: Positional Arguments

By default, Python matches arguments to parameters based on their order.

**Definition:**

```python
def greet(name, message):
    print(f"Hello, {name}! {message}")

# Calling the function
greet("Alice", "How are you?")
# Output: Hello, Alice! How are you?
```

In this example:

- `"Alice"` is the first positional argument, matching the `name` parameter.
- `"How are you?"` is the second positional argument, matching the `message` parameter.

**Order Matters:** If you swap the order, the meaning changes:

```python
greet("How are you?", "Alice")
# Output: Hello, How are you?! Alice
```

This clearly demonstrates that the _position_ of the argument determines which parameter it's assigned to.

---

### 2. The Basics: Keyword Arguments

Keyword arguments allow you to explicitly name the parameter you're passing a value to.

**Definition:** (Same as above)

```python
def greet(name, message):
    print(f"Hello, {name}! {message}")

# Calling the function using keyword arguments
greet(name="Bob", message="Nice to see you.")
# Output: Hello, Bob! Nice to see you.
```

Here:

- `name="Bob"` explicitly assigns `"Bob"` to the `name` parameter.
- `message="Nice to see you."` explicitly assigns `"Nice to see you."` to the `message` parameter.

**Order Doesn't (Strictly) Matter:** With keyword arguments, the order is less important because the name clarifies the intent.

```python
greet(message="Welcome!", name="Charlie")
# Output: Hello, Charlie! Welcome!
```

---

### 3. Mixing Positional and Keyword Arguments

You can combine both types in a single function call, but with a strict rule:

**Rule:** All positional arguments _must_ come before any keyword arguments.

```python
def describe_person(name, age, city):
    print(f"{name} is {age} years old and lives in {city}.")

# Valid mixes
describe_person("David", 30, city="New York")
# Output: David is 30 years old and lives in New York.

describe_person("Eve", city="London", age=25) # Order of keywords doesn't matter
# Output: Eve is 25 years old and lives in London.

# INVALID: Keyword argument before positional
# describe_person(name="Frank", 40, "Paris") # This will raise a SyntaxError
```

---

### 4. Default Arguments

Default arguments allow you to provide a default value for a parameter. If the caller doesn't provide a value for that parameter, the default is used.

**Definition:**

```python
def send_email(to_address, subject="No Subject", body=""):
    print(f"To: {to_address}")
    print(f"Subject: {subject}")
    print(f"Body: {body}\n")

# Using defaults
send_email("alice@example.com")
# Output:
# To: alice@example.com
# Subject: No Subject
# Body:

send_email("bob@example.com", "Meeting Reminder")
# Output:
# To: bob@example.com
# Subject: Meeting Reminder
# Body:

# Overriding defaults using positional arguments
send_email("charlie@example.com", "Important News", "Check your inbox for details.")
# Output:
# To: charlie@example.com
# Subject: Important News
# Body: Check your inbox for details.

# Overriding defaults using keyword arguments
send_email("david@example.com", body="This is the main content.", subject="Update")
# Output:
# To: david@example.com
# Subject: Update
# Body: This is the main content.
```

**Rule for Default Arguments:** Parameters with default values _must_ come after parameters without default values.

```python
# Valid
def func(a, b, c=1, d=2): pass

# INVALID
# def func(a, b=1, c, d): pass # SyntaxError: non-default argument follows default argument
```

This rule ensures that Python can unambiguously map positional arguments to parameters.

---

### 5. Arbitrary Positional Arguments (`*args`)

Sometimes you don't know how many positional arguments a function will receive. The `*args` syntax allows a function to accept a variable number of positional arguments. These arguments are collected into a tuple.

**Definition:**

```python
def calculate_sum(*numbers):
    total = 0
    for num in numbers:
        total += num
    print(f"The sum is: {total}")

calculate_sum(1, 2, 3)          # numbers = (1, 2, 3)
# Output: The sum is: 6
calculate_sum(10, 20, 30, 40, 50) # numbers = (10, 20, 30, 40, 50)
# Output: The sum is: 150
calculate_sum()                 # numbers = ()
# Output: The sum is: 0
```

- `*numbers` will capture all extra positional arguments into a tuple named `numbers`.
- You can mix `*args` with regular positional arguments, but `*args` usually comes near the end of the parameter list.

```python
def describe_items(category, *items):
    print(f"Category: {category}")
    if items:
        print("Items:")
        for item in items:
            print(f"- {item}")
    else:
        print("No items in this category.")

describe_items("Fruits", "Apple", "Banana", "Cherry")
# Output:
# Category: Fruits
# Items:
# - Apple
# - Banana
# - Cherry

describe_items("Vegetables")
# Output:
# Category: Vegetables
# No items in this category.
```

---

### 6. Arbitrary Keyword Arguments (`**kwargs`)

Similarly, if you don't know how many keyword arguments a function will receive, `**kwargs` allows it to accept a variable number of them. These arguments are collected into a dictionary.

**Definition:**

```python
def print_profile(**details):
    print("--- User Profile ---")
    for key, value in details.items():
        print(f"{key.replace('_', ' ').title()}: {value}")
    print("--------------------")

print_profile(name="Alice", age=30, city="New York")
# Output:
# --- User Profile ---
# Name: Alice
# Age: 30
# City: New York
# --------------------

print_profile(product="Laptop", price=1200, brand="Dell", year=2023)
# Output:
# --- User Profile ---
# Product: Laptop
# Price: 1200
# Brand: Dell
# Year: 2023
# --------------------
```

- `**details` will capture all extra keyword arguments into a dictionary named `details`.
- `**kwargs` must always be the _last_ parameter in a function definition.

---

### 7. Parameter Order in Function Definitions

When defining a function, there's a specific order in which different types of parameters must appear:

1.  **Positional-or-Keyword Parameters:** Regular parameters that can be passed by position or keyword (e.g., `name`, `message`).
2.  **`*args`:** Arbitrary positional arguments.
3.  **Keyword-Only Parameters:** Parameters that _must_ be passed by keyword (introduced with `*` or `*args`).
4.  **`**kwargs`:\*\* Arbitrary keyword arguments.

**Example Illustrating Order:**

```python
def complex_function(
    p1,                  # Positional-or-Keyword (required)
    p2="default_p2",     # Positional-or-Keyword (with default)
    *args,               # Arbitrary positional arguments
    kw1,                 # Keyword-only (required)
    kw2="default_kw2",   # Keyword-only (with default)
    **kwargs             # Arbitrary keyword arguments
):
    print(f"p1: {p1}")
    print(f"p2: {p2}")
    print(f"args: {args}")
    print(f"kw1: {kw1}")
    print(f"kw2: {kw2}")
    print(f"kwargs: {kwargs}")

# Valid call
complex_function(
    10,                   # p1 (positional)
    20,                   # p2 (positional, overriding default)
    30, 40,               # *args
    kw1="value_kw1",      # kw1 (keyword-only)
    kw2="value_kw2",      # kw2 (keyword-only, overriding default)
    extra_arg="hello",    # **kwargs
    another_arg=True
)
# Output:
# p1: 10
# p2: 20
# args: (30, 40)
# kw1: value_kw1
# kw2: value_kw2
# kwargs: {'extra_arg': 'hello', 'another_arg': True}

# Another valid call
complex_function("a", kw1="b")
# Output:
# p1: a
# p2: default_p2
# args: ()
# kw1: b
# kw2: default_kw2
# kwargs: {}

# INVALID calls (will raise TypeError):
# complex_function() # Missing p1, kw1
# complex_function(10, 20, kw2="x", kw1="y", 30) # Positional after keyword
# complex_function(10, 20, 30, 40) # Missing required keyword-only arg kw1
```

---

### 8. Unpacking Arguments (The `*` and `**` Operators in Function Calls)

The `*` and `**` operators aren't just for defining functions; they can also be used when _calling_ them.

- **`*iterable`:** Unpacks an iterable (list, tuple, set, string) into separate positional arguments.
- **`**dictionary`:\*\* Unpacks a dictionary into separate keyword arguments.

```python
def show_items(a, b, c):
    print(f"a: {a}, b: {b}, c: {c}")

# Unpacking a list into positional arguments
my_list = [1, 2, 3]
show_items(*my_list)
# Output: a: 1, b: 2, c: 3

# Unpacking a tuple
my_tuple = ('x', 'y', 'z')
show_items(*my_tuple)
# Output: a: x, b: y, c: z

def show_info(name, age, city):
    print(f"{name}, {age}, {city}")

# Unpacking a dictionary into keyword arguments
my_dict = {'name': 'Grace', 'age': 28, 'city': 'Berlin'}
show_info(**my_dict)
# Output: Grace, 28, Berlin

# Mixing explicit arguments with unpacking
def show_details(id, name, type, **kwargs):
    print(f"ID: {id}, Name: {name}, Type: {type}")
    print(f"Other details: {kwargs}")

product_details = {'name': 'Laptop', 'type': 'Electronics', 'price': 1500, 'brand': 'HP'}
show_details(101, **product_details)
# Output:
# ID: 101, Name: Laptop, Type: Electronics
# Other details: {'price': 1500, 'brand': 'HP'}
```

### 9. Positional-Only Parameters (`/`) - Advanced

Introduced in Python 3.8, you can specify parameters that _must_ be passed positionally. This is useful for API design where you want to prevent users from accidentally relying on keyword argument names that might change in the future, or when parameter names aren't meant to be part of the public API.

You define positional-only parameters by placing a `/` (slash) after them. Any parameter before the `/` must be positional.

```python
def create_point(x, y, /):
    print(f"Point created at ({x}, {y})")

create_point(10, 20)  # Valid: positional
# Output: Point created at (10, 20)

# create_point(x=10, y=20) # TypeError: create_point() got some positional-only arguments passed as keyword arguments: 'x, y'
```

You can combine positional-only, positional-or-keyword, and keyword-only parameters:

```python
def example_args(pos_only_arg, /, normal_arg, *, kw_only_arg):
    print(f"Positional-only: {pos_only_arg}")
    print(f"Normal: {normal_arg}")
    print(f"Keyword-only: {kw_only_arg}")

example_args(100, 200, kw_only_arg=300)
# Output:
# Positional-only: 100
# Normal: 200
# Keyword-only: 300

# example_args(pos_only_arg=100, normal_arg=200, kw_only_arg=300) # TypeError for pos_only_arg
# example_args(100, kw_only_arg=300, normal_arg=200) # Valid: order of keywords doesn't matter for normal_arg too
# example_args(100, 200, 300) # TypeError: kw_only_arg must be passed as keyword
```

---

### 10. Keyword-Only Parameters (`*` or `*args` before them) - Advanced

We briefly touched on this earlier, but `*` (or `*args` if you also need arbitrary positional args) in the parameter list means that all subsequent parameters _must_ be passed as keyword arguments.

```python
def configure_settings(log_level, *, debug_mode=False, timeout=30):
    print(f"Log Level: {log_level}")
    print(f"Debug Mode: {debug_mode}")
    print(f"Timeout: {timeout} seconds")

configure_settings("INFO", debug_mode=True)
# Output:
# Log Level: INFO
# Debug Mode: True
# Timeout: 30 seconds

# configure_settings("DEBUG", True) # TypeError: configure_settings() takes 1 positional argument but 2 were given
# configure_settings("ERROR", timeout=60)
# Output:
# Log Level: ERROR
# Debug Mode: False
# Timeout: 60 seconds
```

This is excellent for functions with many optional parameters, making calls clearer and preventing accidental misinterpretations of positional arguments.

### Why is this important?

- **Readability:** Keyword arguments make code self-documenting. `send_email(to="john@example.com", subject="Hello")` is clearer than `send_email("john@example.com", "Hello")` if you don't know the function signature.
- **Flexibility:** Default arguments allow functions to be called with fewer arguments, making them more versatile. `*args` and `**kwargs` allow for highly flexible function designs.
- **Maintainability:** When function signatures change (e.g., adding a new parameter), keyword arguments can often reduce the need to update all existing calls, especially if the new parameter has a default. Positional-only and keyword-only parameters help enforce how a function should be used, making its API more stable.
- **Preventing Errors:** Explicitly naming arguments helps prevent errors when arguments are inadvertently swapped, especially for parameters of the same type.

Mastering positional and keyword arguments, along with `*args`, `**kwargs`, positional-only, and keyword-only parameters, is key to writing professional and effective Python code.


````python

---

## Python Positional and Keyword Arguments: From Basic to Advanced

### 1. Introduction: What are Arguments?

When you define a function, you specify **parameters** in its signature. When you call that function, you provide **arguments** for those parameters.

Arguments can be passed in two primary ways:

1.  **Positional Arguments**: Their meaning is determined by their position (order) in the function call.
2.  **Keyword Arguments**: Their meaning is determined by their explicit keyword (name) in the function call, regardless of position.

---

### 2. Basic Positional Arguments

* **Definition**: Arguments assigned to parameters based solely on their order.
* **Order Matters**: The first argument in the call matches the first parameter in the definition, the second matches the second, and so on.

**Example:**

```python
def greet(name, message):
    """Greets a person with a message."""
    print(f"Hello, {name}! {message}")

# Calling with positional arguments
greet("Alice", "Nice to meet you!")
# Output: Hello, Alice! Nice to meet you!

greet("Bob", "How are you?")
# Output: Hello, Bob! How are you?

# Order matters: If you swap them, the meaning changes
greet("How are you?", "Charlie")
# Output: Hello, How are you?! Charlie
# (Clearly not what we intended)
```

---

### 3. Basic Keyword Arguments

* **Definition**: Arguments assigned to parameters by explicitly naming the parameter in the function call.
* **Order Irrelevant (within keyword args)**: As long as the parameter name is used, the order of keyword arguments doesn't matter.

**Example:**

```python
def describe_person(name, age, city):
    """Describes a person with their name, age, and city."""
    print(f"{name} is {age} years old and lives in {city}.")

# Calling with keyword arguments
describe_person(name="David", age=30, city="New York")
# Output: David is 30 years old and lives in New York.

# Order does not matter for keyword arguments:
describe_person(city="London", name="Eve", age=25)
# Output: Eve is 25 years old and lives in London.
```

---

### 4. Mixing Positional and Keyword Arguments

* **Rule**: Any positional arguments **must** come before any keyword arguments in the function call.
* **Best Practice**: Once you start using keyword arguments, stick to them for the rest of the arguments in that call for clarity.

**Example:**

```python
def create_product(product_id, name, price, quantity):
    """Creates a product entry."""
    print(f"Product ID: {product_id}, Name: {name}, Price: ${price:.2f}, Quantity: {quantity}")

# Valid mixing: Positional first, then keyword
create_product(101, "Laptop", price=1200.50, quantity=5)
# Output: Product ID: 101, Name: Laptop, Price: $1200.50, Quantity: 5

create_product(202, name="Mouse", price=25.00, quantity=20)
# Output: Product ID: 202, Name: Mouse, Price: $25.00, Quantity: 20

# Invalid mixing: Keyword first, then positional
# create_product(product_id=303, "Keyboard", price=75.00, quantity=10)
# SyntaxError: positional argument follows keyword argument
```

---

### 5. Default Arguments

* **Definition**: Parameters can be assigned default values in the function definition. If an argument is not provided for such a parameter during the call, its default value is used.
* **Flexibility**: Allows functions to be called with fewer arguments, making them more versatile.
* **Rule**: Parameters with default values must come *after* parameters without default values in the function definition.

**Example:**

```python
def send_email(to_address, subject="No Subject", body="", sender="noreply@example.com"):
    """Simulates sending an email."""
    print(f"From: {sender}")
    print(f"To: {to_address}")
    print(f"Subject: {subject}")
    print(f"Body: {body}\n---")

# Using all defaults for optional arguments
send_email("receiver@example.com")
# Output:
# From: noreply@example.com
# To: receiver@example.com
# Subject: No Subject
# Body:
# ---

# Overriding some defaults using positional arguments (if order allows)
send_email("jane@example.com", "Meeting Reminder")
# Output:
# From: noreply@example.com
# To: jane@example.com
# Subject: Meeting Reminder
# Body:
# ---

# Overriding defaults using keyword arguments (recommended for clarity)
send_email("john@example.com", body="Please find the report attached.", subject="Report Available")
# Output:
# From: noreply@example.com
# To: john@example.com
# Subject: Report Available
# Body: Please find the report attached.
# ---

# Overriding sender using keyword argument
send_email("admin@example.com", sender="support@mycompany.com", subject="Urgent Issue")
# Output:
# From: support@mycompany.com
# To: admin@example.com
# Subject: Urgent Issue
# Body:
# ---
```

---

### 6. Arbitrary Positional Arguments (`*args`)

* **Definition**: When you don't know in advance how many positional arguments a function will receive, you can use `*args`.
* **Behavior**: It collects all extra positional arguments into a **tuple**.
* **Naming Convention**: `args` is just a convention; you could use `*items` or `*numbers`, but `*args` is widely accepted.

**Example:**

```python
def calculate_sum(*numbers):
    """Calculates the sum of an arbitrary number of numbers."""
    total = 0
    for num in numbers:
        total += num
    print(f"Numbers: {numbers}, Sum: {total}")

calculate_sum(1, 2, 3)
# Output: Numbers: (1, 2, 3), Sum: 6

calculate_sum(10, 20, 30, 40, 50)
# Output: Numbers: (10, 20, 30, 40, 50), Sum: 150

calculate_sum() # Can be called with no extra arguments
# Output: Numbers: (), Sum: 0

# You can combine with regular positional arguments
def process_data(header, *values):
    print(f"Header: {header}")
    print(f"Values: {values}")

process_data("Temperatures", 25, 27, 26.5, 28)
# Output:
# Header: Temperatures
# Values: (25, 27, 26.5, 28)
```

---

### 7. Arbitrary Keyword Arguments (`**kwargs`)

* **Definition**: When you don't know in advance how many keyword arguments a function will receive, you can use `**kwargs`.
* **Behavior**: It collects all extra keyword arguments into a **dictionary**. The keys of the dictionary are the argument names, and the values are their corresponding values.
* **Naming Convention**: `kwargs` is just a convention.

**Example:**

```python
def build_profile(name, **details):
    """Builds a user profile with arbitrary additional details."""
    profile = {'name': name}
    profile.update(details) # Add all details from the kwargs dictionary
    print("Profile:", profile)

build_profile("Sarah", age=28, city="San Francisco", occupation="Engineer")
# Output: Profile: {'name': 'Sarah', 'age': 28, 'city': 'San Francisco', 'occupation': 'Engineer'}

build_profile("Mike", email="mike@example.com")
# Output: Profile: {'name': 'Mike', 'email': 'mike@example.com'}

build_profile("Lisa")
# Output: Profile: {'name': 'Lisa'}
```

---

### 8. Combining `*args` and `**kwargs`

* **Standard Order of Parameters**: When defining a function that uses all types of arguments, the standard (and required) order is:
    1.  Standard positional arguments
    2.  Positional-only arguments (Python 3.8+)
    3.  `*args`
    4.  Keyword-only arguments (Python 3.0+)
    5.  `**kwargs`

* **Typical Signature**: `def function(arg1, arg2, *args, kw_only_arg, **kwargs):` (with `kw_only_arg` being optional)

**Example:**

```python
def configure_device(device_id, *settings, debug_mode=False, **extra_config):
    """
    Configures a device with ID, a list of settings,
    an optional debug mode, and extra keyword configurations.
    """
    print(f"Device ID: {device_id}")
    print(f"Settings: {settings}")
    print(f"Debug Mode: {debug_mode}")
    print(f"Extra Config: {extra_config}")
    print("---")

configure_device(101, "mode:auto", "channel:5", debug_mode=True, timeout=30, retries=3)
# Output:
# Device ID: 101
# Settings: ('mode:auto', 'channel:5')
# Debug Mode: True
# Extra Config: {'timeout': 30, 'retries': 3}
# ---

configure_device(202, "status:on")
# Output:
# Device ID: 202
# Settings: ('status:on',)
# Debug Mode: False
# Extra Config: {}
# ---
```

---

### 9. Positional-Only Arguments (`/`) (Python 3.8+)

* **Purpose**: To force certain parameters to *only* accept positional arguments, preventing them from being called with keywords.
* **Syntax**: Parameters listed before a `/` (slash) in the function signature are positional-only.
* **Use Cases**:
    * When the parameter name isn't part of the function's public API (e.g., internal implementation detail).
    * To allow changing parameter names in the future without breaking user code.
    * For consistency with existing C APIs.

**Example:**

```python
def divide(numerator, denominator, /): # numerator and denominator are positional-only
    """Divides numerator by denominator. Must use positional arguments."""
    return numerator / denominator

print(f"Result: {divide(10, 2)}")
# Output: Result: 5.0

# Invalid call: Cannot use keyword arguments for positional-only parameters
# print(divide(numerator=10, denominator=2))
# TypeError: divide() got some positional-only arguments passed as keyword arguments: 'numerator, denominator'
```

---

### 10. Keyword-Only Arguments (`*` or `*, arg`) (Python 3.0+)

* **Purpose**: To force certain parameters to *only* accept keyword arguments, preventing them from being called positionally.
* **Syntax**:
    * If `*args` is present, any parameters after `*args` are keyword-only.
    * If `*args` is *not* present, a single `*` (asterisk) can be used to indicate that all subsequent parameters are keyword-only.
* **Use Cases**:
    * Improve readability for functions with many parameters or complex options.
    * Ensure parameters are always explicitly named, reducing ambiguity.
    * Prevent accidental reordering of arguments.

**Example (using `*`):**

```python
def send_notification(message, *, recipient, priority="normal"): # recipient and priority are keyword-only
    """Sends a notification to a specific recipient with a given priority."""
    print(f"Sending '{message}' to {recipient} with priority: {priority}")

send_notification("System update available", recipient="admin@example.com")
# Output: Sending 'System update available' to admin@example.com with priority: normal

send_notification("Low battery", recipient="mobile_user@example.com", priority="high")
# Output: Sending 'Low battery' to mobile_user@example.com with priority: high

# Invalid call: Cannot pass recipient or priority positionally
# send_notification("Error", "user@example.com")
# TypeError: send_notification() takes 1 positional argument but 2 were given
```

**Example (using `*args` implicitly making arguments keyword-only):**

```python
def log_event(event_type, *details, timestamp=None, user_id=None):
    """Logs an event with optional details and keyword-only metadata."""
    log_entry = f"[{event_type}]"
    if details:
        log_entry += f" Details: {details}"
    if timestamp:
        log_entry += f" Timestamp: {timestamp}"
    if user_id:
        log_entry += f" User ID: {user_id}"
    print(log_entry)

log_event("Login", "Successful", "From IP: 192.168.1.1", timestamp="2023-10-26 10:00:00", user_id=123)
# Output: [Login] Details: ('Successful', 'From IP: 192.168.1.1') Timestamp: 2023-10-26 10:00:00 User ID: 123

# `timestamp` and `user_id` MUST be passed as keywords because they come after `*details`.
# log_event("Logout", timestamp="...", user_id=...) # This works
# log_event("Logout", "...", "...") # This also works, but '...' go into *details
```

---

### 11. Argument Unpacking (Splitting Arguments)

You can use the `*` and `**` operators not just in function definitions, but also when *calling* functions, to unpack iterables and dictionaries into arguments.

**11.1. Unpacking Iterables (`*`) for Positional Arguments**

* A list or tuple can be unpacked into individual positional arguments.

**Example:**

```python
def print_coords(x, y, z):
    print(f"X: {x}, Y: {y}, Z: {z}")

coordinates = [10, 20, 30]
print_coords(*coordinates) # Unpacks [10, 20, 30] into 10, 20, 30
# Output: X: 10, Y: 20, Z: 30

numbers_to_sum = (5, 8, 12, 1)
# Using our earlier calculate_sum function
calculate_sum(*numbers_to_sum) # Unpacks the tuple
# Output: Numbers: (5, 8, 12, 1), Sum: 26
```

**11.2. Unpacking Dictionaries (`**`) for Keyword Arguments**

* A dictionary can be unpacked into keyword arguments, where dictionary keys become parameter names.

**Example:**

```python
def configure_network(ip_address, subnet_mask, gateway, dns_server):
    print(f"IP: {ip_address}, Subnet: {subnet_mask}, Gateway: {gateway}, DNS: {dns_server}")

network_settings = {
    'ip_address': '192.168.1.100',
    'subnet_mask': '255.255.255.0',
    'gateway': '192.168.1.1',
    'dns_server': '8.8.8.8'
}

configure_network(**network_settings) # Unpacks the dictionary
# Output: IP: 192.168.1.100, Subnet: 255.255.255.0, Gateway: 192.168.1.1, DNS: 8.8.8.8

# Combine with other arguments
def process_config(config_name, **params):
    print(f"Processing config: {config_name}")
    print("Parameters:", params)

additional_settings = {'timeout': 60, 'retries': 5}
process_config("Web Server", port=80, **additional_settings)
# Output:
# Processing config: Web Server
# Parameters: {'port': 80, 'timeout': 60, 'retries': 5}
```

---

### 12. Practical Use Cases and Best Practices

* **Readability**:
    * Use **keyword arguments** for parameters that are not immediately obvious in their meaning, especially boolean flags (e.g., `is_active=True` vs. just `True`).
    * If a function has many optional parameters with default values, using keyword arguments for those you *do* provide makes the call much clearer.
* **Flexibility**:
    * `*args` and `**kwargs` are invaluable when designing functions that need to accept an unpredictable number of inputs, especially when acting as wrappers or proxies for other functions.
* **Clarity of API**:
    * **Positional-only arguments** (`/`) clarify that certain parameters are strictly about their order and that their names are not part of the stable public API, allowing you to refactor names without breaking user code.
    * **Keyword-only arguments** (`*` or `*, arg`) enforce that parameters are always explicitly named, preventing calling mistakes and improving self-documentation.
* **Argument Order**: Always follow the standard parameter order in function definitions for consistency:
    `positional_only_args, /, positional_or_keyword_args, *args, keyword_only_args, **kwargs`
* **Default Values**: Use default values to make functions more convenient for common use cases while allowing customization when needed.
* **Unpacking**: Use `*` and `**` when calling functions to programmatically pass arguments, which is common when dealing with configuration dictionaries or lists of items.

By mastering these concepts, you gain significant control over how your functions are defined and called, leading to more maintainable and user-friendly Python code.
````

## Basic Concepts

Python functions can accept arguments in two main ways:

- **Positional arguments**: Matched by position
- **Keyword arguments**: Matched by name

```python
def greet(name, message):
    print(f"{message}, {name}!")

# Positional arguments
greet("Alice", "Hello")  # "Hello, Alice!"

# Keyword arguments
greet(message="Hi", name="Bob")  # "Hi, Bob!"
```

## Positional Arguments

Arguments passed in the order of the function's parameters.

```python
def power(base, exponent):
    return base ** exponent

result = power(2, 3)  # 8 (2^3)
```

### Important Characteristics:

- Order matters
- Required unless default value is specified
- Must come before keyword arguments in function calls

## Keyword Arguments

Arguments identified by parameter names.

```python
def describe_pet(animal, name):
    print(f"I have a {animal} named {name}.")

# Using keyword arguments (order doesn't matter)
describe_pet(name="Whiskers", animal="cat")
```

### Benefits:

- Improves code readability
- Allows skipping optional arguments
- Order-independent

## Default Arguments

Parameters with default values become optional.

```python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")

greet("Alice")  # Uses default: "Hello, Alice!"
greet("Bob", "Hi")  # Overrides default: "Hi, Bob!"
```

### Important Notes:

- Default parameters are evaluated when the function is defined, not when called
- Mutable defaults (like lists/dicts) can lead to unexpected behavior

```python
# Dangerous default value example
def add_item(item, items=[]):  # Same list used across calls!
    items.append(item)
    return items

# Better approach:
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```

## Variable-Length Arguments

### \*args - Variable positional arguments

Collects extra positional arguments into a tuple.

```python
def sum_numbers(*args):
    return sum(args)

print(sum_numbers(1, 2, 3))  # 6
print(sum_numbers(4, 5, 6, 7))  # 22
```

### \*\*kwargs - Variable keyword arguments

Collects extra keyword arguments into a dictionary.

```python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=25, city="NY")
# name: Alice
# age: 25
# city: NY
```

## Argument Unpacking

### Unpacking positional arguments with \*

```python
def multiply(a, b, c):
    return a * b * c

numbers = [2, 3, 4]
print(multiply(*numbers))  # 24 (equivalent to multiply(2, 3, 4))
```

### Unpacking keyword arguments with \*\*

```python
def greet(name, message):
    print(f"{message}, {name}!")

params = {"name": "Alice", "message": "Hello"}
greet(**params)  # "Hello, Alice!"
```

## Advanced Patterns

### Positional-only parameters (Python 3.8+)

```python
def power(x, y, /):  # Parameters before / are positional-only
    return x ** y

power(2, 3)  # OK
power(x=2, y=3)  # TypeError
```

### Keyword-only parameters

```python
def greet(*, name, message):  # Parameters after * are keyword-only
    print(f"{message}, {name}!")

greet(name="Alice", message="Hi")  # OK
greet("Alice", "Hi")  # TypeError
```

### Combined example

```python
def complex_func(a, b, /, c, d, *, e, f):
    print(a, b, c, d, e, f)

complex_func(1, 2, 3, d=4, e=5, f=6)  # Valid
complex_func(1, 2, c=3, d=4, e=5, f=6)  # Also valid
```

### Type hints with arguments

```python
from typing import Optional

def greet(name: str, message: str = "Hello") -> str:
    return f"{message}, {name}!"

def process_data(*args: int, **kwargs: Optional[str]) -> None:
    pass
```

## Best Practices

1. **Order of parameters**:

   - Positional-only (if any)
   - Regular positional
   - \*args (if needed)
   - Keyword-only (if any)
   - \*\*kwargs (if needed)

   ```python
   def good_func(pos1, pos2, /, standard, *args, kwd_only, **kwargs):
       pass
   ```

2. **When to use keyword arguments**:

   - For clarity with many parameters
   - When you want to skip optional parameters
   - With boolean flags that affect behavior

3. **Avoid mutable default arguments** unless you specifically need that behavior

4. **Use \* and ** to make your API more flexible\*\* when appropriate

5. **Document your functions** to explain expected arguments, especially when using advanced patterns

6. **Consider type hints** to make argument expectations clear

```python
def create_user(
```

````python
# --- 1. Basic Positional Arguments ---
print("--- 1. Basic Positional Arguments ---")

def greet(name, age):
    """
    This function takes two positional arguments: name and age.
    Their meaning is determined by their order in the function call.
    """
    print(f"Hello, {name}! You are {age} years old.")

# Calling with positional arguments
greet("Alice", 30)
# Output: Hello, Alice! You are 30 years old.

# What happens if the order is wrong?
greet(30, "Bob")
# Output: Hello, 30! You are Bob years old. (Demonstrates order matters)
print("-" * 30)

# --- 2. Basic Keyword Arguments ---
print("--- 2. Basic Keyword Arguments ---")

def display_product_info(product_name, price):
    """
    This function displays product information.
    Arguments can be passed using their keywords.
    """
    print(f"Product: {product_name}, Price: ${price:.2f}")

# Calling with keyword arguments
display_product_info(product_name="Laptop", price=1200.50)
# Output: Product: Laptop, Price: $1200.50

# Order of keyword arguments doesn't matter
display_product_info(price=75.99, product_name="Mouse")
# Output: Product: Mouse, Price: $75.99
print("-" * 30)

# --- 3. Mixing Positional and Keyword Arguments ---
print("--- 3. Mixing Positional and Keyword Arguments ---")

def create_user(username, email, role="user"):
    """
    Creates a user with username, email, and an optional role.
    Demonstrates mixing positional and keyword arguments.
    """
    print(f"User Created: Username='{username}', Email='{email}', Role='{role}'")

# Valid: Positional arguments before keyword arguments
create_user("charlie_user", "charlie@example.com", role="admin")
# Output: User Created: Username='charlie_user', Email='charlie@example.com', Role='admin'

create_user("diana_user", "diana@example.com") # Using default for role
# Output: User Created: Username='diana_user', Email='diana@example.com', Role='user'

# Invalid: Keyword argument after a positional argument (SyntaxError)
# Uncomment the line below to see the error:
# create_user(username="invalid_user", "invalid@example.com")
# Output: SyntaxError: positional argument follows keyword argument
print("-" * 30)

# --- 4. Default Arguments ---
print("--- 4. Default Arguments ---")

def log_message(message, level="INFO", timestamp=None):
    """
    Logs a message with an optional level and timestamp.
    Demonstrates parameters with default values.
    """
    if timestamp is None:
        from datetime import datetime
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{timestamp}] [{level}] {message}")

# Using all defaults (except required message)
log_message("Application started.")
# Output: [2025-05-25 13:18:17] [INFO] Application started. (Timestamp will vary)

# Overriding level
log_message("Error occurred in module X.", level="ERROR")
# Output: [2025-05-25 13:18:17] [ERROR] Error occurred in module X.

# Overriding all
log_message("User logged out.", level="DEBUG", timestamp="2024-01-15 10:30:00")
# Output: [2024-01-15 10:30:00] [DEBUG] User logged out.

# Default arguments must come after non-default arguments in definition:
# Invalid:
# def bad_func(a, b="default", c): pass # SyntaxError: non-default argument follows default argument
print("-" * 30)

# --- 5. Arbitrary Positional Arguments (`*args`) ---
print("--- 5. Arbitrary Positional Arguments (`*args`) ---")

def print_shopping_list(store_name, *items):
    """
    Prints a shopping list for a given store.
    `*items` collects any number of additional positional arguments into a tuple.
    """
    print(f"\nShopping List for: {store_name}")
    if items:
        for item in items:
            print(f"- {item}")
    else:
        print("No items added.")

print_shopping_list("Walmart", "Milk", "Eggs", "Bread")
# Output:
# Shopping List for: Walmart
# - Milk
# - Eggs
# - Bread

print_shopping_list("Target", "Apples")
# Output:
# Shopping List for: Target
# - Apples

print_shopping_list("Local Grocer")
# Output:
# Shopping List for: Local Grocer
# No items added.
print("-" * 30)

# --- 6. Arbitrary Keyword Arguments (`**kwargs`) ---
print("--- 6. Arbitrary Keyword Arguments (`**kwargs`) ---")

def configure_device(**settings):
    """
    Configures a device based on keyword arguments.
    `**settings` collects any number of additional keyword arguments into a dictionary.
    """
    print("\n--- Device Configuration ---")
    if settings:
        for key, value in settings.items():
            print(f"{key.replace('_', ' ').title()}: {value}")
    else:
        print("No specific settings provided.")
    print("--------------------------")

configure_device(device_id="A1B2C3", power_mode="high", network="wifi_home")
# Output:
# --- Device Configuration ---
# Device Id: A1B2C3
# Power Mode: High
# Network: Wifi_Home
# --------------------------

configure_device(brightness=80, color_temp=5000)
# Output:
# --- Device Configuration ---
# Brightness: 80
# Color Temp: 5000
# --------------------------

configure_device()
# Output:
# --- Device Configuration ---
# No specific settings provided.
# --------------------------
print("-" * 30)

# --- 7. Parameter Order in Function Definitions ---
print("--- 7. Parameter Order in Function Definitions ---")

def full_featured_func(
    required_pos,             # 1. Positional-or-Keyword (required)
    default_pos="default_val",# 1. Positional-or-Keyword (with default)
    *dynamic_pos_args,        # 2. Arbitrary positional arguments
    keyword_only_req,         # 3. Keyword-only (required)
    keyword_only_default="kwd",# 3. Keyword-only (with default)
    **dynamic_kwd_args        # 4. Arbitrary keyword arguments
):
    """
    This function demonstrates the strict order of parameters in a definition.
    """
    print(f"\nRequired Positional: {required_pos}")
    print(f"Default Positional: {default_pos}")
    print(f"Dynamic Positional (*args): {dynamic_pos_args}")
    print(f"Required Keyword-Only: {keyword_only_req}")
    print(f"Default Keyword-Only: {keyword_only_default}")
    print(f"Dynamic Keyword (**kwargs): {dynamic_kwd_args}")

# Valid call adhering to the order rules
full_featured_func(
    10,                       # required_pos
    "custom_default",         # default_pos
    100, 200, 300,            # dynamic_pos_args
    keyword_only_req="I'm required!", # keyword_only_req (MUST be keyword)
    extra_data="foo",
    user_id=123               # dynamic_kwd_args
)

# Call with minimal arguments (meeting requirements)
full_featured_func(
    "simple_val",
    keyword_only_req="another_required"
)

# Invalid Calls (uncomment to see TypeErrors/SyntaxErrors):
# full_featured_func(keyword_only_req="Oops", "positional_arg_here") # SyntaxError: positional argument follows keyword argument
# full_featured_func(10) # TypeError: missing 1 required keyword-only argument: 'keyword_only_req'
print("-" * 30)

# --- 8. Unpacking Arguments (using * and ** in function CALLS) ---
print("--- 8. Unpacking Arguments (using * and ** in function CALLS) ---")

def process_data(value1, value2, value3):
    print(f"Processing: {value1}, {value2}, {value3}")

# Unpacking a list
my_list_data = [10, 20, 30]
process_data(*my_list_data)
# Output: Processing: 10, 20, 30

# Unpacking a tuple
my_tuple_data = ("alpha", "beta", "gamma")
process_data(*my_tuple_data)
# Output: Processing: alpha, beta, gamma

def configure_server(ip, port, protocol, **options):
    print(f"Server at {ip}:{port} using {protocol}")
    print(f"Options: {options}")

# Unpacking a dictionary for keyword arguments
server_config = {
    'ip': '192.168.1.1',
    'port': 8080,
    'protocol': 'HTTP',
    'max_connections': 100,
    'secure': True
}
configure_server(**server_config)
# Output:
# Server at 192.168.1.1:8080 using HTTP
# Options: {'max_connections': 100, 'secure': True}

# Mixing explicit and unpacked arguments
partial_config = {'protocol': 'HTTPS', 'timeout': 60}
configure_server("10.0.0.5", 443, **partial_config)
# Output:
# Server at 10.0.0.5:443 using HTTPS
# Options: {'timeout': 60}
print("-" * 30)

# --- 9. Positional-Only Parameters (`/`) --- Advanced ---
print("--- 9. Positional-Only Parameters (`/`) --- Advanced ---")

def calculate_distance(x1, y1, x2, y2, /):
    """
    Calculates Euclidean distance between two points.
    All arguments before '/' MUST be passed positionally.
    """
    print(f"Calculating distance between ({x1}, {y1}) and ({x2}, {y2})")
    # Actual calculation logic would go here
    return ((x2 - x1)**2 + (y2 - y1)**2)**0.5

print(f"Distance: {calculate_distance(0, 0, 3, 4):.2f}")
# Output:
# Calculating distance between (0, 0) and (3, 4)
# Distance: 5.00

# Invalid: Trying to pass positional-only args as keywords
# Uncomment the line below to see the error:
# calculate_distance(x1=0, y1=0, x2=3, y2=4)
# Output: TypeError: calculate_distance() got some positional-only arguments passed as keyword arguments: 'x1, y1, x2, y2'

# Mixing positional-only, positional-or-keyword, and keyword-only
def advanced_settings(
    threshold_value, /, # positional-only
    mode,                # positional-or-keyword
    *,                   # everything after this is keyword-only
    enable_feature=False,
    log_file="app.log"
):
    print(f"\nThreshold: {threshold_value}")
    print(f"Mode: {mode}")
    print(f"Feature Enabled: {enable_feature}")
    print(f"Log File: {log_file}")

advanced_settings(100, "fast", enable_feature=True)
# Output:
# Threshold: 100
# Mode: fast
# Feature Enabled: True
# Log File: app.log

advanced_settings(200, mode="slow", log_file="debug.log")
# Output:
# Threshold: 200
# Mode: slow
# Feature Enabled: False
# Log File: debug.log

# Invalid calls:
# advanced_settings(threshold_value=100, mode="fast") # TypeError: positional-only argument
# advanced_settings(100, "fast", True) # TypeError: enable_feature must be keyword-only
print("-" * 30)

# --- 10. Keyword-Only Parameters (`*` or `*args` before them) --- Advanced ---
print("--- 10. Keyword-Only Parameters (`*` or `*args` before them) --- Advanced ---")

def process_report(report_id, *, output_format="pdf", detailed=False):
    """
    Processes a report. output_format and detailed MUST be keyword arguments.
    """
    print(f"\nProcessing Report ID: {report_id}")
    print(f"Output Format: {output_format}")
    print(f"Detailed: {detailed}")

process_report(123)
# Output:
# Processing Report ID: 123
# Output Format: pdf
# Detailed: False

process_report(456, output_format="csv", detailed=True)
# Output:
# Processing Report ID: 456
# Output Format: csv
# Detailed: True

# Invalid: Trying to pass keyword-only args as positional
# Uncomment the line below to see the error:
# process_report(789, "xml", True)
# Output: TypeError: process_report() takes 1 positional argument but 3 were given

# Using *args to force keyword-only for subsequent args
def create_config_entry(name, *values, description, version="1.0"):
    """
    Creates a configuration entry. 'description' and 'version' are keyword-only
    because they come after *values.
    """
    print(f"\nConfig Name: {name}")
    print(f"Values: {values}")
    print(f"Description: {description}") # Must be keyword
    print(f"Version: {version}")        # Must be keyword

create_config_entry("API_Key", "abc", "def", description="Authentication key")
# Output:
# Config Name: API_Key
# Values: ('abc', 'def')
# Description: Authentication key
# Version: 1.0

create_config_entry("Logging", "level: DEBUG", "path: /var/log", description="App logging settings", version="2.1")
# Output:
# Config Name: Logging
# Values: ('level: DEBUG', 'path: /var/log')
# Description: App logging settings
# Version: 2.1

# Invalid:
# create_config_entry("DB_Conn", "user: root", "password: 123", "Database connection details")
# Output: TypeError: create_config_entry() missing 1 required keyword-only argument: 'description'
print("-" * 30)
````
