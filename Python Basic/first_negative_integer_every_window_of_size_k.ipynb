# First negative integer in every window of size k

**Medium**

**Company**: Amajon

Given an array and a positive integer k, find the first negative integer for each window(contiguous subarray) of size k. If a window does not contain a negative integer, then print 0 for that window.

# Examples:

```python
Input: arr[] = [-8, 2, 3, -6, 1] , k = 2
Output: [-8, 0, -6, -6]

```

**Explanation**:

- First negative integer for each window of size 2
- [-8, 2] = -8
- [2, 3]= 0 (does not contain a negative integer)
- [3, -6] = -6
- [-6, 10] = -6

```python
Input: arr[] = [12, -1, -7, 8, -15, 30, 16, 28], k = 3
Output: [-1, -1, -7, -15, -15, 0]
```

**Explanation**:

- First negative integer for each window of size 3
- [ 12, -1, -7] = -1
- [-1,-7, 8] = -1
- [-7, 8, -15] = -7
- [8, -15, 30] = -15
- [-15, 30, 16] = -15
- [30, 16, 28] = 0


````python
from collections import deque
from typing import List

class Solution:
    def printFirstNegativeInteger(self, arr: List[int], k: int) -> List[int]:
        n = len(arr)
        dll = deque()
        result = []
        i = 0
        j = 0

        while j < n:
            if arr[j] < 0:
                dll.append(arr[j])
            
            if j - i + 1 == k:
                result.append(dll[0] if dll else 0)
                if arr[i] < 0 and dll:
                    dll.popleft()
                i += 1
            j += 1

        return result
    
	
def test_solution():
    sol = Solution()

    # Basic test case
    assert sol.printFirstNegativeInteger([12, -1, -7, 8, -15, 30, 16, 28], 3) == [-1, -1, -7, -15, -15, 0]

    # Edge case: no negatives
    assert sol.printFirstNegativeInteger([1, 2, 3, 4, 5], 2) == [0, 0, 0, 0]

    # Edge case: all negatives
    assert sol.printFirstNegativeInteger([-5, -10, -20, -30], 2) == [-5, -10, -20]

    # Edge case: k = 1
    assert sol.printFirstNegativeInteger([-1, 2, -3, 4], 1) == [-1, 0, -3, 0]

    # Edge case: k > len(arr)
    assert sol.printFirstNegativeInteger([1, -2], 5) == []

test_solution()
print("All tests passed!")
````

````python
from collections import deque
from typing import List

class Solution:
    def printFirstNegativeInteger(self, arr: List[int], k: int) -> List[int]:
        n = len(arr)
        deq = deque()
        result = []

        # First window
        for i in range(k):
            if arr[i] < 0:
                deq.append(i)

        # Remaining windows
        for i in range(k, n):
            result.append(arr[deq[0]] if deq else 0)

            # Remove indices out of current window
            while deq and deq[0] < i - k + 1:
                deq.popleft()

            if arr[i] < 0:
                deq.append(i)

        # Handle last window
        result.append(arr[deq[0]] if deq else 0)

        return result
    
def test_solution():
    sol = Solution()
    assert sol.printFirstNegativeInteger([12, -1, -7, 8, -15, 30, 16, 28], 3) == [-1, -1, -7, -15, -15, 0]
    assert sol.printFirstNegativeInteger([1, 2, 3, 4, 5], 2) == [0, 0, 0, 0]
    assert sol.printFirstNegativeInteger([-5, -10, -20, -30], 2) == [-5, -10, -20]
    assert sol.printFirstNegativeInteger([-1, 2, -3, 4], 1) == [-1, 0, -3, 0]
    assert sol.printFirstNegativeInteger([1, -2], 5) == []  # k > len(arr)
    print("All tests passed!")

test_solution()
````
