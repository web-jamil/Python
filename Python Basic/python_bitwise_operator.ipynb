### **Python Bitwise Operators (Basic to Advanced)**

Bitwise operators in Python allow you to manipulate individual bits of numbers. These operators work on binary representations of integers. Understanding bitwise operators is useful for tasks like cryptography, network programming, and performance optimization in lower-level programming.

---

### **1. Types of Bitwise Operators in Python**

Here are the main bitwise operators in Python:

1. **AND (`&`)**
2. **OR (`|`)**
3. **XOR (`^`)**
4. **NOT (`~`)**
5. **Left Shift (`<<`)**
6. **Right Shift (`>>`)**

---

### **2. Basic Bitwise Operators**

#### **1. Bitwise AND (`&`)**

- The `&` operator compares each bit of two numbers. If both bits are `1`, the result is `1`; otherwise, it's `0`.

**Example:**

```python
a = 5  # (in binary: 0101)
b = 3  # (in binary: 0011)

result = a & b
print(result)  # Output: 1 (in binary: 0001)
```

- **Explanation**:
  ```
  0101 (5)
  & 0011 (3)
  ---------
    0001 (1)
  ```

#### **2. Bitwise OR (`|`)**

- The `|` operator compares each bit of two numbers. If either bit is `1`, the result is `1`; otherwise, it's `0`.

**Example:**

```python
a = 5  # (in binary: 0101)
b = 3  # (in binary: 0011)

result = a | b
print(result)  # Output: 7 (in binary: 0111)
```

- **Explanation**:
  ```
  0101 (5)
  | 0011 (3)
  ---------
    0111 (7)
  ```

#### **3. Bitwise XOR (`^`)**

- The `^` operator compares each bit of two numbers. If the bits are different, the result is `1`; if they are the same, the result is `0`.

**Example:**

```python
a = 5  # (in binary: 0101)
b = 3  # (in binary: 0011)

result = a ^ b
print(result)  # Output: 6 (in binary: 0110)
```

- **Explanation**:
  ```
  0101 (5)
  ^ 0011 (3)
  ---------
    0110 (6)
  ```

#### **4. Bitwise NOT (`~`)**

- The `~` operator is a unary operator that inverts all the bits of a number. It changes `0` to `1` and `1` to `0`. In Python, this is the two's complement representation, so the result is `-(n+1)`.

**Example:**

```python
a = 5  # (in binary: 0101)

result = ~a
print(result)  # Output: -6 (because ~5 = -(5+1) = -6)
```

- **Explanation**:  
  For `a = 5 (binary: 0101)`, `~a` results in the two's complement, which is `-6`.

---

### **3. Bitwise Shift Operators**

#### **1. Left Shift (`<<`)**

- The `<<` operator shifts the bits of a number to the left by a specified number of positions. It effectively multiplies the number by `2^n`, where `n` is the number of positions.

**Example:**

```python
a = 5  # (in binary: 0101)

result = a << 1  # Shift left by 1
print(result)  # Output: 10 (in binary: 1010)
```

- **Explanation**:

  ```
  0101 (5) << 1
  ---------
  1010 (10)
  ```

- **Left shift by 1** is equivalent to multiplying by 2.

#### **2. Right Shift (`>>`)**

- The `>>` operator shifts the bits of a number to the right by a specified number of positions. It effectively divides the number by `2^n`, where `n` is the number of positions.

**Example:**

```python
a = 5  # (in binary: 0101)

result = a >> 1  # Shift right by 1
print(result)  # Output: 2 (in binary: 0010)
```

- **Explanation**:

  ```
  0101 (5) >> 1
  ---------
  0010 (2)
  ```

- **Right shift by 1** is equivalent to dividing by 2.

### **4. Advanced Bitwise Operations**

#### **1. Swapping Two Numbers Using XOR**

You can swap two numbers without using a temporary variable by using the XOR operator.

**Example:**

```python
a = 5  # (in binary: 0101)
b = 3  # (in binary: 0011)

# Swapping using XOR
a = a ^ b
b = a ^ b
a = a ^ b

print("a:", a)  # Output: 3
print("b:", b)  # Output: 5
```

- **Explanation**:
  After the first XOR (`a = a ^ b`), `a` contains the XOR of `a` and `b`. Then, `b = a ^ b` retrieves the original value of `a`. Finally, `a = a ^ b` retrieves the original value of `b`.

#### **2. Checking if a Number is Power of 2**

A number is a power of 2 if and only if it has exactly one `1` bit in its binary representation. This can be efficiently checked using bitwise AND.

**Example:**

```python
def is_power_of_2(n):
    return n > 0 and (n & (n - 1)) == 0

print(is_power_of_2(16))  # True, because 16 is 2^4
print(is_power_of_2(18))  # False, because 18 is not a power of 2
```

- **Explanation**:  
  If `n` is a power of 2, the result of `(n & (n - 1))` will be `0`.

#### **3. Counting Set Bits (1s)**

To count the number of `1` bits (set bits) in an integer, you can use the `bin()` function in combination with `str.count()`, or use a bitwise operation.

**Example:**

```python
def count_set_bits(n):
    count = 0
    while n:
        count += n & 1
        n >>= 1
    return count

print(count_set_bits(5))  # Output: 2 (binary: 0101)
```

- **Explanation**:  
  The function counts the number of `1` bits by repeatedly checking if the least significant bit is `1` and then right-shifting the number.

#### **4. Clearing Specific Bits**

To clear (set to `0`) specific bits in a number, you can use the bitwise AND with a mask where the bits to be cleared are `0`.

**Example:**

```python
n = 15  # (in binary: 1111)
mask = 13  # (in binary: 1101)

result = n & mask
print(result)  # Output: 13 (in binary: 1101)
```

- **Explanation**:  
  The mask `1101` clears the third bit of `n`.

## **Bitwise AND Operator (`&`) in Python**

### **1. Basic Usage**

The bitwise AND operator works on the binary representation of integers. It compares each bit of two numbers and produces a new number based on the AND logic.

#### **Syntax:**

```python
result = a & b
```

```
  0101 (5)
& 0011 (3)
-------
  0001 (1)
```

---

### **2. How Bitwise AND Works**

The bitwise AND operator compares each bit of the two numbers and applies the AND logic. Here's how it works step by step:

1. Convert the integers to their binary representation.
2. Align the bits of the two numbers.
3. Perform the AND operation on each pair of corresponding bits.
4. Convert the resulting binary number back to a decimal integer.

#### **Example:**

```python
a = 12  # Binary: 1100
b = 10  # Binary: 1010
result = a & b
print(result)  # Output: 8
```

**Explanation:**

```
  1100 (12)
& 1010 (10)
-------
  1000 (8)
```

It compares the corresponding bits of two numbers and returns a new number whose bits are set to `1` where exactly one of the input bits is `1`. Otherwise, the result bit is `0`.

This guide will cover everything about the bitwise XOR operator, from basic usage to advanced techniques.

---

### **1. Basic Usage**

#### **Syntax**

```python
result = a ^ b
```

- `a` and `b` are integers.
- The `^` operator performs a bitwise XOR operation on their binary representations.

#### **How It Works**

- The XOR operation compares each bit of the two numbers:
  - If the bits are different, the result is `1`.

the **bitwise NOT operator (`~`)**, a fundamental unary operator in many programming languages that work directly at the bit level.

**Core Theory:**

The bitwise NOT operator performs a logical negation on each bit of its operand. For every bit in the operand:

- If the bit is **0**, the result in the corresponding bit position becomes **1**.
- If the bit is **1**, the result in the corresponding bit position becomes **0**.

Think of it as flipping each individual bit.

**Representation and Two's Complement:**

The behavior of the bitwise NOT operator is particularly important when dealing with signed integers due to how they are typically represented in computers: **two's complement**.

1.  **Binary Representation:** Every integer is stored in binary form, a sequence of 0s and 1s. The number of bits used depends on the data type (e.g., 8-bit, 16-bit, 32-bit integers).

2.  **Signed Integers and the Sign Bit:** In signed integer representations, the most significant bit (MSB), the leftmost bit, is often reserved as the sign bit:

    - **0** indicates a positive number.
    - **1** indicates a negative number.

3.  **Two's Complement:** Negative numbers are usually represented using the two's complement system. To find the two's complement of a positive number:
    a. Invert all the bits (apply bitwise NOT).
    b. Add 1 to the result.

**How Bitwise NOT Interacts with Two's Complement:**

When you apply the bitwise NOT operator (`~`) to a signed integer, the result is the **bitwise complement** of its binary representation. This directly affects the sign bit and the magnitude of the number when interpreted in two's complement.

Let's consider an 8-bit signed integer example:

- **Number:** `5`
- **Binary (8-bit):** `00000101`
- **`~5` (Bitwise NOT):** `11111010`

Now, how do we interpret `11111010` as a signed 8-bit integer in two's complement? Since the MSB is `1`, it's a negative number. To find its magnitude, we reverse the two's complement process:

1.  **Subtract 1:** `11111010 - 1 = 11111001`
2.  **Invert all bits:** `~11111001 = 00000110`

`00000110` in decimal is `6`. Therefore, `~5` in an 8-bit signed integer system is typically interpreted as `-6`.

**General Pattern:**

For a signed integer `x`, the result of `~x` is often `-(x + 1)`. This is a direct consequence of the two's complement representation.

Let's try to understand why this pattern emerges:

Assume a positive number `x` has a binary representation `B`. Its two's complement representation is simply `B`.

1.  **`~x` (Bitwise NOT of `B`)** gives us the bitwise complement of `B`, let's call it `B'`.

2.  To find the decimal value of `B'` as a two's complement negative number, we need to:
    a. Invert `B'`: `~B' = B`
    b. Add 1: `B + 1`

So, the magnitude of the negative number represented by `B'` is `B + 1`, and since the MSB of `B'` will be `1` (because the MSB of `B` was `0`), it represents a negative number. Therefore, `~x` is interpreted as `-(x + 1)`.

**Unsigned Integers:**

If the operand is an unsigned integer, the bitwise NOT operator simply inverts all the bits. The result is then interpreted as another unsigned integer. There's no sign bit or two's complement involved in the same way.

Example (8-bit unsigned):

- **Number:** `5`
- **Binary (8-bit unsigned):** `00000101`
- **`~5` (Bitwise NOT):** `11111010`

In unsigned decimal, `11111010` is `250`.

**Key Properties and Use Cases:**

1.  **Flipping Bits:** The primary purpose is to invert the bits of a value.

2.  **Creating Masks:** It can be used to create masks with specific bits set or unset. For example, `~0` (all bits 0) will result in a mask with all bits 1 (within the size of the data type).

3.  **Checking for All Bits Set:** You can use `~0` to get a value with all bits set and compare against it.

4.  **Working with Bit Flags:** In scenarios where individual bits represent boolean flags, `~` can be used to invert the state of all flags at once.

5.  **Relationship with Other Operators:**
    - `~x` is not the same as logical NOT (`not x` in Python, `!` in C++), which evaluates the truthiness of a value (non-zero is typically true, zero is false).

Alright, let's dive deep into the concept of **Bitmasking**, a powerful technique in computer science, especially in the realm of algorithms and data structures.

**What is Bitmasking?**

Bitmasking is a technique that uses the bits of an integer to represent a set of boolean flags or small integer values. Each bit in the integer corresponds to a specific element or condition. By manipulating these bits using bitwise operators, you can efficiently perform operations on these sets or flags.

Think of an integer as a sequence of light switches (the bits), where each switch can be either ON (1) or OFF (0), representing the presence or absence of a particular item or the state of a condition.

**Why Use Bitmasking?**

1.  **Space Efficiency:** Instead of using an array of booleans or small integers, you can often represent the same information using a single integer, thus saving memory, especially when dealing with a moderate number of elements (typically up to the number of bits in an integer, e.g., 32 or 64).

2.  **Speed and Efficiency:** Bitwise operations are extremely fast as they are directly supported by the processor's hardware. Operations like checking if an element is present, adding an element, removing an element, or performing set-like operations can be done very quickly.

3.  **Concise Representation:** Bitmasks provide a compact and often elegant way to represent and manipulate sets and states, leading to shorter and potentially easier-to-understand code.

4.  **Facilitates Certain Algorithmic Techniques:** Bitmasking is crucial for certain algorithmic approaches, particularly in dynamic programming (e.g., DP on subsets) and combinatorial problems.

**Core Concepts and Operations:**

1.  **Representing Sets:**

    - Assume you have a set of `n` elements. You can use an integer of `n` bits (or more if needed) to represent any subset of these elements.
    - If the `i`-th bit (0-indexed from the right) is 1, it means the `i`-th element is present in the subset.
    - If the `i`-th bit is 0, it means the `i`-th element is absent.

    **Example:** Let's say we have a set of 3 elements {A, B, C} with indices 0, 1, and 2 respectively.

    - `000` (binary) = 0 (decimal) represents the empty set {}.
    - `001` (binary) = 1 (decimal) represents the set {A}.
    - `010` (binary) = 2 (decimal) represents the set {B}.
    - `011` (binary) = 3 (decimal) represents the set {A, B}.
    - `100` (binary) = 4 (decimal) represents the set {C}.
    - `101` (binary) = 5 (decimal) represents the set {A, C}.
    - `110` (binary) = 6 (decimal) represents the set {B, C}.
    - `111` (binary) = 7 (decimal) represents the set {A, B, C}.

2.  **Basic Bitwise Operations for Set Manipulation:**

    - **Checking if the i-th element is present:**

      ```python
      mask = ... # Your bitmask
      i = ...    # Index of the element

      if (mask >> i) & 1:
          print(f"Element at index {i} is present.")
      else:
          print(f"Element at index {i} is absent.")
      ```

      Explanation: `(mask >> i)` shifts the `i`-th bit to the least significant position. ANDing with `1` isolates that bit.

    - **Adding the i-th element to the set:**

      ```python
      mask = ...
      i = ...

      mask = mask | (1 << i)
      ```

      Explanation: `(1 << i)` creates a mask with only the `i`-th bit set. The OR operation sets the `i`-th bit of `mask` without affecting other bits.

    - **Removing the i-th element from the set:**

      ```python
      mask = ...
      i = ...

      mask = mask & ~(1 << i)
      ```

      Explanation: `~(1 << i)` creates a mask with all bits set to 1 except the `i`-th bit. The AND operation clears the `i`-th bit of `mask`.

    - **Checking if the set is empty:**

      ```python
      mask = ...

      if mask == 0:
          print("The set is empty.")
      ```

    - **Creating a set with all elements present:**

      ```python
      n = ... # Number of elements
      all_present_mask = (1 << n) - 1
      ```

      Explanation: `(1 << n)` creates a number with the `n`-th bit set (and all lower bits 0). Subtracting 1 sets all bits from 0 to `n-1` to 1.

    - **Union of two sets (represented by masks):**

      ```python
      mask1 = ...
      mask2 = ...
      union_mask = mask1 | mask2
      ```

    - **Intersection of two sets:**

      ```python
      mask1 = ...
      mask2 = ...
      intersection_mask = mask1 & mask2
      ```

    - **Difference of two sets (mask1 - mask2):**
      ```python
      mask1 = ...
      mask2 = ...
      difference_mask = mask1 & (~mask2)
      ```

3.  **Iterating Through Subsets (Crucial Technique):**

    A very common application of bitmasking is to iterate through all possible subsets of a set of `n` elements. Since there are `2^n` subsets, you can use a loop from `0` to `2^n - 1`. Each value in the loop represents a bitmask corresponding to a unique subset.

    ```python
    n = 3 # Number of elements
    elements = ['A', 'B', 'C']

    for i in range(1 << n): # Iterate from 0 to 2^n - 1
        subset = []
        for j in range(n):
            if (i >> j) & 1:
                subset.append(elements[j])
        print(f"Subset for mask {bin(i)}: {subset}")
    ```

    This technique is fundamental for solving problems involving exploring all combinations or selections.

4.  **Submask Iteration (Iterating through all submasks of a given mask):**

    Sometimes, you need to iterate through all the subsets of the set represented by a particular bitmask. This can be done efficiently:

    ```python
    mask = 13 # Binary: 1101 (representing a set with elements at indices 0, 2, 3)
    submask = mask
    while submask > 0:
        print(f"Submask: {bin(submask)}")
        submask = (submask - 1) & mask
    print(f"Submask: {bin(0)}") # Don't forget the empty submask
    ```

    This loop efficiently generates all submasks of `mask`.

1.  **Dynamic Programming on Subsets:** Problems where the state of the DP can be represented by a subset of elements. 4.  **Game Theory:** Representing the state of a game or the set of available moves using bitmasks.

1. **AND (`&`)**

     - Mask: `0110` (binary)
     - Value: `1010` (binary)
     - Result: `1010 & 0110 = 0010` (binary)
   - **Usage**: To check whether certain bits are set.

2. **OR (`|`)**

   - **Purpose**: Used to set bits (set to 1).
   - **Theory**: Bits are set to 1 if at least one of the corresponding bits in the operands is 1.
   - **Example**:
     - Mask: `0110`
     - Value: `1010`
     - Result: `1010 | 0110 = 1110`
   - **Usage**: To add flags or permissions to a set.

3. **XOR (`^`)**

   - **Purpose**: Used to toggle bits.
   - **Theory**: Bits are set to 1 if they differ (one is 0, the other is 1).
   - **Example**:
     - Mask: `0110`
     - Value: `1010`
     - Result: `1010 ^ 0110 = 1100`
   - **Usage**: To toggle flags or to revert bits back to their previous state.

4. **NOT (`~`)**

   - **Purpose**: Used to invert bits.
   - **Theory**: Converts all 0s to 1s and all 1s to 0s.
   - **Example**:
     - Value: `1010`
     - Result: `~1010 = 0101` for 4-bit representation.
   - **Usage**: To inverse a bitmask or to negate conditions.

5. **Left Shift (`<<`)**

   - **Purpose**: To shift bits to the left, effectively multiplying the number by 2 for each shift.
   - **Theory**: Bits shifted out on the left are discarded, and new bits on the right become 0.
   - **Example**:
     - Value: `0001` (1)
     - Result: `0001 << 1 = 0010` (2)
   - **Usage**: When creating a bitmask of specific powers of two.

6. **Right Shift (`>>`)**
   - **Purpose**: To shift bits to the right, effectively dividing the number by 2 for each shift.
   - **Theory**: For unsigned types, zeros are shifted in from the left; for signed types, the sign bit is extended.
   - **Example**:
     - Value: `0010` (2)
     - Result: `0010 >> 1 = 0001` (1)
   - **Usage**: To manipulate and check bit positions.

Let's delve into the bitmask operation for **checking if the i-th element is present** within a set represented by a bitmask. This is a fundamental operation in bitmasking techniques.

**Core Theory: Bitmasks and Set Representation**

Before we dive into the operation itself, let's briefly reiterate the core idea of using bitmasks to represent sets:

- An integer (`mask`) acts as a container for the set information.
- Each bit position (typically 0-indexed from the right, representing powers of 2: 2⁰, 2¹, 2², ...) corresponds to a specific element in our conceptual set.
- If the bit at the `i`-th position is `1`, it signifies that the `i`-th element is considered to be _present_ in the set.
- If the bit at the `i`-th position is `0`, it signifies that the `i`-th element is _absent_ from the set.

**The Membership Test Operation: `(mask >> i) & 1`**

To check if the `i`-th element is present in the set represented by the `mask`, we use the following bitwise operation:

```
(mask >> i) & 1
```

Let's break down this operation step by step with its underlying theory:

1.  **Right Shift Operator (`>> i`)**:

    - The right shift operator (`>>`) shifts the bits of the `mask` to the right by `i` positions.
    - Each right shift effectively divides the number by 2 (integer division).
    - The bits shifted out from the right end are discarded.
    - The bits vacated on the left end are filled with `0` if the number is unsigned or non-negative signed, and with the sign bit if the number is negative signed (arithmetic right shift). In the context of checking presence in a set, we're primarily concerned with the relative positions of the bits, and the left-fill behavior doesn't directly impact the bit we're interested in after the shift.

    - **Purpose in Membership Test:** By right-shifting the `mask` by `i` positions, we are effectively moving the `i`-th bit of the original `mask` to the **0-th position** (the least significant bit). This makes it easy to isolate and examine the value of that specific bit.

    - **Example:**
      Let `mask` be `13` (binary `1101`), representing a set where elements at indices 0, 2, and 3 are present. If we want to check if the element at index `2` is present (`i = 2`):
      `mask >> i` becomes `13 >> 2`, which is binary `1101 >> 2 = 0011` (decimal `3`). Now, the bit that was originally at the 2nd position is now at the 0th position.

2.  **Bitwise AND Operator (`& 1`)**:

    - The bitwise AND operator (`&`) compares the corresponding bits of two operands. The resulting bit is `1` only if both corresponding bits are `1`; otherwise, the resulting bit is `0`.
    - We are performing a bitwise AND of the result of the right shift (`mask >> i`) with the integer `1`.
    - The binary representation of `1` (for typical integer sizes) has all bits as `0` except for the least significant bit (0-th position), which is `1`. For example, in 8 bits, `1` is `00000001`.

    - **Purpose in Membership Test:** By ANDing with `1`, we are effectively isolating the value of the **least significant bit** of the right-shifted `mask`. Since the `i`-th bit of the original `mask` is now at this least significant position, the result of the AND operation will be:

      - `1` if the `i`-th bit of the original `mask` was `1`.
      - `0` if the `i`-th bit of the original `mask` was `0`.

    - **Continuing the Example:**
      After the right shift, we had `0011` (decimal `3`). Now we perform `0011 & 0001`:
      ```
        0011
      & 0001
      ------
        0001
      ```
      The result is `1`, indicating that the element at index `2` was indeed present in the original `mask`.

**Putting it Together:**

The combined operation `(mask >> i) & 1` efficiently checks the presence of the `i`-th element by:

1.  Positioning the `i`-th bit to the rightmost end using the right shift.
2.  Isolating the value of that rightmost bit using the bitwise AND with `1`.

**Return Value:**

The expression `(mask >> i) & 1` will evaluate to:

- `1` (True-like value) if the `i`-th bit of `mask` is `1`, meaning the `i`-th element is present in the set.
- `0` (False-like value) if the `i`-th bit of `mask` is `0`, meaning the `i`-th element is absent from the set.

This operation is a cornerstone of many algorithms that utilize bitmasking to represent and manipulate sets of items or states. Its efficiency and simplicity make it a valuable tool in a programmer's arsenal.

Let's dive deep into the bitmask operation for **setting a bit** at a specific position within an integer. This is a fundamental operation in bit manipulation and is crucial when using bitmasks to represent sets or flags.

**Core Theory: Bitmasks and Setting Bits**

As we know, in bitmasking:

- An integer (`mask`) holds information as a sequence of bits.
- Each bit position corresponds to a specific element or flag.
- Setting a bit means changing its value to `1`, indicating the presence of an element or the truth of a flag.

**The Set Bit Operation: `mask | (1 << i)`**

To set the bit at the `i`-th position (0-indexed from the right) in a `mask`, we use the bitwise OR operator (`|`) in conjunction with a left shift operation:

```
mask | (1 << i)
```

Let's break down this operation and the underlying theory:

1.  **Left Shift Operator (`1 << i`)**:

    - The left shift operator (`<<`) shifts the bits of the left operand (`1` in this case) to the left by the number of positions specified by the right operand (`i`).
    - Starting with the binary representation of `1` (which has only the 0-th bit set: `...0001`), shifting it left by `i` positions creates a new value where only the `i`-th bit is set to `1`, and all other bits are `0`.

    - **Purpose in Setting a Bit:** `(1 << i)` generates a specific bitmask that has a `1` only at the exact bit position we want to set in our target `mask`.

    - **Example:**
      If we want to set the 3rd bit (index `i = 3`) in a mask:
      `1 << 3` becomes binary `0001 << 3 = 1000` (decimal `8`). This mask has a `1` only at the 3rd bit position.

2.  **Bitwise OR Operator (`|`)**:

    - The bitwise OR operator (`|`) compares the corresponding bits of two operands. The resulting bit is `1` if at least one of the corresponding bits is `1`; otherwise, the resulting bit is `0`.
    - We are performing a bitwise OR between our original `mask` and the mask generated by `(1 << i)`.

    - **Purpose in Setting a Bit:**

      - If the `i`-th bit in the original `mask` was `0`, the OR operation with the `1` at the `i`-th position from `(1 << i)` will change it to `1`.
      - If the `i`-th bit in the original `mask` was already `1`, the OR operation with another `1` will leave it as `1`.
      - All other bit positions in the original `mask` will remain unchanged because the corresponding bits in `(1 << i)` are `0`, and `bit | 0` results in `bit`.

    - **Continuing the Example:**
      Let our original `mask` be `5` (binary `0101`), and we want to set the 3rd bit (`i = 3`). We calculated `(1 << 3)` as `8` (binary `1000`). Now we perform `mask | (1 << 3)`:
      ```
        0101 (mask = 5)
      | 1000 (1 << 3 = 8)
      ------
        1101 (decimal 13)
      ```
      The resulting mask is `13` (binary `1101`), where the 3rd bit has now been set to `1`. The original 0th and 2nd bits remained `1`, and the 1st bit remained `0`.

**In Summary:**

The operation `mask | (1 << i)` reliably sets the `i`-th bit of the `mask` to `1` without affecting the values of any other bits. It works by creating a specific bitmask with a `1` only at the target position and then using the OR operation to merge this `1` into the original `mask`.

**Use Cases:**

- **Adding Elements to a Set:** When using a bitmask to represent a set, setting the `i`-th bit corresponds to adding the `i`-th element to the set.
- **Setting Flags:** In situations where individual bits represent boolean flags, setting a bit turns a specific flag to `True` or activates a certain state.
- **Building Bitmasks:** You can start with a mask of `0` and iteratively set bits to construct a bitmask representing a desired combination of elements or flags.

This operation is a fundamental building block for effectively manipulating sets and states using bitmasking techniques in various algorithms and data structures.


Let's explore the bitmask operation for **clearing a bit** at a specific position within an integer. Clearing a bit means setting its value to `0`. This is another fundamental operation in bit manipulation, essential when using bitmasks to represent and modify sets or flags.

**Core Theory: Bitmasks and Clearing Bits**

In bitmasking:

- An integer (`mask`) represents a collection of elements or flags through its binary bits.
- Each bit position corresponds to a specific element or flag.
- Clearing the `i`-th bit means forcing its value to `0`, indicating the absence of an element or the falsity of a flag.

**The Clear Bit Operation: `mask & ~(1 << i)`**

To clear the bit at the `i`-th position (0-indexed from the right) in a `mask`, we use the bitwise AND operator (`&`) in combination with the bitwise NOT operator (`~`) and a left shift operation:

```
mask & ~(1 << i)
```

Let's break down this operation and the underlying theory:

1.  **Left Shift Operator (`1 << i`)**:

    - As we discussed in "setting a bit," `(1 << i)` creates an integer with a `1` only at the `i`-th bit position and `0` elsewhere. This mask isolates the bit we are interested in.

    - **Example:** To target the 2nd bit (index `i = 2`), `(1 << 2)` results in binary `0001 << 2 = 0100` (decimal `4`).

2.  **Bitwise NOT Operator (`~`)**:

    - The bitwise NOT operator (`~`) inverts all the bits of its operand. Every `0` becomes `1`, and every `1` becomes `0`.

    - **Purpose in Clearing a Bit:** Applying the NOT operator to the mask `(1 << i)` creates a new mask where all bits are `1` _except_ for the `i`-th bit, which becomes `0`. This is the crucial mask we'll use to clear the corresponding bit in our original `mask`.

    - **Continuing the Example:**
      For `i = 2`, `(1 << 2)` is `0100`. Applying the NOT operator `~(0100)` (assuming an 8-bit integer for illustration) gives `11111011`. Notice that the 2nd bit is now `0`, and all other bits are `1`.

3.  **Bitwise AND Operator (`&`)**:

    - The bitwise AND operator (`&`) compares the corresponding bits of two operands. The resulting bit is `1` only if both corresponding bits are `1`; otherwise, the resulting bit is `0`.
    - We are performing a bitwise AND between our original `mask` and the inverted mask `~(1 << i)`.

    - **Purpose in Clearing a Bit:**

      - If the `i`-th bit in the original `mask` was `1`, the AND operation with the `0` at the `i`-th position in `~(1 << i)` will change it to `0`.
      - If the `i`-th bit in the original `mask` was already `0`, the AND operation with another `0` will leave it as `0`.
      - All other bit positions in the original `mask` will remain unchanged because they are being ANDed with `1` in the `~(1 << i)` mask (`bit & 1` results in `bit`).

    - **Continuing the Example:**
      Let our original `mask` be `7` (binary `0111`), and we want to clear the 2nd bit (`i = 2`). We calculated `~(1 << 2)` as `11111011`. Now we perform `mask & ~(1 << 2)`:
      ```
        0111 (mask = 7)
      & 11111011 (~(1 << 2))
      ----------
        0011 (decimal 3)
      ```
      The resulting mask is `3` (binary `0011`), where the 2nd bit has now been cleared to `0`. The original 0th and 1st bits remained `1`.

**In Summary:**

The operation `mask & ~(1 << i)` reliably clears the `i`-th bit of the `mask` to `0` without affecting the values of any other bits. It works by creating a specific bitmask where all bits are `1` except for the target position, and then using the AND operation to force the target bit in the original mask to `0`.

**Use Cases:**

- **Removing Elements from a Set:** When using a bitmask to represent a set, clearing the `i`-th bit corresponds to removing the `i`-th element from the set.
- **Resetting Flags:** In situations where individual bits represent boolean flags, clearing a bit turns a specific flag to `False` or deactivates a certain state.
- **Masking Operations:** Creating masks with specific bits turned off for filtering or other bitwise manipulations.

This operation is a fundamental building block for effectively managing sets and states using bitmasking techniques in various algorithms and data structures. Understanding how it works is crucial for manipulating bit-level data precisely.


Let's explore the bitmask operation for **toggling a bit** at a specific position within an integer. Toggling a bit means flipping its value: if it's `0`, it becomes `1`, and if it's `1`, it becomes `0`. This is a useful operation when you need to switch the state of a flag or the presence of an element in a set represented by a bitmask.

**Core Theory: Bitmasks and Toggling Bits**

In bitmasking:

- An integer (`mask`) holds information as a sequence of bits.
- Each bit position corresponds to a specific element or flag.
- Toggling the `i`-th bit means inverting its current value.

**The Toggle Bit Operation: `mask ^ (1 << i)`**

To toggle the bit at the `i`-th position (0-indexed from the right) in a `mask`, we use the bitwise XOR operator (`^`) in combination with a left shift operation:

```
mask ^ (1 << i)
```

Let's break down this operation and the underlying theory:

1.  **Left Shift Operator (`1 << i`)**:

    - As we've discussed before, `(1 << i)` creates an integer with a `1` only at the `i`-th bit position and `0` elsewhere. This mask specifically targets the bit we want to toggle.

    - **Example:** To toggle the 1st bit (index `i = 1`), `(1 << 1)` results in binary `0001 << 1 = 0010` (decimal `2`).

2.  **Bitwise XOR Operator (`^`)**:

    - The bitwise XOR (exclusive OR) operator (`^`) compares the corresponding bits of two operands. The resulting bit is `1` if the corresponding bits are _different_, and `0` if they are the _same_.

    - **Purpose in Toggling a Bit:**

      - If the `i`-th bit in the original `mask` was `0`, XORing it with the `1` at the `i`-th position from `(1 << i)` will result in `1` (0 ^ 1 = 1).
      - If the `i`-th bit in the original `mask` was `1`, XORing it with the `1` at the `i`-th position from `(1 << i)` will result in `0` (1 ^ 1 = 0).
      - All other bit positions in the original `mask` will remain unchanged because they are being XORed with `0` in the `(1 << i)` mask (`bit ^ 0` results in `bit`).

    - **Example:**
      Let our original `mask` be `5` (binary `0101`), and we want to toggle the 1st bit (`i = 1`). We calculated `(1 << 1)` as `2` (binary `0010`). Now we perform `mask ^ (1 << 1)`:

      ```
        0101 (mask = 5)
      ^ 0010 (1 << 1 = 2)
      ------
        0111 (decimal 7)
      ```

      The resulting mask is `7` (binary `0111`). The 1st bit, which was originally `0`, has been toggled to `1`. The other bits (0th and 2nd) remained unchanged.

      Now, let's toggle the 1st bit again on the new mask `7` (binary `0111`):

      ```
        0111 (mask = 7)
      ^ 0010 (1 << 1 = 2)
      ------
        0101 (decimal 5)
      ```

      The 1st bit, which was now `1`, has been toggled back to `0`, returning the mask to its original value.

**In Summary:**

The operation `mask ^ (1 << i)` reliably flips the value of the `i`-th bit in the `mask` without affecting any other bits. It leverages the property of the XOR operator where XORing with `1` inverts a bit, and XORing with `0` leaves a bit unchanged.

**Use Cases:**

- **Toggling Elements in a Set:** When using a bitmask to represent a set, toggling the `i`-th bit effectively adds the `i`-th element if it's not present, and removes it if it is present.
- **Flipping Flags:** In situations where individual bits represent boolean flags, toggling a bit switches the state of a specific flag (True to False, or False to True).
- **Implementing Switches or States:** In algorithms or data structures where you need to easily flip between two states represented by a bit.

The toggle bit operation is a concise and efficient way to modify the state of individual bits within a bitmask, making it a valuable tool for various bit manipulation tasks.

This is a fundamental operation in bit manipulation, allowing you to query the state of individual flags or the presence of elements in a set represented by a bitmask.

**Core Theory: Bitmasks and Checking Set Bits**

As we know, in bitmasking:

- An integer (`mask`) stores information as a sequence of bits.
- Each bit position corresponds to a specific element or flag.
- A bit being "set" means its value is `1`, indicating presence or a `True` state.

**The Check if Bit is Set Operation: `(mask >> i) & 1`**

To check if the bit at the `i`-th position (0-indexed from the right) in a `mask` is set to `1`, we use the bitwise right shift operator (`>>`) followed by the bitwise AND operator (`&`):

```
(mask >> i) & 1
```

Let's break down this operation and the underlying theory:

1.  **Right Shift Operator (`>> i`)**:

    - The right shift operator (`>>`) shifts the bits of the `mask` to the right by `i` positions.
    - As discussed previously, this effectively moves the bit at the `i`-th position to the **0-th position** (the least significant bit).

    - **Purpose in Checking a Set Bit:** By shifting, we bring the bit we want to inspect to the very end, making it easy to isolate.

    - **Example:**
      Let `mask` be `13` (binary `1101`), representing a set where elements at indices 0, 2, and 3 are present. If we want to check if the bit at index `3` is set (`i = 3`):
      `mask >> i` becomes `13 >> 3`, which is binary `1101 >> 3 = 0001` (decimal `1`). The bit originally at the 3rd position is now at the 0th position.

2.  **Bitwise AND Operator (`& 1`)**:

    - The bitwise AND operator (`&`) compares the corresponding bits of two operands. The result is `1` only if both corresponding bits are `1`; otherwise, the result is `0`.
    - We are performing a bitwise AND of the result of the right shift (`mask >> i`) with the integer `1` (binary `...0001`).

    - **Purpose in Checking a Set Bit:** By ANDing with `1`, we isolate the value of the least significant bit of the right-shifted `mask`. If this bit is `1`, it means the `i`-th bit in the original `mask` was also `1` (set). If this bit is `0`, the `i`-th bit in the original `mask` was `0` (not set).

    - **Continuing the Example:**
      After the right shift, we had `0001` (decimal `1`). Now we perform `0001 & 0001`:

      ```
        0001
      & 0001
      ------
        0001
      ```

      The result is `1`, indicating that the bit at index `3` was indeed set in the original `mask`.

      Now, let's check if the bit at index `1` is set in the same `mask` (`1101`).
      `mask >> 1` becomes `13 >> 1`, which is binary `1101 >> 1 = 0110` (decimal `6`).
      Then, `0110 & 0001`:

      ```
        0110
      & 0001
      ------
        0000
      ```

      The result is `0`, indicating that the bit at index `1` was not set in the original `mask`.

**Return Value and Interpretation:**

The expression `(mask >> i) & 1` will evaluate to:

- `1` (or a True-like value in boolean contexts) if the `i`-th bit of `mask` is `1` (the bit is set).
- `0` (or a False-like value in boolean contexts) if the `i`-th bit of `mask` is `0` (the bit is not set).

**Use Cases:**

- **Checking Presence in a Set:** When using a bitmask to represent a set, this operation determines if a specific element (corresponding to the `i`-th bit) is present in the set.
- **Querying Flags:** If individual bits represent boolean flags, this operation checks the current state of a particular flag (whether it's `True` or `False`).
- **Conditional Logic Based on Bits:** Algorithms often use this check to make decisions based on the state of specific bits within a bitmask.

This simple yet powerful operation is fundamental to working with bitmasks and allows you to efficiently inspect the individual components of the information encoded within an integer.


Alright, let's dive deep into the concept of **counting set bits** (also known as counting the number of '1's) in the binary representation of an integer. This is a fundamental operation in bit manipulation with various applications in computer science.

**Core Theory: What are Set Bits?**

The binary representation of an integer consists of a sequence of 0s and 1s. A "set bit" is simply a bit that has a value of `1`. Counting set bits means determining the total number of `1`s present in the binary form of a given integer.

**Why Count Set Bits?**

Counting set bits is useful in several scenarios:

- **Hamming Weight:** The number of set bits is also known as the Hamming weight or population count of an integer.
- **Error Detection and Correction:** In coding theory, the Hamming weight is used to determine the Hamming distance between codewords.
- **Cryptography:** Some cryptographic algorithms rely on the number of set bits.
- **Combinatorics and Subsets:** When using bitmasks to represent subsets, the number of set bits directly corresponds to the size of the subset.
- **Performance Optimization:** In some cases, counting set bits can help optimize certain computations.
- **Low-Level Programming:** Understanding the bit pattern of data is crucial in embedded systems and system programming.

**Common Algorithms and Techniques for Counting Set Bits:**

Here are several approaches to count set bits, ranging from straightforward to more optimized:

1.  **Iterative Checking (Simple Approach):**

    - **Theory:** Iterate through each bit of the integer and check if it's a `1`.
    - **Algorithm:**

      1.  Initialize a counter to 0.
      2.  While the integer is greater than 0:
          a. Check the least significant bit using the bitwise AND operator (`n & 1`). If it's 1, increment the counter.
          b. Right shift the integer by 1 (`n >>= 1`) to examine the next bit.
      3.  Return the counter.

    - **Code (Python):**

      ```python
      def count_set_bits_iterative(n):
          count = 0
          while n > 0:
              if n & 1:
                  count += 1
              n >>= 1
          return count
      ```

    - **Time Complexity:** O(log n), where n is the integer. In the worst case, you might need to examine all the bits of the integer. For a k-bit integer, it's O(k).

2.  **Brian Kernighan's Algorithm (Optimized):**

    - **Theory:** This algorithm is more efficient because it clears the rightmost set bit in each iteration. The number of iterations is equal to the number of set bits.
    - **Algorithm:**

      1.  Initialize a counter to 0.
      2.  While the integer is greater than 0:
          a. Perform the bitwise AND operation `n &= (n - 1)`. This operation clears the rightmost set bit of `n`.
          b. Increment the counter.
      3.  Return the counter.

    - **Code (Python):**

      ```python
      def count_set_bits_kernighan(n):
          count = 0
          while n > 0:
              n &= (n - 1)
              count += 1
          return count
      ```

    - **Time Complexity:** O(number of set bits). In the best case (n=0), it's O(0). In the worst case (all bits are set), for a k-bit integer, it's O(k). This is often better than the iterative approach when the number of set bits is small compared to the total number of bits.

3.  **Lookup Table (Space-Time Tradeoff):**

    - **Theory:** Precompute the number of set bits for all possible values within a small range (e.g., for all 8-bit numbers). Then, for a larger integer, process it in chunks of that size and use the lookup table to find the count for each chunk.
    - **Algorithm (for 32-bit integer using an 8-bit lookup table):**

      1.  Create a lookup table `lookup` of size 256, where `lookup[i]` stores the number of set bits in the 8-bit integer `i`.
      2.  For a given 32-bit integer `n`:
          a. Extract the first 8 bits: `n & 0xFF`. Use this as an index into `lookup` and add the result to a counter.
          b. Right shift `n` by 8 bits: `n >>= 8`.
          c. Repeat steps a and b three more times to process all 4 bytes (32 bits).
      3.  Return the counter.

    - **Code (Python):**

      ```python
      lookup = [bin(i).count('1') for i in range(256)]

      def count_set_bits_lookup(n):
          count = 0
          count += lookup[n & 0xFF]
          count += lookup[(n >> 8) & 0xFF]
          count += lookup[(n >> 16) & 0xFF]
          count += lookup[(n >> 24) & 0xFF]
          return count
      ```

    - **Time Complexity:** O(number of chunks), which is constant (O(1)) for a fixed-size integer (e.g., 32 or 64 bits). The precomputation of the lookup table takes O(2^k), where k is the size of the chunk (e.g., O(256) for 8-bit chunks), which is also constant.
    - **Space Complexity:** O(2^k) for the lookup table.

4.  **Pairwise Addition (Digit DP Inspired):**

    - **Theory:** This approach uses a series of bitwise operations to sum the counts of set bits in adjacent groups of bits.
    - **Algorithm (for a 32-bit integer):**

      1.  Count set bits in each group of 1 bit (which is just the bit itself).
      2.  Sum the counts of adjacent pairs of 1-bit groups into 2-bit groups.
      3.  Sum the counts of adjacent pairs of 2-bit groups into 4-bit groups.
      4.  Continue this process until you have the total count for the entire 32 bits.

    - **Code (Python):**

      ```python
      def count_set_bits_pairwise(n):
          n = (n & 0x55555555) + (n >> 1 & 0x55555555)
          n = (n & 0x33333333) + (n >> 2 & 0x33333333)
          n = (n & 0x0F0F0F0F) + (n >> 4 & 0x0F0F0F0F)
          n = (n & 0x00FF00FF) + (n >> 8 & 0x00FF00FF)
          n = (n & 0x0000FFFF) + (n >> 16 & 0x0000FFFF)
          return n
      ```

    - **Time Complexity:** O(number of bit groups), which is constant (O(1)) for a fixed-size integer.
    - **Space Complexity:** O(1).

5.  **Built-in Functions (Language Dependent):**

    - Many programming languages provide built-in functions or intrinsic instructions that are highly optimized for counting set bits.
    - **Example (Python):** `bin(n).count('1')` (though this involves string conversion, which might not be the most efficient). A more direct approach might be available in lower-level libraries or through specific CPU instructions.
    - **Example (C++):** `__builtin_popcount(n)` (GCC and Clang), `std::bitset<...>(n).count()` (from `<bitset>`).

    - **Time Complexity:** Typically O(1) or very close to it, as these are often implemented using hardware-level optimizations.
    - **Space Complexity:** O(1).

**Choosing the Right Approach:**

- For simplicity and understanding, the **iterative checking** method is a good starting point.
- **Brian Kernighan's algorithm** is often a good balance of efficiency and relatively simple implementation.
- If performance is critical and you are dealing with fixed-size integers, the **lookup table** or **pairwise addition** methods can offer constant time complexity.
- Leveraging **built-in functions** is generally the most efficient approach if available in your programming language.

Understanding the theory behind each method allows you to choose the most appropriate one for your specific needs and constraints. In competitive programming and performance-critical applications, optimized techniques like Brian Kernighan's or built-in functions are often preferred.
