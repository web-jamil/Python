**comparator কিভাবে কাজ করে (ইনভার্টিং এবং নন-ইনভার্টিং কেস)**

নিশ্চয়ই! Comparator (কমপ্যারেটর) একটি খুবই গুরুত্বপূর্ণ ইলেকট্রনিক্স উপাদান। নিচে বাংলায় ইনভার্টিং এবং নন-ইনভার্টিং কমপ্যারেটর কীভাবে কাজ করে তা সহজভাবে ব্যাখ্যা করা হলো।

---

### **১. কমপ্যারেটর কি?**

**কমপ্যারেটর** হল একটি বিশেষ ধরনের অপ-এম্প (Op-Amp) সার্কিট, যেটি তার **দইটি ইনপুট টার্মিনালে** ভোল্টেজের তুলনা করে। তুলনার ভিত্তিতে এটি তার **আউটপুটে** একটি **হাই (HIGH)** বা **লো (LOW)** ভোল্টেজ প্রদান করে। সাধারণত:

- **যদি নন-ইনভার্টিং ইনপুট (+) ইনভার্টিং ইনপুট (-) এর চেয়ে বড় হয়**, তাহলে আউটপুট হয় **হাই** (যেমন, +Vcc বা positive saturation)।
- **যদি নন-ইনভার্টিং ইনপুট (+) ইনভার্টিং ইনপুট (-) এর চেয়ে ছোট হয়**, তাহলে আউটপুট হয় **লো** (যেমন, -Vcc বা negative saturation)।

এটির কাজ অনেকটা একটি ডিজিটাল সুইচের মতো, যা অ্যানালগ সিগন্যালকে ডিজিটাল সিগন্যালে (১ বা ০) রূপান্তর করে।

---

### **২. নন-ইনভার্টিং কমপ্যারেটর (Non-Inverting Comparator)**

এই কনফিগারেশনে, আমরা যে ভোল্টেজের সাথে তুলনা করতে চাই (সেটাকে **রেফারেন্স ভোল্টেজ, Vref** বলে) সেটাকে **ইনভার্টিং (-) টার্মিনালে** দেই। আর যে সিগন্যালটি পরিমাপ বা তুলনা করতে চাই (সেটাকে **ইনপুট ভোল্টেজ, Vin** বলে) সেটাকে **নন-ইনভার্টিং (+) টার্মিনালে** দেই।

![Non-Inverting Comparator](https://www.electronics-tutorials.ws/wp-content/uploads/2018/05/opamp-comparator1.gif)

**কাজের পদ্ধতি:**

1. **যখন Vin (ইনপুট) > Vref (রেফারেন্স):**

   - যেহেতু নন-ইনভার্টিং টার্মিনাল (+) এর ভোল্টেজ ইনভার্টিং টার্মিনাল (-) এর চেয়ে বেশি, তাই আউটপুট **পজিটিভ স্যাচুরেশন (Positive Saturation)**-এ চলে যায়। অর্থাৎ, আউটপুট হয় **হাই (HIGH)**।

2. **যখন Vin (ইনপুট) < Vref (রেফারেন্স):**
   - যেহেতু নন-ইনভার্টিং টার্মিনাল (+) এর ভোল্টেজ ইনভার্টিং টার্মিনাল (-) এর চেয়ে কম, তাই আউটপুট **নেগেটিভ স্যাচুরেশন (Negative Saturation)**-এ চলে যায়। অর্থাৎ, আউটপুট হয় **লো (LOW)**।

**সরল বাংলায়:** ইনপুট ভোল্টেজ রেফারেন্স ভোল্টেজকে অতিক্রম (cross) করলেই আউটপুটের অবস্থান (state) পরিবর্তন হয়। ইনপুট বাড়লে আউটপুট হাই হয়।

---

### **৩. ইনভার্টিং কমপ্যারেটর (Inverting Comparator)**

এই কনফিগারেশনে, **রেফারেন্স ভোল্টেজ (Vref)**-কে **নন-ইনভার্টিং (+) টার্মিনালে** দেই। আর **ইনপুট ভোল্টেজ (Vin)**-কে **ইনভার্টিং (-) টার্মিনালে** দেই।

![Inverting Comparator](https://www.electronics-tutorials.ws/wp-content/uploads/2018/05/opamp-comparator2.gif)

**কাজের পদ্ধতি:**

1. **যখন Vin (ইনপুট) < Vref (রেফারেন্স):**

   - যেহেতু নন-ইনভার্টিং টার্মিনাল (+) এর ভোল্টেজ এখন ইনভার্টিং টার্মিনাল (-) এর চেয়ে বেশি, তাই আউটপুট **পজিটিভ স্যাচুরেশন (Positive Saturation)**-এ চলে যায়। অর্থাৎ, আউটপুট হয় **হাই (HIGH)**।

2. **যখন Vin (ইনপুট) > Vref (রেফারেন্স):**
   - যেহেতু নন-ইনভার্টিং টার্মিনাল (+) এর ভোল্টেজ এখন ইনভার্টিং টার্মিনাল (-) এর চেয়ে কম, তাই আউটপুট **নেগেটিভ স্যাচুরেশন (Negative Saturation)**-এ চলে যায়। অর্থাৎ, আউটপুট হয় **লো (LOW)**।

**সরল বাংলায়:** নন-ইনভার্টিং কমপ্যারেটরের ঠিক উল্টো। ইনপুট ভোল্টেজ রেফারেন্স ভোল্টেজকে অতিক্রম করলে আউটপুটের অবস্থান পরিবর্তন হয়, কিন্তু এবার ইনপুট বাড়লে আউটপুট _লো_ হয়। এটাই হল **ইনভার্টিং** হওয়ার মূল অর্থ – আউটপুটের প্রতিক্রিয়াটি ইনপুটের বিপরীত (inverted)।

---

### **৪. সারাংশ (সাইড বাই সাইড কমপ্যারিজন)**

| বৈশিষ্ট্য                 | **নন-ইনভার্টিং কমপ্যারেটর** | **ইনভার্টিং কমপ্যারেটর**    |
| :------------------------ | :-------------------------- | :-------------------------- |
| **রেফারেন্স (Vref)**      | ইনভার্টিং (-) টার্মিনালে    | নন-ইনভার্টিং (+) টার্মিনালে |
| **ইনপুট (Vin)**           | নন-ইনভার্টিং (+) টার্মিনালে | ইনভার্টিং (-) টার্মিনালে    |
| **আউটপুট যখন Vin > Vref** | **হাই (HIGH)**              | **লো (LOW)**                |
| **আউটপুট যখন Vin < Vref** | **লো (LOW)**                | **হাই (HIGH)**              |
| **সরল অর্থ**              | ইনপুট বড় হলে আউটপুট হাই    | ইনপুট বড় হলে আউটপুট লো     |

### **৫. ব্যবহারিক উদাহরণ**

ধরুন, আপনার একটি সার্কিট আছে যা একটি ব্যাটারির ভোল্টেজ মনিটর করবে। ব্যাটারির ভোল্টেজ ১০V-এর নিচে নামলেই যেন একটি LED জ্বলে।

- **রেফারেন্স ভোল্টেজ (Vref):** ১০V (একটি জেনার ডায়োড বা ভোল্টেজ ডিভাইডার দিয়ে বানানো)।
- **ইনপুট ভোল্টেজ (Vin):** ব্যাটারির ভোল্টেজ।

আপনি চান ব্যাটারির ভোল্টেজ (Vin) ১০V (Vref) এর নিচে নামলেই LED জ্বলুক।

- **কোন কমপ্যারেটর ব্যবহার করবেন?** **ইনভার্টিং কমপ্যারেটর**।
  - **কারণ:** যখন Vin (ব্যাটারি) < Vref (১০V), তখন আমরা চাই আউটপুট _হাই_ হোক যাতে LED জ্বলে। ইনভার্টিং কমপ্যারেটর ঠিক এই কাজটি করে (উপরের টেবিল দেখুন)।

আশা করি, ইনভার্টিং এবং নন-ইনভার্টিং কমপ্যারেটরের কাজের পদ্ধতি এখন আপনার কাছে পরিষ্কার হয়েছে
