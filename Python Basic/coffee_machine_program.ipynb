from abc import ABC, abstractmethod

# Constants for coin values
COIN_VALUES = {
    "quarters": 0.25,

    "nickels": 0.05,
    "pennies": 0.01

# Abstract Base Class for Coffee
class Coffee(ABC):
    @abstractmethod
    def cost(self):
        pass

    @abstractmethod
    def ingredients(self):
        pass

# Different types of coffee
class Espresso(Coffee):
    def cost(self):
        return 1.5

    def ingredients(self):
        return {"water": 50, "coffee": 18}

class Latte(Coffee):
    def cost(self):
        return 2.5

    def ingredients(self):
        return {"water": 200, "milk": 150, "coffee": 24}

class Cappuccino(Coffee):
    def cost(self):
        return 3.0

    def ingredients(self):
        return {"water": 250, "milk": 100, "coffee": 24}

# Singleton Coffee Machine
class CoffeeMachine:
    _instance = None  # Singleton instance

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(CoffeeMachine, cls).__new__(cls)
            cls._instance._initialize()
        return cls._instance

    def _initialize(self):
        self.resources = {
            "water": 500,
            "milk": 300,
            "coffee": 100,
            "money": 0.0
        }
        self.menu = {
            "espresso": Espresso(),
            "latte": Latte(),
            "cappuccino": Cappuccino()
        }

    def print_report(self):
        """Displays the current machine resources."""
        print("\n🔹 Coffee Machine Status 🔹")
        for item, amount in self.resources.items():
            unit = "ml" if item in ["water", "milk"] else "g" if item == "coffee" else "$"
            print(f"{item.capitalize()}: {amount}{unit}")
        print()

    def is_resources_sufficient(self, coffee):
        """Checks if there are enough resources to make the selected coffee."""
        for ingredient, required_amount in coffee.ingredients().items():
            if self.resources[ingredient] < required_amount:
                print(f"❌ Sorry, not enough {ingredient}.")
                return False
        return True

    def process_payment(self, coffee):
        """Handles payment and returns True if successful."""
        print(f"💰 {coffee.__class__.__name__} costs ${coffee.cost():.2f}")
        print("Insert coins: ")
        total_inserted = sum(int(input(f"{coin.capitalize()}: ")) * value for coin, value in COIN_VALUES.items())
        total_inserted = round(total_inserted, 2)

        if total_inserted < coffee.cost():
            print("❌ Not enough money. Refunding...")
            return False
        elif total_inserted > coffee.cost():
            change = round(total_inserted - coffee.cost(), 2)
            print(f"💵 Here is your ${change} in change.")

        self.resources["money"] += coffee.cost()
        return True

    def make_coffee(self, coffee):
        """Prepares the coffee by deducting ingredients."""
        for ingredient, required_amount in coffee.ingredients().items():
            self.resources[ingredient] -= required_amount
        print(f"☕ Here is your {coffee.__class__.__name__}. Enjoy! 🎉\n")

    def run(self):
        """Main loop for the coffee machine."""
        while True:
            choice = input("🔸 What would you like? (espresso/latte/cappuccino/report/off): ").lower()

            if choice == "off":
                print("⚡ Shutting down the coffee machine...")
                break
            elif choice == "report":
                self.print_report()
            elif choice in self.menu:
                coffee = self.menu[choice]
                if self.is_resources_sufficient(coffee) and self.process_payment(coffee):
                    self.make_coffee(coffee)
            else:
                print("❓ Invalid choice. Please choose again.")

# Run the Coffee Machine
if __name__ == "__main__":

````python
# Coffee Machine Resources
resources = {
    "water": 300,  # in ml
    "milk": 200,   # in ml
    "coffee": 100,  # in g
    "money": 0.0    # in $
}

# Menu with drink ingredients and cost
MENU = {
    "espresso": {
        "ingredients": {
            "water": 50,
            "coffee": 18,
        },
        "cost": 1.5,
    },
    "latte": {
        "ingredients": {
            "water": 200,
            "milk": 150,
            "coffee": 24,
        },
        "cost": 2.5,
    },
    "cappuccino": {
        "ingredients": {
            "water": 250,
            "milk": 100,
            "coffee": 24,
        },
        "cost": 3.0,
    }
}

# Coin and dollar values
MONEY_VALUES = {
    "dollars": 1.00,
    "quarters": 0.25,
    "dimes": 0.10,
    "nickles": 0.05,
    "pennies": 0.01,
}


def print_report():
    """Prints the current resources and money in the coffee machine."""
    print(f"Water: {resources['water']}ml")
    print(f"Milk: {resources['milk']}ml")
    print(f"Coffee: {resources['coffee']}g")
    print(f"Money: ${resources['money']:.2f}")


def check_resources(drink):
    """Checks if there are enough resources to make the selected drink."""
    for ingredient, amount in MENU[drink]["ingredients"].items():
        if resources[ingredient] < amount:
            print(f"Sorry, there is not enough {ingredient}.")
            return False
    return True


def get_valid_input(prompt):
    """Gets a valid integer input from the user, handling errors."""
    while True:
        try:
            value = int(input(prompt))
            if value < 0:
                print("Please enter a positive number.")
            else:
                return value
        except ValueError:
            print("Invalid input. Please enter a number.")


def process_payment():
    """Calculates the total value of money inserted by the user (dollars + coins)."""
    total = 0
    print("Please insert money:")
    for money, value in MONEY_VALUES.items():
        count = get_valid_input(f"How many {money}?: ")
        total += count * value
    return total


def check_transaction(drink, money_inserted):
    """Checks if the user has inserted enough money to purchase the drink."""
    cost = MENU[drink]["cost"]
    if money_inserted < cost:
        print("Sorry, that's not enough money. Money refunded.")
        return False
    else:
        change = money_inserted - cost
        if change > 0:
            print(f"Here is ${change:.2f} in change.")
        resources["money"] += cost
        return True


def make_coffee(drink):
    """Deducts the required ingredients from resources and serves the drink."""
    for ingredient, amount in MENU[drink]["ingredients"].items():
        resources[ingredient] -= amount
    print(f"Here is your {drink}. Enjoy!")


def get_valid_choice():
    """Gets a valid drink choice from the user, handling typos."""
    while True:
        choice = input("What would you like? (espresso/latte/cappuccino): ").strip().lower()
        if choice in MENU or choice in ["off", "report"]:
            return choice
        print("Invalid choice. Please enter 'espresso', 'latte', or 'cappuccino'.")


def coffee_machine():
    """Main function to run the coffee machine."""
    while True:
        # Prompt user with validation
        choice = get_valid_choice()

        # Turn off the machine
        if choice == "off":
            print("Turning off the coffee machine. Goodbye!")
            break

        # Print report
        elif choice == "report":
            print_report()

        # Make a drink
        elif choice in MENU:
            drink = choice
            if check_resources(drink):
                money_inserted = process_payment()
                if check_transaction(drink, money_inserted):
                    make_coffee(drink)


# Run the coffee machine
coffee_machine()
````

````python
import time

class CoffeeMachine:
    def __init__(self):
        self.menu = {
            "espresso": {"ingredients": {"water": 50, "coffee": 18}, "cost": 1.5},
            "latte": {"ingredients": {"water": 200, "milk": 150, "coffee": 24}, "cost": 2.5},
            "cappuccino": {"ingredients": {"water": 250, "milk": 100, "coffee": 24}, "cost": 3.0},
        }
        self.resources = {"water": 300, "milk": 200, "coffee": 100}
        self.profit = 0
        self.is_on = True
        self.last_activity = time.time()

    def report(self):
        print("\nMachine Report:")
        for key, value in self.resources.items():
            unit = "ml" if key != "coffee" else "g"
            print(f"{key.capitalize()}: {value}{unit}")
        print(f"Profit: ${self.profit}\n")
    
    def is_resource_sufficient(self, ingredients):
        for item, amount in ingredients.items():
            if amount > self.resources.get(item, 0):
                print(f"Sorry, not enough {item}.")
                return False
        return True

    def process_payment(self, cost):
        print("Payment Options: [1] Cash [2] Card [3] Mobile Wallet")
        choice = input("Choose payment method: ")
        
        if choice == "1":
            print("Insert coins:")
            total = (int(input("Quarters: ")) * 0.25 +
                     int(input("Dimes: ")) * 0.10 +
                     int(input("Nickels: ")) * 0.05 +
                     int(input("Pennies: ")) * 0.01)
        elif choice in ["2", "3"]:
            total = cost  # Assume full payment is made for simplicity
            print("Payment successful via digital method.")
        else:
            print("Invalid payment method. Try again.")
            return False
        
        if total >= cost:
            change = round(total - cost, 2)
            if change > 0:
                print(f"Here is your change: ${change}")
            self.profit += cost
            return True
        else:
            print("Insufficient funds. Money refunded.")
            return False

    def make_coffee(self, name, ingredients):
        for item, amount in ingredients.items():
            self.resources[item] -= amount
        print(f"Enjoy your {name} ☕️!\n")
        self.last_activity = time.time()

    def refill_resources(self):
        for item in self.resources:
            self.resources[item] += int(input(f"Add {item} amount: "))
        print("Resources refilled!\n")

    def auto_shutdown(self):
        if time.time() - self.last_activity > 60:
            print("Energy-saving mode activated. Machine shutting down...")
            self.is_on = False

    def start(self):
        while self.is_on:
            self.auto_shutdown()
            choice = input("What would you like? (espresso/latte/cappuccino/report/refill/off): ")
            
            if choice == "off":
                print("Shutting down...")
                self.is_on = False
            elif choice == "report":
                self.report()
            elif choice == "refill":
                self.refill_resources()
            elif choice in self.menu:
                drink = self.menu[choice]
                if self.is_resource_sufficient(drink["ingredients"]):
                    if self.process_payment(drink["cost"]):
                        self.make_coffee(choice, drink["ingredients"])
            else:
                print("Invalid selection, please try again.")

if __name__ == "__main__":
    machine = CoffeeMachine()
    machine.start()
````
