## Python Stack Data Structure: All About with Theory and Syntax

### What is a Stack?

A stack is a linear data structure that follows a particular order in which operations are performed. The order is **LIFO (Last In, First Out)** or **FILO (First In, Last Out)**. Imagine a stack of plates: the last plate you put on top is the first one you take off.

**Key Characteristics:**

- **LIFO/FILO:** The element that was inserted most recently is the first one to be removed.
- **Linear:** Elements are arranged in a sequential manner.
- **Abstract Data Type (ADT):** A stack is a conceptual model, defined by its behavior (operations) rather than its specific implementation.

### Core Operations of a Stack

The primary operations associated with a stack are:

1.  **Push:** Adds an element to the top of the stack.
2.  **Pop:** Removes the top element from the stack and returns it. If the stack is empty, it typically raises an error or returns a special value (e.g., `None`).
3.  **Peek (or Top):** Returns the top element of the stack without removing it. If the stack is empty, it typically raises an error.
4.  **isEmpty (or Is_Empty):** Checks if the stack is empty. Returns `True` if empty, `False` otherwise.
5.  **Size:** Returns the number of elements in the stack.

### Where are Stacks Used? (Applications)

Stacks are surprisingly common in computer science:

- **Function Call Management:** Compilers and interpreters use stacks to manage function calls. When a function is called, its state (local variables, return address) is "pushed" onto the call stack. When it returns, its state is "popped."
- **Expression Evaluation:** Used in evaluating arithmetic expressions (infix to postfix/prefix conversion, postfix evaluation).
- **Undo/Redo Functionality:** Many applications use a stack to keep track of operations for undo/redo features.
- **Browser History:** Web browsers use a stack to manage the "back" and "forward" functionality.
- **Parentheses/Bracket Matching:** As seen in the "Valid Parentheses" problem, stacks are perfect for checking if parentheses, braces, and brackets are balanced.
- **Depth-First Search (DFS):** In graph and tree traversal algorithms, DFS can be implemented using a stack.
- **Backtracking Algorithms:** Used to keep track of choices made so that they can be "undone" if a path leads to a dead end.

### Python Implementations of a Stack

Python doesn't have a built-in, dedicated `Stack` class like some other languages (e.g., Java's `java.util.Stack`). However, its highly versatile built-in data structures can be easily used to implement a stack effectively.

#### 1\. Using a `list` (Most Common and Recommended for General Use)

Python's list provides `append()` for push and `pop()` for pop operations, making it a natural fit for a stack.

**Theory:**
Elements are added to the end of the list and removed from the end of the list. Since `append()` and `pop()` from the end of a list are $O(1)$ (constant time) operations on average, this is very efficient.

**Syntax and Example:**

```python
# Create an empty stack
my_stack = []

# --- Push Operation ---
# Use append() to add elements to the top (end) of the stack
my_stack.append(10)
my_stack.append(20)
my_stack.append(30)
print(f"After pushes: {my_stack}") # Output: After pushes: [10, 20, 30]

# --- Pop Operation ---
# Use pop() to remove and return the top (last) element
popped_element = my_stack.pop()
print(f"Popped element: {popped_element}") # Output: Popped element: 30
print(f"Stack after pop: {my_stack}")     # Output: Stack after pop: [10, 20]

# --- Peek (Top) Operation ---
# Access the last element without removing it using index [-1]
if my_stack: # Always check if the stack is not empty before peeking
    top_element = my_stack[-1]
    print(f"Top element (peek): {top_element}") # Output: Top element (peek): 20
else:
    print("Stack is empty, cannot peek.")

# --- isEmpty Operation ---
# Check if the list is empty
is_empty = not my_stack
print(f"Is stack empty? {is_empty}") # Output: Is stack empty? False

my_stack.pop() # Pop 20
my_stack.pop() # Pop 10
print(f"Stack after more pops: {my_stack}") # Output: Stack after more pops: []

is_empty = not my_stack
print(f"Is stack empty? {is_empty}") # Output: Is stack empty? True

# Trying to pop from an empty stack will raise an IndexError
try:
    my_stack.pop()
except IndexError:
    print("Cannot pop from an empty stack (IndexError).")

# --- Size Operation ---
# Use len() to get the size of the stack
print(f"Size of stack: {len(my_stack)}") # Output: Size of stack: 0 (after pops)
```

**Pros:**

- Simple and intuitive.
- Built-in, no need for imports.
- Efficient for typical stack operations (average $O(1)$).

**Cons:**

- Appending or popping from the _beginning_ of a list (`insert(0, ...)` or `pop(0)`) is $O(N)$ because it requires shifting all other elements. This is why you must use `append()` and `pop()` from the _end_ for stack behavior.
- Does not enforce stack behavior strictly (can access elements by index, etc., which isn't standard stack behavior).

#### 2\. Using `collections.deque` (Double-Ended Queue)

`deque` (pronounced "deck") is a list-like container with fast appends and pops from both ends. This makes it slightly more efficient and robust for stack operations, especially in multi-threaded contexts, compared to a plain list.

**Theory:**
A deque is implemented as a doubly-linked list of blocks of elements. This structure allows constant time (amortized $O(1)$) operations for adding or removing elements from either end.

**Syntax and Example:**

```python
from collections import deque

# Create an empty stack (deque)
my_stack = deque()

# --- Push Operation ---
# Use append() to add elements to the right (top) of the deque
my_stack.append(100)
my_stack.append(200)
my_stack.append(300)
print(f"After pushes: {my_stack}") # Output: After pushes: deque([100, 200, 300])

# --- Pop Operation ---
# Use pop() to remove and return the element from the right (top)
popped_element = my_stack.pop()
print(f"Popped element: {popped_element}") # Output: Popped element: 300
print(f"Stack after pop: {my_stack}")     # Output: Stack after pop: deque([100, 200])

# --- Peek (Top) Operation ---
# Access the rightmost element using index [-1]
if my_stack:
    top_element = my_stack[-1]
    print(f"Top element (peek): {top_element}") # Output: Top element (peek): 200
else:
    print("Stack is empty, cannot peek.")

# --- isEmpty Operation ---
is_empty = not my_stack
print(f"Is stack empty? {is_empty}") # Output: Is stack empty? False

my_stack.pop()
my_stack.pop()
print(f"Stack after more pops: {my_stack}") # Output: Stack after more pops: deque([])

is_empty = not my_stack
print(f"Is stack empty? {is_empty}") # Output: Is stack empty? True

# Trying to pop from an empty deque will raise an IndexError
try:
    my_stack.pop()
except IndexError:
    print("Cannot pop from an empty stack (IndexError).")

# --- Size Operation ---
print(f"Size of stack: {len(my_stack)}") # Output: Size of stack: 0
```

**Pros:**

- More efficient for both ends ($O(1)$ for `append()` and `pop()`).
- Thread-safe: `deque` operations are atomic, making them suitable for multi-threaded environments.

**Cons:**

- Requires importing `deque` from `collections`.
- Still doesn't strictly enforce stack behavior (can access elements by index or from the other end).

#### 3\. Using `queue.LifoQueue` (Thread-Safe, Meant for Queues but Acts as Stack)

The `queue` module in Python is primarily designed for concurrent programming (handling data exchange between multiple threads). `LifoQueue` specifically implements a LIFO queue, which is exactly a stack.

**Theory:**
`LifoQueue` uses a list internally and adds necessary locking mechanisms to ensure thread safety. This means it's slower than `list` or `deque` for single-threaded use due to the overhead of locks, but essential for concurrent scenarios.

**Syntax and Example:**

```python
from queue import LifoQueue

# Create an empty stack (LifoQueue)
# Max size can be set, 0 means infinite size
my_stack = LifoQueue(maxsize=0)

# --- Put (Push) Operation ---
# Use put() to add elements. It blocks if the queue is full (if maxsize > 0).
my_stack.put(1000)
my_stack.put(2000)
my_stack.put(3000)
print(f"After puts (internal representation): {list(my_stack.queue)}") # Output: After puts (internal representation): [1000, 2000, 3000]

# --- Get (Pop) Operation ---
# Use get() to remove and return the top element. It blocks if the queue is empty.
popped_element = my_stack.get()
print(f"Got element: {popped_element}") # Output: Got element: 3000
print(f"Stack size after get: {my_stack.qsize()}") # Output: Stack size after get: 2

# --- Peek (Top) Operation ---
# LifoQueue does not have a direct peek method.
# You can access it via my_stack.queue[-1] but this bypasses thread-safety.
# For thread-safe peek, you'd usually get() and then put() it back, or check qsize() first.
if not my_stack.empty():
    # This is not strictly a 'peek' without blocking or removing
    # It's an internal access and not recommended for thread-safe operations.
    # For a true peek with LifoQueue, you'd usually implement it as get() then put() back.
    print(f"Top element (internal peek): {my_stack.queue[-1]}")
else:
    print("Stack is empty, cannot peek.")


# --- empty() Operation ---
is_empty = my_stack.empty()
print(f"Is stack empty? {is_empty}") # Output: Is stack empty? False

my_stack.get() # Get 2000
my_stack.get() # Get 1000
print(f"Stack size after more gets: {my_stack.qsize()}") # Output: Stack size after more gets: 0

is_empty = my_stack.empty()
print(f"Is stack empty? {is_empty}") # Output: Is stack empty? True

# Trying to get from an empty stack (blocking call)
# my_stack.get() # This would block indefinitely without a timeout
try:
    my_stack.get(timeout=0.1) # Try to get with a timeout
except Exception as e:
    print(f"Cannot get from an empty stack (timeout/empty exception: {e}).")


# --- qsize() Operation ---
print(f"Size of stack: {my_stack.qsize()}") # Output: Size of stack: 0
```

**Pros:**

- **Thread-safe:** Best choice for multi-threaded applications where multiple threads need to access the same stack.
- Can set a `maxsize`.

**Cons:**

- Slightly slower due to locking mechanisms, not ideal for single-threaded applications where performance is critical.
- Methods are `put()` and `get()` instead of `push()` and `pop()`, which is less intuitive for stack terminology.
- No direct `peek()` method.

### Choosing the Right Implementation

- **For most general-purpose, single-threaded applications (like competitive programming or simple scripts):** Use a **`list`**. It's the simplest, most Pythonic, and usually performs well enough.
- **For single-threaded applications where slight performance gains or thread-safety might be a future concern (or specific library needs it):** Use `collections.deque`.
- **For multi-threaded or concurrent applications:** Use `queue.LifoQueue`.

### Stack in Action: Example (Reverse a String)

```python
def reverse_string_using_stack(s: str) -> str:
    """
    Reverses a string using a stack.
    """
    stack = []
    for char in s:
        stack.append(char) # Push each character onto the stack

    reversed_chars = []
    while stack:
        reversed_chars.append(stack.pop()) # Pop characters to reverse order

    return "".join(reversed_chars)

print(reverse_string_using_stack("hello")) # Output: olleh
print(reverse_string_using_stack("Python")) # Output: nohtyP
print(reverse_string_using_stack("")) # Output:
```

## Python Stack Data Structure: All Methods with Syntax

### I. Using a Python `list` (Most Common Approach)

When using a standard Python `list` as a stack, you leverage its built-in methods.

**1. `append(element)` (Simulates `push`)**

- **Purpose:** Adds an element to the top (end) of the stack.
- **Syntax:** `list_name.append(element)`
- **Time Complexity:** Amortized $O(1)$ (constant time). On rare occasions, it might be $O(N)$ if the underlying array needs to be resized, but this is averaged out over many operations.

<!-- end list -->

```python
my_stack = []
my_stack.append('A') # Stack: ['A']
my_stack.append('B') # Stack: ['A', 'B']
```

**2. `pop()` (Simulates `pop`)**

- **Purpose:** Removes and returns the element from the top (end) of the stack.
- **Syntax:** `list_name.pop()`
- **Return Value:** The element that was removed.
- **Raises:** `IndexError` if the stack is empty.
- **Time Complexity:** $O(1)$ (constant time).

<!-- end list -->

```python
my_stack = ['A', 'B', 'C']
top_element = my_stack.pop() # top_element is 'C', Stack: ['A', 'B']
next_element = my_stack.pop() # next_element is 'B', Stack: ['A']

try:
    empty_stack = []
    empty_stack.pop() # This will raise an IndexError
except IndexError as e:
    print(f"Error: {e}")
```

**3. `list_name[-1]` (Simulates `peek` / `top`)**

- **Purpose:** Returns the element at the top (end) of the stack _without removing it_.
- **Syntax:** `list_name[-1]`
- **Return Value:** The top element.
- **Raises:** `IndexError` if the stack is empty.
- **Time Complexity:** $O(1)$ (constant time).

<!-- end list -->

```python
my_stack = ['X', 'Y', 'Z']
if my_stack: # Always check if not empty before peeking!
    peeked_element = my_stack[-1] # peeked_element is 'Z', Stack remains ['X', 'Y', 'Z']
    print(f"Peeked: {peeked_element}")
else:
    print("Stack is empty, cannot peek.")
```

**4. `not list_name` or `len(list_name) == 0` (Simulates `isEmpty`)**

- **Purpose:** Checks if the stack is empty.
- **Syntax:** `not list_name` (more Pythonic) or `len(list_name) == 0`
- **Return Value:** `True` if empty, `False` otherwise.
- **Time Complexity:** $O(1)$ (constant time).

<!-- end list -->

```python
my_stack = []
is_empty = not my_stack # True
print(f"Is stack empty? {is_empty}")

my_stack.append(1)
is_empty = not my_stack # False
print(f"Is stack empty? {is_empty}")
```

**5. `len(list_name)` (Simulates `size`)**

- **Purpose:** Returns the number of elements currently in the stack.
- **Syntax:** `len(list_name)`
- **Return Value:** An integer representing the number of elements.
- **Time Complexity:** $O(1)$ (constant time).

<!-- end list -->

```python
my_stack = ['P', 'Q']
stack_size = len(my_stack) # stack_size is 2
print(f"Stack size: {stack_size}")
```

### III. Using `queue.LifoQueue` (Thread-Safe Stack)

This is part of Python's `queue` module, designed for inter-thread communication. `LifoQueue` specifically implements a LIFO (Last-In, First-Out) structure, making it a thread-safe stack. It has different method names more aligned with queue terminology.

**1. `put(item, block=True, timeout=None)` (Simulates `push`)**

- **Purpose:** Adds an item to the stack.
- **Syntax:** `lifo_queue_instance.put(item)`
- **`block`:** If `True` (default), blocks until a free slot is available if `maxsize` is set and the queue is full. If `False`, raises `queue.Full` immediately.
- **`timeout`:** If `block` is `True`, waits for `timeout` seconds before raising `queue.Full`.
- **Time Complexity:** $O(1)$ typically, but can vary depending on blocking.

<!-- end list -->

```python
from queue import LifoQueue

my_lifo_stack = LifoQueue(maxsize=3) # Stack with max capacity of 3
my_lifo_stack.put('First')
my_lifo_stack.put('Second')
my_lifo_stack.put('Third')
print(f"Stack size after puts: {my_lifo_stack.qsize()}") # Output: 3

# This would block if maxsize was hit and no item was removed
# my_lifo_stack.put('Fourth', block=False) # Would raise queue.Full if maxsize was 3 and stack full
```

**2. `get(block=True, timeout=None)` (Simulates `pop`)**

- **Purpose:** Removes and returns an item from the top of the stack.
- **Syntax:** `lifo_queue_instance.get()`
- **`block`:** If `True` (default), blocks until an item is available if the queue is empty. If `False`, raises `queue.Empty` immediately.
- **`timeout`:** If `block` is `True`, waits for `timeout` seconds before raising `queue.Empty`.
- **Return Value:** The item that was retrieved.
- **Time Complexity:** $O(1)$ typically, but can vary depending on blocking.

<!-- end list -->

```python
from queue import LifoQueue
my_lifo_stack = LifoQueue()
my_lifo_stack.put('Last')
my_lifo_stack.put('Middle')
my_lifo_stack.put('First')

item1 = my_lifo_stack.get() # item1 is 'First' (LIFO, so 'First' was put in last)
item2 = my_lifo_stack.get() # item2 is 'Middle'
```

**3. `qsize()` (Simulates `size`)**

- **Purpose:** Returns the approximate number of items currently in the stack.
- **Syntax:** `lifo_queue_instance.qsize()`
- **Return Value:** An integer.
- **Time Complexity:** $O(1)$.

<!-- end list -->

```python
from queue import LifoQueue
my_lifo_stack = LifoQueue()
my_lifo_stack.put(1)
print(f"Stack size: {my_lifo_stack.qsize()}") # Output: 1
```

**4. `empty()` (Simulates `isEmpty`)**

- **Purpose:** Returns `True` if the stack is empty, `False` otherwise.
- **Syntax:** `lifo_queue_instance.empty()`
- **Time Complexity:** $O(1)$.

<!-- end list -->

```python
from queue import LifoQueue
my_lifo_stack = LifoQueue()
is_stack_empty = my_lifo_stack.empty() # True
my_lifo_stack.put('item')
is_stack_empty = my_lifo_stack.empty() # False
```

**5. `full()` (Specific to `LifoQueue` with `maxsize`)**

- **Purpose:** Returns `True` if the stack has reached its `maxsize`, `False` otherwise.
- **Syntax:** `lifo_queue_instance.full()`
- **Time Complexity:** $O(1)$.

<!-- end list -->

```python
from queue import LifoQueue
my_lifo_stack = LifoQueue(maxsize=1)
my_lifo_stack.put('A')
is_stack_full = my_lifo_stack.full() # True
```

**Note on Peek with `LifoQueue`:**
`LifoQueue` does _not_ have a standard `peek()` method. This is because its primary purpose is thread-safe data transfer, where accessing elements without removing them can introduce race conditions or make the state difficult to reason about concurrently. If you absolutely need a peek for debugging or specific single-threaded scenarios, you might access `my_lifo_stack.queue[-1]`, but be aware that this bypasses the thread-safety mechanisms and is generally discouraged in concurrent programming. For typical algorithm problems, `list` or `deque` are better if you need `peek`.


````python
class Stack:
    def __init__(self):
        """
        Initializes an empty stack using a Python list.
        """
        self._items = []

    def push(self, item):
        """
        Adds an item to the top of the stack.
        :param item: The item to be added to the stack.
        """
        self._items.append(item)

    def pop(self):
        """
        Removes and returns the item from the top of the stack.
        Raises an IndexError if the stack is empty.
        :return: The item removed from the top of the stack.
        """
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._items.pop()

    def peek(self):
        """
        Returns the item at the top of the stack without removing it.
        Raises an IndexError if the stack is empty.
        :return: The item at the top of the stack.
        """
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self._items[-1]

    def is_empty(self):
        """
        Checks if the stack is empty.
        :return: True if the stack is empty, False otherwise.
        """
        return len(self._items) == 0

    def size(self):
        """
        Returns the number of items in the stack.
        :return: The number of items in the stack.
        """
        return len(self._items)

    def __str__(self):
        """
        Returns a string representation of the stack.
        """
        return f"Stack: {self._items}"

# Example Usage:
if __name__ == "__main__":
    my_stack = Stack()

    print(f"Is stack empty? {my_stack.is_empty()}")
    print(f"Stack size: {my_stack.size()}")

    my_stack.push(10)
    my_stack.push(20)
    my_stack.push(30)
    print(my_stack)

    print(f"Is stack empty? {my_stack.is_empty()}")
    print(f"Stack size: {my_stack.size()}")

    print(f"Top element (peek): {my_stack.peek()}")
    print(my_stack)

    popped_item = my_stack.pop()
    print(f"Popped item: {popped_item}")
    print(my_stack)

    my_stack.push(40)
    print(my_stack)

    print(f"Stack size: {my_stack.size()}")

    # Pop all elements
    while not my_stack.is_empty():
        print(f"Popped: {my_stack.pop()}")
    print(my_stack)
    print(f"Is stack empty? {my_stack.is_empty()}")

    # Trying to pop from an empty stack (will raise an error)
    try:
        my_stack.pop()
    except IndexError as e:
        print(f"Error: {e}")

    # Trying to peek from an empty stack (will raise an error)
    try:
        my_stack.peek()
    except IndexError as e:
        print(f"Error: {e}")
````

````python
from collections import deque

class Stack:
    def __init__(self):
        """
        Initializes an empty stack using collections.deque.
        A deque (double-ended queue) is optimized for appends and pops
        from both ends, making it highly efficient for stack operations.
        """
        self._items = deque()

    def push(self, element):
        """
        Adds an element to the top of the stack.
        This operation uses deque.append(), which is O(1).

        Syntax: deque_object.append(element)
        Time Complexity: O(1) (constant time).
        """
        self._items.append(element)

    def pop(self):
        """
        Removes and returns the element from the top of the stack.
        This operation uses deque.pop(), which is O(1).

        Syntax: deque_object.pop()
        Time Complexity: O(1) (constant time).
        Raises: IndexError if the stack is empty.
        """
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._items.pop()

    def peek(self):
        """
        Returns the element at the top of the stack without removing it.
        This operation uses deque[-1], which is O(1).

        Syntax: deque_object[-1]
        Time Complexity: O(1) (constant time).
        Raises: IndexError if the stack is empty.
        """
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self._items[-1]

    def is_empty(self):
        """
        Checks if the stack is empty.
        This operation uses 'not deque_object' or 'len(deque_object) == 0', both O(1).

        Syntax Option 1: not deque_object
        Syntax Option 2: len(deque_object) == 0
        Time Complexity: O(1) (constant time).
        """
        return not self._items
        # return len(self._items) == 0 # Also valid and O(1)

    def size(self):
        """
        Returns the number of elements in the stack.
        This operation uses len(deque_object), which is O(1).

        Syntax: len(deque_object)
        Time Complexity: O(1) (constant time).
        """
        return len(self._items)

    def __str__(self):
        """
        Returns a string representation of the stack.
        """
        return f"Stack (using deque): {list(self._items)}" # Convert to list for cleaner print

# Example Usage:
if __name__ == "__main__":
    my_stack = Stack()

    print("--- Initial State ---")
    print(f"Stack: {my_stack}")
    print(f"Is empty? {my_stack.is_empty()}")
    print(f"Size: {my_stack.size()}")

    print("\n--- Push Operations ---")
    my_stack.push("Task A")
    my_stack.push(101)
    my_stack.push({'id': 1, 'status': 'new'})
    print(f"Stack after pushes: {my_stack}")
    print(f"Is empty? {my_stack.is_empty()}")
    print(f"Size: {my_stack.size()}")

    print("\n--- Peek Operation ---")
    try:
        top_element = my_stack.peek()
        print(f"Top element: {top_element}")
    except IndexError as e:
        print(f"Error peeking: {e}")
    print(f"Stack after peek: {my_stack}") # Stack remains unchanged

    print("\n--- Pop Operations ---")
    popped_item = my_stack.pop()
    print(f"Popped: {popped_item}")
    print(f"Stack after 1st pop: {my_stack}")
    print(f"Size: {my_stack.size()}")

    popped_item = my_stack.pop()
    print(f"Popped: {popped_item}")
    print(f"Stack after 2nd pop: {my_stack}")
    print(f"Size: {my_stack.size()}")

    print("\n--- Emptying and Error Handling ---")
    popped_item = my_stack.pop()
    print(f"Popped: {popped_item}")
    print(f"Stack after 3rd pop: {my_stack}")
    print(f"Is empty? {my_stack.is_empty()}")
    print(f"Size: {my_stack.size()}")

    # Attempt to pop from an empty stack
    try:
        my_stack.pop()
    except IndexError as e:
        print(f"Caught error when popping from empty stack: {e}")

    # Attempt to peek from an empty stack
    try:
        my_stack.peek()
    except IndexError as e:
        print(f"Caught error when peeking from empty stack: {e}")

    print("\n--- Direct usage of collections.deque for stack behavior ---")
    # You can also use deque directly without wrapping it in a class if preferred
    direct_stack = deque()
    direct_stack.append("item1") # push
    direct_stack.append("item2") # push
    print(f"Direct deque stack: {direct_stack}")
    print(f"Top element: {direct_stack[-1]}") # peek
    print(f"Popped: {direct_stack.pop()}") # pop
    print(f"Direct deque stack after pop: {direct_stack}")
    print(f"Is empty: {not direct_stack}")
````

````python
from collections import deque

# 1. Creating a deque
# An empty deque
dq_empty = deque()
print(f"Empty deque: {dq_empty}")

# Initializing with an iterable (e.g., a list)
dq_initial = deque([1, 2, 3, 4, 5])
print(f"Deque initialized with a list: {dq_initial}")

# Creating a deque with a maximum length (maxlen)
# When maxlen is specified, new elements added will discard elements from the opposite end.
dq_fixed_size = deque(maxlen=3)
print(f"Fixed-size deque (maxlen=3): {dq_fixed_size}")

# 2. Adding Elements (O(1) time complexity)

# append(element): Adds an element to the right end (tail)
dq_empty.append('A')
dq_empty.append('B')
print(f"\nAfter append('A') and 'B': {dq_empty}")

# appendleft(element): Adds an element to the left end (head)
dq_empty.appendleft('Z')
print(f"After appendleft('Z'): {dq_empty}")

# extend(iterable): Extends the deque by appending elements from an iterable to the right
dq_initial.extend([6, 7, 8])
print(f"After extend([6, 7, 8]): {dq_initial}")

# extendleft(iterable): Extends the deque by appending elements from an iterable to the left.
# Note: Elements are added in reversed order from the iterable when using extendleft.
dq_initial.extendleft(['Y', 'X'])
print(f"After extendleft(['Y', 'X']): {dq_initial}")

# Demonstrating maxlen with push operations
dq_fixed_size.append(1)
dq_fixed_size.append(2)
dq_fixed_size.append(3)
print(f"Fixed-size deque after 1, 2, 3: {dq_fixed_size}") # deque([1, 2, 3])

dq_fixed_size.append(4) # '1' is automatically discarded from the left
print(f"Fixed-size deque after append(4): {dq_fixed_size}") # deque([2, 3, 4])

dq_fixed_size.appendleft(0) # '4' is automatically discarded from the right
print(f"Fixed-size deque after appendleft(0): {dq_fixed_size}") # deque([0, 2, 3])


# 3. Removing Elements (O(1) time complexity)

# pop(): Removes and returns an element from the right end (tail)
popped_right = dq_empty.pop()
print(f"\nPopped from right ({popped_right}): {dq_empty}")

# popleft(): Removes and returns an element from the left end (head)
popped_left = dq_empty.popleft()
print(f"Popped from left ({popped_left}): {dq_empty}")

# remove(value): Removes the first occurrence of a specified value (O(N) time complexity)
dq_remove_example = deque([10, 20, 30, 20, 40])
print(f"\nOriginal for remove example: {dq_remove_example}")
dq_remove_example.remove(20)
print(f"After removing first '20': {dq_remove_example}")

# clear(): Removes all elements from the deque (O(N) time complexity)
dq_clear_example = deque([1, 2, 3])
print(f"Before clear: {dq_clear_example}")
dq_clear_example.clear()
print(f"After clear: {dq_clear_example}")


# 4. Accessing Elements (O(1) for ends, O(N) for middle)

# Accessing by index (like a list)
print(f"\nDeque for indexing: {dq_initial}")
print(f"First element (dq_initial[0]): {dq_initial[0]}")
print(f"Last element (dq_initial[-1]): {dq_initial[-1]}")
# Note: Accessing elements in the middle (e.g., dq_initial[len(dq_initial)//2])
# is not O(1) for deques, it's O(N) because deques are implemented as
# doubly linked lists of fixed-size blocks, not contiguous arrays like lists.
# For fast random access, use a list.


# 5. Other Useful Operations

# len(deque): Returns the number of elements (O(1) time complexity)
print(f"\nSize of dq_initial: {len(dq_initial)}")

# count(value): Returns the number of occurrences of a value (O(N) time complexity)
dq_count_example = deque([1, 2, 1, 3, 1, 4])
print(f"Deque for counting: {dq_count_example}")
print(f"Count of '1': {dq_count_example.count(1)}")

# rotate(n): Rotates the deque by n steps. Positive n rotates right, negative n rotates left. (O(k) time complexity, where k is the rotation amount)
dq_rotate_example = deque([1, 2, 3, 4, 5])
print(f"\nOriginal for rotate: {dq_rotate_example}")
dq_rotate_example.rotate(1) # Rotates 1 to the right (last element becomes first)
print(f"After rotate(1): {dq_rotate_example}") # deque([5, 1, 2, 3, 4])
dq_rotate_example.rotate(-2) # Rotates 2 to the left (first two elements go to the end)
print(f"After rotate(-2): {dq_rotate_example}") # deque([2, 3, 4, 5, 1])

# copy(): Creates a shallow copy of the deque
dq_copy = dq_rotate_example.copy()
print(f"Copied deque: {dq_copy}")

# reverse(): Reverses the elements of the deque in place (O(N) time complexity)
dq_reverse_example = deque([1, 2, 3, 4, 5])
print(f"\nOriginal for reverse: {dq_reverse_example}")
dq_reverse_example.reverse()
print(f"After reverse(): {dq_reverse_example}") # deque([5, 4, 3, 2, 1])

# Converting deque to a list or tuple
list_from_deque = list(dq_initial)
print(f"\nConverted to list: {list_from_deque}")
tuple_from_deque = tuple(dq_initial)
print(f"Converted to tuple: {tuple_from_deque}")

# Deque as a Stack (LIFO - Last In, First Out)
stack_dq = deque()
stack_dq.append('task1') # Push
stack_dq.append('task2') # Push
print(f"\nStack (LIFO): {stack_dq}")
print(f"Popped from stack: {stack_dq.pop()}") # Pop
print(f"Stack after pop: {stack_dq}")

# Deque as a Queue (FIFO - First In, First Out)
queue_dq = deque()
queue_dq.append('clientA') # Enqueue
queue_dq.append('clientB') # Enqueue
print(f"\nQueue (FIFO): {queue_dq}")
print(f"Dequeued from queue: {queue_dq.popleft()}") # Dequeue
print(f"Queue after dequeue: {queue_dq}")
````

````python
class Stack:
    def __init__(self):
        """
        Initializes an empty stack using a standard Python list.
        This is the most common and straightforward way to implement
        a stack in Python due to the list's efficient append() and pop()
        methods at the end.
        """
        self._items = []

    def push(self, item):
        """
        Adds an item to the top of the stack.
        This operation directly corresponds to Python list's append() method.

        Syntax: list_name.append(item)
        Time Complexity: Amortized O(1).
                         Appending an item to the end of a Python list generally takes
                         constant time. Occasionally, when the underlying array needs
                         to be reallocated and copied to accommodate more elements,
                         it can take O(N) time (where N is the current size of the list),
                         but this worst-case scenario is rare and the average cost
                         (amortized) over many operations remains O(1).
        """
        self._items.append(item)

    def pop(self):
        """
        Removes and returns the item from the top of the stack.
        This operation directly corresponds to Python list's pop() method
        when called without an index.

        Syntax: list_name.pop()
        Time Complexity: O(1).
                         Removing the last item from a Python list is a constant time operation.
        Raises: IndexError if the stack is empty (as list.pop() would on an empty list).
        """
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._items.pop()

    def peek(self):
        """
        Returns the item at the top of the stack without removing it.
        This operation uses Python list's negative indexing (list_name[-1]).

        Syntax: list_name[-1]
        Time Complexity: O(1).
                         Accessing an item by index (including negative indices like -1
                         for the last element) in a Python list is a constant time operation.
        Raises: IndexError if the stack is empty (as list[-1] would on an empty list).
        """
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self._items[-1]

    def is_empty(self):
        """
        Checks if the stack is empty.
        This can be efficiently done by checking the length of the list
        or by leveraging Python's truthiness of empty lists.

        Syntax Option 1 (Pythonic): not list_name
        Syntax Option 2 (Explicit): len(list_name) == 0
        Time Complexity: O(1) for both options.
        """
        return not self._items # More Pythonic
        # return len(self._items) == 0 # Also valid and clear

    def size(self):
        """
        Returns the number of items currently in the stack.
        This operation uses Python's built-in len() function on the list.

        Syntax: len(list_name)
        Time Complexity: O(1).
                         Getting the length of a Python list is a constant time operation.
        """
        return len(self._items)

    def __str__(self):
        """
        Returns a string representation of the stack, showing its current contents.
        The top of the stack is considered the rightmost element in the list display.
        """
        # Representing the stack with the top being on the right
        return f"Stack (top on right): {self._items}"

    def __repr__(self):
        """
        Returns an official string representation for developers.
        """
        return f"Stack({self._items})"

# --- Example Usage ---
if __name__ == "__main__":
    print("--- Initializing the Stack ---")
    my_stack = Stack()
    print(f"Current stack: {my_stack}")
    print(f"Is stack empty? {my_stack.is_empty()}")
    print(f"Stack size: {my_stack.size()}")

    # Trying to pop/peek from an empty stack
    try:
        my_stack.pop()
    except IndexError as e:
        print(f"Error: {e}")

    try:
        my_stack.peek()
    except IndexError as e:
        print(f"Error: {e}")

    print("\n--- Pushing Elements onto the Stack ---")
    my_stack.push("Task 1")
    print(f"Pushed 'Task 1': {my_stack}")
    print(f"Stack size: {my_stack.size()}")
    print(f"Top element: {my_stack.peek()}")

    my_stack.push(42)
    print(f"Pushed 42: {my_stack}")
    print(f"Stack size: {my_stack.size()}")
    print(f"Top element: {my_stack.peek()}")

    my_stack.push({'id': 101, 'status': 'pending'})
    print(f"Pushed {{'id': 101, 'status': 'pending'}}: {my_stack}")
    print(f"Stack size: {my_stack.size()}")
    print(f"Top element: {my_stack.peek()}")

    print("\n--- Popping Elements from the Stack ---")
    popped_item_1 = my_stack.pop()
    print(f"Popped: {popped_item_1}")
    print(f"Stack after pop: {my_stack}")
    print(f"Stack size: {my_stack.size()}")
    print(f"Top element: {my_stack.peek()}")

    popped_item_2 = my_stack.pop()
    print(f"Popped: {popped_item_2}")
    print(f"Stack after pop: {my_stack}")
    print(f"Stack size: {my_stack.size()}")
    print(f"Top element: {my_stack.peek()}")

    print("\n--- Emptying the Stack ---")
    while not my_stack.is_empty():
        item = my_stack.pop()
        print(f"Popped: {item}. Remaining: {my_stack}")
    print(f"Is stack empty? {my_stack.is_empty()}")
    print(f"Stack size: {my_stack.size()}")

    print("\n--- Final Checks ---")
    try:
        my_stack.pop()
    except IndexError as e:
        print(f"Error: {e}")
````

````python
import queue
import threading
import time

class ThreadSafeStack:
    def __init__(self, maxsize=0):
        """
        Initializes a thread-safe stack using queue.LifoQueue.
        LifoQueue implements a LIFO (Last-In, First-Out) queue,
        which is exactly the behavior of a stack.
        It is specifically designed for multi-threaded environments,
        handling locking internally to prevent race conditions.

        :param maxsize: Maximum number of items allowed in the queue.
                        If 0 (default), the queue size is infinite.
        """
        self._stack = queue.LifoQueue(maxsize=maxsize)

    def push(self, item):
        """
        Adds an item to the top of the stack.
        This operation is thread-safe.

        Syntax: LifoQueue_object.put(item, block=True, timeout=None)
        block: If True (default), blocks if the queue is full until a free slot is available.
        timeout: If block is True, waits for up to 'timeout' seconds.
                 If timeout occurs, raises queue.Full.
        Time Complexity: O(1) in practice (ignoring contention/blocking).
        """
        print(f"[{threading.current_thread().name}] Pushing: {item}")
        try:
            self._stack.put(item)
            print(f"[{threading.current_thread().name}] Pushed: {item}")
        except queue.Full:
            print(f"[{threading.current_thread().name}] Stack is full, could not push {item}")

    def pop(self):
        """
        Removes and returns an item from the top of the stack.
        This operation is thread-safe.

        Syntax: LifoQueue_object.get(block=True, timeout=None)
        block: If True (default), blocks if the queue is empty until an item is available.
        timeout: If block is True, waits for up to 'timeout' seconds.
                 If timeout occurs, raises queue.Empty.
        Time Complexity: O(1) in practice (ignoring contention/blocking).
        :return: The item removed from the top of the stack.
        :raises queue.Empty: If the stack is empty and block=False, or timeout occurs.
        """
        print(f"[{threading.current_thread().name}] Attempting to pop...")
        try:
            item = self._stack.get()
            print(f"[{threading.current_thread().name}] Popped: {item}")
            return item
        except queue.Empty:
            print(f"[{threading.current_thread().name}] Stack is empty, could not pop.")
            raise

    def peek(self):
        """
        Returns the item at the top of the stack without removing it.
        Note: LifoQueue does not have a direct 'peek' method.
        This implementation uses an internal hack (accessing _queue[0] which is dangerous
        due to internal implementation details and not truly thread-safe for peek),
        or by temporarily popping and then pushing back, which can affect order if other
        threads are active. A truly thread-safe peek often requires a lock if not built-in.
        For simplicity, we'll mimic by temporarily getting and putting back.
        A more robust peek might involve a separate lock or a custom implementation.

        This 'peek' is illustrative but has potential race conditions if multiple
        threads call it concurrently without additional external locking,
        or if the `_queue` internal structure changes in future Python versions.
        For production, if peek is critical, consider `queue.Queue` with `with` statements
        and a `_peek_data` attribute or a dedicated `threading.Lock`.
        """
        if self.is_empty():
            raise IndexError("peek from empty stack")
        # WARNING: Accessing _queue[0] directly is generally discouraged
        # as it bypasses the queue's internal locking mechanisms for put/get.
        # It works for LIFOQueue's internal list but can lead to race conditions
        # if not careful or if the internal implementation changes.
        # For a truly safe peek without temporary pop/push:
        # You'd need to acquire the internal lock. Example (not public API):
        # with self._stack.mutex:
        #     return self._stack.queue[self._stack.unfinished_tasks - 1]
        # Or, for simplest safe peek for demonstration with no other concurrent operations:
        # We will use get/put for a 'safe' peek in this example, though it's less efficient.
        item = self._stack.get(block=False) # Get without blocking
        self._stack.put(item) # Put it back immediately
        return item


    def is_empty(self):
        """
        Checks if the stack is empty.
        This operation is thread-safe.

        Syntax: LifoQueue_object.empty()
        Time Complexity: O(1).
        """
        return self._stack.empty()

    def size(self):
        """
        Returns the number of items in the stack.
        This operation is thread-safe.

        Syntax: LifoQueue_object.qsize()
        Time Complexity: O(1).
        """
        return self._stack.qsize()

    def __str__(self):
        """
        Returns a string representation of the stack (approximate view).
        Note: Accessing _queue directly for printing is generally discouraged
        in production code as it breaks encapsulation and can be unsafe in
        multi-threaded scenarios without proper locking.
        It's used here for demonstration purposes only.
        """
        # It's safer to not expose the internal list directly in multi-threaded environments.
        # This is just for a simple demonstration print.
        # For a truly safe representation, you might have to temporarily drain and refill,
        # which isn't practical.
        # The qsize() method gives a reliable count.
        return f"Stack (LifoQueue, current size: {self.size()})"

# --- Example Usage in a Multi-threaded Context ---

if __name__ == "__main__":
    print("--- Demonstrating Thread-Safe Stack (LifoQueue) ---")
    my_stack = ThreadSafeStack(maxsize=3) # Stack with a limited size

    def producer_task(stack, items_to_produce):
        thread_name = threading.current_thread().name
        print(f"[{thread_name}] Producer started.")
        for item in items_to_produce:
            stack.push(item)
            time.sleep(0.1) # Simulate work
        print(f"[{thread_name}] Producer finished.")

    def consumer_task(stack, items_to_consume):
        thread_name = threading.current_thread().name
        print(f"[{thread_name}] Consumer started.")
        consumed_count = 0
        while consumed_count < items_to_consume:
            try:
                item = stack.pop()
                consumed_count += 1
                time.sleep(0.2) # Simulate work
            except queue.Empty:
                print(f"[{thread_name}] Stack is currently empty, waiting...")
                time.sleep(0.5) # Wait a bit before trying again
        print(f"[{thread_name}] Consumer finished.")

    # Create threads
    producer_items = ['A', 'B', 'C', 'D', 'E', 'F']
    producer_thread = threading.Thread(target=producer_task, args=(my_stack, producer_items), name="ProducerThread")

    consumer_items_count = 6 # Consumer will try to get 6 items
    consumer_thread = threading.Thread(target=consumer_task, args=(my_stack, consumer_items_count), name="ConsumerThread")

    # Start threads
    producer_thread.start()
    consumer_thread.start()

    # Wait for threads to complete
    producer_thread.join()
    consumer_thread.join()

    print("\n--- All threads finished ---")
    print(f"Final stack size: {my_stack.size()}")
    print(f"Is stack empty? {my_stack.is_empty()}")

    # Demonstrate peek (with caution as noted in the method)
    if not my_stack.is_empty():
        try:
            print(f"Attempting to peek after threads: {my_stack.peek()}")
        except IndexError as e:
            print(f"Error during peek: {e}")
    else:
        print("Stack is empty, cannot peek.")

    # Demonstrate pushing beyond maxsize and handling Empty/Full
    print("\n--- Demonstrating Full/Empty exceptions (non-blocking) ---")
    small_stack = ThreadSafeStack(maxsize=1)
    small_stack.push("Item1") # Stack is full now

    print("Attempting to push when full (non-blocking):")
    try:
        small_stack._stack.put("Item2", block=False) # Use _stack.put directly for non-blocking
    except queue.Full:
        print("Caught queue.Full as expected!")

    print("Attempting to pop when empty (non-blocking):")
    try:
        small_stack.pop() # Pop "Item1"
        small_stack._stack.get(block=False) # Use _stack.get directly for non-blocking
    except queue.Empty:
        print("Caught queue.Empty as expected!")
````
