### Python Keyword Arguments

Keyword arguments allow you to pass arguments to a function by explicitly naming the parameters. This makes the function calls more readable and flexible.

#### 1\. Functions

```python
# Function with both positional and keyword arguments
def create_profile(name, age, city="New York"):
    # 'name' and 'age' are typically positional.
    # 'city' has a default value, but can be set with a keyword.
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")

# Calling the function using keyword arguments
# The order of keyword arguments doesn't matter.
create_profile(age=30, name="Alice")
# Output:
# Name: Alice
# Age: 30
# City: New York

# Mixing positional and keyword arguments.
# Positional arguments must come first.
create_profile("Bob", age=25, city="Los Angeles")
# Output:
# Name: Bob
# Age: 25
# City: Los Angeles

# Incorrect usage: positional argument after keyword argument raises a TypeError.
# create_profile(age=25, "Bob")  # This will fail.
```

#### 2\. Functions with Arbitrary Keyword Arguments (`**kwargs`)

- \*\*`**kwargs` allows a function to accept an arbitrary number of keyword arguments.
- It collects them into a dictionary where the keys are the argument names and the values are the argument values.
- The `**` syntax is a Python convention.

<!-- end list -->

```python
def print_kwargs(**kwargs):
    # 'kwargs' is a dictionary containing all the keyword arguments passed.
    for key, value in kwargs.items():
        print(f"Key: {key}, Value: {value}")

# Calling the function with various keyword arguments
print_kwargs(name="Charlie", age=40, job="Engineer")
# Output:
# Key: name, Value: Charlie
# Key: age, Value: 40
# Key: job, Value: Engineer

# You can combine positional arguments with **kwargs.
def print_details(required_arg, **kwargs):
    print(f"Required Argument: {required_arg}")
    print_kwargs(**kwargs)

print_details("Important", city="London", country="UK")
# Output:
# Required Argument: Important
# Key: city, Value: London
# Key: country, Value: UK
```

#### 3\. Class Constructors (`__init__`) and Methods

- Keyword arguments work identically in class methods and constructors.
- `**kwargs` is particularly useful in constructors for creating flexible classes that can handle a variety of attributes.

<!-- end list -->

```python
class Person:
    # A constructor using **kwargs for flexibility.
    def __init__(self, **kwargs):
        # The __dict__ attribute of an object stores its attributes.
        self.__dict__.update(kwargs)
        # This allows us to create objects with any attributes we pass in.

    def display_info(self):
        print("Person Info:")
        for key, value in self.__dict__.items():
            print(f"- {key.capitalize()}: {value}")

# Creating instances with different sets of attributes using keyword arguments.
person1 = Person(name="David", age=25, city="Paris")
person1.display_info()
# Output:
# Person Info:
# - Name: David
# - Age: 25
# - City: Paris

person2 = Person(name="Eve", job="Artist")
person2.display_info()
# Output:
# Person Info:
# - Name: Eve
# - Job: Artist
```

#### 4\. Keyword-Only Arguments

- You can enforce that certain arguments must be passed by keyword, not by position.
- This is done by placing them after a `*` or a `*args` in the function signature.

<!-- end list -->

```python
def send_email(subject, *, recipient, sender="noreply@example.com"):
    # 'subject' is positional.
    # 'recipient' and 'sender' MUST be called by keyword.
    # The '*' in the signature indicates that all arguments after it are keyword-only.
    print(f"From: {sender}")
    print(f"To: {recipient}")
    print(f"Subject: {subject}")

# Correct usage:
send_email("Meeting Reminder", recipient="team@example.com")
# Output:
# From: noreply@example.com
# To: team@example.com
# Subject: Meeting Reminder

# Incorrect usage: attempting to pass 'recipient' as a positional argument.
# send_email("Meeting Reminder", "team@example.com") # This will raise a TypeError.
```


````python
# 1. Functions with Keyword Arguments

# A function with both positional and keyword arguments.
def create_user(username, email, is_active=True):
    # `username` and `email` are typically positional arguments.
    # `is_active` has a default value but can be set with a keyword.
    print(f"Username: {username}")
    print(f"Email: {email}")
    print(f"Is Active: {is_active}")

# Calling the function using keyword arguments.
# The order of keyword arguments does not matter.
create_user(email="alice@example.com", username="alice")
# Output:
# Username: alice
# Email: alice@example.com
# Is Active: True

# Mixing positional and keyword arguments.
# Positional arguments must always come before keyword arguments.
create_user("bob", "bob@example.com", is_active=False)
# Output:
# Username: bob
# Email: bob@example.com
# Is Active: False


# 2. Functions with Arbitrary Keyword Arguments (**kwargs)

# The `**kwargs` syntax allows a function to accept any number of keyword arguments.
# They are collected into a dictionary.
def display_profile(**kwargs):
    # 'kwargs' is a dictionary containing all the keyword arguments passed.
    print("User Profile:")
    for key, value in kwargs.items():
        print(f"- {key.capitalize()}: {value}")

# Calling the function with various keyword arguments.
display_profile(name="Charlie", age=30, occupation="Developer")
# Output:
# User Profile:
# - Name: Charlie
# - Age: 30
# - Occupation: Developer

# You can combine positional arguments with `**kwargs`.
def combined_args(required_arg, **optional_kwargs):
    print(f"Required Argument: {required_arg}")
    display_profile(**optional_kwargs)

combined_args("Important ID", city="London", country="UK")
# Output:
# Required Argument: Important ID
# User Profile:
# - City: London
# - Country: UK


# 3. Class Constructors and Methods

class Product:
    # A constructor (`__init__`) using `**kwargs` for flexibility.
    def __init__(self, **attributes):
        # We can dynamically set instance attributes from the dictionary.
        for key, value in attributes.items():
            setattr(self, key, value)
    
    # A method that can also use keyword arguments.
    def display_details(self, show_id=True):
        print("Product Details:")
        if show_id:
            print(f"Product ID: {self.id}")
        print(f"Name: {self.name}")
        print(f"Price: ${self.price:.2f}")

# Creating instances with different sets of attributes using keyword arguments.
product1 = Product(id="P123", name="Laptop", price=1200.50)
product1.display_details()
# Output:
# Product Details:
# Product ID: P123
# Name: Laptop
# Price: $1200.50

product2 = Product(id="P456", name="Mouse", price=25.00)
product2.display_details(show_id=False)
# Output:
# Product Details:
# Name: Mouse
# Price: $25.00


# 4. Keyword-Only Arguments

# Use a `*` to specify that all arguments that follow must be keyword arguments.
def configure_settings(config_file, *, timeout=30, max_retries=5):
    # `config_file` is a positional argument.
    # `timeout` and `max_retries` must be called by keyword.
    print(f"Config file: {config_file}")
    print(f"Timeout: {timeout}s")
    print(f"Max Retries: {max_retries}")

# Correct usage:
configure_settings("config.ini", max_retries=10, timeout=60)
# Output:
# Config file: config.ini
# Timeout: 60s
# Max Retries: 10

# Incorrect usage: trying to pass `timeout` as a positional argument will fail.
# configure_settings("config.ini", 60) # Raises a TypeError.
````

````python
# --- Unlimited Keyword Arguments in Functions ---

def display_user_info(**user_details):
    """
    This function accepts an unlimited number of keyword arguments.
    The arguments are collected into a dictionary named 'user_details'.
    """
    print(f"Inside the function, 'user_details' is of type: {type(user_details)}")
    print(f"Keyword arguments received: {user_details}")
    
    print("--- User Details ---")
    for key, value in user_details.items():
        print(f"{key.capitalize()}: {value}")
    print("--------------------")

# Calling the function with different numbers of keyword arguments
display_user_info(name="Alice", age=30, city="New York")
print("\n")

display_user_info(book="Python Crash Course")
print("\n")

display_user_info()
print("\n" + "="*50 + "\n")


def print_user_profile(username, **options):
    """
    **kwargs can be used alongside other arguments.
    'username' is a required argument, and all other keyword arguments
    go into the 'options' dictionary.
    """
    print(f"Profile for user: {username}")
    if options:
        for key, value in options.items():
            print(f"  - {key}: {value}")
    else:
        print("  - No additional options provided.")

print_user_profile("Alice", logged_in=True, last_visit="2025-08-05")
print("\n")

print_user_profile("Bob")
print("\n" + "="*50 + "\n")


# --- Unlimited Keyword Arguments in Classes ---

class Profile:
    """
    Demonstrates using **kwargs in a class constructor (__init__).
    """
    def __init__(self, name, **attributes):
        self.name = name
        # The keyword arguments are stored as a dictionary
        self.attributes = attributes
        print(f"Created a new Profile for {self.name}")

    def display_profile(self):
        print(f"--- Profile for {self.name} ---")
        for key, value in self.attributes.items():
            print(f"{key.capitalize()}: {value}")
        print("--------------------")

# Creating class instances with different sets of keyword arguments
profile_1 = Profile(name="Alice", age=30, occupation="Developer")
profile_1.display_profile()
print("-" * 20)

profile_2 = Profile(name="Bob", hobby="Cycling")
profile_2.display_profile()
print("-" * 20)


class ReportGenerator:
    """
    Demonstrates using **kwargs in a regular class method.
    """
    def generate_report(self, title, **data):
        """Generates a formatted report from keyword arguments."""
        print(f"--- Report: {title} ---")
        if data:
            for key, value in data.items():
                print(f"  - {key}: {value}")
        else:
            print("  - No data provided.")
        print("--------------------")

report_gen = ReportGenerator()
report_gen.generate_report("Monthly Sales", month="August", total_sales=10000, new_customers=50)
print("\n")

report_gen.generate_report("Summary Report")
````
