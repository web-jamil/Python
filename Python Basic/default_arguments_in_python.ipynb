### Python Default Arguments

Default arguments in Python allow you to assign a default value to a function parameter. If a function is called without a value for that parameter, the default value is used. If a value is provided, it overrides the default.

#### 1\. Functions

Default arguments are defined in the function signature using the assignment operator (`=`).

```python
# Function with a default argument
def greet(name, message="Hello"):
    # `name` is a required argument, `message` has a default value.
    print(f"{message}, {name}!")

# Calling the function with and without the optional argument
greet("Alice")  # Output: "Hello, Alice!" (uses the default message)
greet("Bob", "Hi")  # Output: "Hi, Bob!" (overrides the default)
```

**Important Caveat: Mutable Default Arguments**

A common mistake is using mutable objects (like lists, dictionaries, or sets) as default arguments. The default value is created only once, at the time the function is defined. Subsequent calls to the function will use and modify the _same_ object.

```python
# Bad example: using a mutable list as a default argument
def append_to_list(value, my_list=[]):
    my_list.append(value)
    return my_list

# Let's see the unexpected behavior
list1 = append_to_list(1)  # list1 is [1]
print(list1)

list2 = append_to_list(2)  # list2 is [1, 2]! It's not a new list.
print(list2)

# To avoid this, use a sentinel value (like None)
def append_to_list_correct(value, my_list=None):
    if my_list is None:
        my_list = []  # A new list is created for each call where none is provided
    my_list.append(value)
    return my_list

list3 = append_to_list_correct(1)  # list3 is [1]
print(list3)

list4 = append_to_list_correct(2)  # list4 is a new list, [2]
print(list4)
```

#### 2\. Class Methods and Constructors

Default arguments work the same way in class methods, including the constructor (`__init__`).

```python
class Car:
    # The constructor `__init__` can have default arguments.
    def __init__(self, color="Red", model="Sedan"):
        self.color = color
        self.model = model

    # Regular methods can also have default arguments.
    def display_info(self, detailed=False):
        if detailed:
            print(f"This is a {self.color} {self.model}.")
        else:
            print(f"Model: {self.model}")

# Creating objects with and without default arguments
car1 = Car()  # Uses defaults: "Red" and "Sedan"
print(f"Car 1: {car1.color}, {car1.model}")

car2 = Car("Blue", "SUV")  # Overrides defaults
print(f"Car 2: {car2.color}, {car2.model}")

car2.display_info()  # Uses the default `detailed=False`
car2.display_info(detailed=True)  # Overrides the default
```

#### 3\. Other Sections (Lambda Functions)

Default arguments can also be used in lambda functions, which are small, anonymous functions.

```python
# A lambda function with a default argument
# The syntax is `lambda argument_name=default_value: expression`
add = lambda x, y=10: x + y

# Calling the lambda function
result1 = add(5)       # Uses default y=10, result1 is 15
result2 = add(5, 20)   # Overrides default, result2 is 25

print(f"Result 1: {result1}")
print(f"Result 2: {result2}")
```

#### Summary

- **Syntax:** `parameter_name = default_value` in the function/method signature.
- **Behavior:** The default value is used if the argument is omitted in the function call.
- **Best Practice:** Avoid using mutable objects as default arguments. Use a sentinel value like `None` and initialize the mutable object inside the function body instead.
- **Order:** Default arguments must be placed after all non-default arguments in the function signature. This is because Python matches arguments by position first. For example, `def func(a=1, b):` is a syntax error.


````python



# 1. Functions with Default Arguments

# A function with a required argument 'name' and a default argument 'greeting'
def greet(name, greeting="Hello"):
    # If the 'greeting' argument is not provided, "Hello" is used.
    print(f"{greeting}, {name}!")

# Calling the function with and without the default argument.
greet("Alice")          # Output: "Hello, Alice!"
greet("Bob", "Good morning")  # Output: "Good morning, Bob!"


# 2. Important Caveat: Mutable Default Arguments

# BAD EXAMPLE: The default list is created once and shared across all calls.
def append_item_bad(item, my_list=[]):
    # This list is the same object for every function call without 'my_list' specified.
    my_list.append(item)
    print(my_list)

# Let's see the unexpected behavior.
append_item_bad(1)  # Output: [1]
append_item_bad(2)  # Output: [1, 2] - The list from the first call was modified.
append_item_bad(3)  # Output: [1, 2, 3]


# CORRECT WAY: Use None as a sentinel and create a new mutable object inside.
def append_item_good(item, my_list=None):
    # Check if a list was provided. If not, create a new one.
    if my_list is None:
        my_list = []  # A new list is created each time.
    my_list.append(item)
    print(my_list)

# Now, the behavior is as expected.
append_item_good(1)  # Output: [1]
append_item_good(2)  # Output: [2] - A new list was created.
append_item_good(3)  # Output: [3]


# 3. Class Constructors (`__init__`) and Methods

class Car:
    # The constructor can have default arguments.
    def __init__(self, color="Red", engine="V6"):
        self.color = color
        self.engine = engine
    
    # Class methods can also have default arguments.
    def start_engine(self, key_present=True):
        if key_present:
            print(f"The {self.color} car's {self.engine} engine is starting.")
        else:
            print("No key found.")

# Creating objects using the default arguments.
car1 = Car()
car1.start_engine()  # Uses the default `key_present=True`.

# Creating an object by overriding the defaults.
car2 = Car(color="Blue", engine="Electric")
car2.start_engine(key_present=False) # Overrides the default.


# 4. Lambda Functions

# A lambda function with a default argument 'z'.
# The syntax is 'lambda arguments: expression'
add_three_numbers = lambda x, y, z=0: x + y + z

# Calling the lambda with and without the default argument.
result1 = add_three_numbers(1, 2)       # Uses default z=0, result1 is 3.
result2 = add_three_numbers(1, 2, 3)    # Overrides default, result2 is 6.

print(f"Result 1: {result1}")
print(f"Result 2: {result2}")


# 5. Argument Order

# REQUIRED: All non-default arguments must come before default arguments.
# This function signature is CORRECT.
def correct_order(a, b, c=3, d=4):
    print(a, b, c, d)

correct_order(1, 2)  # Output: 1 2 3 4
correct_order(1, 2, 10, 20) # Output: 1 2 10 20

# INCORRECT: This will raise a `SyntaxError`.
# def incorrect_order(a=1, b):
#     pass
````

````python
# 🌟 Default arguments in Python function definitions

def greet(name="Guest"):
    print(f"Hello, {name}!")

greet("Jamil")     # Output: Hello, Jamil!
greet()            # Output: Hello, Guest


# 🧮 Multiple default arguments
def power(base, exponent=2):
    return base ** exponent

print(power(3))       # Output: 9 (uses default exponent)
print(power(3, 3))    # Output: 27 (overrides default)


# 🧠 Default with mutable types (be careful!)
def append_item(item, items=[]):
    items.append(item)
    return items

print(append_item("a"))  # Output: ['a']
print(append_item("b"))  # Output: ['a', 'b'] — unexpected!

# ✅ Safer way using None
def append_item_safe(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(append_item_safe("a"))  # Output: ['a']
print(append_item_safe("b"))  # Output: ['b']


# 🚫 Default arguments must follow non-defaults
# def example(x=1, y): ❌ SyntaxError
def example(y, x=1):    # ✅ Correct
    return x + y

print(example(2))       # Output: 3
````

````python
# --- Understanding "Unlimited Default Arguments" in Python ---
# This is not a standard Python feature. You can have an unlimited number of
# positional arguments (*args) or keyword arguments (**kwargs), but you can't
# define an unlimited number of parameters with individual default values
# in a function signature.

# However, you can achieve a similar effect for keyword arguments
# by using **kwargs and providing defaults inside the function body.

# --- Case 1: Fixed Number of Default Arguments (The Standard Python Way) ---
# A function can have any number of parameters, each with a default value.

def print_settings(theme="dark", font_size=12, show_sidebar=True):
    """
    This function has 3 parameters, all with default values.
    """
    print("--- User Settings ---")
    print(f"Theme: {theme}")
    print(f"Font Size: {font_size}")
    print(f"Show Sidebar: {show_sidebar}")
    print("---------------------\n")

# Calling with all default values
print_settings()

# Calling and overriding some defaults
print_settings(theme="light", font_size=16)

# Calling and overriding all defaults by position
print_settings("light", 16, False)

# Calling and overriding all defaults by keyword
print_settings(show_sidebar=False, theme="light", font_size=16)

print("="*50)


# --- Case 2: Simulating "Unlimited" Default Arguments with **kwargs ---
# This is a common and powerful pattern for handling flexible configurations.

DEFAULT_REPORT_OPTIONS = {
    'sort_by': 'date',
    'format': 'pdf',
    'show_header': True,
    'pages': 'all'
}

def generate_report(**options):
    """
    This function accepts an unlimited number of keyword arguments.
    It uses a predefined dictionary of defaults and overrides them
    with any provided options.
    """
    print("--- Generating Report ---")
    
    # Create a copy of the defaults
    report_settings = DEFAULT_REPORT_OPTIONS.copy()
    
    # Update the settings with any provided keyword arguments
    report_settings.update(options)
    
    # Access the final settings, which now have the desired defaults
    print(f"Report will be sorted by: {report_settings['sort_by']}")
    print(f"Format: {report_settings['format']}")
    print(f"Show Header: {report_settings['show_header']}")
    print(f"Pages to include: {report_settings['pages']}")
    print("---------------------\n")

# Calling with no arguments, using all defaults
generate_report()

# Calling with some arguments, overriding defaults
generate_report(sort_by="title", pages="1-10")

# Calling with a new, unknown argument. It's stored in 'options'
generate_report(font_size=10, format="html")

print("="*50)


# --- Classes with Default Arguments ---

class User:
    def __init__(self, username, is_admin=False):
        """
        __init__ method with a fixed default argument.
        """
        self.username = username
        self.is_admin = is_admin
        print(f"User created: {self.username} (Admin: {self.is_admin})\n")

# Using the default argument
user1 = User("normal_user")

# Overriding the default argument
user2 = User("admin_user", is_admin=True)

print("="*50)


# --- Classes Simulating "Unlimited" Default Arguments with **kwargs ---

class Configurator:
    DEFAULT_CONFIG = {
        'api_key': 'default_key',
        'timeout': 30,
        'retries': 3
    }
    
    def __init__(self, **kwargs):
        """
        __init__ method that handles unlimited keyword arguments with defaults.
        """
        # Create an instance-specific copy of the defaults
        self.config = self.DEFAULT_CONFIG.copy()
        
        # Update with any provided arguments
        self.config.update(kwargs)
        
        print("--- New Configurator Instance ---")
        print("Final configuration:")
        for key, value in self.config.items():
            print(f"  - {key}: {value}")
        print("---------------------\n")

# Using all default arguments
config1 = Configurator()

# Overriding some default arguments
config2 = Configurator(api_key="my_new_key", retries=5)

# Providing a new argument not in the defaults
config3 = Configurator(api_key="another_key", log_level="INFO")
````
