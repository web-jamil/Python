A **monotonic data structure** is a specialized version of a standard data structure (like a stack, queue, or deque) where the elements are always kept in a specific, sorted order—either strictly increasing or decreasing. This property makes them incredibly efficient for solving problems that involve finding the "next greater/smaller element" or determining a maximum/minimum value within a dynamic window.

---

### Monotonic Stack

A monotonic stack is a stack where the elements are ordered from the bottom to the top. When you add a new element, you first pop any existing elements that violate the monotonic property.

- **Monotonic Increasing Stack** 📈: Elements are in increasing order from the bottom up. When a new element comes, you pop elements from the top of the stack that are **greater** than the new element. This ensures the stack remains sorted from smallest to largest.

  - **Use Case:** Finding the **Next Smaller Element** to the right for each element in an array.

- **Monotonic Decreasing Stack** 📉: Elements are in decreasing order from the bottom up. When a new element comes, you pop elements from the top of the stack that are **smaller** than the new element. This keeps the stack sorted from largest to smallest.
  - **Use Case:** Finding the **Next Greater Element** to the right for each element in an array, or calculating the largest rectangle in a histogram.

The key insight is that by popping elements, you're discarding irrelevant data. For example, in a monotonic decreasing stack, any element you pop is smaller than the current element. This means the popped element can never be the "next greater element" for any future element, so you can safely discard it.

---

### Monotonic Queue and Deque

A monotonic queue or deque maintains the same ordering principle but is typically implemented using a **deque (double-ended queue)**, which allows you to efficiently add and remove elements from both the front and the back. This flexibility is what makes it so powerful.

- **What it is:** A monotonic queue is a queue-like structure that keeps its elements in a sorted order. Unlike a standard queue which is strictly FIFO (First-In, First-Out), a monotonic queue can remove elements from both the front and the back to maintain its property.

- **Increasing/Decreasing Deque Logic:**

  - **Increasing Deque** 📈: The front of the deque holds the smallest element and the back holds the largest. When you add a new element, you pop from the back any elements that are larger than the new one. This keeps the smallest element at the front, which is ideal for finding the minimum in a window.
  - **Decreasing Deque** 📉: The front of the deque holds the largest element and the back holds the smallest. When a new element is added, you pop from the back any elements that are smaller than the new one. This keeps the largest element at the front, which is great for finding the maximum in a window.

- **Primary Use Case:** Monotonic deques are the go-to solution for **Sliding Window** problems where you need to find the maximum or minimum value in a window of a fixed size `k` in $O(n)$ time. The deque stores the indices of the elements, and by checking the front, you can get the current max/min and also remove elements that have fallen out of the window.

In summary, all these structures are about **maintaining order to enable fast lookups**. While a monotonic stack is limited to push/pop from one end, a monotonic deque allows for more complex, window-based operations with the same $O(n)$ efficiency.

It's a powerful technique that often transforms a brute-force $O(n^2)$ solution into a highly efficient $O(n)$ solution.

### The Core Theory of Monotonicity

The central idea behind monotonic data structures is to maintain a sorted (monotonic) order of elements as you traverse an input array or list. This sorted order allows you to make an immediate, optimal decision about a new element without having to re-examine all previous elements.

When you iterate through an array, you're essentially processing elements one by one. A monotonic stack or deque stores a subset of the previously seen elements. This subset is carefully chosen to be **relevant** for future calculations. An element is considered "irrelevant" if a newer element can solve the same future problems more optimally.

For example, consider a monotonic decreasing stack used to find the next greater element. When a new element arrives, any elements in the stack that are smaller than it become irrelevant. The new element is a better candidate for being the "next greater element" for future elements than the smaller, older elements. So, you pop the smaller elements, making the stack more concise and efficient. This simple process guarantees that each element is pushed and popped at most once, leading to an overall linear time complexity. 📈

### Common Use Cases in DSA

Monotonic data structures are not a general-purpose tool but are highly effective for a specific class of problems. Here are the most common scenarios:

- **Next Greater/Smaller Element:** This is the classic application.

  - **Next Greater Element:** For each number in an array, find the first number to its right that is greater than it. A **monotonic decreasing stack** is the ideal tool for this.
  - **Next Smaller Element:** For each number in an array, find the first number to its right that is smaller than it. A **monotonic increasing stack** is used here.

- **Largest Rectangle in Histogram:** Given an array of integers representing bar heights in a histogram, find the area of the largest rectangle. A monotonic stack can efficiently find the left and right boundaries for each bar that would form a rectangle.

- **Sliding Window Maximum/Minimum:** When you need to find the maximum or minimum value within a fixed-size window that slides across an array. A **monotonic deque** is perfect for this. It keeps the deque ordered so that the maximum or minimum element is always at one of the ends (front or back), allowing for constant-time lookups.

- **Trapping Rain Water:** This problem involves calculating the amount of water that can be trapped between bars of a histogram. A monotonic stack can be used to solve this by finding the "walls" that trap the water.

In all these cases, the core principle is the same: use the monotonic property to eliminate redundant comparisons and make optimal decisions in a single pass through the data.

### Monotonic Increasing Stack 📈

An **increasing monotonic stack** is one where the elements are always sorted in ascending order from the bottom to the top. When a new element comes, you pop all elements from the top of the stack that are **greater than** the new element. You do this because those popped elements can never be the "next smaller element" for any future element—the new, smaller element is a better candidate.

**When to Use It:**

- To find the **Next Smaller Element** to the right for each element in an array.
- To find the **Previous Smaller Element** to the left.
- Problems related to finding the area of a rectangle or histogram where you need to find the nearest smaller bar on both sides.

**Example Logic (Next Smaller Element):**

1.  Initialize an empty stack.
2.  Iterate through the array.
3.  For each element, `x`:
    - While the stack is not empty and the top element is `> x`, pop the stack. The popped element's next smaller element is `x`.
    - Push `x` onto the stack.

### Monotonic Decreasing Stack 📉

A **decreasing monotonic stack** is a stack where the elements are always sorted in descending order from the bottom to the top. When a new element arrives, you pop all elements from the top of the stack that are **smaller than** the new element. This is because the popped elements are now "blocked" by the new, larger element and can no longer be the "next greater element" for future elements.

**When to Use It:**

- To find the **Next Greater Element** to the right for each element in an array.
- To find the **Previous Greater Element** to the left.
- In the **Largest Rectangle in Histogram** problem, to find the nearest smaller bar to the left and right.

**Example Logic (Next Greater Element):**

1.  Initialize an empty stack.
2.  Iterate through the array.
3.  For each element, `x`:
    - While the stack is not empty and the top element is `< x`, pop the stack. The popped element's next greater element is `x`.
    - Push `x` onto the stack.

### Key Concept: The "Relevance" Principle

The power of a monotonic stack lies in its ability to discard irrelevant information. The stack only stores potential candidates for a future answer. For instance, if you're looking for the next greater element and you encounter a new, larger number, any smaller numbers currently in your stack are no longer the "next greater" for anything that comes later. You can discard them, ensuring that each element is processed in a streamlined, efficient manner.


A monotonic queue, often implemented with a deque (double-ended queue), is a data structure where elements are maintained in a specific sorted order, either increasing or decreasing. Unlike a monotonic stack, which is optimized for finding the next greater or smaller element, a monotonic queue is designed for problems that involve a **sliding window**, allowing for efficient $O(1)$ lookups of the maximum or minimum element in the current window.

### The Core Concept

The fundamental difference between a monotonic stack and a monotonic queue is the **sliding window property**. While a stack processes elements in a LIFO (Last-In, First-Out) manner and is excellent for a single pass, a queue must handle both insertions and deletions from its ends to manage a dynamic window of elements. A monotonic queue maintains its order by adding new elements to the back and removing them from both the back (to preserve the monotonic property) and the front (to discard elements that have slid out of the window). This makes it perfect for problems where you need to track the maximum or minimum value within a sub-array of a fixed size.

---

### Monotonic Decreasing Queue 📉

A monotonic **decreasing** queue stores elements in descending order from front to back. The element at the front of the queue is always the maximum element in the current window.

**How it works:**

1.  **Add a new element:** Before adding a new element to the back of the queue, you pop all smaller elements from the back. This ensures the queue maintains its decreasing order.
2.  **Remove an old element:** You check if the element at the front of the queue is outside the current window. If it is, you pop it from the front.

**When to Use It:**
The classic use case for a monotonic decreasing queue is the **Sliding Window Maximum** problem, where you need to find the maximum value for each sub-array of a fixed size `k`.

---

### Monotonic Increasing Queue 📈

A monotonic **increasing** queue stores elements in ascending order from front to back. The element at the front of the queue is always the minimum element in the current window.

**How it works:**

1.  **Add a new element:** Before adding a new element to the back of the queue, you pop all larger elements from the back. This keeps the queue in increasing order.
2.  **Remove an old element:** You check if the element at the front is outside the window. If so, you remove it.

**When to Use It:**
This is used for problems like **Sliding Window Minimum**, where you need to find the minimum element in a moving window of size `k`.

A **monotonic deque** is a double-ended queue (deque) that maintains its elements in a specific sorted order, either strictly increasing or strictly decreasing. It's essentially the same concept as a monotonic queue, but using the term "deque" explicitly highlights its ability to add and remove elements from both ends. This dual-end functionality is what makes it the ideal data structure for solving **sliding window** problems in $O(n)$ time.

---

### Key Difference from a Monotonic Stack

While a monotonic stack is great for finding the next greater/smaller element for a single pass through an array, a monotonic deque is designed for scenarios where a "window" of elements is moving. The deque's two ends are used for different purposes:

- **Back of the deque:** Used to maintain the monotonic order as new elements enter the window.
- **Front of the deque:** Used to remove elements that are no longer part of the current window.

This combination of operations allows the deque to always have the maximum or minimum element of the current sliding window at its front, giving you a constant-time $O(1)$ lookup.

---

### When to Use a Monotonic Deque

You should use a monotonic deque whenever you encounter a problem that requires finding the **maximum or minimum value within a sliding window of a fixed size `k`**. The classic example of this is the "Sliding Window Maximum" problem.

---

### Example: Sliding Window Maximum

Let's say you have an array `nums = [1, 3, -1, -3, 5, 3, 6, 7]` and a window size `k = 3`. You want to find the maximum in each window.

1.  **Initialize an empty deque** to store indices, not values. This is important so you can check if an element is still in the window based on its index. The deque will be monotonic decreasing, with the largest element's index at the front.

2.  **Iterate through the array `nums`:**
    - **Add new elements:** For each new number `nums[i]`, you pop all indices from the back of the deque where `nums[deque.back()] <= nums[i]`. This ensures the deque's order is maintained. Then, you push the new index `i` to the back.
    - **Remove old elements:** If the index at the front of the deque is `i - k`, it means that element has slid out of the window. You pop that index from the front.
    - **Record the maximum:** The maximum element for the current window is always `nums[deque.front()]`. You add this to your result array.

Following this process, you get the output `[3, 3, 5, 5, 6, 7]` in a single $O(n)$ pass.

### Key Scenarios for Using Monotonicity

#### 1. Next Greater/Smaller Element

This is the most common application. A monotonic stack is used to efficiently find the next greater or smaller element for every element in an array. A brute-force approach with nested loops would be $O(n^2)$, but a monotonic stack solves it in **$O(n)$** time.

- **Next Greater Element:** Use a **monotonic decreasing stack**. When you encounter a new element `x`, you pop all elements from the stack that are smaller than `x`. For each popped element, `x` is its next greater element.
- **Next Smaller Element:** Use a **monotonic increasing stack**. When you encounter a new element `x`, you pop all elements from the stack that are greater than `x`. For each popped element, `x` is its next smaller element.

#### 2. Sliding Window Problems

When a problem asks for the maximum or minimum value in a sub-array of a fixed size `k` that slides across the main array, a monotonic deque is the perfect tool.

- **Sliding Window Maximum:** Use a **monotonic decreasing deque**. The deque's front will always contain the index of the maximum element in the current window. As the window slides, you add new elements to the back and remove old elements from the front.
- **Sliding Window Minimum:** Use a **monotonic increasing deque**. The deque's front will always hold the index of the minimum element in the current window.

---

### When to Avoid Monotonicity

Monotonic data structures are highly specialized. They are not a general-purpose solution and should not be used for problems that don't fit the patterns described above. For instance, if you need to find all elements greater than a given element (not just the first one), a simple iteration or sorting approach may be more suitable.
