A **queue** is a linear data structure that follows the **First-In, First-Out (FIFO)** principle. This means the first element added to the queue is the first one to be removed. Think of it like a line at a store—the first person in line is the first person to be served.

A queue has two main operations:

- **Enqueue:** Adding an element to the back (or tail) of the queue.
- **Dequeue:** Removing an element from the front (or head) of the queue.

Unlike a stack, where you only access the top element, a queue requires access to both ends.

---

### Python Implementation

In Python, you can implement a queue using a list, but it's inefficient for large queues because removing the first element (`list.pop(0)`) requires shifting all other elements, resulting in an O(n) time complexity.

A much better approach is to use the `collections.deque` class, which is optimized for fast appends and pops from both ends, giving you O(1) time complexity for enqueue and dequeue operations.

Here's how to implement a queue using `collections.deque`:

```python
from collections import deque

# Create an empty queue
queue = deque()

# Enqueue operation: add elements to the right (back) of the queue
print("Enqueuing elements: 10, 20, 30")
queue.append(10)
queue.append(20)
queue.append(30)
print("Queue after enqueue:", queue) # Output: deque([10, 20, 30])

# Peek operation: view the front element without removing it
# The front of the queue is the leftmost element
print("Front of the queue (peek):", queue[0]) # Output: 10

# Dequeue operation: remove elements from the left (front) of the queue
print("Dequeuing element:", queue.popleft())  # Output: 10
print("Queue after dequeue:", queue) # Output: deque([20, 30])

print("Dequeuing element:", queue.popleft())  # Output: 20
print("Queue after dequeue:", queue) # Output: deque([30])

# Check if the queue is empty
print("Is the queue empty?", not queue) # Output: False

print("Dequeuing element:", queue.popleft()) # Output: 30
print("Queue after dequeue:", queue) # Output: deque([])

print("Is the queue empty?", not queue) # Output: True

```

## **2. Python Implementations of Queue**

Python doesn’t have a dedicated queue in core syntax, but the **`collections`** and **`queue`** modules provide efficient implementations.

---

### **A. Using `collections.deque` (Fast & Recommended for Simple Queues)**

```python
from collections import deque

# Create a queue
queue = deque()

# Enqueue
queue.append("A")
queue.append("B")
queue.append("C")   # queue = ['A', 'B', 'C']

# Dequeue
front = queue.popleft()  # Removes 'A'
print("Removed:", front)

# Peek front element
print("Front element:", queue[0])

# Check size
print("Size:", len(queue))

# Check empty
print("Empty?", len(queue) == 0)
```

---

### **B. Using `queue.Queue` (Thread-Safe, Used in Multithreading)**

```python
from queue import Queue

# Create queue with max size
q = Queue(maxsize=5)

# Enqueue
q.put(10)
q.put(20)
q.put(30)

# Dequeue
print(q.get())  # 10 removed

# Size
print("Size:", q.qsize())

# Empty check
print("Empty?", q.empty())

# Full check
print("Full?", q.full())
```

**Notes:**

- `queue.Queue` is **thread-safe** — suitable for producer/consumer problems.
- `put()` and `get()` can be **blocking** (wait until item available/space available).

---

### **C. Using `queue.LifoQueue` (Stack Behavior)**

```python
from queue import LifoQueue

stack_like_queue = LifoQueue()

stack_like_queue.put(1)
stack_like_queue.put(2)
stack_like_queue.put(3)

print(stack_like_queue.get())  # Removes 3 (LIFO)
```

---

### **D. Using `queue.PriorityQueue` (Priority-Based Removal)**

```python
from queue import PriorityQueue

pq = PriorityQueue()

# Lower number = higher priority
pq.put((1, "Task A"))
pq.put((3, "Task C"))
pq.put((2, "Task B"))

while not pq.empty():
    print(pq.get())  # (1, 'Task A'), (2, 'Task B'), (3, 'Task C')
```

- **`PriorityQueue`** → When elements must be processed by priority.

# Python Queue Data Structure: Theory and Concepts

## Introduction to Queues

A queue is a fundamental linear data structure that follows the **First-In-First-Out (FIFO)** principle. This means the first element added to the queue will be the first one to be removed, similar to a real-world queue (like people waiting in line).

## Basic Queue Operations

1. **Enqueue (Add)**: Add an item to the end of the queue
2. **Dequeue (Remove)**: Remove an item from the front of the queue
3. **Peek/Front**: Get the front item without removing it
4. **IsEmpty**: Check if the queue is empty
5. **Size**: Get the number of items in the queue

## Queue Implementation in Python

Python offers several ways to implement queues:

### 1. Using List (Not Recommended for Performance)

```python
queue = []

# Enqueue
queue.append('a')
queue.append('b')
queue.append('c')

# Dequeue
item = queue.pop(0)  # 'a' is removed
```

_Note_: Using `pop(0)` on a list is O(n) operation, making it inefficient for large queues.

### 2. Using collections.deque (Recommended)

```python
from collections import deque

queue = deque()

# Enqueue
queue.append('a')
queue.append('b')
queue.append('c')

# Dequeue
item = queue.popleft()  # 'a' is removed - O(1) operation
```

`deque` is optimized for fast appends and pops from both ends, making it ideal for queue implementation.

### 3. Using queue.Queue (Thread-Safe)

```python
from queue import Queue

q = Queue(maxsize=3)

# Enqueue
q.put('a')
q.put('b')
q.put('c')

# Check if full
print(q.full())  # True

# Dequeue
item = q.get()  # 'a' is removed

# Check if empty
print(q.empty())  # False
```

This implementation is thread-safe and useful for multi-threaded programming.

## Types of Queues

### 1. Simple Queue

Basic FIFO queue as described above.

### 2. Circular Queue (Ring Buffer)

- Fixed size
- When full, oldest elements are overwritten
- Efficient memory usage

```python
from collections import deque

class CircularQueue:
    def __init__(self, size):
        self.queue = deque(maxlen=size)

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        return self.queue.popleft()

    def is_full(self):
        return len(self.queue) == self.queue.maxlen
```

### 3. Priority Queue

- Elements are dequeued based on priority
- In Python, can be implemented using `heapq` module

```python
import heapq

class PriorityQueue:
    def __init__(self):
        self._queue = []
        self._index = 0

    def push(self, item, priority):
        heapq.heappush(self._queue, (-priority, self._index, item))
        self._index += 1

    def pop(self):
        return heapq.heappop(self._queue)[-1]
```

### 4. Double-Ended Queue (Deque)

- Allows insertion and removal from both ends
- Implemented using `collections.deque`

## Practical Example: BFS Implementation

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

# Example usage
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

bfs(graph, 'A')  # Output: A B C D E F
```

### Key Concepts

#### FIFO Principle 🔄

The **FIFO** principle is the core concept of a queue. It dictates the order in which elements are processed. This is different from a stack, which uses a Last-In, First-Out (LIFO) principle.

#### Enqueue and Dequeue

- **Enqueue**: The operation of adding an element to the **rear** (or end) of the queue.
- **Dequeue**: The operation of removing an element from the **front** (or beginning) of the queue.

#### Front and Rear

- **Front**: The element at the beginning of the queue, ready to be dequeued.
- **Rear**: The element at the end of the queue, where new elements are enqueued.

---

### Queue Implementation in Python

While you can implement a queue from scratch using lists, Python's built-in `collections` module provides a more efficient and thread-safe option: `deque`.

#### Using `collections.deque`

The `deque` (double-ended queue) class is optimized for fast appends and pops from both ends, making it an excellent choice for a queue.

```python
from collections import deque

# Create an empty queue
q = deque()

# Enqueue elements
q.append('A')  # Adds to the rear
q.append('B')
q.append('C')

print(q)  # Output: deque(['A', 'B', 'C'])

# Dequeue elements
front_element = q.popleft()  # Removes from the front
print(front_element)  # Output: A

print(q)  # Output: deque(['B', 'C'])

# Access the front element without removing it
if q:  # Check if the queue is not empty
    print(q[0])  # Output: B

# Check if the queue is empty
print(not q) # Output: False

# Get the size of the queue
print(len(q))  # Output: 2
```

Using `popleft()` is the standard way to dequeue from a `deque` object, as it's an O(1) operation, meaning it's very fast regardless of the queue's size.

## 1. Basic List Implementation (Not Recommended for Production)

```python
# Initialize
queue = []

# Enqueue (add to rear)
queue.append(1)
queue.append(2)
queue.append(3)
print("After enqueue:", queue)  # [1, 2, 3]

# Dequeue (remove from front)
item = queue.pop(0)
print("Dequeued item:", item)  # 1
print("After dequeue:", queue)  # [2, 3]

# Peek (front element)
front = queue[0] if queue else None
print("Front element:", front)  # 2

# Check empty
is_empty = len(queue) == 0
print("Is empty:", is_empty)  # False

# Size
size = len(queue)
print("Size:", size)  # 2
```

## 2. Deque Implementation (Recommended)

```python
from collections import deque

# Initialize
queue = deque()

# Enqueue
queue.append(10)
queue.append(20)
queue.append(30)
print("Queue:", queue)  # deque([10, 20, 30])

# Dequeue
item = queue.popleft()
print("Dequeued:", item)  # 10
print("Queue:", queue)  # deque([20, 30])

# Peek
front = queue[0] if queue else None
print("Front:", front)  # 20

# Check empty
is_empty = len(queue) == 0

# Size
size = len(queue)

# Additional deque operations
queue.appendleft(5)  # Add to front (not standard queue op)
print("After appendleft:", queue)  # deque([5, 20, 30])

last = queue.pop()  # Remove from end (not standard queue op)
print("Popped from end:", last)  # 30
```

## 3. Queue Module (Thread-Safe)

```python
from queue import Queue

# Initialize with maxsize
q = Queue(maxsize=3)

# Enqueue
q.put(1)
q.put(2)
q.put(3)
print("Full?", q.full())  # True

# Dequeue
item = q.get()
print("Dequeued:", item)  # 1
print("Full?", q.full())  # False

# Peek (non-standard, requires workaround)
if not q.empty():
    front = q.queue[0]
    print("Front:", front)  # 2

# Check empty
is_empty = q.empty()

# Size
size = q.qsize()

# Blocking operations
import threading
import time

def worker(q):
    item = q.get()  # Blocks until item available
    print("Worker got:", item)
    q.task_done()

q = Queue()
threading.Thread(target=worker, args=(q,)).start()
time.sleep(0.5)
q.put(100)  # Worker will process this
q.join()  # Wait for all tasks to be processed
```

## 4. Priority Queue Implementation

```python
import heapq

# Initialize
heap = []

# Enqueue with priority
heapq.heappush(heap, (2, 'code'))
heapq.heappush(heap, (1, 'eat'))
heapq.heappush(heap, (3, 'sleep'))
print("Heap:", heap)  # [(1, 'eat'), (2, 'code'), (3, 'sleep')]

# Dequeue (always gets lowest priority)
item = heapq.heappop(heap)
print("Dequeued:", item)  # (1, 'eat')

# Peek
front = heap[0] if heap else None
print("Front:", front)  # (2, 'code')

# Check empty
is_empty = len(heap) == 0

# Size
size = len(heap)
```

## 5. Circular Queue (Ring Buffer)

```python
class CircularQueue:
    def __init__(self, size):
        self.size = size
        self.queue = [None] * size
        self.front = self.rear = -1

    def enqueue(self, item):
        if (self.rear + 1) % self.size == self.front:
            print("Queue is full")
            return
        elif self.front == -1:  # First element
            self.front = self.rear = 0
        else:
            self.rear = (self.rear + 1) % self.size
        self.queue[self.rear] = item

    def dequeue(self):
        if self.front == -1:
            print("Queue is empty")
            return
        item = self.queue[self.front]
        if self.front == self.rear:  # Last element
            self.front = self.rear = -1
        else:
            self.front = (self.front + 1) % self.size
        return item

    def peek(self):
        if self.front == -1:
            return None
        return self.queue[self.front]

    def is_empty(self):
        return self.front == -1

    def is_full(self):
        return (self.rear + 1) % self.size == self.front

    def __str__(self):
        if self.front == -1:
            return "[]"
        if self.rear >= self.front:
            return str(self.queue[self.front:self.rear+1])
        else:
            return str(self.queue[self.front:] + self.queue[:self.rear+1])

# Usage
cq = CircularQueue(3)
cq.enqueue(1)
cq.enqueue(2)
cq.enqueue(3)
print("Queue:", cq)  # [1, 2, 3]
cq.enqueue(4)  # Queue is full
print("Dequeue:", cq.dequeue())  # 1
cq.enqueue(4)
print("Queue:", cq)  # [2, 3, 4]
```

## 6. Double-Ended Queue (Deque) Advanced Operations

```python
from collections import deque

d = deque([1, 2, 3])

# Add to both ends
d.appendleft(0)  # [0, 1, 2, 3]
d.append(4)      # [0, 1, 2, 3, 4]

# Remove from both ends
left = d.popleft()  # 0
right = d.pop()     # 4

# Rotation
d.rotate(1)   # [3, 1, 2]
d.rotate(-1)  # [1, 2, 3]

# Maxlen (circular behavior)
d = deque(maxlen=3)
d.append(1)
d.append(2)
d.append(3)
d.append(4)  # [2, 3, 4] (1 is automatically removed)

# Extend operations
d.extend([5, 6])  # [4, 5, 6]
d.extendleft([3, 2])  # [2, 3, 4, 5, 6] (note order)
```

## 7. Multi-Threaded Queue with Producer-Consumer Pattern

```python
from queue import Queue
import threading
import time
import random

def producer(q, count):
    for i in range(count):
        item = random.randint(1, 100)
        q.put(item)
        print(f"Produced {item}")
        time.sleep(random.random())

def consumer(q, name):
    while True:
        item = q.get()
        if item is None:  # Sentinel value to exit
            break
        print(f"{name} consumed {item}")
        q.task_done()
        time.sleep(random.random() * 2)

# Create queue
q = Queue()

# Start producer
prod_thread = threading.Thread(target=producer, args=(q, 10))
prod_thread.start()

# Start consumers
cons_threads = []
for i in range(3):
    t = threading.Thread(target=consumer, args=(q, f"Consumer-{i+1}"))
    t.start()
    cons_threads.append(t)

# Wait for producer to finish
prod_thread.join()

# Add sentinels to stop consumers
for _ in range(3):
    q.put(None)

# Wait for all consumers to finish
for t in cons_threads:
    t.join()

print("All tasks completed")
```
