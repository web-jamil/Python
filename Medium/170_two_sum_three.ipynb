# 170. Two Sum III - Data structure design

Design a data structure that accepts a stream of integers and checks if it has a pair of integers that sum up to a particular value.

Implement the TwoSum class:

TwoSum() Initializes the TwoSum object, with an empty array initially.
void add(int number) Adds number to the data structure.
boolean find(int value) Returns true if there exists any pair of numbers whose sum is equal to value, otherwise, it returns false.

# Example 1:

```
Input
["TwoSum", "add", "add", "add", "find", "find"]
[[], [1], [3], [5], [4], [7]]
Output
[null, null, null, null, true, false]

Explanation
TwoSum twoSum = new TwoSum();
twoSum.add(1);   // [] --> [1]
twoSum.add(3);   // [1] --> [1,3]
twoSum.add(5);   // [1,3] --> [1,3,5]
twoSum.find(4);  // 1 + 3 = 4, return true
twoSum.find(7);  // No two integers sum up to 7, return false
```

# Constraints:

- 105 <= number <= 105
- 231 <= value <= 231 - 1
- At most 104 calls will be made to add and find.


**1. Naive Approach (Using a List):**

- **Data Structure:** A simple list (or array) to store the incoming numbers.
- **`add(number)`:** Append the `number` to the list.
- **`find(value)`:** Iterate through all possible pairs of numbers in the list. For each pair `(num1, num2)`, check if `num1 + num2 == value`. If a pair is found, return `true`. If the loop completes without finding a pair, return `false`.

  ```python
  class TwoSumNaive:
      def __init__(self):
          self.numbers = []

      def add(self, number: int) -> None:
          self.numbers.append(number)

      def find(self, value: int) -> bool:
          n = len(self.numbers)
          for i in range(n):
              for j in range(i + 1, n):
                  if self.numbers[i] + self.numbers[j] == value:
                      return True
          return False
  ```

- **Time Complexity:**
  - `add`: O(1)
  - `find`: O(n^2), where n is the number of elements.
- **Space Complexity:** O(n) to store the numbers.
- **Pros:** Simple to implement.
- **Cons:** `find` operation is very inefficient for a large number of elements.

**2. Using a Set (for Faster `find`):**

- **Data Structure:** A list to store the numbers and a set to quickly check for the existence of a number.
- **`add(number)`:** Append the `number` to the list and add it to the set.
- **`find(value)`:** Iterate through the list of numbers. For each `num`, calculate the `complement = value - num`. Check if the `complement` exists in the set. To avoid using the same number twice (unless duplicates are allowed and should be counted), you need to be careful.

  ```python
  class TwoSumSet:
      def __init__(self):
          self.numbers = []
          self.seen = set()

      def add(self, number: int) -> None:
          self.numbers.append(number)
          self.seen.add(number)

      def find(self, value: int) -> bool:
          for num in self.numbers:
              complement = value - num
              if complement in self.seen:
                  if complement != num or self.numbers.count(num) > 1:
                      return True
          return False
  ```

- **Time Complexity:**
  - `add`: O(1)
  - `find`: O(n), where n is the number of elements. The `count` operation in `find` can be O(n) in the worst case for a list.
- **Space Complexity:** O(n) to store the numbers and the set.
- **Pros:** `find` operation is significantly faster than the naive approach.
- **Cons:** The `find` operation still iterates through the list. Handling duplicates requires extra care.

**3. Using a Hash Map (Frequency Count):**

- **Data Structure:** A hash map (dictionary in Python) to store the frequency of each added number.
- **`add(number)`:** Increment the count of `number` in the hash map. If the number is not present, add it with a count of 1.
- **`find(value)`:** Iterate through the keys (unique numbers) in the hash map. For each `num`, calculate the `complement = value - num`. Check if the `complement` exists as a key in the hash map.

  - If `num == complement`, then we need to check if the frequency of `num` is greater than or equal to 2.
  - If `num != complement`, then we just need to check if the `complement` exists in the hash map with a frequency greater than 0.

  ```python
  class TwoSumHashmap:
      def __init__(self):
          self.counts = {}

      def add(self, number: int) -> None:
          self.counts[number] = self.counts.get(number, 0) + 1

      def find(self, value: int) -> bool:
          for num in self.counts:
              complement = value - num
              if complement in self.counts:
                  if num == complement:
                      if self.counts[num] >= 2:
                          return True
                  else:
                      return True
          return False
  ```

- **Time Complexity:**
  - `add`: O(1) on average.
  - `find`: O(n), where n is the number of unique elements in the hash map (which can be at most the total number of added elements).
- **Space Complexity:** O(n) to store the counts in the hash map.
- **Pros:** Efficient `find` operation. Handles duplicates naturally.
- **Cons:** Requires extra space for the hash map.

**4. Sorted List with Two Pointers:**

- **Data Structure:** A sorted list to store the incoming numbers.
- **`add(number)`:** Append the `number` to the list and then maintain the sorted order (e.g., by inserting it in the correct position or sorting after each addition).
- **`find(value)`:** Use the two-pointer technique. Initialize two pointers, `left` at the beginning of the list and `right` at the end.

  - If `numbers[left] + numbers[right] == value`, return `true`.
  - If `numbers[left] + numbers[right] < value`, increment `left` to consider larger numbers.
  - If `numbers[left] + numbers[right] > value`, decrement `right` to consider smaller numbers.
  - If `left >= right`, no such pair exists, return `false`.

  ```python
  class TwoSumSortedList:
      def __init__(self):
          self.numbers = []

      def add(self, number: int) -> None:
          import bisect
          bisect.insort(self.numbers, number) # Maintain sorted order

      def find(self, value: int) -> bool:
          left, right = 0, len(self.numbers) - 1
          while left < right:
              current_sum = self.numbers[left] + self.numbers[right]
              if current_sum == value:
                  return True
              elif current_sum < value:
                  left += 1
              else:
                  right -= 1
          return False
  ```

- **Time Complexity:**
  - `add`: O(n) in the worst case if using `bisect.insort`. If simply appending and sorting later, `add` is O(1) and a subsequent sort would be O(n log n).
  - `find`: O(n) in the worst case.
- **Space Complexity:** O(n) to store the numbers.
- **Pros:** Efficient `find` operation after the list is sorted.
- **Cons:** `add` operation can be relatively expensive if the list needs to be kept sorted continuously.

**Comparison of Approaches:**

| Approach    | `add` Time Complexity | `find` Time Complexity | Space Complexity | Pros                                      | Cons                                            |
| ----------- | --------------------- | ---------------------- | ---------------- | ----------------------------------------- | ----------------------------------------------- |
| Naive       | O(1)                  | O(n^2)                 | O(n)             | Simple to implement                       | Inefficient `find` for large datasets           |
| Set         | O(1)                  | O(n)                   | O(n)             | Faster `find` than naive                  | Still iterates through the list in `find`       |
| Hash Map    | O(1) (average)        | O(n)                   | O(n)             | Efficient `find`, handles duplicates well | Requires extra space for the hash map           |
| Sorted List | O(n) (with insertion) | O(n)                   | O(n)             | Efficient `find` after sorting            | `add` can be expensive to maintain sorted order |

**Choosing the Right Approach:**

The best approach depends on the frequency of `add` and `find` operations:

- If `add` operations are much more frequent than `find` operations, the naive approach might be acceptable for small datasets.
- If `find` operations are frequent, the **Hash Map** approach generally offers the best performance due to its O(1) average `add` and O(n) `find` time complexity (where n is the number of unique elements).
- The **Set** approach is a good alternative if you don't need to handle duplicates in a specific way during the `find` operation.
- The **Sorted List with Two Pointers** can be efficient for `find` if you can afford the cost of maintaining a sorted list during `add` operations. If additions happen in batches, you could append and sort less frequently.

Considering the constraints (at most 10^4 calls to `add` and `find`), the **Hash Map approach is likely the most practical and efficient solution** for this problem. It provides a good balance between the time complexities of the two operations.
