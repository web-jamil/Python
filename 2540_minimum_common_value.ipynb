**2540. Minimum Common Value**

**Easy**

**Companies**: Microsoft

Given two integer arrays nums1 and nums2, sorted in non-decreasing order, return the minimum integer common to both arrays. If there is no common integer amongst nums1 and nums2, return -1.

Note that an integer is said to be common to nums1 and nums2 if both arrays have at least one occurrence of that integer.

**Example 1:**

```python
Input: nums1 = [1,2,3], nums2 = [2,4]
Output: 2
```

**Explanation**: The smallest element common to both arrays is 2, so we return 2.

**Example 2:**

```python
Input: nums1 = [1,2,3,6], nums2 = [2,3,4,5]
Output: 2
```

**Explanation:** There are two common elements in the array 2 and 3 out of which 2 is the smallest, so 2 is returned.

**Constraints:**

- 1 <= nums1.length, nums2.length <= 105
- 1 <= nums1[i], nums2[j] <= 109
- Both nums1 and nums2 are sorted in non-decreasing order.


````python

'''
**Algorithm:**

1. Insert all elements of `nums1` into a hash set for O(1) lookups.
2. Iterate through each element in `nums2` and check if it exists in the set.
3. The first common element found is the smallest (since both arrays are sorted and we traverse `nums2` in order).
4. If no common element is found, return -1.

**Time Complexity:** O(m + n)  
**Space Complexity:** O(m) for the hash set.


'''

class Solution:
    def getCommon(self, nums1: List[int], nums2: List[int]) -> int:
        st = set(nums1)
        for num in nums2:
            if num in st:
                return num
        return -1
````

````python
'''
**Algorithm:**

1. Iterate through each element in `nums1` (the smaller array if we optimize, but here we use `nums1`).
2. For each element in `nums1`, perform a binary search in `nums2`.
3. The first common element found (smallest in `nums1` that exists in `nums2`) is the answer.
4. If no common element is found, return -1.

**Time Complexity:** O(m log n)  
**Space Complexity:** O(1)
'''
class Solution:
    def getCommon(self, nums1: List[int], nums2: List[int]) -> int:
        def binary_search(arr, target):
            left, right = 0, len(arr) - 1
            while left <= right:
                mid = left + (right - left) // 2
                if arr[mid] == target:
                    return True
                elif arr[mid] < target:
                    left = mid + 1
                else:
                    right = mid - 1
            return False
        
        for num in nums1:
            if binary_search(nums2, num):
                return num
        return -1
````

````python
'''

**Algorithm:**

1. Use two pointers `i` and `j` for `nums1` and `nums2` respectively.
2. Since both arrays are sorted, traverse both arrays:
   - If `nums1[i] == nums2[j]`, return the common element.
   - If `nums1[i] < nums2[j]`, increment `i` to move to a larger value in `nums1`.
   - Otherwise, increment `j` to move to a larger value in `nums2`.
3. If no common element is found, return -1.

**Time Complexity:** O(m + n)  
**Space Complexity:** O(1)



'''
class Solution:
    def getCommon(self, nums1: List[int], nums2: List[int]) -> int:
        i, j = 0, 0
        m, n = len(nums1), len(nums2)
        while i < m and j < n:
            if nums1[i] == nums2[j]:
                return nums1[i]
            elif nums1[i] < nums2[j]:
                i += 1
            else:
                j += 1
        return -1
````
