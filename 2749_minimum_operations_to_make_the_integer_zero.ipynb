**2749. Minimum Operations to Make the Integer Zero**

**Medium**

**Companies**

You are given two integers num1 and num2.

In one operation, you can choose integer i in the range [0, 60] and subtract 2i + num2 from num1.

Return the integer denoting the minimum number of operations needed to make num1 equal to 0.

If it is impossible to make num1 equal to 0, return -1.

**Example 1:**

```python
Input: num1 = 3, num2 = -2
Output: 3
```

**Explanation:** We can make 3 equal to 0 with the following operations:

- We choose i = 2 and subtract 22 + (-2) from 3, 3 - (4 + (-2)) = 1.
- We choose i = 2 and subtract 22 + (-2) from 1, 1 - (4 + (-2)) = -1.
- We choose i = 0 and subtract 20 + (-2) from -1, (-1) - (1 + (-2)) = 0.
  It can be proven, that 3 is the minimum number of operations that we need to perform.\

**Example 2:**

```python
Input: num1 = 5, num2 = 7
Output: -1
```

**Explanation:** It can be proven, that it is impossible to make 5 equal to 0 with the given operation.

**Constraints:**

- 1 <= num1 <= 109
- -109 <= num2 <= 109


````python
class Solution:
  def makeTheIntegerZero(self, num1: int, num2: int) -> int:
    # If k operations are used, num1 - [(num2 + 2^{i_1}) + (num2 + 2^{i_2}) +
    # ... + (num2 + 2^{i_k})] = 0. So, num1 - k * num2 = (2^{i_1} + 2^{i_2} +
    # ... + 2^{i_k}), where i_1, i_2, ..., i_k are in the range [0, 60].
    # Note that for any number x, we can use "x's bit count" operations to make
    # x equal to 0. Additionally, we can also use x operations to deduct x by
    # 2^0 (x times), which also results in 0.

    for ops in range(61):
      target = num1 - ops * num2
      if target.bit_count() <= ops <= target:
        return ops

    return -1
````

````python
from itertools import count
class Solution:
    def makeTheIntegerZero(self, num1: int, num2: int) -> int:
        for k in count(1):
            x = num1 - k * num2
            if x < 0:
                break
            if x.bit_count() <= k <= x:
                return k
        return -1
````

````python
class Solution:
    def makeTheIntegerZero(self, num1: int, num2: int) -> int:
        """
        Approach 1: Brute-force simulation with popcount check
        -----------------------------------------------
        We try increasing values of t starting from 0.
        For each t:
        - Compute val = num1 - t * num2
        - If val < 0, break (no need to continue)
        - If popcount(val) <= t and t <= val: return t
        - Else increment t

        Time Complexity: O(log(num1) * t)
        Space Complexity: O(1)
        """

        t = 0
        while True:
            val = num1 - t * num2
            if val < 0:
                return -1
            if bin(val).count('1') <= t <= val:
                return t
            t += 1

        return -1  # Should never reach here

````

````python

class Solution:
    def makeTheIntegerZero(self, num1: int, num2: int) -> int:
        """
        Approach 2: Optimized with strict upper bound on t
        --------------------------------------------------
        We try values of t from 1 to 36 (based on constraints and binary limits).
        For each t:
        - Compute val = num1 - t * num2
        - If val < 0, break (can't reduce num1 further)
        - If popcount(val) <= t and t <= val: return t

        Why 36?
        Because the maximum number of bits needed to represent any val (given 10^9 constraint)
        is around 36 (i.e., log2(10^9) ≈ 30, with a bit of buffer).

        Time Complexity: O(1), constant number of iterations
        Space Complexity: O(1)
        """

        for t in range(1, 37):
            val = num1 - t * num2
            if val < 0:
                return -1
            if bin(val).count('1') <= t <= val:
                return t

        return -1
````
