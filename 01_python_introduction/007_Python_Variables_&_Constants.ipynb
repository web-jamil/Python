# 1. Python Variables

A variable is a named location used to **store data in the memory**. Variable also known as **identifier** and used to hold value.
```python
>>>number = 90
```

# 2. Constants

A constant is a type of variable whose value cannot be changed. It is helpful to think of constants as containers that hold information which cannot be changed later.

You can think of constants as a bag to store some books which cannot be replaced once placed inside the bag.


### Assigning value to constant in Python

In Python, constants are usually declared and assigned in a module. Here, the module is a new file containing variables, functions, etc which is imported to the main file. Inside the module, constants are written in all capital letters and underscores separating the words.


#### Example 1: Declaring and assigning value to a constant

Create a **constant.py**:

```python
>>>PI = 3.14
>>>GRAVITY = 9.8
```

Create a **main.py**:

```python
>>>import constant
>>>print(constant.PI)
>>>print(constant.GRAVITY)

3.14
9.8
```

In the above program, we create a **constant.py** module file. Then, we assign the constant value to **`PI`** and **`GRAVITY`**. After that, we create a **main.py** file and import the **`constant`** module. Finally, we print the constant value.

> **Note**: In reality, we don't use constants in Python. Naming them in all capital letters is a convention to separate them from variables, however, it does not actually prevent reassignment.


## Rules and Naming Convention for Variables and constants

The examples you have seen so far have used **short**, terse variable names like m and n. But variable names can be more **verbose**. In fact, it is usually beneficial if they are because it makes the purpose of the variable more evident at first glance.

1. Constant and variable names should have a combination of letters in lowercase (a to z) or uppercase (**A to Z**) or digits (**0 to 9**) or an underscore **`_`**. For example:

```python
snake_case
MACRO_CASE
camelCase
CapWords
```

2. Create a name that makes sense. For example, **`vowel`** makes more sense than **`v`**.

3. If you want to create a variable name having two words, use underscore to separate them. For example:

```python
my_name
current_salary
```

4. Use capital letters possible to declare a constant. For example:

```python
PI
G
MASS
SPEED_OF_LIGHT
TEMP
```

5. Never use special symbols like **!**, **@**, **#**, **$** <b> % </b>, etc.

6. Don't start a variable name with a digit.

> **Note**: One of the additions to Python 3 was full Unicode support, which allows for **Unicode** characters in a variable name as well. You will learn about Unicode in greater depth in a future tutorial.


For example, all of the following are valid variable names:


````python
name = "Bob"
Age = 54
has_W2 = True
print(name, Age, has_W2)
````

But this one is not, **because a variable name can’t begin with a digit**:


````python
1099_filed = False    # cannot start name of a variable with a number.
````


There is nothing stopping you from creating two different variables in the same program called age and Age, or for that matter agE. But it is probably **ill-advised**. ### 6. **Variable Scope and Lifetime**

The scope of a variable refers to where the variable is accessible in the code, and the lifetime refers to how long the variable exists.

#### **Global Variables**

A variable defined outside any function or class is called a **global variable**. It is accessible throughout the program.

```python
x = 10  # Global variable

def print_value():
    print(x)

print_value()  # Output: 10
```

#### **Local Variables**

A variable defined inside a function is a **local variable**. It can only be accessed within the function where it is defined.

```python
def my_function():
    y = 20  # Local variable
    print(y)

my_function()  # Output: 20
print(y)        # Error: NameError: name 'y' is not defined
```

#### **Nonlocal Variables**

A **nonlocal variable** is used in nested functions to refer to variables in the nearest enclosing scope that is not global.

```python
def outer_function():
    x = 10
    def inner_function():
        nonlocal x
        x = 20
    inner_function()
    print(x)  # Output: 20

outer_function()
```

#### **Built-in Variables**

Python has a set of built-in variables that you can use directly, such as `__name__`, `__doc__`, `__file__`, etc.

```python
print(__name__)  # Output depends on how the script is run
```

### 7. **Reassigning Variables**

In Python, variables are dynamically typed, meaning you can reassign them to different data types during execution.

```python
x = 10        # x is an integer
x = "Hello"   # Now x is a string
x = 3.14      # Now x is a float
```

#### **Dynamic Typing in Python**

Python does not require variable type declaration before assignment. It infers the type at runtime.

```python
a = 5        # a is an integer
a = "Hello"  # a is now a string
```

### 8. **Constants and Variables**

Python does not have a constant type. However, a convention is used for naming constants: **ALL_CAPS**. This tells the programmer that the value should not be modified.

```python
PI = 3.14159  # This should be treated as a constant
```

### 9. **Advanced Topics**

#### **Variables in Functions (Argument Passing)**

When passing variables to functions, Python passes **references** to objects for mutable types (e.g., lists), and copies for immutable types (e.g., integers).

```python
def modify_value(x):
    x = 5
    return x

a = 10
print(modify_value(a))  # Output: 5
print(a)  # Output: 10 (a is unchanged)
```

#### **Variable Length Arguments**

You can pass a variable number of arguments to functions using `*args` (for non-keyword arguments) and `**kwargs` (for keyword arguments).

```python
def print_args(*args, **kwargs):
    print(args)   # tuple of arguments
    print(kwargs) # dictionary of keyword arguments

print_args(1, 2, 3, name="Alice", age=25)
```

#### **Mutable vs Immutable Variables**

In Python, **mutable** types (e.g., lists, dictionaries) can be modified after assignment, while **immutable** types (e.g., integers, strings, tuples) cannot.

```python
# Immutable: String
x = "hello"
x = x.upper()  # x now points to "HELLO"

# Mutable: List
y = [1, 2, 3]
y[0] = 100  # y is modified
```

### 1. **What Are Variables in Python?**

In programming, a **variable** is a symbolic name for a data value. Python is dynamically typed, meaning you don't need to specify the type of data the variable will hold when you declare it. A variable can hold values of various data types, including integers, strings, lists, and dictionaries.

---

### 2. **Python Variable Naming Rules**

#### **Rules for Identifiers**

In Python, **identifiers** are names used to identify variables, functions, classes, and other objects. Variable names must adhere to the following rules:

1. **Start with a letter or an underscore**: A variable name must begin with a letter (A-Z, a-z) or an underscore (`_`), but **not a number**.
   - Valid: `age`, `_name`, `my_var`
   - Invalid: `2nd_name`, `9name`
2. **Contain letters, numbers, or underscores**: After the first character, the name can contain letters, numbers, and underscores.

   - Valid: `var_1`, `my_var_2`
   - Invalid: `@my_var`, `#name`

3. **No reserved keywords**: You cannot use Python’s reserved keywords (also known as **keywords**) as variable names. These are words that are predefined in Python, such as `if`, `else`, `def`, `return`, etc.

   - Valid: `x`, `age`, `name`
   - Invalid: `def`, `class`, `try`, `else`

4. **Case sensitive**: Python is case-sensitive, meaning that `myvar`, `Myvar`, and `MYVAR` would be considered different variables.
   - `myVar` and `myvar` are two different identifiers.

#### **Python Reserved Keywords**

Reserved keywords are predefined words that have a special meaning in Python. They are used to define the syntax and structure of the language. These cannot be used as identifiers (variable names).

**Examples of reserved keywords**:

```python
False, None, True, and, as, assert, async, await, break, class, continue, def,
del, elif, else, except, finally, for, from, global, if, import, in, is, lambda,
nonlocal, not, or, pass, raise, return, try, while, with, yield
```

You can check the full list of reserved keywords using the `keyword` module:

```python
import keyword
print(keyword.kwlist)
```

### 3. **Python Variable Naming Conventions**

#### **PEP 8 Guidelines**

PEP 8 is the style guide for Python code. It defines best practices for writing clean and readable Python code, including variable naming conventions. Some key points from PEP 8 are:

1. **Use lowercase for variable names**: By convention, variable names should be written in lowercase letters, with words separated by underscores (`_`).

   ```python
   my_variable = 10
   total_price = 25.5
   ```

2. **Use underscores to separate words**: In variable names with multiple words, use underscores to separate them. This is called **snake_case**.

   ```python
   first_name = "John"
   student_age = 20
   ```

3. **Avoid using single character names**: Avoid using single characters for variable names, unless they are counters in loops or used in a very limited scope.

   ```python
   count = 10  # Bad example: don't use 'c' or 'x' without context
   ```

4. **Use meaningful names**: The name of the variable should describe the value it holds.

   ```python
   user_name = "Alice"  # Good example
   temp = "Bob"         # Bad example: unclear what "temp" represents
   ```

5. **Constants should be in all uppercase**: Constants are variables whose value should not change. They are usually written in all uppercase letters with underscores separating words.
   ```python
   MAX_RETRIES = 5
   PI = 3.14159
   ```

#### **Case Styles**

There are several common naming conventions used in Python, and the choice of style often depends on the context:

- **snake_case**: Used for variable names, function names, and method names.

  ```python
  my_variable = 10
  def my_function():
      pass
  ```

- **camelCase**: Often used in other languages, but not recommended in Python. Some Python libraries may use it for class names or other objects.

  ```python
  myVariable = 10  # Not recommended in Python
  ```

- **PascalCase**: Used for class names in Python.

  ```python
  class MyClass:
      pass
  ```

- **UPPERCASE**: Used for constants.
  ```python
  MAX_CONNECTIONS = 100
  ```

### 4. **Best Practices for Naming Variables**

1. **Descriptive and Meaningful Names**: Always choose descriptive names that convey the purpose of the variable.

   ```python
   user_age = 30      # Good
   a = 30             # Bad
   ```

2. **Avoid Abbreviations**: Abbreviations can make your code hard to understand, especially for others reading your code. Avoid abbreviating variable names unless they are widely recognized.

   ```python
   total_amount = 1000   # Good
   amt = 1000            # Okay for short names but avoid excessive abbreviation
   ```

3. **Don’t Overuse Underscores**: While underscores are used to separate words in variable names (snake_case), don't overdo it with too many underscores.

   ```python
   # Good: customer_name
   # Bad: customer__name  (Extra underscores make the name harder to read)
   ```

4. **Avoid Using Python Built-ins**: Do not use Python's built-in names for variable names. For example, don't name a variable `list`, `str`, or `int`.

5. **Use Consistent Naming**: Stick to one naming convention throughout your code to maintain consistency. For example, if you choose snake_case for variable names, avoid switching to camelCase in the middle of the code.

### 5. **Advanced Variable Naming Tips**

#### **Variable Naming for Global Scope**

When using variables in a global scope, try to minimize their usage to avoid accidental changes in the global state. It’s recommended to use lowercase names for global variables, and constants should be in uppercase.

```python
GLOBAL_VARIABLE = 10
```

#### **Avoiding Name Collisions**

Python doesn't have block-level scoping like some other languages. Be mindful of naming variables that may collide with global or built-in variable names. It's a good idea to avoid using names like `str`, `list`, `int`, etc.

```python
# Avoid this:
list = [1, 2, 3]

# Instead, use:
my_list = [1, 2, 3]
```

#### **Meaningful Contextual Naming**

Sometimes, especially in larger programs, it's good to have more specific names that describe **what** the variable holds within the specific context of your program. For example:

- **Use the context of the variable to make the name more descriptive**:

  ```python
  customer_email = "john.doe@example.com"   # Clear
  email = "john.doe@example.com"             # Less clear in certain contexts
  ```

- **Use naming schemes based on data types or structures**:
  ```python
  user_data = {"name": "Alice", "age": 25}  # For a dictionary of user data
  item_list = ["apple", "banana", "cherry"]  # For a list of items
  ```

## **4. Advanced Naming Conventions**

### **1. Private Variables**

- Prefix a variable name with a single underscore (`_`) to indicate that it is intended for internal use.
- Example: `_internal_variable`.

#### **Example**

```python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance  # Private variable

    def deposit(self, amount):
        self._balance += amount

account = BankAccount(100)
account.deposit(50)
print(account._balance)  # Output: 150 (accessible but not recommended)
```

---

### **2. Name Mangling**

- Prefix a variable name with double underscores (`__`) to make it private and avoid name clashes in subclasses.
- Python internally changes the name to `_ClassName__variable`.

#### **Example**

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Name mangling

    def deposit(self, amount):
        self.__balance += amount

account = BankAccount(100)
account.deposit(50)
print(account._BankAccount__balance)  # Output: 150 (name mangling)
```

---

### **3. Special Methods**

- Use double underscores at the beginning and end of variable names for special methods.
- Example: `__init__`, `__str__`.

#### **Example**

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Person: {self.name}"

person = Person("Alice")
print(person)  # Output: Person: Alice
```

### **4. Dynamic Variable Names**

- Use dictionaries or the `globals()`/`locals()` functions to create dynamic variable names.

#### **Example**

```python
# Using a dictionary
data = {}

print(data["name"])  # Output: Alice

# Using globals()
globals()["age"] = 25
print(age)  # Output: 25
```
