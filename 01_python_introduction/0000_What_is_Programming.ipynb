### 1. **Introduction to Programming**

#### 1.1. What is Programming?

Programming is the act of instructing a computer on how to perform specific tasks. It involves writing code in a programming language that the computer can understand. The code specifies the operations the computer should carry out and the data it should manipulate.

### 2. **Basic Programming Concepts**

#### 2.1. Syntax and Semantics

- **Syntax** refers to the structure or grammar of the programming language (the rules that define how code must be written).
- **Semantics** refers to the meaning of the code (what the code does when executed).

For example, in Python:

```python
x = 10  # Syntax: valid assignment, Semantics: assigns 10 to x
```

#### 2.2. Variables and Data Types

- **Variables** are used to store values.
- **Data Types** define the kind of data a variable can hold, such as integers (`int`), floating-point numbers (`float`), strings (`str`), and booleans (`bool`).

Example:

```python
x = 5        # int
y = 3.14     # float
name = "Alice" # string
is_active = True # boolean
```

#### 2.3. Operators

Operators are used to perform operations on variables and values:

- **Arithmetic Operators**: `+`, `-`, `*`, `/`, `%`
- **Comparison Operators**: `==`, `!=`, `<`, `>`, `<=`, `>=`
- **Logical Operators**: `and`, `or`, `not`

Example:

```python
a = 10
b = 5
print(a + b)  # Output: 15 (Arithmetic Operator)
```

#### 2.4. Input and Output

- **Input**: Taking data from the user using `input()`.
- **Output**: Displaying data using `print()`.

Example:

```python
name = input("Enter your name: ")  # Taking user input
print(f"Hello, {name}!")  # Output the result
```

#### 2.5. Control Flow: If-else, Loops

- **If-else Statements** allow branching based on conditions.
- **Loops** allow repeating a block of code multiple times. Python supports `for` loops (for iterating over sequences) and `while` loops (until a condition is met).

Example:

```python
x = 10
if x > 5:
    print("x is greater than 5")
else:
    print("x is not greater than 5")

for i in range(3):
    print(i)  # Output: 0, 1, 2
```

### 3. **Functions and Modular Programming**

#### 3.1. Functions

Functions allow you to group related code into reusable blocks. They are defined using the `def` keyword.

Example:

```python
def greet(name):
    return f"Hello, {name}!"
```

#### 3.2. Arguments and Return Values

Functions can accept **arguments** (input values) and return values using the `return` statement.

Example:

```python
def add(a, b):
    return a + b

result = add(3, 5)  # Output: 8
```

#### 3.3. Recursion

A function is **recursive** if it calls itself to solve a problem.

Example (factorial calculation):

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

print(factorial(5))  # Output: 120
```

#### 3.4. Scope and Lifetime of Variables

- **Local variables**: Variables defined inside a function.
- **Global variables**: Variables defined outside any function and accessible from anywhere in the program.

### 5. **Object-Oriented Programming (OOP)**

#### 5.1. Classes and Objects

- **Class**: A blueprint for creating objects (instances).
- **Object**: An instance of a class.

Example:

```python
class Dog:
    def __init__(self, name):
        self.name = name

dog1 = Dog("Buddy")
print(dog1.name)  # Output: Buddy
```

#### 5.2. Inheritance and Polymorphism

- **Inheritance**: Allows a new class to inherit properties and behaviors of an existing class.
- **Polymorphism**: Enables objects to be treated as instances of their parent class.

Example:

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Woof")

dog = Dog()
dog.speak()  # Output: Woof
```

## **4. Advanced Concepts**

### **1. Generators and Iterators**

- **Generators**: Functions that produce a sequence of values lazily using `yield`.
- **Iterators**: Objects that allow traversal through a sequence.

#### **Example**

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

for num in fibonacci(5):
    print(num)  # Output: 0 1 1 2 3
```

---

### **2. Decorators**

- Functions that modify the behavior of other functions.

#### **Example**

```python
def my_decorator(func):
    def wrapper():
        print("Before the function call")
        func()
        print("After the function call")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
# Output:
# Before the function call
# Hello!
# After the function call
```
