## Type Conversion

The process of converting the value of one data type (integer, string, float, etc.) to another data type is called type conversion. Python has two types of type conversion.

1. **Implicit Type Conversion**
2. **Explicit Type Conversion**


### 1. Implicit Type Conversion

In Implicit type conversion, Python automatically converts one data type to another data type. This process doesn't need any user involvement.

Let's see an example where Python promotes the conversion of the lower data type (integer) to the higher data type (float) to avoid data loss.


````python
# Example 1: Converting integer to float

num_int = 123
num_flo = 1.23

num_new = num_int + num_flo

print("datatype of num_int:",type(num_int))
print("datatype of num_flo:",type(num_flo))

print("Value of num_new:",num_new)
print("datatype of num_new:",type(num_new))
````

**Explanation**:

In the above program,

1. We add two variables **`num_int`** and **`num_flo`**, storing the value in **`num_new`**.
2. We will look at the data type of all three objects respectively.
3. In the output, we can see the data type of **`num_int`** is an **`integer`** while the data type of **`num_flo`** is a **`float`**.
4. Also, we can see the **`num_new`** has a **`float`** data type because Python always converts smaller data types to larger data types to avoid the loss of data.


Now, let's try adding a string and an integer, and see how Python deals with it.


````python
# Example 2: Addition of string(higher) data type and integer(lower) datatype

num_int = 123
num_str = "456"

print("Data type of num_int:",type(num_int))
print("Data type of num_str:",type(num_str))

print(num_int+num_str)  # ERROR! cannot add two different data types (int and string)
````

**Explanation**:

In the above program,

1. We add two variables **`num_int`** and **`num_str`**.
2. As we can see from the output, we got **`TypeError`**. Python is not able to use Implicit Conversion in such conditions.
3. However, Python has a solution for these types of situations which is known as Explicit Conversion.


### 2. Explicit Type Conversion

In Explicit Type Conversion, users convert the data type of an object to required data type. We use the predefined functions like **`int()`**, **`float()`**, **`str()`**, etc to perform explicit type conversion.

This type of conversion is also called typecasting because the user casts (changes) the data type of the objects.

**Syntax:**

```python
<required_datatype>(expression)
```

Typecasting can be done by assigning the required data type function to the expression.


````python
# int to float
num_int = 10
print('num_int',num_int)         # 10
num_float = float(num_int)
print('num_float:', num_float)   # 10.0
````

````python
# float to int
gravity = 9.81
print(int(gravity))             # 9
````

````python
# Example 3: Addition of string and integer using explicit conversion

num_int = 123
num_str = "456"       # there is a number inside " " so I CAN change it to integer
#num_str = "jason"    # there is a name inside " " so I cannot change to integer

print("Data type of num_int:",type(num_int))
print("Data type of num_str before Type Casting:",type(num_str))

num_str = int(num_str) # change from "string" data type to integer datatype
print("Data type of num_str after Type Casting:",type(num_str))

num_sum = num_int + num_str  # Now we can add two same data types (int and int)

print("Sum of num_int and num_str:",num_sum)
print("Data type of the sum:",type(num_sum))
````

**Explanation**:

In the above program,

1. We add **`num_str`** and **`num_int`** variable.
2. We converted **`num_str`** from string(higher) to integer(lower) type using **`int()`** function to perform the addition.
3. After converting **`num_str`** to an integer value, Python is able to add these two variables.
4. We got the **`num_sum`** value and data type to be an integer.

### **Implicit Type Conversion in Python: From Basic to Advanced**

#### **Introduction:**

Implicit type conversion, also known as **type coercion**, occurs when Python automatically converts one data type to another without explicitly being told to do so by the programmer. Python performs this conversion to prevent data loss, maintain consistency, and ensure the program can execute without errors when performing operations with different types of data.

For example, when performing operations involving different types of numbers (integers and floats), Python automatically converts integers to floats to prevent truncation or loss of precision.

---

### **1. Basic Implicit Type Conversion**

Implicit conversion typically occurs in situations where Python can automatically decide that it’s safe to convert one type to another, usually from a lower data type to a higher data type.

#### **Basic Examples:**

```python
# Integer and Float
x = 10      # Integer
y = 2.5     # Float
result = x + y   # The integer is automatically converted to float
print(result)  # Output: 12.5
```

In the above example, the integer `10` is automatically converted to a float before being added to `2.5`. The result is a float (`12.5`).

---

### **2. Automatic Conversion Between Numbers**

Python automatically converts between numeric types when necessary, like between `int` (integer) and `float` (floating-point number), `int` and `complex`, and so on.

#### **Integer to Float Conversion:**

```python
x = 5         # Integer
y = 2.0       # Float
result = x + y   # Python automatically converts the integer to float
print(result)  # Output: 7.0
```

Here, Python automatically converts `x` (an integer) to a float so that the addition can happen safely without any data loss.

#### **Integer to Complex Number Conversion:**

If you perform operations between an integer and a complex number, Python converts the integer to a complex number for the operation.

```python
x = 3           # Integer
y = 1 + 2j      # Complex number
result = x + y  # Python automatically converts the integer to complex
print(result)  # Output: (3+2j)
```

Python automatically converts `x` (an integer) to a complex number with zero imaginary part to perform the addition.

---

### **3. Implicit Type Conversion in String Operations**

#### **String and Integer:**

If you try to add an integer to a string directly, you will get an error. However, in certain contexts like concatenating strings, Python may perform implicit conversions. But in arithmetic operations like addition or multiplication, Python will not implicitly convert a string to a number. You’ll need to perform explicit conversion using functions like `int()`, `str()`, etc.

```python
x = "Age: "
y = 25
result = x + str(y)  # Explicit conversion of integer to string
print(result)  # Output: Age: 25
```

In this case, you explicitly convert `y` to a string before concatenation. Python will not automatically convert the integer `25` to a string.

---

### **4. Implicit Conversion with Booleans**

Python treats `True` as `1` and `False` as `0`. When you use booleans in arithmetic operations, Python automatically converts them to integers.

#### **Boolean to Integer:**

```python
x = True   # Boolean
y = 10     # Integer
result = x + y   # Boolean `True` is treated as 1
print(result)  # Output: 11
```

Here, `True` is implicitly converted to `1` and added to `10`. The result is `11`.

#### **Boolean in Arithmetic Operations:**

```python
x = False  # Boolean
y = 100    # Integer
result = x + y   # Boolean `False` is treated as 0
print(result)  # Output: 100
```

In this case, `False` is implicitly converted to `0`, and the result is `100`.

---

### **5. Collections and Iterables:**

Python performs implicit type conversion when working with collections (e.g., lists, sets, tuples) and iterables. However, you need to be careful since Python won't automatically convert data types like `str` to `int` or vice versa.

#### **List + Integer Conversion:**

If you try to multiply a list by an integer, Python will repeat the list instead of performing a mathematical multiplication.

```python
x = [1, 2, 3]  # List
y = 3          # Integer
result = x * y  # Python repeats the list 3 times
print(result)  # Output: [1, 2, 3, 1, 2, 3, 1, 2, 3]
```

Python implicitly converts the integer `3` into a repetition factor and applies it to the list.

---

### **6. Advanced Examples:**

#### **Implicit Type Conversion with Functions**

When you pass mixed types of arguments into functions that accept multiple arguments, Python will perform implicit conversions to make the argument types compatible.

```python
def add(x, y):
    return x + y

# Function works with mixed types
result = add(5, 3.5)
print(result)  # Output: 8.5
```

Here, Python automatically converts `5` (integer) to `5.0` (float) before performing the addition.

#### **Combining Lists and Tuples**

Python can also combine different collection types, like lists and tuples, during operations like concatenation.

```python
x = [1, 2, 3]    # List
y = (4, 5, 6)    # Tuple
result = x + list(y)  # Convert tuple to list and concatenate
print(result)  # Output: [1, 2, 3, 4, 5, 6]
```

In this case, the tuple `y` is implicitly converted to a list before concatenation.

---

### **7. Implicit Type Conversion in Operations with Objects**

If you define custom classes and overload operators (e.g., `+`, `-`, etc.), Python will use implicit type conversion if necessary.

#### **Custom Example with Operator Overloading:**

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

point1 = Point(1, 2)
point2 = Point(3, 4)

result = point1 + point2  # The __add__ method is implicitly called
print(result)  # Output: (4, 6)
```

Here, the `+` operator implicitly calls the `__add__()` method of the `Point` class, allowing the addition of two `Point` objects.

---

### **8. Limitations of Implicit Type Conversion:**

1. **Incompatibility in Operations:**

   - Python will not implicitly convert data types like a string to an integer when performing arithmetic operations.

   ```python
   x = "10"
   y = 5
   # This will raise an error:
   # TypeError: can only concatenate str (not "int") to str
   result = x + y
   ```

   To handle this, explicit type conversion must be used (`int(x)`).

2. **Non-Automatic Conversion Between Unrelated Types:**
   - You cannot implicitly convert a list to a set, or a tuple to a string.
   - This would raise a `TypeError` without explicit conversion.

---

### **9. Conclusion:**

- **Implicit Type Conversion** is automatically performed by Python when necessary, converting smaller data types to larger data types (e.g., from `int` to `float`).
- **Automatic Promotion**: Python promotes integers to floats, or booleans to integers, to maintain correctness in operations involving mixed data types.
- **Explicit Conversion**: Sometimes, Python will require explicit type conversion if the types are incompatible or cannot be converted automatically.
- **Limitations**: Implicit conversions do not occur when types are incompatible, and Python does not convert types between very different data types (e.g., string to integer).

Implicit type conversion is a fundamental concept in Python that helps developers write more concise and error-free code, but it’s essential to understand when Python does and does not apply automatic conversions.

### **Explicit Type Conversion in Python: From Basic to Advanced**

#### **Introduction:**

Explicit type conversion, also known as **type casting**, is the process of manually converting one data type into another. Unlike implicit type conversion, where Python automatically converts types when needed, explicit type conversion requires the programmer to explicitly tell Python to convert a value from one type to another. This is done using **Python's built-in functions** such as `int()`, `float()`, `str()`, etc.

---

### **1. Basic Concept of Explicit Type Conversion**

Explicit type conversion is necessary when:

- You need to change the type of a value to perform a specific operation.
- You want to handle incompatible types manually.
- You want to ensure type consistency.

The common functions used for explicit type conversion in Python are:

- `int()`: Converts a value to an integer type.
- `float()`: Converts a value to a float type.
- `str()`: Converts a value to a string type.
- `bool()`: Converts a value to a boolean type.
- `list()`: Converts a value to a list.
- `tuple()`: Converts a value to a tuple.
- `set()`: Converts a value to a set.
- `dict()`: Converts a value to a dictionary (from a pair of key-value tuples).

---

### **2. Converting Between Numbers**

#### **Converting to Integer (`int()`)**:

The `int()` function is used to convert a value to an integer type. The value can be a float, a string representation of a number, or a boolean.

```python
# Converting float to integer
x = 3.14
y = int(x)
print(y)  # Output: 3

# Converting string to integer
x = "123"
y = int(x)
print(y)  # Output: 123

# Converting boolean to integer
x = True
y = int(x)
print(y)  # Output: 1

x = False
y = int(x)
print(y)  # Output: 0
```

- The `int()` function truncates the decimal part when converting from a float to an integer.

#### **Converting to Float (`float()`)**:

The `float()` function is used to convert a value to a floating-point number.

```python
# Converting integer to float
x = 5
y = float(x)
print(y)  # Output: 5.0

# Converting string to float
x = "123.45"
y = float(x)
print(y)  # Output: 123.45
```

- The `float()` function also works with integers and strings containing decimal points.

---

### **3. Converting Between Strings and Numbers**

#### **Converting to String (`str()`)**:

The `str()` function is used to convert a number, boolean, or other objects to a string.

```python
# Converting integer to string

y = str(x)
print(y)  # Output: "123"

# Converting float to string
x = 3.14
y = str(x)
print(y)  # Output: "3.14"

```

- The `str()` function converts any object into a string, making it useful for formatting output or concatenating with other strings.

#### **Converting a String to a Number**:

Strings representing valid numbers can be explicitly converted to integers or floats.

