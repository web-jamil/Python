## Command vs. Edit Modes

But first...something key to be aware of: Jupyter Notebooks have two different keyboard input modes:

1. **Command mode** - binds the keyboard to notebook level actions. Indicated by a grey cell border with a <font color=teal>**blue**</font> left margin.
2. **Edit mode** - when you're typing in a cell. Indicated by a green cell border with a <font color=green>**green**</font> left margin.

### Tables

Tables aren't part of the core Markdown spec, but they are part of GFM and Markdown Here supports them. They are an easy way of adding tables to your email -- a task that would otherwise require copy-pasting from another application.


Colons can be used to align columns.

| Tables   |      Are      |   Cool |
| :------- | :-----------: | -----: |
| col 3 is | right-aligned | \$1900 |
| col 2 is |   centered    |   \$12 |
| col 1 is | left-aligned  |    \$1 |

The outer pipes **`|`** are optional, and you don't need to make the raw Markdown line up prettily. You can also use inline Markdown.

| Markdown | Less      | Pretty     |
| -------- | --------- | ---------- |
| _Still_  | `renders` | **nicely** |
| 1        | 2         | 3          |

## **1. Basic Keyboard Shortcuts**

### **1. Command Mode vs. Edit Mode**

- **Command Mode**: The notebook is in command mode when no cell is actively being edited. The cell border is blue.
- **Edit Mode**: The notebook is in edit mode when a cell is being edited. The cell border is green.

### **2. Switching Between Modes**

- **Enter Edit Mode**: Press `Enter` or double-click a cell.
- **Enter Command Mode**: Press `Esc`.

---

### 1. **Introduction to Jupyter Notebooks**

#### What is Jupyter?

Jupyter Notebook is an open-source web application that allows you to create and share documents that contain live code, equations, visualizations, and narrative text. It is widely used for data analysis, machine learning, and scientific research.

#### Why Use Keyboard Shortcuts?

Keyboard shortcuts allow you to navigate and control your Jupyter Notebooks without needing to rely on the mouse. This speeds up workflows and makes data analysis tasks more efficient.

---

### 2. **Basic Shortcuts (Command Mode vs Edit Mode)**

Jupyter Notebooks have **two modes** that influence the available shortcuts:

- **Command Mode** (press `Esc`): In this mode, the notebook behaves like a command-line interface, allowing you to perform actions on the entire cell (e.g., moving, deleting, running).
- **Edit Mode** (press `Enter`): In this mode, you can edit the content of the cell (code or text).

---

### Common Shortcuts in **Command Mode** (Press `Esc` to activate)

- **Enter Edit Mode**: `Enter`
- **Run Cell**: `Shift + Enter` (Runs the current cell and moves to the next one)
- **Run Cell and Insert Below**: `Alt + Enter` (Runs the current cell and inserts a new cell below)
- **Run Cell and Stay in Current Cell**: `Ctrl + Enter`
- **Insert Cell Above**: `A`
- **Insert Cell Below**: `B`
- **Delete Cell**: `D, D` (Press `D` twice rapidly)
- **Undo Delete**: `Z`
- **Copy Cell**: `C`
- **Cut Cell**: `X`
- **Paste Cell Below**: `V`
- **Paste Cell Above**: `Shift + V`
- **Merge Cells**: `Shift + M` (Merges the selected cells)
- **Split Cell**: `Ctrl + Shift + -` (Splits a cell into two at the cursor position)
- **Change Cell Type to Code**: `Y`
- **Change Cell Type to Markdown**: `M`
- **Change Cell Type to Raw NBConvert**: `R`
- **Toggle Cell Output**: `O`
- **Toggle Line Numbers in Code Cells**: `L`
- **Move Cell Up**: `K` or `Up Arrow`
- **Move Cell Down**: `J` or `Down Arrow`
- **Save Notebook**: `Ctrl + S`
- **Interrupt Kernel**: `I, I` (Press `I` twice rapidly)
- **Restart Kernel**: `0, 0` (Press `0` twice rapidly)
- **Show Command Palette**: `Cmd + Shift + P` (Mac) or `Ctrl + Shift + P` (Windows/Linux)
- **Open Console**: `Ctrl + Shift + C`

---

### Common Shortcuts in **Edit Mode** (Press `Enter` to activate)

- **Autocomplete**: `Tab` (Autocomplete variables, functions, or imports)
- **Indent Code**: `Tab` (Indent selected code)
- **Unindent Code**: `Shift + Tab` (Unindent selected code)
- **Find and Replace**: `Ctrl + F` (Find text in the current cell), `Ctrl + H` (Replace)
- **Undo**: `Ctrl + Z`
- **Redo**: `Ctrl + Shift + Z`
- **Go to Beginning of Line**: `Home`
- **Go to End of Line**: `End`
- **Move Cursor to Previous Word**: `Ctrl + Left Arrow`
- **Move Cursor to Next Word**: `Ctrl + Right Arrow`
- **Delete Word Before Cursor**: `Ctrl + Backspace`
- **Delete Word After Cursor**: `Ctrl + Delete`
- **Cut Text**: `Ctrl + X`
- **Copy Text**: `Ctrl + C`
- **Paste Text**: `Ctrl + V`

---

### 3. **Cell Navigation and Management**

- **Move to Next Cell**: `Down Arrow` or `J`
- **Move to Previous Cell**: `Up Arrow` or `K`
- **Select Multiple Cells**: `Shift + Down Arrow` / `Shift + Up Arrow`
- **Delete Multiple Cells**: Select multiple cells using `Shift + Up Arrow` / `Shift + Down Arrow`, then press `D, D`
- **Select All Cells**: `Ctrl + A` (In Command Mode)

---

### 4. **Markdown and Code Cell Operations**

- **Switch Cell Type to Code**: `Y`
- **Switch Cell Type to Markdown**: `M`
- **Insert Markdown in the Cell**: To add text, headers, lists, or even LaTeX equations inside a Markdown cell. For example:
  - `# Header 1`
  - `## Header 2`
  - `* Bullet point`
  - `**Bold Text**`
  - `$$ Equation $$`
- **Render Markdown Cell**: `Shift + Enter` (To see the formatted Markdown output)

---

### 5. **Advanced Shortcuts**

- **Find and Replace in Entire Notebook**: `Ctrl + Shift + F`
- **Kernel Restart and Clear Output**: `Ctrl + .`
- **Clear All Output**: From the Jupyter Notebook menu: `Cell` > `All Output` > `Clear`
- **Toggle Cell Editor**: `Ctrl + Shift + -` (Split cell)
- **Toggle Auto Save**: Jupyter automatically saves notebooks, but you can turn off auto-saving through the File menu.

---

### 6. **Shortcut Customization and Tips**

#### How to View and Customize Shortcuts

1. **View Current Shortcuts**:
   - In Command Mode, click on **Help** → **Keyboard Shortcuts**. This will display all the available shortcuts in a pop-up window.
2. **Customize Shortcuts**:
   - You can customize Jupyter's keyboard shortcuts by modifying the configuration file (`jupyter_notebook_config.py`). This requires some knowledge of Jupyter's configuration system, and you can add custom shortcuts by using the Jupyter extension system.

### Practical Usage Tips

1. **Use Jupyter's Command Mode** as much as possible to quickly perform cell-related operations like moving, deleting, copying, and running cells.
2. **Combine Command Mode Shortcuts with Edit Mode**: You can navigate the notebook with command-mode shortcuts and edit code with edit-mode shortcuts.

3. **Master Markdown for Documentation**: Use Markdown cells to keep your code organized and well-documented. Use LaTeX equations to present formulas clearly.

4. **Leverage Code Completion** (`Tab` key): This is incredibly useful for writing code quickly, as it auto-completes function names, variable names, and even imports.

5. **Restart the Kernel** (`0, 0`) after making significant changes or when running into memory issues to clear up any issues with long-running computations.
