## What is Name in Python?

If you have ever read 'The Zen of Python' (type **`import this`** in the Python interpreter), the last line states, **Namespaces are one honking great idea -- let's do more of those!** So what are these mysterious namespaces? Let us first look at what name is.

Name (also called identifier) is simply a name given to objects. Everything in Python is an **[object](https://github.com/milaan9/06_Python_Object_Class/blob/main/002_Python_Classes_and_Objects.ipynb)**. Name is a way to access the underlying object.

For example, when we do the assignment **`a = 2`**, **`2`** is an object stored in memory and **`a`** is the name we associate it with. We can get the address (in RAM) of some object through the **[built-in function](https://github.com/milaan9/04_Python_Functions/tree/main/002_Python_Functions_Built_in)** **`id()`**. Let's look at how to use it.


````python
# Note: You may get different values for the id

a = 9
print('id(9) =', id(9))

print('id(a) =', id(a))
````

Here, both refer to the same object **`9`**, so they have the same **`id()`**. Let's make things a little more interesting.


````python
# Note: You may get different values for the id

a = 2
print('id(a) =', id(a))

a = a+1
print('id(a) =', id(a))

print('id(3) =', id(3))

b = 2
print('id(b) =', id(b))
print('id(2) =', id(2))
````

Initially, an object **`2`** is created and the name a is associated with it, when we do **`a = a+1`**, a new object **`3`** is created and now a is associated with this object.

Note that **`id(a)`** and **`id(3)`** have the same values.

Furthermore, when **`b = 2`** is executed, the new name **`b`** gets associated with the previous object **`2`**.

This is efficient as Python does not have to create a new duplicate object. This dynamic nature of name binding makes Python powerful; a name could refer to any type of object.

All these are valid and **`a`** will refer to three different types of objects in different instances. **[Functions](https://github.com/milaan9/04_Python_Functions/blob/main/001_Python_Functions.ipynb)** are objects too, so a name can refer to them as well.


````python
def printHello():
    print("Hello")

a = printHello

a()
````

The same name **`a`** can refer to a function and we can call the function using this name.


## What is a Namespace in Python?

Now that we understand what names are, we can move on to the concept of namespaces.

To simply put it, a namespace is a collection of names.

In Python, you can imagine a namespace as a mapping of every name you have defined to corresponding objects.

Different namespaces can co-exist at a given time but are completely isolated.

A namespace containing all the built-in names is created when we start the Python interpreter and exists as long as the interpreter runs.

This is the reason that built-in functions like **`id()`**, **`print()`** etc. are always available to us from any part of the program. Each **[module](https://github.com/milaan9/04_Python_Functions/blob/main/007_Python_Function_Module.ipynb)** creates its own global namespace.

These different namespaces are isolated. Hence, the same name that may exist in different modules do not collide.

Modules can have various functions and classes. A local namespace is created when a function is called, which has all the names defined in it. Similar, is the case with class. Following diagram may help to clarify this concept.

## Python Variable Scope

Although there are various unique namespaces defined, we may not be able to access all of them from every part of the program. The concept of scope comes into play.

A scope is the portion of a program from where a namespace can be accessed directly without any prefix.

At any given moment, there are at least three nested scopes.

1. Scope of the current function which has local names
2. Scope of the module which has global names
3. Outermost scope which has built-in names

When a reference is made inside a function, the name is searched in the local namespace, then in the global namespace and finally in the built-in namespace.

If there is a function inside another function, a new scope is nested inside the local scope.


Example of Scope and Namespace in Python

```python
>>>def outer_function():
>>>    b = 30
>>>    def inner_func():
>>>        c = 60

>>>a = 90
```

Here, the variable **`a`** is in the global namespace. Variable **`b`** is in the local namespace of **`outer_function()`** and **`c`** is in the nested local namespace of **`inner_function()`**.

When we are in **`inner_function()`**, **`c`** is local to us, **`b`** is nonlocal and **`a`** is global. We can read as well as assign new values to **`c`** but can only read **`b`** and **`a`** from **`inner_function()`**.

If we try to assign as a value to **`b`**, a new variable **`b`** is created in the local namespace which is different than the nonlocal **`b`**. The same thing happens when we assign a value to **`a`**.

However, if we declare **`a`** as global, all the reference and assignment go to the global **`a`**. Similarly, if we want to rebind the variable **`b`**, it must be declared as nonlocal. The following example will further clarify this.


````python
def outer_function():
    a = 30

    def inner_function():
        a = 60
        print('a =', a)

    inner_function()
    print('a =', a)


a = 90
outer_function()
print('a =', a)
````

In this program, three different variables **`a`** are defined in separate namespaces and accessed accordingly. While in the following program,


````python
def outer_function():
    global a
    a = 30

    def inner_function():
        global a
        a = 60
        print('a =', a)

    inner_function()
    print('a =', a)


a = 90
outer_function()
print('a =', a)
````

Here, all references and assignments are to the global **`a`** due to the use of keyword **`global`**.


### **Python Namespace and Scoping: From Basic to Advanced**

In Python, **namespace** and **scoping** are fundamental concepts that determine the context in which variables and functions are accessible. Understanding these concepts is crucial for writing efficient and bug-free code.

---

### **1. What is a Namespace?**

A **namespace** in Python is a container that holds a collection of objects (like variables, functions, classes, etc.) and their names (also called identifiers). Each object in Python is associated with a unique name within a particular namespace.

In simple terms, a **namespace** is a collection of variable names and their associated values.

#### **Types of Namespaces**

There are several types of namespaces in Python:

- **Built-in Namespace**: Contains all the built-in objects, functions, and exceptions (e.g., `len()`, `print()`, `Exception`). This namespace is available at all times and cannot be modified.
- **Global Namespace**: Contains names that are defined at the level of the main program. For example, variables or functions defined outside of any function or class.

- **Local Namespace**: Refers to names that are defined within a function or method. Each function has its own local namespace.

- **Enclosing Namespace**: This exists when a function is nested inside another function, which creates an enclosing scope for the inner function.

#### **Example:**

```python
# Built-in Namespace
print(len("Hello"))  # len is built-in

# Global Namespace
a = 10  # a is defined in the global namespace

def func():
    # Local Namespace
    b = 20  # b is defined in the local namespace of func()
    print(a)  # Can access global namespace
    print(b)  # Can access local namespace

func()
```

In this example:

- `a` is part of the global namespace.
- `b` is part of the local namespace inside `func()`.
- `len()` is part of the built-in namespace.

---

### **2. What is Scoping?**

**Scoping** in Python refers to the visibility or accessibility of variables in different parts of the program. Scoping determines where a variable can be accessed.

#### **LEGB Rule (Local, Enclosing, Global, Built-in)**

Python uses the **LEGB rule** to resolve variable names:

1. **L (Local)**: Variables defined inside the current function.
2. **E (Enclosing)**: Variables defined in any enclosing function (nested function).
3. **G (Global)**: Variables defined in the main program.
4. **B (Built-in)**: Variables in the built-in namespace (e.g., `len()`, `max()`).

#### **Example:**

```python
a = 10  # Global scope

def outer():
    a = 20  # Enclosing scope

    def inner():
        a = 30  # Local scope
        print(a)  # Prints the local variable

    inner()

outer()
```

**Output:**

```
30
```

- When the `inner()` function is executed, it accesses the **local scope** of `inner()` and prints the value `30`.

---

### **3. Variable Lookup Order (LEGB)**

The **LEGB** rule is used to determine which variable should be used when there is a name conflict. Python will search for a variable in the following order:

1. **Local**: First, it looks for the variable in the **local scope**.
2. **Enclosing**: If the variable is not found, Python will look in the **enclosing scope** (if any).
3. **Global**: If not found, it will look in the **global scope**.
4. **Built-in**: If it is still not found, Python will search in the **built-in** namespace.

#### **Example:**

```python
a = 5  # Global scope

def outer():
    a = 10  # Enclosing scope
    def inner():
        a = 15  # Local scope
        print(a)  # Local variable is used

    inner()

outer()
```

**Output:**

```
15
```

- Here, Python follows the LEGB rule and finds the variable `a` in the **local scope** of the `inner()` function, which is `15`.

---

### **4. Global and Local Variables**

In Python, variables declared outside a function are **global** variables, and those declared inside a function are **local** variables. You can use the `global` keyword to modify global variables inside a function.

#### **Global Variables Example:**

```python
x = 10  # Global variable

def func():
    print(x)  # Accessing global variable inside a function

func()  # Output: 10
```

#### **Modifying Global Variables Using `global` Keyword**:

```python
x = 10  # Global variable

def func():
    global x
    x = 20  # Modifying the global variable

func()
print(x)  # Output: 20
```

- Without the `global` keyword, modifying `x` inside the function would result in creating a local variable `x`, leaving the global `x` unchanged.

---

### **5. Nested Functions and Enclosing Scope**

When one function is nested inside another, the inner function can access variables from the enclosing function's scope. This is called **enclosing scope**.

#### **Example:**

```python
def outer():
    a = 20  # Enclosing scope variable

    def inner():
        print(a)  # Accessing variable from enclosing scope

    inner()

outer()  # Output: 20
```

- The `inner()` function can access the variable `a` from the **enclosing scope** (`outer()` function).

---

### **6. Nonlocal Keyword**

The `nonlocal` keyword is used to refer to variables in the **enclosing scope** but not in the global scope. It allows you to modify a variable in the enclosing scope.

#### **Example:**

```python
def outer():
    a = 10  # Enclosing scope variable

    def inner():
        nonlocal a  # Accessing and modifying variable from the enclosing scope
        a = 20  # Modifying variable 'a' in the enclosing scope

    inner()
    print(a)  # Output: 20

outer()
```

- The `nonlocal` keyword allows `inner()` to modify `a` from the **enclosing** scope (in this case, the `outer()` function).

---

### **7. Built-in Scope**

The **built-in scope** refers to Python's standard library and built-in functions. This scope is always available and contains functions like `print()`, `len()`, and constants like `None`.

#### **Example:**

```python
print(len("Hello"))  # len is a built-in function
```

- Here, `len()` is accessed from the **built-in namespace**.

---

### **8. Python Scoping and Closures**

A **closure** occurs when a function **remembers** the values from the enclosing scope even after the outer function has finished executing. Closures can be used to preserve state across function calls.

#### **Example:**

```python
def outer(x):
    def inner(y):
        return x + y
    return inner

closure = outer(10)
print(closure(5))  # Output: 15
```

- The `inner()` function has access to the variable `x` from its enclosing function `outer()`, even though `outer()` has finished execution.

---

### **9. Best Practices and Summary**

- **Avoid global variables**: Try to limit the use of global variables, as they can make code harder to debug and maintain.
- **Use `nonlocal` carefully**: While `nonlocal` can be powerful, it should be used sparingly, as it can create confusion and make the code harder to understand.
- **Local variables are fast**: Variables in local scope are faster to access than global or built-in variables. Use local variables wherever possible for efficiency.
- **Understand closures**: Closures are a powerful feature in Python, useful for maintaining state and creating higher-order functions.

---

### **Summary of Key Terms:**

- **Namespace**: A container for variables and their values. Includes **local**, **global**, **enclosing**, and **built-in** namespaces.
- **Scope**: The context in which a variable is accessible. Python follows the **LEGB rule** to resolve variable names.
- **Global Variables**: Variables declared outside a function, accessible from any part of the code.
- **Local Variables**: Variables declared inside a function, only accessible within that function.
- **Nonlocal Keyword**: Used to refer to variables in the **enclosing scope**.
- **Closure**: A function that retains access to variables from its enclosing scope even after the enclosing function has finished executing.

Understanding **namespace** and **scoping** will help you write more efficient, readable, and maintainable Python code. These concepts are also fundamental to understanding the inner workings of Python functions, decorators, and closures.


In Python, **namespaces** and **scoping** are fundamental concepts that govern how variables and objects are organized, accessed, and managed in a program. Understanding these concepts is crucial for writing efficient and bug-free code. Below is a comprehensive guide to Python namespaces and scoping, from basic to advanced concepts:

---

### **1. Namespaces**

A **namespace** is a mapping between names (identifiers) and objects (variables, functions, classes, etc.). It ensures that names are unique and do not conflict with each other.

#### **Types of Namespaces**

1. **Built-in Namespace**

   - Contains built-in functions and types (e.g., `print`, `len`, `int`).
   - Available as soon as the Python interpreter starts.
   - Accessed using the `__builtins__` module.

2. **Global Namespace**

   - Contains names defined at the top level of a script or module.
   - Created when the script/module is executed.
   - Accessed using the `globals()` function.

3. **Local Namespace**

   - Contains names defined inside a function or block.
   - Created when the function is called and destroyed when the function exits.
   - Accessed using the `locals()` function.

4. **Enclosing Namespace**
   - Relevant for nested functions.
   - Contains names from the enclosing (outer) function.
   - Accessed using the `nonlocal` keyword.

---

### **2. Scoping**

**Scoping** defines the region of a program where a namespace is directly accessible. Python uses the **LEGB rule** to resolve names:

1. **L (Local)**

   - Names defined inside the current function.
   - Example:
     ```python
     def func():
         x = 10  # Local scope
         print(x)
     func()
     ```

2. **E (Enclosing)**

   - Names in the local scopes of enclosing functions (for nested functions).
   - Example:
     ```python
     def outer():
         x = 10  # Enclosing scope
         def inner():
             print(x)  # Accesses x from outer function
         inner()
     outer()
     ```

3. **G (Global)**

   - Names defined at the top level of a script or module.
   - Example:
     ```python
     x = 10  # Global scope
     def func():
         print(x)  # Accesses x from global scope
     func()
     ```

4. **B (Built-in)**
   - Names predefined in the built-in namespace.
   - Example:
     ```python
     print(len("Hello"))  # Accesses len from built-in scope
     ```

---

### **3. Variable Scope Rules**

1. **Local Variables**

   - Variables defined inside a function are local to that function.
   - Example:
     ```python
     def func():
         x = 10  # Local variable
         print(x)
     func()
     ```

2. **Global Variables**

   - Variables defined outside all functions are global.
   - Use the `global` keyword to modify a global variable inside a function.
   - Example:
     ```python
     x = 10  # Global variable
     def func():
         global x
         x = 20  # Modifies global x
     func()
     print(x)  # Output: 20
     ```

3. **Nonlocal Variables**
   - Used in nested functions to modify a variable in the enclosing (non-global) scope.
   - Example:
     ```python
     def outer():
         x = 10  # Enclosing scope
         def inner():
             nonlocal x
             x = 20  # Modifies x in outer function
         inner()
         print(x)  # Output: 20
     outer()
     ```

---

### **4. Namespace Lifetime**

1. **Built-in Namespace**

   - Created when the Python interpreter starts and lasts until it exits.

2. **Global Namespace**

   - Created when the module is imported and lasts until the interpreter exits.

3. **Local Namespace**
   - Created when a function is called and destroyed when the function exits.

---

### **5. Advanced Concepts**

1. **Closures**

   - A closure is a function that retains access to variables from its enclosing scope, even after the outer function has finished executing.
   - Example:
     ```python
     def outer():
         x = 10
         def inner():
             print(x)  # Accesses x from outer function
         return inner
     closure = outer()
     closure()  # Output: 10
     ```

2. **Class Namespace**

   - Classes have their own namespace, which includes class variables and methods.
   - Example:
     ```python
     class MyClass:
         x = 10  # Class variable
         def method(self):
             print(self.x)  # Accesses class variable
     obj = MyClass()
     obj.method()  # Output: 10
     ```

3. **Module Namespace**

   - Each module has its own global namespace.
   - Example:

     ```python
     # module.py
     x = 10
     def func():
         print(x)
     ```

     ```python
     # main.py
     import module
     print(module.x)  # Accesses x from module namespace
     module.func()  # Output: 10
     ```

4. **Dynamic Namespace Manipulation**

   - Use `globals()` and `locals()` to dynamically access or modify namespaces.
   - Example:
     ```python
     x = 10
     print(globals()['x'])  # Output: 10
     ```

5. **Name Mangling**

   - Python uses name mangling to make class attributes private (not truly private).
   - Example:
     ```python
     class MyClass:
         def __init__(self):
             self.__private = 10  # Name mangling: _MyClass__private
     obj = MyClass()
     print(obj._MyClass__private)  # Output: 10
     ```

6. **Scope Resolution in Nested Functions**
   - Python uses the LEGB rule to resolve names in nested functions.
   - Example:
     ```python
     x = 10  # Global scope
     def outer():
         x = 20  # Enclosing scope
         def inner():
             x = 30  # Local scope
             print(x)  # Output: 30
         inner()
     outer()
     ```

---

### **6. Common Pitfalls**

1. **UnboundLocalError**

   - Occurs when a local variable is referenced before assignment.
   - Example:
     ```python
     x = 10
     def func():
         print(x)  # UnboundLocalError
         x = 20
     func()
     ```

2. **Shadowing Global Variables**

   - Avoid accidentally shadowing global variables with local variables.
   - Example:
     ```python
     x = 10
     def func():
         x = 20  # Shadows global x
         print(x)  # Output: 20
     func()
     print(x)  # Output: 10
     ```

3. **Namespace Pollution**
   - Avoid adding unnecessary names to the global namespace.
   - Example:
     ```python
     # Bad practice
     x = 10
     y = 20
     z = 30
     ```
