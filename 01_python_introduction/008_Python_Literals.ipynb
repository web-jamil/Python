# Python Literals

**Literal** is a raw data given in a **variable** or **constant**. In Python, there are various types of literals they are as follows:


## 1. Numeric Literals

Numeric Literals are **immutable (unchangeable)**. Numeric literals can belong to 3 different numerical types **`Integer`**, **`Float`** and **`Complex`**.

<div>

Number data types in Python:

1. Integers: Integer(negative, zero and positive) numbers

   - Example:
     - ... -3, -2, -1, 0, 1, 2, 3 ...

2. Floating Point Numbers(Decimal numbers)

   - Example:
     - ... -3.5, -2.25, -1.0, 0.0, 1.1, 2.2, 3.5 ...

3. Complex Numbers
   - Example:
     - 1 + j, 2 + 3j, 1 - 1j


````python
# Example:

a = 0b1010 #Binary Literals
b = 100 #Decimal Literal
c = 0o310 #Octal Literal
d = 0x12c #Hexadecimal Literal

# Integer Literal
int_1 = 10
int_2 = 99


# Float Literal
float_1 = 10.5
float_2 = 1.5e2


# Complex Literal
x = 3.14j

print(a, b, c, d)
print(int_1, int_2)
print(float_1, float_2)
print(x, x.imag, x.real)
````

In the above program

- We assigned integer literals into different variables. Here, **`a`** is **binary** literal, **`b`** is a **decimal** literal,**`c`** is an **octal** literal and **`d`** is a **hexadecimal** literal.

- When we print the variables, all the literals are converted into **decimal** values.

- **`10.5`** and **`1.5e2`** are **floating point** literals. **`1.5e2`** is **expressed** with **exponential** and is **equivalent** to **`1.5 * 10^2`**.

- We assigned a complex literal i.e **`3.14j`** in variable **`x`**. Then we use **imaginary** literal (x.imag) and **real** literal (x.real) to create imaginary and real part of **complex number**.

## 2. String literals (unicode character strings)

A **string** literal is a **sequence of characters** surrounded by **quotes**. We can use both **single**, **double** or **triple** quotes for a string. And, a **character literal** is a single character surrounded by single or double quotes.


````python
# Example:

a = '''Apple'''

b = """Apple"""

c = 'Apple'

d = "Apple"

e = Apple     # cannot write string with out quotes ('', " ", """ """, ''' ''')

print(a)
print(b)
print(c)
print(d)
print(e)
````

````python
# Example:

strings = "This is Python"
char = "C"
multiline_str = """This is a multiline string with more than one line code."""
unicode = u"\u00dcnic\u00f6de"
raw_str = r"raw \n string"

print(strings)
print(char)
print(multiline_str)
print(unicode)
print(raw_str)
````

In the above program,

- **`This is Python`** is a string literal and **`C`** is a **character** literal.

- The value with **triple-quote """** assigned in the **`multiline_str`** is **multi-line** string literal.

- The **`u"\u00dcnic\u00f6de"`** is a **unicode literal** which supports characters other than English. In this case, **`\u00dc`** represents **`Ü`** and **`\u00f6`** represents **`ö`**.

- **`r"raw \n string"`** is a raw string literal.


## 3. Boolean literals

A Boolean literal can have any of the two values: **`True`** or **`False`**.


````python
# Example:

#REMEMBER  True == 1   False == 0

x = (1 == True)
y = (1 == False)
a = True + 6
b = False + 90

print("x is", x)
print("y is", y)
print("a:", a)
print("b:", b)
````

In the above program, we use boolean literal **`True`** and **`False`**. In Python, **`True`** represents the value as **`1` and `False` as `0`**. The value of `x` is `True` because **`1` is equal to `True`**. And, the value of **`y`** is **`False`** because **`1` is not equal to `False`**.

Similarly, we can use the **`True`** and **`False`** in numeric expressions as the value. The value of **`a`** is **`6`** because we add **`True`** which has value of **`1` with `6`**. Similarly, **`b`** is **`90`** because we add the **`False`** having value of **`0` with `90`**.


````python
# Example:

9.0           # a simple floating point number
1e100         # a googol as floating point number
-1234567890   # an integer
True or False # the two possible boolean values
'This is a string'
"It's another string"
print("""Triple quotes (also with '''), allow strings to break over multiple lines.
Alternatively \n is a newline character (\t for tab, \\ is a single backslash)""")
````

## Special literals

Python contains one **special** literal i.e., **`None`**. We use it to specify to that field that is not created.


````python
# Example:

juice = "Available"
soup = None
def menu(x):
    if x == juice:
        print(juice)
    else:
        print(soup)
menu(juice)
menu(soup)
````

In the above program, we define a **`menu` function**. Inside **`menu`**, when we set parameter as **`drink`** then, it displays **`Available`**. And, when the parameter is **`food`**, it displays **`None`**.

### **Numeric Literals in Python: All About from Basic to Advanced**

In Python, **numeric literals** represent numbers in the code directly. They are used to assign numeric values to variables or perform calculations. Numeric literals in Python come in different types and are classified based on their format. Python supports integer, floating-point, and complex numeric literals.

This guide will provide a detailed understanding of numeric literals in Python, from the basics to advanced concepts.

---

### **1. What Are Numeric Literals?**

A **numeric literal** is a constant value used in Python to represent numbers directly in the source code. Python supports three types of numeric literals:

- **Integer literals** (whole numbers)
- **Floating-point literals** (numbers with a decimal point)
- **Complex number literals** (numbers with real and imaginary parts)

---

### **2. Types of Numeric Literals in Python**

#### **A. Integer Literals**

Integer literals represent whole numbers without a fractional part. Python allows integers to be written in different number bases, such as:

- Decimal (base 10)
- Binary (base 2)
- Octal (base 8)
- Hexadecimal (base 16)

##### **1. Decimal Integer Literals**

Decimal literals are the most common type of integer and are written using digits from 0 to 9.

**Syntax:**

```python
decimal_literal = 42  # A normal decimal number
```

##### **2. Binary Integer Literals**

Binary literals represent numbers in base 2, using only `0` and `1`. They are written with a prefix `0b` or `0B`.

**Syntax:**

```python
binary_literal = 0b1010  # Binary for 10
```

##### **3. Octal Integer Literals**

Octal literals represent numbers in base 8, using digits from `0` to `7`. They are written with a prefix `0o` or `0O`.

**Syntax:**

```python
octal_literal = 0o12  # Octal for 10 (8 in decimal)
```

##### **4. Hexadecimal Integer Literals**

Hexadecimal literals represent numbers in base 16, using digits from `0` to `9` and letters from `a` to `f` (or `A` to `F`). They are written with a prefix `0x` or `0X`.

**Syntax:**

```python
hexadecimal_literal = 0x1F  # Hexadecimal for 31
```

##### **Examples:**

```python
decimal = 100
binary = 0b1101    # 13 in decimal
octal = 0o75       # 61 in decimal
hexadecimal = 0x1A # 26 in decimal
```

#### **B. Floating-Point Literals**

Floating-point literals represent numbers that have a decimal point. These can also be expressed in scientific notation (e.g., `1e3`).

##### **1. Standard Floating-Point Literals**

Floating-point literals are written with digits and a decimal point.

**Syntax:**

```python
float_literal = 3.14
```

##### **2. Scientific Notation (Exponential)**

In scientific notation, numbers are represented as a base value followed by an exponent, using the `e` or `E` notation. It means the base number should be multiplied by 10 raised to the power of the exponent.

**Syntax:**

```python
scientific_literal = 1.23e4  # Equivalent to 1.23 * 10^4 = 12300
```

##### **Examples:**

```python
standard_float = 3.14
scientific_float = 2.5e3  # 2.5 * 1000 = 2500
```

#### **C. Complex Number Literals**

Complex number literals are used to represent numbers with both real and imaginary parts. In Python, a complex number is written as `real_part + imaginary_part`, where the imaginary part is suffixed with a `j` or `J`.

##### **Syntax:**

```python
complex_literal = 3 + 4j  # 3 is the real part, 4j is the imaginary part
```

##### **Examples:**

```python
complex_num = 3 + 4j
another_complex = 2 - 5j
```

---

### **3. Advanced Concepts of Numeric Literals**

#### **A. Numeric Types and Type Conversion**

Python automatically assigns the appropriate numeric type to a value, whether it is an integer, floating-point number, or complex number. You can also explicitly convert between numeric types using Python's built-in functions.

##### **Examples:**

```python
# Integer to float
a = 10
b = float(a)  # 10.0

# Float to integer
x = 3.14
y = int(x)  # 3 (The decimal part is truncated)

# Complex to integer or float (Not possible directly, need to extract real part)
z = 5 + 2j
real_part = z.real  # 5.0
imaginary_part = z.imag  # 2.0
```

#### **B. Large Numbers and Underscore for Readability**

Python supports large integers without any size limit (apart from available memory), and allows you to use underscores (`_`) to improve readability of large numbers.

##### **Examples:**

```python
# Large integer literals
large_number = 1_000_000  # 1000000

# Large floating-point literals
large_float = 1.23e6  # 1230000.0

print(large_number)  # Output: 1000000
print(large_float)  # Output: 1230000.0
```

#### **C. Precision and Rounding of Floating-Point Numbers**

Floating-point numbers are represented in binary internally, and this can lead to rounding errors. To handle precision, Python provides several libraries and techniques, such as `round()` and the `decimal` module.

##### **Example:**

### **Boolean Literals in Python: From Basic to Advanced**

Boolean literals represent a simple data type in Python that can have one of two possible values: `True` or `False`. They are used in logical expressions, conditions, and control flow statements to represent truth values.

In Python, boolean literals are a core part of the language and play a fundamental role in decision-making and condition checking.

---

### **1. What are Boolean Literals?**

In Python, boolean literals are predefined constants that represent two truth values:

- **True**: Represents a logical true value.
- **False**: Represents a logical false value.

Both `True` and `False` are the only two boolean literals available in Python, and they are capitalized.

**Boolean literals** are often used in conditional expressions, boolean operations, and in the evaluation of expressions in control structures like `if`, `while`, and `for` loops.

---

### **2. Basic Syntax of Boolean Literals**

In Python, boolean literals are simply the keywords `True` and `False`.

#### **Syntax:**

```python
# True boolean literal
is_valid = True

# False boolean literal
is_invalid = False
```

**Examples:**

```python
is_sunny = True
is_raining = False
```

These boolean literals (`True` and `False`) are used to represent binary truth values.

---

### **3. Boolean Context in Python**

Python has several contexts where boolean literals are implicitly used. The values `True` and `False` are commonly returned by conditional checks, comparisons, and logical operations.

#### **A. Comparison Operators**

Comparison operators compare values and return boolean results.

**Syntax:**

```python
a = 10
b = 20
result = a < b  # result will be True
```

**Example:**

```python
print(5 == 5)   # True
print(5 != 3)   # True
print(3 > 5)    # False
```

#### **B. Logical Operators**

Logical operators such as `and`, `or`, and `not` evaluate expressions that result in boolean values.

- `and`: Returns `True` if both operands are `True`.
- `or`: Returns `True` if at least one operand is `True`.
- `not`: Reverses the truth value of the operand.

**Example:**

```python
x = True
y = False

print(x and y)   # False
print(x or y)    # True
print(not x)     # False
```

#### **C. In Conditional Statements**

Boolean literals are often used in conditional structures such as `if`, `while`, and `for`.

**Example:**

```python
is_raining = False

if is_raining:
    print("Take an umbrella")
else:
    print("No need for an umbrella")
```

**Output:**

```
No need for an umbrella
```

#### **D. Evaluating Truthiness**

Python evaluates all objects in boolean context. Non-zero numbers, non-empty sequences (e.g., lists, tuples, strings), and `True` are considered **truthy**. Zero, empty sequences, and `False` are considered **falsy**.

##### **Examples:**

```python
# Truthy
print(bool(1))       # True
print(bool("Hello")) # True
print(bool([1, 2]))  # True

# Falsy
print(bool(0))       # False
print(bool(""))      # False
print(bool(None))    # False
```

#### **E. Boolean Context in Functions**

Functions like `any()` and `all()` use boolean values in their operations to return a result based on the truthiness of elements in an iterable.

**Examples:**

```python
values = [True, False, True]

# any() returns True if at least one element is True
print(any(values))  # True

# all() returns True if all elements are True
print(all(values))  # False
```

### **4. Boolean Literals in Expressions**

Boolean literals (`True` and `False`) can be part of more complex expressions and play an important role in conditionals, loops, and logical expressions.

#### **A. Logical Expressions**

You can directly use boolean literals in logical expressions:

```python

y = False

```

#### **B. In `if` Statements**

```python
    print("This will always be printed.")
```

```python

    print("This will never be printed.")
```

#### **C. As Return Values**

You can also return boolean literals from functions:

```python
    return num % 2 == 0

print(is_even(7))  # False
```

---

### **5. Boolean Literals in Classes and Custom Objects**

You can also define how objects behave in boolean contexts by implementing special methods like `__bool__()` and `__len__()`.

#### **B. Using `__len__()` to Customize Truth Value**

   ```python

       # Do something
   ```


   - `None`, empty strings, `0`, and empty containers (e.g., lists, tuples, sets) are treated as `False` in a boolean context. However, other objects might have special rules, so always be clear about what you expect.

```python
text = "This is an octal character: \101"
print(text)
```

**Output:**

```
This is an octal character: A
```

#### **4.2. Hexadecimal Escape Sequence (`\xhh`)**

A hexadecimal escape sequence (`\xhh`) allows you to represent a character using its hexadecimal value.

```python
# Example
text = "This is a hexadecimal character: \x41"
print(text)
```

**Output:**

```
This is a hexadecimal character: A
```

---

### **5. Raw Strings (`r"..."` or `R"..."`)**

In Python, you can use raw strings to prevent escape sequences from being processed. This is especially useful when working with regular expressions or file paths, where backslashes are common and you want to avoid the need for double escaping.

```python
# Example
raw_string = r"Hello\nWorld"
print(raw_string)
```

**Output:**

```
Hello\nWorld
```

In a normal string, `\n` would be treated as a newline, but with a raw string, the backslash is preserved.

### **6. Multiline Strings**

In Python, you can use triple quotes (`"""` or `'''`) to create strings that span multiple lines. Newlines within the string are included automatically without needing an escape character.

```python
# Example
multiline_string = """This is
a multiline
string."""
print(multiline_string)
```

**Output:**

```
This is
a multiline
string.
```

### **7. Escape Sequences in Regular Expressions**

In regular expressions, many characters (such as `*`, `+`, `?`, and `.`) have special meanings. To use these characters literally in regular expressions, you must escape them with a backslash `\`.

```python
import re

# Example
pattern = r"\."
text = "This is a dot: ."
match = re.search(pattern, text)
print(match.group())  # Output: .
```

---

### **8. Escape Sequences in File Paths**

In file paths, especially in Windows, backslashes (`\`) are used. However, Python interprets backslashes as escape characters, so to represent a file path correctly, you can either use double backslashes `\\` or raw strings `r"..."`.

```python
# Windows path
file_path = "C:\\Users\\Username\\Documents\\file.txt"
# Or use raw string for simplicity
file_path = r"C:\Users\Username\Documents\file.txt"
```

---

### **9. Advanced Use: Combining Escape Sequences**

You can combine multiple escape sequences in a single string to achieve more complex formatting.

```python
# Example
text = "Hello\tWorld\nNew Line\nBackslash: \\"
print(text)
```

**Output:**

```
Hello   World
New Line
Backslash: \
```

### **10. Conclusion**

Escape characters are essential in Python for handling special characters that would otherwise interfere with the syntax or formatting of strings. By understanding and using escape characters effectively, you can manipulate text in powerful ways, whether it’s for formatting output, working with file paths, or processing regular expressions.

In summary:

- **Basic escape characters**: `\n`, `\t`, `\\`, `\'`, `\"`
- **Advanced escape characters**: `\b`, `\r`, `\f`, `\v`, `\u`, `\x`, `\o`
- **Raw strings** (`r"..."`) for ignoring escape sequences
