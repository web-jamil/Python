## 1. Python Output Using `print()` function

We use the **`print()`** function to output data to the standard output device (screen). We can also **[output data to a file](https://github.com/milaan9/05_Python_Files/blob/main/001_Python_File_Input_Output.ipynb)**, but this will be discussed later.

An example of its use is given below.


````python
# Example 1:

print('This sentence is output to the screen')
````

````python
# Example 2:

a = 9
print('The value of a is', a)
````

In the second **`print()`** statement, we can notice that space was added between the **[string](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String.ipynb)** and the value of variable **`a`**. This is by default, but we can change it.


**Syntax** of the **`print()`** function is:

```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```

Here, objects is the value(s) to be printed.

The **`sep`** separator is used between the values. It defaults into a space character.

After all values are printed, **`end`** is printed. It defaults into a new line.

The **`file`** is the object where the values are printed and its default value is **`sys.stdout`** (screen). Here is an example to illustrate this.


````python
print(1, 2, 3, 4)
print(1, 2, 3, 4, sep='#')   # It will separate your elements with '#'
print(1, 2, 3, 4, sep='*', end='&')  # It will separate your elements with '*' and end with '&'
````

## Output formatting

Sometimes we would like to format our output to make it look attractive. This can be done by using the **`str.format()`** method. This method is visible to any string object.


````python
x = 6; y = 12
print('The value of x is {} and y is {}'.format(x,y))
````

Here, the curly braces **`{}`** are used as placeholders. We can specify the order in which they are printed by using numbers (tuple index).


````python
print('I love {0} and {1}'.format('Mango','Banana'))
print('I love {1} and {0}'.format('Mango','Banana'))
````

We can even use keyword arguments to format the string.


````python
print('Hello {name}, {greeting}!'.format(greeting = 'Good morning', name = 'Mark'))
````

We can also format strings like the old **`sprintf()`** style used in C programming language. We use the **`%`** operator to accomplish this.


````python
x = 12.34567890
print('The value of x is %0.2f' %x)  # "%0.2" means only 2 number after decimal

print('The value of x is %0.3f' %x)  # "%0.3" means only 3 number after decimal
````

### Accept multiline input from a user

As you know, the input() function does not allow the user to enter put lines separated by a newline. If the user tries to enter multiline input, it prints back only the first line. Let’s see how to gets multiple line input.

As you know, the input() function does not allow the user to enter put lines separated by a newline. If the user tries to enter multiline input, it prints back only the first line. Let’s see how to gets multiple line input.

We can use for loop. In each iteration, we can get input string from the user and then .join() them using \n, or you can also concatenate each input string using + operator separated by \n.


````python
MultiLine = []
print("Tell me about yourself")
while True:
    line = input()
    if line:
        MultiLine.append(line)
    else:
        break
finalText = '\n'.join(MultiLine)
print("\n")
print("Final text input")
print(finalText)
````

### **Python Input Functions: From Basic to Advanced**

In Python, the **input function** is used to capture user input from the console. This user input can then be used in programs to perform tasks based on user decisions, preferences, or dynamic data. The built-in function for accepting user input is `input()`.

---

### **1. The `input()` Function: Basic Usage**

#### **Basic Usage of `input()`**:

The `input()` function in Python reads a line from input, converts it into a string (by default), and returns it.

```python
# Basic Example
name = input("Enter your name: ")
print("Hello, " + name + "!")
```

**Explanation**:

- `input()` is used to prompt the user for input.
- The prompt string `"Enter your name: "` is optional, but it can be used to provide instructions or a message to the user.
- The returned value is always a string, even if the user enters a number.

**Sample Output**:

```
Enter your name: Alice
Hello, Alice!
```

---

### **2. Converting Input to Other Data Types**

Since the `input()` function always returns a string, you may need to convert the input into another data type like an integer or float to use it for mathematical operations.

#### **Converting Input to an Integer**:

```python
# Converting input to an integer
age = int(input("Enter your age: "))
print("Your age is:", age)
```

**Explanation**:

- `int()` is used to convert the input into an integer. If the input is not a valid integer (like "abc"), Python will raise a `ValueError`.

**Sample Output**:

```
Enter your age: 25
Your age is: 25
```

#### **Converting Input to a Float**:

```python
# Converting input to a float
height = float(input("Enter your height in meters: "))
print("Your height is:", height)
```

**Explanation**:

- `float()` converts the input string to a floating-point number.

**Sample Output**:

```
Enter your height in meters: 5.9
Your height is: 5.9
```

---

### **3. Handling Invalid Inputs**

You might want to ensure that the user inputs valid data before proceeding. This can be done using exception handling (`try` and `except` blocks).

```python
# Handling invalid input using try-except
try:
    age = int(input("Enter your age: "))
except ValueError:
    print("Invalid input! Please enter a valid integer.")
```

**Explanation**:

- If the user enters a non-integer value, Python will catch the `ValueError` and display a custom error message.

**Sample Output**:

```
Enter your age: twenty-five
Invalid input! Please enter a valid integer.
```

---

### **4. Multiple Inputs in One Line**

Sometimes, you may want to take multiple inputs on the same line, separated by spaces. You can do this using `split()`.

```python
# Taking multiple inputs in one line
name, age = input("Enter your name and age separated by a space: ").split()
age = int(age)  # Convert age to an integer
print("Name:", name)
print("Age:", age)
```

**Explanation**:

- `split()` splits the input string at spaces and returns a list of substrings. You can assign the list elements to variables using tuple unpacking.

**Sample Output**:

```
Enter your name and age separated by a space: John 30
Name: John
Age: 30
```

#### **Using `split()` with custom delimiters**:

You can specify custom delimiters for splitting the input. For example, using commas:

```python
# Splitting input with a comma
numbers = input("Enter numbers separated by commas: ").split(',')
print("Numbers:", numbers)
```

**Sample Output**:

```
Enter numbers separated by commas: 1,2,3,4,5
Numbers: ['1', '2', '3', '4', '5']
```

---

### **5. Using `input()` in a Loop**

You can repeatedly ask for user input by using `input()` inside a loop. This is useful when you want to prompt for several inputs in a row, or if you want to allow the user to quit the input process at any time.

```python
# Using a loop to accept multiple inputs
while True:
    response = input("Do you want to continue? (yes/no): ").lower()
    if response == "no":
        print("Exiting the program.")
        break
    elif response == "yes":
        print("Continuing the program...")
    else:
        print("Invalid input. Please enter 'yes' or 'no'.")
```

**Explanation**:

- The loop continues until the user types `"no"`, and breaks out of the loop.
- The `.lower()` method is used to ensure that input is case-insensitive.

**Sample Output**:

```
Do you want to continue? (yes/no): yes
Continuing the program...
Do you want to continue? (yes/no): no
Exiting the program.
```

### **6. Advanced Techniques**

#### **Reading Input from Multiple Users (Using `input()` in Lists)**

```python
# Collecting multiple inputs from the user
user_count = int(input("How many users are there? "))
users = []
for _ in range(user_count):
    user = input("Enter username: ")
    users.append(user)

print("Users entered:", users)
```

**Explanation**:

- This program asks the user for the number of users, then collects usernames in a loop and stores them in a list.

**Sample Output**:

```
How many users are there? 3
Enter username: Alice
Enter username: Bob
Enter username: Charlie
Users entered: ['Alice', 'Bob', 'Charlie']
```

#### **Advanced User Input (Reading from Files)**

While `input()` is commonly used to get input from the user interactively, Python can also accept input from a file using `open()` and `read()` methods.

```python
# Reading input from a file
with open('input.txt', 'r') as file:
    data = file.read()
print(data)
```

**Explanation**:

- The file is opened using `open()`, and the content is read using `read()`. This is useful when you want to process large input that is not interactive.

### **7. Python `input()` and `eval()`**

You can also use the `eval()` function to evaluate user input as Python code. This is powerful but should be used carefully as it can execute arbitrary code, which might be unsafe.

```python
# Evaluating a mathematical expression
expression = input("Enter a mathematical expression: ")
result = eval(expression)
print("The result is:", result)
```

**Explanation**:

- `eval()` evaluates the input string as a Python expression.

**Sample Output**:

```
Enter a mathematical expression: 3 * (4 + 5)
The result is: 27
```

- **Warning**: Using `eval()` on untrusted user input can lead to code injection vulnerabilities. Always sanitize the input if using `eval()`.

## **1. Basic Usage of `print()`**

### **Syntax**

```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```

- **`sep`**: Separator between objects (default is a space `' '`).
- **`end`**: Character to be printed at the end (default is a newline `'\n'`).
- **`file`**: The file-like object where the output is written (default is `sys.stdout`).
- **`flush`**: Whether to forcibly flush the output stream (default is `False`).

### **Python `print()` Function: From Basic to Advanced**

The `print()` function in Python is used to display output to the console. It is one of the most frequently used functions in Python for debugging, providing information to the user, and for logging purposes. Let's break down the `print()` function and explore its basic usage, parameters, and advanced functionalities.

---

### **1. Basic Usage of `print()`**

#### **Syntax**:

```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```

- `*objects`: One or more objects to be printed. These can be strings, numbers, lists, etc.
- `sep`: The separator between multiple objects. By default, it's a space (' ').
- `end`: The string appended after the last object. By default, it's a newline (`\n`).
- `file`: The file object where the output will be written. Default is `sys.stdout`, which is the console.
- `flush`: Whether to forcibly flush the stream. Default is `False`.

#### **Basic Example**:

```python
# Basic print example
print("Hello, World!")
```

**Output**:

```
Hello, World!
```

---

### **2. Printing Multiple Objects**

The `print()` function can print multiple objects at once. By default, it separates the objects by a space.

```python
# Printing multiple objects
name = "Alice"
age = 25
print("Name:", name, "Age:", age)
```

**Output**:

```
Name: Alice Age: 25
```

You can control the separator between multiple objects using the `sep` argument.

```python
# Custom separator
print("Name:", name, "Age:", age, sep=' | ')
```

**Output**:

```
Name: | Alice | Age: | 25
```

---

### **3. Changing the End Character**

By default, the `print()` function ends with a newline character (`\n`). You can change this by using the `end` argument.

```python
# Changing the end character
print("Hello,", end=" ")
print("World!")
```

**Output**:

```
Hello, World!
```

In this case, the first `print()` statement ends with a space instead of a newline, so the second `print()` continues on the same line.

---

### **4. Printing to a File**

The `print()` function can print output to a file instead of the console by using the `file` argument.

```python
# Printing to a file
with open("output.txt", "w") as f:
    print("Hello, World!", file=f)
```

This will write the string "Hello, World!" to the `output.txt` file. If the file does not exist, it will be created; if it does exist, it will be overwritten.

---

### **5. Flushing the Output Buffer**

The `flush` argument can be used to forcibly flush the output buffer, which is useful when you want to ensure that the output is written immediately, especially when working with large data or in real-time applications.

```python
# Flushing the output buffer
import time
print("Starting...", end=" ", flush=True)
time.sleep(2)
print("Done!")
```

Here, the `flush=True` ensures that the message "Starting..." is printed immediately, even though there is a `sleep(2)` delay before the "Done!" message is printed.

### **6. Printing Variables and Expressions**

#### **Using `str.format()`**:

**Output**:

```

```

This is useful when you have a larger number of variables to print and want to format them in a readable way.

### **Output Formatting in Python**

Output formatting in Python is an essential part of displaying data in a clean, readable, and user-friendly way. Python offers various ways to format output. These methods allow you to align text, format numbers, include placeholders, handle special characters, and much more.

We'll cover the various techniques and features for formatting output in Python, ranging from basic to advanced.

---

### **1. Basic String Concatenation**

You can concatenate multiple strings using the `+` operator. However, this method can become cumbersome when working with multiple variables.

```python
name = "Alice"
age = 25
print("Name: " + name + ", Age: " + str(age))
```

**Output**:

```
Name: Alice, Age: 25
```

Here, `str(age)` converts the integer `age` to a string so that it can be concatenated with the other strings.

---

### **2. The `print()` Function and String Concatenation with Commas**

The `print()` function in Python automatically adds a space between arguments when multiple items are printed together, avoiding the need to manually concatenate strings.

```python
name = "Alice"
age = 25
print("Name:", name, "Age:", age)
```

**Output**:

```
Name: Alice Age: 25
```

This is cleaner than manual string concatenation.

---

### **3. Using `%` Operator for Formatting**

The `%` operator is a classic way to format strings. It works by substituting placeholders (like `%s` for strings or `%d` for integers) with values.

```python
name = "Alice"
age = 25
print("Name: %s, Age: %d" % (name, age))
```

**Output**:

```
Name: Alice, Age: 25
```

- `%s`: String format specifier.
- `%d`: Integer format specifier.

You can also use other format specifiers like `%f` for floating-point numbers and `%x` for hexadecimal representation.

---

### **4. The `str.format()` Method**

Introduced in Python 2.7 and Python 3.0, `str.format()` provides a more modern and powerful way to format strings. It allows you to insert variables into a string using curly braces `{}` as placeholders.

```python
name = "Alice"
age = 25
print("Name: {}, Age: {}".format(name, age))
```

**Output**:

```
Name: Alice, Age: 25
```

You can also specify the order of the arguments explicitly using indexes or keywords.

```python
print("Name: {0}, Age: {1}".format(name, age))
```

**Output**:

```
Name: Alice, Age: 25
```

Using **keyword arguments**:

```python
print("Name: {name}, Age: {age}".format(name="Alice", age=25))
```

**Output**:

```
Name: Alice, Age: 25
```

---

### **5. f-strings (Formatted String Literals)**

Introduced in Python 3.6, **f-strings** provide a more concise and readable way to format strings. The `f` in front of the string allows you to directly embed expressions inside curly braces `{}`.

```python
name = "Alice"
age = 25
print(f"Name: {name}, Age: {age}")
```

**Output**:

```
Name: Alice, Age: 25
```

You can even use expressions within the curly braces:

```python
a = 5
b = 10
print(f"The sum of {a} and {b} is {a + b}")
```

**Output**:

```
The sum of 5 and 10 is 15
```

F-strings are efficient and provide an easy-to-read way of formatting strings.

---

### **6. Alignment and Padding**

Python allows you to align text and pad it with specific characters using the `str.format()` method or f-strings.

#### **Using `str.format()`**:

```python
# Align text with spaces
print("{:<10} {:>10} {:^10}".format("Left", "Right", "Center"))
```

**Output**:

```
Left       Right    Center
```

- `:<10` left-aligns the text with a width of 10 characters.
- `:>10` right-aligns the text with a width of 10 characters.
- `:^10` centers the text with a width of 10 characters.

#### **Using f-strings**:

```python
print(f"{'Left':<10} {'Right':>10} {'Center':^10}")
```

**Output**:

```
Left       Right    Center
```

### **7. Padding Numbers**

You can pad numbers with zeros (or other characters) to ensure they have a certain width. This is especially useful for formatting numbers, such as for timestamps or serial numbers.

```python

print(f"{num:03}")  # Padding with leading zeros to make the number 3 digits long
```

**Output**:

```
007
```

### **8. Decimal Precision and Floating-Point Formatting**

For floating-point numbers, you can control the number of decimal places printed using format specifiers.

#### **Using `str.format()`**:

```python

```

**Output**:

```
3.14
```

#### **Using f-strings**:

```python

```

**Output**:

```
3.14
```

You can also format numbers in scientific notation using `e` or `E`:


**Output**:

```
1.23e+06
```
