## 1. Python Statement

Instructions that a Python interpreter can execute are called statements. For example, **`a = 1`** is an assignment statement. **`if`** statement, **`for`** statement, **`while`** statement, etc. are other kinds of statements which will be discussed later.


### Multi-line statement

In Python, the end of a statement is marked by a newline character. But we can make a statement extend over multiple lines with the line continuation character **`\`**.

- Statements finish at the end of the line:
  - Except when there is an open bracket or paranthesis:

```python
>>> 1+2
>>> +3  #illegal continuation of the sum
```

- A single backslash at the end of the line can also be used to indicate that a statement is still incomplete

```python
>>> 1 + \
>>>    2 + 3 # this is also okay
```

For example:


````python
1+2    # assignment line 1
+3     # assignment line 2

# Python is only calculating assignment line 1
````

```python
1+2 \       # '\' means assignment line is continue to next line
+3
```

> **Note:** This would result in error because there should be no comment **`#`** or space after **`\`**.


````python
# Method 1:

1+2 \
+3
````

````python
1 + 2 + 3 + \
4 + 5 + 6 + \
7 + 8 + 9
````

This is an explicit line continuation. In Python, line continuation is implied inside:

1. parentheses **`( )`**,
   For Example:
   ```python
   (1+2 + 3) # perfectly OK even with spaces

```
2. brackets **`[ ]`**, and
3. braces **`{ }`**.
For instance, we can implement the above multi-line statement as:
```


````python
# Method 2:

(1+2
+3)
````

````python
(1 + 2 + 3 +
4 + 5 + 6 +
7 + 8 + 9)
````

Here, the surrounding parentheses **`( )`** do the line continuation implicitly. Same is the case with **`[ ]`** and **`{ }`**. For example:


````python
['red',
'blue',
    'green', '99']
````

## 3. Python Comments

Comments are very important while writing a program. They describe what is going on inside a program, so that a person looking at the source code does not have a hard time figuring it out.

You might forget the key details of the program you just wrote in a month's time. So taking the time to explain these concepts in the form of comments is always fruitful.

In Python, we use the hash **`#`** symbol to start writing a comment.

It extends up to the newline character. Comments are for programmers to better understand a program. Python Interpreter ignores comments.

Generally, comments will look something like this:

```python
#This is a Comment
```

Because comments do not **execute**, when you run a program you will not see any indication of the comment there. Comments are in the source code for **humans** to **read**, not for **computers to execute**.


````python
#This is a Comment
````

### 4. Docstrings in Python

A docstring is short for documentation string.

**[Python Docstrings](https://github.com/milaan9/04_Python_Functions/blob/main/Python_Docstrings.ipynb)** (documentation strings) are the **[string](https://github.com/milaan9/02_Python_Datatypes/blob/main/002_Python_String.ipynb)** literals that appear right after the definition of a function, method, class, or module.

Triple quotes are used while writing docstrings. For example:

```python
>>>def double(num):
>>>    """Function to double the value"""
>>>    return 3*num
```

Docstrings appear right after the definition of a function, class, or a module. This separates docstrings from multiline comments using triple quotes.

The docstrings are associated with the object as their **`__doc__`** attribute.

So, we can access the docstrings of the above function with the following lines of code:


````python
def double(num):
    """Function to double the value"""
    return 3*num
print(double.__doc__)
````

### 1. **What is a Python Statement?**

A Python statement is an instruction that the interpreter executes. A program is a collection of these instructions. Statements in Python perform various actions such as variable assignment, function calls, loops, and conditionals.

Here’s an example of a basic Python statement:

```python
x = 10  # This is an assignment statement
```

In the above example, the statement assigns the value `10` to the variable `x`.

---

### 2. **Types of Statements in Python**

Python supports several types of statements. Below are the most common ones:

#### **a. Expression Statements**

An expression statement is an expression that can be evaluated and it does something, but doesn't directly return any value (like a print function or a variable assignment).

```python
x = 10 + 5  # This is an expression statement
```

This expression calculates the sum of `10` and `5` and assigns it to the variable `x`.

#### **b. Assignment Statements**

An assignment statement is used to assign a value to a variable.

```python
x = 10  # Assignment statement
y = 20
```

Here, the values `10` and `20` are assigned to variables `x` and `y`, respectively.

#### **c. Control Flow Statements**

Control flow statements alter the execution flow of a program. These statements allow decision-making and branching.

- **if-else Statements**:

```python
x = 10
if x > 5:
    print("x is greater than 5")
else:
    print("x is less than or equal to 5")
```

- **elif Statements**:

```python
x = 10
if x > 10:
    print("x is greater than 10")
elif x == 10:
    print("x is equal to 10")
else:
    print("x is less than 10")
```

#### **d. Looping Statements**

Looping statements allow repetitive execution of a block of code.

- **For Loop**:

```python
for i in range(5):
    print(i)  # Prints numbers from 0 to 4
```

- **While Loop**:

```python
x = 0
while x < 5:
    print(x)  # Prints numbers from 0 to 4
    x += 1
```

#### **e. Function Call Statements**

In Python, calling a function is considered a statement. When you call a function, it performs the defined actions.

```python
def greet(name):
    print(f"Hello, {name}")

greet("Alice")  # Function call statement
```

#### **f. Import Statements**

In Python, import statements are used to import modules or specific components of modules.

```python
import math  # Import the entire math module
print(math.sqrt(16))  # Use math module to calculate the square root of 16
```

You can also import specific items from a module:

```python
from math import sqrt
print(sqrt(16))  # Directly using the sqrt function
```

#### **g. Exception Handling Statements**

Python allows you to handle errors using exception handling statements.

- **try-except Statements**:

```python
try:
    num = int(input("Enter a number: "))
except ValueError:
    print("That's not a valid number!")
```

- **else and finally**:

```python
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Division by zero is not allowed!")
else:
    print("Division was successful")
finally:
    print("This will always be executed.")
```

#### **h. Conditional (Ternary) Statements**

A conditional expression (ternary operator) allows you to assign a value based on a condition in a single line.

```python
x = 10
y = 20
result = x if x > y else y
print(result)  # Output: 20 (since x is not greater than y)
```

### 3. **Compound and Simple Statements**

- **Simple Statements**: These are statements that can be written in one line. Examples include assignments, expressions, or imports.

```python

```

- **Compound Statements**: These are statements that can span multiple lines and are used to group a series of actions. These include loops, conditionals, and function definitions.

```python
if x > 5:
    print("x is greater than 5")  # Compound statement with indentation
```

### 4. **Best Practices for Writing Python Statements**

1. **Indentation Matters**: Python uses indentation to group statements. Make sure your code is properly indented (usually 4 spaces).
2. **Avoid Nested Conditionals**: Avoid having too many nested `if` statements. It makes the code harder to read and maintain. Instead, consider using `elif` or breaking your logic into smaller functions.

3. **Use Readable Variables**: Make your variable names meaningful, so they reflect their purpose in the program.

4. **Comment Your Code**: Always comment your statements when they are complex or involve business logic that might not be immediately clear.

5. **Keep Statements Simple**: Try to write simple, easy-to-understand statements. If a statement is too long, consider breaking it up.

### 5. **Advanced Topics Related to Statements**

- **List Comprehensions**: These allow you to create lists in a more concise and readable way.

```python
squares = [x ** 2 for x in range(10)]  # List comprehension
print(squares)  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
```

- **Lambda Functions in Statements**: A lambda function is an anonymous function defined with the `lambda` keyword.

```python
double = lambda x: x * 2
print(double(5))  # Output: 10
```

- **Decorators**: A decorator is a function that modifies the behavior of another function. You can use decorators to add additional functionality to statements.

```python
def decorator_function(original_function):
    def wrapper_function():
        print("Wrapper executed this before {}".format(original_function.__name__))
        return original_function()
    return wrapper_function

@decorator_function
def display():
    return "Display function executed"

print(display())  # This will display the wrapped function result
```

## **1. Indentation in Python**

### **What is Indentation?**

- Indentation refers to the spaces or tabs at the beginning of a line of code.
- In Python, indentation is **mandatory** and is used to define blocks of code (e.g., loops, conditionals, functions).

### **Why is Indentation Important?**

- **Readability**: Proper indentation makes the code easier to read and understand.
- **Structure**: Indentation defines the logical structure of the code.
- **Syntax**: Python uses indentation instead of braces (`{}`) to define code blocks.

---

### **Basic Rules for Indentation**

1. **Consistency**: Use the same number of spaces or tabs for each level of indentation.
2. **Standard Practice**: Use **4 spaces** per indentation level (recommended by PEP 8).
3. **Avoid Mixing Spaces and Tabs**: Mixing spaces and tabs can lead to errors. Stick to one.

---

### **Examples of Indentation**

#### **1. Conditional Statements**

```python
x = 10
if x > 5:
    print("x is greater than 5")  # Indented block
else:
    print("x is less than or equal to 5")  # Indented block
```

#### **2. Loops**

```python
for i in range(5):
    print(i)  # Indented block
```

#### **3. Functions**

```python
def greet(name):
    print(f"Hello, {name}!")  # Indented block
```

---

### **Common Indentation Errors**

1. **Inconsistent Indentation**:

   ```python
   if x > 5:
       print("x is greater than 5")
     print("This will cause an error")  # Inconsistent indentation
   ```

2. **Missing Indentation**:

   ```python
   if x > 5:
   print("x is greater than 5")  # Missing indentation
   ```

3. **Mixing Spaces and Tabs**:
   ```python
   if x > 5:
       print("x is greater than 5")  # Spaces
       print("This will cause an error")  # Tabs
   ```

## **2. Comments in Python**

### **What are Comments?**

- Comments are lines of text in the code that are ignored by the Python interpreter.
- They are used to explain the code, making it easier to understand.

### **Why Use Comments?**

- **Documentation**: Explain the purpose of the code.
- **Debugging**: Temporarily disable code for testing.
- **Collaboration**: Help other developers understand your code.

---

### **Types of Comments**

#### **1. Single-Line Comments**

- Use the `#` symbol to create a single-line comment.

#### **Example**

```python

```

---

#### **2. Multi-Line Comments**

- Python does not have a specific syntax for multi-line comments. Use multiple `#` symbols or a multi-line string (`"""` or `'''`).

#### **Example**

```python

# using multiple single-line comments.

using a multi-line string.
"""
```

---

### **Best Practices for Comments**

1. **Be Clear and Concise**: Write comments that are easy to understand.
2. **Avoid Redundancy**: Do not state the obvious (e.g., `x = 10  # Assign 10 to x`).
3. **Update Comments**: Keep comments up-to-date with the code.
4. **Use Docstrings**: Use docstrings to document functions, classes, and modules.

---

### **Docstrings**

- Docstrings are multi-line strings used to document functions, classes, and modules.
- They are enclosed in triple quotes (`"""` or `'''`) and are accessible via the `__doc__` attribute.

#### **Example**

```python
def add(a, b):
    """
    This function adds two numbers.

    Parameters:
    a (int): The first number.
    b (int): The second number.

    Returns:
    int: The sum of a and b.
    """
    return a + b

print(add.__doc__)
```

### 1. **What is Indentation in Python?**

Indentation refers to the spaces or tabs at the beginning of a line of code. In Python, indentation is used to define the structure and scope of code blocks, such as loops, conditionals, functions, and classes.

#### **Importance of Indentation:**

- Python uses indentation to define code blocks instead of curly braces `{}` or keywords (like `begin` or `end` in other languages). Proper indentation is critical because Python relies on it to understand the hierarchy and flow of the program.
- Inconsistent indentation (mixing tabs and spaces) can lead to `IndentationError` or unexpected behavior.

#### **How Python Uses Indentation:**

- Python uses indentation to indicate a block of code that is executed together. For example, the body of a `for` loop, `if` statement, or a function is indented.
- Python doesn’t require curly braces `{}` to define blocks. Instead, it uses the amount of space (usually 4 spaces) at the beginning of each line to group statements.

#### Example:

```python
if x > 5:
    print("x is greater than 5")  # This is indented under the if block
    y = x * 2                    # This is also part of the same block
else:
    print("x is less than or equal to 5")
```

- Here, the code inside the `if` and `else` blocks is indented, so Python knows these are part of the same logical structure.

#### **Best Practices for Indentation:**

- Use **4 spaces** for indentation. This is the official Python style guide recommended by **PEP 8** (Python Enhancement Proposal).
- Avoid using **tabs** for indentation. If you mix tabs and spaces, Python will raise an `IndentationError`.
- Consistently indent blocks of code, such as function bodies, loops, and conditionals, with 4 spaces.

### 2. **Types of Comments in Python**

Python provides multiple ways to add comments to your code. Comments are essential for explaining what your code does and for marking sections that need attention.

#### **a. Single-line Comments**

A single-line comment starts with a hash symbol `#`. Everything following the `#` symbol on that line is considered a comment and is ignored by the Python interpreter.

```python
# This is a single-line comment
x = 10  # This is an inline comment
```

- **Inline Comments**: These are comments placed at the end of a line of code. They should be used sparingly and should only explain something that is not immediately clear from the code.

```python
y = x * 2  # Multiply x by 2
```

#### **b. Multi-line Comments**

Multi-line comments are typically created by enclosing the comment block in triple quotes (`'''` or `"""`). This is useful for comments that span more than one line or for temporarily disabling code.

```python
'''
This is a multi-line comment.
You can write as much text as you like here.
It will be ignored by Python.
'''
```

- Python doesn’t have a specific multi-line comment syntax. Triple quotes are generally used for docstrings (see below), but they can also serve as multi-line comments.

Alternatively, you can use multiple single-line comments:

```python
# This is a multi-line comment
# spanning multiple lines.
# Each line starts with a hash symbol.
```

#### **c. Docstrings (Documentation Strings)**

A docstring is a special type of multi-line comment used to describe modules, functions, classes, or methods. Docstrings are not ignored by Python and can be accessed using the `.__doc__` attribute. They provide a way to document your code and make it more understandable for others (or yourself).

- Docstrings are enclosed in triple quotes (`"""` or `'''`), and they are generally used to describe the purpose, parameters, and return values of functions or classes.

Example of a function docstring:

```python
def greet(name):
    """
    This function takes a name as input and prints a greeting message.

    Parameters:
    name (str): The name of the person to greet.

    Returns:
    None
    """
    print(f"Hello, {name}!")
```

- The docstring explains what the function does, its parameters, and its return value. You can access this documentation using:

```python
print(greet.__doc__)  # Prints the docstring of the function
```

### 3. **Best Practices for Comments**

While comments are helpful, it is important to write meaningful and concise comments that improve the readability of your code. Here are some best practices for writing comments:

1. **Write Comments to Explain Why, Not What**:

   - Try to explain **why** the code does something, not **what** it does. The code itself should be clear enough to explain the "what."
   - Example:
     ```python
     # Instead of: "Increment the counter by 1"
     counter += 1  # Increment counter for every new item processed
     ```

2. **Be Concise and Clear**:

   - Keep comments short and to the point. Don’t write unnecessary comments that repeat what is already obvious from the code.
   - Example:

     ```python
     # Good comment
     total_sum = sum(numbers)  # Sum all numbers in the list

     # Bad comment
     total_sum = sum(numbers)  # Adding all the numbers in the list to get the total sum
     ```

3. **Comment Complex or Non-Obvious Code**:

   - Add comments to explain complex algorithms or steps that may not be immediately clear to other developers.
   - Example:
     ```python
     # Using dynamic programming to calculate Fibonacci number
     fib = [0, 1]
     for i in range(2, n):
         fib.append(fib[i-1] + fib[i-2])  # Add previous two Fibonacci numbers
     ```

4. **Avoid Obvious Comments**:

   - Don’t over-comment code. Avoid commenting code that is self-explanatory or trivially obvious.
   - Example of a bad comment:
     ```python
     x = 5  # Assign 5 to x
     ```

5. **Use Comments for TODOs**:

   - Use comments to mark parts of your code that still need to be completed or improved.
   - Example:
     ```python
     # TODO: Refactor this code to make it more efficient
     ```

6. **Update Comments Regularly**:
   - Make sure that comments are updated when the code changes. Outdated comments can confuse others (and your future self).
