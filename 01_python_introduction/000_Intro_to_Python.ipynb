### 1. Basic Python Concepts

#### 1.1. What is Python?

Python is a **general-purpose** programming language that allows you to write clean, readable, and efficient code. Its key features include:

- **Dynamic Typing**: You don’t need to explicitly declare the type of a variable.
- **Interpreted**: Python code is executed line-by-line, which makes debugging easier.
- **High-level Language**: Python abstracts away many complex details of the computer, allowing developers to focus on solving problems.
- **Extensive Libraries**: Python comes with a vast standard library and a vibrant ecosystem of third-party packages.
- **Cross-platform**: It works on major operating systems like Windows, Linux, and macOS.

#### 1.2. Python Syntax and Structure

Python uses indentation to define blocks of code, making it visually clean and easy to follow.

```python
def greet(name):
    print(f"Hello, {name}!")

greet("World")  # Output: Hello, World!
```

#### 1.3. Variables and Data Types

Variables in Python do not require explicit type declarations. Python automatically determines the type based on the value assigned.

```python
x = 10          # Integer
y = "Hello"     # String
z = 3.14        # Float
is_active = True  # Boolean
```

#### 1.4. Input/Output

You can take user input using the `input()` function and print outputs using `print()`.

```python
name = input("Enter your name: ")  # Takes user input
print(f"Hello, {name}!")
```

#### 1.5. Basic Operators

Python supports arithmetic, comparison, and logical operators:

```python
a = 5
b = 10

# Arithmetic
print(a + b)  # 15
print(a - b)  # -5
print(a * b)  # 50
print(a / b)  # 0.5

# Comparison
print(a > b)  # False
print(a == b) # False

# Logical
print(a > 3 and b > 8)  # True
```

### 2. Control Flow

#### 2.1. Conditionals (if-else)

Python uses `if`, `elif`, and `else` to handle conditional logic.

```python
x = 5
if x > 10:
    print("Greater than 10")
elif x == 5:
    print("Equal to 5")
else:
    print("Less than 5")
```

#### 2.2. Loops (for, while)

Python provides `for` and `while` loops to iterate over sequences.

```python
# For loop
for i in range(5):
    print(i)

# While loop
count = 0
while count < 5:
    print(count)
    count += 1
```

#### 2.3. Comprehensions

List, dictionary, and set comprehensions allow you to create and manipulate collections in a concise and efficient way.

```python
# List comprehension
squares = [x**2 for x in range(5)]
print(squares)  # [0, 1, 4, 9, 16]

# Dictionary comprehension
squared_dict = {x: x**2 for x in range(5)}
print(squared_dict)  # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
```

### 3. Functions

#### 3.1. Defining Functions

Functions in Python are defined using the `def` keyword.

```python
def greet(name):
    print(f"Hello, {name}!")
```

#### 3.2. Arguments and Return Values

You can pass arguments to functions and return values.

```python
def add(x, y):
    return x + y

result = add(5, 3)
print(result)  # Output: 8
```

#### 3.3. Lambda Functions

A lambda function is a small anonymous function defined using the `lambda` keyword.

```python
square = lambda x: x**2
print(square(4))  # Output: 16
```

#### 3.4. Recursive Functions

A recursive function calls itself to solve problems.

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

print(factorial(5))  # Output: 120
```

### 5. Object-Oriented Programming (OOP)

#### 5.1. Classes and Objects

Python supports OOP concepts, including classes, objects, and methods.

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        print(f"{self.name} barks!")

dog1 = Dog("Buddy")
dog1.bark()  # Output: Buddy barks!
```

#### 5.2. Inheritance and Polymorphism

Inheritance allows you to create new classes based on existing ones, while polymorphism allows you to use methods in different ways.

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Woof!")

dog = Dog()
dog.speak()  # Output: Woof!
```

### 8. Advanced Python Concepts

#### 8.1. Decorators

Decorators are functions that modify the behavior of other functions or methods.

```python
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

#### 8.2. Generators and Iterators

Generators allow you to create iterators in a memory-efficient way using the `yield` keyword.

```python
def my_gen():
    for i in range(3):
        yield i

gen = my_gen()
for value in gen:
    print(value)  # Output: 0 1 2
```

- **Dictionaries**: Key-value pairs.

#### **Example**

```python
# Lists
fruits = ["apple", "banana", "cherry"]
fruits.append("orange")
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'orange']

# Tuples
colors = ("red", "green", "blue")
print(colors[0])  # Output: red

# Dictionaries
person = {"name": "Alice", "age": 25}
print(person["name"])  # Output: Alice
```

## **4. Advanced Concepts**

### **1. Generators and Iterators**

- Generators use `yield` to produce a sequence of values lazily.

#### **Example**

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

for num in fibonacci(5):
    print(num)  # Output: 0 1 1 2 3
```

---

### **2. Decorators**

- Decorators are used to modify the behavior of functions.

#### **Example**

```python
def my_decorator(func):
    def wrapper():
        print("Before the function call")
        func()
        print("After the function call")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
# Output:
# Before the function call
# Hello!
# After the function call
```
